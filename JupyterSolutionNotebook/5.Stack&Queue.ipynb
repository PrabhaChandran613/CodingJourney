## Problem Statement : 137 
Implement a stack using an array with two operations:  
1. **push(x)** – Push integer `x` onto the stack.  
2. **pop()** – Remove and return the top element of the stack. If the stack is empty, return `-1`.  

**Input:**  
- Queries are given in the following format:  
  - `1 x` → Push `x` onto the stack.  
  - `2` → Pop and print the top element from the stack.  
- The input consists of multiple test cases.  

**Output:**  
For each pop operation (`2` query type), print the popped element.  

**Constraints:**  
- Stack size **≤ 1000** (Fixed size array).  
- The number of queries **≤ 10^5**.  
- Each element **fits within 32-bit integer range**.  

**Example**  
**Input:**  
```

6
1 15
1 20
2
1 35
2
2

```

**Output:**  
```

20
35
15

```

**Goal of the Code:**  
Efficiently implement a stack using an array with **O(1) time complexity** for `push()` and `pop()` operations.  

**Approach:**  
- Use an **integer array** `arr[]` of fixed size (`1000`).
- Maintain an integer `top`, initialized to `-1`, representing the index of the top element.  
- **push(x)**:  
  - If the stack is **not full**, increment `top` and store `x` at `arr[top]`.  
  - If full, do nothing (stack overflow is ignored).  
- **pop()**:  
  - If `top == -1`, return `-1` (stack is empty).  
  - Otherwise, return `arr[top]` and decrement `top`.  

**Pseudocode/Algorithm:**  
1. **Initialize** `arr[]` (size 1000) and `top = -1`.  
2. **push(x) Function:**  
   - If `top < 999`, increment `top` and store `x`.  
   - Else, do nothing (stack overflow case).  
3. **pop() Function:**  
   - If `top == -1`, return `-1` (stack is empty).  
   - Otherwise, return `arr[top]` and decrement `top`.  
4. **Process Queries:**  
   - Read multiple queries from input and perform `push()` or `pop()`.  
   - Print results of pop operations.  

**🧮 Flowchart/ Diagram:**  
```sql

Start → Initialize stack (array\[1000], top = -1)
↓
Read Query → Is query type 1 (push)?
→ Yes → Is stack full?
→ No → top++, arr\[top] = x
→ Yes → Do nothing
→ No → Is query type 2 (pop)?
→ Is stack empty?
→ Yes → Print -1
→ No → Print arr\[top], top--
↓
Repeat until all queries are processed
↓
End

```
**Code:**
```java

class MyStack {
    private int[] arr;
    private int top;

    public MyStack() {
        arr = new int[1000];
        top = -1;
    }

    public void push(int x) {
        // Your 
        if(top==arr.length-1){
            return;
        }
      
         arr[++top]=x;
     
    }

    public int pop() {
        // Your Code
        if(top==-1){
            return -1;
        }
        return arr[top--];
    }
}
```

**Time and Space Complexity:**  
- **Time Complexity:**  
  - `push(x)`: **O(1)**  
  - `pop()`: **O(1)**  
  - **Overall:** **O(n)** for `n` queries.  
- **Space Complexity:** **O(1)** (Fixed array of size 1000).  

**Additional Notes/Hints:**  
- Stack follows **LIFO (Last In, First Out)** principle.  
- Using a **fixed-size array** is memory-efficient but lacks dynamic resizing.  
- If **dynamic size** is needed, consider using `ArrayList` or `LinkedList`.  


## Problem Statement: 138
**Two Stacks in an Array** 
You are required to implement two stacks using a single array efficiently. The operations that need to be implemented are:  

1. **twoStacks()** – Initializes the data structure and variables needed to manage two stacks in one array.  
2. **push1(int x)** – Pushes an element `x` into the first stack.  
3. **push2(int x)** – Pushes an element `x` into the second stack.  
4. **pop1()** – Pops an element from the first stack and returns the popped element. If the first stack is empty, it should return `-1`.  
5. **pop2()** – Pops an element from the second stack and returns the popped element. If the second stack is empty, it should return `-1`.  

**Input**  
- The input consists of multiple test cases.  
- Each test case includes a number of queries that specify operations on either of the two stacks.  

**Output**  
- Output consists of values returned by the `pop1()` and `pop2()` operations.  
- If a stack is empty, popping from it should return `-1`.  

**Constraints**  
- The array should be used efficiently without wasting extra space.  
- The time complexity for `push` and `pop` operations should be **O(1)**.  
- Stack operations should handle overflow and underflow conditions gracefully.  

**Example**  
**Input:**  
```

8
1 1 10
1 2 20
1 1 30
2 1
2 2
1 2 40
2 2
2 1

```

**Output:**  
```

30
20
40
10

```

**Goal of the Code**  
The main objective is to implement **two stacks using a single array** while ensuring efficient space utilization and constant-time operations for push and pop.  

**Approach** 
To efficiently manage two stacks in one array, we divide the array into two parts:  
- **Stack 1** grows from the left (index `0` onwards).  
- **Stack 2** grows from the right (index `size-1` backwards).  
- A push operation adds an element at the respective stack's top index.  
- A pop operation removes and returns the top element of the respective stack.  
- We ensure that both stacks do not overlap; otherwise, an overflow condition occurs.  

**Pseudocode/Algorithm** 

1. Initialization (`twoStacks`)  
1. Define an array `arr[]` of fixed size `size`.  
2. Set `top1 = -1` (indicating the top of Stack 1).  
3. Set `top2 = size` (indicating the top of Stack 2).  

2. Push Operations  
**push1(int x)** (Push to Stack 1)  
1. Check if `top1 < top2 - 1`. If true:  
   - Increment `top1` and place `x` at `arr[top1]`.  
   - If false, print "Stack Overflow".  

**push2(int x)** (Push to Stack 2)  
1. Check if `top1 < top2 - 1`. If true:  
   - Decrement `top2` and place `x` at `arr[top2]`.  
   - If false, print "Stack Overflow".  

3. Pop Operations  
**pop1()** (Pop from Stack 1)  
1. If `top1 >= 0`:  
   - Retrieve `arr[top1]` and decrement `top1`.  
   - Return the retrieved value.  
2. Else, return `-1` (Stack 1 is empty).  

**pop2()** (Pop from Stack 2)  
1. If `top2 < size`:  
   - Retrieve `arr[top2]` and increment `top2`.  
   - Return the retrieved value.  
2. Else, return `-1` (Stack 2 is empty).  

**🧮 Flowchart/ Diagram:**  
```sql

Start → Initialize array, top1 = -1, top2 = size
↓
Read next query
↓
Is it push1? → Check space → top1++ → arr\[top1] = x
Is it push2? → Check space → top2-- → arr\[top2] = x
Is it pop1? → top1 >= 0? → print arr\[top1--] else print -1
Is it pop2? → top2 < size? → print arr\[top2++] else print -1
↓
Repeat for all queries
↓
End

```

**Code:**
```java
class twoStacks {
      int size;
    int[] arr;
    int top1, top2;
    
    twoStacks() {
        this.size=100;
        arr=new int[size];
        top1=-1;
        top2=size;
    }

    // Function to push an integer into the stack1.
    void push1(int x) {
      if(top1<top2-1){
         arr[++top1]=x;
      }
      else{
          System.out.println("Stack Overflow");
          return;
      }
    }

    // Function to push an integer into the stack2.
    void push2(int x) {
      if(top1<top2-1){
          arr[--top2]=x;
      }
      else{
           System.out.println("Stack Overflow");
           return;
      }
    }

    // Function to remove an element from top of the stack1.
    int pop1() {
      if(top1>=0){
         int x=arr[top1];
          top1--;
          return x;
      }
      return -1;
    }

    // Function to remove an element from top of the stack2.
    int pop2() {
       if(top2<size){
          int x=arr[top2];
           top2++;
           return x;
       }
       
           return -1;
      
    }
}
```

**Time and Space Complexity** 
- **Time Complexity**: O(1) for all operations (`push1`, `push2`, `pop1`, `pop2`).  
- **Space Complexity**: O(n) where `n` is the size of the array.  

**Additional Notes**  
- This approach ensures efficient space utilization by allowing both stacks to grow dynamically within the array.  
- If the two stacks meet in the middle, it indicates that the array is fully occupied.  
- This implementation prevents unused space, unlike naive implementations that preallocate separate halves of the array for each stack.  



## Problem Statement: **Implement Two Stacks in an Array**

Create a data structure `TwoStacks` that represents two stacks using a **single array**. Both stacks should share the same array space for storing elements.

Implement the following operations using one array to simulate two stacks:

- `push1(int x)` — push element `x` to the **first** stack.
- `push2(int x)` — push element `x` to the **second** stack.
- `pop1()` — pop an element from the **first** stack and return it.
- `pop2()` — pop an element from the **second** stack and return it.

**Example:**
```
push1(2)
push1(3)
push2(4)
pop1()  -> 3
pop2()  -> 4
pop2()  -> -1   (since second stack is empty)
```

**Input:**
- A sequence of `push1`, `push2`, `pop1`, `pop2` operations.

**Output:**  
- For each pop operation, return the popped element or `-1` if the corresponding stack is empty.

**Constraints:**  
- Both stacks should share the same array.
- Efficient space utilization without wasting array space.

**Goal:** 
Efficiently implement two stacks in one array such that no space is wasted, and operations run in constant time.

**Approach:**  

Although the provided code uses two separate Java `Stack` objects, the classic problem is about implementing **two stacks in a single array** to save space. The key idea is:

- Start `stack1` from the **beginning** of the array (index 0) and grow towards the right.
- Start `stack2` from the **end** of the array (last index) and grow towards the left.
- When both stacks meet (cross each other), the array is full.

This allows maximum utilization of array space.

**🧮 Flowchart/Diagram:**

```sql
Array: [ s1 ... ... ... s2 ]

Stack1 pointer grows from left -> right
Stack2 pointer grows from right -> left

If pointers cross, no space left.
```

**Pseudocode/Algorithm:**

```sql
Initialize array arr of size n
Initialize top1 = -1  // stack1 top index
Initialize top2 = n   // stack2 top index

function push1(x):
    if top1 + 1 == top2:
        print "Stack Overflow"
        return
    top1 += 1
    arr[top1] = x

function push2(x):
    if top2 - 1 == top1:
        print "Stack Overflow"
        return
    top2 -= 1
    arr[top2] = x

function pop1():
    if top1 == -1:
        return -1  // stack1 empty
    val = arr[top1]
    top1 -= 1
    return val

function pop2():
    if top2 == n:
        return -1  // stack2 empty
    val = arr[top2]
    top2 += 1
    return val
```

**Code (Given Version - Uses Two Java Stacks)**

```java
import java.util.Stack;

class TwoStacks {
    private Stack<Integer> stack1;
    private Stack<Integer> stack2;

    public TwoStacks() {
        stack1 = new Stack<>();
        stack2 = new Stack<>();
    }

    public void push1(int x) {
        stack1.push(x);
    }

    public void push2(int x) {
        stack2.push(x);
    }

    public int pop1() {
        if (!stack1.isEmpty()) {
            return stack1.pop();
        } else {
            return -1;
        }
    }

    public int pop2() {
        if (!stack2.isEmpty()) {
            return stack2.pop();
        } else {
            return -1;
        }
    }

    public static void main(String[] args) {
        TwoStacks ts = new TwoStacks();
        ts.push1(2);
        ts.push1(3);
        ts.push2(4);

        System.out.print(ts.pop1() + " ");  // Output: 3
        System.out.print(ts.pop2() + " ");  // Output: 4
        System.out.print(ts.pop2() + " ");  // Output: -1 (empty)
    }
}
```

**Time Complexity:**

* Each operation (`push1`, `push2`, `pop1`, `pop2`) runs in **O(1)** time.

**Space Complexity:**

* Uses space proportional to the number of elements pushed (or the size of the underlying array).

**Notes:**

* The classic problem is to implement two stacks in **one fixed-size array** to maximize space usage, but the above code uses Java's `Stack` class which internally uses separate arrays.
* For an array-based solution, pointers from both ends are used as described in the approach.
* The array-based solution is preferred in memory-constrained environments.


## Problem Statement: 139
**Parenthesis Checker**  
Given a string `s` composed of different combinations of parentheses `'(' , ')' , '{' , '}' , '[' , ']'`, verify whether the arrangement of brackets is **valid**.  

A string is considered **valid** if:  
1. Open brackets must be closed by the same type of brackets.  
2. Open brackets must be closed in the correct order.  

**Input**  
- The input consists of multiple test cases.  
- Each test case contains a single string `s` of length `n`.  

**Output**  
- Print `"true"` if the given string is valid, otherwise print `"false"`.  

**Constraints**  
- `1 ≤ n ≤ 10^5` (length of the string)  
- The string only consists of `'(' , ')' , '{' , '}' , '[' , ']'`.  

**Example**  
**Input:**  
```

3
()
({\[]})
(\[)]

```

**Output:**  
```

true
true
false

```

**Goal of the Code**  
To determine if a given string of parentheses is **balanced** by using a **stack-based approach**.  

**Approach**  
To validate the arrangement of brackets:  
1. Use a **stack** to keep track of the last opened bracket.  
2. Traverse the string character by character:  
   - If it is an **opening bracket** (`'('`, `'{'`, `'['`), push it onto the stack.  
   - If it is a **closing bracket** (`')'`, `'}'`, `']'`):  
     - Check if the stack is empty (indicating an unmatched closing bracket).  
     - Pop the top element from the stack and verify if it matches the corresponding opening bracket.  
     - If it does not match, return `"false"`.  
3. After traversing the string, if the stack is **empty**, return `"true"`, otherwise return `"false"`.  

**Pseudocode/Algorithm**    
1. Initialize an empty **stack**.  
2. Traverse each character `ch` in the string `s`:  
   - If `ch` is `'('`, `'{'`, `'['`, push it onto the stack.  
   - If `ch` is `')'`, `'}'`, or `']'`:  
     - If the stack is empty, return `false` (unmatched closing bracket).  
     - Pop the top element from the stack.  
     - Check if the popped element matches the correct opening bracket.  
     - If not, return `false`.  
3. After the loop, return `true` if the stack is empty (all brackets matched), otherwise return `false`.  

**🧮 Flowchart/ Diagram:**  
```sql

Start → Initialize empty stack
↓
For each character in string:
↓
Is it opening bracket? → Push to stack
Is it closing bracket?
→ Is stack empty? → Yes → return false
→ No → Pop from stack
→ Does it match opening?
→ No → return false
↓
After loop → Is stack empty?
→ Yes → return true
→ No → return false
↓
End

```

**Code:**
```java
class Solution {
    static boolean isBalanced(String s) {
      Stack<Character> stack= new Stack<>();
      
      for(char ch: s.toCharArray()){
          if(ch=='(' || ch=='{' || ch=='['){
              stack.push(ch);
          }
          else{
              if(stack.isEmpty()) return false;
          
              char top=stack.pop();
              
              if(ch=='}' && top!='{'  ||
                 ch==')' && top!='('  ||
                 ch==']' && top!='[' ){
                     return false;
                 }
          }
      }
      
      return stack.isEmpty();
      
      
   }
}
```

- **Time Complexity:** **O(n)** (each character is pushed and popped at most once).  
- **Space Complexity:** **O(n)** (in the worst case, all brackets are opening and stored in the stack).  

**Additional Notes**
The solution uses a stack to efficiently track and match opening and closing brackets.

**Edge cases handled:**
Unmatched brackets (e.g., "{{}", "([)]").
Empty string "" (should return "true").
String with only opening brackets ("(((") or only closing brackets (")))").


## Problem Statement: 141
**Get Min from Stack**   
You are given **q queries** where you need to implement a stack with the following functions:  

1. **push(x)** – Insert an integer `x` onto the stack.  
2. **pop()** – Remove the top element from the stack.  
3. **peek()** – Return the top element from the stack. If the stack is empty, return `-1`.  
4. **getMin()** – Retrieve the minimum element from the stack in **O(1) time**. If the stack is empty, return `-1`.  

Each query can be one of the following:  
- `1 x` → Push `x` onto the stack.  
- `2` → Pop the top element from the stack.  
- `3` → Return the top element from the stack. If empty, return `-1`.  
- `4` → Return the minimum element from the stack. If empty, return `-1`.  

**Input**  
- **q** (1 ≤ q ≤ 10^5) - Number of queries.  
- Each query consists of a type (1, 2, 3, or 4) followed by an integer in case of push operation (`1 x`).  

**Output**  
- For each `3` query, print the top element of the stack.  
- For each `4` query, print the minimum element in the stack.  

**Constraints**  
- The stack operations must be performed in **O(1)** time complexity.  
- The stack can store large numbers (within the range of integers).  
- The space complexity should be **O(n)** (no extra space should be used other than the stack itself).  

**Example**
**Input:**
```

8
1 5
1 3
4
3
2
4
3
2

```

**Output:**
```

3
3
5
5

```

**Goal of the Code**  
To efficiently **push, pop, get the top, and find the minimum element in O(1) time** using a **modified stack approach**.  


**Approach**  
A **normal stack** does not support retrieving the minimum element in **O(1)** time.  
To achieve this, we use a **trick with an auxiliary minimum tracking variable** (`minEle`).  

#Optimized Stack Approach:
- Instead of using an **extra stack**, store **modified values** when pushing an element **smaller than minEle**.  
- If the new element `x` is smaller than `minEle`:  
  - Push a **modified value** `2*x - minEle` onto the stack.  
  - Update `minEle = x`.  
- When popping, if the popped value is **less than minEle**, it means `minEle` was updated and needs to be restored.  

#Explanation:
1. **Push(x)**
   - If the stack is empty, set `minEle = x` and push `x`.
   - If `x < minEle`, push `2*x - minEle` instead and update `minEle = x`.
   - Otherwise, push `x` normally.

2. **Pop()**
   - If the popped element is **less than minEle**, it was a special stored value, so update `minEle = 2*minEle - poppedValue`.

3. **Peek()**
   - If the top element is **less than minEle**, return `minEle` (since the stored value is modified). Otherwise, return the top element.

4. **getMin()**
   - Return `minEle` as it stores the **minimum element in O(1) time**.


**Pseudocode/Algorithm**  
 
1. **Initialize**: Create a stack `s` and a variable `minEle` to store the current minimum.  
2. **Push(x)**:
   - If the stack is empty, set `minEle = x` and push `x`.
   - If `x < minEle`, push `2*x - minEle` (modified value) and update `minEle = x`.
   - Else, push `x` normally.
3. **Pop()**:
   - If the popped value is **less than minEle**, update `minEle = 2*minEle - poppedValue`.
4. **Peek()**:
   - If the top element is **less than minEle**, return `minEle` (real value).
   - Else, return the top element.
5. **getMin()**:
   - Return `minEle`.

**🧮 Flowchart/ Diagram:**

```sql
      ┌───────────────┐
      │   Start       │
      └─────┬─────────┘
            ↓
     ┌─────────────┐
     │  Read Query │
     └─────┬───────┘
           ↓
 ┌──────────────────────┐
 │  Query Type = 1?     │─Yes─> Push(x) with min logic
 └─────┬─────────────┬──┘
       │             │
     No│           Query Type = 2? ──Yes─> Pop() with min restore
       ↓             ↓
 Query Type = 3?   Query Type = 4? 
       ↓             ↓
    Peek()        getMin()
       ↓             ↓
    Print           Print
       ↓             ↓
     ┌────────────────┐
     │ More Queries?  │
     └──────┬─────────┘
            ↓
          End
```

**Code:**
```java


class Solution {
    public Solution() {}
    Stack<Integer> s=new Stack<>();
    int minEle=-1;
    // Add an element to the top of Stack
    public void push(int x) {
        // code here
        if(s.isEmpty()){
        minEle=x;
        s.push(x);
        }
        else if(x<minEle){
            s.push(2*x - minEle);
            minEle=x;
        }
        else{
            s.push(x);
        }
    }

    // Remove the top element from the Stack
    public void pop() {
        if(s.isEmpty()){
            return;
        }
        int top=s.pop();
        if(top<minEle){
            minEle=(2*minEle - top);
        }
    }

    // Returns top element of the Stack
    public int peek() {
        if(s.isEmpty()){
            return -1;
        }
        int top=s.peek();
        return (top<minEle) ? minEle : top;

    }

    // Finds minimum element of Stack
    public int getMin() {
        if(s.isEmpty()){
            return -1;
        }
        return minEle;
    }
}
```

## Problem Statement: **Design a Stack that Supports getMin() in O(1) Time**

Design a data structure `SpecialStack` that supports all standard stack operations — `push()`, `pop()`, `peek()` — and an additional operation `getMin()`, which returns the minimum element in the stack. All operations must run in **O(1)** time.
 
Implement a stack with the following functionalities:

Each operation should run in constant time, **O(1)**.

**Example:**
```
push(2)
push(3)
peek()     -> 3
pop()
getMin()   -> 2
push(1)
getMin()   -> 1
```

**Input:**
- A sequence of operations such as push, pop, peek, and getMin.

**Output:**  
- Output depends on the operation, e.g., the top element for peek, the minimum element for getMin, or no output for push/pop.

**Constraints:**  
- Operations must be implemented with **O(1)** time complexity.
- Stack can contain any integer values.
- Space complexity should be optimized as much as possible (using O(n) auxiliary space is acceptable).

**Goal:**  
To efficiently maintain and retrieve the minimum element of the stack at any given time, without scanning the entire stack.

**Approach:**  

We use an auxiliary stack approach where each element pushed to the main stack is stored alongside the minimum element up to that point. This allows retrieving the minimum element in O(1) time.

- Store elements in the stack as pairs: `[value, currentMin]`.
- When pushing a new element, determine the new minimum by comparing the element with the current minimum (top of stack).
- When popping, pop the pair.
- The current minimum is always stored with the top element.

**🧮 Flowchart/Diagram:**

```sql
START
  |
  v
Push(x) --> Is stack empty?
            /          \
           Yes         No
            |           |
   Push [x, x]    Peek top min value
                   Compare x with min
                   Push [x, min(x, topMin)]
  |
Pop() --> Remove top pair
  |
Peek() --> Return top element's value
  |
getMin() --> Return top element's min value
  |
END
```

**Pseudocode/Algorithm:**

```sql
Initialize stack S as empty

function push(x):
    if S is empty:
        S.push([x, x])  // element and min both are x
    else:
        currentMin = S.peek()[1]
        newMin = min(x, currentMin)
        S.push([x, newMin])

function pop():
    if S is not empty:
        S.pop()

function peek():
    if S is empty:
        return -1
    else:
        return S.peek()[0]

function getMin():
    if S is empty:
        return -1
    else:
        return S.peek()[1]
```

**Code**

```java
import java.util.Stack;

class SpecialStack {
    private Stack<int[]> s;

    public SpecialStack() {
        s = new Stack<>();
    }

    // Add an element to the top of Stack
    public void push(int x) {
        int newMin = s.isEmpty() ? x : Math.min(x, s.peek()[1]);
        s.push(new int[]{x, newMin});
    }

    // Remove the top element from the Stack
    public void pop() {
        if (!s.isEmpty()) {
            s.pop();
        }
    }

    // Returns top element of the Stack
    public int peek() {
        return s.isEmpty() ? -1 : s.peek()[0];
    }

    // Finds minimum element of Stack
    public int getMin() {
        return s.isEmpty() ? -1 : s.peek()[1];
    }

    public static void main(String[] args) {
        SpecialStack ss = new SpecialStack();

        ss.push(2);
        ss.push(3);
        System.out.print(ss.peek() + " ");      // Output: 3
        ss.pop();
        System.out.print(ss.getMin() + " ");    // Output: 2
        ss.push(1);
        System.out.print(ss.getMin() + " ");    // Output: 1
    }
}
```

* This technique allows constant time minimum retrieval while keeping stack operations efficient.

**Approach:**  
1. **Use a Stack (Monotonic Decreasing Stack Approach)**:  
   - Traverse the array from **right to left**.  
   - Maintain a **stack** to store elements in **decreasing order** (smallest elements are removed first).  
   - For each element:  
     - **Pop** elements from the stack while the top is **≤ current element** (since they can't be NGE).

**Output**:  
```
\[1, 1, -1]
```

**Explanation**:  
- For `2`, the next smaller is `1`.  
- For `3`, the next smaller is `1`.  
- For `1`, there is no next smaller element, so the answer is `-1`.

**🚀 Approach** 

We can solve this efficiently using a **Monotonic Stack** in **O(n)** time.

**Steps:**
1. **Traverse the array from right to left**.
2. Maintain a **stack** to store candidates for the next smaller elements.
3. For each element:
   - Pop elements from the stack that are **greater than or equal to** the current element.
   - If the stack is not empty after popping, the **top** is the next smaller element.
   - If the stack is empty, **no next smaller** exists (`-1`).
   - **Push the current element** to the stack.

**🧠 Pseudocode**
```plaintext
function nextSmallerElement(arr, n):
    stack = empty
    result = array of size n

    for i from n-1 to 0:
        while stack not empty and stack.top() >= arr[i]:
            stack.pop()

        if stack is empty:
            result[i] = -1
        else:
            result[i] = stack.top()

        stack.push(arr[i])

    return result
```

**✅ Java Implementation**

```java
import java.util.*;

public class Solution {
    static ArrayList<Integer> nextSmallerElement(ArrayList<Integer> arr, int n) {
        ArrayList<Integer> res = new ArrayList<>(Collections.nCopies(n, -1));
        Stack<Integer> st = new Stack<>();

        for (int i = n - 1; i >= 0; i--) {
            while (!st.isEmpty() && st.peek() >= arr.get(i)) {
                st.pop();
            }

            if (!st.isEmpty()) {
                res.set(i, st.peek());
            }

            st.push(arr.get(i));
        }

        return res;
    }

    public static void main(String[] args) {
        ArrayList<Integer> arr = new ArrayList<>(Arrays.asList(2, 3, 1));
        ArrayList<Integer> result = nextSmallerElement(arr, arr.size());
        System.out.println(result); // Output: [1, 1, -1]
    }
}
```

3. **Return** the computed `span[]`.

**Pseudocode/Algorithm**  
```plaintext
1. Initialize a stack `s` (stores indices).
2. Create an array `span[]` of size `n` initialized to 0.
3. Loop through `i = 0` to `n-1`:
   a. While `s` is not empty and `arr[s.peek()] <= arr[i]`, pop from `s`.
   b. If `s` is empty, set `span[i] = i + 1` (since no previous greater value exists).
   c. Else, set `span[i] = i - s.peek()` (difference from last greater price index).
   d. Push `i` onto the stack.
4. Return `span[]`.
```

```java
class Solution {
    public ArrayList<Integer> calculateSpan(int[] arr) {
        // write code here
        int n=arr.length;
        ArrayList<Integer> span=new ArrayList<>(Collections.nCopies(n,0));
        Stack<Integer> s=new Stack<>();
        
        for(int i=0; i<n; i++){
            while(!s.isEmpty() && arr[s.peek()]<=arr[i]){
                s.pop();
            }
            
            if(s.isEmpty()){
                span.set(i,(i+1));
            }
            else{
                span.set(i,(i-s.peek()));
            }
            
            s.push(i);
        }
        return span;
    }
}
```

## Problem: Trapping Rain Water Problem - Tutorial with Illustrations  

The **Trapping Rainwater Problem** states that given an array of `n` non-negative integers `arr[]` representing an **elevation map**, where the width of each bar is **1**, compute how much water it can trap **after raining**.

**Input**  
- An **array** `arr[]` of **non-negative integers** representing heights.  

**Output**  
- An **integer** representing the **total units of rain water** that can be trapped between the bars.

**Constraints** 
- `1 ≤ arr.length ≤ 10^5`  
- `0 ≤ arr[i] ≤ 10^9`  

**Example**

**Input:**

`arr[] = {2, 1, 5, 3, 1, 0, 4}`

**Visualization:**

```
      |           
  |   |       |   
  |   |   |   |   
  | | | | | | | |
-----------------
 0 1 2 3 4 5 6 
```

**Output:**

`6` units of water are trapped in the valleys.

**Goal**  
To compute the **total volume of water trapped** between bars using an efficient algorithm.

**Approach**  
We use a **stack-based approach** to solve this in **O(n)** time.  

**Key Observations:**  
1. Water can be trapped **only if there is a boundary on both left and right** of a lower height bar.
2. The **amount of water above a bar** is determined by the **minimum of left and right boundaries** minus the height of the current bar.

**Stack-Based Algorithm:**  
- Use a **stack** to keep track of indices of bars.  
- Traverse the array:  
  - While the **current bar is higher** than the **top of the stack**, calculate the water trapped.  
  - Pop from stack, and if the stack becomes empty, break (no left boundary).  
  - Use the **distance between the current and new top** of the stack as width.  
  - Use the **min of heights** of current bar and stack top, subtract the popped bar height, and multiply by width to find water trapped.  
- Push the current index into the stack.  

**Pseudocode**  
```sql
Initialize stack, res = 0

For each index i in array:
    While stack is not empty and arr[i] > arr[stack.top]:
        pop = arr[stack.pop()]
        If stack is empty, break
        distance = i - stack.top - 1
        bounded_height = min(arr[stack.top], arr[i]) - pop
        res += distance * bounded_height

    Push i to stack

Return res
```

**Java Implementation**

**Goal of the Code:**  
Efficiently compute the **largest rectangular area** in the histogram using a **monotonic stack** with **O(n) time complexity**.  

**Approach:**  
1. **Use a Monotonic Increasing Stack:**  
   - The stack stores **indices of histogram bars** in **increasing order of height**.  
   - When a **smaller height** is encountered, it means a **rectangular boundary is found**, and we calculate the **area**.  

2. **Processing Steps:**  
   - **Iterate through the histogram (`arr[]`)** while keeping track of indices in a stack.  
   - For **each bar `arr[i]`**, if it is **smaller** than the **bar at top of the stack**,  
     - **Pop from the stack** and compute the **largest possible area** with the popped height as the smallest bar.  
     - **Width Calculation:**  
       - If the stack is **empty**, it means the popped bar was the **smallest**, so width = `i`.  
       - Otherwise, width = `i - s.peek() - 1`.  
   - **Push the current index** into the stack.  
   - **After the loop**, continue popping remaining indices from the stack to compute areas.  

**Pseudocode/Algorithm:**  
1. **Initialize `maxArea = 0`** and an empty **stack**.  
2. **Loop through `0` to `n`:**  
   - If `arr[i] >= top of stack`, **push index `i`**.  
   - Else, while `stack is not empty and arr[stack.top] > arr[i]`:  
     - Pop `top` and calculate **area = height × width**.  
     - Update `maxArea`.  
   - Push `i` into stack.  
3. **Process remaining stack elements** after iteration.  
4. **Return `maxArea`**.  

**Code:**
```java
import java.util.Stack;

public class LargestRectangleHistogram {
    public static long largestRectangleArea(int[] arr) {
        int n = arr.length;
        Stack<Integer> stack = new Stack<>();
        long maxArea = 0;

        for (int i = 0; i < n; i++) {
            // While current bar is smaller than top of stack, calculate area
            while (!stack.isEmpty() && arr[stack.peek()] > arr[i]) {
                int height = arr[stack.pop()];
                int width = stack.isEmpty() ? i : i - stack.peek() - 1;
                long area = (long) height * width;
                maxArea = Math.max(maxArea, area);
            }
            stack.push(i);
        }

        // Process remaining bars in stack
        while (!stack.isEmpty()) {
            int height = arr[stack.pop()];
            int width = stack.isEmpty() ? n : n - stack.peek() - 1;
            long area = (long) height * width;
            maxArea = Math.max(maxArea, area);
        }

        return maxArea;
    }

    // Example usage
    public static void main(String[] args) {
        int[] histogram = {6, 2, 5, 4, 5, 1, 6};
        System.out.println("Largest Rectangle Area: " + largestRectangleArea(histogram)); // Output: 12
    }
}

```

- **Why use a stack?**  
  - It efficiently helps in finding the **nearest smaller height** to define **rectangular boundaries**.  
- **Edge Cases:**  
  - If all bars have **equal height**, the largest rectangle is `n × arr[0]`.  
  - If the bars are **strictly increasing**, the last bar defines the **maximum area**.  
  - If the bars are **strictly decreasing**, each bar defines **multiple smaller rectangles**.

**Example**  
**Input:**  
```

s = "a+b\*(c^d-e)^(f+g\*h)-i"

```

**Output:**  
```

"abcd^e-fgh\*+^\*+i-"

```

**Goal of the Code:**  
Convert an **infix expression** to a **postfix expression** while maintaining **operator precedence**.

**Approach:**  
1. **Use a Stack:**  
   - Operators are stored in a **stack** based on precedence.  
   - Operands are directly added to the result.  

2. **Process Characters Sequentially:**  
   - If it's an **operand (a-z, A-Z, 0-9)** → Add to the result.  
   - If it's an **opening bracket `(`** → Push to the stack.  
   - If it's a **closing bracket `)`** → Pop and append all operators until `(` is encountered.  
   - If it's an **operator (`+`, `-`, `*`, `/`, `^`)**:  
     - **Pop operators with higher or equal precedence** from the stack and append to the result.  
     - Push the current operator onto the stack.  

3. **Pop Remaining Operators:**  
   - After processing the entire string, pop all remaining operators in the stack.

**Pseudocode/Algorithm:**  
1. **Initialize** an empty **stack** and a **result string**.  
2. **Iterate** through the input string:  
   - If it's an **operand**, add it to `result`.  
   - If it's `'('`, push it to the stack.  
   - If it's `')'`, pop from the stack and add to `result` until `'('` is found.  
   - If it's an **operator**, pop from the stack **while it has higher/equal precedence**, then push the current operator.  
3. **Pop all remaining operators** from the stack and add to `result`.  
4. **Return the final `result` string**.

**🧮 Flowchart/ Diagram:**  
```sql

Start
↓
Initialize empty stack and result string
↓
For each character c in input:
├─ If c is operand → append to result
├─ If c == '(' → push to stack
├─ If c == ')' → pop and append until '(' found; pop '('
└─ If c is operator:
While stack not empty and precedence(top) ≥ precedence(c):
Pop and append top
Push c
↓
Pop and append all remaining operators from stack
↓
Return result
End

```

**Code:**
```java
class Solution {
    private static int prec(char a){
        if(a== '^'){
            return 3;
        }
        else if(a=='/' || a=='*'){
            return 2;
        }
        else if(a=='+' || a=='-'){
            return 1;
        }
        else{
            return -1;
        }
    }
    
    // Function to convert an infix expression to a postfix expression.
    public static String infixToPostfix(String s) {
        // Your code here
        Stack<Character> st=new Stack<>();
        StringBuilder result= new StringBuilder();
        
        for(int i=0; i<s.length(); i++){
            char c=s.charAt(i);
            if((c>='a' && c<='z') || (c>='A' && c<='Z') || (c>='0' && c<='9')){
                result.append(c);
            }
            else if(c=='('){
                st.push('(');
            }
            else if(c==')'){
                while(st.peek()!='('){
                    result.append(st.pop());
                }
                st.pop();
            }
            else{
                while(!st.isEmpty() && (prec(c) < prec(st.peek()) ||
                                      prec(c)==prec(st.peek()))){
                        result.append(st.pop());                
              }
              st.push(c);
            }
            
        }
        
        while(!st.isEmpty()){
            result.append(st.pop());
        }
        
        
        return result.toString();
    }
}
```

- **Space Complexity:** `O(N)` (For the stack in the worst case).

- This follows **Reverse Polish Notation (RPN)** principles.  
- Use a **stack-based approach** to handle precedence and order correctly.  
- The final output **does not contain parentheses** since postfix notation does not require them.

**Arithmetic Expression Evaluation (Infix to Result)**
You are given a **mathematical expression** in **infix notation** as a string. Your task is to evaluate the expression and return its result.

**Example:**  
Input: `(1+(2*3))`  
Output: `7`

**Input:**  
A string representing a fully parenthesized arithmetic expression.

**Output:**  
An integer representing the evaluated result of the expression.

**Constraints**
1. The expression **only** contains:
   - **Operators:** `+`, `-`, `*`, `/`
   - **Parentheses:** `(`, `)`
   - **Digits:** `0-9`
2. **Operator Precedence:**
   - `*` and `/` have **higher precedence** than `+` and `-`.
   - `/` is **integer division (floor division)**.
3. The expression is **fully parenthesized**, meaning:
   - It **always starts with** `(`.
   - It **always ends with** `)`.
4. The input expression **will always be valid**.
5. **No division by zero** will occur.
6. The **operands and final result fit in a 32-bit integer**.

**Goal:**  
Compute the integer result of the given infix expression using a stack-based evaluation.

**Approach: Using Stacks**
We use **two stacks**:
1. **Operands Stack (`operands`)**: Stores integer values.
2. **Operators Stack (`operators`)**: Stores `+`, `-`, `*`, `/`, `(`.

**Algorithm**
1. **Iterate through each character in the string**:
   - If it is a **digit**, extract the full number and push it to `operands`.
   - If it is an **opening parenthesis `(`**, push it to `operators`.
   - If it is a **closing parenthesis `)`**, evaluate the expression inside:
     - Keep popping from `operators` and applying them to `operands` until `(` is encountered.
     - Remove `(` from the stack.
   - If it is an **operator (`+`, `-`, `*`, `/`)**, check precedence:
     - If it has **lower or equal precedence** than the top of `operators`, apply the top operator.
     - Push the current operator to `operators`.

2. **Process remaining operators** after iteration:
   - Apply all remaining operations in `operators` to `operands`.

3. **Final result** is the last remaining element in `operands`.

**🧮 Flowchart/ Diagram:**  
- Start → Initialize two empty stacks  
- For each character in expression:  
  - Digit → parse full number → push to operands stack  
  - `(` → push to operators stack  
  - `)` → pop and apply operators until `(` found  
  - Operator → while top operator has higher or equal precedence, apply it → push current operator  
- After processing, apply remaining operators  
- Result = top of operands stack  
- End

**Code Implementation (Java)**
```java
import java.util.* ;
import java.io.*; 

public class Solution {
    // Helper function to apply an operator on two operands
    private static int applyOperation(char op, int b, int a) {
        switch (op) {
            case '+': return a + b;
            case '-': return a - b;
            case '*': return a * b;
            case '/': return a / b;  // Integer division
        }
        return 0;
    }

    // Helper function to define operator precedence
    private static int precedence(char op) {
        if (op == '+' || op == '-') return 1;
        if (op == '*' || op == '/') return 2;
        return 0;
    }

    public static int evaluateArithmeticExpression(String expression) {
        Stack<Integer> operands = new Stack<>();
        Stack<Character> operators = new Stack<>();
        int n = expression.length();

        for (int i = 0; i < n; i++) {
            char ch = expression.charAt(i);

            // If character is a digit, extract full number
            if (Character.isDigit(ch)) {
                int num = 0;
                while (i < n && Character.isDigit(expression.charAt(i))) {
                    num = num * 10 + (expression.charAt(i) - '0');
                    i++;
                }
                i--; // Move back to last digit
                operands.push(num);
            }
            // If opening parenthesis, push to operator stack
            else if (ch == '(') {
                operators.push(ch);
            }
            // If closing parenthesis, evaluate expression inside
            else if (ch == ')') {
                while (operators.peek() != '(') {
                    operands.push(applyOperation(operators.pop(), operands.pop(), operands.pop()));
                }
                operators.pop(); // Remove '(' from stack
            }
            // If an operator, apply precedence rules
            else if (ch == '+' || ch == '-' || ch == '*' || ch == '/') {
                while (!operators.isEmpty() && precedence(ch) <= precedence(operators.peek())) {
                    operands.push(applyOperation(operators.pop(), operands.pop(), operands.pop()));
                }
                operators.push(ch);
            }
        }

        // Process any remaining operations
        while (!operators.isEmpty()) {
            operands.push(applyOperation(operators.pop(), operands.pop(), operands.pop()));
        }

        return operands.pop();
    }
}
```

**Time and Space Complexity**
- **Time Complexity:** `O(N)` (each element is pushed/popped once).
- **Space Complexity:** `O(N)` (storing numbers and operators in stacks).

**Notes:**  
- Input is guaranteed valid and fully parenthesized.  
- Division by zero does not occur.  
- Supports multi-digit integers.  
- Use `applyOperation(op, b, a)` where `a` and `b` are popped operands, order matters.  
- Handles operator precedence correctly with stacks.

**Approach:**

   - **Base Case:** If the stack is empty, push `x` and return.
   - **Recursive Step:** 
     - Pop the **top element** and **recursively call** `insertAtBottom()` for the remaining stack.

**Algorithm:**
1. **If stack is empty**, push `x` and return.
2. Otherwise:

```java
// User function Template for Java
class Solution {
    public Stack<Integer> insertAtBottom(Stack<Integer> st, int x) {
        // Base Case: If stack is empty, push x
        if(st.isEmpty()){
            st.push(x);
            return st;
        }

        // Recursive Step: Pop top element
        int top = st.pop();

        // Recur for the rest of the stack
        st = insertAtBottom(st, x);

        // Push the top element back
        st.push(top);

        return st;
    }
}
```

**The Celebrity Problem**
A **celebrity** at a party is someone who:

**Explanation**
- Person `0` **knows** person `1`, but **not everyone knows** person `1`.
- Person `1` **knows** person `0`, but **a celebrity cannot know anyone**.
- ❌ **No celebrity exists**.

**Constraints**

**Example:**

**Input:**  
- Integer `N` (number of people)  
- `N x N` matrix `mat[][]` where `mat[i][j]` indicates if person i knows person j

**Output:**  
- Index of the celebrity if exists, else `-1`

**Goal:**  
Identify the celebrity who is known by everyone but knows no one.

**Approach: Stack-Based Solution**
**Algorithm**
1. **Push all people (0 to N-1) into a stack**.
2. **Find a potential celebrity**:
   - Pop two people `A` and `B` from the stack.
   - If `A` **knows** `B`, then `A` **cannot** be a celebrity → Push `B` back.
   - If `A` **does not know** `B`, then `B` **cannot** be a celebrity → Push `A` back.
   - Repeat until **only one person remains** in the stack.
3. **Verify if the remaining person is a celebrity**:
   - A celebrity **should not know** anyone (`mat[celeb][i] == 0` for all `i`).
   - A celebrity **should be known by** everyone (`mat[i][celeb] == 1` for all `i ≠ celeb`).
   - If any condition fails, return `-1`.
4. **Return the celebrity's index** if valid.

**🧮 Flowchart/ Diagram:**  
- Push all persons into stack → Pop two persons A, B →  
- If A knows B → discard A, keep B → else discard B, keep A →  
- Repeat until one person remains → Verify if that person is celebrity →  
- Return index or -1

**Pseudocode/Algorithm:**  
1. Push all people (0 to N-1) onto stack.  
2. While stack size > 1:  
   - Pop A and B  
   - If A knows B, push B back  
   - Else push A back  
3. Potential celebrity = top of stack  
4. For all i in [0, N-1], i != celeb:  
   - If celeb knows i OR i does not know celeb → return -1  
5. Return celeb


**Code Implementation (Java)**
```java
import java.util.*;

class Solution {
    public int celebrity(int mat[][]) {
        int n = mat.length;
        Stack<Integer> s = new Stack<>();

        // Step 1: Push all persons into the stack
        for (int i = 0; i < n; i++) {
            s.push(i);
        }

        // Step 2: Find a potential celebrity
        while (s.size() > 1) {
            int A = s.pop();
            int B = s.pop();

            // If A knows B, then A is not a celebrity, push B back
            if (mat[A][B] == 1) {
                s.push(B);
            }
            // If A does not know B, then B is not a celebrity, push A back
            else {
                s.push(A);
            }
        }

        // Step 3: Verify if the remaining person is a celebrity
        int celeb = s.pop();
        for (int i = 0; i < n; i++) {
            if (i == celeb) continue; // Skip self-check

            // A celebrity should not know anyone AND everyone should know them
            if (mat[celeb][i] != 0 || mat[i][celeb] != 1) {
                return -1;
            }
        }

        return celeb;
    }
}

```

   - Insert `x` at `arr[rear]`.

**Key Observations**
1. Unlike an **array-based queue**, a **linked list queue** does not have a fixed size.
2. **Maintain `front` and `rear` pointers**:
   - `front` → Points to the **first element**.
   - `rear` → Points to the **last element**.
3. **Push (`1 x`)**

   - Update the `rear` pointer.
4. **Pop (`2`)**
   - Remove and return the **front** element.

**🧮 Flowchart/ Diagram:**

**Code Implementation (Java)**
```java
class QueueNode {
    int data;
    QueueNode next;
    QueueNode(int a) {
        data = a;
        next = null;
    }
}

class MyQueue {
    QueueNode front, rear;
    
    // Function to push an element into the queue.
    void push(int a) {
        QueueNode newNode = new QueueNode(a);
        
        if (rear == null) {
            front = rear = newNode;
            return;
        }
        
        rear.next = newNode;
        rear = newNode;
    }
    
    // Function to pop front element from the queue.
    int pop() {
        if (front == null) {
            return -1; // Queue is empty
        }
        
        int result = front.data;
        front = front.next;
        
        if (front == null) {
            rear = null; // Queue is now empty
        }
        
        return result;
    }
}
```

**Using a Stack**
```java
import java.util.*;

class Solution {
    // Function to reverse the queue.
    public Queue<Integer> reverseQueue(Queue<Integer> queue) {
        Stack<Integer> stack = new Stack<>();
        
        // Push all elements to stack
        while (!queue.isEmpty()) {
            stack.push(queue.poll());
        }
        
        // Pop from stack and enqueue back to queue
        while (!stack.isEmpty()) {
            queue.add(stack.pop());
        }
        
        return queue;
    }
}

```

**Approach**
**1. Using Queue and Frequency Count (`O(N)`)**
1. Maintain a **queue** to keep track of characters in the order they appear.
2. Use a **frequency array** (`count[26]`) to track occurrences of each character.
3. **For each character**:
   - Increase its frequency.
   - Add to queue if it's the **first occurrence**.
   - Remove elements from the queue **until the front has frequency 1**.
   - Append **either the first non-repeating character or `#`** to the result.

**Pseudocode/Algorithm:**  
```sql
Initialize queue Q
Initialize frequency map freq\[26] = {0}
Initialize result string ans = ""

For each char c in stream:
freq\[c - 'a'] += 1
if freq\[c - 'a'] == 1:
Q.enqueue(c)

while Q is not empty and freq[Q.front - 'a'] > 1:
    Q.dequeue()

if Q is empty:
    ans += '#'
else:
    ans += Q.front

Return ans
```

**🧮 Flowchart/ Diagram:**  
```

\[Start]
↓
For each character in stream:

* Increment frequency
* Add to queue if first occurrence
* While queue front is repeated → remove
* Append front of queue or '#'
  ↓
  \[End]

```

**Java Implementation**
```java
import java.util.*;

class Solution {
    public String FirstNonRepeating(String s) {
        int[] count = new int[26];  // To track character frequencies
        Queue<Character> q = new LinkedList<>();  // Queue for processing
        StringBuilder ans = new StringBuilder();

        for (char ch : s.toCharArray()) {
            count[ch - 'a']++;  // Update frequency

            // If this is the first occurrence, add to queue
            if (count[ch - 'a'] == 1) {
                q.add(ch);
            }

            // Remove characters from queue if they are no longer unique
            while (!q.isEmpty() && count[q.peek() - 'a'] > 1) {
                q.poll();
            }

            // Append first non-repeating character or #
            ans.append(q.isEmpty() ? '#' : q.peek());
        }
        return ans.toString();
    }
}
```

1. **Push all elements** into a **Min-Heap** (Priority Queue).
2. **Extract the two smallest elements**, compute the cost, and insert the new rope back.
3. **Repeat** until one rope remains.
4. **Sum all costs** to get the minimum.

**🧮 Flowchart/ Diagram:**  
```sql

\[Start]
↓
Build Min-Heap from input array
↓
while heap size > 1:
extract two smallest
calculate sum
add cost to total
insert sum back to heap
↓
return totalCost

```

**Pseudocode/Algorithm:**  
```sql

function minCost(arr):
create minHeap and insert all elements
totalCost = 0

while size of minHeap > 1:
    first = extractMin(minHeap)
    second = extractMin(minHeap)
    cost = first + second
    totalCost += cost
    insert(minHeap, cost)

return totalCost

```
**Java Implementation**
```java
import java.util.*;

class Solution {
    // Function to return the minimum cost of connecting the ropes.
    public int minCost(int[] arr) {
        // Min-Heap to get smallest elements first
        PriorityQueue<Integer> pq = new PriorityQueue<>();
        for (int num : arr) {
            pq.add(num);
        }
        
        int totalCost = 0;
        
        while (pq.size() > 1) {
            int first = pq.poll();
            int second = pq.poll();
            
            int sum = first + second;
            totalCost += sum;
            
            pq.add(sum);
        }
        
        return totalCost;
    }
}
```

**Time Complexity**
- **`O(N log N)`**  
  - **Heap operations (`O(log N)`)** happen `N-1` times.
- **Efficient for large inputs!**

**Space Complexity**
- **`O(N)`** (For heap storage)

**Notes:**  
- Min-Heap ensures optimal greedy selection of the smallest ropes first.
- Frequently used in file merging and network design problems.

**Nearly Sorted Array**
Given an array `arr[]`, where each element is **at most `k` positions away** from its target position in a sorted order, sort the array **optimally**.

**Constraints**
- The array is **not completely unsorted** but has elements **displaced by at most `k` positions**.
- Sorting should be **in-place**.

**Explanation**
Each element is misplaced by **at most 3 positions**. We use a **Min-Heap** to efficiently sort.

**Example**  
**Input:**  
```
arr\[] = {6, 5, 3, 2, 8, 10, 9}, k = 3
```

**Output:**  
```
{2, 3, 5, 6, 8, 9, 10}
```

**Goal:**  
Efficiently sort the array using the information that elements are displaced by at most k positions.

**Approach**
**1. Min-Heap (`O(N log K)`)**
1. Use a **Min-Heap (Priority Queue)** to store **at most `k+1` elements**.
2. **Insert the first `k+1` elements** into the heap.
3. **Iterate over the rest of the array**:
   - Extract the **smallest element** and place it in the sorted order.
   - Insert the next element from the array into the heap.
4. **Empty the heap** for remaining elements.

**🧮 Flowchart/ Diagram:**  
```sql

\[Start]
↓
Insert first k+1 elements into Min-Heap
↓
For each remaining element:
→ Remove min from heap and place in array
→ Add current element to heap
↓
After array traversal:
→ Remove remaining elements from heap and place in array
↓
\[End]

```

**Pseudocode/Algorithm:**  
```sql

function nearlySorted(arr, k):
create a Min-Heap
for i = 0 to k:
heap.insert(arr\[i])
index = 0
for i = k+1 to end of arr:
arr\[index++] = heap.removeMin()
heap.insert(arr\[i])
while heap is not empty:
arr\[index++] = heap.removeMin()

```

**Java Implementation**
```java
import java.util.*;

class Solution {
    public void nearlySorted(int[] arr, int k) {
        int n = arr.length;
        PriorityQueue<Integer> pq = new PriorityQueue<>();

        // Add first k+1 elements into the min-heap
        for (int i = 0; i <= k && i < n; i++) {
            pq.add(arr[i]);
        }

        int index = 0;

        // Process remaining elements
        for (int i = k + 1; i < n; i++) {
            arr[index++] = pq.poll();
            pq.add(arr[i]);
        }

        // Empty the remaining elements from the heap
        while (!pq.isEmpty()) {
            arr[index++] = pq.poll();
        }
    }
}
```
**Time Complexity**
- **Insertion & Deletion in Min-Heap** → `O(log k)`
- **Processing `N` elements** → `O(N log k)`
- **Overall Complexity** → **`O(N log k)`** (much better than `O(N log N)` sorting)

**Space Complexity**
- **`O(k)`** (Heap stores at most `k+1` elements)

**Notes:**  
- Optimized for partially sorted data.
- Frequently used in streaming and real-time processing scenarios where sorting windows of data efficiently is needed.

To efficiently find the maximum sum of any contiguous subarray of size `k` using an optimized approach.

**Approach:**
1. Use **Sliding Window Technique**.
2. Calculate the sum of the first window of size `k`.
3. For each subsequent element, slide the window by:
   - Removing the element that is moving out of the window.
   - Adding the new element coming into the window.
4. Track the maximum sum encountered.

**🧮 Flowchart/ Diagram:**  
```sql

\[Start]
↓
Calculate sum of first k elements → current\_sum
↓
Initialize result = current\_sum
↓
For each i = k to n-1:
current\_sum = current\_sum + arr\[i] - arr\[i-k]
result = max(result, current\_sum)
↓
Return result
↓
\[End]

```

**Pseudocode/Algorithm:**
1. Initialize `result` and `current_sum` as the sum of the first window of size `k`.
2. Traverse the array from index `k` to `n`:
   - Update `current_sum` by adding `arr[i]` and removing `arr[i - k]`.
   - Update `result` if `current_sum` is greater.
3. Return `result`.

**Code:**
```java
// User function Template for Java

class Solution {
    public int maximumSumSubarray(int[] arr, int k) {
        // Code here
        if(arr.length<k){
            return -1;
        }
        
        int res=0;
        for(int i=0; i<k; i++){
            res+=arr[i];
        }
        
        int curr_sum=res;
        for(int i=k; i<arr.length; i++){
            curr_sum+=arr[i]-arr[i-k];
            res=Math.max(res,curr_sum);
        }
        
        return res;
    }
}
```

**Count Distinct Elements in Every Window**

**Input:**

- An integer `k` representing the size of the window

**Output:**
- A list of integers representing the count of distinct elements in each window of size `k`.

```
arr\[] = {1, 2, 1, 3, 4, 2, 3}, k = 4
```

**Output:**  
```
3 4 4 3
```

**Goal of the Code:**
To efficiently find and count the number of distinct elements within every window of size `k` using an optimized approach.

**Approach:**
1. Use a **Hash Map** to keep track of element frequencies within the current window.
2. Slide the window by one element at a time:
   - Add the next element to the map.
   - Remove the frequency of the element that is sliding out of the window.
   - Track the count of distinct elements by checking the size of the map.
3. Continue until the end of the array.

**🧮 Flowchart/ Diagram:**  
```sql

\[Start]
↓
Initialize freq map and result list
↓
Add first k elements to map
↓
Add map size to result
↓
For i = k to n-1:

* Increment freq of arr\[i]
* Decrement freq of arr\[i - k]
* Remove if freq becomes 0
* Add map size to result
  ↓
  Return result
  ↓
  \[End]

```

**Pseudocode/Algorithm:**
1. Initialize a Hash Map to keep track of element frequencies.
2. Traverse the first `k` elements and update the map.
3. For each subsequent element:
   - Add the new element to the map.
   - Reduce the count of the outgoing element in the map.
   - If the count of the outgoing element becomes zero, remove it from the map.
   - Store the size of the map as the result for the current window.
4. Return the results.

**Code:**
```java
class Solution {
    ArrayList<Integer> countDistinct(int arr[], int k) {
        // code here
        int n=arr.length;
        ArrayList<Integer> res=new ArrayList<>();
        Map<Integer, Integer> freq=new HashMap<>();
        
        for(int i=0; i<k; i++){
            freq.put(arr[i],freq.getOrDefault(arr[i],0)+1);
        }
        
        res.add(freq.size());
        
        for(int i=k; i<n; i++){
            freq.put(arr[i],freq.getOrDefault(arr[i],0)+1);
            freq.put(arr[i-k], freq.get(arr[i-k])-1);
            
            if(freq.get(arr[i-k])==0){
                freq.remove(arr[i-k]);
            }
            res.add(freq.size());
        }
        
        return res;
    }
}
```

- Use a sliding window technique to ensure we only make changes for elements entering or leaving the window.

- HashMap ensures constant-time updates.
- Efficient for large arrays compared to nested loops.

1. Use a **sliding window technique** along with a **frequency array** to track character occurrences.
2. Process the first `K-1` characters and store their frequencies.
3. For each subsequent character:
   - Add the current character to the frequency map.
   - Count distinct characters.
   - If the distinct count is exactly `K-1`, increment the result.
   - Remove the first character of the previous window from the frequency map.
4. Return the final count of valid substrings.

**🧮 Flowchart/ Diagram:**  
- Initialize frequency array of size 26.  
- Process first `K-1` characters to populate frequency.  
- Slide window from `K-1` to end:  
  - Add current character frequency.  
  - Count distinct characters in frequency array.  
  - If distinct count equals `K-1`, increment result.  
  - Remove frequency of character leaving the window.

2. For `i` in `0` to `K-2`: increment frequency of `S[i]`.

   - Increment frequency of `S[i]`.

   - If distinct == `K-1`, increment answer.  
   - Decrement frequency of `S[i-K+1]`.

**Code:**
```java
//User function Template for Java

class Solution {
    static int countOfSubstrings(String S, int K) {
        // code here
        
        int n=S.length();
        int[] count=new int[26];
        int ans=0;
        
        if(K>n) return 0;
        
        for(int i=0; i<K-1; i++){
            count[S.charAt(i)-'a']++;
        }
        
        for(int i=K-1; i<n; i++){
            count[S.charAt(i)-'a']++;
            
            int distinctCnt=0;
            for(int j1=0; j1<26; j1++){
                if(count[j1]>0) distinctCnt++;
            }
            
            if(distinctCnt==K-1) ans++;
            
            count[S.charAt(i-K+1)-'a']--;
        }
        
        return ans;
        
    }
};
```

```java
import java.util.*;

class Solution {
    public ArrayList<Integer> maxOfMins(int[] arr) {
        int n = arr.length;
        
        ArrayList<Integer> res = new ArrayList<>(Collections.nCopies(n, 0));
        Stack<Integer> s = new Stack<>();
        ArrayList<Integer> lenArr = new ArrayList<>(Collections.nCopies(n, 0));
        
        for (int i = 0; i < n; i++) {
            while (!s.isEmpty() && arr[s.peek()] >= arr[i]) {
                int top = s.pop();
                int windowSize = s.isEmpty() ? i : i - s.peek() - 1;
                lenArr.set(top, windowSize);
            }
            s.push(i);
        }
        
        while (!s.isEmpty()) {
            int top = s.pop();
            int windowSize = s.isEmpty() ? n : n - s.peek() - 1;
            lenArr.set(top, windowSize);
        }
        
        for (int i = 0; i < n; i++) {
            int windowSize = lenArr.get(i) - 1;
            res.set(windowSize, Math.max(res.get(windowSize), arr[i]));
        }
        
        for (int i = n - 2; i >= 0; i--) {
            res.set(i, Math.max(res.get(i), res.get(i + 1)));
        }
        
        return res;
    }
}
```

**Time and Space Complexity**
- **Time Complexity:** `O(N)`, since each element is pushed and popped from the stack at most once.  
- **Space Complexity:** `O(N)`, for storing window sizes and results.  

**Notes:**

* This problem is best solved using the concept of **Next Smaller Element (NSE)** and **Previous Smaller Element (PSE)**.
* For each element, we find the largest window in which it is the minimum, and map it accordingly.
* Post-processing ensures that if a window of size `k` has no entry, it inherits from larger windows (since `max of min` is non-increasing).

**Smallest Distinct Window**  

Given a string **s**, find the length of the **smallest window** that contains **all the characters** of the given string **at least once**.  

**Input:**

**Output:**

**Constraints:**  
- `1 ≤ s.length ≤ 10^5`  

**Example**

**Input:**
`s = "AABBC"`

**Output:**
`3`

**Explanation:**
The distinct characters in `"AABBC"` are `'A'`, `'B'`, and `'C'`. The substring `"ABC"` is the smallest window containing all these distinct characters.


**Goal of the Code:**  
To efficiently determine the **smallest substring length** that contains **all unique characters** of `s` at least once, using the **Sliding Window (Two Pointers) Technique**.  

**Approach**  
The problem can be efficiently solved using a **sliding window approach** with **two pointers**:  

1. **Key Idea:**  
   - The **number of distinct characters** in `s` is fixed (`distinct`).  
   - Use a **frequency map** (`count[]`) to track occurrences of characters in the current window.  
   - Expand the window (`end++`) to include all distinct characters.  
   - Once all distinct characters are in the window (`matched == distinct`), try to shrink it (`start++`) while maintaining all distinct characters.  
   - Update the **minimum window length** found.  

2. **Steps to Solve:**  
   - Compute `distinct`, the number of unique characters in `s`.  
   - Use a frequency array `count[256]` to track character counts.  
   - Expand the window (`end++`) while maintaining a count of **matched** distinct characters.  
   - Once all distinct characters are in the window, try **shrinking** it from the left (`start++`).  
   - Keep track of the **smallest window size** found.  


**Flowchart**

```mermaid
flowchart TD
    A[Start] --> B[Count total distinct characters in the string]
    B --> C[Initialize count array, start, end, matched, minLen]
    C --> D[Increment count of s[end] and move end forward]
    D --> E{Have all distinct chars been matched?}
    E -- No --> D
    E -- Yes --> F[Update minLen if current window is smaller]
    F --> G[Decrease count of s[start] and move start forward]
    G --> H{Still all matched?}
    H -- Yes --> F
    H -- No --> D
    D --> I[Repeat until end of string]
    I --> J[Return minLen]
```

**Pseudocode/Algorithm**
1. Compute `distinct = count of unique characters in s`.  
2. Initialize `start = 0`, `minLen = Integer.MAX_VALUE`, and `matched = 0`.  
3. Use `count[256]` to track occurrences of characters in the current window.  
4. Expand the window (`end++`) until all distinct characters are included.  
5. Once `matched == distinct`:  
   - Try **shrinking** the window from `start++`.  
   - Update `minLen` if the current window is smaller.  
   - If removing `s[start]` reduces `matched`, stop shrinking.  
6. Return `minLen` as the smallest distinct window length.  

**Java Solution**
```java
import java.io.*;

class Solution {
    public int findSubString(String s) {
        int n = s.length();
        if (n <= 1) return n;

        int[] count = new int[256];
        int distinct = (int) s.chars().distinct().count();
        int start = 0, startIdx = -1, matched = 0, minLen = Integer.MAX_VALUE;

        for (int end = 0; end < n; end++) {
            if (++count[s.charAt(end)] == 1) matched++;

            while (matched == distinct) {
                if (end - start + 1 < minLen) {
                    minLen = end - start + 1;
                    startIdx = start;
                }
                if (--count[s.charAt(start++)] == 0) matched--;
            }
        }
        return startIdx == -1 ? -1 : minLen;
    }
}
```

**Time and Space Complexity**
- **Time Complexity:** `O(N)`, since each character is processed at most twice (once for expansion, once for contraction).  
- **Space Complexity:** `O(1)`, since we use a **fixed** frequency array of size `256` (constant space).  

**Additional Notes/Hints:**

* This problem is a classic variant of the **minimum window substring** type problem, with a simplified condition: we just need to include **all unique characters**, not match a target pattern.
* Using an array of size `256` (ASCII range) ensures **constant space** for character frequency tracking.
* Carefully maintaining the `matched` count avoids repeatedly checking the whole count array, which helps keep the time complexity linear.
* The **shrinking phase** of the sliding window is crucial to minimizing the window size efficiently.
* If all characters in the string are already unique, the result is the length of the string itself.

   - Convert `arr[i]` from `0` to `-1`.  
   - Add `arr[i]` to `preSum`.  
   - If `preSum == 0`, update `res = i + 1`.  
   - If `preSum` exists in the HashMap, compute the subarray length and update `res`.  
   - Otherwise, store `preSum` in the HashMap.  
4. Return `res` as the final result.  

**Code:**
```java
class Solution {
    public int maxLen(int[] arr) {
        // Your code here
        
        HashMap<Integer, Integer> mp=new HashMap<>();
        int preSum=0;
        int res=0;
        
        for(int i=0; i<arr.length; i++){
            preSum+=(arr[i]==0)?-1: 1;
            
            if(preSum==0){
                res=i+1;
            }
            
            if(mp.containsKey(preSum)){
                res=Math.max(res,i-mp.get(preSum));
            }
            else{
                mp.put(preSum,i);
            }
            
            
        }
        
        return res;
    }
}
```

**Time and Space Complexity:**

   - Adjust the frequency array as we traverse `txt` while maintaining a **fixed window size** equal to `pat.length()`.

2. **Steps:**

```sql
Start
  |
  v
If pat.length > txt.length: return 0
  |
  v
Initialize frequency array cnt[26] for characters in pat
  |
  v
For i in 0 to pat.length - 2:
  Decrement cnt[txt[i]]
  |
  v
Initialize ans = 0
For i in pat.length - 1 to txt.length - 1:
  Decrement cnt[txt[i]]
  If allZero(cnt):
    ans++
  Increment cnt[txt[i - pat.length + 1]]
  |
  v
Return ans
  |
  v
End
```

**Pseudocode/Algorithm:**  
1. If `pat.length > txt.length`, return `0`.  
2. Initialize a frequency array `cnt[26]` for `pat`.  
3. Iterate over the first `pat.length() - 1` characters of `txt` and update `cnt[]`.  
4. Start a sliding window from `pat.length() - 1` to `txt.length()`:  
   - Update `cnt[]` for the new character.  
   - If `cnt[]` contains only zeroes, increment `ans`.  
   - Restore `cnt[]` by adding back the character that slides out.  
5. Return `ans` as the total count of anagrams found.  

**Code:**
```java
// User function Template for Java

class Solution {

    int search(String pat, String txt) {
        // code here
        
        if(pat.length()>txt.length())  return 0;
        
        int[] cnt=new int[26];
        for(char ch: pat.toCharArray()) cnt[ch-'a']++;
        for(int i=0; i<pat.length()-1; i++)  cnt[txt.charAt(i)-'a']--;
        
        int ans=0;
        for(int i=pat.length()-1; i<txt.length(); i++){
            cnt[txt.charAt(i)-'a']--;
            if(allZero(cnt)) ans++;
            cnt[txt.charAt(i-pat.length()+1)-'a']++;
        }
        return ans;
    }
    
    static boolean allZero(int[] cnt){
        for(int val: cnt) if (val!=0) return false;
        return true;
    }
}
```

   - Update `maxFreq` to store the most frequent character in the window.

**Code:**
```java
// User function Template for Java

class Solution {
    static int characterReplacement(String s, int k) {
        // code here
        int n=s.length();
        HashMap<Character, Integer> freq =new HashMap<>();
        int maxFreq=0, res=0,l=0;
        for(int r=0; r<n; r++){
            freq.put(s.charAt(r),freq.getOrDefault(s.charAt(r),0)+1);
            
            maxFreq=Math.max(maxFreq,freq.get(s.charAt(r)));
            
            if(r-l+1-maxFreq >k){
                freq.put(s.charAt(l),freq.get(s.charAt(l))-1);
                l++;
            }
            res=Math.max(res,r-l+1);
        }
        return res;
    }
}
```

3. Fill the DP table using the recurrence relation.
4. Scan the last column `dp[i][m]` to find the shortest window.
5. Return substring starting from `dp[i][m] - 1` with minimum length.

   - If characters match: `dp[i][j] = dp[i-1][j-1]`

5. Return substring from `dp[i][m]-1` to `i`.  

**Java Solution**
```java
class Solution {
    static String minWindow(String s1, String s2) {
        int n = s1.length(), m = s2.length();
        if (n < m) return "";

        int[][] dp = new int[n + 1][m + 1];

        for (int i = 0; i <= n; i++) dp[i][0] = i + 1;
        for (int j = 1; j <= m; j++) dp[0][j] = 0;

        for (int i = 1; i <= n; i++) {
            for (int j = 1; j <= m; j++) {
                if (s1.charAt(i - 1) == s2.charAt(j - 1)) {
                    dp[i][j] = dp[i - 1][j - 1];
                } else {
                    dp[i][j] = dp[i - 1][j];
                }
            }
        }

        int startIdx = -1, minLen = Integer.MAX_VALUE;

        for (int i = 1; i <= n; i++) {
            if (dp[i][m] > 0) {
                int start = dp[i][m] - 1;
                int len = i - start;
                if (len < minLen) {
                    minLen = len;
                    startIdx = start;
                }
            }
        }

        return startIdx == -1 ? "" : s1.substring(startIdx, startIdx + minLen);
    }
}
```

**Time and Space Complexity**
- **Time Complexity:** `O(n * m)` where `n` and `m` are lengths of `s1` and `s2` respectively.
- **Space Complexity:** `O(n * m)` for the DP table.

**Notes:**  
- The problem differs from typical substring search since `s2` must appear as a subsequence, not necessarily contiguous.  
- Dynamic Programming helps track the earliest start index for subsequences ending at each position efficiently.  
- If multiple minimum windows exist, returning the earliest one requires scanning all possibilities after DP computation.  
- The solution uses `O(n*m)` time and space, which is feasible given the constraints (`n, m <= 1000`).  
- Edge cases include when `s2` is longer than `s1` or when `s2` is empty (should return empty or entire string accordingly).  
- Optimization is possible using two pointers for specific cases, but DP ensures correctness for all inputs.

**Approach**  
The problem can be efficiently solved using a **sliding window** approach:  

1. **Key Idea:**  
   - The order of characters does **not** matter for an anagram; only the **frequency of characters** must match.  
   - Use **two frequency arrays** to store character counts:
     - `pCount[]` → Stores frequency of each character in `p`.  
     - `sCount[]` → Stores frequency of the current window in `s`.  
   - Compare the **two frequency arrays** after processing each window of size `p.length()`.  

2. **Steps to Solve:**  
   - Initialize frequency arrays `pCount[]` and `sCount[]` for `p` and `s`.  
   - Use a **sliding window** of size `p.length()` to track character counts.  
   - When the window size exceeds `p.length()`, remove the leftmost character from `sCount[]`.  
   - If `sCount[]` matches `pCount[]`, store the starting index.  
   - Continue sliding the window across `s`.  

**🧮 Flowchart/ Diagram:**  
1. Initialize frequency arrays for `p` and sliding window in `s`.  
2. Slide window over `s`, update character counts dynamically.  
3. After each slide, compare frequency arrays.  
4. If equal, record start index.  
5. Continue until end of `s`.  

**Pseudocode/Algorithm**
1. Compute the **character frequency** of `p` in `pCount[]`.  
2. Iterate through `s` using a **sliding window** of size `p.length()`:  
   - Add the new character to `sCount[]`.  
   - If the window size exceeds `p.length()`, remove the leftmost character.  
   - If `sCount[]` matches `pCount[]`, store the start index.  
3. Return the list of **starting indices**.  

**Java Solution**
```java
import java.util.*;

class Solution {
    public List<Integer> findAnagrams(String s, String p) {
        List<Integer> result = new ArrayList<>();
        int sLen = s.length(), pLen = p.length();
        if (sLen < pLen) return result;
        
        int[] pCount = new int[26];
        int[] sCount = new int[26];

        // Store frequency of characters in p
        for (char c : p.toCharArray()) {
            pCount[c - 'a']++;
        }

        for (int i = 0; i < sLen; i++) {
            sCount[s.charAt(i) - 'a']++;  // Add new character to window

            if (i >= pLen) {
                sCount[s.charAt(i - pLen) - 'a']--; // Remove leftmost character
            }

            if (Arrays.equals(pCount, sCount)) {
                result.add(i - pLen + 1); // Store starting index
            }
        }
        return result;
    }
}
```

**Notes:**

- `Arrays.equals()` efficiently compares the frequency arrays in O(26) = O(1) time per step.  
- Sliding window of fixed size ensures linear time complexity.  
- Edge cases include when `p` is longer than `s` (return empty list).

The longest substring with exactly 3 unique characters is "cbebebe", length 7.

**Goal of the Code:**  
To efficiently determine the **longest substring** that contains **exactly k distinct characters** using the **Sliding Window (Two Pointers) Technique**.  

**Approach**  
The problem can be efficiently solved using a **sliding window approach** with **two pointers**:  

1. **Key Idea:**  
   - Maintain a **window** `[start, end]` with at most `k` unique characters.  
   - Use a **frequency array** (`count[26]`) to track character occurrences.  
   - Expand the window (`end++`) to include more characters.  
   - If the number of unique characters **exceeds k**, shrink the window (`start++`) until only `k` unique characters remain.  
   - Update the **maximum window size** found whenever `uniqueCount == k`.  

2. **Steps to Solve:**  
   - Initialize `start = 0`, `maxLen = -1`, and `uniqueCount = 0`.  
   - Expand the window (`end++`), updating `count[26]` and `uniqueCount`.  
   - If `uniqueCount > k`, shrink the window (`start++`) until `uniqueCount == k`.  
   - Track the **maximum window size** found.  

**🧮 Flowchart/ Diagram:**  
1. Initialize `start = 0`, `maxLen = -1`, `uniqueCount = 0`, and frequency array `count[26]`.  
2. Expand `end` pointer to include new characters and update counts and `uniqueCount`.  
3. If `uniqueCount > k`, shrink window from `start` until `uniqueCount == k`.  
4. Update `maxLen` whenever `uniqueCount == k`.  
5. Continue until `end` reaches the end of the string.  

**Pseudocode/Algorithm**
1. Initialize `start = 0`, `maxLen = -1`, and `count[26] = {0}`.  
2. Expand the window (`end++`) while updating `count[]`.  
3. If adding a character increases `uniqueCount` beyond `k`, shrink the window (`start++`).  
4. If `uniqueCount == k`, update `maxLen = max(maxLen, end - start + 1)`.  
5. Return `maxLen`.  

**Java Solution**
```java
class Solution {
    public int longestkSubstr(String s, int k) {
        int n = s.length(), start = 0, maxLen = -1, count[] = new int[26];
        int uniqueCount = 0; // Track distinct characters

        for (int end = 0; end < n; end++) {
            if (count[s.charAt(end) - 'a']++ == 0) uniqueCount++; // New distinct char

            while (uniqueCount > k) { // Shrink window if unique chars exceed k
                if (--count[s.charAt(start) - 'a'] == 0) uniqueCount--; 
                start++;
            }

            if (uniqueCount == k) maxLen = Math.max(maxLen, end - start + 1);
        }
        return maxLen;
    }
}
```

**Goal of the Code:**

**🧠 Approach**
This is a **sliding window with hash map frequency counting** problem.

**Step-by-Step Strategy:**
1. **Understand total word length:**
   - If each word is of length `w` and there are `n` words, the window size = `w * n`.

2. **Use a HashMap** to store the **expected count of each word**.

3. **Slide a window** of total size `w * n` across the string `S`.

4. In each window:
   - Break the substring into chunks of size `w`.
   - Count frequencies of each word in the current window.
   - Compare with the expected frequency.
   - If all match, store the start index.

**🧮 Flowchart/ Diagram:**  
1. Calculate total length of concatenation (`wordLen * wordCount`).  
2. Slide window from 0 to `S.length - totalLen`.  
3. For each window, split substring into chunks of size `wordLen`.  
4. Count frequency of words in window and compare with expected frequencies.  
5. If match, record start index.  
6. Continue until end of string.

**📋 Pseudocode**
1. Create a HashMap to store frequency of each word in list L.
2. Loop through S from 0 to length(S) - totalConcatLength: a. Create a temporary map. b. Extract substrings of word length. c. Compare counts with the original map. d. If all match, store current index.
3. Return all matching indices.

**🧑‍💻 Java Solution**
```java
import java.util.ArrayList;
import java.util.Arrays;
import java.util.HashMap;
import java.util.List;

class GFG {
    public static ArrayList<Integer> findSubstring(String S, final List<String> L) {
        int wordLen = L.get(0).length();
        int wordCount = L.size();
        int totalLen = wordLen * wordCount;
        int n = S.length();

        ArrayList<Integer> result = new ArrayList<>();
        if (totalLen > n) return result;

        HashMap<String, Integer> wordFreq = new HashMap<>();
        for (String word : L) {
            wordFreq.put(word, wordFreq.getOrDefault(word, 0) + 1);
        }

        for (int i = 0; i <= n - totalLen; i++) {
            HashMap<String, Integer> seen = new HashMap<>(wordFreq);
            int j = i, count = wordCount;

            while (j < i + totalLen) {
                String sub = S.substring(j, j + wordLen);
                if (!seen.containsKey(sub) || seen.get(sub) == 0) break;
                seen.put(sub, seen.get(sub) - 1);
                count--;
                j += wordLen;
            }

            if (count == 0) result.add(i);
        }

        return result;
    }

    public static void main(String[] args) {
        String S = "barfoothefoobarman";
        ArrayList<String> L = new ArrayList<>(Arrays.asList("foo", "bar"));
        ArrayList<Integer> indices = findSubstring(S, L);
        for (Integer i : indices) {
            System.out.println(i);
        }
    }
}
```

**Output:**

**Goal:**  
Efficiently count subarrays with exactly `k` distinct integers using sliding window technique and mathematical subtraction.

**Approach:**  
Use helper function `atMost(nums, k)` that returns count of subarrays with at most `k` distinct integers.  
Calculate result as:  
`atMost(nums, k) - atMost(nums, k-1)`.

To solve this problem efficiently, we use the **sliding window technique** combined with a **mathematical trick**.

**Key Insight:**
The number of subarrays with **exactly `k`** distinct integers =  
**Subarrays with at most `k` distinct integers** - **Subarrays with at most `k-1` distinct integers**

**Steps:**

1. **Implement a helper function `atMost(nums, k)`**:
   - Counts how many subarrays have at most `k` distinct integers.
   - Uses a HashMap to keep track of frequency of elements in the current window.
   - Expands the window to the right (`right` pointer) and shrinks from the left (`left` pointer) when the unique count exceeds `k`.

2. **Main function**:
   - Result = `atMost(nums, k)` - `atMost(nums, k-1)`

**🧮 Flowchart/ Diagram:**  
- Initialize left pointer at 0, right pointer moves forward.  
- Maintain frequency map of elements in current window.  
- Expand window with right pointer, increment frequency.  
- While unique elements > k, shrink window from left.  
- Accumulate count of valid subarrays ending at right.  
- Repeat until right reaches end of array.

**Pseudocode/Algorithm:**  
```sql

function atMost(nums, k):
if k == 0: return 0
freq = empty map
left = 0, count = 0
for right in 0 to len(nums)-1:
freq\[nums\[right]] += 1
while size(freq) > k:
freq\[nums\[left]] -= 1
if freq\[nums\[left]] == 0:
remove nums\[left] from freq
left += 1
count += right - left + 1
return count

function subarraysWithKDistinct(nums, k):
return atMost(nums, k) - atMost(nums, k-1)

```

**Java Code:**

```java
class Solution {
    // Helper method to count subarrays with at most k distinct integers
    private int atMost(int[] nums, int k) {
        if (k == 0) return 0;

        HashMap<Integer, Integer> hm = new HashMap<>();
        int left = 0, count = 0;

        for (int right = 0; right < nums.length; right++) {
            hm.put(nums[right], hm.getOrDefault(nums[right], 0) + 1);

            // Shrink the window if we have more than k distinct numbers
            while (hm.size() > k) {
                hm.put(nums[left], hm.get(nums[left]) - 1);
                if (hm.get(nums[left]) == 0) {
                    hm.remove(nums[left]);
                }
                left++;
            }

            // All subarrays ending at 'right' with at most k distinct numbers
            count += right - left + 1;
        }

        return count;
    }

    public int subarraysWithKDistinct(int[] nums, int k) {
        // Subarrays with exactly k = atMost(k) - atMost(k-1)
        return atMost(nums, k) - atMost(nums, k - 1);
    }
}
```

- **Time Complexity**: O(N), where N is the length of the array `nums`.  
  Each element is added and removed from the HashMap at most once.
  
- **Space Complexity**: O(K), for storing at most `k` distinct elements in the HashMap.

**Notes:**  
- The subtraction trick avoids explicitly counting exactly k distinct subarrays.  
- Sliding window ensures each element is added and removed at most once.  
- Works for arbitrary integer arrays, not just small alphabets.

**Design a Least Recently Used (LRU) Cache**  

Implement a data structure that simulates the behavior of an LRU (Least Recently Used) Cache. The cache has a fixed capacity, and supports the following operations:

1. **GET x**: Return the value associated with key `x` if present, otherwise return `-1`.
2. **PUT x y**: Insert or update the value for key `x` to `y`. If the cache exceeds its capacity, evict the least recently used item.

**Input:**  
- An integer `cap` denoting the **capacity** of the cache.  
- A sequence of `Q` queries, where each query is either `GET x` or `PUT x y`.

**Output:**  
- For every `GET` query, output the result of the operation (the value or `-1`).

**Constraints:**  
- `1 ≤ cap ≤ 10^4`  
- `1 ≤ Q ≤ 10^5`  
- `0 ≤ x, y ≤ 10^5`

**Example**  
Input:  
```
cap = 2
Queries:
PUT 1 10
PUT 2 20
GET 1    -> returns 10
PUT 3 30 (evicts key 2)
GET 2    -> returns -1
GET 3    -> returns 30
```
Output:  
```
10
-1
30
```

**Goal of the Code:**  
Efficiently implement the LRU cache that handles large volumes of queries in near constant time using appropriate data structures.

**Approach**
We use a combination of a **HashMap** and a **Doubly Linked List** (using Java's `LinkedList`) to implement the LRU cache efficiently:

**Key Concepts:**
- **HashMap** stores the key-value pairs for O(1) access.
- **LinkedList** tracks the usage order of keys with the most recently used at the front.

**How Operations Work:**
- **GET key**:
  - If key exists, move it to the front of the LRU list and return its value.
  - If key doesn't exist, return -1.
- **PUT key value**:
  - If key already exists, update its value and move it to the front.
  - If key doesn't exist:
    - If capacity is full, remove the last item from the list and map.
    - Insert the new key-value and add to the front of the list.

**🧮 Flowchart/Diagram:**  
1. Start  
2. For each query:  
   - If query is GET(key):  
     - Check if key exists in cacheMap  
       - If yes: Move key to front of lruList and return value  
       - If no: Return -1  
   - If query is PUT(key, value):  
     - If key exists: Update value, move key to front  
     - Else:  
       - If cache is full: Remove last element from lruList and cacheMap  
       - Add new key and value to cacheMap and front of lruList  
3. Repeat until all queries processed  
4. End  

**Pseudocode/Algorithm:**  
```sql

Initialize cacheMap as empty hashmap
Initialize lruList as empty doubly linked list
Set capacity = cap

Function GET(key):
If key not in cacheMap:
Return -1
Else:
Remove key from lruList
Add key to front of lruList
Return cacheMap\[key]

Function PUT(key, value):
If key in cacheMap:
Update cacheMap\[key] = value
Remove key from lruList
Add key to front of lruList
Else:
If cacheMap size == capacity:
Remove lastKey from lruList
Remove lastKey from cacheMap
Add key to front of lruList
Add key-value to cacheMap

```

**Java Solution**
```java
class LRUCache {
    private int cap;
    private Map<Integer, Integer> cacheMap;
    private LinkedList<Integer> lruList;

    LRUCache(int cap) {
        this.cap = cap;
        this.cacheMap = new HashMap<>();
        this.lruList = new LinkedList<>();
    }

    public int get(int key) {
        if (!cacheMap.containsKey(key)) {
            return -1;
        }

        lruList.remove(Integer.valueOf(key));
        lruList.addFirst(key);

        return cacheMap.get(key);
    }

    public void put(int key, int value) {
        if (cacheMap.containsKey(key)) {
            cacheMap.put(key, value);
            lruList.remove(Integer.valueOf(key));
        } else {
            if (cacheMap.size() >= cap) {
                int lruKey = lruList.removeLast();
                cacheMap.remove(lruKey);
            }
            cacheMap.put(key, value);
        }
        lruList.addFirst(key);
    }
}
```

**Time and Space Complexity**
- **Time Complexity:** O(1) per operation (on average).
- **Space Complexity:** O(cap), where cap is the cache capacity.

**Notes:**  
- Using a doubly linked list and hashmap together allows O(1) time complexity for both `get` and `put`.  
- Java’s `LinkedList` remove operation with `Integer.valueOf(key)` runs in O(N) worst-case, so for very large capacity, consider implementing a custom doubly linked list with node references for O(1) removals.  
- The solution assumes keys and values are integers within given constraints.  
- Eviction policy strictly removes the least recently used key when capacity is exceeded.

- Input: `"abcabcbb"`  
- Output: `3`  
- Explanation: The longest substring without repeating characters is `"abc"` with length 3.

**Goal of the Code:**  
To efficiently determine the **maximum length** of a contiguous substring in `s` where all characters are **distinct**, using the **Sliding Window (Two Pointers) Technique**.  

**Approach**  
The problem can be efficiently solved using a **sliding window** approach:  

1. **Key Idea:**  
   - Maintain a **boolean array** (`vis[]`) to track characters present in the window.  
   - Expand the right pointer (`right++`) while keeping track of distinct characters.  
   - If a duplicate character is found, shrink the window by moving `left++` until all characters are distinct again.  
   - Update the maximum length (`res`).  

2. **Steps to Solve:**  
   - Initialize `left = 0`, `right = 0`, and a **boolean array** `vis[26]` to mark visited characters.  
   - Expand the right pointer (`right++`) and mark the character as visited.  
   - If a duplicate character is found, move the left pointer (`left++`) until uniqueness is restored.  
   - Update the **maximum length** (`res`).  
   - Continue until `right` reaches the end of the string.  

**Pseudocode/Algorithm**
1. Initialize `left = 0`, `right = 0`, and `res = 0`.  
2. Use a **boolean array** `vis[26]` to mark characters present in the current window.  
3. Iterate over `s` with `right++`:  
   - If `s[right]` is already in `vis[]`, shrink the window by moving `left++`.  
   - Mark `s[right]` as visited in `vis[]`.  
   - Update `res = max(res, right - left + 1)`.  
4. Return `res` as the maximum substring length.  

**🧮 Flowchart/Diagram:**  
1. Start  
2. Initialize pointers `left = 0`, `right = 0`, and `res = 0`  
3. Create boolean array `vis[26]` initialized to false  
4. While `right < s.length`:  
   - If `vis[s[right]]` is true (duplicate found):  
     - Move `left` forward, mark characters as not visited until duplicate removed  
   - Mark `vis[s[right]]` as true  
   - Update `res = max(res, right - left + 1)`  
   - Increment `right`  
5. End loop  
6. Return `res`  

**Java Solution**
```java
import java.io.*;

class Solution {
    public int longestUniqueSubstr(String s) {
        if (s.length() == 0 || s.length() == 1) {
            return s.length();
        }

        int res = 0, right = 0, left = 0;
        boolean[] vis = new boolean[26];

        while (right < s.length()) {
            while (vis[s.charAt(right) - 'a']) {
                vis[s.charAt(left) - 'a'] = false;
                left++;
            }

            vis[s.charAt(right) - 'a'] = true;
            res = Math.max(res, right - left + 1);
            right++;
        }
        return res;
    }
}
```

**Time and Space Complexity**
- **Time Complexity:** `O(N)`, since each character is processed at most twice (once for expansion, once for contraction).  
- **Space Complexity:** `O(1)`, since we use a **fixed** boolean array of size `26` (constant space).  

**Notes:**  
- This problem is a classic example of the sliding window technique.  
- The boolean array helps in O(1) checking for duplicates, making the approach efficient.  
- The window dynamically adjusts size by moving the left pointer whenever a duplicate is found, ensuring all characters inside the window are unique.  
- This method works efficiently due to the fixed character set (lowercase letters), allowing constant space usage.  

## Problem Statement: 183  
**Subarrays with K Distinct Integers**  

Given an array `arr[]` of **positive integers** and an integer `k`, find the number of **subarrays** in `arr[]` where the **count of distinct integers** is **exactly k**.  

**Input:**  
- An array `arr[]` containing **positive integers**.  
- An integer `k`, representing the **exact** number of distinct integers in a subarray.  

**Output:**  
- An integer representing the number of **subarrays** with **exactly k distinct integers**.  

**Constraints:**  
- `1 <= arr.length <= 10^5`  
- `1 <= arr[i] <= 10^6`  
- `1 <= k <= arr.length`  

**Example:**  
- Input: `arr = [1, 2, 1, 2, 3]`, `k = 2`  
- Output: `7`  
- Explanation: The subarrays with exactly 2 distinct integers are:  
  `[1, 2]`, `[2, 1]`, `[1, 2]`, `[2, 3]`, `[1, 2, 1]`, `[2, 1, 2]`, `[1, 2, 3]`.

**Goal of the Code:**  
To efficiently determine the number of subarrays that contain **exactly k distinct integers** using the **Sliding Window (Two Pointers) Technique**.  

**Approach**  
The problem can be efficiently solved using the **"At Most" trick** with a sliding window:  

1. **Key Observation:**  
   - The **number of subarrays** with **exactly k distinct integers** can be computed as:  
     \[
     \text{exactlyK}(arr, k) = \text{atMost}(arr, k) - \text{atMost}(arr, k-1)
     \]
   - Here, `atMost(arr, k)` gives the count of subarrays with **at most k distinct** integers.

2. **Sliding Window Approach for `atMost(arr, k)`:**  
   - Use two pointers: `left` and `right` to form a dynamic window.  
   - Maintain a **frequency map** to track the count of distinct integers in the window.  
   - If the number of distinct elements exceeds `k`, shrink the window (`left++`).  
   - Add the count of valid subarrays formed at each step (`right - left + 1`).  
   - Compute the result as:  
     \[
     \text{exactlyK} = \text{atMost}(arr, k) - \text{atMost}(arr, k-1)
     \]

**Pseudocode/Algorithm**
1. Define a helper function `atMost(arr, k)`:  
   - Initialize `left = 0`, `right = 0`, and a frequency map.  
   - Expand the right pointer (`right++`), updating the frequency map.  
   - If the number of distinct integers exceeds `k`, shrink the left pointer (`left++`).  
   - Add `(right - left + 1)` to the result.  

2. Compute `exactlyK(arr, k)` using:  
   \[
   \text{atMost}(arr, k) - \text{atMost}(arr, k-1)
   \]
   Sure! Here's the **pseudocode** and a description for the **flowchart** for the problem "Subarrays with K Distinct Integers":

**Flowchart**

1. **Start**
2. **Initialize variables:** `left = 0`, `result = 0`, `freqMap = empty`
3. **For** each `right` in `0` to `arr.length - 1`:

   * Add `arr[right]` to `freqMap` and update frequency
   * If new distinct element added (freq == 1), decrement `k`
4. **While** `k < 0`:

   * Decrease frequency of `arr[left]` in `freqMap`
   * If frequency becomes zero, increment `k`
   * Move `left` pointer forward
5. **Calculate** subarrays count ending at `right`: `result += (right - left + 1)`
6. Repeat step 3 until end of array
7. **Return** `result`
8. **End**

**Java Solution**
```java
// User function Template for Java
import java.util.*;

class Solution {
    private static int atMost(int arr[], int k) {
        int n = arr.length, res = 0, l = 0;
        Map<Integer, Integer> freq = new HashMap<>();
        
        for (int r = 0; r < n; r++) {
            freq.put(arr[r], freq.getOrDefault(arr[r], 0) + 1);
            
            if (freq.get(arr[r]) == 1) {
                k--;  // A new distinct element is added
            }
            
            while (k < 0) {  // Shrink window if distinct elements exceed k
                freq.put(arr[l], freq.get(arr[l]) - 1);
                if (freq.get(arr[l]) == 0) {
                    k++;  // Remove distinct element
                }
                l++;
            }
            
            res += (r - l + 1);
        }
        return res;
    }

    static int exactlyK(int arr[], int k) {
        return atMost(arr, k) - atMost(arr, k - 1);
    }
}
```
**Time and Space Complexity**
- **Time Complexity:** `O(N)`, since each element is processed at most twice (once when expanding, once when shrinking).  
- **Space Complexity:** `O(K)`, for storing frequency counts of at most `K` distinct elements.  

**Notes:**  
- The "atMost" trick helps reduce the problem of counting exactly k distinct elements to two simpler problems of counting at most k and at most k-1 distinct elements.  
- Using a frequency map with sliding window ensures O(N) time complexity.  
- The frequency map is updated dynamically as the window expands and contracts.  
- This approach avoids brute-force enumeration of all subarrays, which would be O(N^2).
