## Problem Statement : 137 
Implement a stack using an array with two operations:  
1. **push(x)** – Push integer `x` onto the stack.  
2. **pop()** – Remove and return the top element of the stack. If the stack is empty, return `-1`.  

**Input:**  
- Queries are given in the following format:  
  - `1 x` → Push `x` onto the stack.  
  - `2` → Pop and print the top element from the stack.  
- The input consists of multiple test cases.  

**Output:**  
For each pop operation (`2` query type), print the popped element.  

**Constraints:**  
- Stack size **≤ 1000** (Fixed size array).  
- The number of queries **≤ 10^5**.  
- Each element **fits within 32-bit integer range**.  

**Goal of the Code:**  
Efficiently implement a stack using an array with **O(1) time complexity** for `push()` and `pop()` operations.  

**Approach:**  
- Use an **integer array** `arr[]` of fixed size (`1000`).
- Maintain an integer `top`, initialized to `-1`, representing the index of the top element.  
- **push(x)**:  
  - If the stack is **not full**, increment `top` and store `x` at `arr[top]`.  
  - If full, do nothing (stack overflow is ignored).  
- **pop()**:  
  - If `top == -1`, return `-1` (stack is empty).  
  - Otherwise, return `arr[top]` and decrement `top`.  

**Pseudocode/Algorithm:**  
1. **Initialize** `arr[]` (size 1000) and `top = -1`.  
2. **push(x) Function:**  
   - If `top < 999`, increment `top` and store `x`.  
   - Else, do nothing (stack overflow case).  
3. **pop() Function:**  
   - If `top == -1`, return `-1` (stack is empty).  
   - Otherwise, return `arr[top]` and decrement `top`.  
4. **Process Queries:**  
   - Read multiple queries from input and perform `push()` or `pop()`.  
   - Print results of pop operations.  

**Time and Space Complexity:**  
- **Time Complexity:**  
  - `push(x)`: **O(1)**  
  - `pop()`: **O(1)**  
  - **Overall:** **O(n)** for `n` queries.  
- **Space Complexity:** **O(1)** (Fixed array of size 1000).  

**Additional Notes/Hints:**  
- Stack follows **LIFO (Last In, First Out)** principle.  
- Using a **fixed-size array** is memory-efficient but lacks dynamic resizing.  
- If **dynamic size** is needed, consider using `ArrayList` or `LinkedList`.  


## Problem Statement: 138
**Two Stacks in an Array** 
You are required to implement two stacks using a single array efficiently. The operations that need to be implemented are:  

1. **twoStacks()** – Initializes the data structure and variables needed to manage two stacks in one array.  
2. **push1(int x)** – Pushes an element `x` into the first stack.  
3. **push2(int x)** – Pushes an element `x` into the second stack.  
4. **pop1()** – Pops an element from the first stack and returns the popped element. If the first stack is empty, it should return `-1`.  
5. **pop2()** – Pops an element from the second stack and returns the popped element. If the second stack is empty, it should return `-1`.  

**Input**  
- The input consists of multiple test cases.  
- Each test case includes a number of queries that specify operations on either of the two stacks.  

**Output**  
- Output consists of values returned by the `pop1()` and `pop2()` operations.  
- If a stack is empty, popping from it should return `-1`.  

**Constraints**  
- The array should be used efficiently without wasting extra space.  
- The time complexity for `push` and `pop` operations should be **O(1)**.  
- Stack operations should handle overflow and underflow conditions gracefully.  

**Goal of the Code**  
The main objective is to implement **two stacks using a single array** while ensuring efficient space utilization and constant-time operations for push and pop.  

**Approach** 
To efficiently manage two stacks in one array, we divide the array into two parts:  
- **Stack 1** grows from the left (index `0` onwards).  
- **Stack 2** grows from the right (index `size-1` backwards).  
- A push operation adds an element at the respective stack's top index.  
- A pop operation removes and returns the top element of the respective stack.  
- We ensure that both stacks do not overlap; otherwise, an overflow condition occurs.  

**Pseudocode/Algorithm** 

#Initialization (`twoStacks`)  
1. Define an array `arr[]` of fixed size `size`.  
2. Set `top1 = -1` (indicating the top of Stack 1).  
3. Set `top2 = size` (indicating the top of Stack 2).  

#Push Operations  
**push1(int x)** (Push to Stack 1)  
1. Check if `top1 < top2 - 1`. If true:  
   - Increment `top1` and place `x` at `arr[top1]`.  
   - If false, print "Stack Overflow".  

**push2(int x)** (Push to Stack 2)  
1. Check if `top1 < top2 - 1`. If true:  
   - Decrement `top2` and place `x` at `arr[top2]`.  
   - If false, print "Stack Overflow".  

#Pop Operations  
**pop1()** (Pop from Stack 1)  
1. If `top1 >= 0`:  
   - Retrieve `arr[top1]` and decrement `top1`.  
   - Return the retrieved value.  
2. Else, return `-1` (Stack 1 is empty).  

**pop2()** (Pop from Stack 2)  
1. If `top2 < size`:  
   - Retrieve `arr[top2]` and increment `top2`.  
   - Return the retrieved value.  
2. Else, return `-1` (Stack 2 is empty).  

**Time and Space Complexity** 
- **Time Complexity**: O(1) for all operations (`push1`, `push2`, `pop1`, `pop2`).  
- **Space Complexity**: O(n) where `n` is the size of the array.  

**Additional Notes**  
- This approach ensures efficient space utilization by allowing both stacks to grow dynamically within the array.  
- If the two stacks meet in the middle, it indicates that the array is fully occupied.  
- This implementation prevents unused space, unlike naive implementations that preallocate separate halves of the array for each stack.

**Parenthesis Checker**  
Given a string `s` composed of different combinations of parentheses `'(' , ')' , '{' , '}' , '[' , ']'`, verify whether the arrangement of brackets is **valid**.  

A string is considered **valid** if:  
1. Open brackets must be closed by the same type of brackets.  
2. Open brackets must be closed in the correct order.  

**Input**  
- The input consists of multiple test cases.  
- Each test case contains a single string `s` of length `n`.  

**Output**  
- Print `"true"` if the given string is valid, otherwise print `"false"`.  

**Constraints**  
- `1 ≤ n ≤ 10^5` (length of the string)  
- The string only consists of `'(' , ')' , '{' , '}' , '[' , ']'`.  

**Goal of the Code**  
To determine if a given string of parentheses is **balanced** by using a **stack-based approach**.  

**Approach**  
To validate the arrangement of brackets:  
1. Use a **stack** to keep track of the last opened bracket.  
2. Traverse the string character by character:  
   - If it is an **opening bracket** (`'('`, `'{'`, `'['`), push it onto the stack.  
   - If it is a **closing bracket** (`')'`, `'}'`, `']'`):  
     - Check if the stack is empty (indicating an unmatched closing bracket).  
     - Pop the top element from the stack and verify if it matches the corresponding opening bracket.  
     - If it does not match, return `"false"`.  
3. After traversing the string, if the stack is **empty**, return `"true"`, otherwise return `"false"`.  

**Pseudocode/Algorithm**    
1. Initialize an empty **stack**.  
2. Traverse each character `ch` in the string `s`:  
   - If `ch` is `'('`, `'{'`, `'['`, push it onto the stack.  
   - If `ch` is `')'`, `'}'`, or `']'`:  
     - If the stack is empty, return `false` (unmatched closing bracket).  
     - Pop the top element from the stack.  
     - Check if the popped element matches the correct opening bracket.  
     - If not, return `false`.  
3. After the loop, return `true` if the stack is empty (all brackets matched), otherwise return `false`.  

- **Time Complexity:** **O(n)** (each character is pushed and popped at most once).  
- **Space Complexity:** **O(n)** (in the worst case, all brackets are opening and stored in the stack).  

**Additional Notes**
The solution uses a stack to efficiently track and match opening and closing brackets.

**Edge cases handled:**
Unmatched brackets (e.g., "{{}", "([)]").
Empty string "" (should return "true").
String with only opening brackets ("(((") or only closing brackets (")))").


## Problem Statement: 140 
Given an array `arr[]` of integers, find the **Next Greater Element (NGE)** for each element in the array.  
- The **NGE** for an element `arr[i]` is the **nearest element on the right** that is **greater** than `arr[i]`.  
- If no greater element exists, the **NGE** is `-1`.  

**Input:**  
- An integer array `arr[]`.  

**Output:**  
- A list of integers where each element represents the next greater element of the corresponding input element.  

**Constraints:**  
- `1 ≤ arr.length ≤ 10^6`  
- `-10^9 ≤ arr[i] ≤ 10^9`  

**Goal of the Code:**  
Efficiently determine the **next greater element** for each array element in **O(n) time complexity** using a **monotonic stack**.  

**Approach:**  
1. **Use a Stack (Monotonic Decreasing Stack Approach)**:  
   - Traverse the array from **right to left**.  
   - Maintain a **stack** to store elements in **decreasing order** (smallest elements are removed first).  
   - For each element:  
     - **Pop** elements from the stack while the top is **≤ current element** (since they can't be NGE).  
     - If the stack is **not empty**, the top element is the **NGE** for the current element.  
     - **Push the current element** onto the stack.  
   - This ensures each element is pushed and popped **at most once**, achieving **O(n) time complexity**.  

**Pseudocode/Algorithm:**  
1. Initialize `result[]` with `-1` for all elements.  
2. Create an empty stack `s`.  
3. Traverse the array **from right to left**:  
   - While `s` is not empty and `s.peek() ≤ arr[i]`, pop from `s`.  
   - If `s` is not empty, set `result[i] = s.peek()`.  
   - Push `arr[i]` onto `s`.  
4. Return `result`.  

**Time and Space Complexity:**  
- **Time Complexity:** **O(n)** (each element is pushed and popped **at most once**).  
- **Space Complexity:** **O(n)** (for the result array and stack).  

**Additional Notes/Hints:**  
- **Why use a stack?**  
  - A **brute force approach** (checking each element’s next greater element) would take **O(n²)** time.  
  - A **stack-based approach** reduces this to **O(n)**.  
- **Edge Cases:**  
  - If the array is **strictly decreasing**, every element’s NGE is `-1`.  
  - If the array is **strictly increasing**, each element’s NGE is the next element.  


## Problem Statement: 141
**Get Min from Stack**   
You are given **q queries** where you need to implement a stack with the following functions:  

1. **push(x)** – Insert an integer `x` onto the stack.  
2. **pop()** – Remove the top element from the stack.  
3. **peek()** – Return the top element from the stack. If the stack is empty, return `-1`.  
4. **getMin()** – Retrieve the minimum element from the stack in **O(1) time**. If the stack is empty, return `-1`.  

Each query can be one of the following:  
- `1 x` → Push `x` onto the stack.  
- `2` → Pop the top element from the stack.  
- `3` → Return the top element from the stack. If empty, return `-1`.  
- `4` → Return the minimum element from the stack. If empty, return `-1`.  

**Input**  
- **q** (1 ≤ q ≤ 10^5) - Number of queries.  
- Each query consists of a type (1, 2, 3, or 4) followed by an integer in case of push operation (`1 x`).  

**Output**  
- For each `3` query, print the top element of the stack.  
- For each `4` query, print the minimum element in the stack.  

**Constraints**  
- The stack operations must be performed in **O(1)** time complexity.  
- The stack can store large numbers (within the range of integers).  
- The space complexity should be **O(n)** (no extra space should be used other than the stack itself).  


**Goal of the Code**  
To efficiently **push, pop, get the top, and find the minimum element in O(1) time** using a **modified stack approach**.  


**Approach**  
A **normal stack** does not support retrieving the minimum element in **O(1)** time.  
To achieve this, we use a **trick with an auxiliary minimum tracking variable** (`minEle`).  

#Optimized Stack Approach:
- Instead of using an **extra stack**, store **modified values** when pushing an element **smaller than minEle**.  
- If the new element `x` is smaller than `minEle`:  
  - Push a **modified value** `2*x - minEle` onto the stack.  
  - Update `minEle = x`.  
- When popping, if the popped value is **less than minEle**, it means `minEle` was updated and needs to be restored.  

#Explanation:
1. **Push(x)**
   - If the stack is empty, set `minEle = x` and push `x`.
   - If `x < minEle`, push `2*x - minEle` instead and update `minEle = x`.
   - Otherwise, push `x` normally.

2. **Pop()**
   - If the popped element is **less than minEle**, it was a special stored value, so update `minEle = 2*minEle - poppedValue`.

3. **Peek()**
   - If the top element is **less than minEle**, return `minEle` (since the stored value is modified). Otherwise, return the top element.

4. **getMin()**
   - Return `minEle` as it stores the **minimum element in O(1) time**.


**Pseudocode/Algorithm**  
 
1. **Initialize**: Create a stack `s` and a variable `minEle` to store the current minimum.  
2. **Push(x)**:
   - If the stack is empty, set `minEle = x` and push `x`.
   - If `x < minEle`, push `2*x - minEle` (modified value) and update `minEle = x`.
   - Else, push `x` normally.
3. **Pop()**:
   - If the popped value is **less than minEle**, update `minEle = 2*minEle - poppedValue`.
4. **Peek()**:
   - If the top element is **less than minEle**, return `minEle` (real value).
   - Else, return the top element.
5. **getMin()**:
   - Return `minEle`.

**Time and Space Complexity**  
- **Time Complexity:** **O(1)** for all operations (`push`, `pop`, `peek`, `getMin`).  
- **Space Complexity:** **O(n)** (no extra space other than stack).  

**Additional Notes**
This approach efficiently retrieves the minimum element in O(1) time.
Instead of using an extra stack, we store modified values when necessary.
The trick: 2*x - minEle helps track the previous minimum when popping.
This method reduces auxiliary space usage while maintaining efficiency.

## Problem Statement : 142 
Given a **circular integer array** `arr[]`, determine the **next greater element (NGE)** for each element in the array.  
- The **NGE** for an element `arr[i]` is the **first greater element** encountered **when traversing circularly**.  
- If no such element exists, return `-1` for that position.  

**Circular Property:**  
- After reaching the last element, continue searching from the beginning **until all elements are checked once**.  

**Input:**  
- An **integer array** `arr[]`.  

**Output:**  
- A **list of integers**, where each element represents the **next greater element** of the corresponding input element.  

**Constraints:**  
- `1 ≤ arr.length ≤ 10^6`  
- `-10^9 ≤ arr[i] ≤ 10^9`  

**Goal of the Code:**  
Efficiently determine the **next greater element** for each array element in **O(n) time complexity** using a **monotonic stack** while handling circular traversal.  

**Approach:**  
1. **Use a Stack for Monotonic Decreasing Order:**  
   - Maintain **indices** in a **stack** to track elements **awaiting their NGE**.  
   - Traverse the array **twice** (`2*n` iterations) to handle circular behavior.  
   - Use **modulus operator (`%`)** to wrap around after reaching the last element.  

2. **Processing Steps:**  
   - Traverse the array **twice** (`0 → 2n-1`).  
   - Use `index = i % n` to get the correct element position.  
   - **For each element:**  
     - **Pop from the stack** while `arr[s.peek()] < arr[index]`, since `arr[index]` is the next greater element.  
     - Assign `arr[index]` as the **NGE** for the popped index.  
     - **Push index into the stack** (only for the **first traversal (`i < n`)**).  
   - Convert `result[]` to an **ArrayList** and return.  

**Pseudocode/Algorithm:**  
1. **Initialize** `result[]` with `-1`.  
2. **Create an empty stack `s`** (stores indices).  
3. **Loop through `2*n` elements:**  
   - Compute `index = i % n` (handles circular indexing).  
   - While `s` is not empty **and** `arr[s.peek()] < arr[index]`:  
     - `result[s.pop()] = arr[index]` (assign NGE).  
   - Push `index` **only if `i < n`**.  
4. **Return `result[]` as an ArrayList**.  

**Time and Space Complexity:**  
- **Time Complexity:** **O(n)** (Each element is pushed and popped **at most once**).  
- **Space Complexity:** **O(n)** (Stack stores at most `n` elements).  

**Additional Notes/Hints:**  
- **Why traverse `2*n` times?**  
  - This ensures elements that don’t find NGE in the first pass can still check previous elements (circular nature).  
- **Edge Cases:**  
  - If the array is **strictly decreasing**, every element’s NGE is `-1`.  
  - If all elements are **equal**, all NGEs are `-1`.  


## Problem Statement: 143
**Stock Span Problem**  
The **stock span problem** is a financial problem where we are given a **series of daily stock prices**, and we need to compute the **span** for each day.  

**Definition of Span:**
The span `arr[i]` of the stock price on a given day **i** is defined as the **maximum number of consecutive days** (including the current day) for which the stock price **is less than or equal** to its price on the current day.  

**Input**  
- An integer **n** representing the number of days.  
- An array `arr[]` of size **n**, where `arr[i]` represents the stock price on the `i-th` day.  

**Output**  
- An array `span[]` of size **n**, where `span[i]` represents the **span of stock price** on the `i-th` day.  

**Constraints**  
- **1 ≤ n ≤ 10^5**  
- **1 ≤ arr[i] ≤ 10^9**  
- The solution should run efficiently in **O(n) time complexity**.  

**Goal of the Code**  
To efficiently **compute the stock span** for each day using a **monotonic stack approach**, ensuring **O(n) time complexity**.  

**Approach**  
A **brute-force approach** would involve checking all previous elements for each day, leading to **O(n²) time complexity**, which is **inefficient for large inputs**.  

**Optimized Stack-Based Approach** (**O(n) Time Complexity**)
- Use a **monotonic decreasing stack** to store **indices** of the stock prices.  
- The stack helps us efficiently determine the number of consecutive days where the stock price is **less than or equal** to the current price.  

**Algorithm:**
1. **Initialize** an empty stack `s` and a result array `span[]`.
2. **Iterate** through the stock prices:
   - **Pop elements** from the stack while the **top index’s stock price** is smaller than or equal to the current stock price.
   - If the stack is empty, set `span[i] = i + 1` (price is greater than all previous days).
   - Else, set `span[i] = i - stack.peek()` (distance to the last greater stock price).
   - Push the **current index** onto the stack.
3. **Return** the computed `span[]`.

**Pseudocode/Algorithm**  
```plaintext
1. Initialize a stack `s` (stores indices).
2. Create an array `span[]` of size `n` initialized to 0.
3. Loop through `i = 0` to `n-1`:
   a. While `s` is not empty and `arr[s.peek()] <= arr[i]`, pop from `s`.
   b. If `s` is empty, set `span[i] = i + 1` (since no previous greater value exists).
   c. Else, set `span[i] = i - s.peek()` (difference from last greater price index).
   d. Push `i` onto the stack.
4. Return `span[]`.
```
**Time Complexity:** O(n) (Each element is pushed and popped from the stack at most once).
**Space Complexity:** O(n) (For the stack storing indices).

**Additional Notes**
The monotonic decreasing stack ensures that each element is processed only once, making the solution highly efficient.
This O(n) solution is the best possible approach for this problem.
This technique is commonly used in Next Greater Element problems and Histogram problems.

**Approach:**  
1. **Use a Stack for Monotonic Decreasing Order:**  
   - Maintain a **stack of indices** to track students **awaiting their classmate selection**.  
   - The stack helps to find the **nearest classmate with lower marks** efficiently.  

2. **Processing Steps:**  
   - Initialize an **array `result[]` with `-1`** (default if no classmate is found).

```java
// User function Template for Java
class Solution {
    public Stack<Integer> insertAtBottom(Stack<Integer> st, int x) {
        // Base Case: If stack is empty, push x
        if(st.isEmpty()){
            st.push(x);
            return st;
        }

        // Recursive Step: Pop top element
        int top = st.pop();

        // Recur for the rest of the stack
        st = insertAtBottom(st, x);

        // Push the top element back
        st.push(top);

        return st;
    }
}


## Problem Statement: 153
**Reverse a Stack Using Recursion**
You are given a **stack `St`** containing `n` integers. Your task is to **reverse** the stack **using recursion**.

**Key Details**
- You **cannot** use extra data structures like arrays or queues.
- The reversal should be **in-place**, meaning you modify the given stack.
- You can use **only recursion** to achieve this.

**Input Format**
- An **integer `n`** (number of elements in the stack).
- A **stack `St`** of size `n` containing `n` integers.

**Output Format**
- Return the **reversed stack**.

**Constraints**
- `1 ≤ n ≤ 10⁴`
- `-10⁹ ≤ St[i] ≤ 10⁹`

**Explanation**
- Initially, the stack is `[1, 2, 3, 4]` (topmost element = `4`).
- After reversing, the stack becomes `[4, 3, 2, 1]`.

**Approach: Using Recursion**
**Algorithm**
1. **Base Case:** If the stack is empty, return.
2. Otherwise:
   - Pop the **top element**.
   - Recursively **reverse the remaining stack**.
   - Insert the popped element **at the bottom** of the reversed stack.

**Helper Function: Insert at Bottom**
- This function **inserts an element at the bottom** of the stack using recursion.
1. **Base Case:** If the stack is empty, push `x` and return.
2. Otherwise:
   - Pop the **top element**.
   - Recursively call `insertAtBottom(x, s)`.
   - Push back the popped element after recursion finishes.

**Time and Space Complexity**
- **Time Complexity:** `O(n²)`
  - We recursively reverse `n` elements (`O(n)`) and insert each element at the bottom (`O(n)`), leading to `O(n²)`.
- **Space Complexity:** `O(n)`
  - Recursive calls use **stack space** proportional to `O(n)`.

**Code Implementation (Recursive Approach)**
```java
class Solution { 
    // Helper function to insert an element at the bottom of the stack
    private static void insertAtBottom(int x, Stack<Integer> s) {
        if (s.isEmpty()) {
            s.push(x);
            return;
        }
        
        // Pop the top element
        int top = s.pop();
        
        // Recur for the remaining stack
        insertAtBottom(x, s);
        
        // Push back the popped element
        s.push(top);
    }

    // Function to reverse the stack
    static void reverse(Stack<Integer> s) {
        if (!s.isEmpty()) {
            // Pop the top element
            int top = s.pop();
            
            // Recur for the remaining stack
            reverse(s);
            
            // Insert the popped element at the bottom
            insertAtBottom(top, s);
        }
    }
}


## Problem Statement: 154
**Sort a Stack Using Recursion**
You are given a **stack `s`** containing `N` integers. Your task is to **sort the stack in descending order** (i.e., **the largest element should be on top**) using recursion.

**Key Details**
- You **cannot use additional data structures** (like arrays or other lists) except for recursive function calls.
- Sorting should be **in-place**.

**Input Format**
- An **integer `N`** (number of elements in the stack).
- A **stack `s`** of size `N` containing `N` integers.

**Output Format**
- Return the **sorted stack** in **descending order**.

**Constraints**
- `1 ≤ N ≤ 100`
- `-10⁶ ≤ s[i] ≤ 10⁶`
- **Expected Time Complexity:** `O(N²)`
- **Expected Auxiliary Space:** `O(N)` (due to recursion).

**Approach: Recursion + Insertion Sort**
**Algorithm**
1. **Base Case:** If the stack is empty, return.
2. Otherwise:
   - Pop the **top element**.
   - Recursively **sort the remaining stack**.
   - Insert the popped element **at the correct position** in the sorted stack.

**Helper Function: Insert in Sorted Order**
- This function **inserts an element at the correct position** in the sorted stack.
1. **Base Case:** If the stack is empty or the **top element is greater than `x`**, push `x` and return.
2. Otherwise:
   - Pop the **top element**.
   - Recursively call `insertSorted(x, s)`.
   - Push back the popped element after recursion finishes.

**Time and Space Complexity**
- **Time Complexity:** `O(N²)`
  - We recursively sort `N` elements (`O(N)`) and insert each element in the correct position (`O(N)`), leading to `O(N²)`.
- **Space Complexity:** `O(N)`
  - Recursive calls use **stack space** proportional to `O(N)`.

**Code Implementation (Recursive Approach)**
```java
class GfG {
    // Helper function to insert an element at the correct position in sorted stack
    private void insertSorted(Stack<Integer> s, int x) {
        // Base case: If stack is empty or top element is greater than x, push x
        if (s.isEmpty() || s.peek() >= x) {
            s.push(x);
            return;
        }
        
        // Remove the top element
        int temp = s.pop();
        
        // Recursively insert x at the correct position
        insertSorted(s, x);
        
        // Push the removed element back
        s.push(temp);
    }

    // Function to sort the stack using recursion
    public Stack<Integer> sort(Stack<Integer> s) {
        // Base case: If stack is empty, return
        if (!s.isEmpty()) {
            // Remove the top element
            int temp = s.pop();
            
            // Recursively sort the remaining stack
            sort(s);
            
            // Insert the removed element at the correct position
            insertSorted(s, temp);
        }
        return s;
    }
}


## Problem Statement: 155
**Arithmetic Expression Evaluation (Infix to Result)**
You are given a **mathematical expression** in **infix notation** as a string. Your task is to evaluate the expression and return its result.

**Constraints**
1. The expression **only** contains:
   - **Operators:** `+`, `-`, `*`, `/`
   - **Parentheses:** `(`, `)`
   - **Digits:** `0-9`
2. **Operator Precedence:**
   - `*` and `/` have **higher precedence** than `+` and `-`.
   - `/` is **integer division (floor division)**.
3. The expression is **fully parenthesized**, meaning:
   - It **always starts with** `(`.
   - It **always ends with** `)`.
4. The input expression **will always be valid**.
5. **No division by zero** will occur.
6. The **operands and final result fit in a 32-bit integer**.

**Approach: Using Stacks**
We use **two stacks**:
1. **Operands Stack (`operands`)**: Stores integer values.
2. **Operators Stack (`operators`)**: Stores `+`, `-`, `*`, `/`, `(`.

**Algorithm**
1. **Iterate through each character in the string**:
   - If it is a **digit**, extract the full number and push it to `operands`.
   - If it is an **opening parenthesis `(`**, push it to `operators`.
   - If it is a **closing parenthesis `)`**, evaluate the expression inside:
     - Keep popping from `operators` and applying them to `operands` until `(` is encountered.
     - Remove `(` from the stack.
   - If it is an **operator (`+`, `-`, `*`, `/`)**, check precedence:
     - If it has **lower or equal precedence** than the top of `operators`, apply the top operator.
     - Push the current operator to `operators`.

2. **Process remaining operators** after iteration:
   - Apply all remaining operations in `operators` to `operands`.

3. **Final result** is the last remaining element in `operands`.

**Time and Space Complexity**
- **Time Complexity:** `O(N)` (each element is pushed/popped once).
- **Space Complexity:** `O(N)` (storing numbers and operators in stacks).

**Code Implementation (Java)**
```java
import java.util.* ;
import java.io.*; 

public class Solution {
    // Helper function to apply an operator on two operands
    private static int applyOperation(char op, int b, int a) {
        switch (op) {
            case '+': return a + b;
            case '-': return a - b;
            case '*': return a * b;
            case '/': return a / b;  // Integer division
        }
        return 0;
    }

    // Helper function to define operator precedence
    private static int precedence(char op) {
        if (op == '+' || op == '-') return 1;
        if (op == '*' || op == '/') return 2;
        return 0;
    }

    public static int evaluateArithmeticExpression(String expression) {
        Stack<Integer> operands = new Stack<>();
        Stack<Character> operators = new Stack<>();
        int n = expression.length();

        for (int i = 0; i < n; i++) {
            char ch = expression.charAt(i);

            // If character is a digit, extract full number
            if (Character.isDigit(ch)) {
                int num = 0;
                while (i < n && Character.isDigit(expression.charAt(i))) {
                    num = num * 10 + (expression.charAt(i) - '0');
                    i++;
                }
                i--; // Move back to last digit
                operands.push(num);
            }
            // If opening parenthesis, push to operator stack
            else if (ch == '(') {
                operators.push(ch);
            }
            // If closing parenthesis, evaluate expression inside
            else if (ch == ')') {
                while (operators.peek() != '(') {
                    operands.push(applyOperation(operators.pop(), operands.pop(), operands.pop()));
                }
                operators.pop(); // Remove '(' from stack
            }
            // If an operator, apply precedence rules
            else if (ch == '+' || ch == '-' || ch == '*' || ch == '/') {
                while (!operators.isEmpty() && precedence(ch) <= precedence(operators.peek())) {
                    operands.push(applyOperation(operators.pop(), operands.pop(), operands.pop()));
                }
                operators.push(ch);
            }
        }

        // Process any remaining operations
        while (!operators.isEmpty()) {
            operands.push(applyOperation(operators.pop(), operands.pop(), operands.pop()));
        }

        return operands.pop();
    }
}


## Problem Statement: 156
**The Celebrity Problem**
A **celebrity** at a party is someone who:
1. **Is known by everyone** (i.e., all other `n-1` people know them).
2. **Knows no one** (i.e., they don't know any other person).

You are given an **N x N** matrix `mat[][]`, where:
- `mat[i][j] = 1` → Person `i` knows Person `j`.
- `mat[i][j] = 0` → Person `i` does not know Person `j`.

Your task is to **find the index** of the celebrity in the party.  
If no celebrity exists, return `-1`.

**Explanation**
- Person `0` **knows** person `1`, but **not everyone knows** person `1`.
- Person `1` **knows** person `0`, but **a celebrity cannot know anyone**.
- ❌ **No celebrity exists**.

**Constraints**
- `1 ≤ N ≤ 100`
- `mat[i][i] = 0` (A person **does not know themselves**).
- `mat[i][j]` is **either 0 or 1**.

**Approach: Stack-Based Solution**
**Algorithm**
1. **Push all people (0 to N-1) into a stack**.
2. **Find a potential celebrity**:
   - Pop two people `A` and `B` from the stack.
   - If `A` **knows** `B`, then `A` **cannot** be a celebrity → Push `B` back.
   - If `A` **does not know** `B`, then `B` **cannot** be a celebrity → Push `A` back.
   - Repeat until **only one person remains** in the stack.
3. **Verify if the remaining person is a celebrity**:
   - A celebrity **should not know** anyone (`mat[celeb][i] == 0` for all `i`).
   - A celebrity **should be known by** everyone (`mat[i][celeb] == 1` for all `i ≠ celeb`).
   - If any condition fails, return `-1`.
4. **Return the celebrity's index** if valid.

**Time and Space Complexity**
| Approach | Time Complexity | Space Complexity |
|----------|---------------|----------------|
| **Stack-Based** | `O(N)` | `O(N)` |

- **Why O(N)?**  
  - Step 2: **Finding the potential celebrity** takes **O(N)**.
  - Step 3: **Verifying the celebrity** takes **O(N)**.
- **Why O(N) Space?**  
  - We use a stack of size **O(N)**.

**Code Implementation (Java)**
```java
import java.util.*;

class Solution {
    public int celebrity(int mat[][]) {
        int n = mat.length;
        Stack<Integer> s = new Stack<>();

        // Step 1: Push all persons into the stack
        for (int i = 0; i < n; i++) {
            s.push(i);
        }

        // Step 2: Find a potential celebrity
        while (s.size() > 1) {
            int A = s.pop();
            int B = s.pop();

            // If A knows B, then A is not a celebrity, push B back
            if (mat[A][B] == 1) {
                s.push(B);
            }
            // If A does not know B, then B is not a celebrity, push A back
            else {
                s.push(A);
            }
        }

        // Step 3: Verify if the remaining person is a celebrity
        int celeb = s.pop();
        for (int i = 0; i < n; i++) {
            if (i == celeb) continue; // Skip self-check

            // A celebrity should not know anyone AND everyone should know them
            if (mat[celeb][i] != 0 || mat[i][celeb] != 1) {
                return -1;
            }
        }

        return celeb;
    }
}



## Problem Statement: 157
**Count the Reversals**
You are given a string `S` consisting of **only** opening `{` and closing `}` curly brackets.  
Your task is to **find the minimum number of reversals** required to convert `S` into a **balanced** expression.

**Reversal Definition**
A reversal means **changing** `{` to `}` or vice-versa.

**Explanation**
- The given string **cannot** be balanced by any number of reversals.

**Constraints**
- `1 ≤ |S| ≤ 10^5`
- The input string consists of **only `{` and `}`**.

**Approach: Using a Stack**
**Algorithm**
1. **Check for odd length**  
   - If `|S|` is **odd**, return `-1` (impossible to balance).
  
2. **Use a Stack to track unbalanced brackets**  
   - Traverse the string:
     - If **opening `{`**, push it to the stack.
     - If **closing `}`**:
       - If the stack is **not empty**, pop an opening `{}` (pair matched).
       - Else, push a `{` (treat this as an **unbalanced bracket**).
  
3. **Calculate the required reversals**  
   - Let `stackSize = count of unbalanced brackets left in stack`
   - If `stackSize` is **odd**, return `-1` (cannot balance).
   - Otherwise:
     - `ans = (unbalanced opening `{`) / 2 + (unbalanced closing `}`) / 2`

**Time and Space Complexity**
| Approach | Time Complexity | Space Complexity |
|----------|---------------|----------------|
| **Stack-Based** | `O(N)` | `O(N)` |

- **Why O(N)?**  
  - We traverse the string once **O(N)**.
  - Stack operations (`push`/`pop`) are **O(1)** each.
- **Why O(N) Space?**  
  - In the worst case, the stack stores **all brackets**.

**Code Implementation (Java)**
```java
import java.util.*;

class Sol {
    int countRev(String s) {
        // Step 1: If the length is odd, it is impossible to balance
        if (s.length() % 2 != 0) return -1;

        Stack<Character> st = new Stack<>();
        int ans = 0;

        // Step 2: Traverse the string
        for (int i = 0; i < s.length(); i++) {
            char ch = s.charAt(i);

            if (ch == '{') {
                st.push(ch);
            } else {
                // If stack is not empty, pop an opening `{`
                if (!st.isEmpty()) {
                    st.pop();
                } else {
                    // If stack is empty, treat `}` as an unbalanced `{`
                    st.push('{');
                    ans++;
                }
            }
        }

        // Step 3: If the stack size is odd, return -1
        if (st.size() % 2 != 0) return -1;

        // Step 4: Add half of remaining unbalanced brackets
        ans += st.size() / 2;
        return ans;
    }
}


## Problem Statement: 158
**Restrictive Candy Crush**
Given a string `s` and an integer `k`, the task is to **reduce the string** by applying the following operation:
- **Choose a group of `k` consecutive identical characters and remove them**.

The operation can be performed **any number of times** until it is **no longer possible**.

**Explanation**
- `"aa"` is removed → `"bbcc"`
- `"bb"` is removed → `"cc"`
- `"cc"` is removed → `""`
- Final result: `""` (empty string)

**Constraints**
- `1 ≤ |s| ≤ 10^5`
- `1 ≤ k ≤ |s|`
- `s` consists of **lowercase English letters** only.

**Optimal Approach: Using a Stack**
**Key Observations**
1. **Use a stack to track consecutive characters and their frequency**.
2. **If a character reaches `k` occurrences, remove them from the stack**.
3. **Build the final string from the stack**.

**Algorithm**
1. **Use a stack** to store **pairs of (character, frequency)**.
2. **Iterate through `s`**:
   - If the stack is **empty** or the top character is **different**, push `(char, 1)`.
   - If the top character matches the current character:
     - Increase the frequency.
     - If frequency reaches `k`, **remove the group**.
3. **Build the final string from the stack**.

**Time and Space Complexity**
| Approach | Time Complexity | Space Complexity |
|----------|---------------|----------------|
| **Stack-Based** | `O(N)` | `O(N)` |

- **Why O(N)?**  
  - We iterate through `s` **once**.
  - Stack operations (`push`/`pop`) are **O(1)** each.
- **Why O(N) Space?**  
  - The stack stores **at most N elements** in the worst case.

**Code Implementation (Java)**
```java
import java.util.*;

class Solution {
    public static String reduced_String(int k, String s) {
        // Edge case: If k is 1, all characters are removed
        if (k == 1) return "";

        // Step 1: Use a stack to track character frequency
        Stack<Pair> stack = new Stack<>();

        // Step 2: Iterate through the string
        for (char ch : s.toCharArray()) {
            if (!stack.isEmpty() && stack.peek().ch == ch) {
                // Increment count if the same character appears consecutively
                stack.peek().freq++;
            } else {
                // Push new character with frequency 1
                stack.push(new Pair(ch, 1));
            }

            // Step 3: If frequency reaches k, remove the character
            if (stack.peek().freq == k) {
                stack.


## Problem Statement: 159
**Queue Using Array**
Implement a **Queue** using an **Array**. The queue should support the following **operations**:

1. **Push (`1 x`)** → Insert element `x` into the queue.
2. **Pop (`2`)** → Remove and return the **front element** of the queue.
   - If the queue is **empty**, return `-1`.

**Note:**  
- Implement only the `push()` and `pop()` functions.  
- The driver code will handle input and output.

**Explanation**
- `1 2` → Push `2`
- `1 3` → Push `3`
- `2` → Pop → **Returns 2**
- `1 4` → Push `4`
- `2` → Pop → **Returns 3**

**Constraints**
- `1 ≤ Q ≤ 100000`
- `1 ≤ x ≤ 100000`


**Approach: Implementing Queue Using an Array**
**Key Observations**
1. We use an **array** `arr[]` to store queue elements.
2. **Maintain `front` and `rear` pointers**:
   - `front` → Points to the **first element**.
   - `rear` → Points to the **next insertion position**.
3. **Push (`1 x`)**
   - Insert `x` at `arr[rear]`.
   - Increment `rear`.
4. **Pop (`2`)**
   - Return `arr[front]` and increment `front`.
   - If queue is **empty**, return `-1`.


**Time and Space Complexity**
| Operation | Time Complexity | Space Complexity |
|-----------|---------------|----------------|
| **Push (`1 x`)** | `O(1)` | `O(N)` |
| **Pop (`2`)** | `O(1)` | `O(N)` |

- **Why O(1)?**  
  - Insertion (`push`) and deletion (`pop`) take **constant time**.
- **Why O(N) Space?**  
  - We use an array of size **N**.


**Code Implementation (Java)**
```java
class MyQueue {
    int front, rear;
    int arr[] = new int[100005]; // Fixed-size array

    MyQueue() {
        front = 0;
        rear = 0;
    }
    
    // Function to push an element x into the queue.
    void push(int x) {
        if (rear == arr.length) return; // Prevent overflow
        arr[rear] = x;
        rear++;
    } 

    // Function to pop an element from queue and return that element.
    int pop() {
        if (front == rear) return -1; // Queue empty
        int result = arr[front];
        front++;
        return result;
    } 
}


## Problem Statement: 160
Implement a **Queue** using a **Linked List**. The queue should support the following **operations**:

1. **Push (`1 x`)** → Insert element `x` into the queue.
2. **Pop (`2`)** → Remove and return the **front element** of the queue.
   - If the queue is **empty**, return `-1`.

**Note:**  
- Implement only the `push()` and `pop()` functions.  
- The driver code will handle input and output.

**Explanation**
- `1 10` → Push `10`
- `1 20` → Push `20`
- `2` → Pop → **Returns 10**
- `1 30` → Push `30`
- `2` → Pop → **Returns 20**

**Constraints**
- `1 ≤ Q ≤ 100000`
- `1 ≤ x ≤ 100000`

**Approach: Implementing Queue Using Linked List**
**Key Observations**
1. Unlike an **array-based queue**, a **linked list queue** does not have a fixed size.
2. **Maintain `front` and `rear` pointers**:
   - `front` → Points to the **first element**.
   - `rear` → Points to the **last element**.
3. **Push (`1 x`)**
   - Create a new node and add it **at the rear**.
   - Update the `rear` pointer.
4. **Pop (`2`)**
   - Remove and return the **front** element.
   - Update the `front` pointer.
   - If the queue becomes empty, set `rear = null`.

**Time and Space Complexity**
| Operation | Time Complexity | Space Complexity |
|-----------|---------------|----------------|
| **Push (`1 x`)** | `O(1)` | `O(N)` |
| **Pop (`2`)** | `O(1)` | `O(N)` |

- **Why O(1)?**  
  - Insertion (`push`) and deletion (`pop`) take **constant time**.
- **Why O(N) Space?**  
  - Each element requires a **node (extra memory for pointers)**.

**Code Implementation (Java)**
```java
class QueueNode {
    int data;
    QueueNode next;
    QueueNode(int a) {
        data = a;
        next = null;
    }
}

class MyQueue {
    QueueNode front, rear;
    
    // Function to push an element into the queue.
    void push(int a) {
        QueueNode newNode = new QueueNode(a);
        
        if (rear == null) {
            front = rear = newNode;
            return;
        }
        
        rear.next = newNode;
        rear = newNode;
    }
    
    // Function to pop front element from the queue.
    int pop() {
        if (front == null) {
            return -1; // Queue is empty
        }
        
        int result = front.data;
        front = front.next;
        
        if (front == null) {
            rear = null; // Queue is now empty
        }
        
        return result;
    }
}


## Problem Statement : 161
**Queue Using Two Stacks**
Implement a **Queue** using **two stacks (s1 & s2)**. The queue should support the following **operations**:

1. **Push (`1 x`)** → Insert element `x` into the queue.
2. **Pop (`2`)** → Remove and return the **front element** of the queue.
   - If the queue is **empty**, return `-1`.

**Note:**  
- Implement only the `Push()` and `Pop()` functions.  
- The driver code will handle input and output.

**Explanation**
- `1 10` → Push `10`
- `1 20` → Push `20`
- `2` → Pop → **Returns 10**
- `1 30` → Push `30`
- `2` → Pop → **Returns 20**

**Constraints**
- `1 ≤ Q ≤ 100000`
- `1 ≤ x ≤ 100000`

**Approach: Implementing Queue Using Two Stacks**
**Key Observations**
1. A **Queue** follows **FIFO (First In, First Out)**.
2. A **Stack** follows **LIFO (Last In, First Out)**.
3. We **simulate FIFO behavior** using two stacks:
   - **s1** → Used for push operations.
   - **s2** → Used for pop operations.

**Push Operation (`1 x`)**
1. Simply **push `x` onto `s1`**.

**Time Complexity:** `O(1)`

**Pop Operation (`2`)**
1. If `s2` is **not empty**, pop from `s2`.
2. Otherwise:
   - Transfer all elements from `s1` → `s2` (reversing order).
   - Pop from `s2`.
3. If both stacks are **empty**, return `-1`.

**Time Complexity:**  
- **Worst Case (`O(N)`)** (when transferring `s1` → `s2`).
- **Amortized (`O(1)`)** over multiple operations.

**Code Implementation (Java)**
```java
class StackQueue {
    Stack<Integer> s1 = new Stack<>();
    Stack<Integer> s2 = new Stack<>();

    // Function to push an element into queue
    void Push(int x) {
        s1.push(x);
    }

    // Function to pop front element from queue
    int Pop() {
        if (s1.isEmpty() && s2.isEmpty()) {
            return -1; // Queue is empty
        }

        if (s2.isEmpty()) {
            while (!s1.isEmpty()) {
                s2.push(s1.pop());
            }
        }
        
        return s2.pop();
    }
}



## Problem Statement: 162
**Stack Using Two Queues**
Implement a **Stack** using **two queues (`q1` and `q2`)**. The stack should support the following **operations**:

1. **Push (`1 x`)** → Insert element `x` into the stack.
2. **Pop (`2`)** → Remove and return the **top element** of the stack.
   - If the stack is **empty**, return `-1`.

**Note:**  
- Implement only the `push()` and `pop()` functions.  
- The driver code will handle input and output.

**Explanation**
- `1 10` → Push `10`
- `1 20` → Push `20`
- `2` → Pop → **Returns 20** (LIFO)
- `1 30` → Push `30`
- `2` → Pop → **Returns 30**

**Constraints**
- `1 ≤ Q ≤ 100000`
- `1 ≤ x ≤ 100000`

**Approach: Implementing Stack Using Two Queues**
**Key Observations**
1. A **Stack** follows **LIFO (Last In, First Out)**.
2. A **Queue** follows **FIFO (First In, First Out)**.
3. We **simulate LIFO behavior** using two queues:
   - **q1** → Stores the stack elements.
   - **q2** → Used as a temporary queue.

**Push Operation (`1 x`)**
1. Simply **enqueue `x` into `q1`**.

**Time Complexity:** `O(1)`

**Pop Operation (`2`)**
1. Move **all elements except the last** from `q1` → `q2`.
2. Remove the **last element from `q1`** (stack's top).
3. Swap `q1` and `q2`.

**Time Complexity:** `O(N)`

**Code Implementation (Java)**
```java
class Queues {
    Queue<Integer> q1 = new LinkedList<>();
    Queue<Integer> q2 = new LinkedList<>();
    
    // Function to push an element into stack using two queues.
    void push(int a) {
        q1.add(a);
    }
    
    // Function to pop an element from stack using two queues. 
    int pop() {
        if (q1.isEmpty()) {
            return -1; // Stack is empty
        }
        
        while (q1.size() > 1) {
            q2.add(q1.poll());
        }
        
        int topElement = q1.poll();
        Queue<Integer> temp = q1;
        q1 = q2;
        q2 = temp;
        
        return topElement;
    }
}



## Problem Statement:163
**Queue Reversal**
Given a **queue** `q` containing integer elements, your task is to **reverse the queue**.

**Explanation**
- The original queue is: **`1 → 2 → 3 → 4 → 5`**
- After reversal: **`5 → 4 → 3 → 2 → 1`**

**Constraints**
- `1 ≤ N ≤ 10^5` (Number of elements in the queue)
- `0 ≤ Elements in queue ≤ 10^6`

**Approach**
We can reverse the queue using **two main approaches**:

**1. Using a Stack (`O(N) Time, O(N) Space`)**
1. **Push all elements** of the queue into a **stack** (LIFO property).
2. **Pop elements** from the stack and enqueue them back into the queue.
3. The queue will now be **reversed**.

🔹 **Time Complexity**: `O(N)` (Each element is pushed and popped once)  
🔹 **Space Complexity**: `O(N)` (For the stack)

**2. Using Recursion (`O(N) Time, O(N) Space`)**
1. **Recursively remove the front element** until the queue is empty.
2. **On returning**, insert each element **at the back**.

🔹 **Time Complexity**: `O(N)` (Each element is enqueued and dequeued once)  
🔹 **Space Complexity**: `O(N)` (Recursive call stack)

**Code Implementation (Java)**
**Using a Stack**
```java
import java.util.*;

class Solution {
    // Function to reverse the queue.
    public Queue<Integer> reverseQueue(Queue<Integer> queue) {
        Stack<Integer> stack = new Stack<>();
        
        // Push all elements to stack
        while (!queue.isEmpty()) {
            stack.push(queue.poll());
        }
        
        // Pop from stack and enqueue back to queue
        while (!stack.isEmpty()) {
            queue.add(stack.pop());
        }
        
        return queue;
    }
}



## Problem Statement: 164
**Gas Station Problem**
There are some **gas stations** along a **circular route**. You are given two integer arrays:
- `gas[]`: Represents the **amount of gas** at each station.
- `cost[]`: Represents the **cost of travelling** to the next station.

You have a car with an **unlimited gas tank** but **start the journey with an empty tank**.  
Your task is to find the **starting gas station index** from which you can complete the circular journey **without running out of gas**.  
- If it is **not possible**, return `-1`.
- If a solution exists, **it is guaranteed to be unique**.

**Explanation**
- If you start at **station 3 (0-based index)**:
  - Gas collected: `4`, cost to next: `1` → Remaining `3`
  - Gas collected: `5`, cost to next: `2` → Remaining `6`
  - Gas collected: `1`, cost to next: `3` → Remaining `4`
  - Gas collected: `2`, cost to next: `4` → Remaining `2`
  - Gas collected: `3`, cost to next: `5` → Remaining `0`
- Successfully returned to station `3`.

**Constraints**
- `1 ≤ N ≤ 10^5`
- `0 ≤ gas[i], cost[i] ≤ 10^4`

**Approach**
**1. Greedy Approach (`O(N) Time, O(1) Space`)**
1. **Calculate total gas and total cost**:
   - If `totalGas < totalCost`, return `-1` (impossible to complete the circuit).
2. **Iterate through stations**:
   - Maintain a `tank` variable to track available gas.
   - If `tank` becomes negative, reset the starting station to `i+1` and reset `tank` to `0`.
3. **Return the `start` station** if `totalGas ≥ totalCost`.

🔹 **Time Complexity**: `O(N)` (Single pass)  
🔹 **Space Complexity**: `O(1)` (Constant space)  

**Code Implementation (Java)**
```java
class Solution {
    public int startStation(int[] gas, int[] cost) {
        int totalGas = 0, totalCost = 0;
        int tank = 0, start = 0;

        for (int i = 0; i < gas.length; i++) {
            totalGas += gas[i];
            totalCost += cost[i];
            tank += gas[i] - cost[i];

            // If we can't reach the next station
            if (tank < 0) {
                start = i + 1;  // Reset the start position
                tank = 0;       // Reset the tank
            }
        }
        return (totalGas >= totalCost) ? start : -1;
    }
}



## Problem Statement: 165
**Stream First Non-Repeating Character**
Given an input **stream** `s` consisting only of **lowercase alphabets**, you need to determine the **first non-repeating character** at each position **from start till i**.

- If there are multiple non-repeating characters, return the **first one**.
- If no such character exists, return `#`.

**Explanation**
- `a` appears → `"a"`
- `a` appears again → `"a#"` (no unique character)
- `b` appears → `"a#b"`
- `c` appears → `"a#bb"`

**Approach**
**1. Using Queue and Frequency Count (`O(N)`)**
1. Maintain a **queue** to keep track of characters in the order they appear.
2. Use a **frequency array** (`count[26]`) to track occurrences of each character.
3. **For each character**:
   - Increase its frequency.
   - Add to queue if it's the **first occurrence**.
   - Remove elements from the queue **until the front has frequency 1**.
   - Append **either the first non-repeating character or `#`** to the result.

**Time Complexity**
- **`O(N)`** (Each character is added and removed from the queue at most once)

**Space Complexity**
- **`O(N)`** (Queue stores unique characters, worst case all distinct)

**Java Implementation**
```java
import java.util.*;

class Solution {
    public String FirstNonRepeating(String s) {
        int[] count = new int[26];  // To track character frequencies
        Queue<Character> q = new LinkedList<>();  // Queue for processing
        StringBuilder ans = new StringBuilder();

        for (char ch : s.toCharArray()) {
            count[ch - 'a']++;  // Update frequency

            // If this is the first occurrence, add to queue
            if (count[ch - 'a'] == 1) {
                q.add(ch);
            }

            // Remove characters from queue if they are no longer unique
            while (!q.isEmpty() && count[q.peek() - 'a'] > 1) {
                q.poll();
            }

            // Append first non-repeating character or #
            ans.append(q.isEmpty() ? '#' : q.peek());
        }
        return ans.toString();
    }
}


## Problem Statement: 166
**Reverse First K Elements of Queue**

**Given an integer** `K` **and a queue of integers, reverse the order of the first** `K` **elements while keeping the rest in the same relative order.** 

- Only the following operations are allowed:
  - `enqueue(x)`: Add an item `x` to the rear of the queue.
  - `dequeue()`: Remove an item from the front of the queue.
  - `size()`: Returns the number of elements in the queue.
  - `front()`: Finds the front item.

**Explanation**
- The first `3` elements `[1, 2, 3]` are reversed to `[3, 2, 1]`.
- The remaining `[4, 5]` stay the same.

**Approach**
**1. Using a Stack and Queue (`O(N)`)**
1. **Extract the first `K` elements** from the queue and **push them into a stack**.
2. **Pop elements from the stack** and **enqueue them back** to the queue (this reverses the order).
3. **Move the remaining elements (`n-k`)** to the back of the queue to **maintain the original relative order**.

**Time Complexity**
- **`O(K) + O(K) + O(N-K) = O(N)`**

**Space Complexity**
- **`O(K)`** (Stack stores first `K` elements)

**Java Implementation**
```java
import java.util.*;

class GfG {
    // Function to reverse first k elements of a queue.
    public Queue<Integer> modifyQueue(Queue<Integer> q, int k) {
        Deque<Integer> stack = new ArrayDeque<>();
        
        // Step 1: Push first K elements into a stack
        for (int i = 0; i < k; i++) {
            stack.push(q.poll());
        }
        
        // Step 2: Pop elements from stack and enqueue back (Reversed order)
        while (!stack.isEmpty()) {
            q.add(stack.pop());
        }
        
        // Step 3: Move the remaining (n-k) elements to maintain relative order
        int size = q.size();
        for (int i = 0; i < size - k; i++) {
            q.add(q.poll());
        }
        
        return q;
    }
}



## Problem Statement: 167 
**Minimum Cost of Ropes**

Given an array `arr[]` representing the lengths of ropes, the goal is to connect all ropes into a single rope with the **minimum total cost**. The **cost** of connecting two ropes is the sum of their lengths. The process continues until one single rope is formed.

**Explanation**
1. Pick `2` and `3`, cost = `2+3=5`, new array: `[4, 5, 6]`
2. Pick `4` and `5`, cost = `4+5=9`, new array: `[6, 9]`
3. Pick `6` and `9`, cost = `6+9=15`, new array: `[15]`
4. Total cost = `5+9+15 = 29`

**Approach**
**1. Using Min-Heap (`O(N log N)`)**
1. **Push all elements** into a **Min-Heap** (Priority Queue).
2. **Extract the two smallest elements**, compute the cost, and insert the new rope back.
3. **Repeat** until one rope remains.
4. **Sum all costs** to get the minimum.

**Time Complexity**
- **`O(N log N)`**  
  - **Heap operations (`O(log N)`)** happen `N-1` times.
- **Efficient for large inputs!**

**Space Complexity**
- **`O(N)`** (For heap storage)

**Java Implementation**
```java
import java.util.*;

class Solution {
    // Function to return the minimum cost of connecting the ropes.
    public int minCost(int[] arr) {
        // Min-Heap to get smallest elements first
        PriorityQueue<Integer> pq = new PriorityQueue<>();
        for (int num : arr) {
            pq.add(num);
        }
        
        int totalCost = 0;
        
        while (pq.size() > 1) {
            int first = pq.poll();
            int second = pq.poll();
            
            int sum = first + second;
            totalCost += sum;
            
            pq.add(sum);
        }
        
        return totalCost;
    }
}


## Problem Statement: 168
**Nearly Sorted Array**
Given an array `arr[]`, where each element is **at most `k` positions away** from its target position in a sorted order, sort the array **optimally**.

**Constraints**
- The array is **not completely unsorted** but has elements **displaced by at most `k` positions**.
- Sorting should be **in-place**.

**Explanation**
Each element is misplaced by **at most 3 positions**. We use a **Min-Heap** to efficiently sort.

**Approach**
**1. Min-Heap (`O(N log K)`)**
1. Use a **Min-Heap (Priority Queue)** to store **at most `k+1` elements**.
2. **Insert the first `k+1` elements** into the heap.
3. **Iterate over the rest of the array**:
   - Extract the **smallest element** and place it in the sorted order.
   - Insert the next element from the array into the heap.
4. **Empty the heap** for remaining elements.

**Time Complexity**
- **Insertion & Deletion in Min-Heap** → `O(log k)`
- **Processing `N` elements** → `O(N log k)`
- **Overall Complexity** → **`O(N log k)`** (much better than `O(N log N)` sorting)

**Space Complexity**
- **`O(k)`** (Heap stores at most `k+1` elements)

**Java Implementation**
```java
import java.util.*;

class Solution {
    public void nearlySorted(int[] arr, int k) {
        int n = arr.length;
        PriorityQueue<Integer> pq = new PriorityQueue<>();

        // Add first k+1 elements into the min-heap
        for (int i = 0; i <= k && i < n; i++) {
            pq.add(arr[i]);
        }

        int index = 0;

        // Process remaining elements
        for (int i = k + 1; i < n; i++) {
            arr[index++] = pq.poll();
            pq.add(arr[i]);
        }

        // Empty the remaining elements from the heap
        while (!pq.isEmpty()) {
            arr[index++] = pq.poll();
        }
    }
}


## Problem Statement: 169
**Count Distinct Elements in Every Window**

Given an integer array `arr[]` and a number `k`, find the count of **distinct elements** in **every window of size `k`** in the array.

**Input:**
- An integer array `arr[]` of size `n`
- An integer `k` representing the size of the window

**Output:**
- A list of integers representing the count of distinct elements in each window of size `k`.

**Constraints:**
- `1 <= n <= 10^5`
- `1 <= k <= n`
- `0 <= arr[i] <= 10^5`

**Goal of the Code:**
To efficiently find and count the number of distinct elements within every window of size `k` using an optimized approach.

**Approach:**
1. Use a **Hash Map** to keep track of element frequencies within the current window.
2. Slide the window by one element at a time:
   - Add the next element to the map.
   - Remove the frequency of the element that is sliding out of the window.
   - Track the count of distinct elements by checking the size of the map.
3. Continue until the end of the array.

**Pseudocode/Algorithm:**
1. Initialize a Hash Map to keep track of element frequencies.
2. Traverse the first `k` elements and update the map.
3. For each subsequent element:
   - Add the new element to the map.
   - Reduce the count of the outgoing element in the map.
   - If the count of the outgoing element becomes zero, remove it from the map.
   - Store the size of the map as the result for the current window.
4. Return the results.

**Time and Space Complexity:**
- **Time Complexity:** `O(N)` where `N` is the size of the array. Each insertion and deletion operation in the map takes `O(1)` on average.
- **Space Complexity:** `O(K)` where `K` is the size of the window (size of the hash map).

**Additional Notes/Hints:**
- Use a sliding window technique to ensure we only make changes for elements entering or leaving the window.



## Problem Statement: 170
**Max Sum Subarray of Size K**
Given an array of integers `arr[]` and a number `k`, return the maximum sum of a contiguous subarray of size `k`.

**Input:**
- An integer array `arr[]` (positive/negative integers).
- An integer `k` (size of the subarray).

**Output:**
- The maximum sum of a contiguous subarray of size `k`.

**Constraints:**
- `1 <= k <= arr.length <= 10^5`
- `-10^4 <= arr[i] <= 10^4`

**Goal of the Code:**
To efficiently find the maximum sum of any contiguous subarray of size `k` using an optimized approach.

**Approach:**
1. Use **Sliding Window Technique**.
2. Calculate the sum of the first window of size `k`.
3. For each subsequent element, slide the window by:
   - Removing the element that is moving out of the window.
   - Adding the new element coming into the window.
4. Track the maximum sum encountered.

**Pseudocode/Algorithm:**
1. Initialize `result` and `current_sum` as the sum of the first window of size `k`.
2. Traverse the array from index `k` to `n`:
   - Update `current_sum` by adding `arr[i]` and removing `arr[i - k]`.
   - Update `result` if `current_sum` is greater.
3. Return `result`.

**Time and Space Complexity:**
- **Time Complexity:** `O(N)` (Single traversal of the array).
- **Space Complexity:** `O(1)` (Constant space for variables).

**Additional Notes/Hints:**
- This approach ensures an optimal solution using **sliding window technique** rather than brute force.

## Problem Statement: 171
**First Negative Integer in Every Window of Size K**

Given an array `arr[]` and a positive integer `k`, find the **first negative integer** in each and every contiguous subarray (window) of size `k`.

If a window does not contain a negative integer, return `0` for that window.

**Input:**
- An integer array `arr[]` (contains positive/negative integers).
- An integer `k` (size of the subarray/window).

**Output:**
- A list of integers representing the first negative number in every window of size `k`. If no negative number exists, return `0`.

**Constraints:**
- `1 <= k <= arr.length <= 10^5`
- `-10^4 <= arr[i] <= 10^4`

**Goal of the Code:**
To efficiently find the first negative integer in each window using an optimized approach.

**Approach:**
1. Use **Sliding Window Technique** with a **Deque (Double-Ended Queue)**.
2. Traverse the array while maintaining a deque to store indices of negative numbers.
3. For each window of size `k`:
   - If the front of the deque is out of the current window, remove it.
   - Add the current index if the element is negative.
   - Store the first negative integer (if present) or `0` if no negative integer exists.
4. Continue until all windows are processed.

**Pseudocode/Algorithm:**
1. Initialize a `Deque<Integer>` to store indices of negative numbers.
2. Traverse the first `k` elements:
   - If the element is negative, store its index in the deque.
3. Process the remaining elements:
   - Remove indices that are out of the current window.
   - Add the new element’s index if it is negative.
   - Store `arr[deque.peek()]` if deque is not empty, else store `0`.
4. Return the result list.

**Time and Space Complexity:**
- **Time Complexity:** `O(N)` (Each element is processed once).
- **Space Complexity:** `O(K)` (Deque stores at most `k` elements).

**Additional Notes/Hints:**
- This approach ensures an optimal solution using **sliding window technique** rather than brute force.
- Deque operations (insertion/removal from both ends) are **O(1)**, making the approach highly efficient.

## Problem Statement: 172
**K Sized Subarray Maximum**

Given an array `arr[]` of integers and an integer `k`, find the **maximum value** for each contiguous subarray of size `k`. The output should be an array of maximum values corresponding to each contiguous subarray.

**Input:**
- An integer array `arr[]`.
- An integer `k` (size of the subarray).

**Output:**
- An array containing the maximum values of all contiguous subarrays of size `k`.

**Constraints:**
- `1 <= k <= arr.length <= 10^5`
- `-10^4 <= arr[i] <= 10^4`

**Goal of the Code:**
To efficiently find the maximum element in each contiguous subarray of size `k` using an optimized approach.

**Approach:**
1. Use a **Deque (Double-Ended Queue)** to efficiently track the maximum value in each window.
2. Process the first `k` elements:
   - Remove elements from the back of the deque that are smaller than the current element.
   - Add the current index to the deque.
3. For each subsequent element:
   - Store the front element of the deque as the maximum for the previous window.
   - Remove indices that are out of the current window.
   - Remove elements from the back that are smaller than the current element.
   - Add the current index to the deque.
4. Store the maximum of the last window.

**Pseudocode/Algorithm:**
1. Initialize a `Deque` to store indices of useful elements.
2. Process the first `k` elements and maintain the deque.
3. Traverse the rest of the array:
   - Append the maximum element from the deque to the result.
   - Remove out-of-window elements from the deque.
   - Maintain decreasing order by removing smaller elements.
   - Add the current element index to the deque.
4. Append the last maximum element to the result.

**Time and Space Complexity:**
- **Time Complexity:** `O(N)` (Each element is processed once).
- **Space Complexity:** `O(K)` (For the deque storing indices).

**Additional Notes/Hints:**
- The use of a **deque** ensures that each element is processed in constant time.
- This approach is much more efficient than brute force `O(N*K)` solutions.



## Problem Statement: 173
**Substrings of Length K with K-1 Distinct Elements**

Given a string `S` consisting of only lowercase alphabets and an integer `K`, find the count of all substrings of length `K` which have exactly `K-1` distinct characters.

**Input:**
- A string `S` of lowercase alphabets.
- An integer `K` (length of the substring).

**Output:**
- An integer representing the count of substrings of length `K` with exactly `K-1` distinct characters.

**Constraints:**
- `1 <= K <= S.length <= 10^5`
- `S` consists of only lowercase English letters.

**Goal of the Code:**
To efficiently count the number of substrings of length `K` that contain exactly `K-1` distinct characters using an optimized approach.

**Approach:**
1. Use a **sliding window technique** along with a **frequency array** to track character occurrences.
2. Process the first `K-1` characters and store their frequencies.
3. For each subsequent character:
   - Add the current character to the frequency map.
   - Count distinct characters.
   - If the distinct count is exactly `K-1`, increment the result.
   - Remove the first character of the previous window from the frequency map.
4. Return the final count of valid substrings.

**Pseudocode/Algorithm:**
1. Initialize an array `count[26]` to store character frequencies.
2. Traverse the first `K-1` characters and update their frequencies.
3. Iterate through the rest of the string:
   - Add the new character to the frequency map.
   - Count distinct characters.
   - If distinct characters == `K-1`, increase the count.
   - Remove the first character of the previous window from the map.
4. Return the final count of valid substrings.

**Time and Space Complexity:**
- **Time Complexity:** `O(N)`, as we iterate through the string once.
- **Space Complexity:** `O(1)`, since we use a fixed-size frequency array (26 letters).

**Additional Notes/Hints:**
- The sliding window technique ensures efficient traversal of the string without redundant calculations.
- Using an array for character frequencies is more efficient than using a HashMap.

## Problem Statement: 174
**Indexes of Subarray Sum**

Given an array `arr[]` containing only non-negative integers, your task is to find a continuous subarray (a contiguous sequence of elements) whose sum equals a specified value `target`. You need to return the **1-based indices** of the leftmost and rightmost elements of this subarray. If no such subarray is found, return `[-1]`.

**Input:**
- An integer array `arr[]` containing non-negative integers.
- An integer `target` representing the desired sum.

**Output:**
- A list containing the **1-based** indices `[left, right]` of the first subarray whose sum equals `target`.
- If no subarray is found, return `[-1]`.

**Constraints:**
- `1 <= arr.length <= 10^5`
- `0 <= arr[i] <= 10^5`
- `1 <= target <= 10^9`

**Goal of the Code:**
To efficiently find the first contiguous subarray whose sum equals `target` using an optimized approach.

**Approach:**
1. Use the **Sliding Window (Two-Pointer) Technique** to maintain a running sum of the subarray.
2. Start with two pointers (`start` and `end`) and a variable `curr_sum` initialized to `0`.
3. Iterate through the array:
   - Expand the window by adding the current element to `curr_sum`.
   - If `curr_sum` exceeds `target`, shrink the window from the left by removing elements until `curr_sum <= target`.
   - If `curr_sum == target`, return the 1-based indices `[start+1, end+1]`.
4. If no subarray is found, return `[-1]`.

**Pseudocode/Algorithm:**
1. Initialize `start = 0`, `curr_sum = 0`.
2. Traverse the array using `end` pointer:
   - Add `arr[end]` to `curr_sum`.
   - While `curr_sum > target`, subtract `arr[start]` and increment `start`.
   - If `curr_sum == target`, return `[start+1, end+1]`.
3. If no subarray is found, return `[-1]`.

**Time and Space Complexity:**
- **Time Complexity:** `O(N)`, as each element is processed at most twice (once added, once removed).
- **Space Complexity:** `O(1)`, as only a few extra variables are used.

**Additional Notes/Hints:**
- The sliding window approach ensures an efficient solution compared to brute force methods (`O(N^2)`).
- The algorithm works optimally for both small and large `target` values.

## Problem Statement: 175
**Subarray with 0 Sum**

Given an array of integers `arr[]`, determine if there is a subarray (of size at least one) with a sum equal to `0`. Return `true` if such a subarray exists, otherwise return `false`.

**Input:**
- An integer array `arr[]` of size `n`.

**Output:**
- A boolean value (`true` or `false`) indicating whether a subarray with sum `0` exists.

**Constraints:**
- `1 <= n <= 10^5`
- `-10^5 <= arr[i] <= 10^5`

**Goal of the Code:**
To efficiently check if any subarray within the given array has a sum of `0` using an optimized approach.

**Approach:**
1. Use a **Hash Set** to store the cumulative sum encountered so far.
2. Iterate through the array while maintaining a running sum.
3. If at any point:
   - The current element is `0`, or
   - The cumulative sum becomes `0`, or
   - The cumulative sum has been seen before in the set,
   then a subarray with sum `0` exists.
4. If no such subarray is found, return `false`.

**Pseudocode/Algorithm:**
1. Initialize an empty `Hash Set` and a variable `sum = 0`.
2. Traverse through the array:
   - Add `arr[i]` to `sum`.
   - If `sum == 0` or `arr[i] == 0` or `sum` is already in the set, return `true`.
   - Otherwise, add `sum` to the set.
3. If loop completes without finding a zero-sum subarray, return `false`.

**Time and Space Complexity:**
- **Time Complexity:** `O(N)`, as we traverse the array once.
- **Space Complexity:** `O(N)`, due to the usage of a `Hash Set`.

**Additional Notes/Hints:**
- The **Hash Set** is used to track previous cumulative sums efficiently.
- If the same cumulative sum appears again, it means the intermediate sum between two indices is `0`, confirming a zero-sum subarray.

## Problem Statement: 176  
**Length of the Longest Substring Without Repeating Characters**  

Given a string `s`, find the length of the **longest substring** that contains **no repeating characters**.  

**Input:**  
- A string `s` consisting of lowercase English letters.  

**Output:**  
- An integer representing the length of the longest substring without repeating characters.  

**Constraints:**  
- `1 <= s.length <= 10^5`  
- `s` consists of only lowercase English letters.  

**Goal of the Code:**  
To efficiently determine the length of the longest contiguous substring that contains only unique characters.  

**Approach:**  
1. Use the **Sliding Window** technique along with a **boolean array (visited characters)** to track seen characters.  
2. Maintain two pointers (`left` and `right`) to define the current substring window.  
3. Expand the `right` pointer to add characters to the window:  
   - If `s[right]` is already in the substring (visited), move `left` forward until `s[right]` is no longer duplicated.  
4. Update the maximum length (`res`) at each step.  
5. Continue until the end of the string.  

**Pseudocode/Algorithm:**  
1. Initialize `res = 0`, `left = 0`, `right = 0`, and a boolean array `vis[26]` to track seen characters.  
2. Iterate through the string while `right < s.length`:  
   - If `s[right]` is already in the substring, move `left` forward and update `vis[]`.  
   - Mark `s[right]` as visited and update `res = max(res, right - left + 1)`.  
   - Move `right` forward.  
3. Return `res` as the length of the longest substring.  

**Time and Space Complexity:**  
- **Time Complexity:** `O(N)`, since each character is processed at most twice (once when added, once when removed).  
- **Space Complexity:** `O(1)`, as the character set is limited to 26 lowercase English letters.  

**Additional Notes/Hints:**  
- This approach ensures that each character is only visited twice, making it optimal for large input sizes.  
- The sliding window method significantly reduces unnecessary comparisons compared to brute force.  


## Problem Statement: 177  
**Largest Subarray with Equal Number of 0s and 1s**  

Given an array `arr[]` consisting of `0s` and `1s`, find and return the **length of the longest contiguous subarray** that contains an **equal number of 0s and 1s**.  

**Input:**  
- An integer array `arr[]` consisting of `0s` and `1s`.  

**Output:**  
- An integer representing the length of the longest subarray with equal numbers of `0s` and `1s`.  

**Constraints:**  
- `1 <= arr.length <= 10^5`  
- `arr[i]` is either `0` or `1`.  

**Goal of the Code:**  
To efficiently determine the longest contiguous subarray where the count of `0s` and `1s` is equal.  

**Approach:**  
1. **Transform the problem:**  
   - Convert `0s` to `-1s`. This allows us to reduce the problem to finding the **longest subarray with sum = 0**.  

2. **Use a HashMap to track prefix sums:**  
   - Maintain a `prefix sum` variable (`preSum`) while traversing the array.  
   - If `preSum == 0`, it means the entire subarray from `0` to `i` has equal `0s` and `1s`. Update the result.  
   - If `preSum` has been seen before at index `j`, then the subarray from `j+1` to `i` also sums to `0`, and we update the maximum length.  
   - Otherwise, store `preSum` with its first occurrence index in the HashMap.  

**Pseudocode/Algorithm:**  
1. Initialize a `HashMap<Integer, Integer>` to store first occurrence indices of `prefix sums`.  
2. Initialize `preSum = 0` and `res = 0`.  
3. Traverse the array:  
   - Convert `arr[i]` from `0` to `-1`.  
   - Add `arr[i]` to `preSum`.  
   - If `preSum == 0`, update `res = i + 1`.  
   - If `preSum` exists in the HashMap, compute the subarray length and update `res`.  
   - Otherwise, store `preSum` in the HashMap.  
4. Return `res` as the final result.  

**Time and Space Complexity:**  
- **Time Complexity:** `O(N)`, as we traverse the array once.  
- **Space Complexity:** `O(N)`, due to the usage of a HashMap to store prefix sums.  

**Additional Notes/Hints:**  
- This approach is optimized using prefix sums and HashMaps, making it significantly faster than brute-force methods.  
- The transformation of `0s` to `-1s` ensures that we only need to track the sum instead of counting `0s` and `1s` separately.  


## Problem Statement: 178  
**Count Occurrences of Anagrams**  

Given a word `pat` and a text `txt`, return the count of occurrences of **anagrams** of `pat` in `txt`.  

**Input:**  
- A string `txt` representing the main text.  
- A string `pat` representing the pattern whose anagrams need to be counted.  

**Output:**  
- An integer representing the count of occurrences of anagrams of `pat` in `txt`.  

**Constraints:**  
- `1 <= pat.length, txt.length <= 10^5`  
- `pat` and `txt` consist of lowercase English letters only.  

**Goal of the Code:**  
To efficiently count the number of anagrams of `pat` found as substrings within `txt`.  

**Approach:**  
1. **Use Sliding Window Technique:**  
   - Maintain a frequency array (`cnt[26]`) to track character occurrences in `pat`.  
   - Adjust the frequency array as we traverse `txt` while maintaining a **fixed window size** equal to `pat.length()`.  
   - If at any window position, the frequency array is all zeroes, it means an anagram is found.  

2. **Steps:**  
   - Initialize a frequency array `cnt[26]` for `pat`.  
   - Iterate through the first `pat.length() - 1` characters in `txt` and decrement their count.  
   - Use a sliding window to check each substring of length `pat.length()` in `txt`:  
     - Decrement the count of the new character in the window.  
     - Check if all counts are zero (`allZero()` function).  
     - Increment the count of the character that is moving out of the window.  
   - Return the total count of valid anagrams.  

**Pseudocode/Algorithm:**  
1. If `pat.length > txt.length`, return `0`.  
2. Initialize a frequency array `cnt[26]` for `pat`.  
3. Iterate over the first `pat.length() - 1` characters of `txt` and update `cnt[]`.  
4. Start a sliding window from `pat.length() - 1` to `txt.length()`:  
   - Update `cnt[]` for the new character.  
   - If `cnt[]` contains only zeroes, increment `ans`.  
   - Restore `cnt[]` by adding back the character that slides out.  
5. Return `ans` as the total count of anagrams found.  

**Time and Space Complexity:**  
- **Time Complexity:** `O(N)`, where `N` is the length of `txt`.  
- **Space Complexity:** `O(1)`, since we use a fixed-size frequency array of size `26`.  

**Additional Notes/Hints:**  
- The **Sliding Window technique** ensures that we efficiently check all substrings without redundant recalculations.  
- The **allZero() function** is used to check if a given substring is an anagram by verifying that all character counts are zero.  
- This approach is significantly faster than brute force methods (`O(N * M)`) and works well for large inputs.  


## Problem Statement: 179  
**Maximize Number of 1's**  

Given a binary array `arr[]` (containing only `0s` and `1s`) and an integer `k`, you are allowed to **flip at most `k` 0s to 1s**.  
Find the **maximum number of consecutive 1's** that can be obtained in the array after performing the operation at most `k` times.  

**Input:**  
- An integer array `arr[]` consisting of only `0s` and `1s`.  
- An integer `k`, representing the maximum number of `0s` that can be flipped.  

**Output:**  
- An integer representing the maximum number of consecutive `1s` that can be obtained.  

**Constraints:**  
- `1 <= arr.length <= 10^5`  
- `0 <= k <= arr.length`  
- `arr[i]` is either `0` or `1`.  

**Goal of the Code:**  
To efficiently determine the maximum possible length of consecutive `1s` in the array after flipping at most `k` `0s` to `1s`.  

**Approach:**  
1. **Use the Sliding Window (Two Pointers) Technique:**  
   - Maintain a **window `[start, end]`** that represents a subarray with at most `k` flipped `0s`.  
   - Expand `end` while counting the number of `0s` in the window.  
   - If the count of `0s` exceeds `k`, shrink the window from the left (`start++`) until the count is within the limit.  
   - Keep track of the **maximum window size** encountered.  

2. **Steps:**  
   - Initialize `start = 0`, `end = 0`, `cnt = 0` (for counting `0s` in the window), and `res = 0` (result).  
   - Traverse the array while expanding `end`:  
     - If `arr[end] == 0`, increment `cnt`.  
     - If `cnt > k`, move `start` forward until `cnt <= k`.  
     - Update `res = max(res, end - start + 1)`.  
   - Return `res` as the maximum consecutive `1s` after flipping at most `k` `0s`.  

**Pseudocode/Algorithm:**  
1. Initialize `res = 0`, `start = 0`, `cnt = 0`.  
2. Iterate `end` from `0` to `arr.length - 1`:  
   - If `arr[end] == 0`, increment `cnt`.  
   - If `cnt > k`, increment `start` while decrementing `cnt` for `0s` removed.  
   - Update `res = max(res, end - start + 1)`.  
3. Return `res`.  

**Time and Space Complexity:**  
- **Time Complexity:** `O(N)`, since each element is processed once.  
- **Space Complexity:** `O(1)`, as only a few variables are used.  

**Additional Notes/Hints:**  
- The **Sliding Window** method ensures an optimal `O(N)` solution instead of `O(N^2)`.  
- This approach efficiently finds the longest sequence of `1s` by dynamically adjusting the window size based on the allowed number of flips (`k`).  


## Problem Statement: 180  
**Find Length of the Longest Subarray Containing at Most Two Distinct Integers**  

Given an array `arr[]` containing positive elements, find the length of the **longest subarray** that contains at most **two distinct integers**.  

**Input:**  
- An integer array `arr[]` consisting of positive elements.  

**Output:**  
- An integer representing the **length of the longest subarray** that contains at most **two distinct integers**.  

**Constraints:**  
- `1 <= arr.length <= 10^5`  
- `1 <= arr[i] <= 10^5`  

**Goal of the Code:**  
To efficiently determine the longest subarray that contains at most two distinct elements using an optimized approach.  

**Approach:**  
1. **Use the Sliding Window (Two Pointers) Technique:**  
   - Maintain a **window `[start, end]`** that contains at most **two distinct elements**.  
   - Use a **HashMap** to store the frequency of the elements in the window.  
   - Expand `end` while ensuring at most **two distinct integers** are present.  
   - If more than two distinct integers exist, shrink the window from `start` until only two remain.  
   - Keep track of the **maximum window size** encountered.  

2. **Steps:**  
   - Initialize `start = 0`, `end = 0`, `size = 0`, and an empty `HashMap`.  
   - Traverse the array while expanding `end`:  
     - Add `arr[end]` to the map and update its frequency.  
     - If more than **two distinct elements** are present, move `start` forward while adjusting frequencies.  
     - If an element's frequency becomes `0`, remove it from the map.  
     - Update `size = max(size, end - start + 1)`.  
   - Return `size` as the maximum valid subarray length.  

**Pseudocode/Algorithm:**  
1. Initialize `size = 0`, `start = 0`, and an empty `HashMap<Integer, Integer>`.  
2. Iterate `end` from `0` to `arr.length - 1`:  
   - Add `arr[end]` to the map with frequency.  
   - If the map contains more than `2` distinct elements, shrink the window from `start`:
     - Decrease frequency of `arr[start]`.  
     - If `arr[start]`'s frequency becomes `0`, remove it from the map.  
     - Move `start` forward.  
   - Update `size = max(size, end - start + 1)`.  
3. Return `size`.  

**Time and Space Complexity:**  
- **Time Complexity:** `O(N)`, since each element is processed once.  
- **Space Complexity:** `O(1)`, as the HashMap stores at most `2` elements at a time.  

**Additional Notes/Hints:**  
- The **Sliding Window** method ensures an optimal `O(N)` solution instead of `O(N^2)`.  
- This approach efficiently finds the longest subarray by dynamically adjusting the window size while maintaining at most two distinct elements.  


## Problem Statement: 181  
**Count Subarrays with K Odd Numbers**  

Given an array `nums` of size `n` and an integer `k`, find the **number of subarrays** that contain exactly `k` odd numbers.  

**Input:**  
- An integer `n` representing the size of the array.  
- An integer array `nums[]` of size `n`.  
- An integer `k` representing the number of odd numbers required in the subarray.  

**Output:**  
- An integer representing the **count of subarrays** that have exactly `k` odd numbers.  

**Constraints:**  
- `1 <= n <= 10^5`  
- `1 <= nums[i] <= 10^5`  
- `1 <= k <= n`  

**Goal of the Code:**  
To efficiently count the number of subarrays that contain **exactly** `k` odd numbers using an optimized approach.  

**Approach:**  
1. **Use a HashMap to store prefix sums** based on the count of odd numbers encountered.  
2. Maintain a counter `odd` to track the cumulative count of odd numbers.  
3. Traverse through the array while updating the `odd` count:  
   - If `nums[i]` is **odd**, increment `odd`.  
   - Check if `prefix.containsKey(odd - k)`, meaning there exists a prefix subarray that contributes exactly `k` odd numbers.  
   - Update the count accordingly.  
   - Store/update `odd` count occurrences in the `HashMap`.  

**Pseudocode/Algorithm:**  
1. Initialize `count = 0`, `odd = 0`, and an empty `HashMap<Integer, Integer>` to store prefix sums.  
2. Insert `{0, 1}` into the `HashMap` to handle cases where `odd == k` at the start.  
3. Iterate through the array:  
   - If `nums[i]` is odd, increase `odd` by `1`.  
   - If `prefix.containsKey(odd - k)`, add `prefix.get(odd - k)` to `count`.  
   - Store/update `odd` in the `HashMap`.  
4. Return `count`.  

**Time and Space Complexity:**  
- **Time Complexity:** `O(N)`, as we traverse the array once.  
- **Space Complexity:** `O(N)`, due to the `HashMap` storing prefix counts.  

**Additional Notes/Hints:**  
- The **prefix sum technique** is useful for counting subarrays with specific properties efficiently.  
- Using a **HashMap**, we ensure that checking and updating previous occurrences is `O(1)`.  


## Problem Statement: 182  
**Longest Repeating Character Replacement**  

Given a string `s` and an integer `k`, you can perform at most `k` operations where you choose any character in `s` and replace it with any other uppercase English character.  

Return the **length of the longest substring** that consists of the **same letter** after performing at most `k` replacements.  

**Input:**  
- A string `s` consisting of only uppercase English letters.  
- An integer `k`, the maximum number of allowed replacements.  

**Output:**  
- An integer representing the **length of the longest substring** that can be obtained after at most `k` character replacements.  

**Constraints:**  
- `1 <= s.length <= 10^5`  
- `0 <= k <= s.length`  
- `s` consists of only uppercase English letters (`A-Z`).  

**Goal of the Code:**  
To efficiently determine the **longest substring** that consists of the **same letter** after at most `k` replacements using a **sliding window approach**.  

**Approach:**  
1. **Use a Sliding Window Technique:**  
   - Maintain two pointers, `left` and `right`, to define a window.  
   - Use a `HashMap` to keep track of character frequencies in the window.  
   - Track the **most frequent character** in the window (`maxFreq`).  

2. **Adjust the window dynamically:**  
   - If the window size (`right - left + 1`) minus `maxFreq` exceeds `k`, shrink the window by moving `left` forward.  
   - Otherwise, expand the window by moving `right` forward.  

3. **Keep track of the maximum window size encountered.**  

**Pseudocode/Algorithm:**  
1. Initialize a `HashMap<Character, Integer>` to store frequency counts.  
2. Use two pointers: `left = 0` and `right = 0`.  
3. Iterate through the string:  
   - Update the frequency of `s[right]` in the map.  
   - Update `maxFreq` to store the most frequent character in the window.  
   - If `(window size - maxFreq) > k`, move `left` to reduce window size.  
   - Update `res` with the maximum length found.  
4. Return `res` after traversal.  

**Time and Space Complexity:**  
- **Time Complexity:** `O(N)`, since each character is processed at most twice (once when expanding, once when shrinking).  
- **Space Complexity:** `O(1)`, since the frequency map stores at most 26 entries (`A-Z`).  

**Additional Notes/Hints:**  
- The key observation is that we only care about the **most frequent character** in the current window.  
- The remaining characters can be converted using at most `k` operations.  
- The sliding window approach ensures an **optimal O(N) solution** instead of checking all substrings (`O(N^2)`).  


## Problem Statement: 183  
**Subarrays with K Distinct Integers**  

Given an array `arr[]` of **positive integers** and an integer `k`, find the number of **subarrays** in `arr[]` where the **count of distinct integers** is **exactly k**.  

**Input:**  
- An array `arr[]` containing **positive integers**.  
- An integer `k`, representing the **exact** number of distinct integers in a subarray.  

**Output:**  
- An integer representing the number of **subarrays** with **exactly k distinct integers**.  

**Constraints:**  
- `1 <= arr.length <= 10^5`  
- `1 <= arr[i] <= 10^6`  
- `1 <= k <= arr.length`  

**Goal of the Code:**  
To efficiently determine the number of subarrays that contain **exactly k distinct integers** using the **Sliding Window (Two Pointers) Technique**.  

**Approach**  
The problem can be efficiently solved using the **"At Most" trick** with a sliding window:  

1. **Key Observation:**  
   - The **number of subarrays** with **exactly k distinct integers** can be computed as:  
     \[
     \text{exactlyK}(arr, k) = \text{atMost}(arr, k) - \text{atMost}(arr, k-1)
     \]
   - Here, `atMost(arr, k)` gives the count of subarrays with **at most k distinct** integers.

2. **Sliding Window Approach for `atMost(arr, k)`:**  
   - Use two pointers: `left` and `right` to form a dynamic window.  
   - Maintain a **frequency map** to track the count of distinct integers in the window.  
   - If the number of distinct elements exceeds `k`, shrink the window (`left++`).  
   - Add the count of valid subarrays formed at each step (`right - left + 1`).  
   - Compute the result as:  
     \[
     \text{exactlyK} = \text{atMost}(arr, k) - \text{atMost}(arr, k-1)
     \]

**Pseudocode/Algorithm**
1. Define a helper function `atMost(arr, k)`:  
   - Initialize `left = 0`, `right = 0`, and a frequency map.  
   - Expand the right pointer (`right++`), updating the frequency map.  
   - If the number of distinct integers exceeds `k`, shrink the left pointer (`left++`).  
   - Add `(right - left + 1)` to the result.  

2. Compute `exactlyK(arr, k)` using:  
   \[
   \text{atMost}(arr, k) - \text{atMost}(arr, k-1)
   \]
   
**Time and Space Complexity**
- **Time Complexity:** `O(N)`, since each element is processed at most twice (once when expanding, once when shrinking).  
- **Space Complexity:** `O(K)`, for storing frequency counts of at most `K` distinct elements.  
**Java Solution**
```java
// User function Template for Java
import java.util.*;

class Solution {
    private static int atMost(int arr[], int k) {
        int n = arr.length, res = 0, l = 0;
        Map<Integer, Integer> freq = new HashMap<>();
        
        for (int r = 0; r < n; r++) {
            freq.put(arr[r], freq.getOrDefault(arr[r], 0) + 1);
            
            if (freq.get(arr[r]) == 1) {
                k--;  // A new distinct element is added
            }
            
            while (k < 0) {  // Shrink window if distinct elements exceed k
                freq.put(arr[l], freq.get(arr[l]) - 1);
                if (freq.get(arr[l]) == 0) {
                    k++;  // Remove distinct element
                }
                l++;
            }
            
            res += (r - l + 1);
        }
        return res;
    }

    static int exactlyK(int arr[], int k) {
        return atMost(arr, k) - atMost(arr, k - 1);
    }
}


## Problem Statement: 184  
**Smallest Window Containing All Characters of Another String**  

Given two strings `s1` and `s2`, find the **smallest window** in `s1` that contains **all characters** (including duplicates) of `s2`. If no such window exists, return `""`. If multiple windows of the same length exist, return the one with the **smallest starting index**.  

**Input:**  
- A string `s1` of **lowercase English letters**.  
- A string `s2` of **lowercase English letters**.  

**Output:**  
- A **smallest substring** in `s1` that contains **all characters** of `s2`.  
- If no such substring exists, return `""`.  

**Constraints:**  
- `1 <= s1.length, s2.length <= 10^5`  

**Goal of the Code:**  
To efficiently find the **smallest window** that contains all characters of `s2` using the **Sliding Window (Two Pointers) Technique**.  

**Approach**  
The problem can be efficiently solved using the **Sliding Window + HashMap** approach:  

1. **Key Idea:**  
   - Maintain a **frequency count** of characters in `s2`.  
   - Use **two pointers (`left` and `right`)** to expand/shrink the window dynamically.  
   - When a valid window is found (i.e., it contains all characters of `s2`), try shrinking it to find the smallest one.  

2. **Sliding Window Approach:**  
   - **Expand the right pointer (`right`)** to include characters from `s1` while maintaining a **count of matched characters**.  
   - If all characters from `s2` are matched, **shrink the left pointer (`left`)** to find the smallest possible window.  
   - **Track the minimum window length** and update the start index accordingly.  

**Pseudocode/Algorithm**
1. Initialize frequency maps for `s2` and `s1`.  
2. Expand the `right` pointer, updating the frequency map for `s1`.  
3. When all characters from `s2` are matched, shrink the `left` pointer.  
4. Track the **smallest window length** and **starting index**.  
5. Return the substring of `s1` corresponding to the smallest window found.  

**Time and Space Complexity**
- **Time Complexity:** `O(N)`, since each character is processed at most twice (once when expanding, once when shrinking).  
- **Space Complexity:** `O(1)`, since we only store character frequencies (limited to 26 lowercase letters).  

**Java Solution**
```java
import java.util.*;

class Solution {
    public static String smallestWindow(String s1, String s2) {
        int len1 = s1.length(), len2 = s2.length();
        if (len1 < len2) return "";

        HashMap<Character, Integer> hashP = new HashMap<>();
        HashMap<Character, Integer> hashS = new HashMap<>();

        for (char c : s2.toCharArray()) {
            hashP.put(c, hashP.getOrDefault(c, 0) + 1);
        }

        int start = 0, startIdx = -1, count = 0, minLen = Integer.MAX_VALUE;

        for (int end = 0; end < len1; end++) {
            char currentChar = s1.charAt(end);
            hashS.put(currentChar, hashS.getOrDefault(currentChar, 0) + 1);

            if (hashP.containsKey(currentChar) && hashS.get(currentChar) <= hashP.get(currentChar)) {
                count++;
            }

            while (count == len2) {
                if (end - start + 1 < minLen) {
                    minLen = end - start + 1;
                    startIdx = start;
                }

                char startChar = s1.charAt(start);
                hashS.put(startChar, hashS.get(startChar) - 1);
                if (hashP.containsKey(startChar) && hashS.get(startChar) < hashP.get(startChar)) {
                    count--;
                }
                start++;
            }
        }
        return startIdx == -1 ? "" : s1.substring(startIdx, startIdx + minLen);
    }
}


## Problem Statement: 185  
**Maximum of Minimum for Every Window Size**  

Given an array of integers `arr[]`, find the **maximum of the minimum values** for every possible window size in the array.  

More formally, for each window size `k`, determine the **smallest element** in all windows of size `k`, and then find the **largest value** among these minimums, where **1 ≤ k ≤ arr.length**.  

**Input:**  
- An array `arr[]` of **integers**.  

**Output:**  
- A list of integers where the `i-th` element represents the **maximum of the minimum values** for all windows of size `i+1`.  

**Constraints:**  
- `1 ≤ arr.length ≤ 10^5`  
- `-10^6 ≤ arr[i] ≤ 10^6`  

**Goal of the Code:**  
To efficiently determine the **maximum of minimum values** for every possible window size using the **Monotonic Stack + Dynamic Programming** technique.  

**Approach**  
The problem can be efficiently solved using **monotonic stacks** and **window size mapping**:  

1. **Key Idea:**  
   - Compute the **next smaller element (NSE)** and **previous smaller element (PSE)** for each index.  
   - Use this to determine the **window size where an element is the minimum**.  
   - Fill a result array where `res[k]` stores the **maximum of minimums** for window size `k+1`.  

2. **Steps to Solve:**  
   - Use a **monotonic stack** to find **next smaller element (NSE)** and **previous smaller element (PSE)**.  
   - Compute the **window size** where each element is the minimum.  
   - Use a result array `res[]` to track the **maximum of minimums**.  
   - Process `res[]` from right to left to ensure the correct values are filled.  

**Pseudocode/Algorithm**
1. Initialize stacks and arrays for **next smaller** and **previous smaller elements**.  
2. Use **monotonic stack** to compute **NSE** and **PSE**.  
3. Compute the **window size** for each element and store its **minimum value**.  
4. Process `res[]` to ensure correct values for all window sizes.  
5. Return the **result array**.  

**Time and Space Complexity**
- **Time Complexity:** `O(N)`, since each element is pushed and popped from the stack at most once.  
- **Space Complexity:** `O(N)`, for storing window sizes and results.  

**Java Solution**
```java
import java.util.*;

class Solution {
    public ArrayList<Integer> maxOfMins(int[] arr) {
        int n = arr.length;
        
        ArrayList<Integer> res = new ArrayList<>(Collections.nCopies(n, 0));
        Stack<Integer> s = new Stack<>();
        ArrayList<Integer> lenArr = new ArrayList<>(Collections.nCopies(n, 0));
        
        for (int i = 0; i < n; i++) {
            while (!s.isEmpty() && arr[s.peek()] >= arr[i]) {
                int top = s.pop();
                int windowSize = s.isEmpty() ? i : i - s.peek() - 1;
                lenArr.set(top, windowSize);
            }
            s.push(i);
        }
        
        while (!s.isEmpty()) {
            int top = s.pop();
            int windowSize = s.isEmpty() ? n : n - s.peek() - 1;
            lenArr.set(top, windowSize);
        }
        
        for (int i = 0; i < n; i++) {
            int windowSize = lenArr.get(i) - 1;
            res.set(windowSize, Math.max(res.get(windowSize), arr[i]));
        }
        
        for (int i = n - 2; i >= 0; i--) {
            res.set(i, Math.max(res.get(i), res.get(i + 1)));
        }
        
        return res;
    }
}


## Problem Statement: 186  
**Smallest Window Containing 0, 1, and 2**  

Given a string `S` consisting of the characters **'0'**, **'1'**, and **'2'**, find the **length of the smallest substring** that contains **all three characters**.  

If no such substring exists, return **-1**.  

**Input:**  
- A string `S` containing only the characters `'0'`, `'1'`, and `'2'`.  

**Output:**  
- An integer representing the **length** of the smallest substring containing **0, 1, and 2**.  

**Constraints:**  
- `1 ≤ S.length ≤ 10^5`  
- `S[i] ∈ {'0', '1', '2'}`  

**Goal of the Code:**  
To efficiently determine the **minimum-length substring** that contains **0, 1, and 2** using the **Sliding Window (Two Pointers) Technique**.  

**Approach**  
The problem can be efficiently solved using a **sliding window** approach:  

1. **Key Idea:**  
   - Use a **frequency array** to track the count of `'0'`, `'1'`, and `'2'` in the current window.  
   - Expand the right pointer (`k`) until all **three characters** are present.  
   - Once found, try shrinking the window from the left (`i`) while maintaining all three characters.  
   - Keep track of the **smallest window length** found.  

2. **Steps to Solve:**  
   - Use two pointers `i` (left) and `k` (right) to form a dynamic **sliding window**.  
   - Use a **frequency array** of size `3` to count occurrences of `'0'`, `'1'`, and `'2'`.  
   - Expand the right pointer `k` until the window contains all three characters.  
   - Shrink the left pointer `i` to find the **smallest valid window**.  
   - Return the **minimum window size**, or `-1` if no such window exists.  

**Pseudocode/Algorithm**
1. Initialize `i = 0`, `k = 0`, and `cnt = 0` (count of distinct numbers in the window).  
2. Use an array `freq[]` of size `3` to track counts of `'0'`, `'1'`, `'2'`.  
3. Expand `k` and update `freq[S[k] - '0']`.  
4. If all three numbers exist in the window (`cnt == 3`), try shrinking `i`.  
5. Keep track of the **minimum window length**.  
6. Return the **minimum window length**, or `-1` if not found.  

**Time and Space Complexity**
- **Time Complexity:** `O(N)`, since each character is processed at most twice (once when expanding, once when shrinking).  
- **Space Complexity:** `O(1)`, since we use only a fixed-size `freq[]` array of length `3`.  

**Java Solution**
```java
import java.util.*;

class Solution {
    public int smallestSubstring(String S) {
        int n = S.length(), i = 0, k = 0, cnt = 0;
        int minLen = Integer.MAX_VALUE;
        int[] freq = new int[3];
        Arrays.fill(freq, 0);
        
        while (k < n) {
            freq[S.charAt(k) - '0']++;
            if (freq[S.charAt(k) - '0'] == 1) {
                cnt++;
            }
            
            if (cnt == 3) {
                while (freq[S.charAt(i) - '0'] > 1) {
                    freq[S.charAt(i) - '0']--;
                    i++;
                }
                minLen = Math.min(minLen, k - i + 1);
                freq[S.charAt(i) - '0']--;
                i++;
                cnt--;
            }
            
            k++;
        }
        
        return (minLen == Integer.MAX_VALUE) ? -1 : minLen;
    }
}


## Problem Statement: 187 
**Find All Anagrams in a String**  

Given two strings **s** and **p**, return an array of all the **start indices** of **p's anagrams** in **s**.  
You may return the answer in **any order**.  

**Input:**  
- A string `s` of lowercase English letters.  
- A string `p` of lowercase English letters.  

**Output:**  
- A list of integers representing the starting indices of `p`'s anagrams in `s`.  

**Constraints:**  
- `1 ≤ s.length, p.length ≤ 3 * 10^4`  
- `s` and `p` consist of lowercase English letters.  

**Goal of the Code:**  
To efficiently determine the starting indices of **all anagrams** of `p` within `s` using the **Sliding Window (Fixed-Size Window) Technique**.  

**Approach**  
The problem can be efficiently solved using a **sliding window** approach:  

1. **Key Idea:**  
   - The order of characters does **not** matter for an anagram; only the **frequency of characters** must match.  
   - Use **two frequency arrays** to store character counts:
     - `pCount[]` → Stores frequency of each character in `p`.  
     - `sCount[]` → Stores frequency of the current window in `s`.  
   - Compare the **two frequency arrays** after processing each window of size `p.length()`.  

2. **Steps to Solve:**  
   - Initialize frequency arrays `pCount[]` and `sCount[]` for `p` and `s`.  
   - Use a **sliding window** of size `p.length()` to track character counts.  
   - When the window size exceeds `p.length()`, remove the leftmost character from `sCount[]`.  
   - If `sCount[]` matches `pCount[]`, store the starting index.  
   - Continue sliding the window across `s`.  

**Pseudocode/Algorithm**
1. Compute the **character frequency** of `p` in `pCount[]`.  
2. Iterate through `s` using a **sliding window** of size `p.length()`:  
   - Add the new character to `sCount[]`.  
   - If the window size exceeds `p.length()`, remove the leftmost character.  
   - If `sCount[]` matches `pCount[]`, store the start index.  
3. Return the list of **starting indices**.  

**Time and Space Complexity**
- **Time Complexity:** `O(N)`, where `N = s.length()`, since we slide through `s` once.  
- **Space Complexity:** `O(1)`, since the character count arrays are fixed at `26`.  

**Java Solution**
```java
import java.util.*;

class Solution {
    public List<Integer> findAnagrams(String s, String p) {
        List<Integer> result = new ArrayList<>();
        int sLen = s.length(), pLen = p.length();
        if (sLen < pLen) return result;
        
        int[] pCount = new int[26];
        int[] sCount = new int[26];

        // Store frequency of characters in p
        for (char c : p.toCharArray()) {
            pCount[c - 'a']++;
        }

        for (int i = 0; i < sLen; i++) {
            sCount[s.charAt(i) - 'a']++;  // Add new character to window

            if (i >= pLen) {
                sCount[s.charAt(i - pLen) - 'a']--; // Remove leftmost character
            }

            if (Arrays.equals(pCount, sCount)) {
                result.add(i - pLen + 1); // Store starting index
            }
        }
        return result;
    }
}


## Problem Statement:  188
**Longest Substring with Distinct Characters**  

Given a string **s**, find the **length** of the longest substring that contains **all distinct characters**.  

**Input:**  
- A string `s` consisting of lowercase English letters.  

**Output:**  
- An integer representing the length of the **longest substring** with all distinct characters.  

**Constraints:**  
- `1 ≤ s.length ≤ 10^5`  

**Goal of the Code:**  
To efficiently determine the **maximum length** of a contiguous substring in `s` where all characters are **distinct**, using the **Sliding Window (Two Pointers) Technique**.  

**Approach**  
The problem can be efficiently solved using a **sliding window** approach:  

1. **Key Idea:**  
   - Maintain a **boolean array** (`vis[]`) to track characters present in the window.  
   - Expand the right pointer (`right++`) while keeping track of distinct characters.  
   - If a duplicate character is found, shrink the window by moving `left++` until all characters are distinct again.  
   - Update the maximum length (`res`).  

2. **Steps to Solve:**  
   - Initialize `left = 0`, `right = 0`, and a **boolean array** `vis[26]` to mark visited characters.  
   - Expand the right pointer (`right++`) and mark the character as visited.  
   - If a duplicate character is found, move the left pointer (`left++`) until uniqueness is restored.  
   - Update the **maximum length** (`res`).  
   - Continue until `right` reaches the end of the string.  

**Pseudocode/Algorithm**
1. Initialize `left = 0`, `right = 0`, and `res = 0`.  
2. Use a **boolean array** `vis[26]` to mark characters present in the current window.  
3. Iterate over `s` with `right++`:  
   - If `s[right]` is already in `vis[]`, shrink the window by moving `left++`.  
   - Mark `s[right]` as visited in `vis[]`.  
   - Update `res = max(res, right - left + 1)`.  
4. Return `res` as the maximum substring length.  

**Time and Space Complexity**
- **Time Complexity:** `O(N)`, since each character is processed at most twice (once for expansion, once for contraction).  
- **Space Complexity:** `O(1)`, since we use a **fixed** boolean array of size `26` (constant space).  

**Java Solution**
```java
import java.io.*;

class Solution {
    public int longestUniqueSubstr(String s) {
        if (s.length() == 0 || s.length() == 1) {
            return s.length();
        }

        int res = 0, right = 0, left = 0;
        boolean[] vis = new boolean[26];

        while (right < s.length()) {
            while (vis[s.charAt(right) - 'a']) {
                vis[s.charAt(left) - 'a'] = false;
                left++;
            }

            vis[s.charAt(right) - 'a'] = true;
            res = Math.max(res, right - left + 1);
            right++;
        }
        return res;
    }
}


## Problem Statement: 189
**Smallest Distinct Window**  

Given a string **s**, find the length of the **smallest window** that contains **all the characters** of the given string **at least once**.  

**Input:**  
- A string `s` consisting of lowercase and uppercase English letters.  

**Output:**  
- An integer representing the **length** of the smallest substring that contains **all distinct characters** in `s`.  

**Constraints:**  
- `1 ≤ s.length ≤ 10^5`  

**Goal of the Code:**  
To efficiently determine the **smallest substring length** that contains **all unique characters** of `s` at least once, using the **Sliding Window (Two Pointers) Technique**.  

**Approach**  
The problem can be efficiently solved using a **sliding window approach** with **two pointers**:  

1. **Key Idea:**  
   - The **number of distinct characters** in `s` is fixed (`distinct`).  
   - Use a **frequency map** (`count[]`) to track occurrences of characters in the current window.  
   - Expand the window (`end++`) to include all distinct characters.  
   - Once all distinct characters are in the window (`matched == distinct`), try to shrink it (`start++`) while maintaining all distinct characters.  
   - Update the **minimum window length** found.  

2. **Steps to Solve:**  
   - Compute `distinct`, the number of unique characters in `s`.  
   - Use a frequency array `count[256]` to track character counts.  
   - Expand the window (`end++`) while maintaining a count of **matched** distinct characters.  
   - Once all distinct characters are in the window, try **shrinking** it from the left (`start++`).  
   - Keep track of the **smallest window size** found.  

**Pseudocode/Algorithm**
1. Compute `distinct = count of unique characters in s`.  
2. Initialize `start = 0`, `minLen = Integer.MAX_VALUE`, and `matched = 0`.  
3. Use `count[256]` to track occurrences of characters in the current window.  
4. Expand the window (`end++`) until all distinct characters are included.  
5. Once `matched == distinct`:  
   - Try **shrinking** the window from `start++`.  
   - Update `minLen` if the current window is smaller.  
   - If removing `s[start]` reduces `matched`, stop shrinking.  
6. Return `minLen` as the smallest distinct window length.  

**Time and Space Complexity**
- **Time Complexity:** `O(N)`, since each character is processed at most twice (once for expansion, once for contraction).  
- **Space Complexity:** `O(1)`, since we use a **fixed** frequency array of size `256` (constant space).  

**Java Solution**
```java
import java.io.*;

class Solution {
    public int findSubString(String s) {
        int n = s.length();
        if (n <= 1) return n;

        int[] count = new int[256];
        int distinct = (int) s.chars().distinct().count();
        int start = 0, startIdx = -1, matched = 0, minLen = Integer.MAX_VALUE;

        for (int end = 0; end < n; end++) {
            if (++count[s.charAt(end)] == 1) matched++;

            while (matched == distinct) {
                if (end - start + 1 < minLen) {
                    minLen = end - start + 1;
                    startIdx = start;
                }
                if (--count[s.charAt(start++)] == 0) matched--;
            }
        }
        return startIdx == -1 ? -1 : minLen;
    }
}


## Problem Statement: 190
**Longest Substring with K Unique Characters**  

Given a string `s`, find the **length** of the longest substring that contains **exactly** `k` unique characters. If no such substring exists, return `-1`.  

**Input:**  
- A string `s` consisting of lowercase English letters.  
- An integer `k`, representing the number of **distinct** characters allowed in the substring.  

**Output:**  
- An integer representing the **length** of the longest substring containing **exactly k unique characters**.  

**Constraints:**  
- `1 ≤ s.length ≤ 10^5`  
- `1 ≤ k ≤ 26`  

**Goal of the Code:**  
To efficiently determine the **longest substring** that contains **exactly k distinct characters** using the **Sliding Window (Two Pointers) Technique**.  

**Approach**  
The problem can be efficiently solved using a **sliding window approach** with **two pointers**:  

1. **Key Idea:**  
   - Maintain a **window** `[start, end]` with at most `k` unique characters.  
   - Use a **frequency array** (`count[26]`) to track character occurrences.  
   - Expand the window (`end++`) to include more characters.  
   - If the number of unique characters **exceeds k**, shrink the window (`start++`) until only `k` unique characters remain.  
   - Update the **maximum window size** found whenever `uniqueCount == k`.  

2. **Steps to Solve:**  
   - Initialize `start = 0`, `maxLen = -1`, and `uniqueCount = 0`.  
   - Expand the window (`end++`), updating `count[26]` and `uniqueCount`.  
   - If `uniqueCount > k`, shrink the window (`start++`) until `uniqueCount == k`.  
   - Track the **maximum window size** found.  

**Pseudocode/Algorithm**
1. Initialize `start = 0`, `maxLen = -1`, and `count[26] = {0}`.  
2. Expand the window (`end++`) while updating `count[]`.  
3. If adding a character increases `uniqueCount` beyond `k`, shrink the window (`start++`).  
4. If `uniqueCount == k`, update `maxLen = max(maxLen, end - start + 1)`.  
5. Return `maxLen`.  

**Time and Space Complexity**
- **Time Complexity:** `O(N)`, since each character is processed at most twice (once for expansion, once for contraction).  
- **Space Complexity:** `O(1)`, since we use a **fixed** frequency array of size `26` (constant space).  

**Java Solution**
```java
class Solution {
    public int longestkSubstr(String s, int k) {
        int n = s.length(), start = 0, maxLen = -1, count[] = new int[26];
        int uniqueCount = 0; // Track distinct characters

        for (int end = 0; end < n; end++) {
            if (count[s.charAt(end) - 'a']++ == 0) uniqueCount++; // New distinct char

            while (uniqueCount > k) { // Shrink window if unique chars exceed k
                if (--count[s.charAt(start) - 'a'] == 0) uniqueCount--; 
                start++;
            }

            if (uniqueCount == k) maxLen = Math.max(maxLen, end - start + 1);
        }
        return maxLen;
    }
}


## Problem Statement: 191
**Subarrays with K Different Integers**

You are given an integer array `nums` and an integer `k`.  
Return the **number of good subarrays** of `nums`.

A **good subarray** is defined as a **contiguous** part of the array that contains **exactly `k` distinct integers**.

**Approach:**

To solve this problem efficiently, we use the **sliding window technique** combined with a **mathematical trick**.

**Key Insight:**
The number of subarrays with **exactly `k`** distinct integers =  
**Subarrays with at most `k` distinct integers** - **Subarrays with at most `k-1` distinct integers**

**Steps:**

1. **Implement a helper function `atMost(nums, k)`**:
   - Counts how many subarrays have at most `k` distinct integers.
   - Uses a HashMap to keep track of frequency of elements in the current window.
   - Expands the window to the right (`right` pointer) and shrinks from the left (`left` pointer) when the unique count exceeds `k`.

2. **Main function**:
   - Result = `atMost(nums, k)` - `atMost(nums, k-1)`

**Time & Space Complexity:**

- **Time Complexity**: O(N), where N is the length of the array `nums`.  
  Each element is added and removed from the HashMap at most once.
  
- **Space Complexity**: O(K), for storing at most `k` distinct elements in the HashMap.

**Java Code:**

```java
class Solution {
    // Helper method to count subarrays with at most k distinct integers
    private int atMost(int[] nums, int k) {
        if (k == 0) return 0;

        HashMap<Integer, Integer> hm = new HashMap<>();
        int left = 0, count = 0;

        for (int right = 0; right < nums.length; right++) {
            hm.put(nums[right], hm.getOrDefault(nums[right], 0) + 1);

            // Shrink the window if we have more than k distinct numbers
            while (hm.size() > k) {
                hm.put(nums[left], hm.get(nums[left]) - 1);
                if (hm.get(nums[left]) == 0) {
                    hm.remove(nums[left]);
                }
                left++;
            }

            // All subarrays ending at 'right' with at most k distinct numbers
            count += right - left + 1;
        }

        return count;
    }

    public int subarraysWithKDistinct(int[] nums, int k) {
        // Subarrays with exactly k = atMost(k) - atMost(k-1)
        return atMost(nums, k) - atMost(nums, k - 1);
    }
}


## Problem Statement: 192
**First Negative Integer in Every Window of Size K**

Given an integer array `arr[]` and a positive integer `k`, find the **first negative integer** in every window (i.e., contiguous subarray) of size `k`.

If a window does not contain any negative integers, return `0` for that window.

**Explanation:**
- Window 1: [12, -1, -7] → First negative = -1  
- Window 2: [-1, -7, 8] → First negative = -1  
- Window 3: [-7, 8, -15] → First negative = -7  
- Window 4: [8, -15, 30] → First negative = -15  
- Window 5: [-15, 30, 16] → First negative = -15  
- Window 6: [30, 16, 28] → No negative → 0

**Approach:**
We use a **sliding window** approach with the help of a **deque** (double-ended queue) to keep track of negative elements in the current window.

**Steps:**
1. Iterate through the first `k` elements of the array:
   - If the element is negative, store its index in the deque.

2. For the rest of the elements (from index `k` to end):
   - If the deque is not empty, the element at the front is the first negative in the current window.
   - If it is empty, that means there are no negative numbers in the current window → return 0.

3. Before moving the window forward:
   - Remove elements from the front of the deque that are out of the current window's range.
   - Add the current element to the deque **if it is negative**.

4. After the loop ends, handle the last window separately.

**Time and Space Complexity:**
- **Time Complexity**: O(n)  
  Each element is processed at most twice (once added, once removed).
  
- **Space Complexity**: O(k)  
  For storing indices in the deque.

**Java Code:**
```java
class Solution {
    static List<Integer> firstNegInt(int arr[], int k) {
        Deque<Integer> dq = new LinkedList<>();
        List<Integer> res = new ArrayList<>();
        int n = arr.length;

        // Process the first window
        for (int i = 0; i < k; i++) {
            if (arr[i] < 0) dq.addLast(i);
        }

        // Process the rest of the windows
        for (int i = k; i < n; i++) {
            // Add result for the previous window
            if (!dq.isEmpty()) res.add(arr[dq.peekFirst()]);
            else res.add(0);

            // Remove elements out of the current window
            while (!dq.isEmpty() && dq.peekFirst() <= i - k) {
                dq.pollFirst();
            }

            // Add current element if it is negative
            if (arr[i] < 0) dq.addLast(i);
        }

        // Add result for the last window
        if (!dq.isEmpty()) res.add(arr[dq.peekFirst()]);
        else res.add(0);

        return res;
    }
}


## Problem Statement: 193
**Design a Least Recently Used (LRU) Cache**  

Implement a data structure that simulates the behavior of an LRU (Least Recently Used) Cache. The cache has a fixed capacity, and supports the following operations:

1. **GET x**: Return the value associated with key `x` if present, otherwise return `-1`.
2. **PUT x y**: Insert or update the value for key `x` to `y`. If the cache exceeds its capacity, evict the least recently used item.

**Input:**  
- An integer `cap` denoting the **capacity** of the cache.  
- A sequence of `Q` queries, where each query is either `GET x` or `PUT x y`.

**Output:**  
- For every `GET` query, output the result of the operation (the value or `-1`).

**Constraints:**  
- `1 ≤ cap ≤ 10^4`  
- `1 ≤ Q ≤ 10^5`  
- `0 ≤ x, y ≤ 10^5`

**Goal of the Code:**  
Efficiently implement the LRU cache that handles large volumes of queries in near constant time using appropriate data structures.

**Approach**
We use a combination of a **HashMap** and a **Doubly Linked List** (using Java's `LinkedList`) to implement the LRU cache efficiently:

**Key Concepts:**
- **HashMap** stores the key-value pairs for O(1) access.
- **LinkedList** tracks the usage order of keys with the most recently used at the front.

**How Operations Work:**
- **GET key**:
  - If key exists, move it to the front of the LRU list and return its value.
  - If key doesn't exist, return -1.
- **PUT key value**:
  - If key already exists, update its value and move it to the front.
  - If key doesn't exist:
    - If capacity is full, remove the last item from the list and map.
    - Insert the new key-value and add to the front of the list.

**Time and Space Complexity**
- **Time Complexity:** O(1) per operation (on average).
- **Space Complexity:** O(cap), where cap is the cache capacity.

**Java Solution**
```java
class LRUCache {
    private int cap;
    private Map<Integer, Integer> cacheMap;
    private LinkedList<Integer> lruList;

    LRUCache(int cap) {
        this.cap = cap;
        this.cacheMap = new HashMap<>();
        this.lruList = new LinkedList<>();
    }

    public int get(int key) {
        if (!cacheMap.containsKey(key)) {
            return -1;
        }

        lruList.remove(Integer.valueOf(key));
        lruList.addFirst(key);

        return cacheMap.get(key);
    }

    public void put(int key, int value) {
        if (cacheMap.containsKey(key)) {
            cacheMap.put(key, value);
            lruList.remove(Integer.valueOf(key));
        } else {
            if (cacheMap.size() >= cap) {
                int lruKey = lruList.removeLast();
                cacheMap.remove(lruKey);
            }
            cacheMap.put(key, value);
        }
        lruList.addFirst(key);
    }
}
```

To efficiently find the **minimum window** in `s1` where `s2` appears as a **subsequence**, using **dynamic programming**.

**Approach**
We use **dynamic programming** to track the earliest starting index of a valid window that ends at position `i` in `s1` and contains `s2` as a subsequence.

- Let `dp[i][j]` be the start index of the window in `s1` that ends at index `i` and contains the first `j` characters of `s2` as a subsequence.

3. Fill the DP table using the recurrence relation.
4. Scan the last column `dp[i][m]` to find the shortest window.
5. Return substring starting from `dp[i][m] - 1` with minimum length.

- **Space Complexity:** `O(n * m)` for the DP table.

**Java Solution**
```java
class Solution {
    static String minWindow(String s1, String s2) {
        int n = s1.length(), m = s2.length();
        if (n < m) return "";

        int[][] dp = new int[n + 1][m + 1];

        for (int i = 0; i <= n; i++) dp[i][0] = i + 1;
        for (int j = 1; j <= m; j++) dp[0][j] = 0;

        for (int i = 1; i <= n; i++) {
            for (int j = 1; j <= m; j++) {
                if (s1.charAt(i - 1) == s2.charAt(j - 1)) {
                    dp[i][j] = dp[i - 1][j - 1];
                } else {
                    dp[i][j] = dp[i - 1][j];
                }
            }
        }

        int startIdx = -1, minLen = Integer.MAX_VALUE;

        for (int i = 1; i <= n; i++) {
            if (dp[i][m] > 0) {
                int start = dp[i][m] - 1;
                int len = i - start;
                if (len < minLen) {
                    minLen = len;
                    startIdx = start;
                }
            }
        }

        return startIdx == -1 ? "" : s1.substring(startIdx, startIdx + minLen);
    }
}
```

## Problem Statement: 195
**Find Starting Indices of Substrings Formed by Concatenating Words from List**

You are given a string `S` and a list of words `L` (array/vector of strings). Each word in list `L` is of the **same length**.  
Your task is to **find all starting indices** of substrings in string `S` that are **concatenations of all words in list `L`**, appearing **in any order** and **without any intervening characters**.

**Input:**  
- `S = "barfoothefoobarman"`  
- `L = ["foo", "bar"]`  

**Output:**  
- `[0, 9]`  

Explanation: Substrings `"barfoo"` (starts at index 0) and `"foobar"` (starts at index 9) contain all words from `L`.

**✅ Constraints:**
- All words in list `L` are of the same length.
- Words can repeat in `L`.
- Both `S` and `L` consist of lowercase English letters only.
- Return indices in ascending order.

**Goal of the Code:**
To find **all starting positions** in the string `S` where a **concatenation of each word in list `L`** (exactly once per word, in any order) occurs.

**🧠 Approach**
This is a **sliding window with hash map frequency counting** problem.

**Step-by-Step Strategy:**
1. **Understand total word length:**
   - If each word is of length `w` and there are `n` words, the window size = `w * n`.

2. **Use a HashMap** to store the **expected count of each word**.

3. **Slide a window** of total size `w * n` across the string `S`.

4. In each window:
   - Break the substring into chunks of size `w`.
   - Count frequencies of each word in the current window.
   - Compare with the expected frequency.
   - If all match, store the start index.

**📋 Pseudocode**
1. Create a HashMap to store frequency of each word in list L.
2. Loop through S from 0 to length(S) - totalConcatLength: a. Create a temporary map. b. Extract substrings of word length. c. Compare counts with the original map. d. If all match, store current index.
3. Return all matching indices.

**⏱️ Time and Space Complexity**
- **Time Complexity:** `O(N * M)`  
  Where `N` is the length of string `S`, and `M` is number of words in `L`.

- **Space Complexity:** `O(M)`  
  For storing word frequency maps.

**🧑‍💻 Java Solution**
```java
import java.util.ArrayList;
import java.util.Arrays;
import java.util.HashMap;
import java.util.List;

class GFG {
    public static ArrayList<Integer> findSubstring(String S, final List<String> L) {
        int wordLen = L.get(0).length();
        int wordCount = L.size();
        int totalLen = wordLen * wordCount;
        int n = S.length();

        ArrayList<Integer> result = new ArrayList<>();
        if (totalLen > n) return result;

        HashMap<String, Integer> wordFreq = new HashMap<>();
        for (String word : L) {
            wordFreq.put(word, wordFreq.getOrDefault(word, 0) + 1);
        }

        for (int i = 0; i <= n - totalLen; i++) {
            HashMap<String, Integer> seen = new HashMap<>(wordFreq);
            int j = i, count = wordCount;

            while (j < i + totalLen) {
                String sub = S.substring(j, j + wordLen);
                if (!seen.containsKey(sub) || seen.get(sub) == 0) break;
                seen.put(sub, seen.get(sub) - 1);
                count--;
                j += wordLen;
            }

            if (count == 0) result.add(i);
        }

        return result;
    }

    public static void main(String[] args) {
        String S = "barfoothefoobarman";
        ArrayList<String> L = new ArrayList<>(Arrays.asList("foo", "bar"));
        ArrayList<Integer> indices = findSubstring(S, L);
        for (Integer i : indices) {
            System.out.println(i);
        }
    }
}
