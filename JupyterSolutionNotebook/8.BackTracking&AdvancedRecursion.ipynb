## Problem: Permutations of a String

Given a string `s`, which may contain duplicate characters, your task is to generate and return an array of all unique permutations of the string. You can return your answer in any order.

**Example :**

```text
Input: s = "aab"
Output: ["aab", "aba", "baa"]
```

**Constraints:**
1. The input string s will have a length of 1 <= s.length <= 8.
2. The string s will consist of only lowercase English letters.

**Goal:**
Generate all unique permutations of the string `s`.

**🧮 Flowchart/ Diagram:**

* **Backtracking** approach can be visualized by a tree structure where each node explores the next character in the string, and branches expand for each character.

**Pseudocode/Algorithm:**

```java
function genPermutations(n, curr, cnt, res):
    if length of curr == n:
        add curr.toString() to res
        return

    for each character in cnt:
        if count of character is 0:
            continue

        append character to curr
        decrease count of character in cnt
        recursively call genPermutations

        remove last character from curr (backtrack)
        restore count of character in cnt

function findPermutation(s):
    initialize result list res
    initialize a map cnt for counting frequencies of characters
    for each character in s, update cnt
    initialize an empty StringBuilder curr
    call genPermutations(s.length(), curr, cnt, res)
    return res
```

**Approach:**
We need to generate all permutations of the given string `s` and return only the unique permutations.

1. **Backtracking Approach:**

   * Use backtracking to explore all possible permutations of the string.
   * Use a hashmap to count the occurrences of each character in the string to handle duplicates efficiently.
   * Recursively build the permutations, and once a complete permutation is formed, add it to the result list.

2. **Avoid Duplicates:**

   * By using a hashmap, we can keep track of the number of available characters for each permutation.
   * This ensures that duplicate permutations are not generated.

3. **Backtracking Function:**

   * The backtracking function appends a character to the current permutation, updates the hashmap, and recursively generates further permutations.
   * After generating permutations, the character is removed (backtracked) and the hashmap is updated.

**Code Solution:**
```java
import java.util.*;

class Solution {
    static void genPermutations(int n, StringBuilder curr, 
                                 Map<Character, Integer> cnt, List<String> res) {
      
        // If the current permutation is of the required length, add it to the result list
        if (curr.length() == n) {
            res.add(curr.toString());
            return;
        }

        // Explore all possible characters from the map
        for (Map.Entry<Character, Integer> entry : cnt.entrySet()) {
            char c = entry.getKey();
            int count = entry.getValue();

            // Skip this character if there are no more of it left to use
            if (count == 0) 
                continue;

            // Add the character to the current permutation
            curr.append(c);

            // Decrease the count of this character in the map
            cnt.put(c, count - 1);

            // Recursively generate permutations
            genPermutations(n, curr, cnt, res);

            // Backtrack: undo the changes
            curr.deleteCharAt(curr.length() - 1);
            cnt.put(c, count);
        }
    }

    public ArrayList<String> findPermutation(String s) {
        // To store the result
        ArrayList<String> res = new ArrayList<>(); 

        // Count the frequency of each character in the string
        Map<Character, Integer> cnt = new HashMap<>(); 

        for (char c : s.toCharArray()) 
            cnt.put(c, cnt.getOrDefault(c, 0) + 1);

        // StringBuilder to build permutations
        StringBuilder curr = new StringBuilder();
        
        // Start the backtracking function
        genPermutations(s.length(), curr, cnt, res);
        
        return res;
    }
}
```

**Time Complexity:**
The time complexity is approximately O(N!) where N is the length of the string. This is because there are N! possible permutations of the string. However, we reduce the complexity by eliminating duplicates using the cnt map.

**Space Complexity:**
The space complexity is O(N) for storing the string and the result list, where N is the length of the string.

**Notes:**

* The use of a hashmap ensures that duplicate characters are efficiently handled.
* Backtracking is used to explore all possible combinations of characters without repeating permutations.


## Problem: Permutations of Given String

Given a string `s`, the task is to return all permutations of a given string in lexicographically sorted order. A permutation is the rearrangement of all the elements of a string. Duplicate arrangements can exist, so you need to ensure that the result only includes unique permutations.

**Example:**

**Input:**
```
s = "ABC"
```

**Output:**
```
\["ABC", "ACB", "BAC", "BCA", "CAB", "CBA"]
```

**Constraints:**
- The length of the string `s` will be at most 10 characters.
- The string contains only uppercase English letters.

**Goal:**
The goal is to generate all the unique permutations of the string and return them in lexicographically sorted order.

**Approach:**

1. **Recursive Permutation Generation:**
   - A recursive approach is used to generate all permutations of the string. Starting from the first index, swap the character at the current index with every other character in the string.
   - After the swap, the function calls itself recursively to handle the next index.
   - Once the function reaches the base case (index equals the length of the string), the current permutation is added to the result.

2. **Backtracking:**
   - After each recursive call, we backtrack by swapping the characters back to their original position. This allows the next recursive call to explore different combinations.
   
3. **Sorting the Results:**
   - After generating all the permutations, the list is sorted lexicographically before returning.

**🧮 Flowchart/ Diagram:**

```

Start
|
\|--> Call recurPermute with index 0
|
\|--> Swap characters and recursively permute
|
\|--> Base case (index == s.length()): Add permutation to result
|
\|--> Backtrack (swap back)
|
\|--> Sort the result lexicographically
|
\|--> Return the result
|
End

```

**Pseudocode/Algorithm:**
```text
function findPermutation(s):
    ans = []
    str = StringBuilder(s)
    recurPermute(0, str, ans)
    sort(ans)
    return ans

function recurPermute(index, s, ans):
    if index == s.length():
        ans.add(s.toString())
        return
    for i = index to s.length() - 1:
        swap(s, index, i)
        recurPermute(index + 1, s, ans)
        swap(s, index, i)

function swap(s, i, j):
    temp = s.charAt(i)
    s.setCharAt(i, s.charAt(j))
    s.setCharAt(j, temp)
```

**Code:**

```java
import java.util.*;

class GfG {

    // Recursive function to generate 
    // all permutations of string s
    static void recurPermute(int index, StringBuilder s, 
                            List<String> ans) {

        // Base Case
        if (index == s.length()) {
            ans.add(s.toString());
            return;
        }

        // Swap the current index with all
        // possible indices and recur
        for (int i = index; i < s.length(); i++) {
            swap(s, index, i);
            recurPermute(index + 1, s, ans);
            swap(s, index, i);
        }
    }

    // Swap characters at positions i and j
    static void swap(StringBuilder s, int i, int j) {
        char temp = s.charAt(i);
        s.setCharAt(i, s.charAt(j));
        s.setCharAt(j, temp);
    }

    // Function to find all unique permutations
    static List<String> findPermutation(String s) {

        // Stores the final answer
        List<String> ans = new ArrayList<>();
        StringBuilder str = new StringBuilder(s);

        recurPermute(0, str, ans);

        // sort the resultant list
        Collections.sort(ans);

        return ans;
    }

    public static void main(String[] args) {
        String s = "ABC";
        List<String> res = findPermutation(s);
        for (String x : res) {
            System.out.print(x + " ");
        }
    }
}
```

**Time Complexity:**

* **O(n! \* n):** Generating all permutations takes `O(n!)` time. Each permutation takes `O(n)` time to generate and store, where `n` is the length of the string.

**Space Complexity:**

* **O(n!):** The space complexity is determined by the number of unique permutations stored in the list, which is `O(n!)`.

**Notes:**

* This approach uses backtracking and recursion to generate all permutations, ensuring that all possible combinations are explored.
* The sorting step ensures that the permutations are returned in lexicographically sorted order.


## Problem: Permutations II
Given a collection of numbers, `nums`, that might contain duplicates, return all possible unique permutations in any order.

**Input:**
```text
nums = [1,1,2]
```

**Output:**

```text
[[1,1,2], [1,2,1], [2,1,1]]
```

**Example :**
```text
Input: nums = [1,1,2]
Output: [[1,1,2], [1,2,1], [2,1,1]]
```

**Constraints:**
1. 1 <= nums.length <= 8
2. -10^9 <= nums[i] <= 10^9

**Approach:**
To solve this problem, we need to generate all permutations of the given list while ensuring uniqueness by avoiding duplicate permutations due to repeated elements.

**We will use a backtracking approach with the following steps:**

1. **Sort the input array:**
    Sorting helps to easily skip duplicate numbers while generating permutations.

2. **Backtracking:**
    For each number, we try to add it to the current permutation.

    We use a boolean used[] array to track whether a number has been used in the current permutation.

3. **Skip duplicate numbers:**
    If a number is the same as the previous one and the previous number has not been used, it indicates that we are about to create a duplicate permutation, so we skip it.

4. **Base Case:**
    When the length of the current permutation equals the length of the input array, we have generated a valid permutation, and we add it to the result list.
**Goal:**
Generate all unique permutations of an integer array that may contain duplicates.

**🧮 Flowchart/ Diagram:**
A decision tree can be used to visualize the permutation process. At each level of the tree, the next unused number is selected. Duplicates are avoided by ensuring that we skip a number if it's the same as the previous one and the previous hasn't been used.

**Pseudocode/Algorithm:**

```text
function permuteUnique(nums):
    sort(nums)
    used = [false] * len(nums)
    result = []
    backtrack(current=[], used, nums, result)
    return result

function backtrack(current, used, nums, result):
    if length(current) == length(nums):
        result.add(copy of current)
        return

    for i in range 0 to length(nums):
        if used[i] or (i > 0 and nums[i] == nums[i-1] and not used[i-1]):
            continue
        used[i] = true
        current.add(nums[i])
        backtrack(current, used, nums, result)
        current.removeLast()
        used[i] = false
```

**Code Solution:**
```java
import java.util.*;

class Solution {
    public List<List<Integer>> permuteUnique(int[] nums) {
        List<List<Integer>> result = new ArrayList<>();
        
        // Sort the input to handle duplicates
        Arrays.sort(nums);
        
        // Array to track used elements
        boolean[] used = new boolean[nums.length];
        
        // Start backtracking
        backtrack(nums, used, new ArrayList<>(), result);
        
        return result;
    }

    private void backtrack(int[] nums, boolean[] used, List<Integer> current, List<List<Integer>> result) {
        // If the current permutation is of the same length as nums, add to result
        if (current.size() == nums.length) {
            result.add(new ArrayList<>(current));
            return;
        }
        
        // Iterate through the array to build permutations
        for (int i = 0; i < nums.length; i++) {
            
            // Skip the element if it is already used or if it's a duplicate and hasn't been used in the current permutation
            if (used[i] || (i > 0 && nums[i] == nums[i - 1] && !used[i - 1])) {
                continue;
            }
            
            // Mark the element as used
            used[i] = true;
            
            // Add the element to the current permutation
            current.add(nums[i]);
            
            // Recursively build the rest of the permutation
            backtrack(nums, used, current, result);
            
            // Backtrack: unmark the element and remove it from the current permutation
            used[i] = false;
            current.remove(current.size() - 1);
        }
    }
}
```

**Time Complexity: O(N! * N)**
    In the worst case, we generate all permutations of the input array, which takes O(N!) time. For each permutation, we perform additional work (e.g., checking for duplicates) that takes O(N) time. Thus, the total time complexity is O(N! * N).

**Space Complexity: O(N)**
    The space complexity is dominated by the recursive call stack and the space needed to store the result list. The recursive stack depth is at most O(N) since we are building permutations of size N, and the result list will store O(N!) permutations of length N.


**Notes:**

* Sorting the array before backtracking is key to efficiently avoiding duplicate permutations.
* The `used[]` array ensures each index is only used once per permutation.
* Skipping duplicates with `nums[i] == nums[i - 1] && !used[i - 1]` avoids branching into permutations that are already covered.


## Problem: Distinct Permutations of the String | Set 2

Given a string `s`, the task is to return all **unique permutations** of a given string in **lexicographically sorted order**.

**Note:**
A permutation is the rearrangement of all the elements of a string. Duplicate arrangements can exist in a string, so we need to ensure that only unique permutations are included in the result.

**Example:**

**Input:**
```
s = "ABC"
```

**Output:**
```
\["ABC", "ACB", "BAC", "BCA", "CAB", "CBA"]
```

**Constraints:**
- The length of the string `s` will be at most 10 characters.
- The string contains only uppercase English letters.

**Goal:**
The goal is to generate all the **unique permutations** of the string and return them in **lexicographically sorted order**.

**Approach:**

1. **Sorting the String:**
   - First, the string is sorted to ensure that the permutations are generated in lexicographical order.

2. **Recursive Permutation Generation:**
   - A recursive approach is used to generate all permutations of the string. The function swaps characters at the current index with every other character and recursively permutes the remaining characters.
   - This ensures that all permutations of the string are generated.

3. **Handling Unique Permutations:**
   - We use a `Set` to store the permutations, ensuring that only unique permutations are retained (duplicate permutations are automatically discarded by the `Set`).

4. **Sorting the Result:**
   - The result is stored in a `TreeSet`, which naturally sorts the permutations in lexicographical order.

**🧮 Flowchart/ Diagram:**

```
Start
|
\|--> Sort input string
|
\|--> Call recurPermute with index 0 and an empty set
|
\|--> Swap characters and recursively permute
|
\|--> Base case (index == s.length()): Add permutation to result
|
\|--> Backtrack (swap back)
|
\|--> Convert set to a list
|
\|--> Return the result
|
End
```

**Pseudocode/Algorithm:**

```text
function findPermutation(s):
    sort s lexicographically
    ans = []  // Stores the final answer
    res = Set()  // Stores all unique permutations
    recurPermute(0, s, res)
    ans.addAll(res)  // Convert the set to a list
    return ans

function recurPermute(index, s, res):
    if index == s.length():
        res.add(s.toString())
        return
    for i = index to s.length() - 1:
        swap(s, index, i)
        recurPermute(index + 1, s, res)
        swap(s, index, i)  // Backtrack

function swap(s, i, j):
    temp = s[i]
    s[i] = s[j]
    s[j] = temp
```
**Code:**

```java
import java.util.*;

class GfG {

    // Recursive function to generate 
    // all permutations of string s
    static void recurPermute(int index, StringBuilder s, 
                                     Set<String> ans) {

        // Base Case
        if (index == s.length()) {
            ans.add(s.toString());
            return;
        }

        // Swap the current index with all
        // possible indices and recur
        for (int i = index; i < s.length(); i++) {
            char temp = s.charAt(index);
            s.setCharAt(index, s.charAt(i));
            s.setCharAt(i, temp);

            recurPermute(index + 1, s, ans);

            temp = s.charAt(index);
            s.setCharAt(index, s.charAt(i));
            s.setCharAt(i, temp);
        }
    }

    // Function to find all unique permutations
    static List<String> findPermutation(String s) {

        // sort input string
        char[] sorted = s.toCharArray();
        Arrays.sort(sorted);
        s = new String(sorted);

        // Stores the final answer
        List<String> ans = new ArrayList<>();

        // Stores all unique permutations
        Set<String> res = new TreeSet<>();
        recurPermute(0, new StringBuilder(s), res);

        // Copy all unique permutations
        // to the final answer
        ans.addAll(res);
        return ans;
    }

    public static void main(String[] args) {
        String s = "ABC";
        List<String> res = findPermutation(s);
        for (String x : res) {
            System.out.print(x + " ");
        }
    }
}
```

**Time Complexity:**

* **O(n! \* n):** Generating all permutations takes `O(n!)` time. Each permutation requires `O(n)` time to generate and store, where `n` is the length of the string.

**Space Complexity:**

* **O(n!):** The space complexity is determined by the number of unique permutations stored in the set, which is `O(n!)`.

**Notes:**

* The solution uses recursion and backtracking to generate permutations, ensuring that all possible combinations are explored.
* The use of a `Set` guarantees that only unique permutations are included in the final result.
* Sorting the input string at the start ensures that permutations are generated in lexicographical order, and using a `TreeSet` maintains this order in the result.

## Problem Statement: Combination Sum
Given an array `arr[]` and a target number `target`,  
find all **unique combinations** in the array where the sum is equal to the target.

- **Important:** 
  - The same number from the array can be chosen multiple times.
  - The combinations can be returned in any order.

**Input:**
```text
arr = [2, 3, 6, 7], target = 7
```

**Output:**

```text
[[2, 2, 3], [7]]
```

**Objective:**
Write a function to return all unique combinations where chosen numbers sum up to the given `target`.

**Goal:**
Find and return all unique combinations of numbers from the input array that sum up exactly to the given target. Each number in the array can be chosen multiple times.

**Approach:**
1. **Backtracking:**
   - Try to pick an element and stay at the same index (because we can pick the same element again).
   - Explore both:
     - Include the element.
     - Exclude the element and move to the next one.
2. **Base Cases:**
   - If the remaining sum becomes `0`, save the current combination.
   - If the remaining sum becomes negative or array ends, backtrack.

3. **Sorting:**
   - (Optional) Sort the array for consistency but not mandatory for correctness.

**🧮 Flowchart/ Diagram:**

* Start with index 0 and an empty list.
* For each element, make a recursive choice to either:

  * Include it (and keep the index the same),
  * Or exclude it (and move to the next index).
* Stop recursion if the remaining target is zero (valid combination) or becomes negative (invalid path).

**Pseudocode/Algorithm:**

```text
function combinationSum(arr, target):
    sort(arr)
    res = []
    cur = []
    backtrack(index=0, remSum=target, cur, res)

function backtrack(index, remSum, cur, res):
    if remSum == 0:
        res.add(copy of cur)
        return
    if index >= length(arr) or remSum < 0:
        return

    // Include arr[index]
    cur.add(arr[index])
    backtrack(index, remSum - arr[index], cur, res)
    cur.removeLast()
 // Exclude arr[index]
    backtrack(index + 1, remSum, cur, res)
```

**Solution Code:**
```java
import java.util.ArrayList;
import java.util.Arrays;

class Solution {

    // Function to find all combinations of elements
    // in array arr that sum to target.
    static void makeCombination(int[] arr, int remSum, ArrayList<Integer> cur, 
                                       ArrayList<ArrayList<Integer>> res, int index) {
        if (remSum == 0) {
            res.add(new ArrayList<>(cur)); // Found a valid combination
            return;
        }
        if (remSum < 0 || index >= arr.length) {
            return; // No valid combination
        }
        
        // Choose the element at current index
        cur.add(arr[index]);
        makeCombination(arr, remSum - arr[index], cur, res, index); // Stay at current index
        cur.remove(cur.size() - 1); // Backtrack
        
        // Do not choose current index element, move to next index
        makeCombination(arr, remSum, cur, res, index + 1);
    }

    static ArrayList<ArrayList<Integer>> combinationSum(int[] arr, int target) {
        Arrays.sort(arr); // Sort for consistent order
        ArrayList<Integer> cur = new ArrayList<>();
        ArrayList<ArrayList<Integer>> res = new ArrayList<>();
        makeCombination(arr, target, cur, res, 0);

        return res;
    }
}
```
**Time Complexity:**
- **O(2^T)**, where `T` is the target value (in the worst case).
- Every element could either be included or excluded multiple times.

**Space Complexity:**
- **O(target/ min number)** for recursion stack depth.

**Notes:**

* Because repetition of elements is allowed, after choosing an element, we stay on the same index in the recursive call.
* Sorting helps organize the output consistently but is not required for correctness.


## Problem: Combination Sum II
Given a collection of candidate numbers (`candidates`) and a target number (`target`), find all unique combinations in `candidates` where the candidate numbers sum to the target.
Each number in `candidates` may only be used once in the combination.

**Note:** The solution set must not contain duplicate combinations.

**Example :**
```text
Input: candidates = [10,1,2,7,6,5], target = 8
Output: [[1,2,5], [1,7], [2,6]]
```

**Input:**
```text
candidates = [10,1,2,7,6,1,5], target = 8
```
**Output:**

```text
[[1,1,6], [1,2,5], [1,7], [2,6]]
```

**Constraints:**
1. 1 <= candidates.length <= 100
2. 1 <= candidates[i] <= 50
3. 1 <= target <= 30

**Goal:**
Return all unique combinations of numbers from the input array that sum up to the given target. Each element can be used only once in each combination. Avoid returning duplicate combinations.

**Approach:**
We need to find all unique combinations of numbers that sum to the given target. The challenge is to avoid duplicate combinations. We can achieve this by sorting the input array and using backtracking.

**Here are the steps to solve the problem:**

1. **Sort the Array:**
    Sorting helps in efficiently skipping duplicate candidates. It ensures that we process the numbers in an ordered fashion.

2. **Backtracking:**
    We will recursively explore each candidate number and decide whether to include it in the current combination or not.

    Once we include a number in the combination, we reduce the target and move forward to the next candidate.

    If a number is the same as the previous one and hasn’t been used in the current recursion stack, we skip it to avoid duplicates.

3. **Base Case:**
    If the target becomes zero, it means we have found a valid combination, and we add it to the result.

4. **Pruning:**
    If the current candidate is greater than the target, we stop exploring further down that path, as the combination would exceed the target.
**🧮 Flowchart/ Diagram:**

* Sort the input array.
* Use a recursive backtracking function starting from index 0.
* At each step:

  * Skip the element if it’s the same as the previous and on the same recursion level.
  * If the element is greater than the remaining target, stop.
  * If target becomes 0, store the current combination.
  * Otherwise, recurse to the next index.

**Pseudocode/Algorithm:**

```text
function combinationSum2(candidates, target):
    sort(candidates)
    result = []
    backtrack(start=0, target, path=[], result)
    return result

function backtrack(start, target, path, result):
    if target == 0:
        result.add(copy of path)
        return

    for i from start to length of candidates:
        if i > start and candidates[i] == candidates[i-1]:
            continue
        if candidates[i] > target:
            break
        path.add(candidates[i])
        backtrack(i+1, target - candidates[i], path, result)
        path.removeLast()
```

**Code Solution:**
```java
import java.util.*;

class Solution {
    public List<List<Integer>> combinationSum2(int[] candidates, int target) {
        List<List<Integer>> result = new ArrayList<>();
        
        // Sort the array to handle duplicates
        Arrays.sort(candidates);
        
        // Start backtracking to find combinations
        backtrack(candidates, target, 0, new ArrayList<>(), result);
        
        return result;
    }

    private void backtrack(int[] candidates, int target, int start, List<Integer> current, List<List<Integer>> result) {
        // If target is reduced to zero, add the current combination to result
        if (target == 0) {
            result.add(new ArrayList<>(current));
            return;
        }
        
        // Iterate through the candidates
        for (int i = start; i < candidates.length; i++) {
            
            // Skip duplicates
            if (i > start && candidates[i] == candidates[i - 1]) {
                continue;
            }
            
            // If the candidate exceeds the target, break the loop
            if (candidates[i] > target) {
                break;
            }
            
            // Choose the current candidate and explore further
            current.add(candidates[i]);
            // Recur with updated target and next index to ensure each element is used once
            backtrack(candidates, target - candidates[i], i + 1, current, result);
            
            // Backtrack: Remove the last element to explore other combinations
            current.remove(current.size() - 1);
        }
    }
}
```


**Time Complexity: O(2^N)**
In the worst case, the algorithm explores all combinations of candidates. The number of combinations can be up to 2^N where N is the number of candidates. Each combination is processed in linear time.

**Space Complexity: O(N)**
The space complexity is dominated by the recursion stack and the space used to store the result list. The recursion depth can go up to O(N), and the result list stores all valid combinations, which can take up to O(2^N) space in the worst case.

**Notes:**

* Sorting is crucial to allow skipping duplicates and pruning large candidates early.
* Unlike Combination Sum I, we must move to the next index in recursion to ensure elements are not reused.


## Problem: Combination Sum III

Find all valid combinations of `k` numbers that sum up to `n` such that the following conditions are true:
- Only numbers `1` through `9` are used.
- Each number is used at most once.



**Constraints:**
1. 1 <= k <= 9
2. 1 <= n <= 45

**Goal:**
Generate all unique combinations of `k` different numbers from the range 1 to 9 such that the sum equals `n`.

**Approach:**
We need to find all possible combinations of k numbers such that their sum equals n. The challenge is to ensure that each number from 1 through 9 can be used only once, and the same combination should not appear twice in the result.

**Here are the steps to solve the problem:**
1. **Backtracking:**
    Use backtracking to explore all possible combinations of numbers that sum up to n. At each step, we try to include a number from 1 to 9 and recursively check if we can form a valid combination by subtracting the chosen number from n.

2. **Base Case:**
    If the size of the current combination is equal to k and the sum is zero (n == 0), it means we have found a valid combination, and we add it to the result list.

3. **Pruning:**
    If the size of the current combination exceeds k or if the remaining sum n is negative, we prune that path and backtrack.

4. **Recursive Exploration:**
    For each number from the current start value to 9, we explore the possibility of including it in the combination and then move forward to the next number.


**🧮 Flowchart/ Diagram:**

* Start with number 1 and recursively explore including each number up to 9.
* Track the current combination and the remaining target sum.
* If the combination length is `k` and the remaining sum is `0`, it's valid.
* Backtrack to explore alternative paths.

**Pseudocode/Algorithm:**

```text

    result = []
    backtrack(start=1, current=[], target=n, k)
    return result

function backtrack(start, current, target, k):
    if length of current == k and target == 0:
        add copy of current to result
        return
    if length of current > k or target < 0:
        return
    for i from start to 9:
        add i to current
        backtrack(i + 1, current, target - i, k)
        remove last element from current
```

**Code Solution:**

```java
import java.util.*;

class Solution {
    public List<List<Integer>> combinationSum3(int k, int n) {
        List<List<Integer>> result = new ArrayList<>();
        
        // Start backtracking from number 1
        backtrack(k, n, 1, new ArrayList<>(), result);
        
        return result;
    }

    private void backtrack(int k, int n, int start, List<Integer> current, List<List<Integer>> result) {
        // If the combination size is k and the sum is n, it's a valid combination
        if (current.size() == k && n == 0) {
            result.add(new ArrayList<>(current));
            return;
        }
        
        // If the combination size exceeds k or the sum is less than 0, return
        if (current.size() > k || n < 0) {
            return;
        }
        
        // Explore numbers from 'start' to 9
        for (int i = start; i <= 9; i++) {
            current.add(i); // Add current number to the combination
            // Recur with the next number and updated target sum
            backtrack(k, n - i, i + 1, current, result);
            current.remove(current.size() - 1); // Backtrack
        }
    }
}
```
**Time Complexity: O(2^N)**
In the worst case, we need to explore all possible subsets of the set {1, 2, 3, ..., 9}. Since we are looking for combinations, the number of potential subsets can be up to 2^9 due to the recursive branching.

**Space Complexity: O(N)**
The space complexity is dominated by the recursion stack and the space required to store the result list. In the worst case, the recursion depth can go up to N (where N is the length of the combination), and the result can store up to C(9, k) combinations.

**Notes:**

* Since only digits 1 through 9 are allowed, we can safely limit the loop.
* Each number can only appear once, and the result must not contain duplicate sets.

## Problem: Rat in a Maze Problem - I

Consider a rat placed at position `(0, 0)` in an `n x n` square matrix `mat[][]`. The rat's goal is to reach the destination at position `(n-1, n-1)`. The rat can move in four possible directions:  
- 'U' (up)
- 'D' (down)
- 'L' (left)
- 'R' (right)

The matrix contains only two possible values:
- `0`: A blocked cell through which the rat cannot travel.
- `1`: A free cell that the rat can pass through.

Your task is to find all possible paths the rat can take to reach the destination, starting from `(0, 0)` and ending at `(n-1, n-1)`, under the condition that the rat cannot revisit any cell along the same path. Furthermore, the rat can only move to adjacent cells that are within the bounds of the matrix and not blocked.

If no path exists, return an empty list.

**Note:** Return the final result vector in lexicographically smallest order.

**Example :**

```text
Input: 
maze = [
    [1, 0, 0, 0],
    [1, 1, 0, 1],
    [0, 1, 0, 0],
    [1, 1, 1, 1]
]
Output: 
["DRRR", "DDRR"]
```
**Constraints:**
1. 1 <= n <= 10
2. maze[i][j] is either 0 (blocked) or 1 (free).

**Goal:**
Find all the paths from `(0, 0)` to `(n-1, n-1)` in the maze while avoiding blocked cells and ensuring no cell is visited twice along the path.

**Approach:**
We need to find all possible paths that the rat can take to reach the destination (n-1, n-1). The challenge involves traversing the maze while avoiding revisiting cells. The approach involves depth-first search (DFS), where we try all possible directions and backtrack if we encounter a blocked or already visited cell.

**Here are the steps to solve the problem:**
1. **Check Starting Point:**
    If the starting cell (0, 0) is blocked, we return an empty list as no path is possible.

2. **DFS Traversal:**
    Use DFS to explore all possible paths from the starting point (0, 0) to the destination (n-1, n-1).

    For each cell, try to move in all four directions (down, left, right, up), while ensuring the cell is not blocked and hasn't been visited before.

    If the destination (n-1, n-1) is reached, add the current path to the result.

3. **Backtracking:**
    If we cannot proceed further, backtrack to the previous cell and try a different direction.

4. **Lexicographical Sorting:**
    Since the final output needs to be sorted lexicographically, we sort the result before returning it.

**🧮 Flowchart/ Diagram:**

1. Start at `(0,0)`, explore the four possible directions (Down, Left, Right, Up).
2. Use DFS to recursively move towards the destination `(n-1, n-1)`.
3. If the destination is reached, save the current path.
4. If movement is not possible, backtrack to previous cells and try alternative directions.
5. Return the sorted list of valid paths.

**Pseudocode/Algorithm:**

```text
function ratInMaze(maze):
    if maze[0][0] == 0:
        return empty result
    initialize visited array
    call dfs(maze, 0, 0, "", visited)
    return lexicographically sorted result

function dfs(maze, row, col, path, visited):
    if (row, col) is the destination:
        add path to result
    mark (row, col) as visited
    for each direction (D, L, R, U):
        if move is safe:
            add direction to path and call dfs recursively
    backtrack: unmark (row, col) as visited
```

**Code Solution:**
```java
import java.util.*;

class Solution {
    // Function to find all possible paths
    public ArrayList<String> ratInMaze(int[][] maze) {
        int n = maze.length;
        ArrayList<String> result = new ArrayList<>();
        
        // If the starting point is blocked, return an empty result
        if (maze[0][0] == 0) return result;
        
        // Initialize visited array to keep track of visited cells
        boolean[][] visited = new boolean[n][n];
        
        // Start the DFS traversal
        dfs(maze, n, 0, 0, "", result, visited);
        
        // Sort the result in lexicographical order
        Collections.sort(result);
        
        return result;
    }
    
    private void dfs(int[][] maze, int n, int row, int col, String path, ArrayList<String> result, boolean[][] visited) {
        // If we reach the destination, add the path to the result
        if (row == n - 1 && col == n - 1) {
            result.add(path);
            return;
        }
        
        // Mark the current cell as visited
        visited[row][col] = true;
        
        // Explore the four possible directions (down, left, right, up)
        if (isSafe(maze, n, row + 1, col, visited)) {
            dfs(maze, n, row + 1, col, path + "D", result, visited);
        }
        if (isSafe(maze, n, row, col - 1, visited)) {
            dfs(maze, n, row, col - 1, path + "L", result, visited);
        }
        if (isSafe(maze, n, row, col + 1, visited)) {
            dfs(maze, n, row, col + 1, path + "R", result, visited);
        }
        if (isSafe(maze, n, row - 1, col, visited)) {
            dfs(maze, n, row - 1, col, path + "U", result, visited);
        }
        
        // Backtrack: Unmark the current cell as visited
        visited[row][col] = false;
    }
    
    // Helper function to check if the move is safe
    private boolean isSafe(int[][] maze, int n, int row, int col, boolean[][] visited) {
        return (row >= 0 && row < n && col >= 0 && col < n && maze[row][col] == 1 && !visited[row][col]);
    }
}
```

**Time Complexity: O(4^(N^2))**
In the worst case, we might explore every possible path in the maze, which could involve up to 4^(N^2) recursive calls. This comes from the fact that at each step, we have four possible directions to move in a maze of size N x N.

**Space Complexity: O(N^2)**
The space complexity is primarily driven by the recursive stack used in the DFS traversal and the space used to store the visited cells, both of which are proportional to the number of cells in the maze (N x N).

**Notes:**

* The `isSafe` function ensures that only valid and unvisited cells are considered for the next move.
* The result needs to be sorted lexicographically to meet the problem’s requirements.


## Problem Statement: Possible Words From Phone Digits

Given a keypad layout (like on old mobile phones) and an array `arr[]` containing digits,  
your task is to list all possible words (in any order) that can be generated by pressing numbers from the array.

- **Note:** Only digits `2-9` are considered valid (as 0 and 1 don't map to letters).

**Example:**
```text
Input: arr = [2, 3]
Output: ["ad", "ae", "af", "bd", "be", "bf", "cd", "ce", "cf"]
```

**Objective:**
Generate all combinations of letters corresponding to the digits provided. Each digit corresponds to multiple letters, and each combination is formed by selecting one letter from each digit’s corresponding string of characters.

**Approach:**
1. **Mapping:**
   - Each digit from `2` to `9` maps to a string of letters:
     - 2 -> "abc"
     - 3 -> "def"
     - 4 -> "ghi"
     - 5 -> "jkl"
     - 6 -> "mno"
     - 7 -> "pqrs"
     - 8 -> "tuv"
     - 9 -> "wxyz"

2. **Backtracking:**
   - Recursively build words by appending one character corresponding to each digit.
   - When the prefix length equals the number of digits, save the prefix as a complete word.

3. **Handling Invalid Digits:**
   - If the digit is less than 2 or greater than 9, simply skip.

**🧮 Flowchart/ Diagram:**

1. Begin with the first digit.
2. For each digit, recursively append all its possible letters to the current word.
3. Once the length of the word equals the length of the input digits, save the word as a valid combination.
4. Continue until all combinations are explored.

**Pseudocode/Algorithm:**

```text
function possibleWords(arr):
    map each digit to corresponding letters (padMap)
    start recursive helper function possibleWordsRec(arr, 0, "")
    return result with all possible combinations

function possibleWordsRec(arr, index, prefix):
    if index == arr.length:
        add prefix to result
    else:
        for each character corresponding to arr[index]:
            append character to prefix
            call possibleWordsRec with next index
            remove character from prefix (backtrack)
```

**Solution Code:**
```java
import java.util.ArrayList;

class Solution {
    public ArrayList<String> possibleWords(int[] arr) {
        ArrayList<String> res = new ArrayList<>();
        
        // Mapping of digits to corresponding letters
        String[] padMap = {"", "", "abc", "def", "ghi", "jkl",
                           "mno", "pqrs", "tuv", "wxyz"};
      
        StringBuilder prefix = new StringBuilder();

        // Recursive helper function to generate all possible words
        possibleWordsRec(arr, 0, prefix, padMap, res);
        return res;
    }

    static void possibleWordsRec(int[] arr, int index, StringBuilder prefix, 
                                         String[] padMap, ArrayList<String> res) {
      
        if (index == arr.length) {
            res.add(prefix.toString());
            return;
        }

        int digit = arr[index];

        // Ignore digits that don't map to letters (0 or 1)
        if (digit < 2 || digit > 9) {
            possibleWordsRec(arr, index + 1, prefix, padMap, res);
            return;
        }

        // Iterate over all characters corresponding to the current digit
        for (char ch : padMap[digit].toCharArray()) {
            prefix.append(ch);
            possibleWordsRec(arr, index + 1, prefix, padMap, res);
            prefix.deleteCharAt(prefix.length() - 1); // backtrack
        }
    }
}
```

**Time Complexity:**
- **O(4^n)** where `n` is the number of digits, since each digit can map up to 4 letters (for digits like 7 and 9).

**Space Complexity:**
- **O(n)** recursion stack + **O(output size)** for storing results.


**Notes:**

* The digits `0` and `1` are ignored as they do not map to any letters.
* The key idea is backtracking, where we explore each possibility at each step and then backtrack to try other combinations.

## Problem Statement: Subsets

Given an array `arr[]` of distinct positive integers, your task is to find all of its subsets.  
The subsets should be returned in **lexicographical order**.

**Example:**
```text
Input: arr = [1, 2, 3]
Output: [[], [1], [1, 2], [1, 2, 3], [1, 3], [2], [2, 3], [3]]
```

**Objective:**
- Generate **all possible subsets** of the given array.
- Return the list of subsets in **sorted lexicographical order**.

**Approach:**
1. **Recursive Backtracking:**
   - Start from index `0`.
   - At each index, you have two choices:
     - **Include** the current element in the subset.
     - **Exclude** the current element and move forward.

2. **Sorting:**
   - Initially sort the array to maintain order in subsets.
   - After generating all subsets, sort the list of subsets lexicographically.

3. **Lexicographical Order:**
   - First by elements in order.
   - Then by length (if prefix is same).

**🧮 Flowchart/ Diagram:**

1. Start at index 0.
2. For each index, recursively generate subsets by including or excluding the current element.
3. Add the resulting subset to the list of subsets.
4. After generating all subsets, sort them lexicographically.

**Pseudocode/Algorithm:**

```text
function subsets(arr):
    sort arr
    initialize result as empty list
    call subsetRecur(0, arr, result, empty subset)
    sort result lexicographically
    return result

function subsetRecur(i, arr, result, subset):
    if i == arr.length:
        add subset to result
    else:
        include arr[i] in subset
        call subsetRecur with next index
        exclude arr[i] from subset
        call subsetRecur with next index
```

**Solution Code:**
```java
import java.util.ArrayList;
import java.util.Arrays;

class Solution {
    static void subsetRecur(int i, int[] arr,
                            ArrayList<ArrayList<Integer>> res,
                            ArrayList<Integer> subset) {
        if (i == arr.length) {
            res.add(new ArrayList<>(subset));
            return;
        }
        
        // Include arr[i] in the current subset
        subset.add(arr[i]);
        subsetRecur(i + 1, arr, res, subset);
        
        // Exclude arr[i] from the current subset (backtrack)
        subset.remove(subset.size() - 1);
        subsetRecur(i + 1, arr, res, subset);
    }

    public ArrayList<ArrayList<Integer>> subsets(int arr[]) {
        Arrays.sort(arr); // Sort the array for lex order
        
        ArrayList<ArrayList<Integer>> res = new ArrayList<>();
        ArrayList<Integer> subset = new ArrayList<>();
        
        subsetRecur(0, arr, res, subset);

        // Sort all subsets lexicographically
        res.sort((a, b) -> {
            int minLength = Math.min(a.size(), b.size());
            for (int i = 0; i < minLength; i++) {
                int cmp = Integer.compare(a.get(i), b.get(i));
                if (cmp != 0)
                    return cmp;
            }
            return Integer.compare(a.size(), b.size());
        });

        return res;
    }
}
```

**Time Complexity:**
- **O(2^n)** for generating all subsets.
- **O(m log m)** for sorting all `m` subsets (where `m = 2^n`).

**Space Complexity:**
- **O(n)** recursion stack depth, plus space to store results.

**Notes:**

* Sorting the array initially ensures that the subsets are generated in lexicographical order.
* The recursion generates all subsets, and sorting the subsets at the end ensures they follow the lexicographical order.

Given an array `arr[]` of integers of size `N` that might contain duplicates, the task is to find all possible unique subsets of the array. Each subset should be sorted.

**Goals:**
- Generate **all possible unique subsets** of the given array, including the empty subset.

- Return the subsets in **lexicographical order**.

**Approach:**

1. **Sorting the Array:**

   * Sort the array to handle duplicates efficiently, making it easier to skip duplicates during the backtracking process.

2. **Backtracking Approach:**

   * Use backtracking to explore all possible subsets of the array.
   * For each element, recursively decide whether to include it in the current subset.
   * If an element is the same as the previous element and it is not at the start of the current recursion level, skip it to avoid duplicate subsets.

3. **Result Collection:**

   * Maintain a list to store all the unique subsets generated during the backtracking process.

4. **Handling Duplicates:**

   * Ensure duplicates are avoided by skipping elements that are identical to the previous element in the sorted array.


**🧮 Flowchart/ Diagram:**

1. **Sort** the array to ensure duplicates are adjacent.
2. Use **backtracking** to generate all subsets by including or excluding each element.
3. **Skip duplicates** during backtracking by checking if the current element is the same as the previous one.
4. Return the list of unique subsets.

**Pseudocode/Algorithm:**

```text
function AllSubsets(arr, n):
    sort arr
    initialize result as empty list
    call backtrack(arr, 0, empty subset, result)

function backtrack(arr, index, current, result):
    add current to result
    for i from index to arr.length:

        add arr[i] to current
```

**Code Solution:**

```java
import java.util.*;

class solve {
    // Function to find all possible unique subsets.
    public static ArrayList<ArrayList<Integer>> AllSubsets(int arr[], int n) {
        // Sorting the array to handle duplicates
        Arrays.sort(arr);
        
        // List to store all subsets
        ArrayList<ArrayList<Integer>> result = new ArrayList<>();
        
        // Backtracking to find all subsets
        backtrack(arr, 0, new ArrayList<>(), result);
        
        return result;
    }

    // Backtracking helper function
    private static void backtrack(int[] arr, int index, ArrayList<Integer> current, ArrayList<ArrayList<Integer>> result) {
        // Add the current subset to the result
        result.add(new ArrayList<>(current));
        
        // Iterate through the array and explore subsets
        for (int i = index; i < arr.length; i++) {
            // Skip duplicates to avoid generating the same subset
            if (i > index && arr[i] == arr[i - 1]) continue;

            current.add(arr[i]);
            
            // Recursively call backtrack for the next element
            backtrack(arr, i + 1, current, result);
            
            // Remove the last element to backtrack
            current.remove(current.size() - 1); 
        }
    }
```

## Problem: N-Queen Problem

The **N-Queens puzzle** is the problem of placing **n queens** on an **n × n** chessboard such that:
- No two queens attack each other.
- Queens attack if placed in the same row, column, or diagonal.
Given an integer `n`, find all distinct solutions to the N-Queens puzzle.
Each solution is represented as a permutation of `[1, 2, 3, ..., n]`:
- The number at the *ith* position denotes the **row** where the queen is placed in the *ith* column.

You can return the solutions in any order.

**Example**  
**Input:**  
```text
n = 4
````

**Output:**

```text
[
  [2, 4, 1, 3],
  [3, 1, 4, 2]
]
```

**Constraints:**

1. 1 <= n <= 10

**Objective:**
- Find **all distinct** board configurations for the given `n`.
- Represent the board as an array of integers.

**Approach:**

1. **Recursive Backtracking:**

   * Try placing a queen in every row for a given column.
   * Track used rows and diagonals to avoid conflicts.
   * If placing a queen is safe, move to the next column.

2. **Conflict Check:**

   * `rows[i]`: Row `i` already has a queen.
   * `diag1[i + j]`: Major diagonal conflict.
   * `diag2[i - j + n]`: Minor diagonal conflict.

3. **Backtracking:**

   * If the queen can be placed, mark the row and diagonals as occupied.
   * After the recursive call, backtrack by unmarking.

**🧮 Flowchart/ Diagram:**

* \[Diagram of recursive backtracking for N-Queens (Placeholder)]

**Pseudocode/Algorithm:**

```text
function nQueen(n):
    Initialize an empty result list
    Initialize an empty board
    Create boolean arrays to track row and diagonal conflicts

    function nQueenUtil(j):
        If j > n, add the current board to the result
        For each row i from 1 to n:
            If the position is safe (no conflicts):
                Place the queen and mark the row and diagonals
                Recurse to the next column
                Backtrack by removing the queen and restoring the row/diagonal

    Call nQueenUtil(1)
    Return the result
```

**Solution Code:**
```java
import java.util.ArrayList;

class Solution {
    static void nQueenUtil(int j, int n, ArrayList<Integer> board, boolean[] rows,
                           boolean[] diag1, boolean[] diag2, ArrayList<ArrayList<Integer>> res) {

        if (j > n) {
            res.add(new ArrayList<>(board));
            return;
        }

        for (int i = 1; i <= n; ++i) {
            if (!rows[i] && !diag1[i + j] && !diag2[i - j + n]) {
                // Place queen
                rows[i] = diag1[i + j] = diag2[i - j + n] = true;
                board.add(i);

                // Recur for next column
                nQueenUtil(j + 1, n, board, rows, diag1, diag2, res);

                // Backtrack
                board.remove(board.size() - 1);
                rows[i] = diag1[i + j] = diag2[i - j + n] = false;
            }
        }
    }

    public ArrayList<ArrayList<Integer>> nQueen(int n) {
        ArrayList<ArrayList<Integer>> res = new ArrayList<>();
        ArrayList<Integer> board = new ArrayList<>();
        boolean[] rows = new boolean[n + 1];
        boolean[] diag1 = new boolean[2 * n + 1];
        boolean[] diag2 = new boolean[2 * n + 1];
        
        nQueenUtil(1, n, board, rows, diag1, diag2, res);
        return res;
    }
}
```

**Time Complexity:**

* Roughly **O(n!)** due to backtracking for each placement.

**Space Complexity:**

* O(n²) space for recursion and tracking arrays.

**Notes:**

1. Backtracking is the key technique here.
2. Diagonals need to be carefully handled using simple math (i+j and i-j+n).
3. Always backtrack after exploring a path to restore the previous state.


## Problem: N Queen Problem

Given an integer `n`, the task is to find the solution to the N-Queens problem, where `n` queens are placed on an `n x n` chessboard such that no two queens can attack each other.

The N-Queen problem involves placing `N` chess queens on an `N×N` chessboard so that no two queens attack each other.

**Example:**
**Input:**
```
n = 4
```

**Output:**
```
2 4 1 3
```

In this example, the solution places queens in the following positions:
- Queen 1 at column 2 of row 1
- Queen 2 at column 4 of row 2
- Queen 3 at column 1 of row 3
- Queen 4 at column 3 of row 4

This is one of the possible solutions to the 4-Queens problem.

**Constraints:**
- `1 <= n <= 10`: The number of queens (`n`) will always be between 1 and 10.

**Goal:**
The goal is to place `n` queens on an `n x n` chessboard such that no two queens threaten each other. A queen can attack another queen if they are in the same row, column, or diagonal.

**Approach:**
We use a **backtracking** approach to solve the N-Queens problem:
1. **Recursive Backtracking:**
   - Place queens one by one in different columns and check if the placement is safe (i.e., no two queens can attack each other).
   - For each valid placement, recursively place queens in the next row.
   - If all queens are placed safely, return the solution.
   - If no valid position is found, backtrack and try placing queens in different positions.

2. **Safety Check:**
   - For each placement of a queen, we need to ensure that no other queens can attack it. This is checked for the column, upper left diagonal, and upper right diagonal.

3. **Base Case:**
   - If all queens are placed, return the current configuration.
   - If placement is not possible, backtrack and try another position.

**🧮 Flowchart/ Diagram:**
```

Start
|
\|--> Initialize the chessboard
|
\|--> Call placeQueens function starting from row 0
|
\|--> For each column in the current row:
\|     - Check if placing a queen is safe
\|     - Place queen if safe
\|     - Recur to the next row
\|     - If placement fails, backtrack (remove queen)
|
\|--> If all queens are placed, return the configuration
|
End

```

**Pseudocode/Algorithm:**

```text
function isSafe(mat, row, col):
    # Check the column for safety
    for i in 0 to row-1:
        if mat[i][col] == 1:
            return false

    # Check upper left diagonal for safety
    for i, j in (row-1, col-1) to (0, 0):
        if mat[i][j] == 1:
            return false

    # Check upper right diagonal for safety
    for i, j in (row-1, col+1) to (0, n-1):
        if mat[i][j] == 1:
            return false

    return true

function placeQueens(row, mat):
    if row == n:
        return true
    
    for i in 0 to n-1:
        if isSafe(mat, row, i):
            mat[row][i] = 1
            if placeQueens(row + 1, mat):
                return true
            mat[row][i] = 0
    return false

function nQueen(n):
    mat = initialize an n x n chessboard with 0
    if placeQueens(0, mat):
        result = []
        for i in 0 to n-1:
            for j in 0 to n-1:
                if mat[i][j] == 1:
                    result.append(j + 1)
        return result
    else:
        return [-1]
```

**Code:**

```java
import java.util.*;

class GfG {

    // Function to check if it is safe to place
    // the queen at board[row][col]
    static boolean isSafe(int[][] mat, int row, int col) {
        int n = mat.length;

        // Check this col on upper side
        for (int i = 0; i < row; i++)
            if (mat[i][col] == 1)
                return false;

        // Check upper diagonal on left side
        for (int i = row - 1, j = col - 1; i >= 0 && j >= 0; i--, j--)
            if (mat[i][j] == 1)
                return false;

        // Check lower diagonal on left side
        for (int i = row - 1, j = col + 1; j < n && i >= 0; i--, j++)
            if (mat[i][j] == 1)
                return false;

        return true;
    }

    static boolean placeQueens(int row, int[][] mat) {
        int n = mat.length;

        // base case: If all queens are placed then return true
        if (row == n)
            return true;

        // Consider the row and try placing queen in all columns one by one
        for (int i = 0; i < n; i++) {

            // Check if the queen can be placed
            if (isSafe(mat, row, i)) {
                mat[row][i] = 1;
                if (placeQueens(row + 1, mat))
                    return true;
                mat[row][i] = 0;
            }
        }
        return false;
    }

    // Function to find the solution to the N-Queens problem
    static List<Integer> nQueen(int n) {

        // Initialize the board
        int[][] mat = new int[n][n];

        // If the solution exists
        if (placeQueens(0, mat)) {

            // to store the columns of the queens
            List<Integer> ans = new ArrayList<>();
            for (int i = 0; i < n; i++) {
                for (int j = 0; j < n; j++) {
                    if (mat[i][j] == 1) {
                        ans.add(j + 1);
                    }
                }
            }
            return ans;
        } else
            return Collections.singletonList(-1);
    }

    public static void main(String[] args) {
        int n = 4;
        List<Integer> ans = nQueen(n);
        for (int i : ans) {
            System.out.print(i + " ");
        }
    }
}
```

**Time Complexity:**

* **O(n!)**: The time complexity of the N-Queens problem is factorial because we are trying to place each queen in every row and column, leading to a backtracking search with branching at each step.

**Space Complexity:**

* **O(n^2)**: The space complexity is proportional to the size of the chessboard (`n x n`), where we store the board configuration.

**Notes:**

* The N-Queens problem is a classical example of a **backtracking** problem.
* For small values of `n`, this problem is computationally feasible. However, as `n` increases, the number of potential solutions grows factorially.
* The solution provided returns one possible valid solution for the N-Queens problem.

## Problem: N-Queens II

The **n-queens puzzle** is the problem of placing `n` queens on an `n x n` chessboard such that no two queens attack each other.  
Your task is to return the number of distinct solutions to the **n-queens puzzle**.

A queen can attack another queen if they share the same row, column, or diagonal.

**Example :**

```text
Input: n = 4
Output: 2
Explanation: There are two distinct solutions to the 4-queens puzzle as shown.
```

**Constraints:**
1 <= n <= 9

**Goal:**

* Find the number of distinct solutions to the N-Queens problem.

**Approach:**
To solve this problem, we need to place queens on the chessboard such that no two queens attack each other.
The challenge is to find all distinct ways to place the queens without brute-forcing through all possible placements.

1. **Backtracking:**

   * Use backtracking to place queens row by row.
   * At each row, try to place a queen in each column, ensuring no conflict with previously placed queens.

2. **Conflict Checking:**

   * Track columns, diagonals (top-left to bottom-right), and anti-diagonals (top-right to bottom-left) already occupied by a queen.
   * Skip columns or diagonals that are already occupied.

3. **Recursive Backtracking:**

   * Start from the first row and place queens in valid columns.
   * When the last row is reached, count it as a valid solution.

4. **Pruning:**

   * Use sets to track columns and diagonals that are already occupied.
   * Skip positions that would lead to conflicts, pruning redundant checks.

**Code Solution:**
```java
import java.util.*;

class Solution {
    public int totalNQueens(int n) {
        Set<Integer> cols = new HashSet<>();
        Set<Integer> diag1 = new HashSet<>();
        Set<Integer> diag2 = new HashSet<>();
        
        // Start backtracking from row 0
        return backtrack(n, 0, cols, diag1, diag2);
    }

    private int backtrack(int n, int row, Set<Integer> cols, Set<Integer> diag1, Set<Integer> diag2) {
        // If we have placed queens in all rows, we've found a solution
        if (row == n) {
            return 1;
        }
        
        int solutions = 0;
        // Try placing a queen in each column of the current row
        for (int col = 0; col < n; col++) {
            // If the column or diagonals are occupied, skip this column
            if (cols.contains(col) || diag1.contains(row - col) || diag2.contains(row + col)) {
                continue;
            }
            // Place a queen in this column
            cols.add(col);
            diag1.add(row - col);
            diag2.add(row + col);
            
            // Recur to place queens in the next row
            solutions += backtrack(n, row + 1, cols, diag1, diag2);
            
            // Backtrack: Remove the queen and try the next column
            cols.remove(col);
            diag1.remove(row - col);
            diag2.remove(row + col);
        }
        
        return solutions;
    }
}
```


**Time Complexity: O(N!)**
In the worst case, the backtracking algorithm explores all possible placements of queens, which leads to a factorial time complexity (N!), as there are N! ways to arrange n queens on an n x n board.

**Space Complexity: O(N)**
The space complexity is dominated by the space used by the recursion stack and the sets that store column and diagonal information, both of which use O(N) space.

**Notes:**

1. Use sets for efficient conflict checking for columns and diagonals.
2. Backtracking ensures we don't explore invalid paths, reducing redundant calculations.

## Problem: Permutation with Spaces

Given a string `s`, you need to print all possible strings that can be made by placing spaces (zero or one) in between the characters of the string. The output should be printed in sorted increasing order of strings.

**Example :**

```text
Input: "abc"
Output: 
["abc", "ab c", "a bc", "a b c"]
```

**Constraints:**
1. 1 <= s.length <= 10
2. s consists of lowercase English letters.

**Goal:**

* Generate all permutations of the string by inserting spaces between characters.
* Sort the permutations in lexicographical order.

**Approach:**
1. **Backtracking:**
    We will use a backtracking approach to generate all possible permutations of the string by placing spaces between any two consecutive characters.

2. **Recursive Function:**
    We define a recursive function solve() that will take the current string, index, and the string being built so far.

    In each recursive step, we can either append the current character directly to the string or append it with a space.

3. **Sorting the Results:**
    After generating all the permutations, we sort the result list to return them in increasing lexicographical order.

4. **Base Case:**
   The base case of the recursion is when we have processed all characters of the string. At this point, the current string is added to the result list.

**🧮 Flowchart/ Diagram:**

* \[Flowchart showing recursion and space insertion for permutations (Placeholder)]

**Pseudocode/Algorithm:**

```text
function permutation(s):
    Initialize the result list
    Call solve(s, 1, s[0], result)

function solve(s, index, current, result):
    if index == length of s:
        Add current string to result
        return
    Solve recursively with space: solve(s, index + 1, current + " " + s[index], result)
    Solve recursively without space: solve(s, index + 1, current + s[index], result)
```

**Code Solution:**
```java
import java.util.*;

class Solution {

    ArrayList<String> permutation(String s) {
        // Initialize the result list
        ArrayList<String> ans = new ArrayList<>();
        
        // Start the recursive process
        solve(s, 1, s.substring(0, 1), ans);
        
        // Sort the result list in lexicographical order
        Collections.sort(ans);
        
        return ans;
    }
    
    // Recursive helper function to generate permutations
    private void solve(String s, int index, String current, ArrayList<String> ans) {
        // Base case: if we reach the end of the string, add the current permutation
        if (index == s.length()) {
            ans.add(current);
            return;
        }

        // Recursive call with a space between current and the next character
        solve(s, index + 1, current + " " + s.charAt(index), ans);
        
        // Recursive call without a space, directly appending the next character
        solve(s, index + 1, current + s.charAt(index), ans);
    }
}
```

**Time Complexity:**

* The time complexity is **O(2^n \* n log n)**, where `n` is the length of the string. The factor **O(2^n)** arises because there are `2^(n-1)` ways to insert spaces, and **O(n log n)** is due to sorting the resulting list.

**Space Complexity:**

* The space complexity is **O(2^n \* n)** to store all the possible permutations and the recursion stack.

**Notes:**

1. **Backtracking** optimally generates all permutations by considering two possible outcomes (with and without space) at each step.
2. Sorting the results ensures lexicographical order, which is crucial for this problem.
3. Although the problem size is constrained (1 <= s.length <= 10), the solution employs an efficient recursive approach that scales well within these bounds.


## Problem: Generate Parentheses

Given a number `n`, return all the combinations of balanced parentheses of length `n`.

> A sequence of parentheses is **balanced** if:
> - Every opening bracket `(` has a corresponding closing bracket `)`.
> - The brackets are closed in the correct order.

**Examples:**
- Input: `n = 4`
- Output: `["(())", "()()"]`

- Balanced: `(())`, `()()`
- Not Balanced: `)()(`, `))((', `()))`

**Objective:**
- Generate **all valid** combinations of parentheses with exactly `n` total characters.
- Only balanced sequences are valid.

**Solution Approach:**
1. **Backtracking:**
   - Keep track of the number of open and close parentheses used so far.
   - Only add an open parenthesis if we haven't used up all `n/2` opens.
   - Only add a close parenthesis if it won't exceed the number of open parentheses placed.

2. **Conditions:**
   - Add an opening bracket if the number of open brackets used so far is less than `n/2`.
   - Add a closing bracket if the number of closing brackets used is less than the number of open brackets.

3. **End Condition:**
   - When the current string's length is `2 * (n/2)` (since one open + one close = 2 characters), store it.

**Important Notes:**
Here, n in the function generateParentheses(n) is expected to be the total number of characters (not just open brackets).
Hence, the recursive function validParentheses(n/2, 0, "", res) is called, where n/2 is the number of opening brackets.

**🧮 Flowchart/ Diagram:**  
- [Flowchart showing the recursive process of adding opening and closing parentheses (Placeholder)]

**Pseudocode/Algorithm:**  
```text
function validParentheses(n, open, curr, res):
    if current string length == 2 * n:
        add curr to result list
        return
    if open < n:
        validParentheses(n, open + 1, curr + "(", res)
    if current string length - open < open:
        validParentheses(n, open, curr + ")", res)

function generateParentheses(n):
    initialize result list
    call validParentheses(n/2, 0, "", res)
    return result list
```

**Solution Code:**
```java
import java.util.ArrayList;

class Solution {

    static void validParentheses(int n, int open, String curr, ArrayList<String> res) {
        if (curr.length() == 2 * n) {
            res.add(curr);
            return;
        }
        
        // Try to add '('
        if (open < n)
            validParentheses(n, open + 1, curr + "(", res);

        // Try to add ')'
        if (curr.length() - open < open)
            validParentheses(n, open, curr + ")", res);
    }

    public ArrayList<String> generateParentheses(int n) {
        ArrayList<String> res = new ArrayList<>();
        validParentheses(n / 2, 0, "", res);
        return res;
    }
}
```

**Time Complexity:**

* **O(2^n)**, as the backtracking explores all possible balanced sequences of parentheses.

**Space Complexity:**

* **O(n)** for recursion stack and **O(2^n \* n)** for storing the valid sequences.

**Notes:**

* The recursive approach ensures we only build valid combinations by checking at each step whether a parenthesis can be placed without violating the balanced sequence condition.
* The function generates only valid balanced parentheses combinations by controlling when and how parentheses are added.


## Problem: Generate IP Addresses

Given a string `s` containing only digits, your task is to restore it by returning all possible valid IP address combinations. You can return your answer in any order.
A valid IP address must be in the form of `A.B.C.D`, where `A`, `B`, `C`, and `D` are numbers from `0-255` (inclusive).

**Conditions for Validity:**
- Each segment must be a number between 0 and 255 (inclusive).
- The segment cannot be prefixed with `0` unless it is exactly `0`. For example, `01.1.2.11` and `00.11.21.1` are invalid.

**Examples:**
- Input: `"25525511135"`
  - Output: `["255.255.11.135", "255.255.111.35"]`

- Input: `"0000"`
  - Output: `["0.0.0.0"]`

- Input: `"1111"`
  - Output: `["1.1.1.1"]`

**Goal:**  
- Restore a string into all possible valid IP address combinations, formatted as `A.B.C.D`.

**Solution Approach:**
1. **Backtracking Approach**:
   - We need to break the string into four parts.
   - The length of each part must be between 1 to 3 digits, and the value should lie between 0 and 255 (inclusive).
   - Recursively explore each possible split to form valid segments of the IP address.

2. **Base Case**:
   - If we have formed 3 segments, the remaining part of the string should be the 4th segment. We verify if it's valid.

3. **Recursive Exploration**:
   - At each step, try taking 1, 2, or 3 digits from the remaining string, check if they form a valid segment, and then recursively continue with the next segment.

**🧮 Flowchart/ Diagram:**  
- [Diagram placeholder: Illustrate recursive breakdown of the string into 4 segments and checks for validity at each step.]

**Pseudocode/Algorithm:**  
```text
function isValid(segment):
    if segment length is 1: return true
    if segment starts with '0': return false
    if segment value > 255: return false
    return true

function generateIpRec(s, index, currentIp, count, result):
    if index >= length of s:
        return
    if count == 3:
        lastSegment = s.substring(index)
        if isValid(lastSegment):
            result.add(currentIp + lastSegment)
        return

    for i from index to min(index+3, length of s):
        part = s.substring(index, i+1)
        if isValid(part):
            generateIpRec(s, i+1, currentIp + part + ".", count+1, result)

function generateIp(s):
    result = []
    generateIpRec(s, 0, "", 0, result)
    return result
``` 

**Solution Code:**
```java
import java.util.ArrayList;

class Solution {

    // Helper function to check if a part of the IP is valid
    static boolean isValid(String s) {
        int n = s.length();
        if (n == 1) 
            return true; // Single digit is always valid (0-9)
        
        int val = Integer.parseInt(s);
        // Valid if it's not prefixed by 0 and lies between 0 and 255
        if (s.charAt(0) == '0' || val > 255) 
            return false;

        return true;
    }

    // Recursive function to generate valid IP combinations
    static void generateIpRec(String s, int index, String curr, 
                               int cnt, ArrayList<String> res) {
        String temp = "";
        if (index >= s.length()) 
            return;

        if (cnt == 3) {
            // Add the last part of the IP address
            temp = s.substring(index);
            if (temp.length() <= 3 && isValid(temp)) 
                res.add(curr + temp);  // Form the complete IP address

            return;
        }

        // Try splitting into parts of 1, 2, or 3 digits
        for (int i = index; i < Math.min(index + 3, s.length()); i++) {
            temp += s.charAt(i);
            if (isValid(temp)) {
                generateIpRec(s, i + 1, curr + temp + ".", cnt + 1, res);
            }
        }
    }

    // Main function to generate all valid IP addresses
    public ArrayList<String> generateIp(String s) {
        ArrayList<String> res = new ArrayList<>();
        generateIpRec(s, 0, "", 0, res);
        return res;
    }
}
```

**Time Complexity: O(3^4),** as we are recursively splitting the string into up to 3 parts for each segment, and we have 4 segments to form.

**Space Complexity: O(1)** for recursion stack, and O(N) for storing valid IP addresses.

**Notes:**

* The string length must be between 4 and 12 to form a valid IP address.
* This recursive strategy avoids invalid states early by checking segment validity before proceeding.

## Problem: Solve the Sudoku

Given an incomplete Sudoku configuration in terms of a 9x9 2-D integer square matrix `mat[][]`, the task is to solve the Sudoku. It is guaranteed that the input Sudoku will have exactly one solution.
A Sudoku solution must satisfy all of the following rules:

1. Each of the digits 1-9 must occur exactly once in each row.
2. Each of the digits 1-9 must occur exactly once in each column.
3. Each of the digits 1-9 must occur exactly once in each of the 9 3x3 sub-boxes of the grid.

Note: Zeros represent blanks to be filled with numbers 1-9, while non-zero cells are fixed and cannot be changed.

**Input Format:**
- A 9x9 2D integer array `mat[][]`, where each element is from 0 to 9.
- `0` represents an empty cell to be filled.

**Output Format:**
- The same 9x9 matrix `mat[][]` should be modified in-place to represent the solved Sudoku board.

**Constraints:**
- The input will always be a valid Sudoku puzzle with exactly one possible solution.
- `mat.length == 9`
- `mat[i].length == 9`
- `0 <= mat[i][j] <= 9`

**Edge Cases:**
- If the Sudoku is already solved, the algorithm should detect it and make no unnecessary changes.
- If only one cell is blank, the algorithm should efficiently determine the correct digit using constraint satisfaction.

**Example :**
```text
Input:
mat[][] = {
    {5, 3, 0, 0, 7, 0, 0, 0, 0},
    {6, 0, 0, 1, 9, 5, 0, 0, 0},
    {0, 9, 8, 0, 0, 0, 0, 6, 0},
    {8, 0, 0, 0, 6, 0, 0, 0, 3},
    {4, 0, 0, 8, 0, 3, 0, 0, 1},
    {7, 0, 0, 0, 2, 0, 0, 0, 6},
    {0, 6, 0, 0, 0, 0, 2, 8, 0},
    {0, 0, 0, 4, 1, 9, 0, 0, 5},
    {0, 0, 0, 0, 8, 0, 0, 7, 9}
}
Output:
mat[][] = {
    {5, 3, 4, 6, 7, 8, 9, 1, 2},
    {6, 7, 2, 1, 9, 5, 3, 4, 8},
    {1, 9, 8, 3, 4, 2, 5, 6, 7},
    {8, 5, 9, 7, 6, 1, 4, 2, 3},
    {4, 2, 6, 8, 5, 3, 7, 9, 1},
    {7, 1, 3, 9, 2, 4, 8, 5, 6},
    {9, 6, 1, 5, 3, 7, 2, 8, 4},
    {2, 8, 7, 4, 1, 9, 6, 3, 5},
    {3, 4, 5, 2, 8, 6, 1, 7, 9}
}
```
**Solution Approach:**
1. **Backtracking Algorithm:**
   A recursive approach is used to solve the Sudoku by trying possible values for each empty cell.
   For each cell, try placing digits from 1 to 9 and check if placing the digit is safe. If it is, move to the next empty cell.
   If all the cells are filled with valid digits, the Sudoku is solved.

2. **Checking Validity:**
 For each placement, check if the number violates Sudoku constraints:
    It should not repeat in the row.
    It should not repeat in the column.
    It should not repeat in the corresponding 3x3 sub-grid.

3. **Bitmask Optimization:**
Use bitwise operations to track which numbers are already placed in the rows, columns, and 3x3 sub-boxes. This reduces the need for checking every cell.
**Pseudocode/Algorithm:**
```text
function solveSudoku(mat):
    n = 9
    row[9], col[9], box[9] = 0

    // Pre-fill bitmasks based on initial grid
    for i in 0 to 8:
        for j in 0 to 8:
            if mat[i][j] != 0:
                bit = 1 << mat[i][j]
                row[i] |= bit
                col[j] |= bit
                box[i/3*3 + j/3] |= bit

    call backtrack(i=0, j=0)

function backtrack(i, j):
    if i == 9:
        return true
    if j == 9:
        return backtrack(i+1, 0)
    if mat[i][j] != 0:
        return backtrack(i, j+1)

    for num from 1 to 9:
        bit = 1 << num
        boxIndex = i/3*3 + j/3
        if row[i] & bit == 0 and col[j] & bit == 0 and box[boxIndex] & bit == 0:
            mat[i][j] = num
            row[i] |= bit
            col[j] |= bit
            box[boxIndex] |= bit
            
            if backtrack(i, j+1): return true

            mat[i][j] = 0
            row[i] ^= bit
            col[j] ^= bit
            box[boxIndex] ^= bit

    return false
```
**Code Solution:**
```java
class Solution {

    // Helper function to check if a number can be placed safely at mat[i][j]
    static boolean isSafe(int[][] mat, int i, int j, int num, 
                          int[] row, int[] col, int[] box) {
        if ((row[i] & (1 << num)) != 0 || (col[j] & (1 << num)) != 0 || 
            (box[i / 3 * 3 + j / 3] & (1 << num)) != 0)
            return false;
        
        return true;
    }

    // Recursive function to solve the Sudoku
    static boolean sudokuSolverRec(int[][] mat, int i, int j, 
                                   int[] row, int[] col, int[] box) {
        int n = mat.length;
        if (i == n - 1 && j == n)
            return true;  // Base case: If we've reached the end of the grid
        if (j == n) {
            i++;  // Move to the next row
            j = 0;  // Reset column to 0
        }
        if (mat[i][j] != 0)  // Skip fixed cells
            return sudokuSolverRec(mat, i, j + 1, row, col, box);

        // Try placing digits 1 to 9
        for (int num = 1; num <= n; num++) {
            if (isSafe(mat, i, j, num, row, col, box)) {
                mat[i][j] = num;
                row[i] |= (1 << num);  // Mark the number as used in the row
                col[j] |= (1 << num);  // Mark the number as used in the column
                box[i / 3 * 3 + j / 3] |= (1 << num);  // Mark the number as used in the sub-box

                // Recursively move to the next cell
                if (sudokuSolverRec(mat, i, j + 1, row, col, box))
                    return true;
                mat[i][j] = 0;  // Backtrack if not solved
                row[i] &= ~(1 << num);  // Unmark the number in the row
                col[j] &= ~(1 << num);  // Unmark the number in the column
                box[i / 3 * 3 + j / 3] &= ~(1 << num);  // Unmark the number in the sub-box
            }
        }

        return false;  // Return false if no valid number can be placed
    }

    // Function to solve the Sudoku puzzle
    static void solveSudoku(int[][] mat) {
        int n = mat.length;
        int[] row = new int[n];
        int[] col = new int[n];
        int[] box = new int[n];

        // Preprocess the grid to mark the numbers already placed
        for (int i = 0; i < n; i++) {
            for (int j = 0; j < n; j++) {
                if (mat[i][j] != 0) {
                    row[i] |= (1 << mat[i][j]);
                    col[j] |= (1 << mat[i][j]);
                    box[(i / 3) * 3 + j / 3] |= (1 << mat[i][j]);
                }
            }
        }

        // Start solving the Sudoku from the top-left cell
        sudokuSolverRec(mat, 0, 0, row, col, box);
    }
}
```

**Time Complexity: O(9^(N*N))**where N is the grid size (9 in this case). In the worst case, the recursive backtracking could explore all possibilities.
**Space Complexity: O(N),**where N is the grid size (for the bitmask arrays and recursion stack).

**Additional Notes:**

* Bitmasking improves performance by reducing constant-time lookups compared to iterative value checking.
* A unique solution guarantee allows for optimized backtracking without needing to collect multiple answers.

**Diagram Suggestion:**

* Grid showing bitmask arrays `row[]`, `col[]`, and `box[]`, highlighting valid vs invalid placements at a sample step in the recursion.


## Problem: Find Kth Permutation

Given two integers `N` (1 <= N <= 9) and `K`, find the `Kth` permutation sequence of the first `N` natural numbers. Return the answer in string format.

**Example :**
```text
Input: N = 3, K = 3  
Output: "213"
```
**Constraints:**
1. 1 <= N <= 9
2. 1 <= K <= N!

**Example**
**Input:**
- N = 4, K = 9

**Output:**
- "2314"

**Goal:** 
- To compute the exact Kth permutation without generating all previous permutations, using combinatoric logic and factorial arithmetic.

**Approach:**
To solve this problem, we need to figure out how to generate the Kth permutation of the first N natural numbers without generating all possible permutations. The key idea is to break the problem down using the factorial number system.

1. **Factorial Representation:**
    The number of permutations of N elements is N!.

    The first (N-1)! permutations will start with the first number, the next (N-1)! will start with the second number, and so on.

2. **Step-by-Step Calculation:**
    We reduce the problem size by selecting the correct digit for each place, using the factorial to determine which block the Kth permutation falls into.

    We adjust K and the available digits after each selection.

3. **Procedure:**
    Initialize a list of numbers from 1 to N.

    Calculate the factorial values to determine the correct position of each number in the Kth permutation.

    Gradually reduce K as you select digits and update the list of available numbers.

**🧮 Flowchart/ Diagram:**
- Not applicable in text-only format, but can be described as:
  1. Compute factorials up to N.
  2. Use factorial values to index and choose digits step-by-step.
  3. Update K and available digit list dynamically.

**Pseudocode/Algorithm:**
```

function getKthPermutation(N, K):
numbers = \[1, 2, ..., N]
K = K - 1  // Convert to 0-based index
result = ""
factorial = computeFactorial(N-1)

for i from 0 to N-1:
    index = K // factorial
    result += numbers[index]
    remove numbers[index] from list
    K = K % factorial
    if N-i-1 > 0:
        factorial = factorial / (N-i-1)
return result

```

**Code Solution:**
```java
import java.util.*;

class Solution {
    public static String kthPermutation(int n, int k) {
        // Initialize the list of numbers
        List<Integer> numbers = new ArrayList<>();
        for (int i = 1; i <= n; i++) {
            numbers.add(i);
        }
        
        // Decrease k by 1 to make it zero-indexed
        k--; 
        
        StringBuilder result = new StringBuilder();
        int fact = 1;
        
        // Compute (n-1)! to determine the factorials for each step
        for (int i = 1; i < n; i++) {
            fact *= i;
        }
        
        // Find the kth permutation
        for (int i = 0; i < n; i++) {
            int index = k / fact;  // Find which number to pick
            result.append(numbers.get(index));  // Append to the result
            numbers.remove(index);  // Remove the selected number from the list
            
            // Update k and factorial for the next iteration
            if (i < n - 1) {
                k %= fact;
                fact /= (n - i - 1);
            }
        }
        
        return result.toString();  // Return the result as a string
    }
}
```
**Time Complexity: O(N^2)**
For each of the N iterations, we perform a list removal operation, which takes O(N) time in the worst case. Therefore, the overall time complexity is O(N^2).

**Space Complexity: O(N)**
We use a list of size N to store the available digits, so the space complexity is O(N).

**Notes:**
- It’s crucial to zero-index `K` before starting the selection to align with array-based logic.
- List removals can be optimized with linked structures if needed, though N is small (max 9).

## Problem: Find the K-th Permutation Sequence of First N Natural Numbers

Given two integers `N` and `K`, the task is to find the Kth permutation sequence of numbers from 1 to N without using STL functions.

**Example:**
**Input:**
```

N = 3, K = 4

```

**Output:**
```

\[2, 3, 1]

```

**Explanation:**
- The sequence of all permutations of numbers from 1 to 3 in lexicographical order are:
  1. 1 2 3
  2. 1 3 2
  3. 2 1 3
  4. 2 3 1
  5. 3 1 2
  6. 3 2 1

- The 4th permutation in the list is `[2, 3, 1]`.

**Approach:**

**Key Insight:
To find the K-th permutation without generating all permutations:
1. The total number of permutations of `N` distinct elements is `N!` (factorial of N).
2. Each number can be placed at the first position of the permutation, and the remaining numbers will follow the same permutation logic recursively.
3. Given `K`, we can determine which number will occupy the first position, and then recursively solve for the rest of the positions.

**Steps:**
1. **Factorial Number System:** The idea is to divide the problem into subproblems by calculating the factorial of `n-1` at each step. This gives us the number of permutations for each position.
2. **Index Calculation:** For a given `K`, the first number in the permutation can be determined by dividing `K` by `(n-1)!`. The remainder will give us the index of the next number to consider.
3. **Update K:** After choosing a number for the first position, update `K` for the remaining `n-1` elements.

**Procedure:**
1. Start with an array of numbers from `1` to `N`.
2. Calculate the factorial of `N-1` to determine the number of permutations that start with each number.
3. Use the quotient of `K/(N-1)!` to select the first number of the permutation.
4. Recursively find the K-th permutation for the remaining `N-1` numbers.

**Algorithm:**

```text
function findFirstNumIndex(k, n):
    if n == 1:
        return 0
    n -= 1
    n_partial_fact = n

    while k >= n_partial_fact and n > 1:
        n_partial_fact = n_partial_fact * (n - 1)
        n -= 1

    first_num_index = k // n_partial_fact
    k = k % n_partial_fact
    return first_num_index

function findKthPermutation(n, k):
    ans = ""
    s = set()  # to store the natural numbers

    for i = 1 to n:
        s.add(i)

    v = list(s)  # convert set to list
    k -= 1  # Adjust for 0-based indexing

    for i = 0 to n-1:
        index = findFirstNumIndex(k, n - i)

        # Add the number at the found index to the answer
        ans += str(v[index])
        v.remove(index)

        # Update K for the remaining numbers
        k = k % (n - i - 1)

    return ans
```

**Code:**

```java
import java.util.*;

class GFG {

    // Function to find the index of number at the first position of kth sequence of set of size n
    static int findFirstNumIndex(int k, int n) {
        if (n == 1)
            return 0;
        n--;

        int first_num_index;
        int n_partial_fact = n;

        while (k >= n_partial_fact && n > 1) {
            n_partial_fact = n_partial_fact * (n - 1);
            n--;
        }

        // First position of the kth sequence will be occupied by the number present at index = k / (n-1)!
        first_num_index = k / n_partial_fact;

        k = k % n_partial_fact;
        return first_num_index;
    }

    // Function to find the kth permutation of n numbers
    static String findKthPermutation(int n, int k) {
        // Store final answer
        String ans = "";

        HashSet<Integer> s = new HashSet<>();

        // Insert all natural numbers upto n in set
        for (int i = 1; i <= n; i++)
            s.add(i);

        Vector<Integer> v = new Vector<>();
        v.addAll(s);

        // Subtract 1 to get 0 based indexing
        k = k - 1;

        for (int i = 0; i < n; i++) {
            int index = findFirstNumIndex(k, n - i);

            // itr now points to the number at index in set s
            if(index < v.size()) {
                ans += ((v.elementAt(index).toString()));
                v.remove(index);
            }
            else {
                ans += String.valueOf(v.elementAt(0) + 2);
            }

            // Remove current number from the set
        }

        return ans;
    }

    // Driver code
    public static void main(String[] args) {
        int n = 3, k = 4;
        String kth_perm_seq = findKthPermutation(n, k);
        System.out.print(kth_perm_seq + "\n");
    }
}
```

**Time Complexity:**

* **Time Complexity:** O(N^2), where `N` is the number of elements. For each of the `N` positions, we need to find the number at that position and remove it from the list, which takes O(N) time.

**Space Complexity:**

* **Space Complexity:** O(N), due to the space required to store the list of numbers and the answer string.

**Notes:**

This approach efficiently computes the K-th permutation by iteratively selecting the correct number for each position based on the factorial number system, thus avoiding the need to generate all permutations explicitly.


## Problem: Word Search

You are given a two-dimensional matrix `mat[][]` of size `n * m` containing English alphabets and a string `word`. Check if the `word` exists on the matrix. The word can be constructed by using letters from adjacent cells, either horizontally or vertically. The same cell cannot be used more than once.

**Input:**
- A 2D character matrix `mat[][]` of dimensions `n x m`
- A target string `word`

**Output:**
- Return `true` if the word exists in the matrix, otherwise return `false`

**Example :**
```text
Input:
mat[][] = {
    {'A', 'B', 'C', 'E'},
    {'S', 'F', 'C', 'S'},
    {'A', 'D', 'E', 'E'}
}
word = "ABCCED"
Output: true
```

**Example**
**Input:**
```text
mat = \[\['A','B','C','E'],
\['S','F','C','S'],
\['A','D','E','E']]
word = "SEE"
```
**Output:**
```
true
```

**Constraints:** 
- 1 ≤ n, m ≤ 6
- 1 ≤ word.length ≤ 15
- All characters are uppercase English letters

**Goal:** 
- To determine if a given word can be constructed by sequentially adjacent letters (horizontal or vertical) in a grid without reusing the same cell.

**🧮 Flowchart/ Diagram:**
- Start at each cell → match 1st char?
  - Yes → Mark visited → recursively search neighbors for next char
  - No → Move to next cell
- Continue until word is found or all paths exhausted

**Solution Approach:**
1. **Backtracking:**
    The idea is to explore all possible paths starting from each cell in the matrix.

    For each cell, we try to match the first character of the word, and then recursively check for the remaining characters by exploring adjacent cells (up, down, left, and right).

    During the exploration, we mark the visited cells by changing their values temporarily (marking as #) to avoid revisiting them in the current path. After the search completes for a particular cell, we restore the cell to its original state.

2. **Recursive Search:**

    The recursive function findMatch() checks if the current character matches the character in the word at the given index. If it matches, the function proceeds to check the adjacent cells.

3. **Return Condition:**

    If the function finds the entire word by successfully matching all characters in the word, it returns true. If no path can form the word, it returns false.


**Pseudocode/Algorithm:**
```

function isWordExist(mat, word):
for i in range(rows):
for j in range(cols):
if mat\[i]\[j] == word\[0]:
if dfs(mat, word, i, j, 0):
return true
return false

function dfs(mat, word, x, y, index):
if index == length of word:
return true
if x or y out of bounds or mat\[x]\[y] != word\[index]:
return false

temp = mat[x][y]
mark mat[x][y] as visited
for each direction in [up, down, left, right]:
    if dfs(new position, word, index + 1):
        return true
unmark mat[x][y]
return false

```

**Code Solution:**
```java
class Solution {
    // Helper function to find the match for the word starting at (x, y)
    static boolean findMatch(char[][] mat, String word, int x, int y, int wIdx) {
        int wLen = word.length();
        int n = mat.length;
        int m = mat[0].length;
        
        // If we have checked all characters of the word
        if (wIdx == wLen)
            return true;

        // Check if the current cell is out of bounds or if the character does not match
        if (x < 0 || y < 0 || x >= n || y >= m)
            return false;

        if (mat[x][y] == word.charAt(wIdx)) {
            // Temporarily mark the cell as visited
            char temp = mat[x][y];
            mat[x][y] = '#';

            // Explore all possible adjacent cells
            boolean res = findMatch(mat, word, x - 1, y, wIdx + 1) || // Up
                          findMatch(mat, word, x + 1, y, wIdx + 1) || // Down
                          findMatch(mat, word, x, y - 1, wIdx + 1) || // Left
                          findMatch(mat, word, x, y + 1, wIdx + 1);   // Right

            // Restore the cell back to its original state
            mat[x][y] = temp;
            return res;
        }
        
        return false;
    }

    // Main function to check if the word exists in the matrix
    static public boolean isWordExist(char[][] mat, String word) {
        int wLen = word.length();
        int n = mat.length;
        int m = mat[0].length;

        // If the word is longer than the number of cells in the matrix, return false
        if (wLen > n * m)
            return false;

        // Try to find the word starting from each cell in the matrix
        for (int i = 0; i < n; i++) {
            for (int j = 0; j < m; j++) {
                // If the first character of the word matches the cell, start the search
                if (mat[i][j] == word.charAt(0)) {
                    if (findMatch(mat, word, i, j, 0))
                        return true;
                }
            }
        }
        
        return false; // Return false if no match is found
    }
}
```

**Time Complexity: O(n * m * 4^L),** where n is the number of rows, m is the number of columns, and L is the length of the word. In the worst case, for each cell, we perform a search on the adjacent cells (up, down, left, right), and we have to check the entire word.
**Space Complexity: O(L),** where L is the length of the word. The recursion stack depth depends on the length of the word.

**Notes:**
- This is a typical depth-first search (DFS) grid problem with backtracking.
- Modifying the matrix in-place by marking visited cells reduces space overhead but must be restored after recursion.

## Problem: Word Search in a 2D Grid of Characters

Given a 2D grid `m*n` of characters and a word, the task is to find all occurrences of the given word in the grid. A word can be matched in all 8 directions at any point. Word is said to be found in a direction if all characters match in this direction (not in zig-zag form).

The 8 directions are:
- Horizontally Left
- Horizontally Right
- Vertically Up
- Vertically Down
- 4 Diagonal directions

**Note:**  
The returning list should be lexicographically smallest. If the word can be found in multiple directions starting from the same coordinates, the list should contain the coordinates only once.

You are given a 2D grid `m*n` of characters and a word. The task is to find all occurrences of the given word in the grid. The word can be matched in all 8 directions (horizontally, vertically, or diagonally). You need to return all the coordinates where the word starts in the lexicographically smallest order.

**Example 1:**

**Input:**
```text
grid = {{G,E,E,K,S,F,O,R,G,E,E,K,S},
        {G,E,E,K,S,Q,U,I,Z,G,E,E,K},
        {I,D,E,Q,A,P,R,A,C,T,I,C,E}}
word = "GEEKS"
```

**Output:**

```text
{{0,0}, {0,8}, {1,0}}
```

**Example 2:**

**Input:**

```text
grid = {{a,b,a,b},
        {a,b,e,b},
        {e,b,e,b}}
word = "abe"
```

**Output:**

```text
{{0,0},{0,2},{1,0}}
```

**Constraints:**

* `1 <= m, n <= 100`
* `1 <= word.length <= 100`
* The grid contains only lowercase English letters.
* The word can be found multiple times.

**Goal:**

* Find all occurrences of the given word in the 2D grid.
* The list should be lexicographically smallest.
* The word can be matched in all 8 possible directions.

**Approach:**

We will implement two approaches to solve the problem: a **Naive Recursive Approach** and an **Iterative Approach**.

**🧮 Flowchart/Diagram:**

1. **Start**: Begin with the given grid and word.
2. **Traverse the Grid**: For each cell, check if the word can be found starting from that cell in any of the 8 directions.
3. **Match the Word**: If the word matches in a direction, store the coordinates.
4. **Output the Coordinates**: Sort the coordinates lexicographically and return the result.

**Pseudocode/Algorithm:**

1. **Naive Approach (Recursive)**

   * For each cell in the grid, recursively check if the word can be formed by moving in any of the 8 directions.
   * If the word is found, add the starting coordinates to the result list.

2. **Expected Approach (Iterative)**

   * For each cell in the grid, iteratively check all 8 directions to see if the word can be formed.
   * If the word is found, store the starting coordinates and continue.

**\[Naive Approach] Using Recursion – O(m*n*k) Time and O(k) Space**

In this approach, we visit each cell in the grid and check in all eight directions (up, down, left, right, and diagonals) to find the word. If the word is found in a direction, we store the starting point.

**Code:**

```java
// Java program to search a word in a 2D grid
import java.util.*;

class GfG {

    // This function checks if the given coordinate is valid
    static boolean validCoord(int x, int y, int m, int n) {
        if (x >= 0 && x < m && y >= 0 && y < n)
            return true;
        return false;
    }

    // This function searches for the given word in a given direction from the coordinate.
    static boolean findWord(int index, String word, char[][] grid, 
                            int x, int y, int dirX, int dirY) {

        // if word has been found
        if (index == word.length()) return true;

        // if the current coordinate is valid and characters match, then check the next index
        if (validCoord(x, y, grid.length, grid[0].length) && 
        word.charAt(index) == grid[x][y])
            return findWord(index + 1, word, grid, 
                            x + dirX, y + dirY, dirX, dirY);

        return false;
    }

    // This function calls search2D for each coordinate
    static int[][] searchWord(char[][] grid, String word) {
        int m = grid.length;
        int n = grid[0].length;

        List<int[]> ans = new ArrayList<>();

        // x and y are used to set the direction in which word needs to be searched.
        int[] x = { -1, -1, -1, 0, 0, 1, 1, 1 };
        int[] y = { -1, 0, 1, -1, 1, -1, 0, 1 };

        for (int i = 0; i < m; i++) {
            for (int j = 0; j < n; j++) {
                // Search in all 8 directions
                for (int k = 0; k < 8; k++) {
                    // If word is found, then append the coordinates into answer and break.
                    if (findWord(0, word, grid, i, j, x[k], y[k])) {
                        ans.add(new int[] { i, j });
                        break;
                    }
                }
            }
        }

        return ans.toArray(new int[0][]);
    }

    static void printResult(int[][] ans) {
        for (int[] a : ans) {
            System.out.print( "{" + a[0] + "," + a[1] + "}" + " ");
        }
        System.out.println();
    }

    public static void main(String[] args) {
        char[][] grid = { { 'a', 'b', 'a', 'b' },
                          { 'a', 'b', 'e', 'b' },
                          { 'e', 'b', 'e', 'b' } };
        String word = "abe";

        int[][] ans = searchWord(grid, word);

        printResult(ans);
    }
}
```

**Time Complexity:**

* **Time Complexity:** O(m \* n \* k), where `m` is the number of rows, `n` is the number of columns, and `k` is the length of the word.
* **Auxiliary Space:** O(k), due to the recursive stack space.

**\[Expected Approach] Using Iteration – O(m*n*k) Time and O(1) Space**

This is an iterative approach for the recursive solution. Here, for each cell, we try to move in one direction iteratively, checking if the word matches as we move.

**Code:**

```java
// Java program to search word in 2D grid in 8 directions
import java.util.*;

class GfG {

    // This function searches for the given word in all 8 directions from the coordinate.
    static boolean search2D(char[][] grid, int row, int col, String word) {
        int m = grid.length;
        int n = grid[0].length;

        // return false if the given coordinate does not match with first index char.
        if (grid[row][col] != word.charAt(0))
            return false;

        int len = word.length();
        int[] x = { -1, -1, -1, 0, 0, 1, 1, 1 };
        int[] y = { -1, 0, 1, -1, 1, -1, 0, 1 };

        // This loop will search in all the 8 directions
        for (int dir = 0; dir < 8; dir++) {
            int k, currX = row + x[dir], currY = col + y[dir];

            // First character is already checked, match remaining
            for (k = 1; k < len; k++) {
                if (currX >= m || currX < 0 || currY >= n || currY < 0)
                    break;

                if (grid[currX][currY] != word.charAt(k))
                    break;

                currX += x[dir];
                currY += y[dir];
            }

            if (k == len)
                return true;
        }

        return false;
    }

    // This function calls search2D for each coordinate
    static int[][] searchWord(char[][] grid, String word) {
        int m = grid.length;
        int n = grid[0].length;

        // Max possible occurrences
        int[][] ans = new int[m * n][2];
        int count = 0;

        for (int i = 0; i < m; i++) {
            for (int j = 0; j < n; j++) {
                if (search2D(grid, i, j, word)) {
                    ans[count][0] = i;
                    ans[count][1] = j;
                    count++;
                }
            }
        }

        // Resize the array to fit the actual number of found coordinates
        int[][] result = new int[count][2];
        for (int i = 0; i < count; i++) {
            result[i] = ans[i];
        }

        return result;
    }

    static void printResult(int[][] ans) {
        for (int[] coords : ans) {
            System.out.print("{" + coords[0] + "," + coords[1] + "}" + " ");
        }
        System.out.println();
    }

    public static void main(String[] args) {
        char[][] grid = { { 'a', 'b', 'a', 'b' },
                          { 'a', 'b', 'e', 'b' },
                          { 'e', 'b', 'e', 'b' } };

String word = "abe";

    int[][] ans = searchWord(grid, word);

    printResult(ans);
}

}

```

**Time Complexity:**
- **Time Complexity:** O(m * n * k), where `m` is the number of rows, `n` is the number of columns, and `k` is the length of the word.
- **Auxiliary Space:** O(1), as the iteration uses constant space.

## Problem: Find all possible palindromic partitions of a String

Given a string `S`, find all possible palindromic partitions of the given string. A **palindromic partition** is a way of dividing the string such that every substring is a palindrome.

**Input:**
- A string `S` consisting of lowercase English letters.

**Output:**
- A list of all possible palindromic partitions of `S`.

**Example :**
```text
Input: S = "aab"
Output: [["a", "a", "b"], ["aa", "b"]]
```

**Constraints:**
1. 1 <= S.length <= 15
2. S consists of lowercase English letters.

**Goal:** 
- Generate all partitions of the string such that each substring in the partition is a palindrome.

**🧮 Flowchart/ Diagram:**
- Start from index 0
  - For each index, check all substrings ending at or after it
    - If substring is a palindrome → include it in current path
    - Recurse on the remaining string
  - If end of string reached → add current path to result
  - Backtrack and try other options

**Approach:**
To solve this problem, we need to find all the possible ways to partition the string into palindromic substrings.

1. **Backtracking Approach:**
    We use a backtracking approach to generate all possible partitions of the string.

    For each substring, we check whether it is a palindrome.

    If it is a palindrome, we add it to the current partition and recursively try to partition the remaining part of the string.

    When we reach the end of the string, we add the current partition to the result.

2. **Palindrome Check:**
    We use a helper function to check if a given string is a palindrome by comparing characters from both ends.

**Pseudocode/Algorithm:**
```

function allPalindromicPerms(S):
result = \[]
currentPartition = \[]
backtrack(0)
return result

function backtrack(start):
if start == S.length:
add currentPartition to result
return
for end from start+1 to S.length:
substring = S\[start\:end]
if isPalindrome(substring):
currentPartition.append(substring)
backtrack(end)
currentPartition.pop()

function isPalindrome(s):
left, right = 0, s.length - 1
while left < right:
if s\[left] != s\[right]:
return false
left++, right--
return true

```
**Code Solution:**
```java
import java.util.*;

class Solution {
    // Function to find all palindromic partitions
    static ArrayList<ArrayList<String>> allPalindromicPerms(String S) {
        // Initialize result list
        ArrayList<ArrayList<String>> result = new ArrayList<>();
        ArrayList<String> currentPartition = new ArrayList<>();
        
        // Call the helper function to find palindromic partitions
        findPalindromicPartitions(S, 0, currentPartition, result);
        
        // Return all the palindromic partitions
        return result;
    }
    
    // Helper function for backtracking to find palindromic partitions
    static void findPalindromicPartitions(String S, int start, ArrayList<String> currentPartition, ArrayList<ArrayList<String>> result) {
        // If we've reached the end of the string, add the current partition to result
        if (start == S.length()) {
            result.add(new ArrayList<>(currentPartition));
            return;
        }
        
        // Iterate through all possible substrings starting from the current position
        for (int end = start + 1; end <= S.length(); end++) {
            // Extract substring
            String substring = S.substring(start, end);
            
            // If the substring is a palindrome, add it to the current partition
            if (isPalindrome(substring)) {
                currentPartition.add(substring);
                
                // Recursively find partitions for the remaining substring
                findPalindromicPartitions(S, end, currentPartition, result);
                
                // Backtrack: remove the last added substring
                currentPartition.remove(currentPartition.size() - 1);
            }
        }
    }

    // Helper function to check if a string is a palindrome
    static boolean isPalindrome(String s) {
        int left = 0, right = s.length() - 1;
        
        // Compare characters from both ends
        while (left < right) {
            if (s.charAt(left) != s.charAt(right)) {
                return false;
            }
            left++;
            right--;
        }
        
        return true;
    }
}
```

**Time Complexity: O(2^N * N)**
In the worst case, there are 2^N possible partitions of the string (since each character can either be part of a partition or not). For each partition, we are checking all substrings for palindrome properties, which takes O(N) time. Thus, the total time complexity is O(2^N * N).

**Space Complexity: O(N)**
We use extra space for storing the current partition and the result, both of which can have up to N elements. Additionally, the recursion stack may go up to O(N) deep, leading to a space complexity of O(N).

**Notes:**
- Efficient backtracking with pruning helps eliminate unnecessary paths early.
- Palindrome check is optimized with two-pointer approach.



## Problem: Palindrome Partitioning

Given a string `s`, the task is to find the minimum number of cuts needed for palindrome partitioning of the given string. A partitioning of the string is a palindrome partitioning if every sub-string of the partition is a palindrome.

You are given a string `s`. The goal is to partition the string such that each substring is a palindrome, and the number of cuts needed to partition the string is minimized.

For example:
- `"geek"` can be partitioned as `"g|e|ek"` (where each substring is a palindrome), so the minimum cuts required is `2`.

**Example**

**Input:**
```java
s = "geek"
````

2. **Recursive Case:** For substrings that are not palindromes, we partition the string at every possible point and recursively calculate the cuts needed for both parts.
3. **Memoization/Recursion:** The recursive approach explores all the possible partitions, calculating the minimum cuts. To optimize, memoization or dynamic programming could be used to avoid redundant calculations.

1. **isPalindrome(s, i, j)**:

   * Check if the substring from index `i` to `j` is a palindrome by comparing characters at `i` and `j` iteratively.

2. **palPartitionRec(s, i, j)**:

   * Base case: If `i >= j` or the substring is already a palindrome, return `0` (no cuts needed).
   * Otherwise, for each possible partition, recursively calculate the minimum cuts needed for both parts.
   * Return the minimum of all possible partitions.

3. **palPartition(s)**:

**Code:**

```java
        return true;
```

## Problem: Decode the String

Given an encoded string `s`, the task is to decode it. The encoding rule is:
`k[encodedString]`, where the `encodedString` inside the square brackets is being repeated exactly `k` times.

- The input string contains only lowercase English alphabets.
- `k` is guaranteed to be a positive integer.

**Note:**
- The test cases are generated such that the length of the output string will never exceed `10^5`.

**Example :**

```text
Input:
s = "3[a]2[bc]"

Output:
"aaabcbc"
```
**Input:**
- A string `s` containing digits, lowercase letters, and square brackets.

**Output:**
- The fully decoded string as per the rules.

**Goal:**
- Transform the encoded string into its decoded version by handling nested and multiple encodings correctly.

**🧮 Flowchart/ Diagram:**
- Start
  - For each character in the string:
    - If digit → compute number, push to number stack
    - If '[' → push to character stack
    - If ']' → pop until '[', repeat substring, and push result
    - If letter → push to character stack
  - After traversal, pop all characters to get the decoded result
- End

**Solution Approach:**
To solve this problem, we can use two stacks:
1. A stack for numbers: To store the integer k, which represents the number of times the substring needs to be repeated.
2. A stack for characters: To store the characters or substrings as we encounter them.

**The approach follows these steps:**
1. Traverse the string from left to right.
2. If we encounter a digit, calculate the number (which is k).
3. If we encounter [, push it onto the stack as an indicator to start a new substring.
4. If we encounter ], pop characters from the stack to form the substring, repeat it k times, and push it back onto the stack.
5. Once the traversal is complete, the stack will contain the decoded string.

**Pseudocode/Algorithm:**
```

function decodeString(s):
numStack = new Stack()
charStack = new Stack()

for char in s:
    if char is digit:
        accumulate full number and push to numStack
    else if char is '[':
        push char to charStack
    else if char is letter:
        push char to charStack
    else if char is ']':
        temp = ""
        while charStack.top != '[':
            temp = charStack.pop() + temp
        charStack.pop() // remove '['
        repeatCount = numStack.pop()
        repeated = temp * repeatCount
        for c in repeated:
            charStack.push(c)

result = ""
while not charStack.isEmpty():
    result = charStack.pop() + result
return result
```

**Code Solution:**
```java
import java.util.Stack;

class Solution {
    static String decodeString(String s) {
        // Stacks for numbers and characters
        Stack<Integer> numStack = new Stack<>();
        Stack<Character> charStack = new Stack<>();
        StringBuilder res = new StringBuilder();  // To store final result
        String temp = "";
        
        for (int i = 0; i < s.length(); i++) {
            int cnt = 0;
            
            // If current character is a digit
            if (Character.isDigit(s.charAt(i))) {
                // Compute the number (k) for repetitions
                while (Character.isDigit(s.charAt(i))) {
                    cnt = cnt * 10 + (s.charAt(i) - '0');
                    i++;
                }
                i--;
                numStack.push(cnt);  // Push number onto the stack
            } 
            // If current character is ']'
            else if (s.charAt(i) == ']') {
                temp = "";
                cnt = numStack.pop();  // Pop the repetition number

                // Pop characters until we find '['
                while (charStack.peek() != '[') {
                    temp = charStack.pop() + temp;
                }
                charStack.pop();  // Remove '[' from the stack
                
                StringBuilder repeated = new StringBuilder();
                // Repeat the decoded string `cnt` times
                for (int j = 0; j < cnt; j++) {
                    repeated.append(temp);
                }
                
                // Push the repeated string characters back onto the stack
                for (int j = 0; j < repeated.length(); j++) {
                    charStack.push(repeated.charAt(j));
                }
            } 
            // If current character is a letter
            else {
                charStack.push(s.charAt(i));  // Push character to the stack
            }
        }
        
        // Reconstruct the final decoded string
        while (!charStack.isEmpty()) {
            res.insert(0, charStack.pop());
        }

        return res.toString();
    }
}
```

**Time Complexity: O(n),** where n is the length of the string s. We traverse the string once, and each character is pushed and popped from the stacks at most once.

**Space Complexity: O(n),** where n is the length of the string s. In the worst case, we need space for the stack to store the characters and numbers.

**Notes:**
- Handles nested brackets like `"3[a2[c]]"` → `"accaccacc"`
- Number parsing supports multiple digits (e.g., `12[ab]`)
- Repetition is computed after closing bracket is processed

Given a string `s`, you can transform every letter individually to be lowercase or uppercase to create another string.  
Return a list of all possible strings we could create. The output can be returned in any order.

**Constraints:**
1. 1 <= s.length <= 12
2. s consists of lowercase English letters, uppercase English letters, and digits.

**Input:**
- A string `s` containing letters and digits.

**Output:**
- A list of strings containing all possible permutations by changing letter cases.

**Example :**
```text
Input: s = "a1b2"
Output: ["a1b2", "a1B2", "A1b2", "A1B2"]
```

**Goal:**
- Return all combinations of `s` with lowercase and uppercase transformations applied to the letters.

**🧮 Flowchart/ Diagram:**
- Start
  - For each character in the string:
    - If it's a letter:
      - Recursively branch with lowercase
      - Recursively branch with uppercase
    - If it's a digit:
      - Continue recursion with digit unchanged
  - On reaching end of string, add to result list
- End

**Approach:**
To generate all the permutations where the letters can be transformed to either lowercase or uppercase, we can use backtracking. The idea is to explore each character of the string and, if it's a letter, try both lowercase and uppercase possibilities. If it's a digit, no transformation is needed, and we just append it as is.

**Here are the steps to solve the problem:**
1. **Backtracking:**
    We start from the first character and recursively generate all possible permutations.

2. **For each character:**
    If it's a letter, we generate two recursive branches: one where the letter is lowercase and one where it is uppercase.

    If it's a digit, we just append it as is.

3. **Base Case:**

    When we reach the end of the string, we add the current permutation to the result list.

4. **Pruning:**
    We use backtracking to undo the changes after exploring both possibilities for each character.

**Pseudocode/Algorithm:**
```

function letterCasePermutation(s):
result = \[]
backtrack(index=0, current=""):
if index == len(s):
result.append(current)
return
if s\[index] is letter:
backtrack(index + 1, current + s\[index].toLower())
backtrack(index + 1, current + s\[index].toUpper())
else:
backtrack(index + 1, current + s\[index])
call backtrack()
return result

```

**Code Solution:**
```java
import java.util.*;

class Solution {
    public List<String> letterCasePermutation(String s) {
        List<String> result = new ArrayList<>();
        backtrack(s, 0, new StringBuilder(), result);
        return result;
    }

    private void backtrack(String s, int index, StringBuilder current, List<String> result) {
        // Base case: if we have reached the end of the string
        if (index == s.length()) {
            result.add(current.toString());
            return;
        }

        char ch = s.charAt(index);
        
        // If the character is a letter, explore both lowercase and uppercase options
        if (Character.isLetter(ch)) {
            current.append(Character.toLowerCase(ch));  // choose lowercase
            backtrack(s, index + 1, current, result);    // recurse
            current.deleteCharAt(current.length() - 1);  // backtrack
            
            current.append(Character.toUpperCase(ch));  // choose uppercase
            backtrack(s, index + 1, current, result);    // recurse
            current.deleteCharAt(current.length() - 1);  // backtrack
        } else {
            current.append(ch);  // if it's a digit, just append it
            backtrack(s, index + 1, current, result);    // recurse
            current.deleteCharAt(current.length() - 1);  // backtrack
        }
    }
}

```

**Time Complexity: O(2^N)**
The time complexity is exponential because, for each letter in the string, there are two possible transformations (lowercase and uppercase). So, for a string of length N, there are 2^N possible combinations.

**Space Complexity: O(N)**
The space complexity is dominated by the recursion stack and the space used to store the StringBuilder. The recursion depth is O(N) (the length of the string), and the space used for storing the results is also O(2^N) in the worst case.

**Notes:**
- Handles mixed characters like `"a1B2"`
- Efficient use of backtracking with in-place modification
- Backtracking enables restoration of state after exploring each path

## Problem: Sum-string

Given a string of digits, the task is to check if it is a ‘sum-string’. A string `S` is called a sum-string when it is broken down into more than one substring and the rightmost substring can be written as a sum of two substrings before it. The same property should be recursively true for substrings before it.

**Note:** The rightmost substring should not contain leading zeroes.

**Input:**
- A string `str` consisting of digits only.

**Output:**
- Return `1` if the string is a sum-string, else return `0`.

**Example:**

```text
str = "112358"

Output: 1

Explanation:
The string can be broken down as:
"1", "1", "2", "3", "5", "8"
(1 + 1 = 2, 1 + 2 = 3, 2 + 3 = 5, 3 + 5 = 8)

```

**Goal:**
- Validate if the string satisfies the recursive sum property for all its parts.

      - Else try another pair

  - Else return 0

**Solution Approach:**
**Steps:**
1. **Helper Function - string_sum():**

    A helper function to add two strings as numbers. The result is returned as a string. This function takes care of the carry during the addition of the strings, handling cases where the sum exceeds 10.

2. **Recursion - checkSumStrUtil():**

    This recursive function checks if the current substring of the string str follows the sum-string condition.

    It takes the string, the starting index, and the lengths of the two substrings, computes their sum, and compares it with the next substring.

    If the sum matches, it recursively checks the remaining part of the string.

3. **Main Function - isSumString():**

    This function iterates through the string and tries every possible pair of substrings (with different lengths) that can be added. It calls the recursive checkSumStrUtil() to check if a valid sum-string is found.

    If a valid sum-string is found, it returns 1. Otherwise, it returns 0.

```java
class Solution {
    // Helper function to add two strings as numbers and return the result as a string
    public static String string_sum(String str1, String str2) {
        if (str1.length() < str2.length()) {
            String temp = str1;
            str1 = str2;
            str2 = temp;
        }
        int m = str1.length();
        int n = str2.length();
        String ans = "";
        int carry = 0;
        
        for (int i = 0; i < n; i++) {
            int ds = ((str1.charAt(m - 1 - i) - '0') + (str2.charAt(n - 1 - i) - '0') + carry) % 10;
            carry = ((str1.charAt(m - 1 - i) - '0') + (str2.charAt(n - 1 - i) - '0') + carry) / 10;
            ans = Character.toString((char)(ds + '0')) + ans;
        }

        for (int i = n; i < m; i++) {
            int ds = (str1.charAt(m - 1 - i) - '0' + carry) % 10;
            carry = (str1.charAt(m - 1 - i) - '0' + carry) / 10;
            ans = Character.toString((char)(ds + '0')) + ans;
        }

        if (carry != 0) {
            ans = Character.toString((char)(carry + '0')) + ans;
        }
        return ans;
    }

    // Helper function to check if the substring follows the sum-string property
    public static boolean checkSumStrUtil(String str, int beg, int len1, int len2) {
        String s1 = str.substring(beg, beg + len1);
        String s2 = str.substring(beg + len1, beg + len1 + len2);
        String s3 = string_sum(s1, s2);

        int s3_len = s3.length();
        if (s3_len > str.length() - len1 - len2 - beg)
            return false;

        if (s3.equals(str.substring(beg + len1 + len2, beg + len1 + len2 + s3_len))) {
            if (beg + len1 + len2 + s3_len == str.length()) {
                return true;
            }
            return checkSumStrUtil(str, beg + len1, len2, s3_len);
        }
        return false;
    }

    // Main function to check if the string is a sum-string
    public int isSumString(String str) {
        int n = str.length();
        for (int i = 1; i < n; i++)
            for (int j = 1; i + j < n; j++)
                if (checkSumStrUtil(str, 0, i, j))
                    return 1;

        return 0;
    }
}
```

**Space Complexity: O(n),** where n is the length of the string, due to the recursion stack and temporary storage for substrings and results.

**Notes:**
- Leading zeroes are invalid unless the substring is exactly "0".
- Handles recursive validation efficiently with two-level loop and recursion.

## Problem: Word Boggle

Given a dictionary of distinct words and an M x N board where every cell has one character, your task is to find all possible words from the dictionary that can be formed by a sequence of adjacent characters on the board. The movement can be in any of the 8 directions (horizontally, vertically, and diagonally). A cell can be used only once in one word.

**Note:**
- While forming a word, you can move to any of the 8 adjacent cells.
- A cell can only be used once for each word.

**Input:**
- `board`: a 2D character matrix representing the boggle board.
- `dictionary`: an array of distinct strings representing valid words.

**Output:**
- Return an array of words from the dictionary that can be formed from the board.

**Example:**
```text
Input:
board = [
  ['o', 'a', 'n'],
  ['e', 't', 'a'],
  ['i', 'h', 'k'],
  ['r', 'o', 'p']
]
dictionary = ["oath", "pea", "eat", "rain"]

Output:
["eat", "oath"]
```

**Goal:**
- Identify all dictionary words that can be constructed by traversing the board via adjacent cells in any of the 8 directions.

**Solution Approach:**
To solve this problem, we use Depth-First Search (DFS) to explore all possible paths for each word in the dictionary. The approach can be broken down into the following steps:

1. **DFS Search:**
    For each word in the dictionary, we start the search by matching the first character of the word with every character in the board.
    If a match is found, recursively search in all 8 directions (up, down, left, right, and diagonals).
    If a match is found for the entire word, return true. Otherwise, backtrack by marking the cell as visited and continue searching.

2. **Dictionary Matching:**
    We store the words in the dictionary into a set for fast lookup.
    For each word in the dictionary, check if it can be formed on the board using the DFS helper function.

3. **Handling Board Boundaries:**
    Ensure that the DFS search does not go out of bounds of the board.
    Mark cells as visited to avoid reusing the same cell in one word.

**🧮 Flowchart/ Diagram:**
- Start
  - For each word in dictionary
    - For each cell in the board
      - If cell char matches word[0]
        - DFS to find full word
          - If found, add to result
  - Return result list
- End

**Pseudocode/Algorithm:**
```

function wordBoggle(board, dictionary):

for i in 0 to rows:
for j in 0 to cols:

add word to result
break

if index == len(word): return true

unmark board\[i]\[j]

```

**Code Solution**
```java
import java.util.*;

class Solution {
    // Helper function to check if a word exists in the board starting from (r, c)
    public static boolean exist(char board[][], String word, int r, int c) {
        for (int i = 0; i < r; i++) {
            for (int j = 0; j < c; j++) {
                if (board[i][j] == word.charAt(0) && search(board, word, 0, i, j, r, c)) {
                    return true;
                }
            }
        }
        return false;
    }

    // Function to perform DFS search for the word on the board
    public static boolean search(char board[][], String word, int len, int i, int j, int r, int c) {
        if (i < 0 || i >= r || j < 0 || j >= c)
            return false;
        
        if (board[i][j] != word.charAt(len))
            return false;

        if (len == word.length() - 1)
            return true;

        char ch = board[i][j];
        board[i][j] = '@';  // Mark the cell as visited

        boolean ans = 
            search(board, word, len + 1, i - 1, j, r, c) || 
            search(board, word, len + 1, i + 1, j, r, c) || 
            search(board, word, len + 1, i, j - 1, r, c) || 
            search(board, word, len + 1, i, j + 1, r, c) || 
            search(board, word, len + 1, i - 1, j + 1, r, c) || 
            search(board, word, len + 1, i - 1, j - 1, r, c) || 
            search(board, word, len + 1, i + 1, j - 1, r, c) || 
            search(board, word, len + 1, i + 1, j + 1, r, c);

        board[i][j] = ch;  // Backtrack
        return ans;
    }

    // Main function to find all words in the dictionary that can be formed in the board
    public String[] wordBoggle(char board[][], String[] dictionary) {
        int r = board.length;
        int c = board[0].length;

        ArrayList<String> temp = new ArrayList<>();
        Set<String> set = new HashSet<>();

        // Add dictionary words to the set for fast lookup
        for (String word : dictionary) {
            set.add(word);
        }
        int n = set.size();

        // Convert the set back to an array
        String[] dict = new String[n];
        int id = 0;
        for (String s : set) {
            dict[id++] = s;
        }

        // For each word in the dictionary, check if it can be formed on the board
        for (int i = 0; i < dict.length; i++) {
            if (exist(board, dict[i], r, c)) {
                temp.add(dict[i]);
            }
        }

        // Convert the list of found words to an array
        String[] ans = new String[temp.size()];
        int idx = 0;
        for (String word : temp) {
            ans[idx++] = word;
        }

        return ans;
    }
}
```

**Time Complexity: O(R * C * L),** where R is the number of rows, C is the number of columns, and L is the average length of the words in the dictionary. For each word, we explore all the cells of the board.

**Space Complexity: O(R * C),** since we store the board and use extra space for the search recursion stack

**Notes:**
- Efficiently handles checking multiple dictionary words.
- Prevents duplicate results using a set.

## Problem: Boggle (Find All Possible Words in a Board of Characters) | Set 1

Find all the words from the list `words[]` that can be constructed by consecutively linking adjacent characters in the given `n × m` matrix of characters `letters[][]`. You are allowed to move in any of the 8 possible directions (horizontally, vertically, or diagonally), but each cell in the matrix can be used only once per word.

You are given a list of strings `words[]` and an `n × m` matrix of characters `letters[][]`. The objective is to locate all the words from `words[]` that can be constructed by consecutively linking adjacent characters in the matrix. 

- You can move in any of the 8 possible directions (horizontally, vertically, or diagonally).
- Each cell in the matrix can only be used once per word.

**Example**

**Input:**
```java
words = ["geeks", "for", "quiz", "go"]
letters = [
    {'g', 'i', 'z'},
    {'u', 'e', 'k'},
    {'q', 's', 'e'}
]
````

**Output:**

```text
geeks for go
```

**Constraints:**

* `1 <= letters.length, letters[0].length <= 10`
* `1 <= words.length <= 10`
* Word lengths in `words[]` are at most 10.
* The matrix contains lowercase English alphabets.

**Goal:**

* Find all the words from `words[]` that can be formed by traversing the matrix.
* Each word must be constructed by consecutively linking adjacent characters in the matrix.
* Each cell in the matrix can be used only once per word.

**Approach:**

1. **DFS Traversal**:

   * Perform a Depth-First Search (DFS) from each character in the matrix that matches the first character of the word.
   * Try to build the word by moving in all 8 possible directions.
   * If the word is found, mark it as valid.

2. **Backtracking**:

   * If a word cannot be formed, backtrack by unmarking the current cell and trying other possible paths.

3. **Result Set**:

   * Use a set to store valid words to avoid duplicates.

4. **Convert to List**:

   * Convert the set to a list for the final output.

**🧮 Flowchart/ Diagram:**

* A flowchart can be drawn showing the following steps:

  1. Start
  2. Loop through each word in the dictionary.
  3. For each word, start DFS from each cell in the matrix.
  4. If a match is found, add it to the result set.
  5. Return the list of valid words.

**Pseudocode/Algorithm:**

1. **DFS Function**:

   * Start from each character in the matrix.
   * Check if the character matches the current character in the word.
   * If a match is found, recursively explore all 8 possible directions.
   * Mark cells as visited during the search and backtrack when necessary.

2. **Main Function**:

   * For each word in the input list `words[]`, attempt to form the word by calling the DFS function for every cell in the matrix.

**Code:**

```java
import java.util.*;

class GfG {

    // function to perform dfs on the grid
    static boolean dfs(char[][] letters, String word, int i, int j, int index) {

        // check if the current cell is out of bounds
        if (i < 0 || i >= letters.length || j < 0 || j >= letters[0].length) {
            return false;
        }

        // check if the current cell matches 
        // the character in the word
        if (letters[i][j] != word.charAt(index)) {
            return false;
        }

        // check if we have found the complete word
        if (index == word.length() - 1) {
            return true;
        }

        // mark the current cell as visited
        char temp = letters[i][j];
        letters[i][j] = '#';

        // perform dfs on all 8 directions
        for (int row = -1; row <= 1; row++) {
            for (int col = -1; col <= 1; col++) {

                // skip the current cell
                if (row == 0 && col == 0) continue;
                if (dfs(letters, word, i + row, j + col, index + 1)) {

                    // unmark the current cell as visited
                    letters[i][j] = temp;
                    return true;
                }
            }
        }

        // unmark the current cell as visited
        letters[i][j] = temp;

        return false;
    }

    // find all words in a given grid of characters
    // and a given dictionary
    static List<String> findWords(List<String> words, char[][] letters) {
        int n = letters.length, m = letters[0].length;
        int r = words.size();
        List<String> ans = new ArrayList<>();

        // store the unique words in the hashSet
        Set<String> result = new HashSet<>();

        for (int i = 0; i < r; i++) {
            for (int j = 0; j < n; j++) {
                for (int k = 0; k < m; k++) {
                    if (dfs(letters, words.get(i), j, k, 0)) {
                        result.add(words.get(i));
                    }
                }
            }
        }

        // convert the set to vector
        for (String word : result) {
            ans.add(word);
        }

        return ans;
    }

    public static void main(String[] args) {
        List<String> words = Arrays.asList("geeks", "for", "quiz", "go");
        char[][] letters = {
            {'g', 'i', 'z'},
            {'u', 'e', 'k'},
            {'q', 's', 'e'}
        };
        List<String> ans = findWords(words, letters);
        for (String word : ans) {
            System.out.print(word + " ");
        }
    }
}
```

**Time Complexity:**

* **Time Complexity**: O(N \* M \* W \* 8^L), where:

  * N and M are the dimensions of the matrix (`letters[][]`).
  * W is the number of words in the input list `words[]`.
  * L is the average length of the words in the input list.
  * 8^L arises because we can move in any of 8 directions at each step, and for each word, we explore all possible paths.

**Space Complexity:**

* **Space Complexity**: O(N \* M), where N and M are the dimensions of the matrix, as we store the visited state of each cell during the DFS traversal.

**Notes:**

* This approach works efficiently for small to medium-sized matrices and word lists.
* For large matrices or word lists, optimizations such as using a Trie or precomputing possible paths might be necessary to improve performance.


## Problem: Largest Number in K Swaps

Given a number `k` and a string `s` of digits denoting a positive integer, the task is to build the largest number possible by performing swap operations on the digits of `s` at most `k` times.

You are allowed to swap the digits of the number `k` times (or fewer). Your task is to return the largest possible number that can be obtained after at most `k` swaps.

**Input:**
- `s`: A string of digits representing a positive integer.
- `k`: An integer representing the maximum number of swaps allowed.

**Output:**
- Return the largest possible number that can be formed by performing at most `k` swaps on the digits of `s`.

**Example :**

```text
Input: 
s = "12345", k = 1

Output:
"52341"

Explanation:
Performing one swap between 1 and 5 gives the largest possible number.
```

**Goal:**
- Maximize the number by performing at most `k` swaps.

**🧮 Flowchart/ Diagram:**
- Start
  - If k == 0, return current string.
  - For each pair of digits, check if a swap would increase the number.
    - Perform the swap.
    - Recursively reduce k and repeat.
    - Undo the swap after recursion.
  - Return the largest number found.
- End

**Solution Approach:**
The solution relies on performing greedy swaps to maximize the number. We recursively find the largest number by performing the following steps:

   1. If k is 0, return the current number since no more swaps can be made.
   2. For each pair of digits in the string s, we attempt to swap the digits and recursively find the largest number.
   3. After each swap, compare the result with the current maximum, and if the result is larger, update the maximum.
   4. Undo the swap after the recursion step to explore other possibilities.
   5. Repeat until all swaps are exhausted or the best result is found.

**Pseudocode/Algorithm:**
```

function findMax(s, k):
if k == 0:
return s
ans = s
for i = 0 to len(s)-1:
for j = i+1 to len(s):
if s\[i] < s\[j]:
swap(s, i, j)
recResult = findMax(s, k-1)
ans = max(ans, recResult)
swap(s, i, j)
return ans

function findMaximumNum(s, k):
return findMax(s, k)

```

**Code Solution:**
```java
class Solution {
    // Function to swap two characters in the string
    static String swap(String s, int i, int j) {
        char[] arr = s.toCharArray();
        char temp = arr[i];
        arr[i] = arr[j];
        arr[j] = temp;
        return new String(arr);
    }

    // Recursive function to find the largest number after at most k swaps
    static String findMax(String s, int k) {
        if (k == 0) {
            return s; // No more swaps left
        }

        int n = s.length();
        String ans = s; // Initially, the best answer is the current string

        // Try all pairs of digits and swap them
        for (int i = 0; i < n - 1; i++) {
            for (int j = i + 1; j < n; j++) {
                // Swap only if the current digit is less than the digit after it
                if (s.charAt(i) < s.charAt(j)) {
                    // Swap the digits
                    s = swap(s, i, j);
                    
                    // Recur with one less swap available
                    String recResult = findMax(s, k - 1);
                    
                    // If the new result is better, update the answer
                    if (recResult.compareTo(ans) > 0) {
                        ans = recResult;
                    }

                    // Undo the swap to explore other possibilities
                    s = swap(s, i, j);
                }
            }
        }

        return ans; // Return the largest number found
    }

    // Function to find the largest number after k swaps
    public static String findMaximumNum(String s, int k) {
        // Start the recursive function
        return findMax(s, k);
    }
}
```


**Time Complexity: O(n^2 * k),** where n is the length of the string s, and k is the number of swaps allowed. For each swap, we explore all pairs of digits and recursively find the largest result.

**Space Complexity: O(n),** due to the recursive stack and the space used by the string manipulation.

**Notes:**
- The solution uses a recursive approach to explore all possible swaps.
- To optimize, we only swap digits that can increase the number.
- The process backtracks by undoing the swap to explore other possibilities.

## Problem: Partition Array to K Subsets

Given an integer array `arr[]` and an integer `k`, the task is to check if the array `arr[]` could be divided into `k` non-empty subsets with equal sum of elements.

- All elements of this array should be part of exactly one partition.
- The sum of elements in each subset should be equal.

**Constraints:**
 1. The length of the array arr is at most 15.

 2. The number of subsets k is at most 5.

**Input:**
- `arr[]`: An array of integers representing the elements to be partitioned.
- `k`: An integer representing the number of subsets.

**Output:**
- Return `true` if the array can be partitioned into `k` non-empty subsets with equal sum of elements; otherwise, return `false`.

**Example :**
```text
Input: 
arr = [4, 3, 2, 3, 5, 2, 1], k = 4

Output: 
true
```

**Goal:**
- Partition the array into `k` subsets such that the sum of elements in each subset is equal.

**Approach:**
To solve the problem, we can utilize a dynamic programming approach with bitmasking.

1. **Initial Checks:**

    If k == 1, then we can always partition the array into one subset, so return true.

    If the number of elements n in the array is less than k, it's impossible to divide the array into k subsets, so return false.

    Calculate the sum of all elements in the array. If the sum is not divisible by k, return false because the target sum per subset would not be an integer.

2. **Dynamic Programming (DP) with Bitmasking:**

    Use a bitmask to represent the subsets of the array. Each bit of the mask will represent whether the corresponding element in the array is included in the subset or not.

    Use a dp array where dp[mask] represents the current sum of elements in the subset corresponding to the bitmask mask.

    Iterate through all possible bitmasks and update the dp array accordingly. If we can partition the elements such that each subset has the same sum, we return true.

**🧮 Flowchart/ Diagram:**
- Start
  - If k == 1, return true.
  - If the length of the array is less than k, return false.
  - Calculate the total sum of elements in the array.
  - If the total sum is not divisible by k, return false.
  - Initialize a dynamic programming array with bitmasking representation to store possible subset sums.
  - Iterate over all bitmasks and try to form subsets.
  - Check if the full set can be partitioned into k subsets with equal sum.
- End

**Pseudocode/Algorithm:**
```

function isKPartitionPossible(arr, k):
if k == 1:
return true
if len(arr) < k:
return false
sum = sum(arr)
if sum % k != 0:
return false
target = sum / k

dp = array of size 2^n, initialize dp[0] = 0
for mask from 0 to 2^n-1:
    if dp[mask] == -1:
        continue
    for i from 0 to n-1:
        if element i not in current subset and adding i doesn't exceed target:
            dp[mask | (1 << i)] = (dp[mask] + arr[i]) % target
return dp[(1 << n) - 1] == 0

```

**Code Solution:**
```java
import java.util.Arrays;

class Solution {
    public boolean isKPartitionPossible(int[] arr, int k) {
        // Your code here
        int n = arr.length;

        // If k is 1, always possible to partition
        if (k == 1)
            return true;

        // If n is less than k, we can't partition into k subsets
        if (n < k)
            return false;

        // Calculate the sum of all elements in the array
        int sum = 0;
        for (int i = 0; i < n; i++)
            sum += arr[i];

        // If sum is not divisible by k, we cannot partition equally
        if (sum % k != 0)
            return false;
        int target = sum / k;  // Target sum for each subset

        // DP array to store the current sum of the subsets
        int[] dp = new int[1 << n];
        Arrays.fill(dp, -1);
        dp[0] = 0;  // Initial state, empty set has sum 0

        // Iterate through all possible subsets (bitmasks)
        for (int mask = 0; mask < (1 << n); mask++) {
            if (dp[mask] == -1)
                continue;  // Skip invalid state

            // Try adding each element to the subset
            for (int i = 0; i < n; i++) {
                if ((mask & (1 << i)) == 0  // If element i is not in current set
                    && dp[mask] + arr[i] <= target) {  // If adding the element does not exceed target sum
                    // Update the dp state
                    dp[mask | (1 << i)] = (dp[mask] + arr[i]) % target;
                }
            }
        }
        // The full set (all elements) should be partitioned into k subsets with equal sum
        return dp[(1 << n) - 1] == 0;
    }
}
```

**Time Complexity: O(2^n * n),** where n is the length of the array. We iterate over all subsets (2^n possible subsets), and for each subset, we check n elements.

**Space Complexity: O(2^n),** due to the bitmasking representation of subsets.

**Notes:**
- The bitmask represents the inclusion or exclusion of elements in the subset.
- Dynamic programming helps to efficiently check all possible partitions of the array.
- The space complexity is O(2^n) due to the bitmasking array that stores the state of all subsets.

## Problem: Partition of a Set into K Subsets with Equal Sum

Given an integer array `arr[]` and an integer `k`, the task is to check if it is possible to divide the given array into `k` non-empty subsets of equal sum such that every array element is part of a single subset.

**Example:**
**Input:**
```
arr\[] = { 2, 1, 4, 5, 3, 3 }
k = 3
```

**Output:**
```
true
```

In this example, it is possible to divide the array into 3 subsets such that the sum of each subset is equal.

**Constraints:**
- `1 ≤ n ≤ 16` (where `n` is the number of elements in the array)
- `1 ≤ k ≤ n` (the number of subsets)
- Elements in the array can be positive integers.

**Goal:**
The goal is to determine whether it is possible to partition the given array `arr[]` into `k` subsets such that each subset has the same sum.

**Approach:**

1. **Check for Initial Conditions:**
   - If `k == 1`, it is always possible to partition the array into one subset (the entire array).
   - If the number of elements `n` is less than `k` or if the total sum of the array is not divisible by `k`, then it is impossible to partition the array into `k` subsets.

2. **Backtracking Approach:**
   - We use a backtracking technique to try forming `k` subsets. Starting with an empty subset, we attempt to place elements in each subset, ensuring that no subset exceeds the target sum (which is the total sum divided by `k`).
   - We recursively try to assign each element to one of the subsets, and if we reach a situation where all `k` subsets are filled with equal sum, we return `true`.

3. **Subset Formation:**
   - The algorithm proceeds by trying to place each element in the current subset, backtracking when it can't fit.
   - If a subset is filled, it moves to the next subset and continues the process.

4. **Optimization:**
   - We use a boolean array `taken[]` to track which elements have already been included in a subset.
   - We also use an array `subsetSum[]` to track the sum of each subset.

**🧮 Flowchart/ Diagram:**
```

Start
|
\|--> Calculate sum of all elements in the array
|
\|--> If sum % k != 0 or n < k, return false (partition not possible)
|
\|--> Initialize subsetSum\[0] with the last element
|
\|--> Call isKPartitionPossible function
|
\|--> Try to form k subsets using backtracking
|
\|--> If all subsets are filled with the target sum, return true
|
End

```

**Pseudocode/Algorithm:**

```text
function isKPartitionPossible(arr, subsetSum, taken, target, k, n, currIdx, limitIdx):
    if subsetSum[currIdx] == target:
        if currIdx == k - 2:
            return true
        return isKPartitionPossible(arr, subsetSum, taken, target, k, n, currIdx + 1, n - 1)

    for i in limitIdx to 0:
        if taken[i]:
            continue
        
        temp = subsetSum[currIdx] + arr[i]
        if temp <= target:
            taken[i] = true
            subsetSum[currIdx] += arr[i]
            
            if isKPartitionPossible(arr, subsetSum, taken, target, k, n, currIdx, i - 1):
                return true

            taken[i] = false
            subsetSum[currIdx] -= arr[i]
    return false

function isKPartitionPossible(arr, k):
    n = length of arr
    sum = sum of all elements in arr

    if k == 1:
        return true

    if n < k or sum % k != 0:
        return false

    target = sum / k
    subsetSum = array of size k initialized to 0
    taken = array of size n initialized to false

    subsetSum[0] = arr[n - 1]
    taken[n - 1] = true

    return isKPartitionPossible(arr, subsetSum, taken, target, k, n, 0, n - 1)
```

**Code:**

```java
import java.util.Arrays;

class GfG {

    static boolean isKPartitionPossible(int[] arr, int[] subsetSum,
                                         boolean[] taken, int target, int k,
                                         int n, int currIdx, int limitIdx) {

        // If the current subset sum matches the target
        if (subsetSum[currIdx] == target) {

            // If all but one subset are filled, the last
            // subset is guaranteed to work
            if (currIdx == k - 2)
                return true;
            return isKPartitionPossible(arr, subsetSum,
                                        taken, target, k, n,
                                        currIdx + 1, n - 1);
        }

        // Try including each element in the current subset
        for (int i = limitIdx; i >= 0; i--) {
            if (taken[i])  
                continue;

            int temp = subsetSum[currIdx] + arr[i];
            if (temp <= target) {

                // Only proceed if it doesn't exceed the target
                taken[i] = true;
                subsetSum[currIdx] += arr[i];

                if (isKPartitionPossible(arr, subsetSum,
                                         taken, target, k,
                                         n, currIdx, i - 1))
                    return true;

                // Backtrack
                taken[i] = false;
                subsetSum[currIdx] -= arr[i];
            }
        }
        return false;
    }

    static boolean isKPartitionPossible(int[] arr, int k) {
        int n = arr.length;
        int sum = Arrays.stream(arr).sum();

        // If only one subset is needed, it's always possible
        if (k == 1)
            return true;

        // Check if partition is impossible
        if (n < k || sum % k != 0)
            return false;

        int target = sum / k;
        int[] subsetSum = new int[k];
        boolean[] taken = new boolean[n];

        // Initialize the first subset with the last element
        subsetSum[0] = arr[n - 1];
        taken[n - 1] = true;

        // Recursively check for partitions
        return isKPartitionPossible(arr, subsetSum, taken,
                                    target, k, n, 0, n - 1);
    }

    public static void main(String[] args) {
        int[] arr = { 2, 1, 4, 5, 3, 3 };
        int k = 3;

        if (isKPartitionPossible(arr, k))
            System.out.println("true");
        else
            System.out.println("false");
    }
}
```

**Time Complexity:**

* **O(2^n)**: The time complexity of the algorithm is exponential due to the backtracking approach, where we recursively try to place each element into the subsets.

**Space Complexity:**

* **O(n + k)**: The space complexity is mainly for storing the `taken[]` array (of size `n`) and the `subsetSum[]` array (of size `k`).

**Notes:**

* This problem is an example of a **backtracking problem** where we try to partition the array into subsets, checking each possibility recursively.
* The solution works for smaller values of `n` and `k`. As the number of elements grows, the problem becomes computationally expensive.
