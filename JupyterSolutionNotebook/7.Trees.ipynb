## Problem Statement: 332
**Burning Tree**

Given a **binary tree** and a node data called **target**, find the **minimum time required to burn the complete binary tree** if the target node is set on fire. It is known that in 1 second all nodes connected to a given node get burned. That is its left child, right child, and parent.

**Example**
**Input:**
     1
    / \
   2   3
  / \   / \
 4   5  6  7
/ \
8  9

- **Target node value:** `5`

**Output:**
4

**Explanation:**
- At time 0, node `5` is set on fire.
- At time 1, nodes `2`, `4`, and `6` catch fire.
- At time 2, nodes `1`, `3`, and `8` catch fire.
- At time 3, nodes `7` and `9` catch fire.
- At time 4, all nodes are burned.

The minimum time required to burn the entire tree is `4` seconds.

**Constraints**
- `1 ≤ Number of nodes ≤ 10^4`
- `-10^4 ≤ Node value ≤ 10^4`
- The tree contains unique values.

**Approach**
1. **Breadth-First Search (BFS)**:
   - First, perform a BFS to **map each node to its parent node** using a queue. This will help us know which nodes are connected to the target.
   
2. **Starting Point**:
   - The **target node** is set on fire first. We need to calculate the time it takes for the fire to propagate through the entire tree.
   
3. **Propagation of Fire**:
   - Each second, the fire spreads to all nodes that are **connected** to the nodes that are already on fire. This includes the **parent, left child, and right child** of the current nodes.

4. **Time Calculation**:
   - The BFS will continue to propagate the fire level by level, and the time taken will be equal to the number of levels traversed by BFS.
   
5. **Time Complexity**:
   - **O(n)** where `n` is the number of nodes in the tree. Each node is visited exactly once.
   
6. **Space Complexity**:
   - **O(n)** due to the space used by the queue and hash maps.

**Java Code**
```java
class Solution {
    /*class Node {
        int data;
        Node left;
        Node right;

        Node(int data) {
            this.data = data;
            left = null;
            right = null;
        }
    }*/

    public static int minTime(Node root, int target) {
        // If the root is null, return -1
        if (root == null)
            return -1;

        // Queue for BFS traversal
        Queue<Node> q = new LinkedList<>();
        q.add(root);
        Node tar = null;

        // HashMap to map child nodes to their parent nodes
        Map<Node, Node> par = new HashMap<>();
        par.put(root, null);

        // Perform BFS to find the target node and map parent nodes
        while (!q.isEmpty()) {
            Node curr = q.poll();

            // Set tar = curr if the current node's data is equal to the target
            if (curr.data == target)
                tar = curr;

            // Map the left child to its parent
            if (curr.left != null) {
                par.put(curr.left, curr);
                q.add(curr.left);
            }

            // Map the right child to its parent
            if (curr.right != null) {
                par.put(curr.right, curr);
                q.add(curr.right);
            }
        }

        // HashMap to check if a node has been visited or not
        Map<Node, Boolean> vis = new HashMap<>();

        // Variable to track the time
        int ans = -1;

        // Start the BFS from the target node
        q.add(tar);

        // Perform BFS to propagate the fire to all nodes
        while (!q.isEmpty()) {
            int size = q.size();
            while (size-- > 0) {
                Node curr = q.poll();
                vis.put(curr, true);

                // Add left child to the queue if it's not visited
                if (curr.left != null && !vis.getOrDefault(curr.left, false))
                    q.add(curr.left);
                
                // Add right child to the queue if it's not visited
                if (curr.right != null && !vis.getOrDefault(curr.right, false))
                    q.add(curr.right);
                
                // Add parent node to the queue if it's not visited
                if (par.get(curr) != null && !vis.getOrDefault(par.get(curr), false))
                    q.add(par.get(curr));
            }
            ans++; // Increment the time after each level of BFS
        }

        // Return the minimum time required to burn the entire tree
        return ans;
    }
}


## Problem Statement: 333
**Max and Min element in Binary Tree**

Given a **Binary Tree**, find the **maximum** and **minimum** elements in it.

**Example**
**Input:**
     1
    / \
   2   3
  / \
 4   5

**Output:**
Maximum: 5 Minimum: 1

**Explanation:**
- The maximum element in the tree is `5`.
- The minimum element in the tree is `1`.

**Constraints**
- `1 ≤ Number of nodes ≤ 10^4`
- `-10^4 ≤ Node value ≤ 10^4`

**Approach**
We can solve this problem by performing a **Depth First Search (DFS)** traversal of the binary tree and keeping track of the maximum and minimum values encountered during the traversal.

1. **Recursive Approach**:
   - We will recursively traverse both left and right subtrees.
   - For each node, we compare the node value with the results from the left and right subtrees and update the maximum and minimum values accordingly.

2. **Time Complexity**:
   - **O(n)** where `n` is the number of nodes in the tree, as each node is visited once.

3. **Space Complexity**:
   - **O(h)** where `h` is the height of the tree, due to the recursive call stack.

**Java Code**
```java
class Solution {
    // Function to find the maximum value in the binary tree
    public static int findMax(Node node) {
        if (node == null)
            return Integer.MIN_VALUE; // Base case: empty tree has minimum value

        int res = node.data; // Start with the node's value

        // Recursively find the maximum value in the left and right subtrees
        int lres = findMax(node.left);
        int rres = findMax(node.right);

        // Compare the current node's value with the maximum values of the left and right subtrees
        if (lres > res)
            res = lres;
        if (rres > res)
            res = rres;

        return res; // Return the maximum value found
    }

    // Function to find the minimum value in the binary tree
    public static int findMin(Node node) {
        if (node == null)
            return Integer.MAX_VALUE; // Base case: empty tree has maximum value

        int res = node.data; // Start with the node's value

        // Recursively find the minimum value in the left and right subtrees
        int lres = findMin(node.left);
        int rres = findMin(node.right);

        // Compare the current node's value with the minimum values of the left and right subtrees
        if (lres < res)
            res = lres;
        if (rres < res)
            res = rres;

        return res; // Return the minimum value found
    }
}


## Problem Statement: 334
**Kth Largest Element in BST**

Given a **Binary Search Tree (BST)**, your task is to complete the function which will return the **kth largest element** without modifying the BST.

**Example**
**Input:**
     5
    / \
   3   8
  / \  / \
 2  4 7   9


**k = 3**

**Output:**
7

**Explanation:**
- The tree has the following inorder traversal: `2, 3, 4, 5, 7, 8, 9`.
- The 3rd largest element in this tree is `7`.

**Constraints**
- `1 ≤ Number of nodes ≤ 10^4`
- `1 ≤ k ≤ Number of nodes`

**Approach**
To find the **kth largest element** in a BST, we can perform a **reverse inorder traversal**. The reverse inorder traversal of a BST visits nodes in descending order (right subtree, root, left subtree). By keeping track of the number of nodes visited, we can easily return the **kth largest** element.

1. **Iterative Approach**:
   - Use a **stack** to simulate the recursion of an inorder traversal.
   - Traverse the tree in reverse (right, root, left) until we visit the **kth largest element**.
   
2. **Time Complexity**:
   - **O(h + k)**, where `h` is the height of the tree and `k` is the position of the largest element to be found.
   - The height of the tree `h` will be visited while pushing nodes to the stack.

3. **Space Complexity**:
   - **O(h)** for the stack, where `h` is the height of the tree.

**Java Code**
```java
class Solution {
    // Function to return the Kth largest element in the given BST rooted at 'root'
    public int kthLargest(Node root, int k) {
        // Stack to simulate the inorder traversal
        Stack<Node> st = new Stack<>();
        Node curr = root;
        int count = 0;

        // Iterative reverse inorder traversal
        while (curr != null || !st.isEmpty()) {
            // Traverse to the rightmost node
            while (curr != null) {
                st.push(curr);
                curr = curr.right;
            }

            // Pop from the stack (current node is the next largest element)
            curr = st.pop();
            count++;  // Increment the counter for the number of nodes visited
            
            // If count reaches k, we have found the kth largest element
            if (count == k) {
                return curr.data;
            }

            // Move to the left subtree
            curr = curr.left;
        }

        return -1;  // In case k is out of bounds (not enough elements)
    }
}


## Problem Statement: 335
**Check for BST**

Given the **root** of a binary tree, check whether it is a **Binary Search Tree (BST)** or not.

**Note:**  
- A **BST** is defined as follows:
  - The left subtree of a node contains only nodes with keys less than the node's key.
  - The right subtree of a node contains only nodes with keys greater than the node's key.
  - Both the left and right subtrees must also be **Binary Search Trees**.

- **BSTs cannot contain duplicate nodes.**

**Example**
**Input:**
     5
    / \
   3   8
  / \  / \
 2  4 6   9

**Output:**
true

**Explanation:**
- The tree satisfies the properties of a Binary Search Tree (BST):
  - Left subtree of node `5`: `[3, 2, 4]` (all are less than `5`).
  - Right subtree of node `5`: `[8, 6, 9]` (all are greater than `5`).
  - All subtrees of the nodes are also binary search trees.

**Constraints**
- `1 ≤ Number of nodes ≤ 10^4`
- `-10^4 ≤ Node value ≤ 10^4`

**Approach**
To check if a binary tree is a **Binary Search Tree (BST)**, we need to validate the BST property recursively for every node in the tree. The key condition for a node to be part of a BST is that:
1. The left child must have a value smaller than the node.
2. The right child must have a value greater than the node.

We can use a helper function that checks if each node's value lies within a valid range (bounded by `min` and `max` values):
- For the left child, the upper bound becomes the current node's value minus 1.
- For the right child, the lower bound becomes the current node's value plus 1.

**Time and Space Complexity**
- **Time Complexity**:  
  - **O(n)**, where `n` is the number of nodes in the tree, since we visit each node exactly once.
  
- **Space Complexity**:  
  - **O(h)**, where `h` is the height of the tree, due to the recursive call stack (in the worst case, the height can be equal to `n` if the tree is skewed).

**Java Code**
```java
class Solution {
    // Helper function to check the BST property
    static boolean isBSTUtil(Node node, int min, int max) {
        // If the node is null, it is a valid BST
        if (node == null) return true;
        
        // If the node's value is out of the valid range, return false
        if (node.data < min || node.data > max) return false;
        
        // Recursively check the left and right subtrees
        return isBSTUtil(node.left, min, node.data - 1) &&
               isBSTUtil(node.right, node.data + 1, max);
    }

    // Function to check if the tree is a BST
    boolean isBST(Node root) {
        // Call the helper function with initial range
        return isBSTUtil(root, Integer.MIN_VALUE, Integer.MAX_VALUE);
    }
}


## Problem Statement: 336
**Closest in BST**

Given a **Binary Search Tree (BST)** and an integer `k`, find the least absolute difference between any node value of the BST and the given integer `k`.

**Example**
**Input:**
    10
   /  \
  5    15
 / \     \
2   7    20


**k = 12**

**Output:**
2

**Explanation:**
- The closest node value to `12` in the BST is `10`, and the absolute difference is `|12 - 10| = 2`.
- Hence, the output is `2`.

**Constraints**
- `1 ≤ Number of nodes ≤ 10^4`
- `-10^4 ≤ Node value ≤ 10^4`
- `-10^4 ≤ k ≤ 10^4`

**Approach**
To find the closest value to the given integer `k` in the BST, we can leverage the BST properties. A Binary Search Tree has the following properties:
1. Left child values are smaller than the parent node's value.
2. Right child values are greater than the parent node's value.

Using this property, we can perform a **binary search** starting from the root:
1. For each node, we calculate the absolute difference between the node's value and `k`.
2. If the current node's value is closer to `k` than the previously found closest value, we update the closest value.
3. If the current node's value is greater than `k`, we move to the left child.
4. If the current node's value is less than `k`, we move to the right child.
5. If the current node's value is equal to `k`, we return `0` as the absolute difference is `0`.

**Time and Space Complexity**
- **Time Complexity**:  
  - **O(h)**, where `h` is the height of the tree, because we perform a traversal that goes down one branch of the tree, which takes time proportional to the height of the tree.
  
- **Space Complexity**:  
  - **O(1)**, since we are only using a constant amount of space (ignoring recursion stack space for simplicity).

**Java Code**
```java
class Solution {
    // Function to find the least absolute difference between any node
    // value of the BST and the given integer.
    static int minDiff(Node root, int k) {
        // Initialize the closest value to the root's data
        int closest = root.data;

        // Perform a binary search in the BST
        while (root != null) {
            // Update the closest value if current node is closer to k
            if (Math.abs(root.data - k) < Math.abs(closest - k)) {
                closest = root.data;
            }

            // Move to the left or right child based on the comparison with k
            if (root.data > k) {
                root = root.left;
            } else if (root.data < k) {
                root = root.right;
            } else {
                // If current node's value is equal to k, return 0 as the closest
                return 0;
            }
        }

        // Return the absolute difference between closest value and k
        return Math.abs(closest - k);
    }
}


## Problem Statement: 337
**Count BST nodes that lie in a given range**

Given a **Binary Search Tree (BST)** and a range `l` to `h` (both inclusive), return the number of nodes in the BST whose values lie within the given range.

**Example**
**Input:**
10
/
5 15 / \
3 7 18

**l = 7**, **h = 15**

**Output:**
3

**Explanation:**
- The nodes in the range [7, 15] are: 10, 7, and 15.
- Hence, the output is `3`.

**Constraints**
- `1 ≤ Number of nodes ≤ 10^4`
- `-10^4 ≤ Node value ≤ 10^4`
- `-10^4 ≤ l ≤ h ≤ 10^4`

**Approach**
We can leverage the properties of a **Binary Search Tree (BST)** to efficiently solve this problem:
1. **If the node value is within the range [l, h]**:
   - Count the node and recursively check both left and right subtrees.
2. **If the node value is less than `l`**:
   - Skip the left subtree and only check the right subtree because all nodes in the left subtree will have values less than `l`.
3. **If the node value is greater than `h`**:
   - Skip the right subtree and only check the left subtree because all nodes in the right subtree will have values greater than `h`.

By using this approach, we can traverse the tree in a way that skips unnecessary subtrees, improving the efficiency of the solution.

**Time and Space Complexity**
- **Time Complexity**:  
  - **O(h)**, where `h` is the height of the tree. In the worst case (unbalanced tree), we may visit all nodes, but in a balanced tree, we may only visit a portion of the nodes.
  
- **Space Complexity**:  
  - **O(h)** for the recursive call stack, where `h` is the height of the tree.

**Java Code**
```java
class Solution {
    // Function to count the number of BST nodes that lie within the range [l, h].
    int getCount(Node root, int l, int h) {
        // Base case: If the root is null, return 0
        if (root == null) return 0;

        // If the node's value lies within the range [l, h]
        if (root.data <= h && root.data >= l)
            return 1 + getCount(root.left, l, h)  // Count the left subtree
                   + getCount(root.right, l, h); // Count the right subtree
        else if (root.data < l)  // If the node's value is less than 'l', ignore the left subtree
            return getCount(root.right, l, h); // Only check the right subtree
        else  // If the node's value is greater than 'h', ignore the right subtree
            return getCount(root.left, l, h); // Only check the left subtree
    }
}


## Problem Statement: 338
**Largest BST**

Given a binary tree, your task is to find the size of the largest subtree within this binary tree that also satisfies the properties of a **Binary Search Tree (BST)**. The size of a subtree is defined as the number of nodes it contains.

**Constraints**
- A subtree is considered a **BST** if:
  1. For every node, the left child is less than the node.
  2. The right child is greater than the node.
  3. There are no duplicate values in the subtree.

**Example**
**Input:**
    10
   /  \
  5    15
 / \     \
1   8     7


**Output:**
3

**Explanation:**
- The largest BST subtree is the one rooted at node `5` with nodes `5, 1, 8`, which is a valid BST.
- The size of this subtree is `3`.

**Approach**
To solve this problem, we need to:
1. **Recursively traverse** the tree to check whether each subtree satisfies the properties of a Binary Search Tree (BST).
2. For each node, compute:
   - The size of the subtree.
   - The minimum and maximum values in the subtree.
   - Whether the subtree rooted at that node is a BST.
3. Maintain the maximum size of the valid BST subtree found during the traversal.

**Key Considerations**
- A **Binary Search Tree (BST)** must have:
  - All nodes in the left subtree are less than the root node.
  - All nodes in the right subtree are greater than the root node.
- We need to handle cases where the tree is unbalanced or contains nodes that break the BST properties.

**Time and Space Complexity**
- **Time Complexity**:  
  - **O(n)** where `n` is the number of nodes in the tree, because each node is visited exactly once during the traversal.
  
- **Space Complexity**:  
  - **O(h)** where `h` is the height of the tree, due to the recursive call stack. In the worst case (unbalanced tree), `h = n`.

**Java Code**
```java
class Solution {
    // Helper class to store information about the tree
    static class Info {
        int size;        // Size of the subtree
        int min;         // Minimum value in the subtree
        int max;         // Maximum value in the subtree
        int maxSize;     // Size of the largest BST subtree found
        boolean isBST;   // True if the subtree is a BST

        // Constructor to initialize the values
        Info(int size, int min, int max, int maxSize, boolean isBST) {
            this.size = size;
            this.min = min;
            this.max = max;
            this.maxSize = maxSize;
            this.isBST = isBST;
        }
    }

    // Helper function to traverse the tree and return the information for each subtree
    static Info helper(Node node) {
        // Base case: If the node is null, return a valid BST with size 0
        if (node == null) {
            return new Info(0, Integer.MAX_VALUE, Integer.MIN_VALUE, 0, true);
        }

        // Recursively check the left and right subtrees
        Info left = helper(node.left);
        Info right = helper(node.right);

        // Create a current Info object for the current node
        Info curr = new Info(0, 0, 0, 0, false);
        curr.size = 1 + left.size + right.size;  // Count the current node and its subtrees

        // Check if the current subtree is a valid BST
        if (left.isBST && right.isBST && node.data > left.max && node.data < right.min) {
            curr.min = Math.min(node.data, left.min);  // Update the minimum value
            curr.max = Math.max(node.data, right.max); // Update the maximum value
            curr.maxSize = curr.size;  // Update the max size of the BST
            curr.isBST = true;         // Mark the subtree as a BST
        } else {
            // If it's not a BST, take the max size of the left or right subtree
            curr.maxSize = Math.max(left.maxSize, right.maxSize);
            curr.isBST = false;  // Mark the subtree as not a BST
        }

        // Return the Info object for the current node
        return curr;
    }

    // Function to return the size of the largest BST subtree
    static int largestBst(Node root) {
        // Call the helper function and return the max size of the BST
        return helper(root).maxSize;
    }
}


## Problem Statement: 339
**Lowest Common Ancestor in a BST**

Given a Binary Search Tree (with all values unique) and two nodes `n1` and `n2` (`n1 != n2`), you may assume that both nodes exist in the tree. Find the **Lowest Common Ancestor (LCA)** of the given two nodes in the BST.

**Definition:**
- **Lowest Common Ancestor (LCA)** of two nodes `n1` and `n2` is the lowest node in the BST that has both `n1` and `n2` as descendants. We allow a node to be a descendant of itself.

**Example**
**Input:**
    20
   /  \
 10    30
/  \   /  \
5  15  25  35


**Output:**
20

**Explanation:**
- The Lowest Common Ancestor (LCA) of nodes `5` and `15` is `10`.
- The LCA of nodes `25` and `35` is `30`.
- The LCA of nodes `5` and `35` is `20`.

**Approach**
To find the LCA of two nodes in a Binary Search Tree (BST), we can use the following property:
- If both `n1` and `n2` are smaller than the current node's value, then the LCA must lie in the left subtree.
- If both `n1` and `n2` are larger than the current node's value, then the LCA must lie in the right subtree.
- If one node is smaller and the other is larger, then the current node is the LCA.

We start from the root of the tree and traverse down. The first node where the paths of `n1` and `n2` diverge is the LCA.

**Time and Space Complexity**
- **Time Complexity:**  
  - **O(h)** where `h` is the height of the tree. In the worst case, we may need to traverse from the root to the leaf node, which takes `O(h)` time.
  
- **Space Complexity:**  
  - **O(1)** since we are only using a constant amount of space for the traversal.

**Java Code**
```java
class Solution {
    // Function to find the Lowest Common Ancestor (LCA)
    Node LCA(Node root, Node n1, Node n2) {
        // Traverse the tree to find the LCA
        while (root != null) {
            // If both n1 and n2 are smaller, LCA lies in the left subtree
            if (root.data > n1.data && root.data > n2.data)
                root = root.left;
            // If both n1 and n2 are larger, LCA lies in the right subtree
            else if (root.data < n1.data && root.data < n2.data)
                root = root.right;
            // If one of n1 or n2 matches root, root is the LCA
            else
                break;
        }

        // Return the LCA node
        return root;
    }
}


## Problem Statement: 340
**Merge Two BSTs**

Given two Binary Search Trees (BSTs), your task is to return the elements of both BSTs in sorted order after merging them.

**Example**
**Input:**

BST 1: BST 2: 4 5 / \ /
2 6 3 7 /
1

**Output:**
[1, 2, 3, 4, 5, 6, 7]

**Explanation:**
- The elements of both BSTs are merged into a sorted array in ascending order.

**Approach**
1. **Inorder Traversal**:  
   - Since a Binary Search Tree (BST) is sorted by definition, we can perform an inorder traversal on both trees to retrieve the nodes in sorted order.
   - Perform inorder traversal on both trees to obtain two sorted arrays of node values.

2. **Merging Sorted Arrays**:  
   - Use the merge technique (commonly used for merging two sorted arrays) to merge the two sorted arrays into a single sorted array.

3. **Return the Merged Array**:  
   - The resulting merged array will contain all the elements from both BSTs in sorted order.

**Time and Space Complexity**
- **Time Complexity:**  
  - **O(N + M)** where `N` is the number of nodes in the first tree and `M` is the number of nodes in the second tree. This is because we perform two inorder traversals (one for each tree) and then merge two sorted lists.

- **Space Complexity:**  
  - **O(N + M)** for the storage of the two sorted arrays and the merged array.

**Java Code**
```java
class Solution {
    // Function to return a list of integers denoting the node
    // values of both the BST in a sorted order.
    public ArrayList<Integer> merge(Node root1, Node root2) {
        // Initialize two lists to hold the inorder traversal of both trees
        ArrayList<Integer> list1 = new ArrayList<>();
        ArrayList<Integer> list2 = new ArrayList<>();

        // Perform inorder traversal on both BSTs
        inorder(root1, list1);
        inorder(root2, list2);

        // Merge both sorted lists and return the result
        return mergeSortedLists(list1, list2);
    }

    // Helper function to perform inorder traversal and store the result in the list
    private void inorder(Node root, ArrayList<Integer> list) {
        if (root == null) return;
        inorder(root.left, list);  // Traverse left subtree
        list.add(root.data);       // Add the node data to the list
        inorder(root.right, list); // Traverse right subtree
    }

    // Helper function to merge two sorted lists
    private ArrayList<Integer> mergeSortedLists(ArrayList<Integer> list1, ArrayList<Integer> list2) {
        ArrayList<Integer> merged = new ArrayList<>();
        int i = 0, j = 0;

        // Merge both lists until one of them is exhausted
        while (i < list1.size() && j < list2.size()) {
            if (list1.get(i) < list2.get(j)) {
                merged.add(list1.get(i++));
            } else {
                merged.add(list2.get(j++));
            }
        }

        // Add remaining elements from list1
        while (i < list1.size()) {
            merged.add(list1.get(i++));
        }

        // Add remaining elements from list2
        while (j < list2.size()) {
            merged.add(list2.get(j++));
        }

        return merged;
    }
}


## Problem Statement: 341
**Predecessor and Successor**

Given a Binary Search Tree (BST) and an integer key, you need to find the **in-order predecessor** and **in-order successor** of the given key. If either predecessor or successor is not found, return `NULL` for that.

- **Predecessor** of a node is the largest value node that is smaller than the given key.
- **Successor** of a node is the smallest value node that is greater than the given key.

**Example**
**Input:**
   50
  /  \
30    70
/ \ /
20 40 60 80

Key = 65

**Output:**
Predecessor = 60 Successor = 70

**Approach**
1. **Inorder Traversal:**  
   - In a Binary Search Tree (BST), the inorder traversal produces a sorted sequence of nodes.
   - The predecessor is the largest node that is smaller than the key, and the successor is the smallest node that is greater than the key.

2. **Finding Predecessor and Successor:**
   - Start from the root of the BST and traverse down:
     - If the current node's value is less than the given key, move to the right and update the predecessor.
     - If the current node's value is greater than the given key, move to the left and update the successor.
     - If the current node's value equals the given key, find the leftmost node in its right subtree (successor) and the rightmost node in its left subtree (predecessor).

3. **Edge Cases:**
   - If the key is the smallest in the BST, the predecessor is `NULL`.
   - If the key is the largest in the BST, the successor is `NULL`.

**Time and Space Complexity**
- **Time Complexity:**
  - **O(H)**, where `H` is the height of the BST, because we are traversing the tree from the root to the leaf nodes (in the worst case).
  
- **Space Complexity:**
  - **O(1)**, since we are only using a constant amount of space for the predecessor and successor nodes.

**Java Code**
```java
class Solution {
    public static void findPreSuc(Node root, Node[] pre, Node[] suc, int key) {
        // Start from the root of the BST
        Node curr = root;

        // Traverse the tree and find the predecessor and successor
        while (curr != null) {
            if (curr.data < key) {
                // Update predecessor and move to the right subtree
                pre[0] = curr;
                curr = curr.right;
            }
            else if (curr.data > key) {
                // Update successor and move to the left subtree
                suc[0] = curr;
                curr = curr.left;
            }
            else {
                // If the current node matches the key
                // Find the rightmost node in the left subtree (predecessor)
                if (curr.left != null)
                    pre[0] = rightMost(curr.left);
                
                // Find the leftmost node in the right subtree (successor)
                if (curr.right != null)
                    suc[0] = leftMost(curr.right);
                break;
            }
        }
    }

    // Helper function to find the rightmost node in a given subtree
    static Node rightMost(Node node) {
        while (node.right != null) {
            node = node.right;
        }
        return node;
    }

    // Helper function to find the leftmost node in a given subtree
    static Node leftMost(Node node) {
        while (node.left != null) {
            node = node.left;
        }
        return node;
    }
}


## Problem Statement: 342
**Populate Inorder Successor for all nodes**

Given a Binary Tree, your task is to populate the **next pointer** for all nodes. The next pointer for every node should point to the **inorder successor** of that node. If there is no inorder successor, the next pointer should point to `NULL` (or `-1` as specified in the prompt, but you don't need to explicitly assign `-1`).

**Example**
**Input:**
    10
   /  \
  8    12
 / \     \
4   9     15

**Output:**
Inorder traversal of the tree after populating the next pointers: 4 -> 8 -> 9 -> 10 -> 12 -> 15

**Explanation:**
- In this example, after populating the next pointers, each node's `next` pointer will point to its inorder successor:
  - `4`'s inorder successor is `8`
  - `8`'s inorder successor is `9`
  - `9`'s inorder successor is `10`
  - `10`'s inorder successor is `12`
  - `12`'s inorder successor is `15`
  - `15` has no inorder successor, so its `next` pointer is `NULL`.

**Approach**
1. **Inorder Traversal**:  
   To populate the inorder successor for each node, we perform a reverse inorder traversal (right, root, left).
   
   - During the traversal, each node's `next` pointer will be set to the previously visited node.
   - If there is no previously visited node, the next pointer will remain `NULL`.

2. **Steps**:
   - Start from the root of the tree.
   - Traverse the right subtree recursively.
   - Set the `next` pointer for the current node to the previously visited node.
   - Update the `next` pointer to the current node.
   - Traverse the left subtree recursively.
   
3. **Time Complexity**:  
   The time complexity is **O(n)**, where `n` is the number of nodes in the tree, because we visit each node exactly once during the traversal.

4. **Space Complexity**:  
   The space complexity is **O(h)**, where `h` is the height of the tree, due to the recursion stack used during the traversal. In the worst case, the space complexity is O(n) for a skewed tree.

**Java Code**
```java
class Solution {
    static Node next;

    public void populateNext(Node root) {
        next = null;  // Initialize next to null
        populateNextUtil(root);  // Helper function to traverse the tree
    }

    private void populateNextUtil(Node node) {
        // Base case: If node is null, return
        if (node == null) return;

        // Traverse the right subtree first (reverse inorder)
        populateNextUtil(node.right);

        // Set the next pointer of the current node
        node.next = next;
        
        // Update the next pointer to the current node
        next = node;

        // Traverse the left subtree
        populateNextUtil(node.left);
    }
}


## Problem Statement: 343
**BST to Greater Sum Tree**

Given a Binary Search Tree (BST) with unique node values, transform it into a greater sum tree where each node contains the sum of all nodes greater than that node in the BST. The transformation should be done in-place without creating new nodes.
The value of each node is updated to the sum of all values greater than the node's value in the BST.

**Example**
**Input:**
    4
   / \
  2   6
 / \  / \
1  3 5  7

**Output:**
    22
   /  \
 27    13
/  \   /  \
28 26 18   7

**Explanation:**
- After the transformation:
  - Node `4` becomes `22` because the sum of nodes greater than `4` is `6 + 5 + 7 + 3 + 1 = 22`.
  - Node `2` becomes `27` because the sum of nodes greater than `2` is `6 + 5 + 7 + 3 = 21`, and the node itself gets updated to this value.
  - Similarly, all nodes are updated to the sum of nodes greater than themselves in the tree.

**Approach**
1. **Reverse Inorder Traversal**:
   - A reverse inorder traversal (right, root, left) of the BST ensures that we process nodes in descending order of their values.
   - While performing this traversal, we maintain a running sum of the values encountered so far.
   
2. **Update Node Values**:
   - For each node, we:
     - Add the node’s value to the running sum.
     - Update the node’s value to the current running sum minus its own value (i.e., all values greater than the node's current value).

3. **Time Complexity**:
   - **O(n)**, where `n` is the number of nodes in the BST. We visit each node exactly once in the reverse inorder traversal.

4. **Space Complexity**:
   - **O(h)**, where `h` is the height of the tree, due to the recursion stack used in the traversal. In the worst case, the space complexity is O(n) for a skewed tree.

**Java Code**
```java
class Solution {
    public static void transformTree(Node root) {
        // Initialize the sum array to store the running sum
        int[] sum = {0}; 
        // Start the transformation process
        transformToGreaterSumTree(root, sum);
    }

    static void transformToGreaterSumTree(Node root, int[] sum) {
        // Base case: If the node is null, return
        if (root == null) {
            return;
        }
        
        // Recurse on the right subtree (greater values first)
        transformToGreaterSumTree(root.right, sum);
        
        // Update the current node's value to the sum of nodes greater than it
        sum[0] += root.data;
        root.data = sum[0] - root.data;
        
        // Recurse on the left subtree
        transformToGreaterSumTree(root.left, sum);
    }
}


## Problem Statement: 344
**Construct Tree from Inorder & Preorder**

Given two arrays representing the inorder and preorder traversals of a binary tree, your task is to construct the binary tree and return the root node of the constructed tree.

**Notes:**
- The problem asks for the construction of the binary tree from its inorder and preorder traversals.
- The output should be written in **postorder traversal** format.

**Approach**
To solve this problem, we will utilize the following approach:
1. **Preorder Traversal**: In preorder, the first element is always the root node.
2. **Inorder Traversal**: In inorder, elements to the left of the root node are part of the left subtree, and elements to the right are part of the right subtree.
3. Using these two properties, we can recursively construct the binary tree:
   - Find the root from the preorder array (first element).
   - Find the index of the root in the inorder array.
   - Recursively construct the left and right subtrees by dividing the inorder array at the root's index.

**Steps:**
1. The first element of preorder is the root.
2. Find the index of this root in the inorder array. The elements to the left of this index in inorder are part of the left subtree, and the elements to the right are part of the right subtree.
3. Recur for the left and right subtrees using updated preorder and inorder segments.
4. Return the root node after constructing the left and right subtrees.

**Time and Space Complexity**
**Time Complexity:**
Building the map for the inorder traversal takes O(n), where n is the number of nodes.
For each node, we make a recursive call and do constant work (finding the index in the map), resulting in an overall time complexity of O(n).

**Space Complexity:**
The space complexity is dominated by the recursion stack and the map, both of which are O(n) in the worst case.

**Java Code**
```java
class Solution {
    static Node buildTreeRecur(Map<Integer, Integer> mp, int[] preorder, 
                               int[] preIndex, int left, int right) {
        if (left > right)
            return null;

        // Get the root value from preorder
        int rootVal = preorder[preIndex[0]];
        preIndex[0]++;
        
        // Create the root node
        Node root = new Node(rootVal);
        
        // Get the index of the root node in inorder
        int index = mp.get(rootVal);
        
        // Recursively construct the left and right subtrees
        root.left = buildTreeRecur(mp, preorder, preIndex, left, index - 1);
        root.right = buildTreeRecur(mp, preorder, preIndex, index + 1, right);

        return root;
    }

    public static Node buildTree(int inorder[], int preorder[]) {
        // Create a map to store inorder elements' indices
        Map<Integer, Integer> mp = new HashMap<>();
        for (int i = 0; i < inorder.length; i++)
            mp.put(inorder[i], i);

        // Initialize preIndex to 0 as it's the index in preorder array
        int[] preIndex = {0};
        
        // Call the recursive function to build the tree
        return buildTreeRecur(mp, preorder, preIndex, 0, inorder.length - 1);
    }
}


## Problem Statement: 345
**Construct Tree from Preorder Traversal**

Given two arrays `pre[]` and `preLN[]`, construct a binary tree of size `N`. The array `pre[]` represents the preorder traversal of a binary tree, and the array `preLN[]` contains either 'L' or 'N' for each corresponding node in the tree:

- 'L' means the corresponding node is a **leaf node** (i.e., it has no children).
- 'N' means the corresponding node is a **non-leaf node** (i.e., it has two children).

**Constraints:**
- Every node in the binary tree has either 0 or 2 children.
- The length of the arrays `pre[]` and `preLN[]` is `N`.

**Objective:**
Construct a binary tree from the given `pre[]` and `preLN[]` arrays.

**Approach:**
1. **Preorder Traversal Representation:**
   - The preorder traversal of a binary tree visits the nodes in the order: root, left subtree, right subtree.
   - The `pre[]` array gives the preorder traversal, while the `preLN[]` array indicates whether each node is a leaf ('L') or a non-leaf ('N').

2. **Recursive Construction:**
   - The construction of the binary tree is done recursively. 
   - For each node in `pre[]`, check whether it is a leaf or non-leaf node based on the corresponding value in `preLN[]`.
   - If it's a leaf ('L'), no children need to be assigned, so return that node.
   - If it's a non-leaf ('N'), recursively construct the left and right subtrees.

3. **Traversal Flow:**
   - Start from the first element of `pre[]`, create the node.
   - If the node is a leaf ('L'), return it.
   - If the node is not a leaf ('N'), recursively construct its left and right children.
   - Continue this process until the tree is fully constructed.

4. **Recursive Helper Function:**
   - The helper function takes the current index and recursively constructs the left and right children for the non-leaf nodes.

**Time Complexity:**
The time complexity is O(N), where N is the number of nodes in the binary tree. This is because we are iterating over each node exactly once during the recursive calls.

**Space Complexity:**
The space complexity is O(N) due to the recursion stack in the worst case, where N is the number of nodes in the tree.

**Solution Code:**
```java
class Solution {
    int index = 0;

    // Function to construct the binary tree
    Node constructTree(int n, int pre[], char preLN[]) {
        return constructTreeRecur(pre, preLN, n);
    }

    // Recursive function to construct the tree
    private Node constructTreeRecur(int[] pre, char[] preLN, int n) {
        // If the index exceeds the size of the array, return null
        if (index >= n) return null;

        // Create a new node with the value from the pre[] array
        Node node = new Node(pre[index]);

        // Move to the next index in the preorder array
        if (preLN[index++] == 'L') {
            // If the node is a leaf, return the node
            return node;
        }

        // If the node is not a leaf, recursively construct its left and right children
        node.left = constructTreeRecur(pre, preLN, n);
        node.right = constructTreeRecur(pre, preLN, n);

        return node;
    }
}


## Problem Statement: 346
**Construct Binary Tree from Parent Array**

Given an array `parent[]`, construct a binary tree. The array represents the tree where the indices are the values of the tree nodes, and `parent[i]` denotes the parent node of node `i`. The parent of the root node is denoted as `-1` since the root has no parent.

**Note:**
- If two elements have the same parent, the one that appears first in the array will be the **left child**, and the second will be the **right child**.
- You don't need to print anything; the driver code will print the level order traversal of the returned root node to verify the output.

**Objective:**
Construct the standard linked representation of a binary tree from the given parent array and return the root node of the tree.

**Approach:**
1. **Understanding the Parent Array:**
   - Each index of the `parent[]` array represents a node, and the value at each index denotes the parent of that node.
   - The root node is the one where the value in `parent[]` is `-1`.
   - Each node can have at most two children: the first child will be assigned to the left child, and the second child will be assigned to the right child.

2. **Node Creation:**
   - Iterate over the `parent[]` array and create nodes for each index.
   - If the parent is `-1`, mark that node as the root.
   - For each node, create its parent node if not already created, and attach the node as the left or right child based on the order in the array.

3. **Recursive Node Creation:**
   - Use recursion to create nodes and their parents. When creating a node, if the parent is not yet created, the function will recursively create the parent first.

4. **Tree Construction:**
   - Use a helper array `created[]` to store the created nodes, as we may need to access parent nodes multiple times.
   - Attach each node to its correct position in the binary tree as left or right child based on the order of appearance in the `parent[]` array.

**Time Complexity:**
O(N), where N is the number of nodes in the tree (i.e., the length of the parent[] array). This is because each node is processed once, and each node creation takes constant time.

**Space Complexity:**
O(N) due to the storage used by the created[] array and the recursive call stack for node creation.

**Solution Code:**
```java
class Solution {
    // Function to construct binary tree from parent array.
    public Node createTree(int parent[]) {
        int n = parent.length;
        Node[] created = new Node[n];  // Array to store created nodes
        Node root = null;

        // Iterate over the parent array to create nodes
        for (int i = 0; i < n; i++) {
            createNode(parent, i, created);

            // If parent[i] is -1, this node is the root
            if (parent[i] == -1) {
                root = created[i];
            }
        }

        return root;
    }

    // Helper function to create a node and attach it to its parent
    private void createNode(int[] parent, int i, Node[] created) {
        if (created[i] != null) return;  // If the node is already created, return

        // Create current node
        created[i] = new Node(i);

        if (parent[i] == -1) return;  // If it's the root, no need to attach to parent

        // Create the parent node if not already created
        if (created[parent[i]] == null) {
            createNode(parent, parent[i], created);
        }

        // Attach the current node to its parent
        Node p = created[parent[i]];
        if (p.left == null) {
            p.left = created[i];  // Attach as left child
        } else {
            p.right = created[i];  // Attach as right child
        }
    }
}


## Problem Statement: 347
**Serialize and Deserialize a Binary Tree**

Serialization is the process of converting a tree into a format (such as an array) so that it can be later restored. Deserialization is the reverse process, where a tree is reconstructed from its serialized format.

In this problem, you are required to implement the following two functions:
1. **`serialize()`**: This function stores the tree in an array and returns the array.
2. **`deSerialize()`**: This function deserializes the array back into the tree and returns the root of the tree.

**Note:**
- Multiple nodes can have the same data, and node values are always positive integers.
- Your code will be considered correct if `deSerialize(serialize(input_tree))` returns the same tree as the input tree.
- The driver code will print the in-order traversal of the tree returned by `deSerialize(serialize(input_tree))` to verify the output.

**Approach:**
1. **Serialization:**
   - Perform a **level-order traversal** (BFS) of the binary tree.
   - Use a queue to traverse the tree and store node values in an array.
   - If a node is null, represent it as `-1` in the array (to mark the absence of a node).
   
2. **Deserialization:**
   - Start by reading the first value from the array (this will be the root of the tree).
   - Use a queue to reconstruct the tree. For each node, read the next values to determine its left and right children.
   - If a value is `-1`, the node is null, and no child is created.
   - Continue reading values and constructing the tree until all values are processed.

**Time Complexity:**
**serialize(): O(N)**, where N is the number of nodes in the tree. Each node is processed once during the level-order traversal.
**deSerialize(): O(N),** where N is the number of nodes in the tree. The array is processed once, and for each element, a node is either created or skipped.

**Space Complexity:**
**serialize(): O(N)** for the array used to store the serialized tree and the queue used for level-order traversal.
**deSerialize(): O(N)** for the array used to store the serialized tree and the queue used for tree reconstruction.

**Solution Code:**
```java
class Tree {
    // Function to serialize a tree and return a list containing nodes of tree.
    public ArrayList<Integer> serialize(Node root) {
        ArrayList<Integer> arr = new ArrayList<>(); // ArrayList to store serialized tree nodes

        // Initialize queue for level-order traversal (BFS)
        Queue<Node> q = new LinkedList<>();
        q.add(root);

        while (!q.isEmpty()) {
            Node curr = q.poll();
            if (curr == null) {
                arr.add(-1);  // -1 represents a null node
                continue;
            }
            arr.add(curr.data);  // Add current node's data to the array
            q.add(curr.left);    // Add left child to the queue
            q.add(curr.right);   // Add right child to the queue
        }

        return arr;  // Return the serialized array
    }

    // Function to deserialize a list and construct the tree.
    public Node deSerialize(ArrayList<Integer> arr) {
        if (arr.get(0) == -1) return null;  // If the first element is -1, the tree is empty

        // Create the root node using the first element of the array
        Node root = new Node(arr.get(0));
        Queue<Node> q = new LinkedList<>();
        q.add(root);

        int i = 1;  // Start from the second element in the array
        while (!q.isEmpty() && i < arr.size()) {
            Node curr = q.poll();

            // If the next element is not -1, create a left child and add it to the queue
            if (arr.get(i) != -1) {
                Node left = new Node(arr.get(i));
                curr.left = left;
                q.add(left);
            }
            i++;

            // If the next element is not -1, create a right child and add it to the queue
            if (i < arr.size() && arr.get(i) != -1) {
                Node right = new Node(arr.get(i));
                curr.right = right;
                q.add(right);
            }
            i++;
        }

        return root;  // Return the root node of the deserialized tree
    }
}


## Problem Statement: 348
**Largest Subtree Sum in a Tree**

Given a binary tree, the task is to find the subtree with the maximum sum and return its sum.

**Example:**
**Input:**
     1
    / \
   2   3
  / \   \
 4   5   6


**Output:**
28

**Explanation:**
- The largest subtree sum is for the subtree rooted at node `1`. The sum of this subtree is `1 + 2 + 4 + 5 + 3 + 6 = 28`.

**Approach:**
To solve this problem, we need to find the sum of all subtrees and keep track of the largest subtree sum encountered. A **subtree sum** for a node is defined as the sum of its value, the sum of its left subtree, and the sum of its right subtree.

**Steps:**
1. **Level-Order Traversal (BFS):**
   - Traverse the tree level by level to process each node.
   - Use a queue to perform the level-order traversal.

2. **Calculate Subtree Sum:**
   - For each node, calculate its subtree sum by adding its value to the sums of its left and right subtrees.
   - Store these subtree sums in a hash map where the key is the node and the value is the subtree sum.

3. **Track the Maximum Subtree Sum:**
   - As we calculate the subtree sum for each node, update the maximum sum found so far.

4. **Post-Order Traversal for Subtree Sum Calculation:**
   - Process the tree in a bottom-up manner (post-order traversal). This ensures that we compute the sum of the subtrees before the sum of their parent nodes.


**Time Complexity:**
O(N), where N is the number of nodes in the tree. Each node is processed once during the level-order traversal, and the subtree sum for each node is computed once.

**Space Complexity:**
O(N), where N is the number of nodes. The space complexity is determined by the space used to store the nodes in the queue, the list of levels, and the hash map for subtree sums.

**Solution Code:**
```java
class Solution {
    public static int findLargestSubtreeSum(Node root) {
        // Handle the edge case of an empty tree
        if (root == null)
            return 0;

        int ans = Integer.MIN_VALUE;  // Variable to store the largest sum
        Queue<Node> q = new LinkedList<>();
        List<List<Node>> levels = new ArrayList<>();
        HashMap<Node, Integer> subtreeSum = new HashMap<>();

        q.add(root);

        // Perform a level-order traversal to collect nodes by levels
        while (!q.isEmpty()) {
            int n = q.size();
            List<Node> level = new ArrayList<>();
            while (n-- > 0) {
                Node node = q.poll();
                level.add(node);
                if (node.left != null)
                    q.add(node.left);  // Add left child to the queue
                if (node.right != null)
                    q.add(node.right); // Add right child to the queue
            }
            levels.add(level);  // Add the current level to the levels list
        }

        // Traverse the tree from bottom-up (reverse level order) to calculate subtree sums
        for (int i = levels.size() - 1; i >= 0; i--) {
            for (Node e : levels.get(i)) {
                subtreeSum.put(e, e.data);  // Initialize the subtree sum with the node's data
                if (e.left != null)
                    subtreeSum.put(e, subtreeSum.get(e) + subtreeSum.get(e.left));  // Add left subtree sum
                if (e.right != null)
                    subtreeSum.put(e, subtreeSum.get(e) + subtreeSum.get(e.right)); // Add right subtree sum
                ans = Math.max(ans, subtreeSum.get(e));  // Update the maximum subtree sum
            }
        }

        return ans;  // Return the largest subtree sum
    }
}


## Problem Statement: 349
**Maximum Sum of Non-Adjacent Nodes**

Given a binary tree where each node has a value associated with it, your task is to select a subset of nodes such that the sum of their values is maximized. The condition is that no two selected nodes are directly connected, meaning that if a node is included in the subset, neither its parent nor its children can be included.

**Example:**
**Input:**
     5
    / \
   1   8
  / \   / \
 7   2  3  4


**Output:**
18

**Explanation:**
- The maximum sum of non-adjacent nodes is `5 + 8 + 7 = 18`.

**Approach:**
We need to select non-adjacent nodes from the tree such that their sum is maximized. The key observation here is to use **dynamic programming with memoization** to efficiently compute the maximum sum for each node while keeping track of the maximum sum for its children nodes.

**Steps:**
1. **Recursive Approach with Memoization:**
   - For each node, we have two options:
     1. **Include the node**: If we include the node, we cannot include its immediate children (left and right). Thus, the maximum sum for this case is the value of the node itself plus the sum of non-adjacent nodes from the left and right children (i.e., grand-children).
     2. **Exclude the node**: If we exclude the node, we can include its left and right children, and the sum will be the sum of the maximum sums of the left and right subtrees.

2. **Memoization**:
   - Use a memoization map (`memo`) to store the results of already computed nodes to avoid redundant calculations and optimize the solution.

3. **Base Case**:
   - If the node is `null`, the sum is `0`.

4. **Recursive Formula**:
   - For a node `node`:
     - **Include**: `node.data + getMaxSumUtil(node.left.left) + getMaxSumUtil(node.left.right) + getMaxSumUtil(node.right.left) + getMaxSumUtil(node.right.right)`
     - **Exclude**: `getMaxSumUtil(node.left) + getMaxSumUtil(node.right)`

   Finally, return the maximum of the two options (including or excluding the node).

**Time Complexity:**
O(N), where N is the number of nodes in the tree. Each node is visited once and the results are cached in the memoization map, ensuring that no node is processed more than once.

**Space Complexity:**
O(N), where N is the number of nodes in the tree. The space complexity is due to the memoization map and the recursion stack.

**Solution Code:**
```java
class Solution {
    // Helper function for the recursive calculation of the maximum sum.
    static int getMaxSumUtil(Node node, HashMap<Node, Integer> memo) {
        if (node == null) {
            return 0;  // Base case: If the node is null, return 0.
        }
        if (memo.containsKey(node)) {
            return memo.get(node);  // Return the stored value if already computed.
        }

        // Calculate the maximum sum including the current node
        int incl = node.data;
        if (node.left != null) {
            incl += getMaxSumUtil(node.left.left, memo) + 
                    getMaxSumUtil(node.left.right, memo);  // Include grandchildren
        }
        if (node.right != null) {
            incl += getMaxSumUtil(node.right.left, memo) + 
                    getMaxSumUtil(node.right.right, memo);  // Include grandchildren
        }

        // Calculate the maximum sum excluding the current node
        int excl = getMaxSumUtil(node.left, memo) + 
                   getMaxSumUtil(node.right, memo);  // Include the children without the current node

        // Store the maximum of including or excluding the node
        memo.put(node, Math.max(incl, excl));

        return memo.get(node);
    }

    // Main function to get the maximum sum of non-adjacent nodes in the binary tree
    public int getMaxSum(Node root) {
        HashMap<Node, Integer> memo = new HashMap<>();
        return getMaxSumUtil(root, memo);  // Call the helper function to calculate the result
    }
}


## Problem Statement: 350
**Duplicate Subtrees**
Given a binary tree, your task is to find all duplicate subtrees in the given binary tree. A **duplicate subtree** is defined as two subtrees that have the same structure and the same node values.

- **Input**: A binary tree rooted at `root`.
- **Output**: A list of the roots of all duplicate subtrees.

**Note**: 
- The returned list should be sorted in lexicographically increasing order based on the pre-order traversal of the trees.

**Example:**
**Input:**
    1
   / \
  2   3
 / \   /
4   5 2
       /
      4

**Output:**
[4, 2]

**Explanation:**
- There are two duplicate subtrees:
  - The subtree rooted at node `4`.
  - The subtree rooted at node `2` which contains the subtree rooted at `4`.

**Approach:**

To identify duplicate subtrees, we can use **serialization** to represent subtrees as strings. Two subtrees are duplicates if their serialized forms are identical. We will recursively traverse the tree, serialize each subtree, and use a hash map to store the serialized strings and their occurrences. If a subtree has been seen before, it means we have found a duplicate.

**Steps:**
1. **Serialization**:
   - For each node, serialize its subtree in a way that captures both the structure and values of the nodes.
   - For example, the subtree rooted at a node with value `1`, with left child `2` and right child `3`, can be serialized as: `(2)(1)(3)`.

2. **Map to Track Occurrences**:
   - Use a map to track the count of each serialized subtree. If a subtree appears more than once, it's a duplicate.
   - The key in the map will be the serialized subtree, and the value will be the number of times it has been encountered.

3. **Recursive Traversal**:
   - Recursively traverse the tree, serializing each subtree.
   - If a subtree has been encountered twice (i.e., it is a duplicate), add it to the result list.

4. **Sorting**:
   - After identifying all duplicate subtrees, return the list of duplicate subtrees, sorted by their pre-order traversal.

**Time Complexity:**
O(N), where N is the number of nodes in the binary tree. We visit each node once and serialize its subtree, which takes linear time.

**Space Complexity:**
O(N) for the space used by the map, strToInt, and the ans list. The space is proportional to the number of nodes and the number of distinct subtrees.

**Solution Code:**
```java
class Solution {
    // Helper function to serialize the tree and detect duplicates
    static int serializeTree(Node root, Map<Integer, Integer> map, 
                             Map<String, Integer> strToInt, List<Node> ans, int[] idNum) {
        if (root == null) return 0;  // Return 0 for null nodes
        
        // Recursively serialize the left and right subtrees
        int left = serializeTree(root.left, map, strToInt, ans, idNum);
        int right = serializeTree(root.right, map, strToInt, ans, idNum);
        
        // Generate the string representation of the current subtree
        String val = Integer.toString(root.data);
        String curr = "(" + left + ")" + val + "(" + right + ")";
        
        // If the subtree is already serialized, get its ID, otherwise generate a new ID
        int currId; 
        if (strToInt.containsKey(curr)) {
            currId = strToInt.get(curr);
        } else {
            currId = idNum[0]++;
            strToInt.put(curr, currId);  // Map the serialized string to a unique ID
        }
        
        // Update the occurrence count of the current serialized subtree
        map.put(currId, map.getOrDefault(currId, 0) + 1);
        
        // If the subtree occurs twice, add it to the result list
        if (map.get(currId) == 2) {
            ans.add(root);
        }
        
        return currId;
    }

    // Main function to return the list of duplicate subtrees
    public List<Node> printAllDups(Node root) {
        Map<Integer, Integer> map = new HashMap<>();  // To store the count of each subtree ID
        Map<String, Integer> strToInt = new HashMap<>();  // To map serialized string to unique ID
        
        List<Node> ans = new ArrayList<>();  // To store the result of duplicate subtrees
        int[] idNum = {1};  // To assign unique IDs to subtrees
        
        // Call the helper function to serialize the tree and detect duplicates
        serializeTree(root, map, strToInt, ans, idNum);  
        
        return ans;  // Return the list of duplicate subtrees
    }
}


## Problem Statement: 351
**Flatten Binary Tree to Linked List**

Given the root of a binary tree, flatten the tree into a "linked list" in-place. The "linked list" should use the same `Node` class, where the `right` child pointer points to the next node in the list, and the `left` child pointer is always null.
The linked list should be in the same order as a **pre-order traversal** of the binary tree.

**Example:**
**Input:**
     1
    / \
   2   5
  / \    \
 3   4    6

**Output:**
1 -> 2 -> 3 -> 4 -> 5 -> 6

**Explanation:**
- The pre-order traversal of the tree is `1, 2, 3, 4, 5, 6`.
- The flattened tree becomes a linked list where each node's left child is `null` and the right child points to the next node in the pre-order traversal.

**Approach:**
1. **Pre-order Traversal**: The order of the "linked list" should match the pre-order traversal of the tree, which visits the root first, then the left subtree, and finally the right subtree.

2. **In-place Transformation**:
   - The key idea is to perform the flattening in-place by modifying the tree's pointers directly. The left child pointer should be set to `null`, and the right child pointer should point to the next node in the flattened list.
   - For each node, we recursively flatten the left subtree, then adjust the current node's pointers:
     - Make the left child `null`.
     - Attach the flattened left subtree to the right child of the current node.
     - Ensure that the last node in the flattened left subtree points to the right child.

3. **Recursive Flattening**:
   - Flatten the left subtree recursively.
   - Temporarily store the right child, then attach the left child to the right.
   - Traverse to the end of the newly attached left subtree, and link it to the previously stored right child.

**Time Complexity:**
O(N), where N is the number of nodes in the tree. We visit each node once during the recursion.

**Space Complexity:**
O(H), where H is the height of the tree. This is the space used by the recursion stack. In the worst case, the tree can be skewed, and the height can be O(N), making the space complexity O(N).

**Solution Code:**
```java
class Solution {
    public static void flatten(Node node) {
        // If the node is null, do nothing
        if (node == null)
            return;
        
        // If the node is a leaf (no left or right children), do nothing
        if (node.left == null && node.right == null)
            return;
        
        // Flatten the left subtree if it exists
        if (node.left != null) {
            flatten(node.left);  // Recursively flatten the left subtree
            
            // Temporarily store the right child
            Node tempNode = node.right;
            
            // Move the left subtree to the right of the current node
            node.right = node.left;
            node.left = null;  // Set the left child to null
            
            // Traverse to the rightmost node of the new right subtree
            Node curr = node.right;
            while (curr.right != null)
                curr = curr.right;
            
            // Attach the original right subtree to the rightmost node of the left subtree
            curr.right = tempNode;
        }
        
        // Flatten the right subtree
        flatten(node.right);
    }
}


## Problem Statement: 352
**Delete Node in a Binary Search Tree (BST)**

Given the root node reference of a Binary Search Tree (BST) and a key, delete the node with the given key in the BST. Return the root node reference (possibly updated) of the BST.

The deletion process can be broken into two main stages:

1. **Search for a node to remove.**
2. **If the node is found, delete the node.**

**Approach:**
The process of deleting a node in a Binary Search Tree involves three possible cases based on the number of children the node has:

1. **Node has no children**: In this case, the node is simply removed.
2. **Node has one child**: The node is removed and its child takes its place.
3. **Node has two children**: In this case, we need to find the in-order successor (or predecessor), replace the node's value with the successor's value, and then delete the successor.

**Algorithm:**
1. **Search for the Node**:
   - If the key to be deleted is smaller than the current node’s value, recurse into the left subtree.
   - If the key to be deleted is larger than the current node’s value, recurse into the right subtree.
   - If the key matches the current node’s value, then proceed with deletion.

2. **Deletion Cases**:
   - **Case 1 (No child)**: Simply return `null` as the node has no children.
   - **Case 2 (One child)**: Return the non-null child node to replace the current node.
   - **Case 3 (Two children)**: Find the in-order successor (the smallest node in the right subtree), copy its value to the current node, and recursively delete the successor node.

**Example:**
**Input:**
     5
    / \
   3   6
  / \
 2   4

**Delete Node with key 3:**
1. The node `3` is found. It has two children (`2` and `4`).
2. The in-order successor is the node `4` (smallest node in the right subtree).
3. The value `3` is replaced by `4`.
4. The node `4` is then deleted.

**Output:**
     5
    / \
   4   6
  / 
 2   

**Time Complexity:**
O(H), where H is the height of the tree. In the worst case, we may need to traverse the entire height of the tree to find and delete the node. This is typical for binary search trees.

**Space Complexity:**
O(H), where H is the height of the tree. This is due to the space used by the recursion stack. In the worst case, the tree could be skewed, resulting in a height of O(N).

**Solution Code:**
```java
class Solution {
    public TreeNode deleteNode(TreeNode root, int key) {
        // If the root is null, return null
        if (root == null) return null;

        // Step 1: Search for the node
        if (key < root.val) {
            root.left = deleteNode(root.left, key);
        } else if (key > root.val) {
            root.right = deleteNode(root.right, key);
        } else {
            // Step 2: Node found
            // Case 1: Node has no children
            if (root.left == null && root.right == null) {
                return null;
            }
            // Case 2: Node has one child
            else if (root.left == null) {
                return root.right;
            } else if (root.right == null) {
                return root.left;
            }
            // Case 3: Node has two children
            else {
                // Find the in-order successor (smallest node in the right subtree)
                TreeNode successor = findMin(root.right);
                // Copy the value of the successor to the current node
                root.val = successor.val;
                // Delete the successor node
                root.right = deleteNode(root.right, successor.val);
            }
        }

        // Return the root after deletion
        return root;
    }

    // Helper function to find the minimum node in a subtree
    private TreeNode findMin(TreeNode node) {
        while (node.left != null)
            node = node.left;
        return node;
    }
}


## Problem Statement: 353
**Insert into a Binary Search Tree (BST)**

You are given the root node of a binary search tree (BST) and a value to insert into the tree. Return the root node of the BST after the insertion. It is guaranteed that the new value does not exist in the original BST.

Notice that there may exist multiple valid ways for the insertion, as long as the tree remains a BST after insertion. You can return any of them.

**Approach:**
To insert a new value into a Binary Search Tree (BST), we need to follow the standard BST insertion rule:
- If the value is smaller than the current node's value, we move to the left subtree.
- If the value is greater than the current node's value, we move to the right subtree.

The insertion process stops when we find an empty spot (either left or right) where we can insert the new node.

**Algorithm:**
1. **If the current node is `null`**: The value should be inserted at this position.
2. **If the value is less than the current node's value**: Move to the left child.
3. **If the value is greater than the current node's value**: Move to the right child.
4. **Repeat the process**: Continue traversing the tree until an empty position is found for the new node.

**Time Complexity:**
O(H), where H is the height of the tree. In the worst case, the tree could be skewed (like a linked list), and we would have to traverse all the way down to the leaf node.
For a balanced tree, the height H would be approximately log(N), where N is the number of nodes in the tree.

**Space Complexity:**
O(1) for iterative approach (ignoring recursion stack). The space complexity here is constant since we only use a fixed amount of space for the current node.

**Solution Code:**
```java
class Solution {
    public TreeNode insertIntoBST(TreeNode root, int val) {
        // If the root is null, create a new node with the given value
        if (root == null) return new TreeNode(val);

        TreeNode current = root;

        // Traverse the tree to find the correct spot for insertion
        while (true) {
            if (val < current.val) {
                // Move to the left subtree
                if (current.left == null) {
                    current.left = new TreeNode(val);
                    break;
                } else {
                    current = current.left;
                }
            } else { // val > current.val (since val is guaranteed to be unique)
                // Move to the right subtree
                if (current.right == null) {
                    current.right = new TreeNode(val);
                    break;
                } else {
                    current = current.right;
                }
            }
        }

        return root; // Return the unchanged root
    }
}


## Problem Statement: 354
**Construct Binary Tree from Inorder and Postorder Traversal**

Given two integer arrays `inorder` and `postorder`, where `inorder` is the inorder traversal of a binary tree and `postorder` is the postorder traversal of the same tree, construct and return the binary tree.

**Approach:**
The key idea for constructing the binary tree from inorder and postorder traversals is based on the following properties of these traversals:

1. **Inorder Traversal**: The nodes are visited in the order `left, root, right`. This means the element at the middle of the inorder array is the root of the tree or subtree.
2. **Postorder Traversal**: The nodes are visited in the order `left, right, root`. Therefore, the last element in the postorder array is the root of the tree or subtree.

Using these properties, we can follow this approach:

1. Start with the last element in the `postorder` array as the root.
2. Find the root in the `inorder` array to divide the tree into left and right subtrees.
3. Recursively build the right subtree using the elements after the root in the `inorder` array, and the corresponding elements in the `postorder` array.
4. Recursively build the left subtree using the elements before the root in the `inorder` array, and the corresponding elements in the `postorder` array.

**Algorithm:**
1. **Initialize postIndex**: The postIndex points to the last element in the `postorder` array.
2. **Build a hashmap for inorder indices**: This allows efficient lookup of the root value in the inorder array.
3. **Recursively construct the tree**:
   - The last element of the postorder array is the root.
   - Use the root value to split the inorder array into left and right subtrees.
   - Recursively build the right subtree first (since postorder traversal visits the right subtree before the left subtree).
   - Then, build the left subtree.

**Time Complexity:**
O(N), where N is the number of nodes in the tree. Each node is processed exactly once, and the hashmap lookup takes constant time.
The recursion goes through all the nodes once and splits the arrays efficiently.

**Space Complexity:**
O(N) for the hashmap inorderMap and the recursion stack.
The recursion stack depth will be proportional to the height of the tree, which in the worst case is O(N) (for a skewed tree).

**Solution Code:**
```java
class Solution {
    int postIndex;
    Map<Integer, Integer> inorderMap;

    public TreeNode buildTree(int[] inorder, int[] postorder) {
        postIndex = postorder.length - 1;
        inorderMap = new HashMap<>();

        // Build a hashmap to store value -> index relations for inorder
        for (int i = 0; i < inorder.length; i++) {
            inorderMap.put(inorder[i], i);
        }

        return build(postorder, 0, inorder.length - 1);
    }

    private TreeNode build(int[] postorder, int inStart, int inEnd) {
        if (inStart > inEnd) return null;

        // Get current root value from postorder
        int rootVal = postorder[postIndex--];
        TreeNode root = new TreeNode(rootVal);

        // Build right subtree before left (postorder => left, right, root)
        int inorderIndex = inorderMap.get(rootVal);
        root.right = build(postorder, inorderIndex + 1, inEnd);
        root.left = build(postorder, inStart, inorderIndex - 1);

        return root;
    }
}
