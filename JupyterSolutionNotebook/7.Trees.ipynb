## Problem Statement: 307
**Preorder Traversal**
Given the root of a binary tree, return a list containing the **Preorder Traversal** of the tree.

In preorder traversal, nodes are visited in the following order:
1. Visit the root node.
2. Traverse the left subtree.
3. Traverse the right subtree.

**Example**
**Input Tree:**
  1
 / \
2   3
/
4 5

**Output:**
[1, 2, 4, 5, 3]


**Constraints**
- `1 ≤ Number of nodes ≤ 10⁴`
- Each node contains a unique integer value.

**Goal:**  
To traverse a binary tree in **preorder** (Root → Left → Right) and return a list of the node values in that specific order.

**Approach**
We use a **recursive method** to perform the preorder traversal.  
Steps:
1. Visit and store the value of the current node.
2. Recursively traverse the left subtree.
3. Recursively traverse the right subtree.

**Pseudocode/Algorithm:**  
```sql
function preorderTraversal(node):
if node is null:
return
visit(node)
preorderTraversal(node.left)
preorderTraversal(node.right)

```

**Java Code**
```java
class Solution {
    // Function to return a list containing the preorder traversal of the tree.
    static ArrayList<Integer> preorder(Node root) {
        ArrayList<Integer> result = new ArrayList<>();
        preorderHelper(root, result);
        return result;
    }

    private static void preorderHelper(Node node, ArrayList<Integer> result) {
        if (node == null) return;

        result.add(node.data);               // Visit the root
        preorderHelper(node.left, result);   // Traverse left subtree
        preorderHelper(node.right, result);  // Traverse right subtree
    }
}

```

**Time Complexity**
- **O(n)** — Every node is visited exactly once.

**Space Complexity**
- **O(h)** — Recursive stack space, where `h` is the height of the tree.

**Notes:**  
- This is a depth-first traversal method.
- Preorder traversal is often used to create a copy of the tree or to get a prefix expression of an expression tree.

## Problem Statement: 308
**Inorder Traversal**

Given a binary tree, return its **Inorder Traversal**.

In an **inorder traversal**, the traversal order is:
1. Visit the left subtree.
2. Visit the root node.
3. Visit the right subtree.

**Example**
**Input Tree:**
 1
 / \
2   3
/
4 5

**Output:**
[4, 2, 5, 1, 3]


**Constraints**
- `1 ≤ Number of nodes ≤ 10⁴`
- Each node contains a unique integer value.

**Follow Up**
Try solving this with **O(1) auxiliary space**. This can be achieved using **Morris Inorder Traversal** which avoids the recursive stack or explicit stack used in typical inorder traversal.

**Goal:**  
To perform an inorder traversal of a binary tree and return a list of node values in the **Left → Root → Right** order.

**Approach**
The solution uses **Morris Inorder Traversal** for **O(1) space** (excluding output list).  
This method involves temporarily modifying the tree to create "threads" that link predecessors to successors.

**Steps:**
1. Initialize `curr` as root.
2. While `curr` is not null:
   - If `curr.left` is null:
     - Visit `curr`.
     - Move to `curr.right`.
   - Else:
     - Find the inorder predecessor of `curr`.
     - If predecessor's right is null, set it to `curr` and move `curr` to `curr.left`.
     - If predecessor's right is `curr`, reset it to null, visit `curr`, and move to `curr.right`.

**Pseudocode/Algorithm:**  
1. Initialize an empty result list.
2. Set `curr` to root.
3. While `curr` is not null:
   - If `curr.left` is null:
     - Add `curr.data` to result.
     - Move to `curr.right`.
   - Else:
     - Find the inorder predecessor of `curr`.
     - If predecessor's right is null:
       - Set predecessor's right to `curr`.
       - Move `curr` to `curr.left`.
     - Else:
       - Reset predecessor's right to null.
       - Add `curr.data` to result.
       - Move `curr` to `curr.right`.

**🧮 Flowchart/ Diagram:**  
```plaintext
            [curr == null?]
                  |
                 No
                  |
      [curr.left == null?]
           /          \
         Yes           No
         |              |
Add curr.data     Find predecessor
  to result            |
    |           [pre.right == null?]
Move to            /           \
curr.right       Yes           No
                  |             |
  pre.right = curr     pre.right = null
  curr = curr.left     Add curr.data
                       curr = curr.right
```

**Java Code**
```java
class Solution {
    // Function to return a list containing the inorder traversal of the tree.
    ArrayList<Integer> inOrder(Node root) {
        ArrayList<Integer> res = new ArrayList<>();
        Node curr = root;

        while (curr != null) {
            if (curr.left == null) {
                res.add(curr.data);
                curr = curr.right;
            } else {
                Node prev = curr.left;
                while (prev.right != null && prev.right != curr) {
                    prev = prev.right;
                }

                if (prev.right == null) {
                    prev.right = curr;
                    curr = curr.left;
                } else {
                    prev.right = null;
                    res.add(curr.data);
                    curr = curr.right;
                }
            }
        }
        return res;
    }
}
```

**Code:2**
```java
import java.util.*;

// Structure of a Binary Tree Node
class Node {
    int data;
    Node left, right;

    Node(int v)
    {
        data = v;
        left = right = null;
    }
}

class GfG {
    // Function to print inorder traversal
    public static void printInorder(Node node)
    {
        if (node == null)
            return;

        // First recur on left subtree
        printInorder(node.left);

        // Now deal with the node
        System.out.print(node.data + " ");

        // Then recur on right subtree
        printInorder(node.right);
    }

    public static void main(String[] args)
    {
        Node root = new Node(1);
        root.left = new Node(2);
        root.right = new Node(3);
        root.left.left = new Node(4);
        root.left.right = new Node(5);
        root.right.right = new Node(6);

        printInorder(root);
    }
}
```

**Time Complexity**
- **O(n)** — Each node is visited at most twice.

**Space Complexity**
- **O(1)** — No recursion or stack used, tree is modified temporarily.

**Notes:**  
- Morris Traversal temporarily modifies the tree by creating "threads" to the inorder predecessor, but restores the tree structure afterward.
- This traversal is efficient for memory-constrained environments because it avoids recursion and explicit stack usage.
- For interview purposes, it's important to know both the recursive and Morris methods and understand their trade-offs in space complexity.

## Problem Statement: 306
**Postorder Traversal**

Given the root of a binary tree, return a list containing the **Postorder Traversal** of the tree.

Postorder traversal is a depth-first traversal method where the nodes are visited in the following order:
1. Traverse the left subtree.
2. Traverse the right subtree.
3. Visit the root node.

**Example**
**Input Tree:**
  1
 / \
2   3
/
4 5

**Output:**
[4, 5, 2, 3, 1]


**Input:**  
- Root of a binary tree.  
- For example:  
```

```
1
```

/&#x20;
2   3
/&#x20;
4   5

```

**Output:**  
- A list representing the postorder traversal.  
- For the above tree: `[4, 5, 2, 3, 1]`

**Constraints**
- `1 ≤ Number of nodes ≤ 10⁴`
- Each node contains a unique integer value.

**Goal:**  
To collect and return the node values of a binary tree following the postorder traversal sequence.

**Approach**
Use a **recursive function** to perform postorder traversal:
1. Traverse the left subtree recursively.
2. Traverse the right subtree recursively.
3. Append the root node’s value to the result list.

This is a straightforward recursive implementation that mirrors the definition of postorder traversal.

**🧮 Flowchart/ Diagram:**  

```sql
    [Root]
     ↓
 Traverse Left
     ↓
Traverse Right
     ↓
  Visit Root
```


**Pseudocode/Algorithm:**  
```sql

postorderTraversal(node, result):
if node is null:
return
postorderTraversal(node.left, result)
postorderTraversal(node.right, result)
result.append(node.value)

```

**Java Code**
```java
class Solution {
    // Function to return a list containing the postorder traversal of the tree.
    ArrayList<Integer> postOrder(Node root) {
        ArrayList<Integer> result = new ArrayList<>();
        postOrderHelper(root, result);
        return result;
    }

    private void postOrderHelper(Node node, ArrayList<Integer> result) {
        if (node == null) return;

        postOrderHelper(node.left, result);   // Traverse left subtree
        postOrderHelper(node.right, result);  // Traverse right subtree
        result.add(node.data);                // Visit the root
    }
}
```

**Notes:**  
- Postorder traversal is useful in applications like deleting a tree or evaluating expression trees.
- This method can be implemented iteratively as well using two stacks, although recursion is more intuitive.

**Time Complexity**
- **O(n)** — Each node is visited exactly once.

**Space Complexity**
- **O(h)** — Recursion stack space, where `h` is the height of the tree.


## Problem Statement: 305

Given the root of a binary tree with `n` nodes, your task is to return the **level order traversal** of the tree.  
Level order traversal means traversing the tree **level by level** from left to right (also known as Breadth-First Search or BFS traversal).

**Example**
**Input:**  
A binary tree:
    1
   / \
  2   3
 / \
4   5

**Output:**  
[ [1], [2, 3], [4, 5] ]


**Constraints**
- `1 ≤ n ≤ 10⁴` (Number of nodes in the tree)
- Tree node values are integers.

**Input:**  
A binary tree root node.

**Output:**  
A nested list where each sublist contains the values of one level of the tree.

**Goal:**  
To collect the values of nodes level by level and return them in a structured nested list.

**Approach**
1. **Use a Queue for BFS**:  
   - Enqueue the root node.
   - For each level, dequeue all nodes at the current level, store their values in a list, and enqueue their children (if any).
2. **Repeat** the process level by level until the queue becomes empty.
3. Append each level's list to the final result list.

**🧮 Flowchart/ Diagram:**  
```sql

Initialize queue with root
↓
While queue not empty:
\- Record size of current level
\- For each node in level:
\- Dequeue node
\- Add its value to current level list
\- Enqueue its children
\- Add level list to result

```

**Pseudocode/Algorithm:**  
```sql

levelOrderTraversal(root):
if root is null:
return \[]

queue = new Queue()
result = []

queue.enqueue(root)

while queue is not empty:
    level = []
    size = queue.size()

    for i in 0 to size - 1:
        node = queue.dequeue()
        level.append(node.value)
        if node.left:
            queue.enqueue(node.left)
        if node.right:
            queue.enqueue(node.right)

    result.append(level)

return result
```

**Java Code**
```java
class Solution {
    public ArrayList<ArrayList<Integer>> levelOrder(Node root) {
        ArrayList<ArrayList<Integer>> result = new ArrayList<>();

        if (root == null) return result;

        Queue<Node> queue = new LinkedList<>();
        queue.add(root);

        while (!queue.isEmpty()) {
            int levelSize = queue.size();
            ArrayList<Integer> level = new ArrayList<>();

            for (int i = 0; i < levelSize; i++) {
                Node curr = queue.poll();
                level.add(curr.data);

                if (curr.left != null) queue.add(curr.left);
                if (curr.right != null) queue.add(curr.right);
            }

            result.add(level);
        }

        return result;
    }
}
```

**Time Complexity**
- **O(n)** — Each node is visited exactly once.

**Space Complexity**
- **O(n)** — In the worst case, the queue can hold all nodes at the last level.

**Notes:**  
- Level order traversal is useful for problems involving hierarchical structures.
- It can also be extended for applications such as zigzag traversal or finding the maximum value at each level.


## Problem Statement: 309
**Tree Boundary Traversal**

Given a Binary Tree, find its **Boundary Traversal** in an anti-clockwise direction. The traversal includes:

1. **Left Boundary** — Path from the root to the left-most node excluding leaf nodes. Always choose the left child over the right.
2. **Leaf Nodes** — All leaf nodes in left-to-right order.
3. **Right Boundary (reversed)** — Path from the right-most node to the root excluding leaf nodes. Always choose the right child over the left.

**Note:**  
- Do not include duplicate nodes.  
- If the root is a leaf or the only node in either boundary, handle accordingly.

**Example**
**Input Tree:**
   20
   /  \
 8     22
/ \      \
4 12 25 /
10 14

**Output:**
[20, 8, 4, 10, 14, 25, 22]


**Constraints**
- `1 ≤ Number of nodes ≤ 10⁴`
- Node values are unique integers.

**Input:**  
A binary tree represented by its root node.

**Output:**  
A list of node values representing the boundary in anti-clockwise order.

**Goal:**  
To print the boundary nodes of a binary tree in anti-clockwise order without duplicating any node.

**Approach**
To perform boundary traversal:
1. **Add Root** (if not a leaf).
2. **Add Left Boundary**: Traverse down using the left child first. Exclude leaf nodes.
3. **Add Leaf Nodes**: Perform DFS and add all leaves from left to right.
4. **Add Right Boundary (Reversed)**: Traverse down using the right child first. Exclude leaf nodes. Push to a stack, then pop to reverse the order.

**🧮 Flowchart/ Diagram:**  
```sql

Start
↓
Check if root is not leaf → Add root to result
↓
Traverse left boundary (excluding leaves)
↓
Collect all leaf nodes (DFS or Morris)
↓
Traverse right boundary (excluding leaves) → Reverse before adding
↓
Return final result list

```

**Pseudocode/Algorithm:**  
```sql

boundaryTraversal(root):
if root is null:
return \[]

result = []
if not isLeaf(root):
    result.append(root.data)

addLeftBoundary(root.left, result)
addLeaves(root, result)
addRightBoundary(root.right, result)

return result
```

**Java Code**
```java
class Solution {
    ArrayList<Integer> boundaryTraversal(Node node) {
        ArrayList<Integer> res = new ArrayList<>();
        if (node == null) return res;

        if (!isLeaf(node)) res.add(node.data);
        
        addLeftBoundary(node.left, res);
        addLeaves(node, res);
        addRightBoundary(node.right, res);
        
        return res;
    }

    boolean isLeaf(Node node) {
        return node.left == null && node.right == null;
    }

    void addLeftBoundary(Node node, ArrayList<Integer> res) {
        while (node != null) {
            if (!isLeaf(node)) res.add(node.data);
            node = node.left != null ? node.left : node.right;
        }
    }

    void addRightBoundary(Node node, ArrayList<Integer> res) {
        Stack<Integer> stack = new Stack<>();
        while (node != null) {
            if (!isLeaf(node)) stack.push(node.data);
            node = node.right != null ? node.right : node.left;
        }
        while (!stack.isEmpty()) res.add(stack.pop());
    }

    void addLeaves(Node node, ArrayList<Integer> res) {
        if (isLeaf(node)) {
            res.add(node.data);
            return;
        }
        if (node.left != null) addLeaves(node.left, res);
        if (node.right != null) addLeaves(node.right, res);
    }
}

```
**Code:**
```java
// Java implementation for Boundary Traversal of Binary Tree
// using Iteration with Morris Traversal
import java.util.ArrayList;

class Node {
    int data;
    Node left, right;

    Node(int x) {
        data = x;
        left = right = null;
    }
}

class GfG {

    static boolean isLeaf(Node node) {
        return node.left == null && node.right == null;
    }
	
  	// Function to collect the left boundary nodes
    static void collectBoundaryLeft(Node root, ArrayList<Integer> res) {
        if (root == null)
            return;

        Node curr = root;
        while (!isLeaf(curr)) {
            res.add(curr.data);

            if (curr.left != null)
                curr = curr.left;
            else
                curr = curr.right;
        }
    }

    // Function to collect the leaf nodes using Morris Traversal
    static void collectLeaves(Node root, ArrayList<Integer> res) {
        Node current = root;

        while (current != null) {
            if (current.left == null) {
              
                // If it's a leaf node
                if (current.right == null) 
                    res.add(current.data);

                current = current.right;
            } 
          	else {
              
                // Find the inorder predecessor
                Node predecessor = current.left;
                while (predecessor.right != null && predecessor.right != current) {
                    predecessor = predecessor.right;
                }

                if (predecessor.right == null) {
                    predecessor.right = current;
                    current = current.left;
                } 
              	else {
                  	// If it's predecessor is a leaf node
                    if (predecessor.left == null) 
                        res.add(predecessor.data);
                    
                    predecessor.right = null;
                    current = current.right;
                }
            }
        }
    }


    // Function to collect the right boundary nodes
    static void collectBoundaryRight(Node root, ArrayList<Integer> res) {
        if (root == null)
            return;

        Node curr = root;
        ArrayList<Integer> temp = new ArrayList<>();
        while (!isLeaf(curr)) {
            temp.add(curr.data);

            if (curr.right != null)
                curr = curr.right;
            else
                curr = curr.left;
        }

        for (int i = temp.size() - 1; i >= 0; --i)
            res.add(temp.get(i));
    }

    // Function to perform boundary traversal
    static ArrayList<Integer> boundaryTraversal(Node root) {
        ArrayList<Integer> res = new ArrayList<>();

        if (root == null)
            return res;

        // Add root data if it's not a leaf
        if (!isLeaf(root))
            res.add(root.data);

        // Collect left boundary
        collectBoundaryLeft(root.left, res);

        // Collect leaf nodes
        collectLeaves(root, res);

        // Collect right boundary
        collectBoundaryRight(root.right, res);

        return res;
    }

    public static void main(String[] args) {
        // Hardcoded Binary tree
        //        20
        //       /  \
        //      8    22
        //     / \     \
        //    4   12    25
        //       /  \
        //      10   14
        Node root = new Node(20);
        root.left = new Node(8);
        root.right = new Node(22);
        root.left.left = new Node(4);
        root.left.right = new Node(12);
        root.left.right.left = new Node(10);
        root.left.right.right = new Node(14);
        root.right.right = new Node(25);

        ArrayList<Integer> boundary = boundaryTraversal(root);

        for (int x : boundary)
            System.out.print(x + " ");
    }
}
```

**Time Complexity**
- **O(n)** — Each node is visited at most once.

**Space Complexity**
- **O(h)** — Height of tree for recursion stack and the stack used for right boundary (worst-case O(n)).

**Notes:**  
- Ensure leaves are not added twice if they appear on the boundary.
- For optimal performance and handling large trees, use iterative Morris Traversal for collecting leaves.
- Right boundary is reversed to maintain the anti-clockwise order.


## Problem Statement: 313
**Vertical Tree Traversal**

Given a root of a Binary Tree, find the **vertical traversal** of it starting from the leftmost level to the rightmost level.

If there are multiple nodes passing through a vertical line, then they should be printed as they appear in **level order traversal** of the tree.

**Example**
**Input Tree:**
       1
       / \
      2   3
     / \   \
    4   5   6

**Output:**
[ [4], [2], [1, 5, 6], [3] ]


**Explanation:**
- The vertical traversal starts from the leftmost column of nodes (4) and moves right. 
- For each vertical line, the nodes are listed from top to bottom (level order).

**Constraints**
- `1 ≤ Number of nodes ≤ 10⁴`
- Tree nodes' values are between `-10⁶` to `10⁶`.

**Approach**
1. Use **DFS traversal** to explore the tree.
2. Maintain a **horizontal distance (hd)** for each node:
   - The root starts at `hd = 0`.
   - The left child has `hd = hd - 1`, and the right child has `hd = hd + 1`.
3. Store nodes at each horizontal distance in a **HashMap**.
4. After DFS, sort the nodes by their horizontal distances.
5. For each vertical line, collect nodes in **level order** from top to bottom.

**Time Complexity**
- **O(n log n)** — Where `n` is the number of nodes in the tree. Sorting the vertical levels takes `O(n log n)`.

**Space Complexity**
- **O(n)** — Space is used by the HashMap and recursion stack.

**Java Code**
```java
class Pair {
    Node node;
    int hd;

    Pair(Node n, int h) {
        node = n;
        hd = h;
    }
}

class Solution {
    // DFS function to traverse the tree and record nodes at each horizontal distance
    static void DFS(Node root, int hd, int[] mn, 
                     HashMap<Integer, ArrayList<Integer>> mp) {
        if (root == null) 
            return;
        
        // Add the node to the corresponding horizontal distance in the map
        if (!mp.containsKey(hd)) 
            mp.put(hd, new ArrayList<>());
        
        mp.get(hd).add(root.data);
        mn[0] = Math.min(mn[0], hd);  // Track the minimum horizontal distance
        
        // Recursively call DFS on left and right subtrees
        DFS(root.left, hd - 1, mn, mp);
        DFS(root.right, hd + 1, mn, mp);
    }

    // Function to return the vertical order traversal of the binary tree
    public ArrayList<ArrayList<Integer>> verticalOrder(Node root) {
        HashMap<Integer, ArrayList<Integer>> mp = new HashMap<>();
        int[] mn = {0};  // Store the minimum horizontal distance
        
        // Start DFS traversal
        DFS(root, 0, mn, mp);
        
        // Prepare the result by collecting nodes at each horizontal distance
        ArrayList<ArrayList<Integer>> res = new ArrayList<>();
        int hd = mn[0];
        
        // Collect nodes for each vertical line in level order
        while (mp.containsKey(hd)) {
            res.add(mp.get(hd));
            hd++;
        }

        return res;
    }
}
```

**Top View of Binary Tree**

You are given a binary tree, and your task is to return its **top view**. The top view of a binary tree is the set of nodes visible when the tree is viewed from the top.

**Note:** 
- Return the nodes from the leftmost node to the rightmost node.
- If two nodes are at the same position (horizontal distance) and are outside the shadow of the tree, consider the leftmost node only.

**Example**
**Input Tree:**
        1
      /   \
     2     3
    / \     \
   4   5     6
                \
                 7

**Output:**
[4, 2, 1, 3, 7]


**Explanation:**
- From the top view, the visible nodes from left to right are: `4`, `2`, `1`, `3`, and `7`.

**Constraints**
- `1 ≤ Number of nodes ≤ 10⁴`
- `Tree nodes' values are between `-10⁶` to `10⁶`.

**Input:**  
A binary tree represented by its root node.

**Output:**  
A list of lists, each representing a vertical level from leftmost to rightmost.

**Goal:**  
Group all nodes that lie on the same vertical line and print them from top to bottom, left to right.

**Approach**
1. **Level Order Traversal**:
   - Perform a **breadth-first search (BFS)** on the tree, maintaining the **horizontal distance (hd)** of each node from the root.
   - If a node at horizontal distance `hd` has not been encountered before, it is part of the top view and should be added to the result.
   
2. **Horizontal Distance (hd)**:
   - Nodes to the left of the root have negative `hd`.
   - Nodes to the right of the root have positive `hd`.
   - The root node has an `hd` of 0.
   
3. **Maintain a HashMap**:
   - Use a `HashMap` to store the first node encountered at each horizontal distance (`hd`).
   
4. **Track the Minimum and Maximum hd**:
   - To ensure we capture all nodes from leftmost to rightmost, track the minimum and maximum horizontal distances encountered during the BFS.

**🧮 Flowchart/ Diagram:**  
```sql

Start
↓
Call DFS with root and hd = 0
↓
Store nodes by horizontal distance (hd) in a HashMap
↓
Update minimum hd
↓
Recurse left (hd - 1) and right (hd + 1)
↓
After DFS, collect nodes in order from min to max hd
↓
Return vertical order result

```

**Pseudocode/Algorithm:**  
```sql

verticalOrder(root):
map = {}
minHd = 0
DFS(root, 0, map, minHd)

result = []
for hd from minHd to max key in map:
    result.append(map[hd])

return result

DFS(node, hd, map, minHd):
if node is null:
return
map\[hd].append(node.data)
minHd = min(minHd, hd)
DFS(node.left, hd - 1, map, minHd)
DFS(node.right, hd + 1, map, minHd)

```

**Java Code**
```java
class Solution {
    static class Pair {
        Node node;
        int dist;
        Pair(Node node, int dist) {
            this.node = node;
            this.dist = dist;
        }
    }

    // Function to return a list containing the top view of the binary tree
    static ArrayList<Integer> topView(Node root) {
        // code here
        if (root == null) {
            return new ArrayList<>();
        }
        
        Queue<Pair> queue = new LinkedList<>();
        queue.add(new Pair(root, 0)); // Start with root node and horizontal distance 0
        
        HashMap<Integer, Integer> map = new HashMap<>();
        int minHD = 0;
        int maxHD = 0;
        
        // Perform level order traversal (BFS)
        while (!queue.isEmpty()) {
            Pair current = queue.poll();
            Node currentNode = current.node;
            int hd = current.dist;
            
            // Update the range of horizontal distances
            if (hd < minHD) {
                minHD = hd;
            }
            if (hd > maxHD) {
                maxHD = hd;
            }
            
            // Add the node to the map if it's not already present at this horizontal distance
            if (!map.containsKey(hd)) {
                map.put(hd, currentNode.data);
            }
            
            // Add the left and right children to the queue with updated horizontal distances
            if (currentNode.left != null) {
                queue.add(new Pair(currentNode.left, hd - 1));
            }
            if (currentNode.right != null) {
                queue.add(new Pair(currentNode.right, hd + 1));
            }
        }
        
        // Prepare the result list by collecting nodes from the map based on their horizontal distances
        ArrayList<Integer> topViewList = new ArrayList<>();
        for (int hd = minHD; hd <= maxHD; hd++) {
            if (map.containsKey(hd)) {
                topViewList.add(map.get(hd));
            }
        }
        
        return topViewList;
    }
}
```

**Time Complexity**
- **O(n)** — Where `n` is the number of nodes in the tree. Every node is visited once during the BFS.

**Space Complexity**
- **O(n)** — Space required for the queue in BFS and the map to store horizontal distances.

**Notes:**  
- A level-order based BFS approach can also be used to ensure true level order within vertical levels.
- This solution assumes that node insertion order during DFS simulates level order when there is no ambiguity due to node positions.

   - Left child is at hd - 1
   - Right child is at hd + 1

```java
class Solution {
    class Pair {
        Node node;
        int hd;

        Pair(Node node, int hd) {
            this.node = node;
            this.hd = hd;
        }
    }

    public ArrayList<Integer> bottomView(Node root) {
        if (root == null) return new ArrayList<>();

        Map<Integer, Integer> hdMap = new TreeMap<>();
        Queue<Pair> q = new LinkedList<>();
        q.add(new Pair(root, 0));

        while (!q.isEmpty()) {
            Pair p = q.poll();
            Node curr = p.node;
            int hd = p.hd;

            // Update with latest node at that horizontal distance
            hdMap.put(hd, curr.data);

            if (curr.left != null) {
                q.add(new Pair(curr.left, hd - 1));
            }
            if (curr.right != null) {
                q.add(new Pair(curr.right, hd + 1));
            }
        }

        ArrayList<Integer> result = new ArrayList<>();
        for (int value : hdMap.values()) {
            result.add(value);
        }

        return result;
    }
}
```

**Code:**
```java
// Java program to print right view of Binary
// tree using modified Morris Traversal
import java.util.ArrayList;

class Node {
    int data;
    Node left, right;

    Node(int x) {
        data = x;
        left = right = null;
    }
}

// Function to return the right view of the binary tree
class GfG {
    
    public static ArrayList<Integer> rightView(Node root) {
        
        // To store the right view nodes
        ArrayList<Integer> res = new ArrayList<>();
        
        // Current level of traversal
        int level = 0;

        // Traverse the tree using modified 
        // Morris Traversal
        while (root != null) {
            
            // If the node has a right child,
            // find the inorder predecessor
            if (root.right != null) {
                Node pred = root.right;
                int backDepth = 1;

                // Find the leftmost node in the
                // right subtree
                while (pred.left != null
                                 && pred.left != root) {
                    pred = pred.left;
                    backDepth++;
                }

                // If threading is not yet established
                if (pred.left == null) {
                    
                    // Add the current node to the view if 
                    // visiting the level for the first time
                    if (res.size() == level) {
                        res.add(root.data);
                    }

                    // Establish the thread and move 
                    // to the right subtree
                    pred.left = root;
                    root = root.right;
                    level++;
                } 
                else {
                    
                    // Threading was already done 
                    // (second visit) remove the thread 
                    // and go to the left subtree
                    pred.left = null;
                    root = root.left;
                    level -= backDepth;
                }
            } 
            else {
                
                // If no right child, process the current 
                // node and move to the left child
                if (res.size() == level) {
                    res.add(root.data);
                }
                
                root = root.left;
                level++;
            }
        }

        return res;
    }

    static void printArray(ArrayList<Integer> arr) {
        for (int val : arr) {
            System.out.print(val + " ");
        }
        System.out.println();
    }

    public static void main(String[] args) {
        
        // Representation of the input tree:
        //         1
        //        / \
        //       2   3
        //          / \
        //         4   5 
        Node root = new Node(1);
        root.left = new Node(2);
        root.right = new Node(3);
        root.right.left = new Node(4);
        root.right.right = new Node(5);

        ArrayList<Integer> result = rightView(root);
        
        printArray(result);
    }
}
```

**Time Complexity**
- **O(n)** — Where `n` is the number of nodes in the tree. Every node is visited once.

**Space Complexity**
- **O(h)** — Where `h` is the height of the tree due to the recursion stack.

**Notes:**  
- The iterative approach using modified Morris Traversal avoids recursion and stack usage.  
- Morris traversal adapts inorder threading for right view collection.  
- The recursive approach is simpler and more intuitive but uses recursion stack.

Diagonal traversal considers lines of slope `-1` that pass between nodes. All nodes appearing in the same diagonal (having the same slope) should be grouped together and returned in order. 

If diagonal elements are present in different subtrees, **elements from the left subtree diagonals should come before the right subtree diagonals**.

   / \   /

Return the diagonal traversal of the binary tree as a single list, grouping nodes by their diagonal levels from top-right to bottom-left.

**Approach**
- Use a recursive helper function `diagonalRecur` to traverse the tree diagonally.
- Track the current diagonal level starting from `0`.
- For each node:
  - Add it to the list corresponding to its diagonal level.
  - Recur for the left child with `level + 1` (moves to next diagonal).
  - Recur for the right child with the same `level` (remains on same diagonal).
- After traversal, concatenate all values from the map in ascending order of level keys.

**🧮 Flowchart/ Diagram:**  
- Start at root with diagonal level 0  
- Add node to corresponding diagonal list  
- Move left child to next diagonal (level + 1)  
- Move right child stays on same diagonal (level)  
- After traversal, concatenate all diagonal lists in order  

**Pseudocode/Algorithm:**  
```sql

function diagonalRecur(node, level, map):
if node is null:
return
add node.data to map\[level]
diagonalRecur(node.left, level + 1, map)
diagonalRecur(node.right, level, map)

function diagonal(root):
map = empty map
diagonalRecur(root, 0, map)
result = empty list
level = 0
while map contains level:
append all elements of map\[level] to result
level += 1
return result

```

**Java Code**
```java
class Tree {
    static void diagonalRecur(Node root, int level, HashMap<Integer, ArrayList<Integer>> levelData) {
        if (root == null)
            return;
        levelData.computeIfAbsent(level, k -> new ArrayList<>()).add(root.data);
        diagonalRecur(root.left, level + 1, levelData);  // Move to next diagonal
        diagonalRecur(root.right, level, levelData);     // Stay on same diagonal
    }

    public ArrayList<Integer> diagonal(Node root) {
        ArrayList<Integer> ans = new ArrayList<>();
        HashMap<Integer, ArrayList<Integer>> levelData = new HashMap<>();
        diagonalRecur(root, 0, levelData);

        int level = 0;
        while (levelData.containsKey(level)) {
            ans.addAll(levelData.get(level));
            level++;
        }

        return ans;
    }
}
```

**Goal:**  
Insert a node with the given key into the BST such that the BST property is preserved.

**Approach:**  
- If the tree is empty, create a new node and return it as the root.  
- Otherwise, traverse the tree starting from the root:  
  - Compare the given key with the current node’s key.  
  - If the key is less, go to the left subtree; if greater, go to the right subtree.  
  - If the key is equal to the current node’s key, no insertion is needed (keys are unique).  
- Once a null position is found where the new node can be inserted, create the node and link it as the child of the last non-null node.  

**🧮 Flowchart / Diagram:**  
```sql

Start -> Is root null? -> Yes -> Create new node as root -> End
|
No
|
Compare key with root.key
/           \\
key < root.key     key > root.key
/                  \\
Move to left subtree   Move to right subtree
\|                     |
Repeat until null      Repeat until null
\|                     |
Insert node here        Insert node here
\|                     |
End                   End

```

**Pseudocode / Algorithm:** 

```sql

function insert(root, key):
newNode = Node(key)

if root is null:
    return newNode

parent = null
curr = root

while curr is not null:
    parent = curr
    if key < curr.key:
        curr = curr.left
    else if key > curr.key:
        curr = curr.right
    else:
        return root  // key already exists

if key < parent.key:
    parent.left = newNode
else:
    parent.right = newNode

return root
```

**Code:**

```java
class Node {
    int key;
    Node left, right;

    public Node(int item) {
        key = item;
        left = right = null;
    }
}

class GfG {

    // Function to insert a new node with the given key
    static Node insert(Node root, int x) {
        Node temp = new Node(x);

        // If tree is empty
        if (root == null) {
            return temp;
        }

        // Find the node who is going to have the new node temp as its child
        Node parent = null;
        Node curr = root;
        while (curr != null) {
            parent = curr;
            if (curr.key > x) {
                curr = curr.left;
            } else if (curr.key < x) {
                curr = curr.right;
            } else {
                return root; // Key already exists
            }
        }

        // If x is smaller, make it left child, else right child
        if (parent.key > x) {
            parent.left = temp;
        } else {
            parent.right = temp;
        }
        return root;
    }

    // A utility function to do inorder tree traversal
    static void inorder(Node root) {
        if (root != null) {
            inorder(root.left);
            System.out.print(root.key + " ");
            inorder(root.right);
        }
    }

    // Driver method
    public static void main(String[] args) {
        Node root = null;

        // Creating the following BST
        //      50
        //     /  \
        //    30   70
        //   / \   / \
        //  20 40 60 80

        root = insert(root, 50);
        root = insert(root, 30);
        root = insert(root, 20);
        root = insert(root, 40);
        root = insert(root, 70);
        root = insert(root, 60);
        root = insert(root, 80);

        // Print inorder traversal of the BST
        inorder(root);
    }
}
```

* Worst case: O(N), when the BST becomes skewed (like a linked list).

**Space Complexity:**

* O(1) auxiliary space for iterative insertion.
* O(H) for recursion stack if implemented recursively, where H is height of the BST.

**Notes:**

* The insertion maintains the BST property where left subtree nodes are less than the parent node and right subtree nodes are greater.
* Duplicate keys are not inserted.
* Inorder traversal of BST produces sorted order of the keys.

## Problem Statement: 316
**Height of Binary Tree**

Given a binary tree, find its **height**.
The height of a tree is defined as the number of **edges** on the longest path from the root to a leaf node. A **leaf node** is a node that does not have any children.

**Example**

**Input:**

      1
    /   \
   2     3
  / \
 4   5

**Output:**
3


**Explanation:**
- The longest path from the root to a leaf is `1 -> 2 -> 4`, which contains 3 edges.
- Therefore, the height of the tree is 3.

**Constraints**
- `1 ≤ Number of nodes ≤ 10⁴`
- Node values are between `-10⁶` and `10⁶`.

**Goal:**  
Find the height of the binary tree defined as the number of edges on the longest path from root to leaf.

**Approach**
1. **Level Order Traversal (BFS)**:
   - Perform a **breadth-first search (BFS)** to calculate the height of the tree.
   - For each level, count the nodes. Each time we traverse a level, increase the depth count.

**🧮 Flowchart/ Diagram:**  
- Start at root  
- Recursively calculate height of left subtree  
- Recursively calculate height of right subtree  
- Height = max(leftHeight, rightHeight) + 1 (for current edge)  
- Return height  

**Pseudocode/Algorithm:**  
```sql

function height(node):
if node is null:
return -1
leftHeight = height(node.left)
rightHeight = height(node.right)
return max(leftHeight, rightHeight) + 1

```

**Java Code**
```java
class Solution {
    // Function to find the height of a binary tree.
    int height(Node root) {
        // code here
        if (root == null) return 0;
        Queue<Node> q = new LinkedList<>();
        q.add(root);
        int depth = 0;
        
        // Perform level order traversal (BFS)
        while (!q.isEmpty()) {
            int levelSize = q.size();
            for (int i = 0; i < levelSize; i++) {
                Node curr = q.poll();
                
                // Add left and right children to the queue
                if (curr.left != null) q.add(curr.left);
                if (curr.right != null) q.add(curr.right);
            }
            // Increment depth after each level
            depth++;
        }
        
        // Return the height, which is the depth minus 1 (number of edges)
        return depth - 1;
    }
}
```

```java
// Java program to see if two trees are identical
// using Morris Traversal
import java.util.*;

class Node {
    int data;
    Node left, right;

    Node(int val) {
        data = val;
        left = right = null;
    }
}

class GfG {

    // Function to check if two trees are identical
    // using Morris traversal
    static boolean isIdentical(Node r1, Node r2) {

        // Check if both trees are empty
        if (r1 == null && r2 == null)
            return true;

        // Check if one tree is empty and the other is not
        if (r1 == null || r2 == null)
            return false;

        // Morris traversal to compare both trees
        while (r1 != null && r2 != null) {

            // Compare the data of both current nodes
            if (r1.data != r2.data)
                return false;

            // Morris traversal for the first tree (r1)
            if (r1.left == null) {

                // Move to the right child if no left child
                r1 = r1.right;
            } 
            else {

                // Find the inorder predecessor of r1
                Node pre = r1.left;
                while (pre.right != null && pre.right != r1) {
                    pre = pre.right;
                }

                // Set the temporary link to r1
                if (pre.right == null) {
                    pre.right = r1;
                    r1 = r1.left;
                } 

                // Remove the temporary link and move to right
                else {
                    pre.right = null;
                    r1 = r1.right;
                }
            }

            // Morris traversal for the second tree (r2)
            if (r2.left == null) {

                // Move to the right child if no left child
                r2 = r2.right;
            } 
            else {

                // Find the inorder predecessor of r2
                Node pre = r2.left;
                while (pre.right != null && pre.right != r2) {
                    pre = pre.right;
                }

                // Set the temporary link to r2
                if (pre.right == null) {
                    pre.right = r2;
                    r2 = r2.left;
                } 

                // Remove the temporary link and move to right
                else {
                    pre.right = null;
                    r2 = r2.right;
                }
            }
        }

        // Both trees are identical if both are 
        // null at the end
        return (r1 == null && r2 == null);
    }

    public static void main(String[] args) {

        // Representation of input binary tree 1
        //        1
        //       / \
        //      2   3
        //     /
        //    4
        Node r1 = new Node(1);
        r1.left = new Node(2);
        r1.right = new Node(3);
        r1.left.left = new Node(4);

        // Representation of input binary tree 2
        //        1
        //       / \
        //      2   3
        //     /
        //    4
        Node r2 = new Node(1);
        r2.left = new Node(2);
        r2.right = new Node(3);
        r2.left.left = new Node(4);

        if (isIdentical(r1, r2))
            System.out.println("Yes");
        else
            System.out.println("No");
    }
}
```

**Sum Tree**
Given a Binary Tree, check for the **Sum Tree** for every node except the leaf node. Return `true` if it is a Sum Tree, otherwise return `false`.

A **Sum Tree** is a Binary Tree where the value of a node is equal to the sum of the nodes present in its left subtree and right subtree. An empty tree is also considered a Sum Tree as the sum of an empty tree can be considered to be 0. A leaf node is also considered a Sum Tree.

true

  - For node `10`, `4 + 6 = 10`, and the value of `10` is equal to the sum of its children.
  - For node `3`, `0 + 3 = 3`, and the value of `3` is equal to the sum of its children.
  - Leaf nodes are considered Sum Trees by default.

**Constraints**
- `1 ≤ Number of nodes ≤ 10⁵`
- The binary tree contains unique values.

**Goal:** Verify if a given binary tree is a Sum Tree, where each non-leaf node's value equals the sum of its left and right subtrees.  

**Approach**
**Check for Sum Tree Recursively**:
   - The sum tree condition requires that the value of each non-leaf node should be the sum of the values of its left and right subtrees.
   - For each node, check if it is a leaf node. If it is, return `true`.
   - For non-leaf nodes, recursively check the left and right subtrees to ensure their sum matches the node's value.

**🧮 Flowchart/ Diagram:**  
- Start at root  
- If node is null or leaf -> return true  
- Recursively check left and right subtrees  
- Calculate sum of left subtree: if leaf, use node value; else use 2 * node value  
- Calculate sum of right subtree similarly  
- Check if current node's value == sum of left and right subtree sums  
- Return true if all checks pass; else false  

**Pseudocode/Algorithm:**  
```sql

function isSumTree(node):
if node == null or isLeaf(node):
return true
if isSumTree(node.left) and isSumTree(node.right):
ls = (node.left == null) ? 0 : (isLeaf(node.left) ? node.left.data : 2 \* node.left.data)
rs = (node.right == null) ? 0 : (isLeaf(node.right) ? node.right.data : 2 \* node.right.data)
return node.data == ls + rs
return false

```

**Java Code**
```java
class Solution {
    // Helper function to check if a node is a leaf.
    static boolean isLeaf(Node node) {
        if (node == null)
            return false;
        if (node.left == null && node.right == null)
            return true;
        return false;
    }

    // Function to check if the binary tree is a Sum Tree.
    boolean isSumTree(Node root) {
        int ls, rs;
        // Base case: empty tree or leaf node is considered a Sum Tree.
        if (root == null || isLeaf(root))
            return true;

        // Recursively check the left and right subtrees.
        if (isSumTree(root.left) && isSumTree(root.right)) {
            // Left subtree sum (if leaf, take its data, else double the node value).
            if (root.left == null)
                ls = 0;
            else if (isLeaf(root.left))
                ls = root.left.data;
            else
                ls = 2 * (root.left.data);

            // Right subtree sum (if leaf, take its data, else double the node value).
            if (root.right == null)
                rs = 0;
            else if (isLeaf(root.right))
                rs = root.right.data;
            else
                rs = 2 * (root.right.data);

            // Check if the node value is equal to the sum of left and right subtree nodes.
            return root.data == ls + rs;
        }
        return false;
    }
}
```

**Time Complexity**:
   - **O(n)** where `n` is the number of nodes in the tree. We visit each node once in the worst case.

**Space Complexity**:
   - **O(h)** where `h` is the height of the tree due to recursion stack space.

**Notes:**  
- The doubling (2 * node.left.data) works because in sum tree, non-leaf node's value is sum of its subtree, so subtree sum for non-leaf node is twice the node's value.

```java
class Solution {
    // Function to print level order traversal of the tree (for testing purposes)
    static void levelOrder(Node root) {
        if (root == null) {
            System.out.print("N ");
            return;
        }

        Queue<Node> queue = new LinkedList<>();
        queue.add(root);

        while (!queue.isEmpty()) {
            Node curr = queue.poll();

            if (curr == null) {
                System.out.print("N ");
                continue;
            }
            System.out.print(curr.data + " ");
            queue.add(curr.left);
            queue.add(curr.right);
        }
    }

    // Function to convert the binary tree into its mirror tree.
    void mirror(Node root) {
        // If the tree is empty, return
        if (root == null)
            return;

        // Recursively mirror the left and right subtrees
        mirror(root.left);
        mirror(root.right);

        // Swap the left and right children
        Node temp = root.left;
        root.left = root.right;
        root.right = temp;
    }
}
```

Given two Binary Trees, check whether they are Isomorphic or not.
Two trees are called **isomorphic** if one can be obtained from another by a series of flips, i.e., by swapping left and right children of several nodes. Any number of nodes at any level can have their children swapped.
Two empty trees are considered isomorphic.

**Example**
**Input:**
   1                   1
  / \                 / \
 2   3               3   2
/ \   \             /     \
4 5    6            6      4

**Output:**
True

**Explanation:**
- By swapping the left and right subtrees of the node `3` in the first tree, both trees become isomorphic.

**Constraints**
- `1 ≤ Number of nodes ≤ 10⁵`
- Each node will have a unique value.

**Goal:** Determine whether two binary trees are isomorphic, meaning one can be transformed into the other by swapping left and right children of any number of nodes.

**Approach**
**Recursive Traversal**:
   - The approach is based on recursively comparing the structures of both trees.
   - For each pair of nodes in the two trees:
     - If both nodes are `null`, they are isomorphic.
     - If only one of the nodes is `null`, they are not isomorphic.
     - If the data of the nodes is different, they are not isomorphic.
     - Otherwise, we check for two conditions:
       - If the left subtree of the first tree is isomorphic to the left subtree of the second tree, and the right subtree of the first tree is isomorphic to the right subtree of the second tree.
       - OR, if the left subtree of the first tree is isomorphic to the right subtree of the second tree, and the right subtree of the first tree is isomorphic to the left subtree of the second tree.

**🧮 Flowchart/ Diagram:**
- Start with root nodes of both trees
- If both nodes are null, return true
- If only one node is null, return false
- If node values differ, return false
- Recursively check:
  - Left subtree of first with left subtree of second AND right subtree of first with right subtree of second
  - OR left subtree of first with right subtree of second AND right subtree of first with left subtree of second
- Return true if either condition holds

**Pseudocode/Algorithm:**
```sql

function isIsomorphic(node1, node2):
if node1 == null and node2 == null:
return true
if node1 == null or node2 == null:
return false
if node1.data != node2.data:
return false
return (isIsomorphic(node1.left, node2.left) and isIsomorphic(node1.right, node2.right))
or (isIsomorphic(node1.left, node2.right) and isIsomorphic(node1.right, node2.left))

```

**Java Code**
```java
class Solution {
    // Return True if the given trees are isomorphic, else return False.
    boolean isIsomorphic(Node root1, Node root2) {
        // If both trees are empty, they are isomorphic
        if (root1 == null && root2 == null) {
            return true;
        }
        
        // If only one tree is empty, they are not isomorphic
        if (root1 == null || root2 == null) {
            return false;
        }
        
        // If the data of the nodes is not the same, they are not isomorphic
        if (root1.data != root2.data) {
            return false;
        }
        
        // Check if one of the following two conditions is true:
        // 1. Left subtree of root1 is isomorphic to left subtree of root2, and
        //    right subtree of root1 is isomorphic to right subtree of root2.
        // 2. Left subtree of root1 is isomorphic to right subtree of root2, and
        //    right subtree of root1 is isomorphic to left subtree of root2.
        return (isIsomorphic(root1.left, root2.left) && 
                isIsomorphic(root1.right, root2.right)) ||
               (isIsomorphic(root1.left, root2.right) && 
                isIsomorphic(root1.right, root2.left));
    }
}
```

**Space Complexity**:
   - **O(h)** where `h` is the height of the tree (due to recursive stack).

- The recursion explores both swapped and non-swapped subtree comparisons.

1. **Recursive Traversal**:
   - We will perform a Depth-First Search (DFS) to traverse the binary tree.
   - Starting from the root, we move down each path, and once we reach a leaf node, we add the current path to the result.
   - The function will backtrack after reaching each leaf node, removing the last node in the path to explore other potential paths.

**🧮 Flowchart/ Diagram:**  
- Start at root  
- Traverse down the left subtree, appending nodes to current path  
- Upon reaching a leaf, record the current path  
- Backtrack and traverse right subtree similarly  
- Repeat until all leaf nodes are processed

```sql

function collectPaths(node, path, result):
if node is null:
return
append node.data to path
if node is leaf:
add copy of path to result
else:
collectPaths(node.left, path, result)
collectPaths(node.right, path, result)
remove last node from path (backtrack)

```

```java
class Solution {
    // Helper function to collect paths from root to leaf nodes.
    static void collectPaths(Node node, ArrayList<Integer> path, 
                             ArrayList<ArrayList<Integer>> paths) {
        if (node == null) 
            return;
        
        // Add the current node to the path
        path.add(node.data);
        
        // If it's a leaf node, add the path to the result
        if (node.left == null && node.right == null) {
            paths.add(new ArrayList<>(path));
        } else {
            // Recur for the left and right subtrees
            collectPaths(node.left, path, paths);
            collectPaths(node.right, path, paths);
        }
        
        // Backtrack by removing the current node from the path
        path.remove(path.size() - 1);
    }

    // Main function to return all paths from root to leaf nodes
    public static ArrayList<ArrayList<Integer>> Paths(Node root) {
        ArrayList<ArrayList<Integer>> paths = new ArrayList<>();
        ArrayList<Integer> path = new ArrayList<>();
        collectPaths(root, path, paths);  // Collect all paths
        return paths;
    }
}
```

- `-10^4 ≤ Node value ≤ 10^4`
- `-10^4 ≤ k ≤ 10^4`

**Goal:**  
Count the number of downward paths in the binary tree whose node values sum to `k`.

**Approach**
1. **Recursive Approach with HashMap**:
   - Traverse the tree using Depth First Search (DFS).
   - Maintain a running sum (`currSum`) from the root to the current node.
   - Use a HashMap (`prefSums`) to store the frequency of previous prefix sums.
   - For each node, check if there is a previously seen sum `currSum - k`. This would indicate that there exists a downward path whose sum is equal to `k`.
   - Continue this for both the left and right subtrees.

**🧮 Flowchart/ Diagram:**  
- Initialize prefix sums map with empty state  
- Traverse the tree with DFS, maintaining current prefix sum  
- At each node:  
  - Update current sum by adding node value  
  - Check prefix sums map for (currSum - k) to find valid paths  
  - Update prefix sums map with current sum frequency  
  - Recur for left and right children  
  - Backtrack by decrementing current sum frequency in map  

**Pseudocode/Algorithm:**  
```sql

function countPathsUtil(node, k, currSum, prefSums):
if node is null:
return 0

currSum += node.data
pathCount = 0

if currSum == k:
    pathCount += 1

pathCount += prefSums.get(currSum - k, 0)

prefSums[currSum] = prefSums.get(currSum, 0) + 1

pathCount += countPathsUtil(node.left, k, currSum, prefSums)
pathCount += countPathsUtil(node.right, k, currSum, prefSums)

prefSums[currSum] -= 1

return pathCount
```

**Java Code**
```java
class Solution {
    // Utility function to count the paths.
    static int countPathsUtil(Node node, int k, int currSum, 
                              HashMap<Integer, Integer> prefSums) {
        if (node == null)
            return 0;

        int pathCount = 0;
        // Add current node's value to the current sum.
        currSum += node.data;
        
        // If the current sum is equal to k, it's a valid path.
        if (currSum == k)
            pathCount++;
        
        // Check if there is a previously seen sum that forms a valid path.
        pathCount += prefSums.getOrDefault(currSum - k, 0);
        
        // Store the current sum in the prefix sum map.
        prefSums.put(currSum, prefSums.getOrDefault(currSum, 0) + 1);

        // Explore the left and right subtrees.
        pathCount += countPathsUtil(node.left, k, currSum, prefSums);
        pathCount += countPathsUtil(node.right, k, currSum, prefSums);
        
        // Remove the current sum from the map (backtracking).
        prefSums.put(currSum, prefSums.get(currSum) - 1);

        return pathCount;
    }

    // Main function to return the number of paths that sum to k.
    public int sumK(Node root, int k) {
        // Map to store prefix sums and their frequencies.
        HashMap<Integer, Integer> prefSums = new HashMap<>();
        
        // Start the recursion to count the paths.
        return countPathsUtil(root, k, 0, prefSums);
    }
}
```

**Notes:**

- Handles negative and positive values for nodes and target sum `k`.

   - **Step 5:** Return the nodes sorted in ascending order.

**🧮 Flowchart/ Diagram:**  
- Start DFS from root to find target node  
- When target found, find all nodes at distance k in its subtree  
- While returning, check if nodes in sibling subtree or ancestors are at distance k  
- Collect all qualifying nodes  
- Sort the collected nodes before returning  

**Pseudocode/Algorithm:**  
```sql

function kDistanceRecur(root, target, k, ans):
if root is null:
return -1

if root.data == target:
    findNodes(root, k, ans)
    return 1

leftDist = kDistanceRecur(root.left, target, k, ans)
if leftDist != -1:
    if k - leftDist == 0:
        ans.add(root.data)
    else:
        findNodes(root.right, k - leftDist - 1, ans)
    return leftDist + 1

rightDist = kDistanceRecur(root.right, target, k, ans)
if rightDist != -1:
    if k - rightDist == 0:
        ans.add(root.data)
    else:
        findNodes(root.left, k - rightDist - 1, ans)
    return rightDist + 1

return -1

function findNodes(root, dist, ans):
if root is null:
return
if dist == 0:
ans.add(root.data)
return
findNodes(root.left, dist-1, ans)
findNodes(root.right, dist-1, ans)

```

**Java Code**
```java
class Solution {
    // Helper function to find nodes at a given distance from a node
    static void findNodes(Node root, int dis, ArrayList<Integer> ans) {
        if (root == null)
            return;

        // If we have reached the desired distance, add the node to the result
        if (dis == 0) {
            ans.add(root.data);
            return;
        }

        // Recursively find nodes at distance dis-1 in both left and right subtrees
        findNodes(root.left, dis - 1, ans);
        findNodes(root.right, dis - 1, ans);
    }

    // Recursive function to find the target node and handle the calculation
    static int kDistanceRecur(Node root, int target, int k, ArrayList<Integer> ans) {
        if (root == null)
            return -1;

        // If the target node is found, start finding nodes at distance k
        if (root.data == target) {
            findNodes(root, k, ans);
            return 1;
        }

        // Search for the target node in the left subtree
        int left = kDistanceRecur(root.left, target, k, ans);
        if (left != -1) {
            // If the target is found in the left subtree, handle both the current node and right subtree
            if (k - left == 0)
                ans.add(root.data);
            else
                findNodes(root.right, k - left - 1, ans);
            return left + 1;
        }

        // Search for the target node in the right subtree
        int right = kDistanceRecur(root.right, target, k, ans);

        if (right != -1) {
            // If the target is found in the right subtree, handle both the current node and left subtree
            if (k - right == 0)
                ans.add(root.data);
            else
                findNodes(root.left, k - right - 1, ans);
            return right + 1;
        }

        return -1;
    }

    // Main function to return the list of nodes at distance k from the target node
    public static ArrayList<Integer> KDistanceNodes(Node root, int target, int k) {
        ArrayList<Integer> ans = new ArrayList<>();
        // Call the recursive function to process the tree
        kDistanceRecur(root, target, k, ans);

        // Sort the list of nodes in ascending order
        Collections.sort(ans);

        return ans;
    }
}
```

**Notes:**  
- The solution handles the absence of parent pointers by using recursion and distance calculations.  
- Sorting at the end ensures output is in ascending order.  
- The approach covers all nodes both below and above the target node at the specified distance.

2. **Key Observations:**

   - Therefore, for each node:
     - Find the minimum value among its descendants.
     - Compute the difference between the node's value and this minimum descendant value.
     - Keep track of the maximum such difference.

3. **Recursive Strategy:**
   - Traverse the tree recursively.
   - For each node, recursively find the minimum value in its left and right subtrees.
   - Update the maximum difference by comparing `node.data - minValueAmongDescendants`.
   - Return the minimum value between the current node's value and the minimum from its subtrees to the parent call.

4. **Steps:**
   - Initialize `diff` as `Integer.MIN_VALUE` to keep track of the maximum difference.
   - Define a helper function `maxDiffUtil(Node root)`:

**Solution Code:**
```java
class Tree {
    private int diff;

    // Function to return the maximum difference between any node and its ancestor
    int maxDiff(Node root) {
        if (root == null) {
            return 0;
        }
        diff = Integer.MIN_VALUE;
        maxDiffUtil(root);
        return diff;
    }

    private int maxDiffUtil(Node root) {
        if (root == null) {
            return Integer.MAX_VALUE;
        }

        // Recursively get the minimum value from left and right subtree
        int leftMin = maxDiffUtil(root.left);
        int rightMin = maxDiffUtil(root.right);

        // Find the minimum value among the descendants
        int minDescendant = Math.min(leftMin, rightMin);

        // Update the maximum difference
        diff = Math.max(diff, root.data - minDescendant);

        // Return the minimum value between current node and its descendants
        return Math.min(root.data, minDescendant);
    }
}
```

**Goal:**  
Calculate the minimum time to burn the entire tree starting from the target node.

**Approach**
1. **Breadth-First Search (BFS)**:
   - First, perform a BFS to **map each node to its parent node** using a queue. This will help us know which nodes are connected to the target.
   
2. **Starting Point**:
   - The **target node** is set on fire first. We need to calculate the time it takes for the fire to propagate through the entire tree.
   
3. **Propagation of Fire**:
   - Each second, the fire spreads to all nodes that are **connected** to the nodes that are already on fire. This includes the **parent, left child, and right child** of the current nodes.

4. **Time Calculation**:
   - The BFS will continue to propagate the fire level by level, and the time taken will be equal to the number of levels traversed by BFS.

**🧮 Flowchart/ Diagram:**  
- BFS traversal from root to map parent nodes  
- BFS from target node to propagate fire  
- Count levels until all nodes are burned  

**Pseudocode/Algorithm:**  
```sql

function minTime(root, target):
map each node to its parent using BFS
find target node reference
initialize queue with target node
initialize visited set
time = -1
while queue not empty:
for each node in current level:
mark node visited
enqueue unvisited neighbors (left, right, parent)
increment time
return time

```

**Java Code**
```java
class Solution {
    /*class Node {
        int data;
        Node left;
        Node right;

        Node(int data) {
            this.data = data;
            left = null;
            right = null;
        }
    }*/

    public static int minTime(Node root, int target) {
        // If the root is null, return -1
        if (root == null)
            return -1;

        // Queue for BFS traversal
        Queue<Node> q = new LinkedList<>();
        q.add(root);
        Node tar = null;

        // HashMap to map child nodes to their parent nodes
        Map<Node, Node> par = new HashMap<>();
        par.put(root, null);

        // Perform BFS to find the target node and map parent nodes
        while (!q.isEmpty()) {
            Node curr = q.poll();

            // Set tar = curr if the current node's data is equal to the target
            if (curr.data == target)
                tar = curr;

            // Map the left child to its parent
            if (curr.left != null) {
                par.put(curr.left, curr);
                q.add(curr.left);
            }

            // Map the right child to its parent
            if (curr.right != null) {
                par.put(curr.right, curr);
                q.add(curr.right);
            }
        }

        // HashMap to check if a node has been visited or not
        Map<Node, Boolean> vis = new HashMap<>();

        // Variable to track the time
        int ans = -1;

        // Start the BFS from the target node
        q.add(tar);

        // Perform BFS to propagate the fire to all nodes
        while (!q.isEmpty()) {
            int size = q.size();
            while (size-- > 0) {
                Node curr = q.poll();
                vis.put(curr, true);

                // Add left child to the queue if it's not visited
                if (curr.left != null && !vis.getOrDefault(curr.left, false))
                    q.add(curr.left);
                
                // Add right child to the queue if it's not visited
                if (curr.right != null && !vis.getOrDefault(curr.right, false))
                    q.add(curr.right);
                
                // Add parent node to the queue if it's not visited
                if (par.get(curr) != null && !vis.getOrDefault(par.get(curr), false))
                    q.add(par.get(curr));
            }
            ans++; // Increment the time after each level of BFS
        }

        // Return the minimum time required to burn the entire tree
        return ans;
    }
}
```

**Notes:**  
- Use a map to track parents since tree nodes do not have parent pointers.

**Algorithm:**

**Solution Code:**
```java
class Solution {
    public TreeNode insertIntoBST(TreeNode root, int val) {
        // If the root is null, create a new node with the given value
        if (root == null) return new TreeNode(val);

        TreeNode current = root;

        // Traverse the tree to find the correct spot for insertion
        while (true) {
            if (val < current.val) {
                // Move to the left subtree
                if (current.left == null) {
                    current.left = new TreeNode(val);
                    break;
                } else {
                    current = current.left;
                }
            } else { // val > current.val (since val is guaranteed to be unique)
                // Move to the right subtree
                if (current.right == null) {
                    current.right = new TreeNode(val);
                    break;
                } else {
                    current = current.right;
                }
            }
        }

        return root; // Return the unchanged root
    }
}
```

Find the kth largest value in a BST using reverse inorder traversal.

**Approach**
To find the **kth largest element** in a BST, we can perform a **reverse inorder traversal**. The reverse inorder traversal of a BST visits nodes in descending order (right subtree, root, left subtree). By keeping track of the number of nodes visited, we can easily return the **kth largest** element.

**Iterative Approach**:
   - Use a **stack** to simulate the recursion of an inorder traversal.
   - Traverse the tree in reverse (right, root, left) until we visit the **kth largest element**.

**🧮 Flowchart/ Diagram:**  
- Start from root  
- Traverse right → node → left  
- Count nodes visited  
- Stop when kth node is visited  

**Pseudocode/Algorithm:**  
```sql
function kthLargest(root, k):
stack = empty stack
curr = root
count = 0

while curr is not null or stack is not empty:
    while curr is not null:
        stack.push(curr)
        curr = curr.right

    curr = stack.pop()
    count += 1

    if count == k:
        return curr.data

    curr = curr.left

return -1
```

**Java Code**
```java
class Solution {
    // Function to return the Kth largest element in the given BST rooted at 'root'
    public int kthLargest(Node root, int k) {
        // Stack to simulate the inorder traversal
        Stack<Node> st = new Stack<>();
        Node curr = root;
        int count = 0;

        // Iterative reverse inorder traversal
        while (curr != null || !st.isEmpty()) {
            // Traverse to the rightmost node
            while (curr != null) {
                st.push(curr);
                curr = curr.right;
            }

            // Pop from the stack (current node is the next largest element)
            curr = st.pop();
            count++;  // Increment the counter for the number of nodes visited
            
            // If count reaches k, we have found the kth largest element
            if (count == k) {
                return curr.data;
            }

            // Move to the left subtree
            curr = curr.left;
        }

        return -1;  // In case k is out of bounds (not enough elements)
    }
}
```

   - **O(h)** for the stack, where `h` is the height of the tree.

- Reverse inorder traversal ensures descending order for BST.  
- Efficient without altering the original tree.

**Goal:**  
Check whether a given binary tree satisfies the properties of a Binary Search Tree (BST) — i.e., for every node, all values in its left subtree are less than the node’s value, and all values in the right subtree are greater.

**Approach**
To check if a binary tree is a **Binary Search Tree (BST)**, we need to validate the BST property recursively for every node in the tree. The key condition for a node to be part of a BST is that:
1. The left child must have a value smaller than the node.
2. The right child must have a value greater than the node.

We can use a helper function that checks if each node's value lies within a valid range (bounded by `min` and `max` values):

**Java Code**
```java
class Solution {
    // Helper function to check the BST property
    static boolean isBSTUtil(Node node, int min, int max) {
        // If the node is null, it is a valid BST
        if (node == null) return true;
        
        // If the node's value is out of the valid range, return false
        if (node.data < min || node.data > max) return false;
        
        // Recursively check the left and right subtrees
        return isBSTUtil(node.left, min, node.data - 1) &&
               isBSTUtil(node.right, node.data + 1, max);
    }

    // Function to check if the tree is a BST
    boolean isBST(Node root) {
        // Call the helper function with initial range
        return isBSTUtil(root, Integer.MIN_VALUE, Integer.MAX_VALUE);
    }
}
```

**Code:**
```java
// Java program to check if a tree is 
// BST using Morris Traversal

class Node {
	int data;
	Node left, right;

	Node(int value) {
        data = value;
        left = right = null;
	}
}

class GfG {

    // Function to check if the binary tree 
  	// is a BST using Morris Traversal
    static boolean isBST(Node root) {
        Node curr = root;
        Node pre;
        int prevValue = Integer.MIN_VALUE;

        while (curr != null) {
            if (curr.left == null) {
              
                // Process curr node
                if (curr.data <= prevValue) {
                  
                    // Not in ascending order
                    return false;
                }
                prevValue = curr.data;
                curr = curr.right;
            } else {
              
                // Find the inorder predecessor of curr
                pre = curr.left;
                while (pre.right != null && pre.right != curr) {
                    pre = pre.right;
                }

                if (pre.right == null) {
                  
                    // Create a temporary thread to the curr node
                    pre.right = curr;
                    curr = curr.left;
                } else {
                  
                    // Remove the temporary thread
                    pre.right = null;

                    // Process the curr node
                    if (curr.data <= prevValue) {
                      
                        // Not in ascending order
                        return false;
                    }
                    prevValue = curr.data;
                    curr = curr.right;
                }
            }
        }

        return true;
    }

    public static void main(String[] args) {
      
        // Create a sample binary tree
        //     10
        //    /  \
        //   5    20
        //        / \
        //       9   25
        Node root = new Node(10);
        root.left = new Node(5);
        root.right = new Node(20);
        root.right.left = new Node(9);
        root.right.right = new Node(25);
      
      	if (isBST(root)) {
            System.out.println("True");
        }
        else {
            System.out.println("False");
        }
    }
}
```

2   7    20


**k = 12**

**Output:**

**Constraints**

- `-10^4 ≤ Node value ≤ 10^4`
- `-10^4 ≤ k ≤ 10^4`

**Goal:**  
Find the minimum absolute difference between the given integer `k` and any node value in the BST.

**Approach**
To find the closest value to the given integer `k` in the BST, we can leverage the BST properties. A Binary Search Tree has the following properties:
1. Left child values are smaller than the parent node's value.
2. Right child values are greater than the parent node's value.

Using this property, we can perform a **binary search** starting from the root:
1. For each node, we calculate the absolute difference between the node's value and `k`.
2. If the current node's value is closer to `k` than the previously found closest value, we update the closest value.
3. If the current node's value is greater than `k`, we move to the left child.
4. If the current node's value is less than `k`, we move to the right child.
5. If the current node's value is equal to `k`, we return `0` as the absolute difference is `0`.

**🧮 Flowchart/ Diagram:**  

```
     10
    /  \
   5    15
  / \     \
 2   7    20
```

Path Traversed: 10 -> 15
Closest value = 10 → |10 - 12| = 2


**Pseudocode/Algorithm:**  
```sql

function minDiff(root, k):
closest = root.data
while root is not null:
if abs(root.data - k) < abs(closest - k):
closest = root.data
if root.data == k:
return 0
else if root.data > k:
root = root.left
else:
root = root.right
return abs(closest - k)

```

**Java Code**
```java
class Solution {
    // Function to find the least absolute difference between any node
    // value of the BST and the given integer.
    static int minDiff(Node root, int k) {
        // Initialize the closest value to the root's data
        int closest = root.data;

        // Perform a binary search in the BST
        while (root != null) {
            // Update the closest value if current node is closer to k
            if (Math.abs(root.data - k) < Math.abs(closest - k)) {
                closest = root.data;
            }

            // Move to the left or right child based on the comparison with k
            if (root.data > k) {
                root = root.left;
            } else if (root.data < k) {
                root = root.right;
            } else {
                // If current node's value is equal to k, return 0 as the closest
                return 0;
            }
        }

        // Return the absolute difference between closest value and k
        return Math.abs(closest - k);
    }
}
```

  - **O(1)**, since we are only using a constant amount of space (ignoring recursion stack space for simplicity).

- The closest value found does not necessarily have to be a leaf.

## Problem Statement: 338
**Largest BST**

Given a binary tree, your task is to find the size of the largest subtree within this binary tree that also satisfies the properties of a **Binary Search Tree (BST)**. The size of a subtree is defined as the number of nodes it contains.

**Constraints**
- A subtree is considered a **BST** if:

  2. The right child is greater than the node.
  3. There are no duplicate values in the subtree.

**Example**
**Input:**
    10
   /  \
  5    15
 / \     \
1   8     7


**Output:**
3

**Explanation:**
- The largest BST subtree is the one rooted at node `5` with nodes `5, 1, 8`, which is a valid BST.
- The size of this subtree is `3`.

**Goal:**  
Find the size (number of nodes) of the largest subtree in a binary tree that is also a valid Binary Search Tree (BST).

**Approach**
To solve this problem, we need to:
1. **Recursively traverse** the tree to check whether each subtree satisfies the properties of a Binary Search Tree (BST).
2. For each node, compute:
   - The size of the subtree.
   - The minimum and maximum values in the subtree.
   - Whether the subtree rooted at that node is a BST.
3. Maintain the maximum size of the valid BST subtree found during the traversal.

**Key Considerations**
- A **Binary Search Tree (BST)** must have:
  - All nodes in the left subtree are less than the root node.
  - All nodes in the right subtree are greater than the root node.
- We need to handle cases where the tree is unbalanced or contains nodes that break the BST properties.

**🧮 Flowchart/ Diagram:**  

```
    10
   /  \
  5    15
 / \     \
1   8     7
```

Largest BST Subtree:
5
/&#x20;
1   8

Size = 3

**Pseudocode/Algorithm:**  
```sql

function helper(node):
if node is null:
return Info(size=0, min=+∞, max=-∞, maxSize=0, isBST=true)

left = helper(node.left)
right = helper(node.right)

curr.size = 1 + left.size + right.size

if left.isBST and right.isBST and node.data > left.max and node.data < right.min:
    curr.min = min(node.data, left.min)
    curr.max = max(node.data, right.max)
    curr.maxSize = curr.size
    curr.isBST = true
else:
    curr.maxSize = max(left.maxSize, right.maxSize)
    curr.isBST = false

return curr

function largestBst(root):
return helper(root).maxSize

```

**Java Code**
```java
class Solution {
    // Helper class to store information about the tree
    static class Info {
        int size;        // Size of the subtree
        int min;         // Minimum value in the subtree
        int max;         // Maximum value in the subtree
        int maxSize;     // Size of the largest BST subtree found
        boolean isBST;   // True if the subtree is a BST

        // Constructor to initialize the values
        Info(int size, int min, int max, int maxSize, boolean isBST) {
            this.size = size;
            this.min = min;
            this.max = max;
            this.maxSize = maxSize;
            this.isBST = isBST;
        }
    }

    // Helper function to traverse the tree and return the information for each subtree
    static Info helper(Node node) {
        // Base case: If the node is null, return a valid BST with size 0
        if (node == null) {
            return new Info(0, Integer.MAX_VALUE, Integer.MIN_VALUE, 0, true);
        }

        // Recursively check the left and right subtrees
        Info left = helper(node.left);
        Info right = helper(node.right);

        // Create a current Info object for the current node
        Info curr = new Info(0, 0, 0, 0, false);
        curr.size = 1 + left.size + right.size;  // Count the current node and its subtrees

        // Check if the current subtree is a valid BST
        if (left.isBST && right.isBST && node.data > left.max && node.data < right.min) {
            curr.min = Math.min(node.data, left.min);  // Update the minimum value
            curr.max = Math.max(node.data, right.max); // Update the maximum value
            curr.maxSize = curr.size;  // Update the max size of the BST
            curr.isBST = true;         // Mark the subtree as a BST
        } else {
            // If it's not a BST, take the max size of the left or right subtree
            curr.maxSize = Math.max(left.maxSize, right.maxSize);
            curr.isBST = false;  // Mark the subtree as not a BST
        }

        // Return the Info object for the current node
        return curr;
    }

    // Function to return the size of the largest BST subtree
    static int largestBst(Node root) {
        // Call the helper function and return the max size of the BST
        return helper(root).maxSize;
    }
}
```

**Time and Space Complexity**
- **Time Complexity**:  
  - **O(n)** where `n` is the number of nodes in the tree, because each node is visited exactly once during the traversal.
  
- **Space Complexity**:  
  - **O(h)** where `h` is the height of the tree, due to the recursive call stack. In the worst case (unbalanced tree), `h = n`.

**Notes:**  
- The `Info` class encapsulates all necessary metadata to check BST conditions at each node.
- This method ensures each node is only visited once, providing an efficient bottom-up approach.

- **Lowest Common Ancestor (LCA)** of two nodes `n1` and `n2` is the lowest node in the BST that has both `n1` and `n2` as descendants. We allow a node to be a descendant of itself.

**Example**
**Input:**
    20
   /  \
 10    30
/  \   /  \
5  15  25  35


**Output:**
20

**Explanation:**
- The Lowest Common Ancestor (LCA) of nodes `5` and `15` is `10`.
- The LCA of nodes `25` and `35` is `30`.
- The LCA of nodes `5` and `35` is `20`.

**Goal:**  
Determine the lowest node in a Binary Search Tree (BST) that is an ancestor to both `n1` and `n2`.

**Approach**
To find the LCA of two nodes in a Binary Search Tree (BST), we can use the following property:
- If both `n1` and `n2` are smaller than the current node's value, then the LCA must lie in the left subtree.
- If both `n1` and `n2` are larger than the current node's value, then the LCA must lie in the right subtree.
- If one node is smaller and the other is larger, then the current node is the LCA.

We start from the root of the tree and traverse down. The first node where the paths of `n1` and `n2` diverge is the LCA.

**🧮 Flowchart/ Diagram:**

```
    20
   /  \
 10    30
/  \   / \
5  15 25  35

LCA(5, 15) = 10
LCA(25, 35) = 30
LCA(5, 35) = 20

```

**Pseudocode/Algorithm:**  
```sql

function LCA(root, n1, n2):
while root is not null:
if n1 < root.data and n2 < root.data:
root = root.left
else if n1 > root.data and n2 > root.data:
root = root.right
else:
return root

```

**Java Code**
```java
class Solution {
    // Function to find the Lowest Common Ancestor (LCA)
    Node LCA(Node root, Node n1, Node n2) {
        // Traverse the tree to find the LCA
        while (root != null) {
            // If both n1 and n2 are smaller, LCA lies in the left subtree
            if (root.data > n1.data && root.data > n2.data)
                root = root.left;
            // If both n1 and n2 are larger, LCA lies in the right subtree
            else if (root.data < n1.data && root.data < n2.data)
                root = root.right;
            // If one of n1 or n2 matches root, root is the LCA
            else
                break;
        }

        // Return the LCA node
        return root;
    }
}
```

  - **O(1)** since we are only using a constant amount of space for the traversal.

**Java Code**
```java
class Solution {
    // Function to return a list of integers denoting the node
    // values of both the BST in a sorted order.
    public ArrayList<Integer> merge(Node root1, Node root2) {
        // Initialize two lists to hold the inorder traversal of both trees
        ArrayList<Integer> list1 = new ArrayList<>();
        ArrayList<Integer> list2 = new ArrayList<>();

        // Perform inorder traversal on both BSTs
        inorder(root1, list1);
        inorder(root2, list2);

        // Merge both sorted lists and return the result
        return mergeSortedLists(list1, list2);
    }

    // Helper function to perform inorder traversal and store the result in the list
    private void inorder(Node root, ArrayList<Integer> list) {
        if (root == null) return;
        inorder(root.left, list);  // Traverse left subtree
        list.add(root.data);       // Add the node data to the list
        inorder(root.right, list); // Traverse right subtree
    }

    // Helper function to merge two sorted lists
    private ArrayList<Integer> mergeSortedLists(ArrayList<Integer> list1, ArrayList<Integer> list2) {
        ArrayList<Integer> merged = new ArrayList<>();
        int i = 0, j = 0;

        // Merge both lists until one of them is exhausted
        while (i < list1.size() && j < list2.size()) {
            if (list1.get(i) < list2.get(j)) {
                merged.add(list1.get(i++));
            } else {
                merged.add(list2.get(j++));
            }
        }

        // Add remaining elements from list1
        while (i < list1.size()) {
            merged.add(list1.get(i++));
        }

        // Add remaining elements from list2
        while (j < list2.size()) {
            merged.add(list2.get(j++));
        }

        return merged;
    }
}
```

- **Space Complexity:**  
  - **O(N + M)** for the storage of the two sorted arrays and the merged array.

**Notes:**  
- Inorder traversal guarantees sorted output for each BST.
- The merging step is linear due to the pre-sorted arrays.

## Problem Statement: 341
**Predecessor and Successor**

Given a Binary Search Tree (BST) and an integer key, you need to find the **in-order predecessor** and **in-order successor** of the given key. If either predecessor or successor is not found, return `NULL` for that.

- **Predecessor** of a node is the largest value node that is smaller than the given key.
- **Successor** of a node is the smallest value node that is greater than the given key.

**Example**
**Input:**
   50
  /  \
30    70
/ \ /
20 40 60 80

Key = 65

**Output:**
Predecessor = 60 Successor = 70

**Goal:**  
Find the in-order predecessor and successor of a given key in a Binary Search Tree (BST).

**Approach**
1. **Inorder Traversal:**  
   - In a Binary Search Tree (BST), the inorder traversal produces a sorted sequence of nodes.
   - The predecessor is the largest node that is smaller than the key, and the successor is the smallest node that is greater than the key.

2. **Finding Predecessor and Successor:**
   - Start from the root of the BST and traverse down:
     - If the current node's value is less than the given key, move to the right and update the predecessor.
     - If the current node's value is greater than the given key, move to the left and update the successor.
     - If the current node's value equals the given key, find the leftmost node in its right subtree (successor) and the rightmost node in its left subtree (predecessor).

3. **Edge Cases:**
   - If the key is the smallest in the BST, the predecessor is `NULL`.
   - If the key is the largest in the BST, the successor is `NULL`.

**🧮 Flowchart/ Diagram:**

```
    50
   /  \
 30    70
/ \    / \
```

20  40  60  80

Key = 65

Step 1: Start at root (50)
Step 2: 65 > 50 → move right, predecessor = 50
Step 3: 65 < 70 → move left, successor = 70
Step 4: 65 > 60 → move right, predecessor = 60
Step 5: Node becomes null → stop
Result: Predecessor = 60, Successor = 70

**Pseudocode/Algorithm:**

1. Initialize predecessor and successor as NULL.

2. Start from the root node:
   - While current node is not NULL:
     - If current node's value is less than key:
       - Update predecessor to current node.
       - Move to right subtree.
     - Else if current node's value is greater than key:
       - Update successor to current node.
       - Move to left subtree.
     - Else (current node's value equals key):
       - If left subtree exists:
         - Set predecessor to the rightmost node in the left subtree.
       - If right subtree exists:
         - Set successor to the leftmost node in the right subtree.
       - Break the loop.

3. Return predecessor and successor.

**Java Code**
```java
class Solution {
    public static void findPreSuc(Node root, Node[] pre, Node[] suc, int key) {
        // Start from the root of the BST
        Node curr = root;

        // Traverse the tree and find the predecessor and successor
        while (curr != null) {
            if (curr.data < key) {
                // Update predecessor and move to the right subtree
                pre[0] = curr;
                curr = curr.right;
            }
            else if (curr.data > key) {
                // Update successor and move to the left subtree
                suc[0] = curr;
                curr = curr.left;
            }
            else {
                // If the current node matches the key
                // Find the rightmost node in the left subtree (predecessor)
                if (curr.left != null)
                    pre[0] = rightMost(curr.left);
                
                // Find the leftmost node in the right subtree (successor)
                if (curr.right != null)
                    suc[0] = leftMost(curr.right);
                break;
            }
        }
    }

    // Helper function to find the rightmost node in a given subtree
    static Node rightMost(Node node) {
        while (node.right != null) {
            node = node.right;
        }
        return node;
    }

    // Helper function to find the leftmost node in a given subtree
    static Node leftMost(Node node) {
        while (node.left != null) {
            node = node.left;
        }
        return node;
    }
}
```

**Time and Space Complexity**
- **Time Complexity:**
  - **O(H)**, where `H` is the height of the BST, because we are traversing the tree from the root to the leaf nodes (in the worst case).
  
- **Space Complexity:**
  - **O(1)**, since we are only using a constant amount of space for the predecessor and successor nodes.

**Notes:**  
- The predecessor and successor must be actual nodes in the BST.
- The solution utilizes BST properties to minimize traversal.
- Use mutable references (like arrays in Java) to update values from recursive or iterative logic.

```java
class Solution {
    public static void transformTree(Node root) {
        // Initialize the sum array to store the running sum
        int[] sum = {0}; 
        // Start the transformation process
        transformToGreaterSumTree(root, sum);
    }

    static void transformToGreaterSumTree(Node root, int[] sum) {
        // Base case: If the node is null, return
        if (root == null) {
            return;
        }
        
        // Recurse on the right subtree (greater values first)
        transformToGreaterSumTree(root.right, sum);
        
        // Update the current node's value to the sum of nodes greater than it
        sum[0] += root.data;
        root.data = sum[0] - root.data;
        
        // Recurse on the left subtree
        transformToGreaterSumTree(root.left, sum);
    }
}
```

   - If the key matches the current node’s value, then proceed with deletion.

2. **Deletion Cases**:
   - **Case 1 (No child)**: Simply return `null` as the node has no children.
   - **Case 2 (One child)**: Return the non-null child node to replace the current node.
   - **Case 3 (Two children)**: Find the in-order successor (the smallest node in the right subtree), copy its value to the current node, and recursively delete the successor node.

**🧮 Flowchart/ Diagram:**  
- Start at root  
- Compare key with current node value  
- If key < node value, go left  
- If key > node value, go right  
- If key == node value, handle deletion cases  
- Return updated root after deletion

**Pseudocode/Algorithm:**  
```sql

function deleteNode(root, key):
if root is null:
return null

if key < root.val:
    root.left = deleteNode(root.left, key)
else if key > root.val:
    root.right = deleteNode(root.right, key)
else:
    if root.left is null and root.right is null:
        return null
    else if root.left is null:
        return root.right
    else if root.right is null:
        return root.left
    else:
        successor = findMin(root.right)
        root.val = successor.val
        root.right = deleteNode(root.right, successor.val)

return root

function findMin(node):
while node.left is not null:
node = node.left
return node

```

**Solution Code:**
```java
class Solution {
    public TreeNode deleteNode(TreeNode root, int key) {
        // If the root is null, return null
        if (root == null) return null;

        // Step 1: Search for the node
        if (key < root.val) {
            root.left = deleteNode(root.left, key);
        } else if (key > root.val) {
            root.right = deleteNode(root.right, key);
        } else {
            // Step 2: Node found
            // Case 1: Node has no children
            if (root.left == null && root.right == null) {
                return null;
            }
            // Case 2: Node has one child
            else if (root.left == null) {
                return root.right;
            } else if (root.right == null) {
                return root.left;
            }
            // Case 3: Node has two children
            else {
                // Find the in-order successor (smallest node in the right subtree)
                TreeNode successor = findMin(root.right);
                // Copy the value of the successor to the current node
                root.val = successor.val;
                // Delete the successor node
                root.right = deleteNode(root.right, successor.val);
            }
        }

        // Return the root after deletion
        return root;
    }

    // Helper function to find the minimum node in a subtree
    private TreeNode findMin(TreeNode node) {
        while (node.left != null)
            node = node.left;
        return node;
    }
}
```

**Time Complexity:**
O(H), where H is the height of the tree. In the worst case, we may need to traverse the entire height of the tree to find and delete the node. This is typical for binary search trees.

**Space Complexity:**
O(H), where H is the height of the tree. This is due to the space used by the recursion stack. In the worst case, the tree could be skewed, resulting in a height of O(N).

**Notes:**  
- Deletion in BST is a fundamental operation and involves careful handling of three cases to maintain BST properties.  
- Using the in-order successor to replace a node with two children ensures the tree remains a valid BST.  
- Alternatively, the in-order predecessor (largest node in the left subtree) can also be used instead of the successor.  
- The recursive approach simplifies the implementation but uses stack space proportional to the height of the tree.  
- For balanced BSTs, time and space complexity are O(log n), but for skewed trees, they degrade to O(n).  
- It is important to update parent references correctly during deletion to avoid losing parts of the tree.

```sql

Start with preorder list pre\[low\..high]

If low > high: return null

Create root with pre\[low]

Find index i where pre\[i] > root.data (first greater element)

Recursively build:
root.left = constructUtil(pre, low+1, i-1)
root.right = constructUtil(pre, i, high)

Return root

```

**Pseudocode / Algorithm:**

```sql

function constructUtil(pre, low, high):
if low > high:
return null

root = new Node(pre[low])

if low == high:
    return root

i = low + 1
while i <= high and pre[i] <= root.data:
    i++

root.left = constructUtil(pre, low + 1, i - 1)
root.right = constructUtil(pre, i, high)

return root

function construct(pre):
return constructUtil(pre, 0, pre.length - 1)

```

**Code:**

```java
import java.util.*;

class Node {
    int data;
    Node left, right;

    public Node(int x) {
        data = x;
        left = right = null;
    }
}

class GfG {

    // Recursive function to construct BST from preorder
    public static Node constructUtil(List<Integer> pre, int low, int high) {
        if (low > high)
            return null;

        Node root = new Node(pre.get(low));

        if (low == high)
            return root;

        int i;
        for (i = low + 1; i <= high; i++) {
            if (pre.get(i) > root.data)
                break;
        }

        root.left = constructUtil(pre, low + 1, i - 1);
        root.right = constructUtil(pre, i, high);

        return root;
    }

    // Wrapper function
    public static Node construct(List<Integer> pre) {
        return constructUtil(pre, 0, pre.size() - 1);
    }

    // Utility to print inorder traversal of BST
    public static void printInorder(Node root) {
        if (root == null)
            return;
        printInorder(root.left);
        System.out.print(root.data + " ");
        printInorder(root.right);
    }

    // Driver code
    public static void main(String[] args) {
        List<Integer> pre = Arrays.asList(10, 5, 1, 7, 40, 50);
        Node root = construct(pre);
        printInorder(root);
    }
}
```

   - Use recursion to create nodes and their parents. When creating a node, if the parent is not yet created, the function will recursively create the parent first.

4. **Tree Construction:**

**Solution Code:**
```java
class Solution {
    // Function to construct binary tree from parent array.
    public Node createTree(int parent[]) {
        int n = parent.length;
        Node[] created = new Node[n];  // Array to store created nodes
        Node root = null;

        // Iterate over the parent array to create nodes
        for (int i = 0; i < n; i++) {
            createNode(parent, i, created);

            // If parent[i] is -1, this node is the root
            if (parent[i] == -1) {
                root = created[i];
            }
        }

        return root;
    }

    // Helper function to create a node and attach it to its parent
    private void createNode(int[] parent, int i, Node[] created) {
        if (created[i] != null) return;  // If the node is already created, return

        // Create current node
        created[i] = new Node(i);

        if (parent[i] == -1) return;  // If it's the root, no need to attach to parent

        // Create the parent node if not already created
        if (created[parent[i]] == null) {
            createNode(parent, parent[i], created);
        }

        // Attach the current node to its parent
        Node p = created[parent[i]];
        if (p.left == null) {
            p.left = created[i];  // Attach as left child
        } else {
            p.right = created[i];  // Attach as right child
        }
    }
}
```

```java
class Tree {
    // Function to serialize a tree and return a list containing nodes of tree.
    public ArrayList<Integer> serialize(Node root) {
        ArrayList<Integer> arr = new ArrayList<>(); // ArrayList to store serialized tree nodes

        // Initialize queue for level-order traversal (BFS)
        Queue<Node> q = new LinkedList<>();
        q.add(root);

        while (!q.isEmpty()) {
            Node curr = q.poll();
            if (curr == null) {
                arr.add(-1);  // -1 represents a null node
                continue;
            }
            arr.add(curr.data);  // Add current node's data to the array
            q.add(curr.left);    // Add left child to the queue
            q.add(curr.right);   // Add right child to the queue
        }

        return arr;  // Return the serialized array
    }

    // Function to deserialize a list and construct the tree.
    public Node deSerialize(ArrayList<Integer> arr) {
        if (arr.get(0) == -1) return null;  // If the first element is -1, the tree is empty

        // Create the root node using the first element of the array
        Node root = new Node(arr.get(0));
        Queue<Node> q = new LinkedList<>();
        q.add(root);

        int i = 1;  // Start from the second element in the array
        while (!q.isEmpty() && i < arr.size()) {
            Node curr = q.poll();

            // If the next element is not -1, create a left child and add it to the queue
            if (arr.get(i) != -1) {
                Node left = new Node(arr.get(i));
                curr.left = left;
                q.add(left);
            }
            i++;

            // If the next element is not -1, create a right child and add it to the queue
            if (i < arr.size() && arr.get(i) != -1) {
                Node right = new Node(arr.get(i));
                curr.right = right;
                q.add(right);
            }
            i++;
        }

        return root;  // Return the root node of the deserialized tree
    }
}
```

**Time Complexity:**
**serialize(): O(N)**, where N is the number of nodes in the tree. Each node is processed once during the level-order traversal.
**deSerialize(): O(N),** where N is the number of nodes in the tree. The array is processed once, and for each element, a node is either created or skipped.

**Space Complexity:**
**serialize(): O(N)** for the array used to store the serialized tree and the queue used for level-order traversal.
**deSerialize(): O(N)** for the array used to store the serialized tree and the queue used for tree reconstruction.

**Notes:**
- Using `-1` to represent null nodes ensures the tree structure, including missing children, is preserved during serialization and deserialization.
- Level-order traversal helps serialize the tree in a way that supports straightforward reconstruction.
- This method handles trees with duplicate values correctly since the structure is maintained via position, not just values.

**Largest Subtree Sum in a Tree**

Given a binary tree, the task is to find the subtree with the maximum sum and return its sum.

**Example:**
**Input:**
     1
    / \
   2   3
  / \   \
 4   5   6


**Output:**

**Example:**  
Refer to the given example in the problem statement.

**Input:**  
A binary tree with nodes and their integer values.

**Output:**  
An integer representing the largest subtree sum.

**Goal:**  
Identify and return the maximum sum achievable by any subtree within the binary tree.

**Approach:**
To solve this problem, we need to find the sum of all subtrees and keep track of the largest subtree sum encountered. A **subtree sum** for a node is defined as the sum of its value, the sum of its left subtree, and the sum of its right subtree.

**Steps:**
1. **Level-Order Traversal (BFS):**
   - Traverse the tree level by level to process each node.
   - Use a queue to perform the level-order traversal.

2. **Calculate Subtree Sum:**
   - For each node, calculate its subtree sum by adding its value to the sums of its left and right subtrees.
   - Store these subtree sums in a hash map where the key is the node and the value is the subtree sum.

3. **Track the Maximum Subtree Sum:**
   - As we calculate the subtree sum for each node, update the maximum sum found so far.

4. **Post-Order Traversal for Subtree Sum Calculation:**
   - Process the tree in a bottom-up manner (post-order traversal). This ensures that we compute the sum of the subtrees before the sum of their parent nodes.

**🧮 Flowchart/ Diagram:**  
- Start at the root node.  
- Traverse the tree level-wise and store nodes by levels.  
- Calculate subtree sums from bottom levels up to the root.  
- Track and update maximum subtree sum.

**Pseudocode/Algorithm:**  
```sql

Initialize ans = -∞
Perform level order traversal, store nodes by levels
For each level from bottom to top:
For each node in level:
subtreeSum\[node] = node.data + subtreeSum\[left child] + subtreeSum\[right child]
Update ans if subtreeSum\[node] > ans
Return ans

```

**Solution Code:**
```java
class Solution {
    public static int findLargestSubtreeSum(Node root) {
        // Handle the edge case of an empty tree
        if (root == null)
            return 0;

        int ans = Integer.MIN_VALUE;  // Variable to store the largest sum
        Queue<Node> q = new LinkedList<>();
        List<List<Node>> levels = new ArrayList<>();
        HashMap<Node, Integer> subtreeSum = new HashMap<>();

        q.add(root);

        // Perform a level-order traversal to collect nodes by levels
        while (!q.isEmpty()) {
            int n = q.size();
            List<Node> level = new ArrayList<>();
            while (n-- > 0) {
                Node node = q.poll();
                level.add(node);
                if (node.left != null)
                    q.add(node.left);  // Add left child to the queue
                if (node.right != null)
                    q.add(node.right); // Add right child to the queue
            }
            levels.add(level);  // Add the current level to the levels list
        }

        // Traverse the tree from bottom-up (reverse level order) to calculate subtree sums
        for (int i = levels.size() - 1; i >= 0; i--) {
            for (Node e : levels.get(i)) {
                subtreeSum.put(e, e.data);  // Initialize the subtree sum with the node's data
                if (e.left != null)
                    subtreeSum.put(e, subtreeSum.get(e) + subtreeSum.get(e.left));  // Add left subtree sum
                if (e.right != null)
                    subtreeSum.put(e, subtreeSum.get(e) + subtreeSum.get(e.right)); // Add right subtree sum
                ans = Math.max(ans, subtreeSum.get(e));  // Update the maximum subtree sum
            }
        }

        return ans;  // Return the largest subtree sum
    }
}
```

**Time Complexity:**
O(N), where N is the number of nodes in the tree. Each node is processed once during the level-order traversal, and the subtree sum for each node is computed once.

**Space Complexity:**
O(N), where N is the number of nodes. The space complexity is determined by the space used to store the nodes in the queue, the list of levels, and the hash map for subtree sums.

**Notes:**
- The problem leverages post-order traversal logic by processing nodes from the bottom up to ensure subtree sums are accurately computed before their parents.
- Using a hash map to store subtree sums avoids recalculating sums for nodes multiple times, optimizing performance.
- This approach works efficiently for any binary tree, including those with negative node values, by tracking the maximum sum encountered.

**Maximum Sum of Non-Adjacent Nodes**

Given a binary tree where each node has a value associated with it, your task is to select a subset of nodes such that the sum of their values is maximized. The condition is that no two selected nodes are directly connected, meaning that if a node is included in the subset, neither its parent nor its children can be included.

**Example:**
**Input:**
     5
    / \
   1   8
  / \   / \
 7   2  3  4


**Output:**
18

**Explanation:**
- The maximum sum of non-adjacent nodes is `5 + 8 + 7 = 18`.

**Constraints:**  
- The tree nodes contain positive integer values.  
- Tree size can vary (N nodes).  

**Goal:**  
Maximize the sum of node values by selecting non-adjacent nodes in the binary tree.

**Approach:**
We need to select non-adjacent nodes from the tree such that their sum is maximized. The key observation here is to use **dynamic programming with memoization** to efficiently compute the maximum sum for each node while keeping track of the maximum sum for its children nodes.

**Steps:**
1. **Recursive Approach with Memoization:**
   - For each node, we have two options:
     1. **Include the node**: If we include the node, we cannot include its immediate children (left and right). Thus, the maximum sum for this case is the value of the node itself plus the sum of non-adjacent nodes from the left and right children (i.e., grand-children).
     2. **Exclude the node**: If we exclude the node, we can include its left and right children, and the sum will be the sum of the maximum sums of the left and right subtrees.

2. **Memoization**:
   - Use a memoization map (`memo`) to store the results of already computed nodes to avoid redundant calculations and optimize the solution.

3. **Base Case**:
   - If the node is `null`, the sum is `0`.

4. **Recursive Formula**:
   - For a node `node`:
     - **Include**: `node.data + getMaxSumUtil(node.left.left) + getMaxSumUtil(node.left.right) + getMaxSumUtil(node.right.left) + getMaxSumUtil(node.right.right)`
     - **Exclude**: `getMaxSumUtil(node.left) + getMaxSumUtil(node.right)`

   Finally, return the maximum of the two options (including or excluding the node).

**🧮 Flowchart/ Diagram:**  
- Start at root  
- Recursively explore each node:  
  - Calculate sum including the node (skip children, include grandchildren)  
  - Calculate sum excluding the node (include children)  
- Use memoization to save intermediate results  
- Return max of including or excluding node  

**Pseudocode/Algorithm:**  
```sql

function getMaxSumUtil(node, memo):
if node is null:
return 0
if node in memo:
return memo\[node]

incl = node.data
if node.left:
    incl += getMaxSumUtil(node.left.left, memo) + getMaxSumUtil(node.left.right, memo)
if node.right:
    incl += getMaxSumUtil(node.right.left, memo) + getMaxSumUtil(node.right.right, memo)

excl = getMaxSumUtil(node.left, memo) + getMaxSumUtil(node.right, memo)

memo[node] = max(incl, excl)
return memo[node]

function getMaxSum(root):
memo = empty map
return getMaxSumUtil(root, memo)

```

**Solution Code:**
```java
class Solution {
    // Helper function for the recursive calculation of the maximum sum.
    static int getMaxSumUtil(Node node, HashMap<Node, Integer> memo) {
        if (node == null) {
            return 0;  // Base case: If the node is null, return 0.
        }
        if (memo.containsKey(node)) {
            return memo.get(node);  // Return the stored value if already computed.
        }

        // Calculate the maximum sum including the current node
        int incl = node.data;
        if (node.left != null) {
            incl += getMaxSumUtil(node.left.left, memo) + 
                    getMaxSumUtil(node.left.right, memo);  // Include grandchildren
        }
        if (node.right != null) {
            incl += getMaxSumUtil(node.right.left, memo) + 
                    getMaxSumUtil(node.right.right, memo);  // Include grandchildren
        }

        // Calculate the maximum sum excluding the current node
        int excl = getMaxSumUtil(node.left, memo) + 
                   getMaxSumUtil(node.right, memo);  // Include the children without the current node

        // Store the maximum of including or excluding the node
        memo.put(node, Math.max(incl, excl));

        return memo.get(node);
    }

    // Main function to get the maximum sum of non-adjacent nodes in the binary tree
    public int getMaxSum(Node root) {
        HashMap<Node, Integer> memo = new HashMap<>();
        return getMaxSumUtil(root, memo);  // Call the helper function to calculate the result
    }
}
```

**Time Complexity:**
O(N), where N is the number of nodes in the tree. Each node is visited once and the results are cached in the memoization map, ensuring that no node is processed more than once.

**Space Complexity:**
O(N), where N is the number of nodes in the tree. The space complexity is due to the memoization map and the recursion stack.

**Notes:**
- Use dynamic programming with memoization to avoid recomputing results for nodes.  
- For each node, compute two sums: including the node (skip children, include grandchildren) and excluding the node (include children).  
- The answer is the maximum of these two sums, ensuring no two adjacent nodes are selected.

**Solution Code:**
```java
class Solution {
    // Helper function to serialize the tree and detect duplicates
    static int serializeTree(Node root, Map<Integer, Integer> map, 
                             Map<String, Integer> strToInt, List<Node> ans, int[] idNum) {
        if (root == null) return 0;  // Return 0 for null nodes
        
        // Recursively serialize the left and right subtrees
        int left = serializeTree(root.left, map, strToInt, ans, idNum);
        int right = serializeTree(root.right, map, strToInt, ans, idNum);
        
        // Generate the string representation of the current subtree
        String val = Integer.toString(root.data);
        String curr = "(" + left + ")" + val + "(" + right + ")";
        
        // If the subtree is already serialized, get its ID, otherwise generate a new ID
        int currId; 
        if (strToInt.containsKey(curr)) {
            currId = strToInt.get(curr);
        } else {
            currId = idNum[0]++;
            strToInt.put(curr, currId);  // Map the serialized string to a unique ID
        }
        
        // Update the occurrence count of the current serialized subtree
        map.put(currId, map.getOrDefault(currId, 0) + 1);
        
        // If the subtree occurs twice, add it to the result list
        if (map.get(currId) == 2) {
            ans.add(root);
        }
        
        return currId;
    }

    // Main function to return the list of duplicate subtrees
    public List<Node> printAllDups(Node root) {
        Map<Integer, Integer> map = new HashMap<>();  // To store the count of each subtree ID
        Map<String, Integer> strToInt = new HashMap<>();  // To map serialized string to unique ID
        
        List<Node> ans = new ArrayList<>();  // To store the result of duplicate subtrees
        int[] idNum = {1};  // To assign unique IDs to subtrees
        
        // Call the helper function to serialize the tree and detect duplicates
        serializeTree(root, map, strToInt, ans, idNum);  
        
        return ans;  // Return the list of duplicate subtrees
    }
}
```

**Goal:**  
Flatten the binary tree into a linked list in-place such that the order of nodes follows the pre-order traversal.

**Approach:**
1. **Pre-order Traversal**: The order of the "linked list" should match the pre-order traversal of the tree, which visits the root first, then the left subtree, and finally the right subtree.

2. **In-place Transformation**:
   - The key idea is to perform the flattening in-place by modifying the tree's pointers directly. The left child pointer should be set to `null`, and the right child pointer should point to the next node in the flattened list.
   - For each node, we recursively flatten the left subtree, then adjust the current node's pointers:

     - Ensure that the last node in the flattened left subtree points to the right child.

3. **Recursive Flattening**:
   - Flatten the left subtree recursively.
   - Temporarily store the right child, then attach the left child to the right.

**Solution Code:**
```java
class Solution {
    public static void flatten(Node node) {
        // If the node is null, do nothing
        if (node == null)
            return;
        
        // If the node is a leaf (no left or right children), do nothing
        if (node.left == null && node.right == null)
            return;
        
        // Flatten the left subtree if it exists
        if (node.left != null) {
            flatten(node.left);  // Recursively flatten the left subtree
            
            // Temporarily store the right child
            Node tempNode = node.right;
            
            // Move the left subtree to the right of the current node
            node.right = node.left;
            node.left = null;  // Set the left child to null
            
            // Traverse to the rightmost node of the new right subtree
            Node curr = node.right;
            while (curr.right != null)
                curr = curr.right;
            
            // Attach the original right subtree to the rightmost node of the left subtree
            curr.right = tempNode;
        }
        
        // Flatten the right subtree
        flatten(node.right);
    }
}
```

**Time Complexity:**

**Space Complexity:**
O(H), where H is the height of the tree. This is the space used by the recursion stack. In the worst case, the tree can be skewed, and the height can be O(N), making the space complexity O(N).

**Notes:**  
- The flattening is done **in-place** without using extra data structures like arrays or lists.  
- The approach relies on **pre-order traversal** because the linked list should reflect the root-left-right order.  
- After flattening the left subtree, it's moved to the right pointer of the current node, and the original right subtree is attached to the end of this moved subtree.

## Problem Statement: 354
**Construct Binary Tree from Inorder and Postorder Traversal**

**Example:**  
**Input:**  
inorder = [9, 3, 15, 20, 7]  
postorder = [9, 15, 7, 20, 3]  

**Output:**  
The binary tree:  
```

```
3
```

/&#x20;
9  20
/ &#x20;
15   7

```

**Constraints:**  
- Lengths of inorder and postorder arrays are the same and valid for a binary tree.  
- All node values are unique to allow correct reconstruction.

**Goal:**  
Reconstruct the binary tree from given inorder and postorder traversals.

**Approach:**
The key idea for constructing the binary tree from inorder and postorder traversals is based on the following properties of these traversals:

1. **Inorder Traversal**: The nodes are visited in the order `left, root, right`. This means the element at the middle of the inorder array is the root of the tree or subtree.
2. **Postorder Traversal**: The nodes are visited in the order `left, right, root`. Therefore, the last element in the postorder array is the root of the tree or subtree.

Using these properties, we can follow this approach:

1. Start with the last element in the `postorder` array as the root.
2. Find the root in the `inorder` array to divide the tree into left and right subtrees.
3. Recursively build the right subtree using the elements after the root in the `inorder` array, and the corresponding elements in the `postorder` array.
4. Recursively build the left subtree using the elements before the root in the `inorder` array, and the corresponding elements in the `postorder` array.

**Algorithm:**
1. **Initialize postIndex**: The postIndex points to the last element in the `postorder` array.
2. **Build a hashmap for inorder indices**: This allows efficient lookup of the root value in the inorder array.
3. **Recursively construct the tree**:
   - The last element of the postorder array is the root.
   - Use the root value to split the inorder array into left and right subtrees.
   - Recursively build the right subtree first (since postorder traversal visits the right subtree before the left subtree).
   - Then, build the left subtree.


**🧮 Flowchart/ Diagram:**  
- Start with the last element in postorder as root.  
- Find root in inorder to split left and right subtree arrays.  
- Recursively construct right subtree first, then left subtree.  
- Return constructed tree node.

**Pseudocode/Algorithm:**  
```sql

buildTree(inorder, postorder):
postIndex = last index of postorder
build inorderMap from inorder array

return build(postorder, 0, inorder.length - 1)


build(postorder, inStart, inEnd):
if inStart > inEnd:
return null

rootVal = postorder[postIndex]
postIndex -= 1
root = new TreeNode(rootVal)

inorderIndex = inorderMap[rootVal]

root.right = build(postorder, inorderIndex + 1, inEnd)
root.left = build(postorder, inStart, inorderIndex - 1)

return root
```

**Solution Code:**
```java
class Solution {
    int postIndex;
    Map<Integer, Integer> inorderMap;

    public TreeNode buildTree(int[] inorder, int[] postorder) {
        postIndex = postorder.length - 1;
        inorderMap = new HashMap<>();

        // Build a hashmap to store value -> index relations for inorder
        for (int i = 0; i < inorder.length; i++) {
            inorderMap.put(inorder[i], i);
        }

        return build(postorder, 0, inorder.length - 1);
    }

    private TreeNode build(int[] postorder, int inStart, int inEnd) {
        if (inStart > inEnd) return null;

        // Get current root value from postorder
        int rootVal = postorder[postIndex--];
        TreeNode root = new TreeNode(rootVal);

        // Build right subtree before left (postorder => left, right, root)
        int inorderIndex = inorderMap.get(rootVal);
        root.right = build(postorder, inorderIndex + 1, inEnd);
        root.left = build(postorder, inStart, inorderIndex - 1);

        return root;
    }
}
```

**Time Complexity:**
O(N), where N is the number of nodes in the tree. Each node is processed exactly once, and the hashmap lookup takes constant time.
The recursion goes through all the nodes once and splits the arrays efficiently.

**Space Complexity:**
O(N) for the hashmap inorderMap and the recursion stack.
The recursion stack depth will be proportional to the height of the tree, which in the worst case is O(N) (for a skewed tree).

**Notes:**  
- The last element in postorder traversal is always the root of the current subtree.  
- Using a hashmap for inorder values allows O(1) lookup for root positions, optimizing the recursive tree building process.  
- Building the right subtree before the left is essential because in postorder traversal, nodes are visited in left-right-root order, so we must process the postorder array backwards.  
- The algorithm works correctly only if all node values are unique; duplicates can cause ambiguity in tree reconstruction.  
- The recursion depth and hashmap storage both contribute to O(N) space complexity.  
- This approach guarantees a unique tree construction given valid input arrays.
