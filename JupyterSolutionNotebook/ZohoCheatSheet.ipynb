## Problem Statement:191  
**Index of an Extra Element**  

You are given two sorted arrays `arr1[]` and `arr2[]` of distinct elements. The first array `arr1[]` contains one **extra element** added at some position. Your task is to find the **index** of this extra element.  

**Input:**  
- Two sorted arrays `arr1[]` and `arr2[]` of distinct integers.  
  - `arr1[]` contains one extra element compared to `arr2[]`.  

**Output:**  
- An integer representing the **index** of the extra element in `arr1[]` (0-based indexing).  

**Constraints:**  
- `1 ≤ arr2.length ≤ 10^5`  
- `arr1.length = arr2.length + 1`  
- `-10^9 ≤ arr1[i], arr2[i] ≤ 10^9`  

**Goal of the Code:**  
To efficiently determine the **index of the extra element** using a **Binary Search approach**.  

**Approach**  
The problem can be efficiently solved using **Binary Search**:  

1. **Key Idea:**  
   - Since both arrays are **sorted**, the first mismatch between `arr1[]` and `arr2[]` indicates the **index** of the extra element.  
   - Instead of comparing elements one by one (`O(N)`), we can use **Binary Search (`O(log N)`)** to locate the first mismatch efficiently.  

2. **Steps to Solve:**  
   - Initialize `left = 0` and `right = arr2.length - 1`.  
   - Perform a **binary search** to find the first index where `arr1[mid] ≠ arr2[mid]`.  
   - If `arr1[mid] == arr2[mid]`, search in the **right half** (`left = mid + 1`).  
   - Otherwise, update `index = mid` and search in the **left half** (`right = mid - 1`).  
   - The final `index` holds the **position** of the extra element.  

**Pseudocode/Algorithm**
1. Initialize `left = 0`, `right = arr2.length - 1`, and `index = arr2.length`.  
2. Perform binary search (`while left <= right`):  
   - Compute `mid = (left + right) / 2`.  
   - If `arr1[mid] == arr2[mid]`, move right (`left = mid + 1`).  
   - Otherwise, store `index = mid` and move left (`right = mid - 1`).  
3. Return `index`.  

**Time and Space Complexity**
- **Time Complexity:** `O(log N)`, since we use **Binary Search**.  
- **Space Complexity:** `O(1)`, as only a few variables are used.  

**Java Solution**
```java
class Solution {
    public int findExtra(int a[], int b[]) {
        int n = b.length;
        int index = n, left = 0, right = n - 1;

        while (left <= right) {
            int mid = (left + right) / 2;
            if (a[mid] == b[mid]) {
                left = mid + 1;
            } else {
                index = mid;
                right = mid - 1;
            }
        }
        return index;
    }
}


## Problem Statement: 192
**Make a Distinct Digit Array**  

Given an array `nums[]` of positive integers of size `N`, find **all distinct digits** present in `nums[]` and return them in **sorted order**.  

**Input:**  
- An integer `N` representing the size of the array.  
- An array `nums[]` of `N` positive integers.  

**Output:**  
- An array containing all distinct digits from `nums[]`, **sorted in ascending order**.  

**Constraints:**  
- `1 ≤ N ≤ 10^3`  
- `1 ≤ nums[i] ≤ 10^9`  

**Goal of the Code:**  
Extract all unique digits from the numbers in `nums[]` and return them **sorted**.  

**Approach**
The problem requires extracting **distinct digits** from numbers and sorting them.  

**Efficient Approach (Using a Set)**
1. **Use a `TreeSet` to store unique digits** (sorted automatically).  
2. **Iterate through each number in `nums[]`** and extract digits:  
   - Take the last digit (`n % 10`) and add it to the set.  
   - Remove the last digit (`n /= 10`) and continue.  
3. Convert the `Set<Integer>` to an `int[]` and return the result.  

**Why `TreeSet`?**  
- **Automatically sorts elements in ascending order** (no need for extra sorting).  
- **Prevents duplicate digits** efficiently.  

**Time and Space Complexity**
- **Time Complexity:** `O(N log M)`  
  - Each number has at most `log M` digits (`M = max value in nums[]` = `10^9` → `≤10` digits).  
  - Extracting and inserting digits in `TreeSet` takes `O(log 10) ≈ O(1)`.  
  - Overall, `O(N log M) ≈ O(N)`.  
- **Space Complexity:** `O(1)`, as we store at most `10` unique digits.  

**Java Solution**
```java
class Solution {
    public int[] common_digits(int[] nums) {
        Set<Integer> digitSet = new TreeSet<>();
        
        // Extract digits from each number
        for (int n : nums) {
            while (n > 0) {
                digitSet.add(n % 10);
                n /= 10;
            }
        }
        
        // Convert Set<Integer> to int[]
        int[] arr = digitSet.stream().mapToInt(Integer::intValue).toArray();
        return arr;
    }
}


## Problem Statement: 193
**Sorting Elements of an Array by Frequency**  

Given an array `A[]` of integers, sort the array according to the **frequency** of elements.  
- Elements with **higher frequency come first**.  
- If two elements have the **same frequency**, the **smaller number comes first**.  

**Input Format**  
- The first line contains an integer `T`, the number of test cases.  
- Each test case consists of:  
  - An integer `N`, the size of the array.  
  - An array `A[]` of `N` space-separated integers.  

**Output Format**  
For each test case, print the sorted array in a new line, with elements separated by a space.  

**Example**
**Input**  

**Approach**
We need to sort the elements based on their **frequency**, and if two elements have the same frequency, sort them based on **value**.  

**Steps to Solve:**
1. **Count frequency of each element** using a `HashMap`.  
2. **Use a max-heap (PriorityQueue) to sort** based on:  
   - **Higher frequency first**  
   - **Smaller number first in case of tie**  
3. **Extract elements from heap** and build the sorted array.  

**Time & Space Complexity**
- **Time Complexity:** `O(N log N)`  
  - Counting elements → `O(N)`  
  - Inserting into priority queue → `O(N log N)`  
  - Extracting from priority queue → `O(N log N)`  
- **Space Complexity:** `O(N)` (for frequency map & priority queue)  

**Java Solution**
```java
import java.util.*;

public class Main {
    public static void sortByFrequency(int[] arr) {
        Map<Integer, Integer> freqMap = new HashMap<>();

        // Step 1: Count frequency of each element
        for (int num : arr) {
            freqMap.put(num, freqMap.getOrDefault(num, 0) + 1);
        }

        // Step 2: Use a max-heap (PriorityQueue) with custom comparator
        PriorityQueue<int[]> maxHeap = new PriorityQueue<>(
            (a, b) -> (b[1] == a[1]) ? a[0] - b[0] : b[1] - a[1] // Sort by frequency, then by value
        );

        // Step 3: Add all elements with their frequency to heap
        for (Map.Entry<Integer, Integer> entry : freqMap.entrySet()) {
            maxHeap.add(new int[]{entry.getKey(), entry.getValue()});
        }

        // Step 4: Extract elements from the heap
        List<Integer> result = new ArrayList<>();
        while (!maxHeap.isEmpty()) {
            int[] top = maxHeap.poll();
            for (int i = 0; i < top[1]; i++) {
                result.add(top[0]);
            }
        }

        // Print the sorted array
        for (int num : result) {
            System.out.print(num + " ");
        }
        System.out.println();
    }

    public static void main(String[] args) {
        Scanner sc = new Scanner(System.in);
        int T = sc.nextInt(); // Number of test cases
        while (T-- > 0) {
            int N = sc.nextInt();
            int[] arr = new int[N];
            for (int i = 0; i < N; i++) {
                arr[i] = sc.nextInt();
            }
            sortByFrequency(arr);
        }
        sc.close();
    }
}
