## Problem Statement: Count Number of Hops

A frog jumps either 1, 2, or 3 steps to go to the top. In how many ways can it reach the top of the `n`th step?

**✅ Constraints**
- 1 ≤ n ≤ 10⁵

**💡 Examples**
**Input**:  
`n = 3`

**Output**:  
`4`

**Explanation**:  
The possible ways are:
- 1 step + 1 step + 1 step
- 1 step + 2 steps
- 2 steps + 1 step
- 3 steps

**🛠️ Approaches**
**🔁 1. Recursive Approach (Exponential Time)**
This approach explores all possible jumps at every step.

```java
static int countWays(int n) {
    if (n == 0) return 1;
    if (n < 0) return 0;

    return countWays(n - 1) + countWays(n - 2) + countWays(n - 3);
}
```
⛔ **Drawback:** Highly inefficient for large n due to repeated subproblems.

📦 **2. Memoization (Top-Down DP)**
Store results of subproblems to avoid recomputation.
**CODE**
```java
static int countWays(int n, int[] dp) {
    if (n == 0) return 1;
    if (n < 0) return 0;
    if (dp[n] != -1) return dp[n];

    dp[n] = countWays(n - 1, dp) + countWays(n - 2, dp) + countWays(n - 3, dp);
    return dp[n];
}
```
**Usage:**
```java
int[] dp = new int[n + 1];
Arrays.fill(dp, -1);
return countWays(n, dp);
```
✅ **Time Complexity: O(n)**
✅ **Space Complexity: O(n)**

📊**3. Tabulation (Bottom-Up DP)**
Iteratively build up the result using an array.
```java
static int countWays(int n) {
    int[] dp = new int[n + 1];
    dp[0] = 1;

    for (int i = 1; i <= n; i++) {
        dp[i] = dp[i - 1];
        if (i >= 2) dp[i] += dp[i - 2];
        if (i >= 3) dp[i] += dp[i - 3];
    }

    return dp[n];
}
```
✅ **Time Complexity: O(n)**
✅ **Space Complexity: O(n)**

🔁 **4. Space Optimization (Rolling Array)**
Only the last three values are needed at any time.
```java
static int countWays(int n) {
    int[] ways = new int[3];
    ways[0] = 1; // n = 0
    if (n == 0) return ways[0];
    ways[1] = 1; // n = 1
    if (n == 1) return ways[1];
    ways[2] = 2; // n = 2
    if (n == 2) return ways[2];

    for (int i = 3; i <= n; i++) {
        ways[i % 3] = ways[(i - 1) % 3] + ways[(i - 2) % 3] + ways[(i - 3) % 3];
    }

    return ways[n % 3];
}
```
✅ **Time Complexity: O(n)**
✅ **Space Complexity: O(1)**

🔚 **Final Java Code (Space Optimized)**
```java
class Solution {
    static int countWays(int n) {
        int[] ways = new int[3];
        ways[0] = 1;
        ways[1] = 1;
        ways[2] = 2;

        for (int i = 3; i <= n; i++)
            ways[i % 3] = ways[(i - 1) % 3] + ways[(i - 2) % 3] + ways[(i - 3) % 3];

        return ways[n % 3];
    }
}
```
🧠 **Summary**


| Approach       | Time Complexity | Space Complexity | Suitable For Large `n` |
|----------------|------------------|-------------------|--------------------------|
| Recursion      | O(3^n)           | O(n)              | ❌                      |
| Memoization    | O(n)             | O(n)              | ✅                      |
| Tabulation     | O(n)             | O(n)              | ✅                      |
| Space Optimized| O(n)             | O(1)              | ✅✅                    |

## Problem Statement: Ways to Reach the n’th Stair

There are `n` stairs, and a person standing at the bottom wants to reach the top. The person can climb either **1 stair** or **2 stairs** at a time.  
Your task is to count the number of ways the person can reach the top (**order matters**).

**✅ Constraints**
- 1 ≤ n ≤ 10⁵

**💡 Examples**
**Input**:  
`n = 4`

**Output**:  
`5`

**Explanation**:  
The possible ways are:
- 1 + 1 + 1 + 1  
- 1 + 2 + 1  
- 2 + 1 + 1  
- 1 + 1 + 2  
- 2 + 2

**💠 Approaches**
**🔁 1. Recursive Approach (Exponential Time)**
Try every path by choosing to take either 1 or 2 steps recursively.

```java
int countWays(int n) {
    if (n == 0 || n == 1) return 1;
    return countWays(n - 1) + countWays(n - 2);
}
```

❌ **Drawback:** Too slow for large `n` due to overlapping subproblems.

**📦 2. Memoization (Top-Down DP)**
Use a DP array to store results of already computed subproblems.

```java
int countWays(int n, int[] dp) {
    if (n == 0 || n == 1) return 1;
    if (dp[n] != -1) return dp[n];

    dp[n] = countWays(n - 1, dp) + countWays(n - 2, dp);
    return dp[n];
}
```

**Usage:**
```java
int[] dp = new int[n + 1];
Arrays.fill(dp, -1);
return countWays(n, dp);
```

✅ **Time Complexity:** O(n)  
✅ **Space Complexity:** O(n)

**📊 3. Tabulation (Bottom-Up DP)**
Fill the table iteratively starting from base cases.

```java
int countWays(int n) {
    int[] dp = new int[n + 1];
    dp[0] = 1;
    dp[1] = 1;

    for (int i = 2; i <= n; i++) {
        dp[i] = dp[i - 1] + dp[i - 2];
    }

    return dp[n];
}
```

✅ **Time Complexity:** O(n)  
✅ **Space Complexity:** O(n)

**🧮 4. Space Optimization (Fibonacci Style)**
Only the last two results are needed to compute the next one.

```java
int countWays(int n) {
    if (n == 0 || n == 1) return 1;

    int prev1 = 1, prev2 = 1;

    for (int i = 2; i <= n; i++) {
        int curr = prev1 + prev2;
        prev2 = prev1;
        prev1 = curr;
    }

    return prev1;
}
```

✅ **Time Complexity:** O(n)  
✅ **Space Complexity:** O(1)

**🔚 Final Java Code (Space Optimized)**
```java
class Solution {
    int countWays(int n) {
        int prev1 = 1;
        int prev2 = 1;

        for (int i = 2; i <= n; i++) {
            int curr = prev1 + prev2;
            prev2 = prev1;
            prev1 = curr;
        }

        return prev1;
    }
}
```

**🧠 Summary**

| Approach         | Time Complexity | Space Complexity | Suitable For Large `n` |
|------------------|------------------|-------------------|--------------------------|
| Recursion        | O(2^n)           | O(n)              | ❌                      |
| Memoization      | O(n)             | O(n)              | ✅                      |
| Tabulation       | O(n)             | O(n)              | ✅                      |
| Space Optimized  | O(n)             | O(1)              | ✅✅                    |

## Problem Statement: Count Ways to N'th Stair (Order Does Not Matter)

A person is standing at the bottom of a staircase with `n` stairs. They can climb either 1 stair or 2 stairs at a time. You need to count how many distinct ways the person can reach the top of the stairs.

**Note:**
- **Order does not matter**, meaning combinations like `{1 2 1}`, `{2 1 1}`, and `{1 1 2}` are considered the same.

**✅ Constraints:**
- 1 ≤ n ≤ 10⁵

**💡 Examples:**
**Input:**
```
n = 4
```
**Output:**
```
3
```
**Explanation:**
- The distinct combinations are:
  - 1+1+1+1
  - 1+1+2
  - 2+2
Even though `1+2+1`, `2+1+1`, etc. sum to 4, they are the same as `1+1+2` due to order not mattering.

**🛠️ Approaches**

**🔁 1. Combinatorics Based Observation**
- Since only steps of 1 and 2 are allowed and order doesn't matter, for any `n`, we can pick up to `n / 2` twos (each two replaces two ones).
- So the number of ways is `n / 2 + 1`.

**📄 Code:**
```java
class Solution {
    public int nthStair(int n) {
        return (n / 2) + 1;
    }
}
```

**🧠 Summary Table**

| Value of `n` | Possible Distinct Ways (Ignoring Order) |
|--------------|------------------------------------------|
| 1            | 1 (1)                                    |
| 2            | 2 (1+1, 2)                                |
| 3            | 2 (1+1+1, 1+2)                            |
| 4            | 3 (1+1+1+1, 1+1+2, 2+2)                   |
| 5            | 3 (1+1+1+1+1, 1+1+1+2, 1+2+2)             |
| ...          | ...                                      |

**🔢 Final Formula:**
```
ways = (n / 2) + 1
```
- 🔝 **Time Complexity:** O(1)
- 🔝 **Space Complexity:** O(1)

This approach leverages the fact that the number of distinct unordered partitions into 1s and 2s is exactly `floor(n / 2) + 1`. This is efficient and ideal for large `n`.



## Problem Statement: House Robber

You are a professional robber planning to rob houses along a street. Each house has a certain amount of money stashed. The only constraint stopping you from robbing each of them is that **adjacent houses have security systems connected**, and robbing two adjacent houses will automatically notify the police.
Given an integer array `nums` representing the amount of money in each house, return the **maximum amount of money you can rob tonight without alerting the police**.

**✅ Constraints**
- `1 <= nums.length <= 10^4`
- `0 <= nums[i] <= 400`

**💡 Examples**
**Input:** `nums = [2,7,9,3,1]`  
**Output:** `12`

**Explanation:**
Rob house 1 (money = 2) + rob house 3 (money = 9) + rob house 5 (money = 1) = 2 + 9 + 1 = **12**.

**🛠️ Approaches**
**🔁 1. Recursive Approach (Exponential Time)**
```java
int rob(int[] nums, int i) {
    if (i < 0) return 0;
    return Math.max(rob(nums, i - 1), nums[i] + rob(nums, i - 2));
}
```
⛔ **Drawback:** Exponential time due to overlapping subproblems.

**📦 2. Memoization (Top-Down DP)**
```java
int rob(int[] nums) {
    int[] memo = new int[nums.length];
    Arrays.fill(memo, -1);
    return robFrom(nums, nums.length - 1, memo);
}

int robFrom(int[] nums, int i, int[] memo) {
    if (i < 0) return 0;
    if (memo[i] >= 0) return memo[i];
    memo[i] = Math.max(robFrom(nums, i - 1, memo), nums[i] + robFrom(nums, i - 2, memo));
    return memo[i];
}
```
✅ **Time Complexity: O(n)**  
✅ **Space Complexity: O(n)**

**📊 3. Tabulation (Bottom-Up DP)**
```java
int rob(int[] nums) {
    int n = nums.length;
    if (n == 0) return 0;
    if (n == 1) return nums[0];

    int[] dp = new int[n];
    dp[0] = nums[0];
    dp[1] = Math.max(nums[0], nums[1]);

    for (int i = 2; i < n; i++) {
        dp[i] = Math.max(dp[i - 1], nums[i] + dp[i - 2]);
    }

    return dp[n - 1];
}
```
✅ **Time Complexity: O(n)**  
✅ **Space Complexity: O(n)**

**🔁 4. Space Optimization (Two Variables)**
```java
class Solution {
    public int rob(int[] nums) {
        int n = nums.length;
        if (n == 0) return 0;
        if (n == 1) return nums[0];

        int prev1 = nums[0];
        int prev2 = 0;

        for (int i = 1; i < n; i++) {
            int take = nums[i] + prev2;
            int skip = prev1;
            int curr = Math.max(take, skip);

            prev2 = prev1;
            prev1 = curr;
        }

        return prev1;
    }
}
```
✅ **Time Complexity: O(n)**  
✅ **Space Complexity: O(1)**

**🧠 Summary**
| Approach        | Time Complexity | Space Complexity | Recommended For |
|----------------|------------------|-------------------|------------------|
| Recursion       | O(2^n)           | O(n)              | ❌                |
| Memoization     | O(n)             | O(n)              | ✅                |
| Tabulation      | O(n)             | O(n)              | ✅                |
| Space Optimized | O(n)             | O(1)              | ✅✅              |



## Problem Statement: House Robber II

You are a professional robber planning to rob houses along a street. Each house has a certain amount of money stashed. All houses at this place are arranged in a circle. That means the first house is the neighbor of the last one. Meanwhile, adjacent houses have a security system connected, and it will automatically contact the police if two adjacent houses were broken into on the same night.
Given an integer array `nums` representing the amount of money of each house, return the maximum amount of money you can rob tonight without alerting the police.

**💡 Examples:**
**Input:**  
`nums = [2, 3, 2]`

**Output:**  
`3`

**Explanation:**  
Robbing house 2 is the best choice, as robbing house 1 and house 3 will alert the police.

**🛠️ Approach**
The problem is similar to the "House Robber" problem, but with a twist: since the houses are arranged in a circle, we cannot rob both the first and last houses together. Therefore, we need to solve the problem in two parts:
- Rob houses excluding the first house.
- Rob houses excluding the last house.
Then, take the maximum of the two results.

**📝 Algorithm**
1. **Linear Robbery Approach:**  
   The approach is the same as the House Robber problem. We use dynamic programming to solve this sub-problem.

2. **Circular Robbery:**  
   Since the houses are arranged in a circle, we can either:
   - Rob houses starting from the first house but excluding the last house.
   - Rob houses starting from the second house but excluding the first house.

We solve both these cases and return the maximum of the two.

**Code Implementation**
```java
class Solution {
    public int rob(int[] nums) {
        int n = nums.length;
        if (n == 1) return nums[0];

        // Two cases: exclude the first house or exclude the last house
        return Math.max(robLinear(nums, 0, n - 2), robLinear(nums, 1, n - 1));
    }

    private int robLinear(int[] nums, int start, int end) {
        int prev1 = 0, prev2 = 0;
        for (int i = start; i <= end; i++) {
            int take = nums[i] + prev2;
            int skip = prev1;
            int curr = Math.max(take, skip);
            prev2 = prev1;
            prev1 = curr;
        }
        return prev1;
    }
}
```
**🧠 Explanation:**
1. **Handling Circularity:**

        First, we check the base case where there is only one house (n == 1). In this case, return the money in that house.

        We then break the problem into two linear sub-problems:

            1. Rob houses from the 0th to the (n-2)th house.

            2. Rob houses from the 1st to the (n-1)th house.

        Finally, we take the maximum of these two results, since we cannot rob both the first and last house.

2. **Robbing Linear Houses:**
        In the robLinear function, we apply the classic dynamic programming approach where prev1 represents the maximum money we can rob up to the previous house, and prev2 represents the maximum money up to two houses ago.

        For each house, we decide whether to rob it (take = nums[i] + prev2) or skip it (skip = prev1), and update prev1 and prev2 accordingly.

**⏱️ Time Complexity:**
Time Complexity: O(n), where n is the number of houses (length of nums). We solve the problem in two linear passes.

Space Complexity: O(1), since we only need a constant amount of space to store the previous two values.

**🧠 Space Optimization:**
We optimize space by keeping track of only the last two values instead of maintaining an entire DP array.

**📝 Summary Table**
| Approach            | Time Complexity | Space Complexity | Suitable for Large `n` |
|---------------------|-----------------|------------------|------------------------|
| Circular Robbery    | O(n)            | O(1)             | ✅                     |

## Problem Statement: Count the Paths

Given a Directed Acyclic Graph (DAG) with `V` nodes labeled from `0` to `V-1`. Given edges, start, and destination, count the number of ways to reach from the start node to the destination node. There is a directed edge from vertex `edges[i][0]` to vertex `edges[i][1]`. Your task is to find the number of ways to reach from the start to the destination.

**💡 Example:**
**Input:**
```plaintext
edges = [[0, 1], [0, 2], [1, 3], [2, 3]]
n = 4
start = 0
destination = 3
```
**Output:**
2
**Explanation:**
From start = 0, the possible paths are:
0 → 1 → 3
0 → 2 → 3
Thus, the total number of paths from start to destination is 2.

**🛠️ Approach**
The problem can be solved by using Depth-First Search (DFS) with Memoization. We perform DFS from the start node and explore all the possible paths. To avoid recomputing the same paths repeatedly, we store the result of each node in a memo array, which allows us to return the result immediately if we have already computed it for that node.

**📝 Algorithm**
1. **Graph Representation:**
Convert the list of edges into an adjacency list representation of the graph. This allows us to easily traverse all the neighbors of a node.

2. **DFS Traversal:**
We perform DFS starting from the start node. For each node, we explore all its neighbors and recursively count the number of paths to the destination.

3. **Memoization**:
Use a memo array to store the number of ways to reach the destination from each node, so we do not recompute the same result multiple times.

4. **Base Case:**
If we reach the destination node, return 1 because we've found a valid path.

5. **Recursive Case:**
For each neighboring node, recursively compute the number of paths to the destination and add them up.

**Code Implementation**
```java
class Solution {
    public int possible_paths(int[][] edges, int n, int start, int destination) {
        // Step 1: Create the graph as an adjacency list
        List<List<Integer>> graph = new ArrayList<>();
        for (int i = 0; i < n; i++) graph.add(new ArrayList<>());

        // Step 2: Fill the graph with edges
        for (int[] edge : edges) {
            graph.get(edge[0]).add(edge[1]);
        }

        // Step 3: Use memoization to store the results of subproblems
        Integer[] memo = new Integer[n];
        return dfs(graph, start, destination, memo);
    }

    private int dfs(List<List<Integer>> graph, int node, int dest, Integer[] memo) {
        // Base case: If we have reached the destination, return 1
        if (node == dest) return 1;

        // If the result is already computed, return the cached value
        if (memo[node] != null) return memo[node];

        // Recursive case: Explore all the neighbors
        int count = 0;
        for (int neighbor : graph.get(node)) {
            count += dfs(graph, neighbor, dest, memo);
        }

        // Store the result in the memo array and return it
        memo[node] = count;
        return count;
    }
}
```
**🧠 Explanation:**
1. **Graph Representation:**
We first convert the list of edges into an adjacency list. This helps in traversing the graph easily.

2. **DFS Traversal:**
We start the DFS from the start node. For each node, we recursively explore its neighbors. For every valid path we reach the destination, we add 1 to the count.

3. **Memoization:**
Before exploring the neighbors of a node, we check if we have already computed the number of ways to reach the destination from that node. If it is already computed, we directly use the stored result to avoid recomputation.

4. **Base Case:**
If the current node is the destination, we return 1, indicating that we have found a valid path.

5. **Recursive Case:**
For each neighbor, we recursively call DFS to count the number of paths from that neighbor to the destination.

**⏱️ Time Complexity:**
**Time Complexity: O(V + E)**, where V is the number of vertices and E is the number of edges. Each node and edge is processed once due to the memoization.

**Space Complexity: O(V)**, due to the space used by the graph and memoization array.

**🧠 Space Optimization:**
The space complexity is already optimized with memoization, storing the results of subproblems to avoid recomputation. We are only using an additional memo array, which stores the results for each node.

**📝 Summary Table**
| Approach           | Time Complexity | Space Complexity | Suitable for Large `n` |
|--------------------|-----------------|------------------|------------------------|
| DFS with Memoization | O(V + E)        | O(V)             | ✅                     |

## Problem Statement: Maximum Path Sum in Matrix

You are given a matrix `mat[][]` of size `n x m` where each element is a positive integer. Starting from any cell in the first row, you are allowed to move to the next row, but with specific movement constraints. From any cell `(r, c)` in the current row, you can move to any of the three possible positions:

- (r+1, c-1) — move diagonally to the left.
- (r+1, c) — move directly down.
- (r+1, c+1) — move diagonally to the right.

Find the maximum sum of any path starting from any column in the first row and ending at any column in the last row, following the above movement constraints.

**Constraints:**
- 1 ≤ n, m ≤ 1000
- mat[i][j] ≥ 1

**Example**
**Input:**
```java
mat = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]
```
**Output:**
```java
15
```
**Explanation:** The path with the maximum sum is 3 → 6 → 9, giving a sum of 18.

**Approaches**

**1. Dynamic Programming (Bottom-Up Approach)**

The most efficient approach is to use dynamic programming. The idea is to iterate over the rows and calculate the maximum sum for each cell while considering all three possible directions of movement.

```java
class Solution {
    public int maximumPath(int[][] mat) {
        int n = mat.length, m = mat[0].length;
        int res = Arrays.stream(mat[0]).max().getAsInt();
        
        for (int i = 1; i < n; i++) {
            for (int j = 0; j < m; j++) {
                int up = mat[i - 1][j];
                int left = (j > 0) ? mat[i - 1][j - 1] : 0;
                int right = (j < m - 1) ? mat[i - 1][j + 1] : 0;
                mat[i][j] += Math.max(up, Math.max(left, right));
                res = Math.max(res, mat[i][j]);
            }
        }
        return res;
    }
}
```

**Time Complexity:**  
- **O(n * m)** because we are iterating over the entire matrix once.

**Space Complexity:**  
- **O(1)** if we use space optimization and update the matrix in place.

**2. Recursive Approach (Brute Force)**

A naive approach would be to explore every possible path recursively, starting from any cell in the first row, and recursively exploring all valid paths (down, left-down, right-down) until reaching the last row.

```java
class Solution {
    public int maximumPath(int[][] mat) {
        int n = mat.length, m = mat[0].length;
        int maxSum = Integer.MIN_VALUE;

        // Start recursion from each cell in the first row
        for (int i = 0; i < m; i++) {
            maxSum = Math.max(maxSum, findMaxPath(mat, 0, i, n, m));
        }
        return maxSum;
    }

    private int findMaxPath(int[][] mat, int r, int c, int n, int m) {
        // Base case: if we reach the last row
        if (r == n - 1) {
            return mat[r][c];
        }

        // Move to the next row in three possible directions
        int down = mat[r][c] + findMaxPath(mat, r + 1, c, n, m);
        int left = (c > 0) ? mat[r][c] + findMaxPath(mat, r + 1, c - 1, n, m) : Integer.MIN_VALUE;
        int right = (c < m - 1) ? mat[r][c] + findMaxPath(mat, r + 1, c + 1, n, m) : Integer.MIN_VALUE;

        return Math.max(down, Math.max(left, right));
    }
}
```

**Time Complexity:**  
- **O(3^n)** as we recursively explore three paths at each step.

**Space Complexity:**  
- **O(n)** due to the recursive call stack.

**3. Memoization (Top-Down DP)**

The recursive approach can be optimized by using memoization to store the results of subproblems, which ensures that each subproblem is solved only once.

```java
class Solution {
    public int maximumPath(int[][] mat) {
        int n = mat.length, m = mat[0].length;
        int maxSum = Integer.MIN_VALUE;

        // Memoization table to store intermediate results
        Integer[][] memo = new Integer[n][m];

        // Start recursion from each cell in the first row
        for (int i = 0; i < m; i++) {
            maxSum = Math.max(maxSum, findMaxPath(mat, 0, i, n, m, memo));
        }
        return maxSum;
    }

    private int findMaxPath(int[][] mat, int r, int c, int n, int m, Integer[][] memo) {
        // Base case: if we reach the last row
        if (r == n - 1) {
            return mat[r][c];
        }

        // If this subproblem has already been solved, return the result
        if (memo[r][c] != null) {
            return memo[r][c];
        }

        // Move to the next row in three possible directions
        int down = mat[r][c] + findMaxPath(mat, r + 1, c, n, m, memo);
        int left = (c > 0) ? mat[r][c] + findMaxPath(mat, r + 1, c - 1, n, m, memo) : Integer.MIN_VALUE;
        int right = (c < m - 1) ? mat[r][c] + findMaxPath(mat, r + 1, c + 1, n, m, memo) : Integer.MIN_VALUE;

        // Store the result in the memo table
        memo[r][c] = Math.max(down, Math.max(left, right));

        return memo[r][c];
    }
}
```
**Time Complexity:**  
- **O(n * m)** because we only solve each subproblem once.

**Space Complexity:**  
- **O(n * m)** for the memoization table.

**4. Breadth-First Search (BFS)**

In this approach, we can perform a level-order traversal (BFS) from the first row, processing all possible paths iteratively. However, this method is not as optimal as DP for problems with overlapping subproblems.

**BFS Approach Concept:**
- Initialize a queue with all the positions from the first row.
- For each node in the queue, process all its neighbors (possible moves to the next row).
- Update the sums based on the path sums and continue until all nodes are processed.
- At the end, return the maximum sum found in the last row.

**Summary of Approaches**

| Approach           | Time Complexity | Space Complexity | Suitable For Large `n` |
|--------------------|-----------------|------------------|------------------------|
| Dynamic Programming | O(n * m)        | O(1)             | ✅                     |
| Recursive (Brute)   | O(3^n)          | O(n)             | ❌                     |
| Memoization (DP)    | O(n * m)        | O(n * m)         | ✅                     |
| BFS                 | O(n * m)        | O(n * m)         | ✅                     |

**Conclusion**
- **Dynamic Programming (Bottom-Up)** is the most efficient approach for solving this problem in both time and space.
- **Recursive (Brute Force)** and **Memoization** are less efficient but can be useful for smaller grids.
- **Breadth-First Search (BFS)** is not as efficient as DP and should be avoided for this type of problem.


## Problem Statement: Subset Sum Problem
Given an array of positive integers `arr[]` and a value `sum`, determine if there is a subset of `arr[]` with a sum equal to the given sum.

**Constraints:**
- 1 ≤ arr[i] ≤ 1000
- 1 ≤ n ≤ 1000
- 1 ≤ sum ≤ 1000

**Example**
**Input:**
```java
arr[] = {3, 34, 4, 12, 5, 2}
sum = 9
```

**Output:**
```java
true
```

**Explanation:**
- There is a subset with sum 9: `{4, 5}`.

**Approaches**

**1. Dynamic Programming (Tabulation)**
We can solve this problem using dynamic programming by creating a 2D DP array `dp[i][j]` where `dp[i][j]` is true if a subset with sum `j` can be made using the first `i` elements.

**Java Code:**
```java
class Solution {

    static Boolean isSubsetSum(int arr[], int sum) {
        int n = arr.length;
        boolean[] prev = new boolean[sum + 1];
        boolean[] curr = new boolean[sum + 1];
        prev[0] = true;

        for (int i = 1; i <= n; i++) {
            for (int j = 0; j <= sum; j++) {
                if (j < arr[i - 1]) {
                    curr[j] = prev[j];
                } else {
                    curr[j] = prev[j] || prev[j - arr[i - 1]];
                }
            }
            System.arraycopy(curr, 0, prev, 0, sum + 1);
        }
        return prev[sum];
    }
}
```

**Explanation:**
- We maintain two arrays, `prev` and `curr`, to keep track of possible subset sums for each value in the array.
- The `prev` array stores the results from the previous row, while the `curr` array stores the current row’s results.
- Initially, `prev[0]` is set to true because we can always form the sum 0 (by selecting no elements).

**Time Complexity:**  
- **O(n * sum)** where `n` is the number of elements in the array, and `sum` is the target sum.

**Space Complexity:**  
- **O(sum)** due to the space used by the `prev` and `curr` arrays.

**2. Recursive Approach (Brute Force)**

In a brute force approach, we recursively check if we can form the given sum by including or excluding each element from the array.

**Java Code:**
```java
class Solution {

    static Boolean isSubsetSum(int arr[], int n, int sum) {
        if (sum == 0) return true;
        if (n == 0) return false;

        // If last element is greater than sum, it cannot be included
        if (arr[n - 1] > sum) {
            return isSubsetSum(arr, n - 1, sum);
        }

        // Check if sum can be obtained by including or excluding the last element
        return isSubsetSum(arr, n - 1, sum) || isSubsetSum(arr, n - 1, sum - arr[n - 1]);
    }
}
```

**Explanation:**
- The base cases are:
  - If the sum is 0, we can form the sum by not including any elements.
  - If we have no elements left and the sum is not 0, we cannot form the sum.
- At each recursive step, we either include or exclude the current element.

**Time Complexity:**  
- **O(2^n)** due to the exponential nature of the recursive calls.

**Space Complexity:**  
- **O(n)** due to the recursion stack.

**3. Memoization (Top-Down Dynamic Programming)**

We can optimize the recursive approach by using memoization to store the results of already computed subproblems.

**Java Code:**
```java
class Solution {

    static Boolean isSubsetSum(int arr[], int n, int sum) {
        Integer[][] dp = new Integer[n + 1][sum + 1];
        return isSubsetSumMemo(arr, n, sum, dp);
    }

    static Boolean isSubsetSumMemo(int arr[], int n, int sum, Integer[][] dp) {
        if (sum == 0) return true;
        if (n == 0) return false;

        if (dp[n][sum] != null) return dp[n][sum];

        // If last element is greater than sum, it cannot be included
        if (arr[n - 1] > sum) {
            dp[n][sum] = isSubsetSumMemo(arr, n - 1, sum, dp) ? 1 : 0;
        } else {
            dp[n][sum] = (isSubsetSumMemo(arr, n - 1, sum, dp) || isSubsetSumMemo(arr, n - 1, sum - arr[n - 1], dp)) ? 1 : 0;
        }

        return dp[n][sum] == 1;
    }
}
```

**Explanation:**
- We store results of subproblems in the `dp` table to avoid redundant calculations.
- The table is initialized with `null` values, and for each subproblem, we check if it has already been solved.

**Time Complexity:**  
- **O(n * sum)** due to the memoization table.

**Space Complexity:**  
- **O(n * sum)** for the `dp` table.

**4. Space Optimized Dynamic Programming**

Instead of maintaining a 2D DP table, we can optimize the space by using only a 1D array, since we only need the previous row at each step.

**Java Code:**
```java
class Solution {

    static Boolean isSubsetSum(int arr[], int sum) {
        int n = arr.length;
        boolean[] dp = new boolean[sum + 1];
        dp[0] = true;

        for (int i = 0; i < n; i++) {
            for (int j = sum; j >= arr[i]; j--) {
                dp[j] = dp[j] || dp[j - arr[i]];
            }
        }

        return dp[sum];
    }
}
```

**Explanation:**
- We use a single array `dp[]` where `dp[j]` represents whether a subset with sum `j` is possible.
- We iterate over the array in reverse order to avoid overwriting values that might be needed later in the same iteration.

**Time Complexity:**  
- **O(n * sum)**, similar to the tabulation approach.

**Space Complexity:**  
- **O(sum)**, as we are using only a 1D array.

**Summary of Approaches**

| Approach                 | Time Complexity  | Space Complexity    | Suitable For Large `n` |
|--------------------------|------------------|---------------------|------------------------|
| Brute Force (Recursive)  | O(2^n)           | O(n)                | ❌                     |
| Memoization (Top-Down DP)| O(n * sum)       | O(n * sum)          | ✅                     |
| Dynamic Programming      | O(n * sum)       | O(sum)              | ✅                     |
| Space Optimized DP       | O(n * sum)       | O(sum)              | ✅✅                    |

**Conclusion**

- **Dynamic Programming** (Tabulation or Space Optimized) is the most efficient approach for this problem, with a time complexity of **O(n * sum)**.
- **Brute Force (Recursive)** is not suitable for larger inputs due to its exponential time complexity.
- **Memoization** offers a middle ground, improving the recursive approach by avoiding redundant calculations.

## Problem Statement: Perfect Sum Problem
Given an array `arr` of non-negative integers and an integer `target`, the task is to count all subsets of the array whose sum is equal to the given target.

**Constraints:**
- 1 ≤ arr[i] ≤ 1000
- 1 ≤ n ≤ 1000
- 1 ≤ target ≤ 1000

**Example**
**Input:**
```java
arr[] = {1, 2, 3}
target = 4
```

**Output:**
```java
2
```

**Explanation:**
- The subsets of `{1, 2, 3}` that sum to 4 are `{1, 3}` and `{2, 2}`.

**Approaches**

**1. Dynamic Programming (Tabulation)**
We can solve this problem using dynamic programming, similar to the subset sum problem. We maintain a DP table `dp[i][j]` where `dp[i][j]` will store the number of subsets that can form the sum `j` using the first `i` elements of the array.

**Java Code:**
```java
class Solution {
    // Function to calculate the number of subsets with a given sum
    public int perfectSum(int[] arr, int target) {
        int n = arr.length;
        int[] prev = new int[target + 1];
        int[] curr = new int[target + 1];
        prev[0] = 1; // There's one way to form sum 0 (by choosing no elements)

        for (int i = 1; i <= n; i++) {
            System.arraycopy(prev, 0, curr, 0, target + 1);
            for (int j = 0; j <= target; j++) {
                if (j >= arr[i - 1]) {
                    curr[j] += prev[j - arr[i - 1]];
                }
            }
            System.arraycopy(curr, 0, prev, 0, target + 1);
        }
        return curr[target];
    }
}
```

**Explanation:**
- We maintain two arrays, `prev` and `curr`. The `prev` array stores the results of the previous row, and `curr` stores the results for the current row.
- The logic behind the tabulation is to check if we can include or exclude the current element to form the sum.
- We initialize `prev[0] = 1` because there is one way to form sum 0 (by selecting no elements).

**Time Complexity:**  
- **O(n * target)**, where `n` is the number of elements in the array and `target` is the target sum.

**Space Complexity:**  
- **O(target)**, as we only maintain two arrays of size `target + 1`.

**2. Recursive Approach (Brute Force)**
In the brute force approach, we recursively check if we can form the given target sum by including or excluding each element in the array.

**Java Code:**
```java
class Solution {
    // Function to calculate the number of subsets with a given sum
    public int perfectSum(int[] arr, int target) {
        return countSubsets(arr, arr.length, target);
    }

    private int countSubsets(int[] arr, int n, int target) {
        if (target == 0) return 1;
        if (n == 0) return 0;

        if (arr[n - 1] > target) {
            return countSubsets(arr, n - 1, target);
        } else {
            return countSubsets(arr, n - 1, target) + countSubsets(arr, n - 1, target - arr[n - 1]);
        }
    }
}
```

**Explanation:**
- This approach recursively explores all possible subsets by either including or excluding the current element at each step.
- If we include the current element, we reduce the target by the element's value and proceed with the remaining elements.
- If we exclude the current element, we proceed with the remaining elements without changing the target.

**Time Complexity:**  
- **O(2^n)** due to the exponential number of recursive calls.

**Space Complexity:**  
- **O(n)** due to the recursion stack.

**3. Memoization (Top-Down Dynamic Programming)**
We can optimize the recursive approach using memoization to store the results of already computed subproblems.

**Java Code:**
```java
class Solution {
    // Function to calculate the number of subsets with a given sum
    public int perfectSum(int[] arr, int target) {
        Integer[][] dp = new Integer[arr.length + 1][target + 1];
        return countSubsets(arr, arr.length, target, dp);
    }

    private int countSubsets(int[] arr, int n, int target, Integer[][] dp) {
        if (target == 0) return 1;
        if (n == 0) return 0;

        if (dp[n][target] != null) return dp[n][target];

        if (arr[n - 1] > target) {
            dp[n][target] = countSubsets(arr, n - 1, target, dp);
        } else {
            dp[n][target] = countSubsets(arr, n - 1, target, dp) + countSubsets(arr, n - 1, target - arr[n - 1], dp);
        }
        return dp[n][target];
    }
}
```

**Explanation:**
- We store the results of the recursive calls in the `dp` table to avoid redundant calculations.
- The table is initialized with `null` values, and for each subproblem, we check if it has already been solved.

**Time Complexity:**  
- **O(n * target)** due to the memoization table.

**Space Complexity:**  
- **O(n * target)** for the `dp` table.

**4. Space Optimized Dynamic Programming**
We can optimize the space by using a single array instead of a 2D array, since we only need the previous row at each step.

**Java Code:**
```java
class Solution {
    // Function to calculate the number of subsets with a given sum
    public int perfectSum(int[] arr, int target) {
        int n = arr.length;
        int[] dp = new int[target + 1];
        dp[0] = 1; // There's one way to form sum 0 (by choosing no elements)

        for (int i = 0; i < n; i++) {
            for (int j = target; j >= arr[i]; j--) {
                dp[j] += dp[j - arr[i]];
            }
        }

        return dp[target];
    }
}
```

**Explanation:**
- We use a single array `dp[]` where `dp[j]` represents the number of subsets that can form the sum `j`.
- We iterate over the array in reverse order to avoid overwriting values that might be needed later in the same iteration.

**Time Complexity:**  
- **O(n * target)**, similar to the tabulation approach.

**Space Complexity:**  
- **O(target)**, as we are using only a 1D array.

**Summary of Approaches**
| Approach                 | Time Complexity  | Space Complexity    | Suitable For Large `n` |
|--------------------------|------------------|---------------------|------------------------|
| Brute Force (Recursive)  | O(2^n)           | O(n)                | ❌                     |
| Memoization (Top-Down DP)| O(n * target)    | O(n * target)       | ✅                     |
| Dynamic Programming      | O(n * target)    | O(target)           | ✅                     |
| Space Optimized DP       | O(n * target)    | O(target)           | ✅✅                    |

**Conclusion**
- **Dynamic Programming (Tabulation or Space Optimized)** is the most efficient approach for this problem, with a time complexity of **O(n * target)**.
- **Brute Force (Recursive)** is not suitable for larger inputs due to its exponential time complexity.
- **Memoization** optimizes the recursive approach by avoiding redundant calculations, but it still has the same time complexity as the tabulation approach.


## Problem Statement: Partition Equal Subset Sum
Given an array `arr[]`, determine if it can be partitioned into two subsets such that the sum of elements in both parts is the same.

**Constraints:**
- 1 ≤ arr[i] ≤ 1000
- 1 ≤ n ≤ 1000

**Example**
**Input:**
```java
arr[] = {1, 5, 11, 5}
```

**Output:**
```java
true
```

**Explanation:**
- The array can be partitioned into two subsets `{1, 5, 5}` and `{11}`. The sum of both subsets is `11`, so it can be partitioned into two subsets with equal sum.

**Approaches**

**1. Dynamic Programming (Tabulation)**

We need to check if there is a subset whose sum is half of the total sum of the array. If such a subset exists, then the other subset will automatically have the same sum. 

We use dynamic programming to solve the problem by keeping track of the possible subset sums in a boolean array. 

**Java Code:**
```java
class Solution {
    static boolean equalPartition(int arr[]) {
        int sum = 0; 
        for (int num : arr) {
            sum += num;
        }
        // If sum is odd, it is not possible to partition it into two equal subsets
        if (sum % 2 != 0) return false;

        sum = sum / 2;  // We need to check if there's a subset with sum = sum/2

        int n = arr.length;
        boolean[] prev = new boolean[sum + 1];
        boolean[] curr = new boolean[sum + 1];
        prev[0] = true;  // There's always a way to get sum 0 (by choosing no elements)

        for (int i = 1; i <= n; i++) {
            for (int j = 0; j <= sum; j++) {
                if (j < arr[i - 1]) {
                    curr[j] = prev[j];  // If the current element is larger than the sum, don't include it
                } else {
                    curr[j] = (prev[j] || prev[j - arr[i - 1]]);  // Include or exclude the element
                }
            }
            // Copy the current row to the previous row for the next iteration
            for (int j = 0; j <= sum; j++) {
                prev[j] = curr[j];
            }
        }

        return prev[sum];  // If we can form the sum 'sum/2', return true
    }
}
```
**Explanation:**
- First, we calculate the total sum of the array. If the sum is odd, it is impossible to partition the array into two equal subsets, so we return `false`.
- We then use dynamic programming to check if there is a subset whose sum is half of the total sum. We maintain two arrays `prev` and `curr` where `prev[i]` represents whether a subset sum `i` is possible with the elements considered so far.
- For each element, we either include it in the subset or exclude it. We update the `curr` array accordingly.
- After processing all elements, we check if we can form a subset with the sum equal to `sum / 2`.

**Time Complexity:**  
- **O(n * sum)**, where `n` is the number of elements and `sum` is half of the total sum.

**Space Complexity:**  
- **O(sum)**, as we are using two arrays of size `sum + 1`.

**2. Space Optimized Dynamic Programming**

Instead of using a 2D array to track the subset sums, we can optimize the space complexity by using a 1D array.
**Java Code:**
```java
class Solution {
    static boolean equalPartition(int arr[]) {
        int sum = 0; 
        for (int num : arr) {
            sum += num;
        }
        // If sum is odd, it is not possible to partition it into two equal subsets
        if (sum % 2 != 0) return false;

        sum = sum / 2;  // We need to check if there's a subset with sum = sum/2

        int n = arr.length;
        boolean[] dp = new boolean[sum + 1];
        dp[0] = true;  // There's always a way to get sum 0 (by choosing no elements)

        for (int i = 0; i < n; i++) {
            for (int j = sum; j >= arr[i]; j--) {
                dp[j] = dp[j] || dp[j - arr[i]];  // Include or exclude the element
            }
        }

        return dp[sum];  // If we can form the sum 'sum/2', return true
    }
}
```
**Explanation:**
- The idea is the same as the previous approach, but instead of using two arrays, we use a single array `dp[]`.
- For each element, we update the `dp[]` array in reverse order to avoid overwriting values that may still be needed in the current iteration.

**Time Complexity:**  
- **O(n * sum)**, where `n` is the number of elements and `sum` is half of the total sum.

**Space Complexity:**  
- **O(sum)**, as we only need a 1D array to store the subset sum possibilities.

**Summary of Approaches**
| Approach                     | Time Complexity  | Space Complexity    | Suitable For Large `n` |
|------------------------------|------------------|---------------------|------------------------|
| Brute Force (Recursive)      | O(2^n)           | O(n)                | ❌                     |
| Dynamic Programming (Tabulation) | O(n * sum)     | O(sum)              | ✅                     |
| Space Optimized DP           | O(n * sum)       | O(sum)              | ✅✅                    |

**Conclusion**
- **Dynamic Programming (Tabulation or Space Optimized)** is the most efficient approach for this problem, with a time complexity of **O(n * sum)**, where `sum` is half of the total sum of the array.
- **Brute Force (Recursive)** is not suitable for larger inputs due to its exponential time complexity.
- **Space Optimized DP** is a better approach as it reduces the space complexity to **O(sum)**, making it more suitable for larger inputs.

## Problem Statement: 0-1 Knapsack Problem
Given `n` items, each with a specific weight and value, and a knapsack with a capacity of `W`, determine the maximum value that can be obtained by selecting items such that their total weight does not exceed `W`. Each item can be included **at most once**.

**Constraints:**
- 1 <= n <= 1000
- 1 <= W <= 1000
- 1 <= wt[i], val[i] <= 1000

**Example**
**Input:**
```java
val = {60, 100, 120}
wt = {10, 20, 30}
W = 50
```
**Output:**
```java
220
```
**Explanation:**
- Take items with weight 20 and 30 for a total value of 100 + 120 = 220.

**Approaches**
**1. Recursive + Memoization (Top-down DP)**

**Java Code:**
```java
class Solution {
    static int knapsackRec(int W, int[] val, int[] wt, int n, int[][] memo) {
        if (n == 0 || W == 0)
            return 0;

        if (memo[n][W] != -1)
            return memo[n][W];

        int pick = 0;
        if (wt[n - 1] <= W)
            pick = val[n - 1] + knapsackRec(W - wt[n - 1], val, wt, n - 1, memo);

        int notPick = knapsackRec(W, val, wt, n - 1, memo);

        return memo[n][W] = Math.max(pick, notPick);
    }

    static int knapsack(int W, int val[], int wt[]) {
        int n = val.length;
        int[][] memo = new int[n + 1][W + 1];

        for (int i = 0; i <= n; i++) {
            for (int j = 0; j <= W; j++)
                memo[i][j] = -1;
        }

        return knapsackRec(W, val, wt, n, memo);
    }
}
```

**Time Complexity:** O(n * W)  
**Space Complexity:** O(n * W)

**2. Tabulation (Bottom-up DP)**
**Java Code:**
```java
class Solution {
    static int knapsack(int W, int val[], int wt[]) {
        int n = val.length;
        int[][] dp = new int[n + 1][W + 1];

        for (int i = 1; i <= n; i++) {
            for (int w = 1; w <= W; w++) {
                if (wt[i - 1] <= w) {
                    dp[i][w] = Math.max(val[i - 1] + dp[i - 1][w - wt[i - 1]], dp[i - 1][w]);
                } else {
                    dp[i][w] = dp[i - 1][w];
                }
            }
        }
        return dp[n][W];
    }
}
```

**Time Complexity:** O(n * W)  
**Space Complexity:** O(n * W)

**3. Space Optimized Tabulation**
We can use a single 1D array and update it in reverse.

**Java Code:**
```java
class Solution {
    static int knapsack(int W, int val[], int wt[]) {
        int n = val.length;
        int[] dp = new int[W + 1];

        for (int i = 0; i < n; i++) {
            for (int w = W; w >= wt[i]; w--) {
                dp[w] = Math.max(dp[w], val[i] + dp[w - wt[i]]);
            }
        }
        return dp[W];
    }
}
```

**Time Complexity:** O(n * W)  
**Space Complexity:** O(W)

**Summary of Approaches**

| Approach                   | Time Complexity | Space Complexity | Notes                         |
|---------------------------|-----------------|------------------|-------------------------------|
| Recursion + Memoization   | O(n * W)        | O(n * W)         | Top-down approach             |
| Tabulation                | O(n * W)        | O(n * W)         | Bottom-up DP                  |
| Space Optimized Tabulation| O(n * W)        | O(W)             | Efficient for large inputs    |

**Conclusion**
- The **0/1 Knapsack Problem** is a classic example of dynamic programming.
- Start with recursion + memoization if you're learning DP.
- Use tabulation or space optimization for better efficiency in production scenarios.



## Problem Statement: Max Sum without Adjacents

Given an array `arr` containing positive integers, find the maximum sum of a subsequence such that no two numbers in the subsequence are adjacent in the original array.

**✅ Constraints**
- 1 ≤ arr.length ≤ 10⁵
- 1 ≤ arr[i] ≤ 10⁴

**💡 Examples**
**Input:**  
`arr = [3, 2, 5, 10, 7]`

**Output:**  
`15`

**Explanation:**  
The maximum sum is 3 + 10 + 2 = 15 (choose 3rd and 5th elements).

**🛠️ Approaches**

**🔄 1. Recursive Approach (Exponential Time)**
Try including or excluding each element.

```java
int findMaxSum(int[] arr, int n) {
    if (n == 0) return 0;
    if (n == 1) return arr[0];

    return Math.max(findMaxSum(arr, n - 1), arr[n - 1] + findMaxSum(arr, n - 2));
}
```

⛔ **Drawback:** Exponential time complexity, not efficient for large inputs.

**📃 2. Memoization (Top-Down DP)**
Avoid recomputation by storing intermediate results.

```java
int findMaxSum(int[] arr, int n, int[] dp) {
    if (n == 0) return 0;
    if (n == 1) return arr[0];
    if (dp[n] != -1) return dp[n];

    return dp[n] = Math.max(findMaxSum(arr, n - 1, dp), arr[n - 1] + findMaxSum(arr, n - 2, dp));
}
```

**Usage:**
```java
int[] dp = new int[n + 1];
Arrays.fill(dp, -1);
return findMaxSum(arr, n, dp);
```

✅ **Time Complexity:** O(n)  
✅ **Space Complexity:** O(n)

**📊 3. Tabulation (Bottom-Up DP)**
Build the result iteratively using a dp array.

```java
int findMaxSum(int[] arr) {
    int n = arr.length;
    if (n == 0) return 0;
    if (n == 1) return arr[0];

    int[] dp = new int[n];
    dp[0] = arr[0];
    dp[1] = Math.max(arr[0], arr[1]);

    for (int i = 2; i < n; i++) {
        dp[i] = Math.max(dp[i - 1], arr[i] + dp[i - 2]);
    }

    return dp[n - 1];
}
```

✅ **Time Complexity:** O(n)  
✅ **Space Complexity:** O(n)

**🤝 4. Space Optimized DP**
Use two variables to avoid using O(n) space.

```java
int findMaxSum(int[] arr) {
    int n = arr.length;
    if (n == 0) return 0;
    if (n == 1) return arr[0];

    int secondLast = 0, last = arr[0], res = 0;
    for (int i = 1; i < n; i++) {
        res = Math.max(arr[i] + secondLast, last);
        secondLast = last;
        last = res;
    }

    return res;
}
```

✅ **Time Complexity:** O(n)  
✅ **Space Complexity:** O(1)

**💪 Summary Table**
| Approach             | Time Complexity | Space Complexity | Suitable For Large `n` |
|----------------------|------------------|-------------------|--------------------------|
| Recursion            | O(2^n)           | O(n)              | ❌                      |
| Memoization          | O(n)             | O(n)              | ✅                      |
| Tabulation           | O(n)             | O(n)              | ✅                      |
| Space Optimized      | O(n)             | O(1)              | ✅✅                    |

**Final Optimized Code:**
```java
class Solution {
    int findMaxSum(int arr[]) {
        int n = arr.length;

        if (n == 0)
            return 0;
        if (n == 1)
            return arr[0];

        int secondLast = 0, last = arr[0], res = 0;
        for (int i = 1; i < n; i++) {
            res = Math.max(arr[i] + secondLast, last);
            secondLast = last;
            last = res;
        }

        return res;
    }
}
```



## Problem Statement: Coin Change (Count Ways)

Given an integer array `coins[]` representing different denominations of currency and an integer `sum`, find the number of ways you can make the `sum` by using different combinations of the coins. You can use any coin an infinite number of times.

**✅ Constraints**
- 1 <= `coins.length` <= 1000
- 1 <= `coins[i]` <= 1000
- 1 <= `sum` <= 10^5
- All answers will fit into a 32-bit integer

**💡 Example**
**Input:**  
`coins = [1, 2, 3]`  
`sum = 4`

**Output:**  
`4`

**Explanation:**
- 1+1+1+1
- 1+1+2
- 2+2
- 1+3

**🛠️ Approaches**

**🔁 1. Dynamic Programming (1D Array)**
Use a 1D DP array where `dp[j]` indicates the number of ways to make sum `j` using the given coins.

```java
class Solution {
    public int count(int coins[], int sum) {
        int n = coins.length;
        int[] dp = new int[sum + 1];
        dp[0] = 1;  // Base case: there's one way to make sum 0

        for (int i = 0; i < n; i++) {
            for (int j = coins[i]; j <= sum; j++) {
                dp[j] += dp[j - coins[i]];
            }
        }
        return dp[sum];
    }
}
```

✅ **Time Complexity:** `O(n * sum)`  
✅ **Space Complexity:** `O(sum)`

**📦 2. 2D DP Table (for Learning Purpose)**

```java
class Solution {
    public int count(int coins[], int sum) {
        int n = coins.length;
        int[][] dp = new int[n + 1][sum + 1];

        for (int i = 0; i <= n; i++) dp[i][0] = 1;  // Base case

        for (int i = 1; i <= n; i++) {
            for (int j = 1; j <= sum; j++) {
                if (coins[i - 1] <= j)
                    dp[i][j] = dp[i][j - coins[i - 1]] + dp[i - 1][j];
                else
                    dp[i][j] = dp[i - 1][j];
            }
        }
        return dp[n][sum];
    }
}
```

✅ **Time Complexity:** `O(n * sum)`  
⛔ **Space Complexity:** `O(n * sum)` (Less efficient but good for understanding transitions)

**🧠 Summary**

| Approach         | Time Complexity | Space Complexity | Suitable for Large Input |
|------------------|------------------|-------------------|---------------------------|
| 1D DP            | O(n * sum)       | O(sum)            | ✅✅✅                      |
| 2D DP Table      | O(n * sum)       | O(n * sum)        | ❌                        |

This problem is a classic **Unbounded Knapsack** problem where each coin can be reused multiple times.



## Problem Statement:Coin Change (Minimum Coins)

You are given an array `coins[]`, where each element represents a coin of a different denomination, and a target value `sum`. You have an unlimited supply of each coin type `{coins1, coins2, ..., coinsm}`.
Your task is to determine the minimum number of coins needed to obtain the target sum. If it is not possible to form the sum using the given coins, return `-1`.

**Java Code (Bottom-Up DP)**
```java
class Solution {
    public int minCoins(int coins[], int sum) {
        int[] dp = new int[sum + 1];
        Arrays.fill(dp, Integer.MAX_VALUE);
        dp[0] = 0;

        for (int i = 0; i < coins.length; i++) {
            for (int j = coins[i]; j <= sum; j++) {
                if (dp[j - coins[i]] != Integer.MAX_VALUE)
                    dp[j] = Math.min(dp[j], dp[j - coins[i]] + 1);
            }
        }

        return dp[sum] == Integer.MAX_VALUE ? -1 : dp[sum];
    }
}
```

**Explanation**
- We use a `dp[]` array where `dp[i]` represents the minimum number of coins required to get sum `i`.
- We initialize `dp[0] = 0` (base case).
- For each coin and for each target from `coin` to `sum`, update `dp[j]` using the minimum of current value and `dp[j - coin] + 1`.

**Time and Space Complexity**
- **Time Complexity:** `O(n * sum)` where `n` is the number of coin types.
- **Space Complexity:** `O(sum)`

**Additional Approaches**
**1. Recursive + Memoization (Top-Down DP)**
```java
class Solution {
    public int minCoins(int[] coins, int sum) {
        Integer[] memo = new Integer[sum + 1];
        int result = helper(coins, sum, memo);
        return result == Integer.MAX_VALUE ? -1 : result;
    }

    private int helper(int[] coins, int rem, Integer[] memo) {
        if (rem == 0) return 0;
        if (rem < 0) return Integer.MAX_VALUE;
        if (memo[rem] != null) return memo[rem];

        int min = Integer.MAX_VALUE;
        for (int coin : coins) {
            int res = helper(coins, rem - coin, memo);
            if (res != Integer.MAX_VALUE) {
                min = Math.min(min, res + 1);
            }
        }
        memo[rem] = min;
        return min;
    }
}
```

**2. BFS Approach**
```java
class Solution {
    public int minCoins(int[] coins, int sum) {
        Queue<Integer> queue = new LinkedList<>();
        boolean[] visited = new boolean[sum + 1];
        queue.offer(0);
        visited[0] = true;
        int level = 0;

        while (!queue.isEmpty()) {
            int size = queue.size();
            level++;
            for (int i = 0; i < size; i++) {
                int curr = queue.poll();
                for (int coin : coins) {
                    int next = curr + coin;
                    if (next == sum) return level;
                    if (next < sum && !visited[next]) {
                        visited[next] = true;
                        queue.offer(next);
                    }
                }
            }
        }

        return -1;
    }
}
```

**Notes**
- If it’s impossible to make the given sum using the coins, return `-1`.
- Ensure proper checks to avoid integer overflow due to `Integer.MAX_VALUE + 1`.



## Problem Statement: Rod Cutting

Given a rod of length `n` (size of price) inches and an array of prices, `price`. `price[i]` denotes the value of a piece of length `i`. Determine the maximum value obtainable by cutting up the rod and selling the pieces.

**✅ Constraints**
- 1 <= `price.length` <= 1000
- 1 <= `price[i]` <= 1000
- 1 <= `n` <= 1000

**💡 Example**
**Input:**  
`price = [1, 5, 8, 9, 10, 17, 17, 20, 24, 30]`  
`n = 8`

**Output:**  
`22`

**Explanation:**
- Cut the rod into two pieces of length 2 and 6.
- The value of length 2 is `5` and the value of length 6 is `17`, so the maximum value is `5 + 17 = 22`.

**🛠️ Approaches**

**🔁 1. Dynamic Programming (Bottom-Up Approach)**
Use a 1D DP array where `dp[i]` represents the maximum value obtainable by cutting a rod of length `i`.

```java
class Solution {
    public int cutRod(int[] price) {
        int n = price.length;
        int[] dp = new int[n + 1];
        for (int i = 1; i <= n; i++) {
            for (int j = 1; j <= i; j++) {
                dp[i] = Math.max(dp[i], price[j - 1] + dp[i - j]);
            }
        }

        return dp[n];
    }
}
```

✅ **Time Complexity:** `O(n^2)`  
✅ **Space Complexity:** `O(n)`

**📦 2. Recursive (Top-Down Approach with Memoization)**
This approach uses recursion with memoization to break down the problem into smaller subproblems and store intermediate results to avoid redundant calculations.

```java
class Solution {
    public int cutRod(int[] price) {
        int n = price.length;
        int[] dp = new int[n + 1];
        Arrays.fill(dp, -1);
        return solve(price, n, dp);
    }

    private int solve(int[] price, int n, int[] dp) {
        if (n == 0) return 0;
        if (dp[n] != -1) return dp[n];
        
        int maxVal = Integer.MIN_VALUE;
        for (int i = 1; i <= n; i++) {
            maxVal = Math.max(maxVal, price[i - 1] + solve(price, n - i, dp));
        }
        dp[n] = maxVal;
        return dp[n];
    }
}
```

✅ **Time Complexity:** `O(n^2)`  
✅ **Space Complexity:** `O(n)`

**🧠 Summary**
| Approach            | Time Complexity | Space Complexity | Suitable for Large Input |
|---------------------|-----------------|------------------|--------------------------|
| Bottom-Up DP        | O(n^2)          | O(n)             | ✅✅✅                     |
| Top-Down with Memo  | O(n^2)          | O(n)             | ✅✅✅                     |

This problem is a classic **Unbounded Knapsack** problem where each piece can be chosen multiple times.

**Additional Insights:**
- The problem is about maximizing profit with a given length, which can be seen as breaking down the rod into smaller parts.
- The recursive solution with memoization can lead to faster results compared to the naive recursive approach as it avoids recalculating results for the same subproblem.

**Edge Cases:**
- If `n == 0`, the answer is `0` since no cuts can be made.
- If all prices are zero, the maximum value will be zero.
- The price array should have a non-zero value for every rod length for the solution to be meaningful.


## Problem Statement: Knapsack with Duplicate Items

Given a set of items, each with a weight and a value, represented by the array `wt` and `val` respectively. Also, a knapsack with a weight limit `capacity`. The task is to fill the knapsack in such a way that we can get the maximum profit. Return the maximum profit.

**✅ Constraints**
- 1 <= `wt.length` <= 1000
- 1 <= `val.length` <= 1000
- 1 <= `wt[i]`, `val[i]` <= 1000
- 1 <= `capacity` <= 10^5

**💡 Example**
**Input:**  
`wt = [1, 2, 3]`  
`val = [10, 15, 40]`  
`capacity = 5`

**Output:**  
`50`

**Explanation:**
- The optimal way to fill the knapsack is to take two items of weight 1 and value 10, resulting in a total value of `10 + 10 = 20`.
- Alternatively, taking one item of weight 2 and value 15, and two items of weight 1 and value 10 gives a total value of `15 + 10 + 10 = 50`.

**🛠️ Approaches**

**🔁 1. Dynamic Programming (Bottom-Up Approach)**

Use a 1D DP array `dp[]` where `dp[j]` represents the maximum value achievable with a knapsack of capacity `j`. We iterate through each item and check if it can fit into the knapsack.

```java
class Solution {
    static int knapSack(int val[], int wt[], int capacity) {
        int[] dp = new int[capacity + 1];
        
        for (int i = 0; i < val.length; i++) {
            for (int j = wt[i]; j <= capacity; j++) {
                dp[j] = Math.max(dp[j], val[i] + dp[j - wt[i]]);
            }
        }

        return dp[capacity];
    }
}
```

✅ **Time Complexity:** `O(n * capacity)`  
✅ **Space Complexity:** `O(capacity)`

**📦 2. Recursive + Memoization (Top-Down DP)**
This approach uses recursion with memoization to avoid redundant calculations by storing the results of subproblems.

```java
class Solution {
    static int knapSack(int val[], int wt[], int capacity) {
        Integer[] memo = new Integer[capacity + 1];
        return knapSackRec(val, wt, capacity, memo);
    }
    
    static int knapSackRec(int val[], int wt[], int capacity, Integer[] memo) {
        if (capacity == 0) return 0;
        if (memo[capacity] != null) return memo[capacity];
        
        int maxProfit = 0;
        for (int i = 0; i < val.length; i++) {
            if (wt[i] <= capacity) {
                maxProfit = Math.max(maxProfit, val[i] + knapSackRec(val, wt, capacity - wt[i], memo));
            }
        }
        memo[capacity] = maxProfit;
        return maxProfit;
    }
}
```

✅ **Time Complexity:** `O(n * capacity)`  
✅ **Space Complexity:** `O(capacity)`

**🧠 Summary**
| Approach             | Time Complexity | Space Complexity | Suitable for Large Input |
|----------------------|-----------------|------------------|--------------------------|
| Bottom-Up DP         | O(n * capacity) | O(capacity)      | ✅✅✅                     |
| Top-Down with Memo   | O(n * capacity) | O(capacity)      | ✅✅✅                     |

This problem is a variant of the **Unbounded Knapsack Problem**, where each item can be chosen any number of times.

**Additional Insights:**
- The problem is an extension of the classic knapsack problem, allowing multiple uses of each item. This makes it similar to the "Unbounded Knapsack" problem.
- The solution's time complexity mainly depends on the capacity of the knapsack and the number of items, making dynamic programming a suitable approach.

**Edge Cases:**
- If the `capacity` is 0, the maximum profit is 0 (since no items can be added to the knapsack).
- If the `wt[i]` of all items is greater than the `capacity`, the answer is 0, as no item can fit in the knapsack.


## Problem Statement: Longest Common Subsequence (LCS)

Given two strings `s1` and `s2`, return the length of their longest common subsequence (LCS). If there is no common subsequence, return 0.
A subsequence is a sequence that can be derived from the given string by deleting some or no elements without changing the order of the remaining elements. For example, "ABE" is a subsequence of "ABCDE".

**✅ Constraints**
- 1 <= `s1.length`, `s2.length` <= 1000
- The characters in the strings are lowercase English letters.

**💡 Example**
**Input:**  
`s1 = "abcde"`  
`s2 = "ace"`  

**Output:**  
`3`

**Explanation:**
- The longest common subsequence between "abcde" and "ace" is "ace", which has a length of `3`.

**🛠️ Approaches**

**🔁 1. Dynamic Programming (Bottom-Up Approach)**

Use a 2D DP table `dp[i][j]`, where `dp[i][j]` represents the length of the longest common subsequence of `S1[0...i-1]` and `S2[0...j-1]`.

```java
class Solution {
    static int lcs(String S1, String S2) {
        int m = S1.length();
        int n = S2.length();
        int[][] dp = new int[m + 1][n + 1];
        
        for (int i = 1; i <= m; i++) {
            for (int j = 1; j <= n; j++) {
                if (S1.charAt(i - 1) == S2.charAt(j - 1)) {
                    dp[i][j] = dp[i - 1][j - 1] + 1;
                }
                else {
                    dp[i][j] = Math.max(dp[i - 1][j], dp[i][j - 1]);
                }
            }
        }
        return dp[m][n];
    }
}
```

✅ **Time Complexity:** `O(m * n)` where `m` and `n` are the lengths of `S1` and `S2` respectively.  
✅ **Space Complexity:** `O(m * n)` for the DP table.

**📦 2. Space Optimized Approach (1D DP)**

Since we only need the previous row and current row of the DP table, we can optimize the space complexity by using a 1D array.

```java
class Solution {
    static int lcs(String S1, String S2) {
        int m = S1.length();
        int n = S2.length();
        int[] dp = new int[n + 1];
        
        for (int i = 1; i <= m; i++) {
            int prev = 0;
            for (int j = 1; j <= n; j++) {
                int temp = dp[j];
                if (S1.charAt(i - 1) == S2.charAt(j - 1)) {
                    dp[j] = prev + 1;
                } else {
                    dp[j] = Math.max(dp[j], dp[j - 1]);
                }
                prev = temp;
            }
        }
        return dp[n];
    }
}
```

✅ **Time Complexity:** `O(m * n)`  
✅ **Space Complexity:** `O(n)` (optimized from `O(m * n)`)

**🧠 Summary**
| Approach             | Time Complexity | Space Complexity | Suitable for Large Input |
|----------------------|-----------------|------------------|--------------------------|
| Bottom-Up DP         | O(m * n)        | O(m * n)         | ✅✅✅                     |
| Space Optimized DP   | O(m * n)        | O(n)             | ✅✅✅                     |

This problem is a classic **Dynamic Programming** problem, which is often asked in various coding interviews to assess problem-solving and optimization skills.

**Additional Insights:**
- LCS is used in various applications such as comparing files, bioinformatics (DNA sequencing), and diff algorithms (text comparison).
- The problem can be extended to calculate the LCS of more than two strings.

**Edge Cases:**
- If either `S1` or `S2` is empty, the LCS length is 0.
- If there are no common subsequences, the LCS length will be 0.
- If the two strings are identical, the LCS length is the length of either string.

## Problem Statement: Minimum number of deletions and insertions
Given two strings `s1` and `s2`, the task is to remove or insert the minimum number of characters from/in `s1` to transform it into `s2`. It could be possible that the same character needs to be removed from one point of `s1` and inserted into another point.

**Java Code (Optimized Space DP)**
```java
class Solution {
    static int lcs(String s1, String s2) {
        int m = s1.length();
        int n = s2.length();
        int[][] dp = new int[2][n + 1];

        for (int i = 0; i <= m; i++) {
            int curr = i % 2;

            for (int j = 0; j <= n; j++) {
                if (i == 0 || j == 0)
                    dp[curr][j] = 0;
                else if (s1.charAt(i - 1) == s2.charAt(j - 1))
                    dp[curr][j] = dp[1 - curr][j - 1] + 1;
                else
                    dp[curr][j] = Math.max(dp[1 - curr][j], dp[curr][j - 1]);
            }
        }

        return dp[m % 2][n];
    }

    public int minOperations(String s1, String s2) {
        int m = s1.length();
        int n = s2.length();
        int len = lcs(s1, s2);
        int minDeletions = m - len;
        int minInsertions = n - len;
        return minDeletions + minInsertions;
    }
}
```

**Explanation**
1. **Longest Common Subsequence (LCS)**: We first find the length of the longest common subsequence of `s1` and `s2` using dynamic programming.
2. **Deletions and Insertions**:
   - The number of deletions required is `m - LCS(s1, s2)`, where `m` is the length of `s1`.
   - The number of insertions required is `n - LCS(s1, s2)`, where `n` is the length of `s2`.
3. The total number of operations (deletions + insertions) is the sum of the above two values.

**Time and Space Complexity**
- **Time Complexity:** `O(m * n)` where `m` and `n` are the lengths of the two strings `s1` and `s2`.
- **Space Complexity:** `O(n)` due to the use of a 1D DP array to store the values.

**🛠️ Other Approaches**

**🔁 1. Brute Force Approach (Recursive)**
In this approach, we can use a recursive function to calculate the minimum number of insertions and deletions needed to convert `s1` to `s2`. The recursive approach compares each character and takes one of the following options:
1. If characters match, move to the next character in both strings.
2. If characters don't match, delete from `s1` or insert into `s1` to make the strings match.

```java
class Solution {
    public int minOperations(String s1, String s2) {
        return helper(s1, s2, s1.length(), s2.length());
    }

    private int helper(String s1, String s2, int m, int n) {
        // Base cases
        if (m == 0) return n; // If s1 is empty, we need to insert all characters of s2
        if (n == 0) return m; // If s2 is empty, we need to delete all characters of s1

        // If characters match, no need for any operation
        if (s1.charAt(m - 1) == s2.charAt(n - 1)) {
            return helper(s1, s2, m - 1, n - 1);
        } else {
            // Either delete from s1 or insert into s1
            int delete = helper(s1, s2, m - 1, n);
            int insert = helper(s1, s2, m, n - 1);
            return 1 + Math.min(delete, insert);
        }
    }
}
```

✅ **Time Complexity:** `O(2^(m+n))` (Exponential time complexity)  
⛔ **Space Complexity:** `O(m+n)` (due to recursive stack)

**🔁 2. Tabulation (Bottom-Up DP with Extra Space)**

In this approach, we use a 2D DP table to compute the number of operations needed. The idea is to use dynamic programming to build the solution from smaller subproblems, gradually solving for the full problem.

```java
class Solution {
    public int minOperations(String s1, String s2) {
        int m = s1.length();
        int n = s2.length();
        int[][] dp = new int[m + 1][n + 1];

        // Base cases
        for (int i = 0; i <= m; i++) dp[i][0] = i;
        for (int j = 0; j <= n; j++) dp[0][j] = j;

        // Fill the DP table
        for (int i = 1; i <= m; i++) {
            for (int j = 1; j <= n; j++) {
                if (s1.charAt(i - 1) == s2.charAt(j - 1)) {
                    dp[i][j] = dp[i - 1][j - 1]; // No operation needed
                } else {
                    dp[i][j] = 1 + Math.min(dp[i - 1][j], dp[i][j - 1]); // Either delete or insert
                }
            }
        }

        return dp[m][n];
    }
}
```

✅ **Time Complexity:** `O(m * n)` where `m` and `n` are the lengths of the two strings.  
✅ **Space Complexity:** `O(m * n)` due to the DP table.

**🔁 3. Optimized Space DP (1D DP)**
In this approach, we reduce the space complexity by using only two rows in the DP table. This works because the current row only depends on the previous row and not on further rows.

```java
class Solution {
    public int minOperations(String s1, String s2) {
        int m = s1.length();
        int n = s2.length();
        int[] dp = new int[n + 1];

        // Base case: initialize first row
        for (int j = 0; j <= n; j++) {
            dp[j] = j;
        }

        // Fill the DP array row by row
        for (int i = 1; i <= m; i++) {
            int prev = dp[0]; // Save the previous value for the row transition
            dp[0] = i; // First column is the number of deletions needed

            for (int j = 1; j <= n; j++) {
                int temp = dp[j]; // Save the current value
                if (s1.charAt(i - 1) == s2.charAt(j - 1)) {
                    dp[j] = prev; // No operation needed
                } else {
                    dp[j] = 1 + Math.min(dp[j - 1], dp[j]);
                }
                prev = temp; // Update prev for the next iteration
            }
        }

        return dp[n];
    }
}
```

✅ **Time Complexity:** `O(m * n)`  
✅ **Space Complexity:** `O(n)` (space optimization using a 1D array)

**🧠 Summary of Approaches**

| Approach                    | Time Complexity     | Space Complexity  | Suitable for Large Input |
|-----------------------------|---------------------|-------------------|--------------------------|
| Brute Force (Recursive)     | `O(2^(m+n))`        | `O(m+n)`          | ❌                        |
| Tabulation (2D DP Table)    | `O(m * n)`          | `O(m * n)`        | ❌                        |
| Optimized Space DP (1D DP)  | `O(m * n)`          | `O(n)`             | ✅✅✅                      |

**💡 Conclusion:**
- **Recursive Approach:** Great for understanding the problem, but inefficient for larger inputs due to overlapping subproblems and exponential time complexity.
- **Tabulation (2D DP):** Straightforward but uses a lot of space, making it less efficient for large inputs.
- **Optimized Space DP:** This approach is the most space-efficient while maintaining the same time complexity as the 2D DP table. It is ideal for larger inputs where memory usage is a concern.


## Problem Statement: Shortest Common Supersequence

Given two strings `s1` and `s2`, find the length of the smallest string which has both `s1` and `s2` as its sub-sequences.
- **Note:** `s1` and `s2` can have both uppercase and lowercase English letters.

**Java Code (Dynamic Programming - 2D Table)**
```java
class Solution {
    // Function to find length of shortest common supersequence of two strings.
    public static int shortestCommonSupersequence(String s1, String s2) {
        int m = s1.length();
        int n = s2.length();
        int[][] dp = new int[m + 1][n + 1];
        
        // Build the DP table
        for (int i = 0; i <= m; i++) {
            for (int j = 0; j <= n; j++) {
                if (i == 0)
                    dp[i][j] = j; // If s1 is empty, we need to insert all characters of s2
                else if (j == 0)
                    dp[i][j] = i; // If s2 is empty, we need to insert all characters of s1
                else if (s1.charAt(i - 1) == s2.charAt(j - 1)) 
                    dp[i][j] = 1 + dp[i - 1][j - 1]; // If characters match, no need to insert
                else 
                    dp[i][j] = 1 + Math.min(dp[i - 1][j], dp[i][j - 1]); // Take minimum of insert/remove
            }
        }

        return dp[m][n]; // Return the length of the shortest common supersequence
    }
}
```

**Explanation:**
1. **Base Cases**: 
   - If either of the strings is empty, the length of the shortest common supersequence is simply the length of the other string.
2. **Recursive Relation**:
   - If the characters of `s1` and `s2` match, then no extra character is needed, and we take the diagonal value from the DP table.
   - If they do not match, we take the minimum of either deleting a character from `s1` or `s2` and adding one extra character to the supersequence.
3. **Final Answer**:
   - The final value at `dp[m][n]` represents the length of the shortest common supersequence.

**Time and Space Complexity:**
- **Time Complexity:** `O(m * n)` where `m` and `n` are the lengths of `s1` and `s2`.
- **Space Complexity:** `O(m * n)` due to the DP table.

**🛠️ Other Approaches**
**🔁 1. Brute Force (Recursive)**
In the brute force approach, we use recursion to try all possible combinations of adding characters from both strings. This results in an exponential time complexity due to overlapping subproblems.

```java
class Solution {
    public static int shortestCommonSupersequence(String s1, String s2) {
        return helper(s1, s2, s1.length(), s2.length());
    }

    private static int helper(String s1, String s2, int m, int n) {
        // Base cases
        if (m == 0) return n; // If s1 is empty, return the length of s2
        if (n == 0) return m; // If s2 is empty, return the length of s1

        // If characters match, no operation needed
        if (s1.charAt(m - 1) == s2.charAt(n - 1)) {
            return 1 + helper(s1, s2, m - 1, n - 1);
        } else {
            // Either delete from s1 or s2
            int insert = 1 + helper(s1, s2, m, n - 1);
            int remove = 1 + helper(s1, s2, m - 1, n);
            return Math.min(insert, remove);
        }
    }
}
```

✅ **Time Complexity:** `O(2^(m+n))` (Exponential time complexity due to recursion)  
⛔ **Space Complexity:** `O(m + n)` due to recursive stack.

**🔁 2. Space Optimized DP (1D DP Array)**
The idea here is to optimize space by using a 1D array instead of a 2D DP table. This is possible because each row of the DP table only depends on the previous row.

```java
class Solution {
    public static int shortestCommonSupersequence(String s1, String s2) {
        int m = s1.length();
        int n = s2.length();
        int[] dp = new int[n + 1];

        for (int i = 0; i <= m; i++) {
            int prev = 0; // To store the value of dp[i-1][j-1]
            for (int j = 0; j <= n; j++) {
                int temp = dp[j]; // Store the current dp[j] value for next iteration
                if (i == 0)
                    dp[j] = j; // If s1 is empty, the answer is the length of s2
                else if (j == 0)
                    dp[j] = i; // If s2 is empty, the answer is the length of s1
                else if (s1.charAt(i - 1) == s2.charAt(j - 1))
                    dp[j] = 1 + prev; // Characters match
                else
                    dp[j] = 1 + Math.min(dp[j], dp[j - 1]); // Either insert or remove
                
                prev = temp; // Update prev for next iteration
            }
        }

        return dp[n]; // Final value after the last iteration
    }
}
```

✅ **Time Complexity:** `O(m * n)`  
✅ **Space Complexity:** `O(n)` due to the use of a 1D DP array.

**🧠 Summary of Approaches**
| Approach                    | Time Complexity     | Space Complexity  | Suitable for Large Input |
|-----------------------------|---------------------|-------------------|--------------------------|
| Brute Force (Recursive)     | `O(2^(m+n))`        | `O(m+n)`          | ❌                        |
| Tabulation (2D DP Table)    | `O(m * n)`          | `O(m * n)`        | ❌                        |
| Optimized Space DP (1D DP)  | `O(m * n)`          | `O(n)`             | ✅✅✅                      |

**💡 Conclusion:**
- **Recursive Approach (Brute Force):** Simple to implement but highly inefficient for larger inputs due to exponential time complexity.
- **2D DP Table:** Simple and easy to understand, but uses a lot of space, which can be inefficient for larger inputs.
- **1D DP Array:** This approach is space optimized while maintaining the same time complexity as the 2D DP approach. It is ideal for larger inputs where memory usage is a concern.

## Problem Statement: Longest Common Substring
You are given two strings `s1` and `s2`. Your task is to find the length of the longest common substring among the given strings.

**Java Code (Dynamic Programming - 2D Table)**
```java
class Solution {
    public int longestCommonSubstr(String s1, String s2) {
        int m = s1.length();
        int n = s2.length();
        int[][] LCSuf = new int[m + 1][n + 1];
        int res = 0; 
        
        // Build the DP table
        for (int i = 1; i <= m; i++) {
            for (int j = 1; j <= n; j++) {
                if (s1.charAt(i - 1) == s2.charAt(j - 1)) {
                    LCSuf[i][j] = LCSuf[i - 1][j - 1] + 1; // If characters match, extend the substring
                    res = Math.max(res, LCSuf[i][j]); // Update the result with the maximum length
                } else {
                    LCSuf[i][j] = 0; // Reset the count if characters do not match
                }
            }
        }

        return res; // Return the longest common substring length
    }
}
```

**Explanation:**
1. **Base Case**: 
   - Initialize the DP table `LCSuf` with zeros.
2. **Recursive Relation**:
   - If the characters at `i-1` in `s1` and `j-1` in `s2` are the same, then extend the common substring by 1 and store it in `LCSuf[i][j]`. The result is updated with the maximum substring length.
   - If the characters do not match, reset the length of the current common substring at `LCSuf[i][j]` to `0`.
3. **Final Answer**:
   - The final result is the maximum value in the `LCSuf` table, which corresponds to the length of the longest common substring.

**Time and Space Complexity:**
- **Time Complexity:** `O(m * n)` where `m` and `n` are the lengths of `s1` and `s2`.
- **Space Complexity:** `O(m * n)` for the 2D DP table.

**🛠️ Other Approaches**
**🔁 1. Brute Force (Recursive)**
The brute force solution involves checking all possible substrings of both strings and finding the longest common one. This is inefficient as it takes exponential time and is highly impractical for large inputs.

```java
class Solution {
    public int longestCommonSubstr(String s1, String s2) {
        return helper(s1, s2, s1.length(), s2.length(), 0);
    }

    private static int helper(String s1, String s2, int m, int n, int length) {
        if (m == 0 || n == 0) return length;
        
        if (s1.charAt(m - 1) == s2.charAt(n - 1)) {
            return helper(s1, s2, m - 1, n - 1, length + 1);
        } else {
            return Math.max(length, Math.max(helper(s1, s2, m - 1, n, 0), helper(s1, s2, m, n - 1, 0)));
        }
    }
}
```

✅ **Time Complexity:** `O(2^(m+n))` (Exponential time complexity)  
⛔ **Space Complexity:** `O(m+n)` due to recursive stack.

**🔁 2. Space Optimized DP (1D DP Array)**
Instead of maintaining a full 2D DP table, we can optimize space by using a 1D DP array. The idea is to only keep track of the current and previous row of the DP table, as each row only depends on the row directly above it.

```java
class Solution {
    public int longestCommonSubstr(String s1, String s2) {
        int m = s1.length();
        int n = s2.length();
        int[] dp = new int[n + 1];
        int res = 0; 
        
        // Build the DP table with space optimization
        for (int i = 1; i <= m; i++) {
            int prev = 0; // Variable to store the previous value for transition
            for (int j = 1; j <= n; j++) {
                int temp = dp[j]; // Store the current value for next iteration
                
                if (s1.charAt(i - 1) == s2.charAt(j - 1)) {
                    dp[j] = prev + 1; // Extend the common substring
                    res = Math.max(res, dp[j]); // Update the result with the maximum length
                } else {
                    dp[j] = 0; // Reset the value if characters don't match
                }
                prev = temp; // Update the prev variable for the next iteration
            }
        }

        return res; // Return the longest common substring length
    }
}
```

✅ **Time Complexity:** `O(m * n)`  
✅ **Space Complexity:** `O(n)` for the 1D DP array.

**🧠 Summary of Approaches**
| Approach                    | Time Complexity     | Space Complexity  | Suitable for Large Input |
|-----------------------------|---------------------|-------------------|--------------------------|
| Brute Force (Recursive)     | `O(2^(m+n))`        | `O(m+n)`          | ❌                        |
| Tabulation (2D DP Table)    | `O(m * n)`          | `O(m * n)`        | ❌                        |
| Optimized Space DP (1D DP)  | `O(m * n)`          | `O(n)`             | ✅✅✅                      |

**💡 Conclusion:**
- **Recursive Approach (Brute Force):** Simple to understand but inefficient for larger inputs due to exponential time complexity.
- **2D DP Table:** The most straightforward approach but uses a lot of memory, which might not be feasible for very large inputs.
- **1D DP Array:** Space optimized approach that reduces memory usage while maintaining the time complexity of the 2D DP approach, making it more suitable for large inputs.

## Problem Statement: Longest Palindrome in a String

Given a string `s`, your task is to find the longest palindromic substring within `s`.
A **substring** is a contiguous sequence of characters within a string, defined as `s[i...j]` where `0 ≤ i ≤ j < len(s)`.
A **palindrome** is a string that reads the same forward and backward. More formally, `s` is a palindrome if `reverse(s) == s`.

**Note:** If there are multiple palindromic substrings with the same length, return the first occurrence of the longest palindromic substring from left to right.

**Java Code (Expand Around Center Approach)**
```java
class Solution {
    static String longestPalindrome(String s) {
        int n = s.length();
        if (n == 0) return "";

        int start = 0, maxLen = 1;

        // Iterate through each character in the string
        for (int i = 0; i < n; i++) {

            // Try both odd and even length palindromes
            for (int j = 0; j <= 1; j++) {
                int low = i;
                int high = i + j; // Initialize low and high pointers

                // Expand around the center (low, high) while palindrome condition holds
                while (low >= 0 && high < n && s.charAt(low) == s.charAt(high)) {
                    int currLen = high - low + 1;
                    if (currLen > maxLen) {
                        start = low;
                        maxLen = currLen; // Update the result with the longer palindrome
                    }
                    low--; // Move left pointer outwards
                    high++; // Move right pointer outwards
                }
            }
        }

        // Return the longest palindrome substring
        return s.substring(start, start + maxLen);
    }
}
```

**Explanation:**
1. **Expand Around Center**:
   - The idea is to treat each index (and each pair of consecutive indices) as the center of a potential palindrome.
   - For each possible center (both odd and even length palindromes), expand outwards while the characters on both sides match.
   
2. **Two Types of Centers**:
   - Odd-length palindromes have a single center (e.g., "aba").
   - Even-length palindromes have two centers (e.g., "abba").
   
3. **Updating the Result**:
   - For each valid palindrome, we check if its length is greater than the previously found palindrome. If so, we update the starting index and length of the longest palindrome.
   
4. **Edge Case**:
   - If the input string is empty, return an empty string.

**Time and Space Complexity:**
- **Time Complexity:** `O(n^2)` where `n` is the length of the string, because for each character, we are expanding outwards in both directions (i.e., `O(n)` times for each of `n` centers).
- **Space Complexity:** `O(1)` as we are using only a few integer variables and not requiring additional space proportional to the input size.

**🛠️ Other Approaches**
**🔁 1. Dynamic Programming Approach**
```java
class Solution {
    public String longestPalindrome(String s) {
        int n = s.length();
        if (n == 0) return "";

        boolean[][] dp = new boolean[n][n];
        int start = 0, maxLen = 1;

        // Every single character is a palindrome
        for (int i = 0; i < n; i++) {
            dp[i][i] = true;
        }

        // Check substrings of length 2 and more
        for (int len = 2; len <= n; len++) {
            for (int i = 0; i <= n - len; i++) {
                int j = i + len - 1; // j is the end index of the substring
                if (s.charAt(i) == s.charAt(j)) {
                    if (len == 2 || dp[i + 1][j - 1]) {
                        dp[i][j] = true;
                        if (len > maxLen) {
                            start = i;
                            maxLen = len;
                        }
                    }
                }
            }
        }

        return s.substring(start, start + maxLen);
    }
}
```

**Explanation:**
- We use dynamic programming (DP) to build a 2D table `dp` where `dp[i][j]` represents whether the substring `s[i...j]` is a palindrome.
- The table is filled for substrings of increasing lengths. If `s[i] == s[j]` and the substring `s[i+1...j-1]` is a palindrome, then `s[i...j]` is also a palindrome.
- The final answer is the longest palindrome substring.

**Time and Space Complexity:**
- **Time Complexity:** `O(n^2)` as we are filling a 2D table.
- **Space Complexity:** `O(n^2)` for the 2D DP table.

**🔁 2. Manacher's Algorithm (Optimal Solution)**
Manacher’s algorithm finds the longest palindromic substring in linear time.

```java
class Solution {
    public String longestPalindrome(String s) {
        if (s == null || s.length() == 0) return "";

        StringBuilder sb = new StringBuilder();
        sb.append("#");
        for (char c : s.toCharArray()) {
            sb.append(c);
            sb.append("#");
        }
        
        int n = sb.length();
        int[] p = new int[n]; // Array to store the radius of palindrome around each center
        int center = 0, right = 0; // Initialize the center and right boundary
        
        for (int i = 0; i < n; i++) {
            if (i < right) {
                p[i] = Math.min(right - i, p[2 * center - i]); // Use the mirror property
            }
            // Expand the palindrome around the current center
            while (i - p[i] >= 0 && i + p[i] < n && sb.charAt(i - p[i]) == sb.charAt(i + p[i])) {
                p[i]++;
            }
            // Update the center and right boundary if expanded palindrome goes beyond right
            if (i + p[i] > right) {
                center = i;
                right = i + p[i];
            }
        }
        
        // Find the longest palindrome length and its center
        int maxLen = 0, maxCenter = 0;
        for (int i = 0; i < n; i++) {
            if (p[i] > maxLen) {
                maxLen = p[i];
                maxCenter = i;
            }
        }

        // Extract the substring from the original string
        StringBuilder result = new StringBuilder();
        for (int i = maxCenter - maxLen + 1; i < maxCenter + maxLen; i++) {
            if (sb.charAt(i) != '#') result.append(sb.charAt(i));
        }
        
        return result.toString();
    }
}
```

**Explanation:**
- Manacher’s algorithm transforms the string by inserting a special character (`#`) between each character and at the ends, making it easier to handle both even and odd-length palindromes.
- We then compute the maximum radius of palindromes centered at each character using a dynamic expansion technique.
- The final longest palindrome substring is extracted by locating the center and radius.

**Time and Space Complexity:**
- **Time Complexity:** `O(n)` where `n` is the length of the string, as we only need one pass over the string.
- **Space Complexity:** `O(n)` for the array storing the radius of the palindrome at each center.

**🧠 Summary of Approaches**
| Approach                      | Time Complexity     | Space Complexity  | Suitable for Large Input |
|-------------------------------|---------------------|-------------------|--------------------------|
| Expand Around Center           | `O(n^2)`            | `O(1)`            | ✅                        |
| Dynamic Programming (DP)       | `O(n^2)`            | `O(n^2)`          | ❌                        |
| Manacher’s Algorithm           | `O(n)`              | `O(n)`            | ✅✅✅                      |

**💡 Conclusion:**
- **Expand Around Center:** Simple and efficient in terms of space, but not as fast as Manacher's Algorithm for larger inputs.
- **Dynamic Programming:** Easy to understand and implement, but not efficient in terms of space complexity.
- **Manacher's Algorithm:** The optimal solution with linear time complexity, but slightly more complex to implement.

## Problem Statement: Longest Repeating Subsequence

Given a string `str`, find the length of the longest repeating subsequence such that it can be found twice in the given string.
The two identified subsequences A and B can use the same `i`-th character from string `s` if and only if that `i`-th character has different indices in A and B. For example, A = "xax" and B = "xax" then the index of the first "x" must be different in the original string for A and B.

**Java Code (Dynamic Programming Approach)**

```java
class Solution {
    public int LongestRepeatingSubsequence(String s) {
        int n = s.length();
        int[][] dp = new int[n + 1][n + 1];

        // Fill the DP table
        for (int i = 1; i <= n; i++) {
            for (int j = 1; j <= n; j++) {
                // Check if characters match and indices are different
                if (s.charAt(i - 1) == s.charAt(j - 1) && i != j) {
                    dp[i][j] = 1 + dp[i - 1][j - 1]; // Subsequence found
                } else {
                    dp[i][j] = Math.max(dp[i - 1][j], dp[i][j - 1]); // No subsequence found
                }
            }
        }

        // Return the length of the longest repeating subsequence
        return dp[n][n];
    }
}
```

**Explanation:**
1. **Dynamic Programming Approach:**
   - We use a 2D DP array `dp[i][j]` where `dp[i][j]` represents the length of the longest repeating subsequence in the first `i` characters of `s` and the first `j` characters of `s`.
   - **Base Case:** `dp[0][0] = 0` since an empty string has no repeating subsequences.
   - **State Transition:** If `s[i-1] == s[j-1]` and `i != j`, it means we have found a matching character for a repeating subsequence, so we increment the length of the subsequence by 1 and look at the previous subsequence. Otherwise, we take the maximum of the previous subsequences.
   
2. **Key Constraints:** 
   - The characters must be the same (`s[i-1] == s[j-1]`), but they must come from different indices (`i != j`) to form a valid repeating subsequence.
   
3. **Final Result:**
   - The value at `dp[n][n]` will give us the length of the longest repeating subsequence in the string.

**Time and Space Complexity:**
- **Time Complexity:** `O(n^2)` where `n` is the length of the string. This is because we are filling up a 2D DP table of size `n x n`.
- **Space Complexity:** `O(n^2)` for the 2D DP table.

**🛠️ Other Approaches**

**🔁 1. Space Optimization (Reduced Space Complexity)**

```java
class Solution {
    public int LongestRepeatingSubsequence(String s) {
        int n = s.length();
        int[] prev = new int[n + 1];
        int[] curr = new int[n + 1];

        // Fill the DP table with optimized space
        for (int i = 1; i <= n; i++) {
            for (int j = 1; j <= n; j++) {
                if (s.charAt(i - 1) == s.charAt(j - 1) && i != j) {
                    curr[j] = 1 + prev[j - 1]; // Subsequence found
                } else {
                    curr[j] = Math.max(prev[j], curr[j - 1]); // No subsequence found
                }
            }
            // Update the previous row for the next iteration
            System.arraycopy(curr, 0, prev, 0, n + 1);
        }

        // The final result is in the last cell of the current array
        return curr[n];
    }
}
```

**Explanation:**
- In this approach, we optimize space by using only two arrays (`prev` and `curr`) instead of a full 2D DP table.
- We iterate over the string and update these arrays for each step, copying the current array values into the previous array after each iteration.

**Time and Space Complexity:**
- **Time Complexity:** `O(n^2)` as we are still iterating over the entire string in a nested loop.
- **Space Complexity:** `O(n)` because we only store two arrays of size `n + 1` instead of a 2D table.

**🔁 2. Recursion with Memoization**

```java
class Solution {
    public int LongestRepeatingSubsequence(String s) {
        int n = s.length();
        int[][] memo = new int[n + 1][n + 1];
        
        // Initialize memo array with -1 (indicating uncomputed values)
        for (int[] row : memo) {
            Arrays.fill(row, -1);
        }

        return lrsHelper(s, n, n, memo);
    }
    
    private int lrsHelper(String s, int i, int j, int[][] memo) {
        // Base case: if we've reached the start of the strings, no more subsequences
        if (i == 0 || j == 0) {
            return 0;
        }
        
        // If result is already computed, return it
        if (memo[i][j] != -1) {
            return memo[i][j];
        }

        // If characters match and indices are different
        if (s.charAt(i - 1) == s.charAt(j - 1) && i != j) {
            memo[i][j] = 1 + lrsHelper(s, i - 1, j - 1, memo); // Subsequence found
        } else {
            memo[i][j] = Math.max(lrsHelper(s, i - 1, j, memo), lrsHelper(s, i, j - 1, memo)); // No subsequence found
        }

        return memo[i][j];
    }
}
```

**Explanation:**
- In this approach, we use **recursion with memoization** to store intermediate results and avoid redundant calculations.
- The function `lrsHelper` recursively computes the longest repeating subsequence while using a memoization table (`memo`) to store already computed results for each subproblem.

**Time and Space Complexity:**
- **Time Complexity:** `O(n^2)` as we solve each subproblem once and store the results.
- **Space Complexity:** `O(n^2)` due to the memoization table storing results for each pair of indices.

**🧠 Summary of Approaches**
| Approach                        | Time Complexity     | Space Complexity  | Suitable for Large Input |
|----------------------------------|---------------------|-------------------|--------------------------|
| **Dynamic Programming**          | `O(n^2)`            | `O(n^2)`          | ❌                        |
| **Space Optimization**           | `O(n^2)`            | `O(n)`            | ✅                        |
| **Recursion with Memoization**   | `O(n^2)`            | `O(n^2)`          | ❌                        |

**💡 Conclusion:**
- **Dynamic Programming:** A direct and easy-to-understand solution, but space-intensive.
- **Space Optimization:** Reduces space complexity while keeping the time complexity the same.
- **Recursion with Memoization:** Efficient in terms of space but harder to implement than iterative DP.

## Problem Statement: Edit Distance

Given two strings `s1` and `s2`, return the minimum number of operations required to convert `s1` to `s2`. The possible operations are:
1. Insert a character at any position of the string.
2. Remove any character from the string.
3. Replace any character from the string with any other character.

**Java Code (Dynamic Programming Approach)**
```java
class Solution {
    public int editDistance(String s1, String s2) {
        int m = s1.length();
        int n = s2.length();
        int[][] dp = new int[m + 1][n + 1];

        // Initialize the dp table
        for (int i = 0; i <= m; i++) 
            dp[i][0] = i;  // If s2 is empty, remove all characters from s1
        for (int j = 0; j <= n; j++) 
            dp[0][j] = j;  // If s1 is empty, insert all characters into s1

        // Fill the dp table for the rest of the cases
        for (int i = 1; i <= m; i++) {
            for (int j = 1; j <= n; j++) {
                // If characters match, no operation is needed
                if (s1.charAt(i - 1) == s2.charAt(j - 1)) {
                    dp[i][j] = dp[i - 1][j - 1];
                } else {
                    // Take the minimum of three possible operations: 
                    // Insert, Remove, or Replace
                    dp[i][j] = 1 + Math.min(Math.min(dp[i][j - 1], dp[i - 1][j]), dp[i - 1][j - 1]);
                }
            }
        }

        return dp[m][n];  // The minimum edit distance is in the bottom-right cell of the dp table
    }
}
```

**Explanation:**
1. **Dynamic Programming Approach:**
   - We use a 2D DP table `dp[i][j]` to store the minimum number of operations needed to convert the first `i` characters of `s1` to the first `j` characters of `s2`.
   - **Base Case:** If one of the strings is empty, the only way to convert is by inserting or removing characters. Hence, `dp[i][0] = i` and `dp[0][j] = j`.
   - **State Transition:** 
     - If the characters of `s1[i-1]` and `s2[j-1]` match, no operation is required, so we take the value from the diagonal `dp[i-1][j-1]`.
     - If the characters do not match, we consider three possible operations: 
       - Insert a character (`dp[i][j-1]`)
       - Remove a character (`dp[i-1][j]`)
       - Replace a character (`dp[i-1][j-1]`)
     - The minimum of these three values plus 1 gives us the result.

2. **Final Result:** 
   - The value at `dp[m][n]` gives the minimum number of operations needed to convert `s1` to `s2`.

**Time and Space Complexity:**
- **Time Complexity:** `O(m * n)` where `m` is the length of `s1` and `n` is the length of `s2`. This is because we are filling a 2D DP table of size `m * n`.
- **Space Complexity:** `O(m * n)` for the DP table.

**🛠️ Other Approaches**
**🔁 1. Space Optimization (Reduced Space Complexity)**

```java
class Solution {
    public int editDistance(String s1, String s2) {
        int m = s1.length();
        int n = s2.length();
        int[] prev = new int[n + 1];
        int[] curr = new int[n + 1];

        // Fill the DP table with optimized space
        for (int i = 0; i <= m; i++) {
            for (int j = 0; j <= n; j++) {
                if (i == 0) {
                    curr[j] = j;  // If s1 is empty, insert all characters of s2
                } else if (j == 0) {
                    curr[j] = i;  // If s2 is empty, remove all characters of s1
                } else if (s1.charAt(i - 1) == s2.charAt(j - 1)) {
                    curr[j] = prev[j - 1];  // No operation needed if characters match
                } else {
                    // Take the minimum of three possible operations
                    curr[j] = 1 + Math.min(Math.min(curr[j - 1], prev[j]), prev[j - 1]);
                }
            }
            // Swap the arrays to avoid unnecessary memory allocation
            int[] temp = prev;
            prev = curr;
            curr = temp;
        }

        return prev[n];  // The minimum edit distance is in the last element of the previous array
    }
}
```

**Explanation:**
- In this approach, we optimize space by using only two 1D arrays (`prev` and `curr`) to store the previous and current row of the DP table.
- After processing the current row, we swap the arrays to reuse them, reducing the space complexity.

**Time and Space Complexity:**
- **Time Complexity:** `O(m * n)` as we still iterate through all characters in both strings.
- **Space Complexity:** `O(n)` for the two 1D arrays.

**🔁 2. Recursion with Memoization**
```java
class Solution {
    public int editDistance(String s1, String s2) {
        int m = s1.length();
        int n = s2.length();
        int[][] memo = new int[m + 1][n + 1];
        
        // Initialize memo array with -1 (indicating uncomputed values)
        for (int[] row : memo) {
            Arrays.fill(row, -1);
        }

        return editDistanceHelper(s1, s2, m, n, memo);
    }
    
    private int editDistanceHelper(String s1, String s2, int m, int n, int[][] memo) {
        // Base case: If one string is empty, the result is the length of the other string
        if (m == 0) return n;
        if (n == 0) return m;

        // If the result is already computed, return it
        if (memo[m][n] != -1) {
            return memo[m][n];
        }

        // If characters match, no operation is needed
        if (s1.charAt(m - 1) == s2.charAt(n - 1)) {
            memo[m][n] = editDistanceHelper(s1, s2, m - 1, n - 1, memo);
        } else {
            // Take the minimum of three possible operations and store the result
            memo[m][n] = 1 + Math.min(Math.min(editDistanceHelper(s1, s2, m - 1, n, memo), 
                                               editDistanceHelper(s1, s2, m, n - 1, memo)), 
                                      editDistanceHelper(s1, s2, m - 1, n - 1, memo));
        }

        return memo[m][n];
    }
}
```

**Explanation:**
- In this approach, we use **recursion with memoization** to store intermediate results and avoid redundant calculations.
- The function `editDistanceHelper` recursively computes the minimum number of operations while using a memoization table (`memo`) to store already computed results for each subproblem.

**Time and Space Complexity:**
- **Time Complexity:** `O(m * n)` as we solve each subproblem once and store the results.
- **Space Complexity:** `O(m * n)` due to the memoization table storing results for each pair of indices.

**🧠 Summary of Approaches**
| Approach                        | Time Complexity     | Space Complexity  | Suitable for Large Input |
|----------------------------------|---------------------|-------------------|--------------------------|
| **Dynamic Programming**          | `O(m * n)`          | `O(m * n)`        | ❌                        |
| **Space Optimization**           | `O(m * n)`          | `O(n)`            | ✅                        |
| **Recursion with Memoization**   | `O(m * n)`          | `O(m * n)`        | ❌                        |

**💡 Conclusion:**
- **Dynamic Programming:** The straightforward and easiest approach but requires significant space.
- **Space Optimization:** Reduces space complexity with minimal changes to the algorithm.
- **Recursion with Memoization:** Avoids recomputing overlapping subproblems, but is harder to understand than iterative DP.

## Problem Statement: Distinct Occurrences

You are given two strings `txt` and `pat`, and you are tasked to find the count of distinct occurrences of `pat` as a subsequence in `txt`.
A subsequence is derived from a string by deleting some or no elements, without changing the order of the remaining characters.

**Java Code (Dynamic Programming Approach)**
```java
class Solution {
    public int subseqCount(String txt, String pat) {
        int m = pat.length();
        int n = txt.length();
        if (m > n)
            return 0;
        
        int[][] dp = new int[m + 1][n + 1];

        // Initialization: The empty pattern can be matched in any string in one way
        for (int j = 0; j <= n; j++) {
            dp[0][j] = 1;
        }

        // Filling the DP table
        for (int i = 1; i <= m; i++) {
            for (int j = 1; j <= n; j++) {
                // If characters don't match, carry forward the previous count
                if (pat.charAt(i - 1) != txt.charAt(j - 1)) {
                    dp[i][j] = dp[i][j - 1];
                }
                // If characters match, add the count of the previous subsequence without this character
                else {
                    dp[i][j] = (dp[i][j - 1] + dp[i - 1][j - 1]);
                }
            }
        }

        return dp[m][n];  // Final result is the number of ways the pattern occurs as a subsequence in the text
    }
}
```

**Explanation:**
1. **Dynamic Programming Approach:**
   - We use a 2D DP table `dp[i][j]` where `dp[i][j]` represents the number of ways to match the first `i` characters of `pat` as a subsequence in the first `j` characters of `txt`.
   - **Base Case:** The first row of the DP table (`dp[0][j]`) is initialized to `1` because an empty pattern (`pat`) is a subsequence of any string, including the empty string, in exactly one way.
   - **State Transition:**
     - If the characters of `pat` and `txt` do not match, we carry forward the previous value, meaning we consider the result without including the current character of `txt`.
     - If the characters match, we add the number of ways to include this character (i.e., `dp[i-1][j-1]`) and the number of ways to exclude it (i.e., `dp[i][j-1]`).

2. **Final Result:** 
   - The value at `dp[m][n]` contains the count of distinct occurrences of `pat` as a subsequence in `txt`.

**Time and Space Complexity:**
- **Time Complexity:** `O(m * n)` where `m` is the length of `pat` and `n` is the length of `txt`. This is because we need to fill a DP table of size `m * n`.
- **Space Complexity:** `O(m * n)` for the DP table.

**🛠️ Other Approaches**
**🔁 1. Space Optimization (Reduced Space Complexity)**

```java
class Solution {
    public int subseqCount(String txt, String pat) {

        int n = txt.length();

            return 0;

        int[] dp = new int[n + 1];
        dp[0] = 1;  // An empty pattern can be found in any text in exactly 1 way

        for (int i = 1; i <= m; i++) {
            int[] temp = new int[n + 1];
            for (int j = 1; j <= n; j++) {
                if (pat.charAt(i - 1) != txt.charAt(j - 1)) {
                    temp[j] = dp[j];  // carry forward the previous value
                } else {
                    temp[j] = (dp[j - 1] + dp[j]);  // Add the previous results
                }
            }
            dp = temp;  // Update dp with the new results
        }

        return dp[n];
    }
}
```

**Explanation:**
- This approach reduces the space complexity by using only a single 1D array (`dp`) to store results for the current and previous rows, thus eliminating the need for a full 2D DP table.

**🔁 2. Recursion with Memoization**

```java
class Solution {
    public int subseqCount(String txt, String pat) {
        int m = pat.length();
        int n = txt.length();
        
        int[][] memo = new int[m + 1][n + 1];
        for (int[] row : memo) {
            Arrays.fill(row, -1);  // Initialize memoization table
        }

        return subseqCountHelper(txt, pat, m, n, memo);
    }
    
    private int subseqCountHelper(String txt, String pat, int m, int n, int[][] memo) {
        // Base case: If the pattern is empty, there is exactly one way to match it (the empty subsequence)
        if (m == 0) return 1;
        
        // If the text is empty but pattern is not, no subsequence can be formed
        if (n == 0) return 0;

        // If the result is already computed, return it
        if (memo[m][n] != -1) return memo[m][n];
        
        // If characters match, include this character in the subsequence and check for remaining characters
        if (pat.charAt(m - 1) == txt.charAt(n - 1)) {
            memo[m][n] = subseqCountHelper(txt, pat, m - 1, n - 1, memo) + subseqCountHelper(txt, pat, m, n - 1, memo);
        } else {
            // Otherwise, move to the next character of txt
            memo[m][n] = subseqCountHelper(txt, pat, m, n - 1, memo);
        }
        
        return memo[m][n];
    }
}
```

**Explanation:**
- In this approach, we use **recursion with memoization** to solve the problem. The `subseqCountHelper` function recursively counts the number of subsequences by checking if characters match and either including or excluding the current character.
- The results for subproblems are stored in the `memo` table to avoid redundant computations.

**Time and Space Complexity:**
- **Time Complexity:** `O(m * n)` where `m` is the length of `pat` and `n` is the length of `txt`, due to memoization.
- **Space Complexity:** `O(m * n)` due to the memoization table.

**🧠 Summary of Approaches**

| Approach                        | Time Complexity     | Space Complexity  | Suitable for Large Input |
|----------------------------------|---------------------|-------------------|--------------------------|
| **Dynamic Programming**          | `O(m * n)`          | `O(m * n)`        | ❌                        |
| **Space Optimization**           | `O(m * n)`          | `O(n)`            | ✅                        |
| **Recursion with Memoization**   | `O(m * n)`          | `O(m * n)`        | ❌                        |

**💡 Conclusion:**
- **Dynamic Programming:** A straightforward approach but requires significant space.
- **Space Optimization:** Reduces space complexity to linear, making it more efficient for large inputs.
- **Recursion with Memoization:** A recursive approach that avoids recomputing overlapping subproblems but has higher space complexity due to recursion stack.

## Problem Statement: Wildcard Pattern Matching

Given two strings `pat` and `txt`, which may be of different sizes, you have to return `true` if the wildcard pattern (`pat`) matches with `txt`, else return `false`. All characters of the string `pat` and `txt` always belong to alphanumeric characters.
The wildcard pattern `pat` can include the characters `?` and `*`.
- `?` – matches any single character.
- `*` – matches any sequence of characters (including the empty sequence).

**Note:** The matching should cover the entire `txt` (not partial `txt`).

**Java Code (Dynamic Programming Approach)**
```java
class Solution {
    public boolean wildCard(String txt, String pat) {
        int n = txt.length();
        int m = pat.length();

        boolean[][] dp = new boolean[n + 1][m + 1];

        dp[0][0] = true;

        // If pattern starts with *, the matching should be true for empty text.
        for (int j = 1; j <= m; j++)
            if (pat.charAt(j - 1) == '*')
                dp[0][j] = dp[0][j - 1];

        // Fill the DP table
        for (int i = 1; i <= n; i++) {
            for (int j = 1; j <= m; j++) {
                // If characters match or pattern character is ?, match the previous state
                if (pat.charAt(j - 1) == txt.charAt(i - 1) 
                    || pat.charAt(j - 1) == '?') {
                    dp[i][j] = dp[i - 1][j - 1];
                }
                // If pattern character is *, match either by ignoring * or including the previous text character
                else if (pat.charAt(j - 1) == '*') {
                    dp[i][j] = dp[i][j - 1] || dp[i - 1][j];
                }
            }
        }

        return dp[n][m];  // The last cell will have the final result
    }
}
```

**Explanation:**
1. **Dynamic Programming Approach:**
   - We define a 2D DP table `dp[i][j]` where `dp[i][j]` indicates whether the first `i` characters of `txt` match with the first `j` characters of `pat`.
   - **Base Case:** `dp[0][0] = true`, which means an empty pattern matches an empty text.
   - If the pattern starts with `*`, it can match the empty sequence of text, so we initialize the first row accordingly.
   - **State Transition:**
     - If the current characters of `pat` and `txt` match or the pattern character is `?`, we check the previous state (`dp[i - 1][j - 1]`).
     - If the pattern character is `*`, we either consider it as matching an empty sequence (`dp[i][j - 1]`) or match it with one or more characters (`dp[i - 1][j]`).

2. **Final Result:** 
   - The value at `dp[n][m]` represents whether the entire text (`txt`) matches the pattern (`pat`).

**Time and Space Complexity:**
- **Time Complexity:** `O(n * m)`, where `n` is the length of `txt` and `m` is the length of `pat`. We are filling a DP table of size `n * m`.
- **Space Complexity:** `O(n * m)` for the DP table.

**🛠️ Other Approaches**
**🔁 1. Space Optimization (Reduced Space Complexity)**
```java
class Solution {
    public boolean wildCard(String txt, String pat) {
```

**🔁 2. Recursive Solution (Naive Approach)**
```java
class Solution {
    public boolean wildCard(String txt, String pat) {
        return isMatch(txt, pat, txt.length(), pat.length());
    }

    private boolean isMatch(String txt, String pat, int n, int m) {
        // If both pattern and text are empty, return true
        if (n == 0 && m == 0)
            return true;

        // If pattern is empty but text is not, return false
        if (m == 0)
            return false;

        // If text is empty but pattern is not
        if (n == 0) {
            for (int i = 0; i < m; i++) {
                if (pat.charAt(i) != '*')
                    return false;
            }
            return true;
        }

        // If characters match or pattern has ?, check the next characters
        if (pat.charAt(m - 1) == txt.charAt(n - 1) || pat.charAt(m - 1) == '?') {
            return isMatch(txt, pat, n - 1, m - 1);
        }

        // If pattern has *, we can either:
        // 1. Match the character in the text (skip one character in text)
        // 2. Skip the '*' and check for the next characters in the pattern
        if (pat.charAt(m - 1) == '*') {
            return isMatch(txt, pat, n - 1, m) || isMatch(txt, pat, n, m - 1);
        }

        return false;  // No match
    }
}
```

**Explanation:**
- This is the naive recursive approach where we check each character of `txt` and `pat` and apply the matching rules:
  - If the pattern has `?`, we match it with any character.
  - If the pattern has `*`, we try two possibilities: skipping the `*` or including characters from `txt`.

**Time and Space Complexity:**
- **Time Complexity:** `O(2^(n + m))`, due to the exponential number of recursive calls.
- **Space Complexity:** `O(n + m)` for the recursion stack.

**🧠 Summary of Approaches**
| Approach                        | Time Complexity     | Space Complexity  | Suitable for Large Input |
|----------------------------------|---------------------|-------------------|--------------------------|
| **Dynamic Programming**          | `O(n * m)`          | `O(n * m)`        | ❌                        |
| **Space Optimization**           | `O(n * m)`          | `O(m)`            | ✅                        |
| **Recursive Solution**           | `O(2^(n + m))`      | `O(n + m)`        | ❌                        |

**💡 Conclusion:**
- **Dynamic Programming:** This approach is efficient but requires a 2D table, which may be space-intensive.
- **Space Optimization:** This approach reduces the space complexity by using a 1D array, making it more suitable for large inputs.
- **Recursive Solution:** This approach is simple but inefficient for large inputs due to its exponential time complexity.

## Problem Statement: Longest Valid Parentheses

Given a string `s` consisting of opening and closing parentheses `'('` and `')'`, find the length of the longest valid parenthesis substring.

A parenthesis string is valid if:
- For every opening parenthesis, there is a closing parenthesis.
- The closing parenthesis must be after its opening parenthesis.

**Java Code (Dynamic Programming Approach)**
```java
class Solution {
    static int maxLength(String s) {
        int n = s.length();
        int[] dp = new int[n];
        int maxLen = 0;
        for (int i = 1; i < n; i++) {
            if (s.charAt(i) == ')') {
                if (s.charAt(i - 1) == '(') {
                    if (i >= 2) {
                        dp[i] = dp[i - 2] + 2;
                    } else {
                        dp[i] = 2;
                    }
                }
                else if (i - dp[i - 1] > 0 && s.charAt(i - dp[i - 1] - 1) == '(') {
                    if (i - dp[i - 1] >= 2) {
                        dp[i] = dp[i - 1] + dp[i - dp[i - 1] - 2] + 2;
                    } else {
                        dp[i] = dp[i - 1] + 2;
                    }
                }
                maxLen = Math.max(maxLen, dp[i]);
            }
        }
        return maxLen;
    }
}
```

**Explanation:**
1. **Dynamic Programming Approach:**
   - We maintain an array `dp` where `dp[i]` represents the length of the longest valid parenthesis substring ending at index `i`.
   - If we encounter a closing parenthesis `')'`:
     - If the previous character is `'('`, it forms a valid pair. If the index allows, we add 2 to the previous valid substring length.
     - If the previous character is not `'('` but a valid pair can still be formed with earlier valid substrings, we adjust the value accordingly.
   - We keep track of the maximum length of the valid parentheses substring encountered.

2. **Time and Space Complexity:**
   - **Time Complexity:** `O(n)`, where `n` is the length of the string. We are iterating through the string once.
   - **Space Complexity:** `O(n)` for the `dp` array.

**🛠️ Other Approaches**
**🔁 1. Stack-Based Approach**
```java
class Solution {
    public int maxLength(String s) {
        Stack<Integer> stack = new Stack<>();
        stack.push(-1); // base index for valid parentheses substring
        int maxLen = 0;

        for (int i = 0; i < s.length(); i++) {
            if (s.charAt(i) == '(') {
                stack.push(i);
            } else {
                stack.pop(); // pop the matching '('
                if (!stack.isEmpty()) {
                    maxLen = Math.max(maxLen, i - stack.peek());
                } else {
                    stack.push(i); // if no matching '(', push the current index
                }
            }
        }

        return maxLen;
    }
}
```

**Explanation:**
- A stack is used to keep track of indices of opening parentheses `'('` and matching closing parentheses `')'`.
- Initially, we push `-1` into the stack to handle edge cases.
- When we encounter a closing parenthesis `')'`, we check for a matching opening parenthesis `'('` by popping from the stack. The difference between the current index and the top of the stack gives the length of the valid substring.
- If there is no matching opening parenthesis, we push the current index onto the stack.

**Time and Space Complexity:**
- **Time Complexity:** `O(n)`, where `n` is the length of the string. We iterate through the string once.
- **Space Complexity:** `O(n)` for the stack.

**🔁 2. Two-Pass Approach (Optimized Space)**
```java
class Solution {
    public int maxLength(String s) {
        int n = s.length();
        int left = 0, right = 0;
        int maxLen = 0;

        // Left to Right pass
        for (int i = 0; i < n; i++) {
            if (s.charAt(i) == '(') left++;
            else right++;

            if (left == right) {
                maxLen = Math.max(maxLen, 2 * right);
            } else if (right > left) {
                left = right = 0; // reset when invalid substring is encountered
            }
        }

        left = right = 0;

        // Right to Left pass
        for (int i = n - 1; i >= 0; i--) {
            if (s.charAt(i) == '(') left++;
            else right++;

            if (left == right) {
                maxLen = Math.max(maxLen, 2 * left);
            } else if (left > right) {
                left = right = 0; // reset when invalid substring is encountered
            }
        }

        return maxLen;
    }
}
```

**Explanation:**
- This approach uses two passes over the string, one from left to right and the other from right to left.
- **Left to Right pass:** We count the number of left and right parentheses. When the counts match, we have a valid substring.
- **Right to Left pass:** We do a similar pass but in reverse to handle edge cases where there are more right parentheses than left.

**Time and Space Complexity:**
- **Time Complexity:** `O(n)`, where `n` is the length of the string. We iterate through the string twice.
- **Space Complexity:** `O(1)` as we only use a constant amount of extra space.

**🧠 Summary of Approaches**
| Approach                      | Time Complexity     | Space Complexity | Suitable for Large Input |
|-------------------------------|---------------------|------------------|--------------------------|
| **Dynamic Programming**        | `O(n)`              | `O(n)`           | ❌                        |
| **Stack-Based Approach**       | `O(n)`              | `O(n)`           | ✅                        |
| **Two-Pass Approach**          | `O(n)`              | `O(1)`           | ✅                        |

**💡 Conclusion:**
- **Dynamic Programming:** Efficient but requires extra space for the `dp` array.
- **Stack-Based Approach:** Space-efficient for larger inputs and works in a single pass over the string.
- **Two-Pass Approach:** Optimal in terms of space and is faster than the stack approach, making it ideal for large inputs.

## Problem Statement: Word Break

You are given a string `s` and a list `dictionary[]` of words. Your task is to determine whether the string `s` can be formed by concatenating one or more words from the dictionary.
Note: From `dictionary[]`, any word can be taken any number of times and in any order.

**Java Code (Dynamic Programming Approach)**
```java
class Solution {
    public boolean wordBreak(String s, String[] dictionary) {
        int n = s.length();
        boolean[] dp = new boolean[n + 1];
        dp[0] = true; // Base case: empty string can always be formed

        for (int i = 0; i <= n; i++) {
            for (String w : dictionary) {
                int start = i - w.length();
                if (start >= 0 && dp[start] && s.substring(start, start + w.length()).equals(w)) {
                    dp[i] = true;
                    break; // Found a valid word, no need to check further
                }
            }
        }

        return dp[n]; // Return whether the entire string can be formed
    }
}
```

**Explanation:**
1. **Dynamic Programming Approach:**
   - We define a boolean array `dp[]` where `dp[i]` indicates whether the substring `s[0...i-1]` can be broken into words from the dictionary.
   - Initially, `dp[0] = true` since an empty string can be formed trivially.
   - We iterate through the string and for each position `i`, check if there is a word from the dictionary that can form a valid substring. If a valid word is found, we mark `dp[i]` as `true`.
   - The final result is stored in `dp[n]`, where `n` is the length of the string `s`.

2. **Time and Space Complexity:**
   - **Time Complexity:** `O(n * m * k)`, where `n` is the length of the string `s`, `m` is the number of words in the dictionary, and `k` is the average length of the words in the dictionary. We are checking each word for each substring.
   - **Space Complexity:** `O(n)`, for the `dp` array that stores whether a substring can be broken.

**🛠️ Other Approaches**
**🔁 1. Trie-Based Approach (for Faster Word Lookup)**
```java
class Solution {
    class TrieNode {
        TrieNode[] children = new TrieNode[26];
        boolean isEndOfWord = false;
    }

    public boolean wordBreak(String s, String[] dictionary) {
        TrieNode root = new TrieNode();

        // Build Trie
        for (String word : dictionary) {
            TrieNode node = root;
            for (char c : word.toCharArray()) {
                if (node.children[c - 'a'] == null) {
                    node.children[c - 'a'] = new TrieNode();
                }
                node = node.children[c - 'a'];
            }
            node.isEndOfWord = true;
        }

        // Use dynamic programming with Trie
        int n = s.length();
        boolean[] dp = new boolean[n + 1];
        dp[0] = true;

        for (int i = 1; i <= n; i++) {
            TrieNode node = root;
            for (int j = i - 1; j >= 0; j--) {
                if (dp[j]) {
                    char c = s.charAt(j);
                    if (node.children[c - 'a'] == null) {
                        break;
                    }
                    node = node.children[c - 'a'];
                    if (node.isEndOfWord) {
                        dp[i] = true;
                        break;
                    }
                }
            }
        }

        return dp[n];
    }
}
```

**Explanation:**
- A Trie is built from the dictionary. Each node of the Trie represents a character in a word.
- We use dynamic programming in combination with Trie. For each position in the string, we attempt to match every possible substring with the words in the Trie.
- When a match is found, we mark the corresponding position in the `dp[]` array as `true`.

**Time and Space Complexity:**
- **Time Complexity:** `O(n * m * k)`, where `n` is the length of the string `s`, `m` is the number of words in the dictionary, and `k` is the average length of the words.
- **Space Complexity:** `O(m * k)`, where `m` is the number of words and `k` is the average length of the words in the dictionary.

**🔁 2. Backtracking (Recursive Approach)**
```java
class Solution {
    public boolean wordBreak(String s, String[] dictionary) {
        return wordBreakHelper(s, dictionary, new HashSet<String>());
    }

    private boolean wordBreakHelper(String s, String[] dictionary, Set<String> wordSet) {
        if (s.length() == 0) {
            return true;
        }
        if (wordSet.contains(s)) {
            return false;
        }

        for (String word : dictionary) {
            if (s.startsWith(word)) {
                if (wordBreakHelper(s.substring(word.length()), dictionary, wordSet)) {
                    return true;
                }
            }
        }

        wordSet.add(s);
        return false;
    }
}
```

**Explanation:**
- We use backtracking to recursively check if we can break the string `s` into words from the dictionary.
- Each time we try to break the string, we check if a valid word exists at the beginning of the string. If it does, we recursively try to solve the remaining string.
- If we reach the end of the string, we return `true`, indicating that the string can be formed from the dictionary. If no valid break exists, we return `false`.

**Time and Space Complexity:**
- **Time Complexity:** `O(2^n)` in the worst case, where `n` is the length of the string. This is because we are recursively checking each substring.
- **Space Complexity:** `O(n)`, due to the recursion stack and storing intermediate results in `wordSet`.

**🧠 Summary of Approaches**
| Approach                     | Time Complexity       | Space Complexity    | Suitable for Large Input |
|------------------------------|-----------------------|---------------------|--------------------------|
| **Dynamic Programming**       | `O(n * m * k)`        | `O(n)`              | ❌                        |
| **Trie-Based Approach**      | `O(n * m * k)`        | `O(m * k)`           | ✅                        |
| **Backtracking**             | `O(2^n)`              | `O(n)`               | ❌                        |

**💡 Conclusion:**
- **Dynamic Programming:** Efficient but not ideal for very large inputs as it requires checking many substrings and can be memory intensive.
- **Trie-Based Approach:** Efficient for dictionary-based problems and well-suited for large inputs as it avoids repeated lookups.
- **Backtracking:** Suitable for smaller inputs but not efficient for large strings due to exponential time complexity.

## Problem Statement: Word Break - 2

You are given a string `s` and a dictionary `dict[]` of valid words. You need to return all possible ways to break the string `s` into sentences such that each word in the sentence is a valid dictionary word.
Note: You are allowed to use a valid word multiple times in the sentence.

**Java Code (Dynamic Programming with Backtracking)**
```java
class Solution {
    static String[] wordBreak(String[] dict, String s) {
        // Convert the dictionary to a HashSet for fast lookups
        HashSet<String> st = new HashSet<>(Arrays.asList(dict));
        int n = s.length();
        
        // DP array to store the possible sentences
        List<List<String>> dp = new ArrayList<>();
        for (int i = 0; i <= n; i++) {
            dp.add(new ArrayList<>());
        }
        dp.get(n).add(""); // Base case: empty string can be broken into empty sentence

        // Iterate over the string from end to start
        for (int i = n - 1; i >= 0; --i) {
            for (int j = i + 1; j <= n; ++j) {
                String word = s.substring(i, j);
                
                if (st.contains(word)) {
                    // For each valid word found, add the resulting sentence
                    for (String sub : dp.get(j)) {
                        if (sub.isEmpty()) {
                            dp.get(i).add(word); // Direct word if no subsequent words
                        } else {
                            dp.get(i).add(word + " " + sub); // Add word to the next sentence
                        }
                    }
                }
            }
        }

        return dp.get(0).toArray(new String[0]);
    }
}
```

**Explanation:**
1. **Dynamic Programming with Backtracking:**
   - We define a DP table `dp[i]` where each entry contains a list of sentences that can be formed by breaking the substring `s[i...]` into valid words from the dictionary.
   - Starting from the end of the string, we try to find valid words by slicing the string into substrings. If the substring exists in the dictionary, we add it to the possible sentences for the current position.
   - For each valid word, we recursively add the sentence formed from that word, appending it to the previous valid sentences stored in the DP table.

2. **Base Case:**
   - For the empty substring (from index `n`), we initialize `dp[n]` with an empty string to represent the valid sentence formed by the empty string.

3. **Iterating Through the String:**
   - We iterate over the string in reverse and, for each substring, check if it is in the dictionary. If it is, we add it to the sentences of the subsequent positions in the DP table.

4. **Result:**
   - Finally, the sentences formed by breaking the entire string are stored in `dp[0]`, which we return as the result.

**Time and Space Complexity:**
- **Time Complexity:**  
  `O(n^2 * m)`, where `n` is the length of the string and `m` is the average length of the words in the dictionary. For each substring, we check if it exists in the dictionary, and this process can involve iterating through all possible substrings and words.
  
- **Space Complexity:**  
  `O(n^2)`, due to the DP table that stores possible sentences for each substring.

**🛠️ Other Approaches:**
**🔁 1. Trie-Based Approach**
Using a Trie data structure can help optimize searching for valid words in the dictionary. Each node in the Trie will represent a character, and we can efficiently search for valid substrings by traversing the Trie instead of checking the entire dictionary for each substring.

```java
class Solution {
    class TrieNode {
        TrieNode[] children = new TrieNode[26];
        boolean isEndOfWord = false;
    }

    public String[] wordBreak(String[] dict, String s) {
        TrieNode root = new TrieNode();
        for (String word : dict) {
            TrieNode node = root;
            for (char c : word.toCharArray()) {
                if (node.children[c - 'a'] == null) {
                    node.children[c - 'a'] = new TrieNode();
                }
                node = node.children[c - 'a'];
            }
            node.isEndOfWord = true;
        }

        // Same dynamic programming logic as before, but optimized with Trie
        int n = s.length();
        List<List<String>> dp = new ArrayList<>();
        for (int i = 0; i <= n; i++) {
            dp.add(new ArrayList<>());
        }
        dp.get(n).add("");

        for (int i = n - 1; i >= 0; --i) {
            TrieNode node = root;
            for (int j = i; j < n; ++j) {
                char c = s.charAt(j);
                if (node.children[c - 'a'] == null) {
                    break;
                }
                node = node.children[c - 'a'];
                if (node.isEndOfWord) {
                    for (String sub : dp.get(j + 1)) {
                        dp.get(i).add(s.substring(i, j + 1) + (sub.isEmpty() ? "" : " " + sub));
                    }
                }
            }
        }

        return dp.get(0).toArray(new String[0]);
    }
}
```

**Explanation:**
- We use a Trie to store the dictionary words. The Trie helps us efficiently check if a substring exists in the dictionary.
- The DP approach remains the same, but the word lookup is now optimized with the Trie.

**Time and Space Complexity:**
- **Time Complexity:**  
  `O(n^2 * m)`, where `n` is the length of the string and `m` is the average length of the words. Although the time complexity doesn't change much, the Trie helps reduce redundant checks by directly accessing valid words.
  
- **Space Complexity:**  
  `O(n^2)`, for storing the DP table. The Trie requires additional space proportional to the total number of characters in the dictionary.

**🔁 2. Backtracking (Recursive Approach)**
```java
class Solution {
    public List<String> wordBreak(String s, Set<String> dict) {
        return wordBreakHelper(s, dict, new HashMap<>());
    }

    private List<String> wordBreakHelper(String s, Set<String> dict, Map<String, List<String>> memo) {
        if (memo.containsKey(s)) return memo.get(s);
        List<String> res = new ArrayList<>();
        if (s.length() == 0) {
            res.add("");
            return res;
        }
        for (int i = 1; i <= s.length(); i++) {
            String word = s.substring(0, i);
            if (dict.contains(word)) {
                List<String> subList = wordBreakHelper(s.substring(i), dict, memo);
                for (String sub : subList) {
                    res.add(word + (sub.isEmpty() ? "" : " " + sub));
                }
            }
        }
        memo.put(s, res);
        return res;
    }
}
```

**Explanation:**
- This approach uses a recursive function to break down the string into possible valid words, and memoization is used to store already computed results for substrings.
- Each valid word found is combined with the results of recursively breaking down the remaining substring.

**Time and Space Complexity:**
- **Time Complexity:**  
  `O(n^2)`, as we are recursively breaking the string and checking each substring for validity.
  
- **Space Complexity:**  
  `O(n^2)`, due to memoization and storing the possible sentences.

**🧠 Summary of Approaches**
| Approach                     | Time Complexity       | Space Complexity    | Suitable for Large Input |
|------------------------------|-----------------------|---------------------|--------------------------|
| **Dynamic Programming with Backtracking** | `O(n^2 * m)`        | `O(n^2)`            | ✅                        |
| **Trie-Based Approach**      | `O(n^2 * m)`        | `O(n^2)`             | ✅                        |
| **Backtracking with Memoization** | `O(n^2)`           | `O(n^2)`            | ✅                        |

**💡 Conclusion:**
- **Dynamic Programming with Backtracking:** Efficient and simple to implement. Best when working with relatively smaller strings and dictionaries.
- **Trie-Based Approach:** Optimizes word lookup, especially beneficial when the dictionary is large.
- **Backtracking with Memoization:** Suitable for cases with deep recursion but may not be as efficient for very large input due to exponential recursive calls.

## Problem Statement: Word Wrap

You are given an array `arr[]`, where `arr[i]` denotes the number of characters in one word. Let `k` be the limit on the number of characters that can be put in one line (line width). Put line breaks in the given sequence such that the lines are printed neatly.
Assume that the length of each word is smaller than the line width. When line breaks are inserted, there is a possibility that extra spaces are present in each line. The extra spaces include spaces put at the end of every line except the last one.
You have to minimize the following total cost where total cost = Sum of cost of all lines, where cost of line is = (Number of extra spaces in the line)².

**Java Code (Dynamic Programming Approach)**
```java
class Solution {
    public int solveWordWrap(int[] arr, int k) {
        // Code here
        int n = arr.length;
        int[] dp = new int[n + 1];
        Arrays.fill(dp, Integer.MAX_VALUE);
        dp[n] = 0; // Base case: No extra spaces when no words are left

        // Iterate backward through the words
        for (int curr = n - 1; curr >= 0; curr--) {
            int sum = 0;
            // Try placing each word on the current line
            for (int i = curr; i < n; i++) {
                sum += arr[i]; // Sum the word lengths
                int tot = sum + (i - curr); // Total characters including spaces
                // Break if the total characters exceed the line width
                if (tot > k) break;
                // If last word in the string, no cost
                if (i == n - 1) {
                    dp[curr] = Math.min(dp[curr], 0);
                }
                else {
                    int cost = (k - tot) * (k - tot); // Cost for this line
                    dp[curr] = Math.min(dp[curr], cost + dp[i + 1]); // Minimize the total cost
                }
            }
        }
        return dp[0]; // The minimum cost starting from the first word
    }
}
```

**Explanation:**
1. **Dynamic Programming (DP) Table:**
   - We use a DP array `dp[]` where `dp[i]` represents the minimum cost to arrange words starting from the `i`-th word onward.
   - We initialize `dp[n] = 0` as the base case, where no more words are left to arrange.

2. **Iterating Over Words:**
   - We iterate backward through the words. For each word `curr`, we try to place subsequent words in the same line and calculate the total number of characters (including spaces between words).
   - If the total number of characters exceeds the line width `k`, we break out of the loop.

3. **Cost Calculation:**
   - If a line fits, we calculate the "extra space" (the remaining space in the line after placing the words), and then the cost is the square of the extra space.
   - We update `dp[curr]` to be the minimum of its current value and the cost for this line plus the cost of arranging the remaining words.

4. **Result:**
   - The result `dp[0]` gives us the minimum cost for arranging the entire string of words.

**Time and Space Complexity:**
- **Time Complexity:**  
  `O(n^2)`, where `n` is the number of words in the input array. For each word, we potentially look at every subsequent word to calculate costs, leading to a nested loop.

- **Space Complexity:**  
  `O(n)`, due to the DP array used to store the minimum costs for each starting index in the word array.

**🛠️ Other Approaches:**
**🔁 1. Greedy Approach (Not optimal for cost minimization)**
In a greedy approach, you could attempt to place as many words as possible on each line without considering the cost of extra spaces. However, this approach may not result in the minimum cost because it ignores the possibility of extra spaces being squared.

```java
class Solution {
    public int solveWordWrap(int[] arr, int k) {
        int n = arr.length;
        int cost = 0, currentLineLength = 0;

        for (int i = 0; i < n; i++) {
            if (currentLineLength + arr[i] > k) {
                // If the word doesn't fit in the current line, calculate the cost for this line
                cost += Math.pow(k - currentLineLength, 2);
                // Start a new line
                currentLineLength = arr[i];
            } else {
                currentLineLength += arr[i];
            }

            // Add space after each word except the last one
            if (i < n - 1) {
                currentLineLength++;
            }
        }

        // Add the cost for the last line
        cost += Math.pow(k - currentLineLength, 2);

        return cost;
    }
}
```

**Explanation:**
- This greedy approach tries to place as many words as possible on the current line without considering the extra spaces on the last line.
- Once a word doesn't fit in the current line, it moves to the next line, calculating the cost of the previous line.

**Time and Space Complexity:**
- **Time Complexity:** `O(n)`, where `n` is the number of words.
- **Space Complexity:** `O(1)`.

**🔁 2. Backtracking Approach (Less Efficient)**
A backtracking approach could explore all possible ways of placing the words into lines and then calculate the cost for each valid placement. This approach would be very inefficient due to its exponential time complexity.

```java
class Solution {
    public int solveWordWrap(int[] arr, int k) {
        return backtrack(arr, k, 0, 0, 0);
    }

    private int backtrack(int[] arr, int k, int index, int currentLineLength, int totalCost) {
        if (index == arr.length) {
            return totalCost;
        }

        int minCost = Integer.MAX_VALUE;

        // Try adding the current word to the current line
        if (currentLineLength + arr[index] <= k) {
            minCost = Math.min(minCost, backtrack(arr, k, index + 1, currentLineLength + arr[index] + 1, totalCost));
        }

        // Try starting a new line
        int extraSpaces = k - currentLineLength;
        minCost = Math.min(minCost, backtrack(arr, k, index + 1, arr[index], totalCost + extraSpaces * extraSpaces));

        return minCost;
    }
}
```

**Explanation:**
- This backtracking approach explores all possible ways to arrange the words into lines. For each arrangement, it calculates the cost based on the extra spaces in each line.
- This approach is highly inefficient due to its exponential time complexity.

**Time and Space Complexity:**
- **Time Complexity:** `O(2^n)`, where `n` is the number of words. There are 2 choices for each word: either continue on the current line or start a new one.
- **Space Complexity:** `O(n)`, due to the recursion stack.

**🧠 Summary of Approaches**
| Approach                      | Time Complexity      | Space Complexity     | Suitable for Large Input |
|-------------------------------|----------------------|----------------------|--------------------------|
| **Dynamic Programming**        | `O(n^2)`             | `O(n)`               | ✅                       |
| **Greedy Approach**            | `O(n)`               | `O(1)`               | ✅                       |
| **Backtracking Approach**      | `O(2^n)`             | `O(n)`               | ❌                       |

**💡 Conclusion:**
- **Dynamic Programming Approach**: Best for minimizing the cost. Optimal solution with a time complexity of `O(n^2)`.
- **Greedy Approach**: Simple and fast but does not guarantee the minimum cost.
- **Backtracking Approach**: Highly inefficient for large inputs.


## Problem Statement: Interleaved Strings
Given three strings `s1`, `s2`, and `s3`, determine if `s3` is formed by interleaving `s1` and `s2`.

**Conditions for interleaving:**
- Characters from `s1` must appear in the same order in `s3` as they are in `s1`.
- Characters from `s2` must appear in the same order in `s3` as they are in `s2`.
- The length of `s3` must be equal to the combined length of `s1` and `s2`.

**Java Code (Dynamic Programming Approach)**
```java
class Solution {
    public boolean isInterLeave(String s1, String s2, String s3) {
        // Code here
        int m = s1.length();
        int n = s2.length();
        
        // If length of s3 is not equal to the combined length of s1 and s2, return false
        if (m + n != s3.length())
            return false;

        boolean[][] dp = new boolean[m + 1][n + 1];
        dp[0][0] = true;

        // Fill the first row (when s1 is empty)
        for (int j = 1; j <= n; ++j) {
            dp[0][j] = (s2.charAt(j - 1) == s3.charAt(j - 1)) && dp[0][j - 1];
        }

        // Fill the first column (when s2 is empty)
        for (int i = 1; i <= m; ++i) {
            dp[i][0] = (s1.charAt(i - 1) == s3.charAt(i - 1)) && dp[i - 1][0];
        }

        // Fill the rest of the dp table
        for (int i = 1; i <= m; ++i) {
            for (int j = 1; j <= n; ++j) {
                int k = i + j;
                dp[i][j] = (s1.charAt(i - 1) == s3.charAt(k - 1) && dp[i - 1][j])
                           || (s2.charAt(j - 1) == s3.charAt(k - 1) && dp[i][j - 1]);
            }
        }

        // The answer is whether we can form s3 using all of s1 and s2
        return dp[m][n];
    }
}
```

**Explanation:**
1. **Dynamic Programming (DP) Table:**
   - We define a 2D DP table `dp[i][j]` that represents whether it is possible to form the first `i` characters of `s1` and the first `j` characters of `s2` in the first `i + j` characters of `s3`.
   
2. **Base Case:**
   - We initialize `dp[0][0]` to `true`, meaning an empty `s1` and `s2` can form an empty `s3`.

3. **First Row and First Column:**
   - For the first row (`dp[0][j]`), we check if the characters from `s2` match the corresponding characters in `s3`.
   - For the first column (`dp[i][0]`), we check if the characters from `s1` match the corresponding characters in `s3`.

4. **Filling the DP Table:**
   - For each position `dp[i][j]`, we check if:
     - The current character of `s1` matches the corresponding character in `s3` and if the previous state `dp[i-1][j]` is `true`.
     - Or, the current character of `s2` matches the corresponding character in `s3` and if the previous state `dp[i][j-1]` is `true`.

5. **Result:**
   - The value at `dp[m][n]` gives the final result, whether it is possible to interleave `s1` and `s2` to form `s3`.

**Time and Space Complexity:**
- **Time Complexity:**  
  `O(m * n)`, where `m` and `n` are the lengths of `s1` and `s2`. We need to fill a 2D DP table of size `(m+1) x (n+1)`.

- **Space Complexity:**  
  `O(m * n)` for the DP table.

**🛠️ Other Approaches:**
**🔁 1. Recursive Approach (Naive)**

A naive recursive approach could be used to check whether a substring of `s3` can be formed by interleaving substrings of `s1` and `s2`. However, this approach would be inefficient as it would result in many overlapping subproblems, making it have exponential time complexity.

```java
class Solution {
    public boolean isInterLeave(String s1, String s2, String s3) {
        return isInterleaveHelper(s1, s2, s3, 0, 0, 0);
    }

    private boolean isInterleaveHelper(String s1, String s2, String s3, int i, int j, int k) {
        // If we've used all characters in both s1 and s2, return true
        if (i == s1.length() && j == s2.length() && k == s3.length()) {
            return true;
        }

        boolean res = false;

        // Check if the current character in s1 matches s3
        if (i < s1.length() && s1.charAt(i) == s3.charAt(k)) {
            res = res || isInterleaveHelper(s1, s2, s3, i + 1, j, k + 1);
        }

        // Check if the current character in s2 matches s3
        if (j < s2.length() && s2.charAt(j) == s3.charAt(k)) {
            res = res || isInterleaveHelper(s1, s2, s3, i, j + 1, k + 1);
        }

        return res;
    }
}
```

**Explanation:**
- This approach checks all possible ways of interleaving `s1` and `s2` to form `s3`. For each index, it recursively checks if the current character of `s1` or `s2` matches `s3` and moves to the next character accordingly.

**Time and Space Complexity:**
- **Time Complexity:**  
  `O(2^(m + n))`, due to the recursive calls.
  
- **Space Complexity:**  
  `O(m + n)` due to the recursion stack.

**🔁 2. Optimized Space DP Approach**
Instead of using a full 2D DP table, we can optimize the space complexity by using a 1D array, where `dp[j]` stores whether it is possible to form the first `j` characters of `s2` using the first `i` characters of `s1` and `s2` in `s3`.

```java
class Solution {
    public boolean isInterLeave(String s1, String s2, String s3) {
        int m = s1.length();
        int n = s2.length();
        
        if (m + n != s3.length()) return false;

        boolean[] dp = new boolean[n + 1];
        dp[0] = true;

        // Fill the first row (when s1 is empty)
        for (int j = 1; j <= n; ++j) {
            dp[j] = dp[j - 1] && s2.charAt(j - 1) == s3.charAt(j - 1);
        }

        // Fill the rest of the DP table
        for (int i = 1; i <= m; ++i) {
            dp[0] = dp[0] && s1.charAt(i - 1) == s3.charAt(i - 1);
            for (int j = 1; j <= n; ++j) {
                dp[j] = (dp[j] && s1.charAt(i - 1) == s3.charAt(i + j - 1))
                        || (dp[j - 1] && s2.charAt(j - 1) == s3.charAt(i + j - 1));
            }
        }

        return dp[n];
    }
}
```

**Explanation:**
- This approach reduces space complexity by using a 1D array instead of a 2D array. The result is stored in the last entry of the array, which stores whether it's possible to form the first `n` characters of `s2` using both `s1` and `s2` for all positions.

**Time and Space Complexity:**
- **Time Complexity:**  
  `O(m * n)` where `m` and `n` are the lengths of `s1` and `s2`.

- **Space Complexity:**  
  `O(n)` due to the 1D DP array.

**🧠 Summary of Approaches**
| Approach                    | Time Complexity      | Space Complexity     | Suitable for Large Input |
|-----------------------------|----------------------|----------------------|--------------------------|
| **Dynamic Programming (2D)** | `O(m * n)`           | `O(m * n)`           | ✅                       |
| **Recursive (Naive)**        | `O(2^(m + n))`       | `O(m + n)`           | ❌                       |
| **Optimized DP (1D)**        | `O(m * n)`           | `O(n)`               | ✅                       |

**💡 Conclusion:**
- **Dynamic Programming (2D)**: Best for clarity and understanding, but uses more space.
- **Recursive Approach**: Simple but inefficient due to overlapping subproblems.
- **Optimized DP (1D)**: Efficient in terms of space, optimal for large inputs.


## Problem Statement: Matrix Chain Multiplication

You are given an array `arr[]` representing the dimensions of a sequence of matrices, where the `i`th matrix has the dimensions `(arr[i-1] x arr[i])` for `i >= 1`. Your task is to find the most efficient way to multiply these matrices together. The efficient way is the one that involves the least number of scalar multiplications.

**Java Code (Recursive with Memoization Approach)**
```java
class Solution {
     static int minMultRec(int[] arr, int i, int j,
                          int[][] memo)
    {
        if (i + 1 == j)
            return 0;
        if (memo[i][j] != -1)
            return memo[i][j];

        int res = Integer.MAX_VALUE;
        for (int k = i + 1; k < j; k++) {
            int curr = minMultRec(arr, i, k, memo)
                       + minMultRec(arr, k, j, memo)
                       + arr[i] * arr[k] * arr[j];

            res = Math.min(curr, res);
        }
        memo[i][j] = res;
        return res;
    }

    static int matrixMultiplication(int arr[]) {
        // code here
        int n = arr.length;
        int[][] memo = new int[n][n];
        for (int[] row : memo)
            Arrays.fill(row, -1);

        return minMultRec(arr, 0, n - 1, memo);
    }
}
```

**Explanation:**
1. **Recursive Approach:**
   - The problem is recursively solved by trying all possible places to split the matrix chain and calculating the minimum cost of multiplying matrices from index `i` to `j`.
   - For a given range `[i, j]`, the recursion splits the problem into two parts: `[i, k]` and `[k+1, j]`, where `k` is a possible place to split the chain.
   - The cost of multiplying these matrices is then computed and compared to find the minimum.

2. **Memoization:**
   - To optimize the recursion, we use memoization to store the results of subproblems. This avoids recalculating the same subproblem multiple times.
   - A 2D array `memo[][]` is used to store the minimum number of scalar multiplications for a given subproblem `(i, j)`.

3. **Base Case:**
   - The base case is when `i + 1 == j`. In this case, there is only one matrix, so no multiplication is needed, and the cost is 0.

4. **Recursive Case:**
   - For each possible split point `k`, the cost is calculated as the sum of multiplying the matrices from `i` to `k`, the matrices from `k+1` to `j`, and the cost of multiplying the resulting two matrices together.
   - The minimum cost is stored and returned.

5. **Final Result:**
   - The result is stored in `memo[0][n-1]`, which gives the minimum number of scalar multiplications required to multiply all matrices.

**Time and Space Complexity:**
- **Time Complexity:**
  - The time complexity is `O(n^3)` where `n` is the number of matrices. This is because for each pair of `i` and `j`, we iterate over `k` from `i+1` to `j-1`, leading to a cubic complexity in terms of `n`.

- **Space Complexity:**
  - The space complexity is `O(n^2)` due to the memoization table `memo[][]`, which stores the results of all subproblems.

**🛠️ Other Approaches:**
**🔁 1. Dynamic Programming (Bottom-Up Approach)**

Instead of using recursion and memoization, we can solve this problem iteratively using dynamic programming. The idea is to fill a DP table that stores the minimum cost of multiplying matrices from index `i` to `j`.

```java
class Solution {
    static int matrixMultiplication(int arr[]) {
        int n = arr.length;
        int[][] dp = new int[n][n];

        // Fill the DP table
        for (int length = 2; length < n; length++) {
            for (int i = 0; i < n - length; i++) {
                int j = i + length;
                dp[i][j] = Integer.MAX_VALUE;

                for (int k = i + 1; k < j; k++) {
                    dp[i][j] = Math.min(dp[i][j], dp[i][k] + dp[k][j] + arr[i] * arr[k] * arr[j]);
                }
            }
        }

        return dp[0][n - 1];
    }
}
```

**Explanation:**
1. **Dynamic Programming Table:**
   - The `dp[i][j]` table stores the minimum number of scalar multiplications required to multiply matrices from `i` to `j`.

2. **Iterative Calculation:**
   - The outer loop iterates over possible lengths of matrix chains, starting from length 2 to `n-1`.
   - For each length, the second loop calculates the cost of multiplying matrices between indices `i` and `j`, and the third loop checks all possible split points `k` to compute the minimum multiplication cost.

3. **Result:**
   - The final result is stored in `dp[0][n-1]`, which gives the minimum number of multiplications required to multiply all matrices in the chain.

**Time and Space Complexity:**
- **Time Complexity:**
  - The time complexity is `O(n^3)` as we use three nested loops to fill the DP table.

- **Space Complexity:**
  - The space complexity is `O(n^2)` due to the `dp[][]` table.

**🧠 Summary of Approaches**

| Approach                          | Time Complexity      | Space Complexity     | Suitable for Large Input |
|-----------------------------------|----------------------|----------------------|--------------------------|
| **Recursive with Memoization**    | `O(n^3)`             | `O(n^2)`             | ✅                       |
| **Dynamic Programming (Bottom-Up)** | `O(n^3)`             | `O(n^2)`             | ✅                       |

**💡 Conclusion:**
- **Recursive with Memoization**: Clear approach, but not the most efficient for very large inputs due to recursion overhead.
- **Dynamic Programming (Bottom-Up)**: More efficient and avoids recursion stack overflow. It is optimal for large inputs.

## Problem Statement: MaxSkill

You are given a queue of `n` people indexed from `0` to `n-1`. Each person has a rating represented by an array `arr[]`. You are asked to remove all the persons from the queue. If you remove the `i`-th person from the queue, you gain a skill value of `arr[i - 1] * arr[i] * arr[i + 1]`. Return the maximum total skill you can obtain by removing the people optimally.

**Constraints:**
- If `i - 1` or `i + 1` is out of bounds, assume there is an implicit person with a rating of `1` at that position.

**Java Code (Dynamic Programming Approach)**

```java
class Solution {
    public static int maxSkill(int[] arr) {
        // code here
        int n = arr.length;
        int[] nums = new int[n + 2];
        nums[0] = 1;
        nums[n + 1] = 1;
        for (int i = 0; i < n; i++) {
            nums[i + 1] = arr[i];
        }
        
        int[][] dp = new int[n + 2][n + 2];
        
        for (int length = 1; length <= n; length++) {
            for (int left = 1; left <= n - length + 1; left++) {
                int right = left + length - 1;
                
                for (int k = left; k <= right; k++) {
                    int skill = nums[left - 1] * nums[k] * nums[right + 1];
                    int totalSkill = skill + dp[left][k - 1] + dp[k + 1][right];
                    dp[left][right] = Math.max(dp[left][right], totalSkill);
                }
            }
        }
        
        return dp[1][n];
    }
}
```

**Explanation:**

1. **Initial Setup:**
   - We are given an array `arr[]` which represents the ratings of `n` people in the queue. We extend this array by adding two `1`s at the start and end (to account for the out-of-bounds cases where the person is at the start or end of the queue).
   - This new array `nums[]` is of size `n + 2`.

2. **Dynamic Programming Approach:**
   - We define a 2D array `dp[][]` where `dp[left][right]` represents the maximum skill obtained by removing all people in the subarray from `left` to `right`.
   - We iterate through all possible subarrays of people, and for each subarray, we try removing every person `k` between `left` and `right`, calculating the skill value for removing `k` and combining it with the results of the subproblems (i.e., the maximum skill for the left and right parts after removing `k`).

3. **Skill Calculation:**
   - The skill gained by removing the person at index `k` is `nums[left - 1] * nums[k] * nums[right + 1]`, where `nums[left - 1]` and `nums[right + 1]` are the ratings of the implicit persons at the boundaries.
   - We then update the `dp[left][right]` with the maximum skill achieved by removing the person at index `k`.

4. **Final Result:**
   - The result is stored in `dp[1][n]`, which represents the maximum total skill that can be obtained by removing all people from the queue.

**Time and Space Complexity:**

- **Time Complexity:** `O(n^3)`
  - We have three nested loops:
    1. The outer loop iterates over all possible lengths of subarrays.
    2. The second loop iterates over the possible `left` positions of subarrays.
    3. The innermost loop iterates over all possible `k` values within the subarray, calculating the maximum possible skill for each subarray.
  
- **Space Complexity:** `O(n^2)`
  - We use a 2D dynamic programming table `dp[][]` to store the results for all subarrays, which requires `O(n^2)` space.

**🛠️ Other Approaches:**
1. **Greedy Approach (Not Optimal but Simplified)**

In this approach, we attempt to remove the person with the highest possible skill (based on the current state) at each step. This approach doesn't guarantee an optimal solution but can work for smaller cases or as a heuristic.

```java
class Solution {
    public static int maxSkill(int[] arr) {
        int n = arr.length;
        int[] nums = new int[n + 2];
        nums[0] = 1;
        nums[n + 1] = 1;
        
        // Extending the array to include implicit people with a rating of 1 at the boundaries
        for (int i = 0; i < n; i++) {
            nums[i + 1] = arr[i];
        }
        
        int totalSkill = 0;
        boolean[] removed = new boolean[n];  // Keeps track of removed persons
        
        while (true) {
            int maxSkill = Integer.MIN_VALUE;
            int personToRemove = -1;
            
            // Try removing each person and calculate the current skill gain
            for (int i = 0; i < n; i++) {
                if (removed[i]) continue;  // Skip removed persons
                
                int left = (i - 1 >= 0) ? nums[i] : 1;
                int right = (i + 1 < n) ? nums[i] : 1;
                
                int skill = nums[i - 1] * nums[i] * nums[i + 1];
                if (skill > maxSkill) {
                    maxSkill = skill;
                    personToRemove = i;
                }
            }
            
            if (personToRemove == -1) break;  // All persons removed
            
            removed[personToRemove] = true;
            totalSkill += maxSkill;
        }
        
        return totalSkill;
    }
}
```

**Explanation:**
- **Greedy Method**: At each step, we try to remove the person who gives the maximum skill value. 
- This approach is simpler, but it doesn't always provide the optimal solution since removing the person with the highest local skill might not lead to the best global outcome. 

2. **Brute Force Approach (Not Efficient)**

In this approach, we could try all possible ways of removing people and calculate the total skill for each configuration. However, this approach has exponential time complexity (`O(2^n)`), so it is infeasible for large inputs.

```java
class Solution {
    public static int maxSkill(int[] arr) {
        int n = arr.length;
        int[] nums = new int[n + 2];
        nums[0] = 1;
        nums[n + 1] = 1;
        
        // Extending the array to include implicit people with a rating of 1 at the boundaries
        for (int i = 0; i < n; i++) {
            nums[i + 1] = arr[i];
        }
        
        // Brute force: Check all subsets of people
        int maxSkill = 0;
        int totalSubsets = 1 << n;  // 2^n subsets
        
        for (int mask = 0; mask < totalSubsets; mask++) {
            int skill = 0;
            boolean[] removed = new boolean[n];
            
            for (int i = 0; i < n; i++) {
                if ((mask & (1 << i)) != 0) {  // If the i-th person is removed
                    removed[i] = true;
                    skill += nums[i - 1] * nums[i] * nums[i + 1];
                }
            }
            
            maxSkill = Math.max(maxSkill, skill);
        }
        
        return maxSkill;
    }
}
```

**Explanation:**
- **Brute Force**: This approach tries all possible combinations of removed people. For each combination, it calculates the total skill.
- The time complexity is `O(2^n)`, as we are considering all possible subsets of the queue. This becomes impractical for large inputs.

3. **Dynamic Programming (Bottom-Up Approach)**
   - The provided solution uses dynamic programming to solve the problem efficiently by breaking the problem into smaller subproblems and building up the solution from smaller subarrays to the full problem. This approach ensures we calculate the maximum possible skill by considering all possible ways of removing people.

**🧠 Summary of Approaches**

| Approach                          | Time Complexity      | Space Complexity     | Suitable for Large Input |
|-----------------------------------|----------------------|----------------------|--------------------------|
| **Dynamic Programming (Bottom-Up)** | `O(n^3)`             | `O(n^2)`             | ✅                       |
| **Greedy**                         | `O(n^2)` (approx)    | `O(1)`                | ❌                       |
| **Brute Force**                    | `O(2^n)`             | `O(1)`                | ❌                       |

**💡 Conclusion:**
- **Dynamic Programming**: The optimal approach for this problem, which guarantees the best solution by considering all subproblems efficiently.
- **Greedy** and **Brute Force**: Not suitable due to their inefficiency and inability to guarantee the optimal solution.

## Problem Statement: Boolean Parenthesization

You are given a boolean expression `s` containing:
- 'T' --> true
- 'F' --> false 
and following operators between symbols:
- `&` --> boolean AND
- `|` --> boolean OR
- `^` --> boolean XOR

Count the number of ways we can parenthesize the expression such that the value of the expression evaluates to `true`.

**Note:**
- The answer is guaranteed to fit within a 32-bit integer.

**Example:**
**Input:**
- s = "T|F&T^F"

**Output:**
- 4

**Explanation:**
There are 4 ways to parenthesize the expression to make it evaluate to `true`.

**Approaches:**
**1. **Dynamic Programming with Memoization (Top-Down Approach)**

This approach involves using recursion with memoization to break down the problem into smaller subproblems. Each subproblem counts the number of ways to parenthesize the subexpression such that it evaluates to either true or false.

**Code Implementation:**

```java
class Solution {
    static boolean evaluate(int b1, int b2, char op) {
        if (op == '&') {
            return (b1 & b2) == 1;
        }
        else if (op == '|') {
            return (b1 | b2) == 1;
        }
        return (b1 ^ b2) == 1;
    }

    static int countRecur(int i, int j, int req, String s, int[][][] memo) {
        if (i == j) {
            return (req == (s.charAt(i) == 'T' ? 1 : 0)) ? 1 : 0;
        }
        if (memo[i][j][req] != -1) {
            return memo[i][j][req];
        }

        int ans = 0;
        for (int k = i + 1; k < j; k += 1) {
            int leftTrue = countRecur(i, k - 1, 1, s, memo);
            int leftFalse = countRecur(i, k - 1, 0, s, memo);
            int rightTrue = countRecur(k + 1, j, 1, s, memo);
            int rightFalse = countRecur(k + 1, j, 0, s, memo);
                
            if (evaluate(1, 1, s.charAt(k)) == (req == 1)) {
                ans += leftTrue * rightTrue;
            }
            if (evaluate(1, 0, s.charAt(k)) == (req == 1)) {
                ans += leftTrue * rightFalse;
            }
            if (evaluate(0, 1, s.charAt(k)) == (req == 1)) {
                ans += leftFalse * rightTrue;
            }
            if (evaluate(0, 0, s.charAt(k)) == (req == 1)) {
                ans += leftFalse * rightFalse;
            }
        }

        return memo[i][j][req] = ans;
    }

    static int countWays(String s) {
        int n = s.length();
        int[][][] memo = new int[n][n][2];
        for (int[][] mat : memo) {
            for (int[] row : mat) {
                Arrays.fill(row, -1);
            }
        }
        return countRecur(0, n - 1, 1, s, memo);
    }
}
```

**Explanation:**

1. **Recursive Function (`countRecur`)**:
    - This function calculates the number of ways the expression can be parenthesized such that it evaluates to `true` (`req = 1`) or `false` (`req = 0`).
    - Base case: If `i == j` (single character), return `1` if the character matches the required value.
    - Otherwise, for each possible position `k` of the operator, the function recursively computes the number of ways to evaluate the left and right parts of the expression.

2. **Memoization**:
    - A 3D `memo` array is used to store intermediate results, which helps in reducing redundant calculations.

3. **Final Result**:
    - The function `countWays` returns the number of ways to parenthesize the entire expression `s` such that it evaluates to `true`.

2. **Brute Force (Recursive with All Possible Parentheses)**

The brute force approach would involve generating all possible ways to insert parentheses in the expression and evaluating each possibility. However, this approach would be inefficient and result in an exponential time complexity.

**Approach Details:
- Generate all possible combinations of parentheses for the given boolean expression.
- For each combination, evaluate the expression.
- Count the number of times the expression evaluates to `true`.

Since this approach is highly inefficient due to its exponential complexity, it is not recommended for large inputs.

3. **Dynamic Programming (Bottom-Up Approach)**

A more optimal solution can be achieved by using dynamic programming with a bottom-up approach. This method involves filling up a DP table where each entry `dp[i][j][0]` represents the number of ways the subexpression `s[i...j]` evaluates to `false`, and `dp[i][j][1]` represents the number of ways it evaluates to `true`.

This approach eliminates the need for recursion and memoization, but it is still complex and requires careful implementation.

**Code Outline (Pseudo-Code):**
```java
class Solution {
    static int countWays(String s) {
        int n = s.length();
        int[][][] dp = new int[n][n][2];
        
        // Initialize the base cases and fill the DP table
        
        // Use dynamic programming to calculate ways for each subexpression
        
        return dp[0][n-1][1]; // Return number of ways the entire expression evaluates to true
    }
}
```

**Summary of Approaches:**
| **Approach**                       | **Time Complexity** | **Space Complexity** | **Optimal?** |
|------------------------------------|---------------------|----------------------|--------------|
| **Dynamic Programming (Top-Down)** | `O(n^3)`            | `O(n^3)`             | ✅ Yes        |
| **Brute Force**                    | Exponential (`O(2^n)`) | `O(2^n)`             | ❌ No         |
| **Dynamic Programming (Bottom-Up)**| `O(n^3)`            | `O(n^3)`             | ✅ Yes        |

The **Dynamic Programming (Top-Down)** approach is the most efficient and optimal solution for this problem.

## Problem Statement: Palindromic Partitioning
Given a string `s`, a partitioning of the string is a palindrome partitioning if every sub-string of the partition is a palindrome. Determine the fewest cuts needed for palindrome partitioning of the given string.

**Example:**
**Input**:
- s = "aab"

**Output**:
- 1

**Explanation**:
The palindrome partitioning of the string is ["aa", "b"], we need 1 cut.

**Approaches:**

1. **Dynamic Programming with Palindrome Check**

This approach utilizes dynamic programming to store whether a substring is a palindrome and then calculates the fewest cuts needed based on that information.

**Code Implementation:**

```java
class Solution {
    static void generatePal(String s, boolean[][] isPalin) {
        int n = s.length();
        for (int i = 0; i < n; i++) {
            isPalin[i][i] = true;  // A single character is always a palindrome
        }
        for (int len = 2; len <= n; len++) {
            for (int i = 0, j = i + len - 1; j < n; i++, j++) {
                if (s.charAt(i) == s.charAt(j) && 
                    (len == 2 || isPalin[i + 1][j - 1])) {
                    isPalin[i][j] = true;
                }
            }
        }
    }

    static int palPartition(String s) {
        int n = s.length();
        boolean[][] isPalin = new boolean[n][n];

        generatePal(s, isPalin);  // Fill the palindrome matrix
        
        int[] dp = new int[n];  // dp[i] will store the fewest cuts needed for substring s[0...i]
        Arrays.fill(dp, n);  // Initialize with maximum cuts
        dp[0] = 0;  // No cuts are needed for a single character
        
        for (int i = 1; i < n; i++) {
            if (isPalin[0][i]) {  // If the substring from 0 to i is a palindrome, no cut is needed
                dp[i] = 0;
            } else {
                for (int j = i; j >= 1; j--) {
                    if (isPalin[j][i]) {  // If substring s[j...i] is a palindrome
                        dp[i] = Math.min(dp[i], 1 + dp[j - 1]);  // Make a cut at j-1
                    }
                }
            }
        }
        return dp[n - 1];  // Return the minimum cuts required for the entire string
    }
}
```

**Explanation:**

1. **Palindrome Check (generatePal function)**:
   - First, we create a 2D boolean array `isPalin` where `isPalin[i][j]` will be `true` if the substring `s[i...j]` is a palindrome.
   - We initialize the diagonal of the matrix as `true` because any single character is a palindrome.
   - We then expand the substring size and check if `s[i] == s[j]` and the substring `s[i+1...j-1]` is a palindrome.

2. **Dynamic Programming (palPartition function)**:
   - We create a `dp` array where `dp[i]` represents the minimum number of cuts needed for substring `s[0...i]`.
   - For each substring, if the substring is a palindrome, we don't need any cuts. Otherwise, we check all possible cuts and take the minimum.

3. **Time Complexity**:  
   - **Palindrome check (`generatePal`)**: `O(n^2)`
   - **DP calculation (`palPartition`)**: `O(n^2)`
   - **Overall time complexity**: `O(n^2)`.

4. **Space Complexity**:
   - We use a 2D boolean array (`isPalin`) of size `O(n^2)` and a 1D array (`dp`) of size `O(n)`, so the space complexity is `O(n^2)`.

    2. **Brute Force (Backtracking)**

A brute force approach would involve recursively generating all possible partitions of the string and checking if each part is a palindrome. However, this approach has an exponential time complexity and is not optimal for large strings.

**Approach Details:**
- Generate all possible partitions of the string.
- For each partition, check if each substring is a palindrome.
- Count the number of cuts required for valid palindrome partitions.
- This method would lead to a time complexity of `O(2^n)` due to the exponential number of partitions.

3. **Dynamic Programming with Simple Optimization**

An alternate approach involves dynamic programming with a simpler optimization, where instead of calculating the palindrome status for all substrings, we only check the current substring’s palindromic property during the DP loop. This could result in fewer operations.

**Code Outline (Pseudo-Code):
```java
class Solution {
    static int palPartition(String s) {
        int n = s.length();
        int[] dp = new int[n];
        
        // Initialize dp with maximum possible cuts
        Arrays.fill(dp, n);

        for (int i = 0; i < n; i++) {
            for (int j = i; j >= 0; j--) {
                if (isPalindrome(s, j, i)) {
                    dp[i] = Math.min(dp[i], (j > 0 ? dp[j - 1] : 0) + 1);
                }
            }
        }
        return dp[n - 1] - 1;
    }

    static boolean isPalindrome(String s, int start, int end) {
        while (start < end) {
            if (s.charAt(start) != s.charAt(end)) {
                return false;
            }
            start++;
            end--;
        }
        return true;
    }
}
```

**Summary of Approaches:**

| **Approach**                       | **Time Complexity** | **Space Complexity** | **Optimal?** |
|------------------------------------|---------------------|----------------------|--------------|
| **Dynamic Programming (with Palindrome Check)** | `O(n^2)`            | `O(n^2)`             | ✅ Yes        |
| **Brute Force (Backtracking)**     | Exponential (`O(2^n)`) | `O(n)`             | ❌ No         |
| **Dynamic Programming (with Simple Optimization)** | `O(n^2)`            | `O(n)`               | ✅ Yes        |

The **Dynamic Programming (with Palindrome Check)** approach is the most efficient and optimal solution for this problem.

## Problem Statement: Max Rectangle
Given a binary matrix `mat[][]` of size `n * m`, find the maximum area of a rectangle formed only of 1s in the given matrix.

**Example:**
**Input**:
- `mat = [[0, 1, 1, 0], [1, 1, 1, 1], [0, 1, 1, 1]]`

**Output**:
- `6`

**Explanation**:
The largest rectangle of 1s has an area of 6 (formed by the bottom-right submatrix).

**Approaches:**:

1. **Dynamic Programming with Histograms**

This approach treats each row as the base of a histogram where the height of each column is increased by 1 whenever the matrix has a `1` at that position. For each row, we calculate the maximum rectangular area that can be formed using that row as the base.

**Code Implementation:**:

```java
class Solution {
    // Function to calculate maximum rectangular area in a binary matrix
    static int maximalRectangle(char[][] matrix) {
        if (matrix == null || matrix.length == 0 || matrix[0].length == 0)
            return 0;

        int n = matrix.length;
        int m = matrix[0].length;

        // Initialize heights array to store the histogram heights
        int[] heights = new int[m];
        int maxArea = 0;

        // Iterate through each row to treat the row as the base of the histogram
        for (int i = 0; i < n; i++) {
            for (int j = 0; j < m; j++) {
                // If the current cell is 1, add to the height
                if (matrix[i][j] == '1') {
                    heights[j]++;
                } else {
                    // Reset the height to 0 if the cell is 0
                    heights[j] = 0;
                }
            }

            // Calculate the maximum area for the current row (acting as a histogram)
            maxArea = Math.max(maxArea, maxHistogramArea(heights));
        }

        return maxArea;
    }

    // Helper function to calculate maximum area of histogram
    static int maxHistogramArea(int[] heights) {
        int n = heights.length;
        int[] stack = new int[n + 1];
        int top = -1;
        int maxArea = 0;

        for (int i = 0; i <= n; i++) {
            while (top >= 0 && (i == n || heights[i] < heights[stack[top]])) {
                int height = heights[stack[top--]];
                int width = (top == -1) ? i : i - stack[top] - 1;
                maxArea = Math.max(maxArea, height * width);
            }
            stack[++top] = i;
        }
        return maxArea;
    }
}
```

**Explanation:**:

1. **Convert Rows to Histograms**:
   - We convert each row in the binary matrix to a histogram where the height of each column is the number of consecutive `1`s in that column from the top row to the current row.
   - For each row, we update the heights array and treat it as a histogram.

2. **Maximum Histogram Area**:
   - For each row's histogram, we calculate the maximum rectangular area using the stack-based approach to find the largest rectangle in a histogram.

3. **Time Complexity**:
   - **Matrix Traversal**: `O(n * m)` where `n` is the number of rows and `m` is the number of columns.
   - **Histogram Calculation**: `O(m)` for each row due to the stack-based approach.
   - **Overall Time Complexity**: `O(n * m)`.

4. **Space Complexity**:
   - The space complexity is `O(m)` for storing the histogram heights array and the stack.

2. **Brute Force (Naive Approach)**

In this approach, we can try to find all possible rectangles in the matrix by iterating over every pair of rows and columns, calculating the area of rectangles with all `1`s, and keeping track of the maximum area.

**Code Implementation:**:

```java
class Solution {
    // Function to calculate maximum area of a rectangle formed by 1s
    static int maximalRectangle(char[][] matrix) {
        int n = matrix.length;
        int m = matrix[0].length;
        int maxArea = 0;

        // Traverse all possible top and bottom rows for rectangle
        for (int top = 0; top < n; top++) {
            for (int bottom = top; bottom < n; bottom++) {
                // Create an array that represents the count of 1s between the rows
                int[] rowSum = new int[m];
                for (int i = top; i <= bottom; i++) {
                    for (int j = 0; j < m; j++) {
                        if (matrix[i][j] == '1') {
                            rowSum[j]++;
                        } else {
                            rowSum[j] = 0;
                        }
                    }
                }

                // Calculate the area of the largest rectangle using this rowSum
                maxArea = Math.max(maxArea, largestRectangle(rowSum));
            }
        }
        return maxArea;
    }

    // Helper function to find largest rectangle in a histogram
    static int largestRectangle(int[] rowSum) {
        int n = rowSum.length;
        int maxArea = 0;
        for (int i = 0; i < n; i++) {
            if (rowSum[i] == 0) continue;
            int width = 1;
            for (int j = i + 1; j < n && rowSum[j] >= rowSum[i]; j++) {
                width++;
            }
            maxArea = Math.max(maxArea, rowSum[i] * width);
        }
        return maxArea;
    }
}
```

**Explanation:**:

1. **Brute Force Over Rows and Columns**:
   - We iterate over all pairs of rows (`top` and `bottom`), and for each such pair, we construct an array (`rowSum`) that stores the number of consecutive `1`s between the rows in each column.

2. **Largest Rectangle in Row**:
   - For each row, we calculate the largest rectangle area formed by `1`s in that row using a simplified version of the largest rectangle in histogram problem.

3. **Time Complexity**:
   - **Row Pair Traversal**: `O(n^2)`, since we are iterating over all pairs of rows.
   - **Column Traversal**: `O(m)` for each row pair.
   - **Overall Time Complexity**: `O(n^2 * m)`.

4. **Space Complexity**:
   - `O(m)` for storing the `rowSum` array.

3. **Optimized Dynamic Programming Approach**

This approach optimizes the histogram-based method by directly calculating the maximal rectangle area for each row, avoiding the need for a stack.

**Code Implementation:**:

```java
class Solution {
    static int maximalRectangle(char[][] matrix) {
        if (matrix == null || matrix.length == 0 || matrix[0].length == 0)
            return 0;

        int n = matrix.length;
        int m = matrix[0].length;
        int[] heights = new int[m];
        int maxArea = 0;

        for (int i = 0; i < n; i++) {
            for (int j = 0; j < m; j++) {
                // If matrix[i][j] is 1, increase the height for the column
                heights[j] = (matrix[i][j] == '1') ? heights[j] + 1 : 0;
            }
            
            maxArea = Math.max(maxArea, largestRectangleInHistogram(heights));
        }
        return maxArea;
    }

    static int largestRectangleInHistogram(int[] heights) {
        int n = heights.length;
        int[] stack = new int[n + 1];
        int top = -1;
        int maxArea = 0;

        for (int i = 0; i <= n; i++) {
            while (top >= 0 && (i == n || heights[i] < heights[stack[top]])) {
                int height = heights[stack[top--]];
                int width = (top == -1) ? i : i - stack[top] - 1;
                maxArea = Math.max(maxArea, height * width);
            }
            stack[++top] = i;
        }
        return maxArea;
    }
}
```

**Explanation:**:
- **Direct Calculation of Heights**:
  - We calculate the histogram heights for each row and use them to calculate the maximal rectangle area in that histogram.
  
- **Efficient Stack-based Maximum Area Calculation**:
  - We use a stack to efficiently calculate the largest rectangle area for each row's histogram.

- **Time Complexity**:
  - **Matrix Traversal**: `O(n * m)`
  - **Histogram Calculation**: `O(m)` for each row.
  - **Overall Time Complexity**: `O(n * m)`.

- **Space Complexity**:
  - `O(m)` for storing the histogram heights array and the stack.

**Summary of Approaches:**:

| **Approach**                    | **Time Complexity** | **Space Complexity** | **Optimal?** |
|----------------------------------|---------------------|----------------------|--------------|
| **Dynamic Programming with Histograms** | `O(n * m)`           | `O(m)`               | ✅ Yes        |
| **Brute Force (Naive Approach)** | `O(n^2 * m)`        | `O(m)`               | ❌ No         |
| **Optimized Dynamic Programming Approach** | `O(n * m)`           | `O(m)`               | ✅ Yes        |

The **Dynamic Programming with Histograms** approach and the **Optimized Dynamic Programming Approach** are the most efficient solutions for this problem, with both having a time complexity of `O(n * m)`.


## Problem Statement: Unique Paths II

You are given an `m x n` integer array `grid`. A robot is initially located at the top-left corner `(0, 0)`. It tries to move to the bottom-right corner `(m - 1, n - 1)`. The robot can only move either **down** or **right** at any point in time.
An obstacle and an empty space are marked as `1` and `0` respectively. A path cannot include any square with an obstacle.
**Return** the number of possible unique paths that the robot can take to reach the bottom-right corner.
The testcases are generated so that the answer will be less than or equal to 2 * 10⁹.

**Example:**
**Input:**
```java
grid = [
  [0,0,0],
  [0,1,0],
  [0,0,0]
]
```

**Output:**
```
2
```

**Explanation:**
There are two possible paths:
- Right → Right → Down → Down
- Down → Down → Right → Right

**Constraints:**
- `m == obstacleGrid.length`
- `n == obstacleGrid[i].length`
- `1 <= m, n <= 100`
- `obstacleGrid[i][j]` is `0` (free) or `1` (obstacle)

**Approaches**
**✅ 1. Dynamic Programming (Tabulation)**

We use a 2D `dp` array where `dp[i][j]` represents the number of unique paths to cell `(i, j)`, taking into account obstacles.

**Java Code:
```java
class Solution {
    public int uniquePathsWithObstacles(int[][] obstacleGrid) {
        int m = obstacleGrid.length;
        int n = obstacleGrid[0].length;

        int[][] dp = new int[m][n];

        if (obstacleGrid[0][0] == 1) return 0;
        dp[0][0] = 1;

        for (int i = 1; i < m; i++) {
            if (obstacleGrid[i][0] == 0 && dp[i-1][0] == 1)
                dp[i][0] = 1;
        }

        for (int j = 1; j < n; j++) {
            if (obstacleGrid[0][j] == 0 && dp[0][j-1] == 1)
                dp[0][j] = 1;
        }

        for (int i = 1; i < m; i++) {
            for (int j = 1; j < n; j++) {
                if (obstacleGrid[i][j] == 0) {
                    dp[i][j] = dp[i-1][j] + dp[i][j-1];
                }
            }
        }

        return dp[m-1][n-1];
    }
}
```

**Time Complexity**: `O(m * n)`  
**Space Complexity**: `O(m * n)`

**✅ 2. Dynamic Programming (Space Optimized)**
We can reduce space to `O(n)` by keeping only the previous row.

**Java Code:**
```java
class Solution {
    public int uniquePathsWithObstacles(int[][] obstacleGrid) {
        int m = obstacleGrid.length;
        int n = obstacleGrid[0].length;

        int[] dp = new int[n];
        dp[0] = obstacleGrid[0][0] == 0 ? 1 : 0;

        for (int i = 0; i < m; i++) {
            for (int j = 0; j < n; j++) {
                if (obstacleGrid[i][j] == 1) {
                    dp[j] = 0;
                } else if (j > 0) {
                    dp[j] += dp[j - 1];
                }
            }
        }

        return dp[n - 1];
    }
}
```

**Time Complexity**: `O(m * n)`  
**Space Complexity**: `O(n)`

**✅ 3. Recursive + Memoization**
We use recursion to explore all paths while caching intermediate results.

**Java Code:**
```java
class Solution {
    Integer[][] memo;

    public int uniquePathsWithObstacles(int[][] obstacleGrid) {
        int m = obstacleGrid.length;
        int n = obstacleGrid[0].length;
        memo = new Integer[m][n];
        return dfs(0, 0, obstacleGrid);
    }

    private int dfs(int i, int j, int[][] grid) {
        int m = grid.length, n = grid[0].length;

        if (i >= m || j >= n || grid[i][j] == 1) return 0;
        if (i == m - 1 && j == n - 1) return 1;

        if (memo[i][j] != null) return memo[i][j];

        memo[i][j] = dfs(i + 1, j, grid) + dfs(i, j + 1, grid);
        return memo[i][j];
    }
}
```

**Time Complexity**: `O(m * n)` (due to memoization)  
**Space Complexity**: `O(m * n)` (memo + recursion stack)

**❌ 4. Plain Recursion (Not Recommended)**
This approach leads to **TLE** (Time Limit Exceeded) due to exponential calls.

**Code Sketch:**
```java
int dfs(int i, int j, int[][] grid) {
    if (i >= m || j >= n || grid[i][j] == 1) return 0;
    if (i == m - 1 && j == n - 1) return 1;
    return dfs(i + 1, j, grid) + dfs(i, j + 1, grid);
}
```

**Time Complexity**: `O(2^(m+n))`  
**Space Complexity**: `O(m + n)` (recursion stack)

**Summary of Approaches**

| Approach             | Time Complexity | Space Complexity | Optimal |
|----------------------|------------------|-------------------|---------|
| DP Tabulation        | `O(m * n)`       | `O(m * n)`        | ✅      |
| DP Space Optimized   | `O(m * n)`       | `O(n)`            | ✅      |
| Memoization (DFS)    | `O(m * n)`       | `O(m * n)`        | ✅      |
| Plain Recursion      | `Exponential`    | `O(m + n)`        | ❌      |

**Conclusion**
This problem is a classic variation of the "Unique Paths" problem that introduces obstacles. The best practice is to use **bottom-up DP** or **top-down memoization** to efficiently solve it in `O(m * n)` time.


## Problem Statement: Minimum Path Sum

You are given an `m x n` grid filled with **non-negative numbers**. Find a path from the **top-left** to the **bottom-right** corner that **minimizes** the sum of all numbers along the path.
You can **only move either down or right** at any point in time.

**Example**
**Input:**
```java
grid = [
  [1,3,1],
  [1,5,1],
  [4,2,1]
]
```

**Output:**
```
7
```

**Explanation:**
The path 1 → 3 → 1 → 1 → 1 minimizes the sum to 7.

**Constraints**
- `m == grid.length`
- `n == grid[i].length`
- `1 <= m, n <= 200`
- `0 <= grid[i][j] <= 100`

**Approaches**
**✅ 1. Dynamic Programming (In-place Tabulation)**
We modify the grid in place.  
Each cell contains the minimum path sum to reach that cell from the top-left.

**Java Code:**
```java
class Solution {
    public int minPathSum(int[][] grid) {
        int m = grid.length;
        int n = grid[0].length;

        for (int j = 1; j < n; j++) {
            grid[0][j] += grid[0][j - 1];
        }

        for (int i = 1; i < m; i++) {
            grid[i][0] += grid[i - 1][0];
        }

        for (int i = 1; i < m; i++) {
            for (int j = 1; j < n; j++) {
                grid[i][j] += Math.min(grid[i - 1][j], grid[i][j - 1]);
            }
        }

        return grid[m - 1][n - 1];
    }
}
```

**Time Complexity**: `O(m * n)`  
**Space Complexity**: `O(1)` (modifying input grid)

**✅ 2. Dynamic Programming (Using Extra Space)**
Use a separate `dp` array to store the minimum path sum.

**Java Code:**
```java
class Solution {
    public int minPathSum(int[][] grid) {
        int m = grid.length, n = grid[0].length;
        int[][] dp = new int[m][n];
        
        dp[0][0] = grid[0][0];

        for (int i = 1; i < m; i++) dp[i][0] = dp[i - 1][0] + grid[i][0];
        for (int j = 1; j < n; j++) dp[0][j] = dp[0][j - 1] + grid[0][j];

        for (int i = 1; i < m; i++) {
            for (int j = 1; j < n; j++) {
                dp[i][j] = Math.min(dp[i - 1][j], dp[i][j - 1]) + grid[i][j];
            }
        }

        return dp[m - 1][n - 1];
    }
}
```

**Time Complexity**: `O(m * n)`  
**Space Complexity**: `O(m * n)`

**✅ 3. Recursive + Memoization (Top-Down)**
A recursive DFS with memoization to cache subproblems.

**Java Code:**
```java
class Solution {
    public int minPathSum(int[][] grid) {
        int m = grid.length, n = grid[0].length;
        Integer[][] memo = new Integer[m][n];
        return dfs(0, 0, grid, memo);
    }

    private int dfs(int i, int j, int[][] grid, Integer[][] memo) {
        if (i >= grid.length || j >= grid[0].length) return Integer.MAX_VALUE;
        if (i == grid.length - 1 && j == grid[0].length - 1) return grid[i][j];
        if (memo[i][j] != null) return memo[i][j];

        int down = dfs(i + 1, j, grid, memo);
        int right = dfs(i, j + 1, grid, memo);

        return memo[i][j] = grid[i][j] + Math.min(down, right);
    }
}
```

**Time Complexity**: `O(m * n)`  
**Space Complexity**: `O(m * n)` (memo + recursion stack)

**❌ 4. Plain Recursion (Without Memoization)**
This method explores all paths and results in **TLE** for large grids.

**Code Sketch:**
```java
int dfs(int i, int j, int[][] grid) {
    if (i >= m || j >= n) return Integer.MAX_VALUE;
    if (i == m - 1 && j == n - 1) return grid[i][j];
    return grid[i][j] + Math.min(dfs(i + 1, j, grid), dfs(i, j + 1, grid));
}
```

**Time Complexity**: `O(2^(m+n))`  
**Space Complexity**: `O(m + n)` (recursion depth)

**Summary of Approaches**
| Approach                  | Time Complexity | Space Complexity | Optimal |
|---------------------------|------------------|-------------------|---------|
| DP (In-place)             | `O(m * n)`       | `O(1)`            | ✅      |
| DP (Extra Space)          | `O(m * n)`       | `O(m * n)`        | ✅      |
| Recursive + Memoization   | `O(m * n)`       | `O(m * n)`        | ✅      |
| Plain Recursion           | `Exponential`    | `O(m + n)`        | ❌      |

**Conclusion**
The optimal solution is to use **in-place DP**, which avoids extra space and runs in `O(m * n)` time.  
Memoized recursion is also a good alternative for a top-down approach.

## Problem Statement: Triangle
You are given a **triangle** array, return the **minimum path sum** from top to bottom.
For each step, you may move to an **adjacent number** of the row below. More formally, if you are on index `i` on the current row, you may move to either index `i` or index `i + 1` on the next row.

**Example**
**Input:**
```java
triangle = [
     [2],
    [3,4],
   [6,5,7],
  [4,1,8,3]
]
```

**Output:**
```
11
```

**Explanation:** The path with minimum sum is `2 → 3 → 5 → 1 = 11`.

**Constraints**
- `1 <= triangle.length <= 200`
- `triangle[0].length == 1`
- `triangle[i].length == triangle[i - 1].length + 1`
- `-10^4 <= triangle[i][j] <= 10^4`

**Approaches**

**✅ 1. Bottom-Up DP (1D Optimized)**
Iterate from the bottom to the top row, updating a 1D `dp` array in-place.

**Java Code:**
```java
class Solution {
    public int minimumTotal(List<List<Integer>> triangle) {
        int n = triangle.size();
        int[] dp = new int[n];

        for (int i = 0; i < n; i++) {
            dp[i] = triangle.get(n - 1).get(i);
        }

        for (int row = n - 2; row >= 0; row--) {
            for (int col = 0; col <= row; col++) {
                dp[col] = triangle.get(row).get(col) + Math.min(dp[col], dp[col + 1]);
            }
        }

        return dp[0];
    }
}
```

**Time Complexity**: `O(n^2)`  
**Space Complexity**: `O(n)`  

**✅ 2. Bottom-Up DP (2D Array)**

Use a 2D `dp` table and fill it from the bottom up.

**Java Code:**
```java
class Solution {
    public int minimumTotal(List<List<Integer>> triangle) {
        int n = triangle.size();
        int[][] dp = new int[n][n];

        for (int j = 0; j < n; j++) {
            dp[n - 1][j] = triangle.get(n - 1).get(j);
        }

        for (int i = n - 2; i >= 0; i--) {
            for (int j = 0; j <= i; j++) {
                dp[i][j] = triangle.get(i).get(j) + Math.min(dp[i + 1][j], dp[i + 1][j + 1]);
            }
        }

        return dp[0][0];
    }
}
```

**Time Complexity**: `O(n^2)`  
**Space Complexity**: `O(n^2)`

**✅ 3. Top-Down DP (Memoization)**

Use recursion with memoization to avoid recomputation.

**Java Code:**
```java
class Solution {
    public int minimumTotal(List<List<Integer>> triangle) {
        int n = triangle.size();
        Integer[][] memo = new Integer[n][n];
        return helper(0, 0, triangle, memo);
    }

    private int helper(int i, int j, List<List<Integer>> triangle, Integer[][] memo) {
        if (i == triangle.size() - 1) {
            return triangle.get(i).get(j);
        }
        if (memo[i][j] != null) return memo[i][j];

        int left = helper(i + 1, j, triangle, memo);
        int right = helper(i + 1, j + 1, triangle, memo);

        return memo[i][j] = triangle.get(i).get(j) + Math.min(left, right);
    }
}
```

**Time Complexity**: `O(n^2)`  
**Space Complexity**: `O(n^2)` (memo + recursion stack)

**❌ 4. Plain Recursion (No Memoization)**

Explores all paths; **inefficient** for large inputs due to exponential time.

**Java Code:**
```java
class Solution {
    public int minimumTotal(List<List<Integer>> triangle) {
        return helper(0, 0, triangle);
    }

    private int helper(int i, int j, List<List<Integer>> triangle) {
        if (i == triangle.size() - 1) {
            return triangle.get(i).get(j);
        }

        int left = helper(i + 1, j, triangle);
        int right = helper(i + 1, j + 1, triangle);

        return triangle.get(i).get(j) + Math.min(left, right);
    }
}
```

**Time Complexity**: `O(2^n)`  
**Space Complexity**: `O(n)`

**Summary of Approaches**
| Approach                  | Time Complexity | Space Complexity | Optimal |
|---------------------------|------------------|-------------------|---------|
| Bottom-Up DP (1D)         | `O(n^2)`         | `O(n)`            | ✅      |
| Bottom-Up DP (2D)         | `O(n^2)`         | `O(n^2)`          | ✅      |
| Top-Down Memoization      | `O(n^2)`         | `O(n^2)`          | ✅      |
| Plain Recursion           | `Exponential`    | `O(n)`            | ❌      |

**Conclusion**
The best solution in terms of space and efficiency is the **1D Bottom-Up DP**, which solves the problem in-place with minimal memory usage.


## Problem Statement: Minimum Falling Path Sum

Given an `n x n` array of integers `matrix`, return the **minimum sum** of any **falling path** through `matrix`.
A **falling path** starts at any element in the first row and chooses the element in the next row that is **either directly below or diagonally left/right**. Specifically, the next element from position `(row, col)` will be:
- `(row + 1, col - 1)`
- `(row + 1, col)`
- `(row + 1, col + 1)`

**Example**
**Input:**
```java
matrix = [
  [2,1,3],
  [6,5,4],
  [7,8,9]
]
```

**Output:**
```
13
```

**Explanation:** The falling path with minimum sum is `1 → 4 → 8 = 13`.

**Constraints**
- `n == matrix.length == matrix[i].length`
- `1 <= n <= 100`
- `-100 <= matrix[i][j] <= 100`

**Approaches**
**✅ 1. Bottom-Up DP with 1D Space Optimization**

Iterate from bottom to top, and only maintain one row at a time.

**Java Code:**
```java
class Solution {
    public int minFallingPathSum(int[][] matrix) {
        int n = matrix.length;
        int[] dp = matrix[n - 1].clone(); 

        for (int row = n - 2; row >= 0; row--) {
            int[] newDp = new int[n];
            for (int col = 0; col < n; col++) {
                int down = dp[col];
                int left = col > 0 ? dp[col - 1] : Integer.MAX_VALUE;
                int right = col < n - 1 ? dp[col + 1] : Integer.MAX_VALUE;

                newDp[col] = matrix[row][col] + Math.min(down, Math.min(left, right));
            }
            dp = newDp;
        }

        int min = Integer.MAX_VALUE;
        for (int val : dp) {
            min = Math.min(min, val);
        }

        return min; 
    }
}
```

**Time Complexity**: `O(n^2)`  
**Space Complexity**: `O(n)`

**✅ 2. Bottom-Up DP with 2D Table**

Use a 2D table and fill from bottom to top.

**Java Code:**
```java
class Solution {
    public int minFallingPathSum(int[][] matrix) {
        int n = matrix.length;
        int[][] dp = new int[n][n];

        for (int col = 0; col < n; col++) {
            dp[n - 1][col] = matrix[n - 1][col];
        }

        for (int row = n - 2; row >= 0; row--) {
            for (int col = 0; col < n; col++) {
                int down = dp[row + 1][col];
                int left = col > 0 ? dp[row + 1][col - 1] : Integer.MAX_VALUE;
                int right = col < n - 1 ? dp[row + 1][col + 1] : Integer.MAX_VALUE;

                dp[row][col] = matrix[row][col] + Math.min(down, Math.min(left, right));
            }
        }

        int min = Integer.MAX_VALUE;
        for (int val : dp[0]) {
            min = Math.min(min, val);
        }

        return min;
    }
}
```

**Time Complexity**: `O(n^2)`  
**Space Complexity**: `O(n^2)`

**✅ 3. Top-Down DP (Memoization)**

Use recursion with memoization to avoid recomputation.

**Java Code:**
```java
class Solution {
    public int minFallingPathSum(int[][] matrix) {
        int n = matrix.length;
        Integer[][] memo = new Integer[n][n];

        int minSum = Integer.MAX_VALUE;
        for (int col = 0; col < n; col++) {
            minSum = Math.min(minSum, dfs(0, col, matrix, memo));
        }
        return minSum;
    }

    private int dfs(int row, int col, int[][] matrix, Integer[][] memo) {
        if (col < 0 || col >= matrix.length) return Integer.MAX_VALUE;
        if (row == matrix.length - 1) return matrix[row][col];
        if (memo[row][col] != null) return memo[row][col];

        int down = dfs(row + 1, col, matrix, memo);
        int left = dfs(row + 1, col - 1, matrix, memo);
        int right = dfs(row + 1, col + 1, matrix, memo);

        memo[row][col] = matrix[row][col] + Math.min(down, Math.min(left, right));
        return memo[row][col];
    }
}
```

**Time Complexity**: `O(n^2)`  
**Space Complexity**: `O(n^2)` (memo + recursion stack)

**❌ 4. Plain Recursion**

Brute-force all paths. Not feasible for large inputs.

**Java Code:**
```java
class Solution {
    public int minFallingPathSum(int[][] matrix) {
        int n = matrix.length;
        int minSum = Integer.MAX_VALUE;
        for (int col = 0; col < n; col++) {
            minSum = Math.min(minSum, dfs(0, col, matrix));
        }
        return minSum;
    }

    private int dfs(int row, int col, int[][] matrix) {
        if (col < 0 || col >= matrix.length) return Integer.MAX_VALUE;
        if (row == matrix.length - 1) return matrix[row][col];

        int down = dfs(row + 1, col, matrix);
        int left = dfs(row + 1, col - 1, matrix);
        int right = dfs(row + 1, col + 1, matrix);

        return matrix[row][col] + Math.min(down, Math.min(left, right));
    }
}
```

**Time Complexity**: `O(3^n)`  
**Space Complexity**: `O(n)`

**Summary of Approaches**
| Approach                | Time Complexity | Space Complexity | Optimal |
|-------------------------|------------------|-------------------|---------|
| Bottom-Up DP (1D)       | `O(n^2)`         | `O(n)`            | ✅      |
| Bottom-Up DP (2D)       | `O(n^2)`         | `O(n^2)`          | ✅      |
| Top-Down Memoization    | `O(n^2)`         | `O(n^2)`          | ✅      |
| Plain Recursion         | `Exponential`    | `O(n)`            | ❌      |

**Conclusion**
For best efficiency, use the **Bottom-Up DP with 1D space optimization** approach. It's fast and uses minimal space.


## Problem Statement: Minimum Falling Path Sum II

Given an `n x n` integer matrix `grid`, return the **minimum sum** of a falling path with **non-zero shifts**.
A **falling path with non-zero shifts** is a choice of exactly one element from each row such that **no two elements chosen in adjacent rows are in the same column**.

**Example**
**Input:**
```java
grid = [
  [2,1,3],
  [6,5,4],
  [7,8,9]
]
```

**Output:**
```
13
```

**Explanation:**  
Choose `1 (row 0, col 1) → 4 (row 1, col 2) → 8 (row 2, col 1) = 13`.  
We ensure no two selected elements are in the same column in adjacent rows.

**Constraints**
- `n == grid.length == grid[i].length`
- `1 <= n <= 200`
- `-99 <= grid[i][j] <= 99`

**Approaches**
**✅ 1. Optimized DP (Track First & Second Min Values)**

This is the **optimal approach**. At each row, keep track of the smallest and second smallest values from the previous row. For each column, if the column index equals the index of the minimum, take the second minimum instead.

**Java Code:**
```java
class Solution {
    public int minFallingPathSum(int[][] grid) {
        int n = grid.length;
        int[] dp = new int[n];
        for (int j = 0; j < n; j++) dp[j] = grid[0][j];

        for (int i = 1; i < n; i++) {
            int[] newDp = new int[n];
            int min1 = Integer.MAX_VALUE, min2 = Integer.MAX_VALUE;
            int min1Index = -1;
            for (int j = 0; j < n; j++) {
                if (dp[j] < min1) {
                    min2 = min1;
                    min1 = dp[j];
                    min1Index = j;
                } else if (dp[j] < min2) {
                    min2 = dp[j];
                }
            }
            for (int j = 0; j < n; j++) {
                if (j == min1Index) {
                    newDp[j] = grid[i][j] + min2;
                } else {
                    newDp[j] = grid[i][j] + min1;
                }
            }

            dp = newDp;
        }

        int result = Integer.MAX_VALUE;
        for (int val : dp) {
            result = Math.min(result, val);
        }

        return result;
    }
}
```

**Time Complexity**: `O(n^2)`  
**Space Complexity**: `O(n)`

**✅ 2. Classic DP with Full Table (Brute-force Column Comparison)**

Try every valid column in the current row (excluding the previous row's column), and build a 2D DP table.

**Java Code:**
```java
class Solution {
    public int minFallingPathSum(int[][] grid) {
        int n = grid.length;
        int[][] dp = new int[n][n];
        for (int j = 0; j < n; j++) dp[0][j] = grid[0][j];

        for (int i = 1; i < n; i++) {
            for (int j = 0; j < n; j++) {
                dp[i][j] = Integer.MAX_VALUE;
                for (int k = 0; k < n; k++) {
                    if (k != j) {
                        dp[i][j] = Math.min(dp[i][j], grid[i][j] + dp[i-1][k]);
                    }
                }
            }
        }

        int result = Integer.MAX_VALUE;
        for (int j = 0; j < n; j++) {
            result = Math.min(result, dp[n-1][j]);
        }
        return result;
    }
}
```

**Time Complexity**: `O(n^3)`  
**Space Complexity**: `O(n^2)`  
⚠️ Slower for large `n`.

**✅ 3. Top-Down DP with Memoization**

Use recursion with memoization and skip the previous column.

**Java Code:**
```java
class Solution {
    int[][] memo;
    int[][] grid;
    int n;

    public int minFallingPathSum(int[][] grid) {
        this.n = grid.length;
        this.grid = grid;
        this.memo = new int[n][n];
        for (int[] row : memo) Arrays.fill(row, Integer.MAX_VALUE);

        int result = Integer.MAX_VALUE;
        for (int j = 0; j < n; j++) {
            result = Math.min(result, dfs(0, j, -1));
        }

        return result;
    }

    private int dfs(int row, int col, int prevCol) {
        if (col == prevCol) return Integer.MAX_VALUE;
        if (row == n - 1) return grid[row][col];
        if (memo[row][col] != Integer.MAX_VALUE) return memo[row][col];

        int min = Integer.MAX_VALUE;
        for (int nextCol = 0; nextCol < n; nextCol++) {
            if (nextCol != col) {
                min = Math.min(min, dfs(row + 1, nextCol, col));
            }
        }

        return memo[row][col] = grid[row][col] + min;
    }
}
```

**Time Complexity**: `O(n^3)` worst-case  
**Space Complexity**: `O(n^2)` + recursion stack

**Summary**
| Approach                     | Time Complexity | Space Complexity | Notes                          |
|------------------------------|------------------|-------------------|---------------------------------|
| Optimized DP (min1, min2)    | `O(n^2)`         | `O(n)`            | ✅ Best performance             |
| Full DP Table (Brute-force)  | `O(n^3)`         | `O(n^2)`          | ❌ Slower, simple to understand |
| Top-Down with Memoization    | `O(n^3)`         | `O(n^2)`          | ✅ Recursive solution           |

**Conclusion**
To handle this problem efficiently, the **optimized DP using `min1`, `min2` tracking** is the best approach with `O(n^2)` time and space efficiency.

## Problem Statement: Target Sum

You are given an integer array `nums` and an integer `target`.  
You want to build expressions by adding `'+'` or `'-'` before each number and evaluate the expression.
Return the number of different expressions that evaluate to `target`.

**Example**
**Input:**
```java
nums = [1,1,1,1,1]
target = 3
```

**Output:**
```
5
```

**Explanation:**  
There are 5 ways to assign symbols to make the sum 3:  
`-1 +1 +1 +1 +1 = 3`  
`+1 -1 +1 +1 +1 = 3`  
`+1 +1 -1 +1 +1 = 3`  
`+1 +1 +1 -1 +1 = 3`  
`+1 +1 +1 +1 -1 = 3`

**Constraints**
- `1 <= nums.length <= 20`
- `0 <= nums[i] <= 1000`
- The answer is guaranteed to fit in a 32-bit integer.

**Approaches**

**✅ 1. Subset Sum Reduction (Optimized DP)**
Transform the problem into finding subsets with a given sum.  
If total sum is `S`, and target is `T`, then:
```
P - N = target
P + N = sum
=> 2P = target + sum
=> P = (target + sum) / 2
```

**Java Code:**
```java
class Solution {
    public int findTargetSumWays(int[] nums, int target) {
        int sum = 0;
        for (int num : nums) sum += num;

        if ((target + sum) % 2 != 0 || sum < Math.abs(target)) return 0;
        int subsetSum = (target + sum) / 2;

        int[] dp = new int[subsetSum + 1];
        dp[0] = 1;

        for (int num : nums) {
            for (int j = subsetSum; j >= num; j--) {
                dp[j] += dp[j - num];
            }
        }

        return dp[subsetSum];
    }
}
```

**Time Complexity**: `O(n * subsetSum)`  
**Space Complexity**: `O(subsetSum)`

**✅ 2. Top-Down DP with Memoization (Backtracking)**

Try all combinations recursively with memoization.

**Java Code:**
```java
class Solution {
    Map<String, Integer> memo = new HashMap<>();

    public int findTargetSumWays(int[] nums, int target) {
        return dfs(nums, 0, 0, target);
    }

    private int dfs(int[] nums, int index, int sum, int target) {
        if (index == nums.length) return sum == target ? 1 : 0;

        String key = index + "," + sum;
        if (memo.containsKey(key)) return memo.get(key);

        int add = dfs(nums, index + 1, sum + nums[index], target);
        int subtract = dfs(nums, index + 1, sum - nums[index], target);

        memo.put(key, add + subtract);
        return memo.get(key);
    }
}
```

**Time Complexity**: `O(n * sum)`  
**Space Complexity**: `O(n * sum)` (memoization map)

**✅ 3. Plain Backtracking (No Memoization)**

A brute-force recursive approach. Simple but inefficient for large input.

**Java Code:**
```java
class Solution {
    int count = 0;

    public int findTargetSumWays(int[] nums, int target) {
        backtrack(nums, 0, 0, target);
        return count;
    }

    private void backtrack(int[] nums, int index, int sum, int target) {
        if (index == nums.length) {
            if (sum == target) count++;
            return;
        }

        backtrack(nums, index + 1, sum + nums[index], target);
        backtrack(nums, index + 1, sum - nums[index], target);
    }
}
```

**Time Complexity**: `O(2^n)`  
**Space Complexity**: `O(n)` (recursion stack)

**Summary**
| Approach                        | Time Complexity     | Space Complexity    | Notes                                |
|-------------------------------|----------------------|----------------------|---------------------------------------|
| Subset Sum DP                  | `O(n * sum)`         | `O(sum)`             | ✅ Best for large inputs              |
| Top-Down DP with Memoization   | `O(n * sum)`         | `O(n * sum)`         | ✅ Clean recursive solution           |
| Plain Backtracking             | `O(2^n)`             | `O(n)`               | ❌ Inefficient, only for small input  |

**Conclusion**
The **Subset Sum Transformation** is the most efficient and scalable approach.  
Use **Top-Down DP** if you prefer a recursive formulation with memoization.

## Problem Statement: Longest Increasing Subsequence

Given an integer array `nums`, return the length of the **longest strictly increasing subsequence**.
A subsequence is a sequence that can be derived from an array by deleting some or no elements without changing the order of the remaining elements.

**Example**
**Input:**
```java
nums = [10, 9, 2, 5, 3, 7, 101, 18]
```

**Output:**
```
4
```

**Explanation:**  
The longest increasing subsequence is `[2, 3, 7, 101]`.

**Constraints**
- `1 <= nums.length <= 2500`
- `-10^4 <= nums[i] <= 10^4`

**Approaches**
**✅ 1. Dynamic Programming (O(n²))**

We use an array `dp[]` where `dp[i]` represents the length of the LIS ending at index `i`.

**Java Code:**
```java
class Solution {
    public int lengthOfLIS(int[] nums) {
        int n = nums.length;
        int[] dp = new int[n];
        Arrays.fill(dp, 1);  
        int maxLen = 1;

        for (int i = 1; i < n; i++) {
            for (int j = 0; j < i; j++) {
                if (nums[j] < nums[i]) {
                    dp[i] = Math.max(dp[i], dp[j] + 1);
                }
            }
            maxLen = Math.max(maxLen, dp[i]);
        }

        return maxLen;
    }
}
```

**Time Complexity**: `O(n²)`  
**Space Complexity**: `O(n)`

**✅ 2. Binary Search + Greedy (O(n log n))**

Use a list to build the smallest tail of all increasing subsequences of a given length.

**Java Code:**
```java
class Solution {
    public int lengthOfLIS(int[] nums) {
        List<Integer> sub = new ArrayList<>();
        for (int num : nums) {
            int i = Collections.binarySearch(sub, num);
            if (i < 0) i = -(i + 1);
            if (i == sub.size()) sub.add(num);
            else sub.set(i, num);
        }
        return sub.size();
    }
}
```

**Time Complexity**: `O(n log n)`  
**Space Complexity**: `O(n)`  
**Explanation**: Maintains a dynamic list of "ends" of potential increasing subsequences.

**✅ 3. Recursion with Memoization (Top-Down DP)**

Try all possible subsequences recursively, caching the results to avoid recomputation.

**Java Code:**
```java
class Solution {
    Integer[][] memo;

    public int lengthOfLIS(int[] nums) {
        int n = nums.length;
        memo = new Integer[n][n + 1];
        return dfs(nums, 0, -1);
    }

    private int dfs(int[] nums, int curr, int prev) {
        if (curr == nums.length) return 0;

        if (memo[curr][prev + 1] != null) return memo[curr][prev + 1];

        int taken = 0;
        if (prev == -1 || nums[curr] > nums[prev]) {
            taken = 1 + dfs(nums, curr + 1, curr);
        }

        int notTaken = dfs(nums, curr + 1, prev);
        memo[curr][prev + 1] = Math.max(taken, notTaken);

        return memo[curr][prev + 1];
    }
}
```

**Time Complexity**: `O(n²)`  
**Space Complexity**: `O(n²)`

**Summary**
| Approach                    | Time Complexity | Space Complexity | Notes                          |
|----------------------------|-----------------|------------------|---------------------------------|
| DP (Tabulation)            | O(n²)           | O(n)             | Most straightforward            |
| Binary Search + Greedy     | O(n log n)      | O(n)             | Fastest, but trickier to grasp |
| Recursion + Memoization    | O(n²)           | O(n²)            | Clean recursive logic           |

**Conclusion**
For large inputs, prefer the **binary search + greedy** approach. For clear understanding, the **DP (O(n²))** solution is ideal.

## Problem Statement: Largest Divisible Subset
Given a set of **distinct positive integers** `nums`, return the **largest subset** such that for every pair `(a, b)` in the subset:
- `a % b == 0` or `b % a == 0`
If there are multiple solutions, return **any** of them.

**Example**
**Input:**
```java
nums = [1, 2, 4, 8]
```

**Output:**
```
[1, 2, 4, 8]
```
All pairs are divisible.

**Constraints**
- `1 <= nums.length <= 1000`
- `1 <= nums[i] <= 2 * 10^9`
- All the integers in `nums` are **unique**

**Approaches**

**✅ 1. Dynamic Programming with Tracking (O(n²))**

**Idea:**
- Sort the array
- Use `dp[i]` to store the size of the largest divisible subset ending at index `i`
- Use `prev[i]` to track the previous index for reconstruction

**Java Code:**
```java
class Solution {
    public List<Integer> largestDivisibleSubset(int[] nums) {
        int n = nums.length;
        Arrays.sort(nums);

        int[] dp = new int[n];
        int[] prev = new int[n];
        Arrays.fill(dp, 1); 
        Arrays.fill(prev, -1);

        int maxLen = 1;
        int maxIndex = 0;

        for (int i = 1; i < n; i++) {
            for (int j = 0; j < i; j++) {
                if (nums[i] % nums[j] == 0 && dp[j] + 1 > dp[i]) {
                    dp[i] = dp[j] + 1;
                    prev[i] = j;
                }
            }
            if (dp[i] > maxLen) {
                maxLen = dp[i];
                maxIndex = i;
            }
        }

        List<Integer> result = new ArrayList<>();
        while (maxIndex != -1) {
            result.add(nums[maxIndex]);
            maxIndex = prev[maxIndex];
        }

        return result;
    }
}
```

**✅ 2. Recursive + Memoization (Top-down DP)**

Less common in practice due to TLE risk, but still viable conceptually.

- Try all subsets recursively
- Cache results for subproblems

⚠ Not efficient for large input sizes.

**✅ 3. Greedy + HashMap (Variation Idea)**

You can also try mapping subset sizes per element with a map, but the classic DP approach is more intuitive and reliable.

**Summary**
| Approach        | Time Complexity | Space Complexity | Notes                          |
|----------------|-----------------|------------------|---------------------------------|
| DP (O(n²))      | O(n²)           | O(n)             | Best practical approach         |
| Recursive + Memo| Exponential     | O(n²)            | Risk of TLE on large inputs     |

**Tips**
- Sorting ensures that if `a % b == 0`, then `a >= b`
- Always track the previous index for reconstruction
- Start reconstruction from the index with the maximum subset length

**Conclusion**
This problem is a great example of using **DP with reconstruction**. The key is sorting the array first and leveraging divisible properties smartly during DP transitions.

## Problem Statement: Longest String Chain

You are given an array of words where each word consists of lowercase English letters.
A word `A` is a **predecessor** of word `B` if and only if we can **insert exactly one letter** anywhere in `A` without changing the order of the other characters to make it equal to `B`.
Return the length of the longest possible word chain using the given list of words.

**Example**
**Input:**
```java
words = ["a","b","ba","bca","bda","bdca"]
```

**Output:** `4`  
**Explanation:** One possible word chain is `"a" -> "ba" -> "bda" -> "bdca"`.

**Constraints**
- `1 <= words.length <= 1000`
- `1 <= words[i].length <= 16`
- `words[i]` consists only of lowercase English letters

**Approaches**

**✅ 1. Dynamic Programming with HashMap (Top-down)**

**Idea:**
- Sort words by length
- Use a HashMap `dp` to store the maximum chain length ending at each word
- For each word, try removing one character at a time to find all possible predecessors

**Java Code:**
```java
class Solution {
    public int longestStrChain(String[] words) {
        Arrays.sort(words, (a, b) -> a.length() - b.length());
        Map<String, Integer> dp = new HashMap<>();
        int maxChainLength = 1;
        for (String word : words) {
            dp.put(word, 1); 
            for (int i = 0; i < word.length(); i++) {
                String predecessor = word.substring(0, i) + word.substring(i + 1); 
                if (dp.containsKey(predecessor)) {
                    dp.put(word, Math.max(dp.get(word), dp.get(predecessor) + 1));
                }
            }
            maxChainLength = Math.max(maxChainLength, dp.get(word));
        }
        return maxChainLength;
    }
}
```

**✅ 2. Recursive + Memoization**

Not as efficient in practice as DP, but conceptually valid.

- For each word, try to recursively build the longest chain of predecessors
- Use memoization to avoid recomputation

```java
class Solution {
    Map<String, Integer> memo;
    Set<String> wordSet;

    public int longestStrChain(String[] words) {
        memo = new HashMap<>();
        wordSet = new HashSet<>(Arrays.asList(words));
        int maxLen = 0;

        for (String word : words) {
            maxLen = Math.max(maxLen, dfs(word));
        }
        return maxLen;
    }

    private int dfs(String word) {
        if (memo.containsKey(word)) return memo.get(word);
        int max = 1;
        for (int i = 0; i < word.length(); i++) {
            String pred = word.substring(0, i) + word.substring(i + 1);
            if (wordSet.contains(pred)) {
                max = Math.max(max, 1 + dfs(pred));
            }
        }
        memo.put(word, max);
        return max;
    }
}
```

**Summary**

| Approach           | Time Complexity | Space Complexity | Notes                           |
|--------------------|------------------|-------------------|----------------------------------|
| DP + HashMap       | O(N * L²)        | O(N * L)          | Best and most efficient method  |
| DFS + Memoization  | O(N * L²)        | O(N * L)          | Intuitive but slightly slower   |

- `N` is the number of words
- `L` is the average word length

**Tips**

- Sort words by length to ensure all predecessors are processed first
- Use substring manipulation to generate possible predecessors
- A single word is a valid chain of length 1

**Conclusion**

This problem is a great exercise in dynamic programming with smart ordering and string manipulation. Sorting + memoization or tabulation with a map makes it efficient and clean.

## Problem Statement: Number of Longest Increasing Subsequence
Given an integer array `nums`, return the **number of longest increasing subsequences**.
A subsequence must be **strictly increasing**.

**Example**
**Input:**
```java
nums = [1,3,5,4,7]
```

**Output:**
```java
2
```

**Explanation:**
The LIS is `[1,3,4,7]` and `[1,3,5,7]`, both of length 4.

**Constraints**
- `1 <= nums.length <= 2000`
- `-10⁶ <= nums[i] <= 10⁶`

**Approach**
**✅ Dynamic Programming (LIS + Count)**

We need to calculate:
1. `length[i]`: the length of the longest increasing subsequence ending at index `i`
2. `count[i]`: the number of such subsequences ending at `i`

We iterate through the array and for each pair `(j, i)` where `j < i` and `nums[j] < nums[i]`, we update:
- `length[i]` if `length[j] + 1 > length[i]`
- `count[i]` based on how `length[i]` changes

**Java Code:**
```java
class Solution {
    public int findNumberOfLIS(int[] nums) {
        if (nums.length == 0) return 0;

        int n = nums.length;
        int[] length = new int[n]; 
        int[] count = new int[n];
        
        Arrays.fill(length, 1);
        Arrays.fill(count, 1);
        
        int maxLength = 1;

        for (int i = 1; i < n; i++) {
            for (int j = 0; j < i; j++) {
                if (nums[i] > nums[j]) {
                    if (length[j] + 1 > length[i]) {
                        length[i] = length[j] + 1;
                        count[i] = count[j];
                    } else if (length[j] + 1 == length[i]) {
                        count[i] += count[j];
                    }
                }
            }
            maxLength = Math.max(maxLength, length[i]);
        }

        int result = 0;
        for (int i = 0; i < n; i++) {
            if (length[i] == maxLength) {
                result += count[i];
            }
        }

        return result;
    }
}
```

**Time & Space Complexity**
| Metric            | Value          |
|-------------------|----------------|
| Time Complexity   | `O(n^2)`       |
| Space Complexity  | `O(n)`         |

**Other Approaches:**
1. **Optimized Dynamic Programming with Binary Search (Using a `TreeMap` or `List`)**

In the optimized approach, we use a combination of **Binary Search** and **Dynamic Programming** to solve the problem in **O(n log n)** time complexity.

**Explanation:**
- Use a dynamic programming array `dp` where `dp[i]` will store the length of the longest increasing subsequence that ends at `i`.
- Use a **TreeMap** or **List** to keep track of the count of the longest subsequences found at each length.
- For each new number, we perform a **binary search** to find its appropriate position and update both the length and count of the subsequences.
- Finally, sum all the subsequences that end with the maximum LIS length.

**Java Code:**
```java
import java.util.*;

class Solution {
    public int findNumberOfLIS(int[] nums) {
        if (nums.length == 0) {
            return 0;
        }
        
        List<List<Integer>> dp = new ArrayList<>();
        int maxLength = 0;
        
        for (int num : nums) {
            List<Integer> temp = new ArrayList<>();
            for (int i = 0; i < dp.size(); i++) {
                if (dp.get(i).get(dp.get(i).size() - 1) < num) {
                    temp.add(num);
                    dp.add(temp);
                }
            }
            maxLength = Math.max(maxLength, temp.size());
        }
        
        return dp.size();
    }
}
```

2. **Backtracking + Memoization (Recursive Approach)**

In this approach, we perform **Backtracking** to explore all possible subsequences, while using **Memoization** to cache results and avoid redundant calculations.

**Explanation:**

- Use a recursive function to explore all increasing subsequences, trying to build them one by one.
- Store intermediate results in a memoization table to avoid recalculating the same subsequences.
- At each step, either include the next number in the sequence or skip it.

**Java Code:**
```java
class Solution {
    public int findNumberOfLIS(int[] nums) {
        int n = nums.length;
        Integer[][] memo = new Integer[n][n];
        return findWays(nums, 0, -1, memo);
    }

    private int findWays(int[] nums, int index, int prevIndex, Integer[][] memo) {
        if (index == nums.length) {
            return 1;
        }
        
        if (memo[index][prevIndex + 1] != null) {
            return memo[index][prevIndex + 1];
        }

        int include = 0;
        if (prevIndex == -1 || nums[index] > nums[prevIndex]) {
            include = findWays(nums, index + 1, index, memo);
        }

        int exclude = findWays(nums, index + 1, prevIndex, memo);

        memo[index][prevIndex + 1] = include + exclude;
        return memo[index][prevIndex + 1];
    }
}
```

3. **Top-Down Approach (Using a Single Array for Storing Subsequence Counts)**

Another variation of the dynamic programming approach, which keeps track of the subsequence counts in a **single array**, also has an optimization for the subsequence count.

**Explanation:**

- For each element in the list, look for all the previous elements that can extend the subsequence. 
- If a subsequence can be extended, increment its count in a `count[]` array.
- At the end, return the sum of the counts of subsequences that reach the maximum length.

**Java Code:**

```java
class Solution {
    public int findNumberOfLIS(int[] nums) {
        int n = nums.length;
        int[] dp = new int[n];  // stores the length of the longest subsequence
        int[] count = new int[n];  // stores the count of longest subsequences
        
        Arrays.fill(dp, 1);  // Minimum length for each element is 1 (the element itself)
        Arrays.fill(count, 1);  // Each element is initially its own subsequence
        
        int maxLen = 1;
        for (int i = 1; i < n; i++) {
            for (int j = 0; j < i; j++) {
                if (nums[i] > nums[j]) {
                    if (dp[j] + 1 > dp[i]) {
                        dp[i] = dp[j] + 1;
                        count[i] = count[j];
                    } else if (dp[j] + 1 == dp[i]) {
                        count[i] += count[j];
                    }
                }
            }
            maxLen = Math.max(maxLen, dp[i]);
        }
        
        int result = 0;
        for (int i = 0; i < n; i++) {
            if (dp[i] == maxLen) {
                result += count[i];
            }
        }
        
        return result;
    }
}
```

**Comparison of Approaches**
| Approach                        | Time Complexity | Space Complexity |
|----------------------------------|-----------------|------------------|
| **Dynamic Programming (DP)**     | O(n²)           | O(n)             |
| **Optimized DP with Binary Search** | O(n log n)     | O(n)             |
| **Backtracking + Memoization**   | O(2^n)          | O(n)             |
| **Top-Down DP Approach**         | O(n²)           | O(n)             |

- **Dynamic Programming (DP)**: Easy to implement and effective for smaller problem sizes. The time complexity is **O(n²)**, which can be slow for larger `n`.
- **Optimized DP with Binary Search**: This is faster with **O(n log n)** time complexity, but requires more careful implementation with binary search.
- **Backtracking + Memoization**: Can be inefficient with an exponential time complexity but works well for smaller problem sizes with fewer elements.
- **Top-Down DP**: Similar to the standard DP approach but ensures better memory utilization.


**Summary**
| Component   | Explanation                                      |
|-------------|--------------------------------------------------|
| `length[]`  | Stores LIS length at each index                  |
| `count[]`   | Stores number of LIS ending at each index        |
| Final step  | Sum all `count[i]` where `length[i] == maxLength`|

**Key Insight**
This is a variation of the classic Longest Increasing Subsequence (LIS) problem where we also track **how many** such sequences exist, not just their length.

## Problem Statement: Longest Palindromic Subsequence

Given a string `s`, find the length of the longest palindromic subsequence in `s`.
A **subsequence** is a sequence that can be derived from another sequence by deleting some or no elements without changing the order of the remaining elements.

**Approach 1: Dynamic Programming (Bottom-Up)**
**Explanation:**
- We use dynamic programming to find the longest palindromic subsequence.
- We create a 2D array `dp` where `dp[i][j]` represents the length of the longest palindromic subsequence in the substring `s[i..j]`.
- Base Case: Every individual character is a palindrome of length 1, so `dp[i][i] = 1` for all `i`.
- If characters at positions `i` and `j` are the same, then we add 2 to the result of the subproblem `dp[i + 1][j - 1]`. Otherwise, we take the maximum of the two possible subproblems: `dp[i + 1][j]` and `dp[i][j - 1]`.

**Java Code:**
```java
class Solution {
    public int longestPalindromeSubseq(String s) {
        int n = s.length();
        int[][] dp = new int[n][n];
        
        // Base case: single characters are palindromes
        for (int i = 0; i < n; i++) {
            dp[i][i] = 1;
        }

        // Solve for substrings of length 2 to n
        for (int len = 2; len <= n; len++) {
            for (int i = 0; i <= n - len; i++) {
                int j = i + len - 1; 
                
                if (s.charAt(i) == s.charAt(j)) {
                    dp[i][j] = dp[i + 1][j - 1] + 2;
                } else {
                    dp[i][j] = Math.max(dp[i + 1][j], dp[i][j - 1]);
                }
            }
        }
        
        return dp[0][n - 1];
    }
}
```

**Time Complexity:**
- **O(n^2)**: We fill a 2D DP table where `n` is the length of the string.

**Space Complexity:**
- **O(n^2)**: We use a 2D DP array to store the intermediate results.

**Approach 2: Dynamic Programming (Space Optimization)**

**Explanation:**
- We can optimize the space used by the above dynamic programming approach.
- Instead of storing the entire 2D DP array, we only need the current row and the previous row to compute the result.
- This reduces the space complexity from **O(n^2)** to **O(n)**.

**Java Code:**

```java
class Solution {
    public int longestPalindromeSubseq(String s) {
        int n = s.length();
        int[] dp = new int[n];
        int[] prev = new int[n];

        // Base case: single characters are palindromes
        for (int i = 0; i < n; i++) {
            dp[i] = 1;
        }

        // Solve for substrings of length 2 to n
        for (int len = 2; len <= n; len++) {
            for (int i = 0; i < n - len + 1; i++) {
                int j = i + len - 1;
                if (s.charAt(i) == s.charAt(j)) {
                    dp[j] = prev[j - 1] + 2;
                } else {
                    dp[j] = Math.max(dp[j], dp[j - 1]);
                }
            }
            System.arraycopy(dp, 0, prev, 0, n);
        }

        return dp[n - 1];
    }
}
```

**Time Complexity:**
- **O(n^2)**: We still iterate over all the pairs of indices, but now we only keep track of two rows instead of the entire table.

**Space Complexity:**
- **O(n)**: We use two 1D arrays instead of a 2D array.

**Approach 3: Recursive with Memoization**

**Explanation:**
- In this approach, we use recursion to explore all possible subsequences, and memoization to store the results of subproblems to avoid recalculating them.
- We recursively check if the characters at positions `i` and `j` are the same, and either include them in the subsequence or explore further by excluding one character at a time.

**Java Code:**

```java
class Solution {
    public int longestPalindromeSubseq(String s) {
        int n = s.length();
        Integer[][] memo = new Integer[n][n];
        return helper(s, 0, n - 1, memo);
    }

    private int helper(String s, int i, int j, Integer[][] memo) {
        if (i >= j) {
            return i == j ? 1 : 0;
        }
        
        if (memo[i][j] != null) {
            return memo[i][j];
        }

        if (s.charAt(i) == s.charAt(j)) {
            memo[i][j] = helper(s, i + 1, j - 1, memo) + 2;
        } else {
            memo[i][j] = Math.max(helper(s, i + 1, j, memo), helper(s, i, j - 1, memo));
        }

        return memo[i][j];
    }
}
```

**Time Complexity:**
- **O(n^2)**: Memoization ensures that each subproblem is solved only once.

**Space Complexity:**
- **O(n^2)**: We store results of subproblems in a memoization table.

**Approach 4: Bottom-Up Dynamic Programming with String Reversal**

**Explanation:**
- This approach uses the idea of **longest common subsequence (LCS)**, where we treat the problem as finding the LCS between the string and its reverse.
- The LCS between the string `s` and its reverse is essentially the longest palindromic subsequence of `s`.

**Java Code:**
```java
class Solution {
    public int longestPalindromeSubseq(String s) {
        String reverseS = new StringBuilder(s).reverse().toString();
        return longestCommonSubsequence(s, reverseS);
    }

    private int longestCommonSubsequence(String s1, String s2) {
        int m = s1.length();
        int n = s2.length();
        int[][] dp = new int[m + 1][n + 1];

        for (int i = 1; i <= m; i++) {
            for (int j = 1; j <= n; j++) {
                if (s1.charAt(i - 1) == s2.charAt(j - 1)) {
                    dp[i][j] = dp[i - 1][j - 1] + 1;
                } else {
                    dp[i][j] = Math.max(dp[i - 1][j], dp[i][j - 1]);
                }
            }
        }

        return dp[m][n];
    }
}
```

**Time Complexity:**
- **O(n^2)**: We compute the LCS for two strings of length `n`.

**Space Complexity:**
- **O(n^2)**: We use a 2D array to store the LCS values.

**Comparison of Approaches**
| Approach                              | Time Complexity | Space Complexity |
|---------------------------------------|-----------------|------------------|
| **Dynamic Programming (Bottom-Up)**   | O(n²)           | O(n²)            |
| **Space Optimized Dynamic Programming** | O(n²)           | O(n)             |
| **Recursive with Memoization**        | O(n²)           | O(n²)            |
| **Bottom-Up with String Reversal (LCS)** | O(n²)         | O(n²)            |

**Conclusion:**
- The **Bottom-Up Dynamic Programming** is the most intuitive and efficient approach for solving this problem in **O(n²)** time complexity.
- The **Space Optimized DP** approach reduces the space complexity to **O(n)**, making it a better choice for larger strings.
- The **Recursive with Memoization** approach is slower due to the overhead of recursion and memoization, but it is simple and effective for smaller problems.
- The **LCS with String Reversal** is another approach to solve this problem, which is effective if you're already familiar with the LCS problem.



## Problem Statement: Minimum Insertion Steps to Make a String Palindrome
Given a string `s`, in one step you can insert any character at any index of the string.
Return the minimum number of steps to make the string a palindrome.
A **Palindrome** string is one that reads the same backward as well as forward.

**Approach 1: Dynamic Programming**

**Explanation:**
- We can solve this problem using dynamic programming by first finding the **longest palindromic subsequence** (LPS) of the string.
- The minimum insertions required will be the difference between the length of the string and the length of its longest palindromic subsequence: `minInsertions = n - LPS`.
- We use a 2D DP table where `dp[i][j]` represents the length of the longest palindromic subsequence in the substring `s[i..j]`.

**Steps:**
1. Initialize a 2D DP table where `dp[i][i] = 1` (base case: a single character is a palindrome of length 1).
2. For substrings of length 2 to `n`, if the characters at positions `i` and `j` are the same, then `dp[i][j] = dp[i+1][j-1] + 2`. Otherwise, take the maximum between the two possible subproblems `dp[i+1][j]` and `dp[i][j-1]`.
3. The length of the longest palindromic subsequence will be found in `dp[0][n-1]`.
4. The minimum insertions required will be `n - LPS`.

**Java Code:**

```java
class Solution {
    public int minInsertions(String s) {
        int n = s.length();
        int[][] dp = new int[n][n];

        // Base case: single characters are palindromes of length 1
        for (int i = 0; i < n; i++) {
            dp[i][i] = 1;
        }

        // Solve for substrings of length 2 to n
        for (int len = 2; len <= n; len++) {
            for (int i = 0; i <= n - len; i++) {
                int j = i + len - 1;
                
                if (s.charAt(i) == s.charAt(j)) {
                    dp[i][j] = dp[i + 1][j - 1] + 2;
                } else {
                    dp[i][j] = Math.max(dp[i + 1][j], dp[i][j - 1]);
                }
            }
        }
        
        // Length of Longest Palindromic Subsequence (LPS)
        int lps = dp[0][n - 1];
        
        // Minimum insertions required
        return n - lps;
    }
}
```

**Time Complexity:**
- **O(n²)**: We iterate through all pairs of indices `i` and `j` to fill the DP table.

**Space Complexity:**
- **O(n²)**: We use a 2D array to store the values of the DP table.

**Approach 2: Recursive with Memoization (Top-Down)**

**Explanation:**
- We can also solve this problem using recursion with memoization to avoid recalculating subproblems.
- The recursive formula is similar to the DP approach:
  - If the characters `s[i]` and `s[j]` are the same, then the result for `s[i..j]` will be `2 + longestPalindromicSubsequence(s, i+1, j-1)`.
  - If `s[i] != s[j]`, then we need to either skip `s[i]` or `s[j]`, so the result will be `max(longestPalindromicSubsequence(s, i+1, j), longestPalindromicSubsequence(s, i, j-1))`.

**Java Code:**

```java
class Solution {
    public int minInsertions(String s) {
        int n = s.length();
        Integer[][] memo = new Integer[n][n];
        return n - longestPalindromicSubsequence(s, 0, n - 1, memo);
    }

    private int longestPalindromicSubsequence(String s, int i, int j, Integer[][] memo) {
        if (i >= j) {
            return (i == j) ? 1 : 0;
        }
        
        if (memo[i][j] != null) {
            return memo[i][j];
        }
        
        if (s.charAt(i) == s.charAt(j)) {
            memo[i][j] = longestPalindromicSubsequence(s, i + 1, j - 1, memo) + 2;
        } else {
            memo[i][j] = Math.max(longestPalindromicSubsequence(s, i + 1, j, memo), longestPalindromicSubsequence(s, i, j - 1, memo));
        }
        
        return memo[i][j];
    }
}
```

**Time Complexity:**
- **O(n²)**: Each subproblem is solved once, and there are `O(n²)` subproblems.

**Space Complexity:**
- **O(n²)**: We store the results of subproblems in the `memo` table.

**Approach 3: Greedy and Two-Pointer Approach (Not Optimal)**

**Explanation:**
- A **greedy approach** could be considered, where we use two pointers to check for the first and last mismatch and try to insert characters optimally. However, this approach does not guarantee correctness and is not as efficient as the dynamic programming solution.

**Java Code (Greedy Example, For Reference):**

```java
class Solution {
    public int minInsertions(String s) {
        int left = 0, right = s.length() - 1;
        int insertions = 0;
        
        while (left < right) {
            if (s.charAt(left) == s.charAt(right)) {
                left++;
                right--;
            } else {
                insertions++;
                // move one of the pointers
                if (s.charAt(left) == s.charAt(right - 1)) {
                    right--;
                } else {
                    left++;
                }
            }
        }
        
        return insertions;
    }
}
```

**Time Complexity:**
- **O(n)**: Linear time complexity for this greedy approach.

**Space Complexity:**
- **O(1)**: Constant space complexity as no additional space is required.

**Comparison of Approaches**
| Approach                         | Time Complexity | Space Complexity |
|----------------------------------|-----------------|------------------|
| **Dynamic Programming (Bottom-Up)** | O(n²)           | O(n²)            |
| **Recursive with Memoization**   | O(n²)           | O(n²)            |
| **Greedy and Two-Pointer (Not Optimal)** | O(n)        | O(1)             |

**Conclusion:**
- The **Dynamic Programming** approach is the most efficient and optimal solution to this problem, both in terms of time complexity (`O(n²)`) and space complexity (`O(n²)`).
- **Recursive with Memoization** provides a good alternative but also has the same time and space complexity as the bottom-up approach.
- The **Greedy and Two-Pointer** approach can be a faster solution in some cases, but it does not guarantee correctness, so it's less reliable.

## Problem Statement: Best Time to Buy and Sell Stock

You are given an array `prices` where `prices[i]` is the price of a given stock on the `i-th` day.
You want to maximize your profit by choosing a single day to buy one stock and a different day in the future to sell that stock.
**Return** the maximum profit you can achieve from this transaction. If you cannot achieve any profit, return `0`.

**Approach 1: One Pass (Greedy Approach)**
**Explanation**
- Traverse the array once while tracking the **minimum price so far** and the **maximum profit**.
- For each price, calculate the potential profit (current price - min price so far).
- Update the maximum profit if this is better than the current best.

**Code (Java)**

```java
class Solution {
    public int maxProfit(int[] prices) {
        if (prices.length < 2) return 0;

        int minPrice = prices[0];
        int maxProfit = 0;

        for (int i = 1; i < prices.length; i++) {
            if (prices[i] < minPrice) {
                minPrice = prices[i];
            } else {
                int profit = prices[i] - minPrice;
                maxProfit = Math.max(maxProfit, profit);
            }
        }

        return maxProfit;
    }
}
```

**Time Complexity**
- **O(n)**: Single pass through the array.

**Space Complexity**
- **O(1)**: Constant space.

**Approach 2: Brute Force**

**Explanation**
- Check every pair of days `(i, j)` where `j > i` and calculate the profit `prices[j] - prices[i]`.
- Keep track of the maximum profit seen.

**Code (Java)**

```java
class Solution {
    public int maxProfit(int[] prices) {
        int maxProfit = 0;
        for (int i = 0; i < prices.length; i++) {
            for (int j = i + 1; j < prices.length; j++) {
                int profit = prices[j] - prices[i];
                if (profit > maxProfit) {
                    maxProfit = profit;
                }
            }
        }
        return maxProfit;
    }
}
```

**Time Complexity**
- **O(n²)**: Nested loop over all pairs.

**Space Complexity**
- **O(1)**

**Approach 3: Dynamic Programming (Overkill for This Problem)**

**Explanation**
- Define `dp[i]` as the maximum profit we can get by selling on day `i`.
- Track the minimum price seen so far as you iterate.

**Code (Java)**

```java
class Solution {
    public int maxProfit(int[] prices) {
        int n = prices.length;
        if (n == 0) return 0;

        int[] dp = new int[n];
        int minPrice = prices[0];
        int maxProfit = 0;

        for (int i = 1; i < n; i++) {
            minPrice = Math.min(minPrice, prices[i]);
            dp[i] = Math.max(dp[i - 1], prices[i] - minPrice);
            maxProfit = Math.max(maxProfit, dp[i]);
        }

        return maxProfit;
    }
}
```

**Time Complexity**
- **O(n)**

**Space Complexity**
- **O(n)** (can be reduced to O(1) using just two variables)

**Comparison of Approaches**
| Approach         | Time Complexity | Space Complexity | Notes                          |
|------------------|------------------|------------------|--------------------------------|
| Greedy (1 Pass)  | O(n)             | O(1)             | Optimal and most efficient     |
| Brute Force      | O(n²)            | O(1)             | Inefficient, only for learning |
| Dynamic Programming | O(n)         | O(n)             | Overkill for this problem      |

**Conclusion**
- The **Greedy One-Pass** solution is the most efficient and widely used method to solve this problem.
- The **Brute Force** method helps understand the core concept but is not viable for large inputs.
- **DP** is useful in more complex stock trading problems but unnecessary here.

## Problem Statement: Best Time to Buy and Sell Stock II

You are given an integer array `prices` where `prices[i]` is the price of a given stock on the `i-th` day.
On each day, you may decide to buy and/or sell the stock. You can only hold at most one share of the stock at any time. However, you can buy it and then immediately sell it on the same day.
**Return** the maximum profit you can achieve.

**Approach 1: Greedy**

**Explanation**
- To maximize profit, buy whenever there's a price increase.
- Add all positive price differences (profit from every upward movement).

**Code (Java)**

```java
class Solution {
    public int maxProfit(int[] prices) {
        int profit = 0;
        for (int i = 1; i < prices.length; i++) {
            if (prices[i] > prices[i - 1]) {
                profit += prices[i] - prices[i - 1];
            }
        }
        return profit;
    }
}
```

**Time Complexity**
- **O(n)**: Single pass through prices.

**Space Complexity**
- **O(1)**

**Approach 2: Peak-Valley Approach**

**Explanation**
- Find valleys (local minima) to buy and peaks (local maxima) to sell.
- Accumulate profit as `peak - valley`.

**Code (Java)**

```java
class Solution {
    public int maxProfit(int[] prices) {
        int i = 0;
        int profit = 0;
        int n = prices.length;
        while (i < n - 1) {
            while (i < n - 1 && prices[i] >= prices[i + 1]) i++; // find valley
            int valley = prices[i];

            while (i < n - 1 && prices[i] <= prices[i + 1]) i++; // find peak
            int peak = prices[i];

            profit += peak - valley;
        }
        return profit;
    }
}
```

**Time Complexity**
- **O(n)**

**Space Complexity**
- **O(1)**

**Approach 3: Dynamic Programming**

**Explanation**
- Maintain two states:
  - `dp[i][0]`: Max profit on day `i` if not holding a stock.
  - `dp[i][1]`: Max profit on day `i` if holding a stock.

**Code (Java)**

```java
class Solution {
    public int maxProfit(int[] prices) {
        int n = prices.length;
        if (n == 0) return 0;

        int[][] dp = new int[n][2];
        dp[0][0] = 0;            // not holding stock
        dp[0][1] = -prices[0];   // holding stock

        for (int i = 1; i < n; i++) {
            dp[i][0] = Math.max(dp[i - 1][0], dp[i - 1][1] + prices[i]);
            dp[i][1] = Math.max(dp[i - 1][1], dp[i - 1][0] - prices[i]);
        }

        return dp[n - 1][0];
    }
}
```

**Time Complexity**
- **O(n)**

**Space Complexity**
- **O(n)** (can be reduced to **O(1)** using two variables)

**Space Optimized DP**

```java
class Solution {
    public int maxProfit(int[] prices) {
        int hold = -prices[0];
        int cash = 0;

        for (int i = 1; i < prices.length; i++) {
            int prevCash = cash;
            cash = Math.max(cash, hold + prices[i]);
            hold = Math.max(hold, prevCash - prices[i]);
        }

        return cash;
    }
}
```

**Summary**
| Approach            | Time Complexity | Space Complexity | Description                          |
|---------------------|------------------|------------------|--------------------------------------|
| Greedy              | O(n)             | O(1)             | Best and most concise solution       |
| Peak-Valley         | O(n)             | O(1)             | Simulates buy-sell points            |
| Dynamic Programming | O(n)             | O(n) → O(1)      | Useful when considering constraints  |

**Conclusion**
- The **Greedy** approach is the most efficient for this problem.
- The **DP** approach is educational and extensible for more complex constraints (e.g., cooldowns or fees).

**Problem Statement: Best Time to Buy and Sell Stock III**

You are given an array `prices` where `prices[i]` is the price of a given stock on the `i-th` day.
Find the **maximum profit** you can achieve. You may complete **at most two transactions**.
> Note: You may not engage in multiple transactions simultaneously (i.e., you must sell the stock before you buy again).

**Approach 1: Dynamic Programming (2D DP Table)**

**Idea**
We define `dp[k][i]` as the maximum profit using at most `k` transactions up to day `i`.
We loop `k` from `1` to `2` and use `maxDiff` to keep track of the best profit we can get by buying stock on a previous day.

**Code (Java)**

```java
class Solution {
    public int maxProfit(int[] prices) {
        if (prices.length < 2) {
            return 0;
        }

        int n = prices.length;
        int[][] dp = new int[3][n];
        
        for (int k = 1; k <= 2; k++) {
            int maxDiff = -prices[0];
            for (int i = 1; i < n; i++) {
                dp[k][i] = Math.max(dp[k][i - 1], prices[i] + maxDiff);
                maxDiff = Math.max(maxDiff, dp[k - 1][i] - prices[i]);
            }
        }

        return dp[2][n - 1];
    }
}
```

**Time Complexity**
- **O(n \* k)** = O(n), since k = 2

**Space Complexity**
- **O(n \* k)** = O(n)

**Approach 2: Optimized DP with Constant Space**

**Idea**
Instead of keeping a 2D table, maintain variables representing the best values for each transaction step:
- `buy1`, `sell1`, `buy2`, `sell2`

**Code (Java)**

```java
class Solution {
    public int maxProfit(int[] prices) {
        int buy1 = Integer.MIN_VALUE, sell1 = 0;
        int buy2 = Integer.MIN_VALUE, sell2 = 0;

        for (int price : prices) {
            buy1 = Math.max(buy1, -price);
            sell1 = Math.max(sell1, buy1 + price);
            buy2 = Math.max(buy2, sell1 - price);
            sell2 = Math.max(sell2, buy2 + price);
        }

        return sell2;
    }
}
```

**Time Complexity**
- **O(n)**

**Space Complexity**
- **O(1)**

**Summary**
| Approach                 | Time Complexity | Space Complexity | Description                            |
|--------------------------|-----------------|------------------|----------------------------------------|
| 2D DP Table              | O(n)            | O(n)             | Tracks max profit for 1st and 2nd txn  |
| Optimized Constant Space | O(n)            | O(1)             | Uses variables to track state changes  |

**Notes**
- This problem is a special case of **Best Time to Buy and Sell Stock IV**, where **k = 2**.
- The optimized solution is both fast and elegant.

## Problem Statement: Best Time to Buy and Sell Stock IV

You are given an integer array `prices` where `prices[i]` is the price of a given stock on the `i-th` day, and an integer `k`.
Find the maximum profit you can achieve. You may complete **at most k transactions**.
> Note: You may not engage in multiple transactions simultaneously (i.e., you must sell the stock before you buy again).

**Approach 1: Dynamic Programming (2D DP Table)**

**Idea**
Use a DP table where `dp[t][d]` is the maximum profit on day `d` with at most `t` transactions.
For each transaction count `t`, we keep track of the best `maxDiff = dp[t-1][j] - prices[j]` so far to efficiently calculate `dp[t][i]`.

**Optimization for High `k`:**
If `k >= prices.length / 2`, the problem becomes **Best Time to Buy and Sell Stock II**, where we can complete unlimited transactions.

**Java Code**

```java
class Solution {
    public int maxProfit(int k, int[] prices) {
        if (prices == null || prices.length == 0) {
            return 0;
        }

        int n = prices.length;
        if (k >= n / 2) {
            return unlimitedTransactions(prices);
        }

        int[][] dp = new int[k + 1][n];
        for (int t = 1; t <= k; t++) {
            int maxDiff = -prices[0];
            for (int d = 1; d < n; d++) {
                dp[t][d] = Math.max(dp[t][d - 1], prices[d] + maxDiff);
                maxDiff = Math.max(maxDiff, dp[t - 1][d] - prices[d]);
            }
        }

        return dp[k][n - 1];
    }

    private int unlimitedTransactions(int[] prices) {
        int profit = 0;
        for (int i = 1; i < prices.length; i++) {
            if (prices[i] > prices[i - 1]) {
                profit += prices[i] - prices[i - 1];
            }
        }
        return profit;
    }
}
```

**Time Complexity**
- **O(k * n)**

**Space Complexity**
- **O(k * n)**

**Approach 2: Optimized DP with Rolling Arrays**

**Idea**
We can optimize space to **O(n)** using two rolling arrays: one for the current transaction and one for the previous.

**Code Snippet**

```java
int[] prev = new int[n];
int[] curr = new int[n];
for (int t = 1; t <= k; t++) {
    int maxDiff = -prices[0];
    for (int d = 1; d < n; d++) {
        curr[d] = Math.max(curr[d - 1], prices[d] + maxDiff);
        maxDiff = Math.max(maxDiff, prev[d] - prices[d]);
    }
    int[] temp = prev;
    prev = curr;
    curr = temp;
}
```

**Time Complexity**
- **O(k * n)**

**Space Complexity**
- **O(n)**

**Summary**
| Approach                     | Time Complexity | Space Complexity | Notes                                   |
|------------------------------|-----------------|------------------|-----------------------------------------|
| Full 2D DP Table             | O(k \* n)       | O(k \* n)        | Simple and intuitive                    |
| Rolling Arrays Optimization  | O(k \* n)       | O(n)             | Memory optimized                        |
| Greedy (Unlimited txns case) | O(n)            | O(1)             | Used when k ≥ n/2                       |

**Key Insight**
If `k` is large relative to the number of days, this becomes the simpler **unlimited transactions** problem — essentially turning this hard problem into an easy greedy one.

## Problem Statement: Best Time to Buy and Sell Stock with Cooldown

You are given an array `prices` where `prices[i]` is the price of a given stock on the `i-th` day.
Find the maximum profit you can achieve. You may complete as many transactions as you like (i.e., buy and sell one share of the stock multiple times) **with the following restrictions**:
- After you sell your stock, you **cannot buy stock on the next day** (i.e., cooldown one day).
- You may not engage in multiple transactions simultaneously (i.e., you must sell the stock before you buy again).

**Approach 1: Dynamic Programming (Three States)**
**Intuition**

On each day, you can be in one of three states:
1. **Buy state**: Holding a stock.
2. **Sell state**: Just sold the stock.
3. **Cooldown state**: Doing nothing (rest day).

**Transitions:**
- `buy[i] = max(buy[i-1], cooldown[i-1] - prices[i])`
- `sell[i] = max(sell[i-1], buy[i-1] + prices[i])`
- `cooldown[i] = max(cooldown[i-1], sell[i-1])`

**Java Code**

```java
class Solution {
    public int maxProfit(int[] prices) {
        if (prices == null || prices.length == 0) return 0;

        int n = prices.length;
        int[] buy = new int[n];
        int[] sell = new int[n];
        int[] cooldown = new int[n];

        buy[0] = -prices[0];
        sell[0] = 0;
        cooldown[0] = 0;

        for (int i = 1; i < n; i++) {
            buy[i] = Math.max(buy[i - 1], cooldown[i - 1] - prices[i]);
            sell[i] = Math.max(sell[i - 1], buy[i - 1] + prices[i]);
            cooldown[i] = Math.max(cooldown[i - 1], sell[i - 1]);
        }

        return Math.max(sell[n - 1], cooldown[n - 1]);
    }
}
```

**Time Complexity**
- **O(n)** — One pass through the price array.

**Space Complexity**
- **O(n)** — Arrays for each state.

**Approach 2: Space Optimized Dynamic Programming**

**Idea**
Instead of using arrays, we can optimize the space to constant by storing only previous values of each state.

**Code Snippet**
```java
class Solution {
    public int maxProfit(int[] prices) {
        int buy = -prices[0];
        int sell = 0;
        int cooldown = 0;

        for (int i = 1; i < prices.length; i++) {
            int prevBuy = buy;
            int prevSell = sell;
            int prevCooldown = cooldown;

            buy = Math.max(prevBuy, prevCooldown - prices[i]);
            sell = Math.max(prevSell, prevBuy + prices[i]);
            cooldown = Math.max(prevCooldown, prevSell);
        }

        return Math.max(sell, cooldown);
    }
}
```

**Time Complexity**
- **O(n)**

**Space Complexity**
- **O(1)** — Only a few variables used.

**Summary**

| Approach                        | Time Complexity | Space Complexity | Notes                             |
|---------------------------------|------------------|-------------------|-----------------------------------|
| DP with state arrays            | O(n)             | O(n)              | Clear state transitions           |
| Space-optimized DP              | O(n)             | O(1)              | Efficient with constant space     |

**Key Insight**
The cooldown constraint introduces a "rest" day after selling, requiring careful state tracking for buying decisions. Using state transitions based on previous days allows elegant and efficient solutions.


## Problem Statement:Minimum Cost to Cut a Stick

You are given a wooden stick of length `n` units. The stick is labeled from `0` to `n`. You are also given an integer array `cuts` where `cuts[i]` denotes a position you should perform a cut.
You can change the order of the cuts as you wish. The cost of one cut is the **length of the stick being cut**. After a cut, the stick is split into two smaller sticks.
Return the **minimum total cost** to perform all the cuts.

**Example:**
**Input:**  
`n = 7`, `cuts = [1, 3, 4, 5]`  
**Output:** `16`

**Explanation:**  
One optimal way is to cut at 3 → cost 7, then cut at 5 → cost 4, then cut at 4 → cost 3, then cut at 1 → cost 2 → Total = 16.

**Approach 1: Bottom-Up Interval DP**

**Idea:**
- Add `0` and `n` to the `cuts` array for boundaries.
- Sort the extended cuts.
- Use DP where `dp[i][j]` represents the minimum cost to cut between `cuts[i]` and `cuts[j]`.

**Java Code:**
```java
class Solution {
    public int minCost(int n, int[] cuts) {
        int[] extendedCuts = new int[cuts.length + 2];
        extendedCuts[0] = 0;
        extendedCuts[extendedCuts.length - 1] = n;

        for (int i = 0; i < cuts.length; i++) {
            extendedCuts[i + 1] = cuts[i];
        }
        Arrays.sort(extendedCuts);

        int[][] dp = new int[extendedCuts.length][extendedCuts.length];

        for (int len = 2; len < extendedCuts.length; len++) {
            for (int i = 0; i + len < extendedCuts.length; i++) {
                int j = i + len;
                dp[i][j] = Integer.MAX_VALUE;

                for (int k = i + 1; k < j; k++) {
                    dp[i][j] = Math.min(dp[i][j],
                        dp[i][k] + dp[k][j] + extendedCuts[j] - extendedCuts[i]);
                }
            }
        }

        return dp[0][extendedCuts.length - 1];
    }
}
```

**Time and Space Complexity**

- **Time:** O(m³), where `m = cuts.length + 2`
- **Space:** O(m²)

**Approach 2: Top-Down Recursive with Memoization**

**Idea:**
- Same concept as DP but solved recursively.
- Use memoization to store results of subproblems.

**Java Code:**
```java
class Solution {
    public int minCost(int n, int[] cuts) {
        int[] extendedCuts = new int[cuts.length + 2];
        extendedCuts[0] = 0;
        extendedCuts[extendedCuts.length - 1] = n;
        for (int i = 0; i < cuts.length; i++) {
            extendedCuts[i + 1] = cuts[i];
        }

        Arrays.sort(extendedCuts);
        int[][] memo = new int[extendedCuts.length][extendedCuts.length];

        return dfs(extendedCuts, 0, extendedCuts.length - 1, memo);
    }

    private int dfs(int[] cuts, int left, int right, int[][] memo) {
        if (left + 1 >= right) return 0;
        if (memo[left][right] != 0) return memo[left][right];

        int minCost = Integer.MAX_VALUE;
        for (int i = left + 1; i < right; i++) {
            int cost = cuts[right] - cuts[left] +
                       dfs(cuts, left, i, memo) +
                       dfs(cuts, i, right, memo);
            minCost = Math.min(minCost, cost);
        }

        memo[left][right] = minCost;
        return minCost;
    }
}
```

**Time and Space Complexity**
- **Time:** O(m³) due to all sub-intervals and cuts within them
- **Space:** O(m²) for memo table

**Summary**
| Approach                     | Time Complexity | Space Complexity | Notes                              |
|-----------------------------|-----------------|------------------|------------------------------------|
| Bottom-Up Interval DP       | O(m³)           | O(m²)            | Most efficient and recommended     |
| Top-Down Memoization (DFS)  | O(m³)           | O(m²)            | More intuitive but less performant |

Where `m = cuts.length + 2` (after adding 0 and n)


## Problem Statement: Partition Array for Maximum Sum

Given an integer array `arr`, partition the array into (contiguous) subarrays of length at most `k`. After partitioning, each subarray's values are changed to become the maximum value of that subarray.
Return the **largest sum** of the array after partitioning.
Test cases are generated such that the answer fits in a **32-bit integer**.

**Example**
**Input:**
```java
arr = [1,15,7,9,2,5,10], k = 3
```
**Output:**
```
84
```
**Explanation:**
The best partition is [15,15,15],[9,9,9],[10], which yields sum 84.

**Approach 1: Bottom-Up Dynamic Programming**

**Idea**
- Use a DP array where `dp[i]` represents the maximum sum of subarrays ending at index `i`.
- For every `i`, look back at most `k` elements and evaluate the maximum value in that range.
- Update `dp[i]` using previously computed values.

**Java Code**
```java
class Solution {
    public int maxSumAfterPartitioning(int[] arr, int k) {
        int n = arr.length;
        int[] dp = new int[n];

        for (int i = 0; i < n; i++) {
            int maxVal = 0;
            for (int j = i; j >= Math.max(0, i - k + 1); j--) {
                maxVal = Math.max(maxVal, arr[j]);
                int sum = (j == 0 ? 0 : dp[j - 1]) + maxVal * (i - j + 1);
                dp[i] = Math.max(dp[i], sum);
            }
        }

        return dp[n - 1];
    }
}
```

**Time and Space Complexity**
- **Time:** O(n * k)
- **Space:** O(n)

**Approach 2: Top-Down Dynamic Programming with Memoization**

**Idea**
- Use recursion with memoization.
- At every index, try all partition lengths up to `k`, and recursively compute the result.

**Java Code**
```java
class Solution {
    public int maxSumAfterPartitioning(int[] arr, int k) {
        int n = arr.length;
        Integer[] memo = new Integer[n];
        return dfs(0, arr, k, memo);
    }

    private int dfs(int start, int[] arr, int k, Integer[] memo) {
        if (start == arr.length) return 0;
        if (memo[start] != null) return memo[start];

        int maxVal = 0, maxSum = 0;
        for (int i = start; i < Math.min(arr.length, start + k); i++) {
            maxVal = Math.max(maxVal, arr[i]);
            maxSum = Math.max(maxSum, maxVal * (i - start + 1) + dfs(i + 1, arr, k, memo));
        }

        return memo[start] = maxSum;
    }
}
```

**Time and Space Complexity**
- **Time:** O(n * k)
- **Space:** O(n) for memo + O(n) recursion stack

**Summary**

| Approach                     | Time Complexity | Space Complexity | Notes                            |
|-----------------------------|-----------------|------------------|----------------------------------|
| Bottom-Up DP                | O(n * k)        | O(n)             | Efficient and easy to implement |
| Top-Down with Memoization   | O(n * k)        | O(n)             | Clear recursive structure       |


## Problem Statement: Count Square Submatrices with All Ones
Given a m * n matrix of ones and zeros, return how many square submatrices have all ones.

**Example:**
**Input:**
```plaintext
matrix = [
  [0,1,1,1],
  [1,1,1,1],
  [0,1,1,1]
]
```

**Output:**
```plaintext
15
```

**Explanation:**
There are 10 squares of side 1, 4 squares of side 2, and 1 square of side 3. Total: 10 + 4 + 1 = 15.

**Solution Approaches**

**Approach 1: Dynamic Programming (Bottom-up)**

We will use dynamic programming to solve this problem efficiently. The idea is to maintain a 2D DP table where each entry `dp[i][j]` represents the size of the largest square submatrix with all ones that ends at position `(i, j)`.

**Algorithm:**
1. Initialize a 2D array `dp` of the same size as the input matrix.
2. Iterate through the matrix and for each element `matrix[i][j]`:
   - If the element is 1, calculate the size of the largest square ending at `(i, j)` by checking the minimum of the three adjacent squares:
     - `dp[i-1][j]` (top)
     - `dp[i][j-1]` (left)
     - `dp[i-1][j-1]` (top-left diagonal)
   - Update `dp[i][j]` to be `min(dp[i-1][j], dp[i][j-1], dp[i-1][j-1]) + 1`.
3. Sum up all the values in the `dp` array to get the total count of square submatrices.

**Code Implementation:**

```java
class Solution {
    public int countSquares(int[][] matrix) {
        int m = matrix.length;
        int n = matrix[0].length;
        
        int[][] dp = new int[m][n];
        int count = 0;
        for (int i = 0; i < m; i++) {
            for (int j = 0; j < n; j++) {
                if (matrix[i][j] == 1) {
                    if (i > 0 && j > 0) {
                        dp[i][j] = Math.min(Math.min(dp[i-1][j], dp[i][j-1]), dp[i-1][j-1]) + 1;
                    } else {
                        dp[i][j] = 1;
                    }
                    count += dp[i][j];
                }
            }
        }
        
        return count;
    }
}
```

**Explanation:**
1. The `dp` array stores the size of the largest square submatrix of ones ending at each cell.
2. If `matrix[i][j]` is 1, we check the adjacent cells (`dp[i-1][j]`, `dp[i][j-1]`, `dp[i-1][j-1]`) and take the minimum of these three values and add 1 to find the largest square submatrix ending at `(i, j)`.
3. For each cell in the matrix, we add `dp[i][j]` to the `count` variable to keep track of the total number of square submatrices with all ones.

**Time Complexity:**
- **Time Complexity**: O(m * n), where `m` is the number of rows and `n` is the number of columns in the matrix. We traverse each element of the matrix once.
- **Space Complexity**: O(m * n), due to the space used by the `dp` table.

**Alternative Approaches:**

**Approach 2: Optimized with Space Reduction (1D Array)**

Instead of using a 2D array for the dynamic programming table, we can reduce the space complexity by using a 1D array to store the current row’s values. This approach uses less space and works by storing the previously calculated values, which allows us to only keep track of the current row and the previous row.

**Algorithm:**
1. Use a 1D array `dp` of size `n` to store the current row's values.
2. As we move through the matrix, we calculate the size of the largest square submatrix at each cell based on the previous row's values and the current row's values.
3. To update the `dp` array correctly, we need to store the previous value from the previous row before updating `dp[i][j]`.

**Code Implementation:**

```java
class Solution {
    public int countSquares(int[][] matrix) {
        int m = matrix.length;
        int n = matrix[0].length;
        
        int[] dp = new int[n];
        int prev = 0; // Variable to store the value of dp[i-1][j-1]
        int count = 0;
        
        for (int i = 0; i < m; i++) {
            for (int j = 0; j < n; j++) {
                int temp = dp[j];
                if (matrix[i][j] == 1) {
                    if (i > 0 && j > 0) {
                        dp[j] = Math.min(Math.min(dp[j], dp[j-1]), prev) + 1;
                    } else {
                        dp[j] = 1;
                    }
                    count += dp[j];
                } else {
                    dp[j] = 0;
                }
                prev = temp;  // Save the previous dp[j] value for the next iteration
            }
        }
        
        return count;
    }
}
```

**Explanation:**
1. The `dp` array is used to store the size of the largest square submatrix ending at each position in the current row.
2. The `prev` variable is used to keep track of the previous value of `dp[i-1][j-1]` as we update the `dp` array.
3. The space complexity is reduced to O(n), as we are only using a 1D array instead of a 2D array.

**Time Complexity:**
- **Time Complexity**: O(m * n), where `m` is the number of rows and `n` is the number of columns in the matrix. We still iterate over each element in the matrix.
- **Space Complexity**: O(n), since we are using only a 1D array instead of a 2D array for dynamic programming.

**Approach 3: Brute Force (Not Recommended for Large Inputs)**

A brute force approach would involve checking every possible square submatrix starting from each cell in the matrix. For each square, we would check if all elements inside it are 1. This approach would be inefficient for larger matrices.

**Algorithm:**
1. For each cell `(i, j)` in the matrix, try all possible square submatrices that start at `(i, j)`.
2. For each square, check if all elements are 1.
3. Count how many square submatrices with all ones exist.

**Time Complexity:**
- **Time Complexity**: O(m^2 * n^2), where `m` is the number of rows and `n` is the number of columns. This is because for each cell, we check all possible squares and validate each element.
- **Space Complexity**: O(1), as no extra space is required except for a few variables.

**Conclusion:**
- The **Dynamic Programming** approach with a 2D table is efficient and works well for moderate-sized matrices.
- The **Optimized Space** approach reduces the space complexity to O(n) and is the preferred approach for large matrices.
- The **Brute Force** approach should only be used for small matrices due to its inefficiency.


## Problem Statement: Count all possible paths from top left to bottom right

You are given a matrix of size `m x n` and you need to find the number of unique paths from the top-left corner (position `(0, 0)`) to the bottom-right corner (position `(m-1, n-1)`), given the constraints that from each cell, you can only move either **right** or **down**.

**Input:**
- Two integers `m` and `n` representing the number of rows and columns of the matrix.

**Output:**
- The number of possible paths from the top-left corner to the bottom-right corner.

**Example:**
**Input:**
```java
m = 3, n = 3
```

**Output:**
```java
6
```

**Explanation:** The 6 unique paths in a `3 x 3` grid can be represented as:
- Right, Right, Down, Down
- Right, Down, Right, Down
- Right, Down, Down, Right
- Down, Right, Right, Down
- Down, Right, Down, Right
- Down, Down, Right, Right

**Approaches:**

**1. Dynamic Programming (Bottom-Up Approach)**

In this approach, we use a **2D DP table** to store the number of ways to reach each cell from the top-left corner. At each position `(i, j)` in the matrix, the number of ways to reach it is the sum of the number of ways to reach the cell directly above it `(i-1, j)` and the cell directly to the left of it `(i, j-1)`.

**Steps:**
1. **Base Cases:**
   - There is exactly one way to reach any cell in the first row (move only right). So, set `dp[0][j] = 1` for all `j`.
   - There is exactly one way to reach any cell in the first column (move only down). So, set `dp[i][0] = 1` for all `i`.
   
2. **Recursive Relation:**
   - For each cell `(i, j)` (starting from `(1, 1)`), the number of ways to reach this cell is the sum of the number of ways to reach the cell directly above it and the cell directly to the left of it:
     ```
     dp[i][j] = dp[i-1][j] + dp[i][j-1]
     ```
   
3. **Final Result:**
   - The number of unique paths to the bottom-right corner is stored in `dp[m-1][n-1]`.

**Code:**

```java
class Solution {
    int numberOfPaths(int m, int n) {
        // Initialize a 2D DP table
        int[][] dp = new int[m][n];
        
        // Set base cases: there is only one way to reach any cell in the first row or first column
        for (int i = 0; i < m; i++) dp[i][0] = 1;
        for (int j = 0; j < n; j++) dp[0][j] = 1;
        
        // Fill the DP table
        for (int i = 1; i < m; i++) {
            for (int j = 1; j < n; j++) {
                dp[i][j] = dp[i-1][j] + dp[i][j-1]; // Total paths are the sum of top and left cells
            }
        }
        
        // The result is the number of ways to reach the bottom-right corner
        return dp[m-1][n-1];
    }
}
```

**Explanation:**
1. **Initialization:** The first row and the first column are initialized to 1, because there is only one way to reach any cell in the first row (by moving right) or in the first column (by moving down).
2. **Filling the DP Table:** For each cell `(i, j)`, we compute the number of ways to reach it by adding the values of the cell directly above it (`dp[i-1][j]`) and the cell directly to the left of it (`dp[i][j-1]`).
3. **Result:** The value in `dp[m-1][n-1]` will give the total number of unique paths from the top-left to the bottom-right.

**Time Complexity:**
- **O(m * n)**: We are iterating over every cell in the `m x n` matrix to compute the number of ways to reach each cell.

**Space Complexity:**
- **O(m * n)**: We are using a 2D array `dp` of size `m x n` to store the number of paths.

**2. Optimized Dynamic Programming (Space Optimization)**

In this optimized approach, we notice that we only need the current row and the previous row to calculate the number of paths. Thus, we can reduce the space complexity by using a 1D array instead of a 2D array.

**Steps:**
1. Use a 1D array `dp[]` of size `n`.
2. Initialize `dp[0] = 1` because there is exactly one way to reach any cell in the first column.
3. Iterate through the matrix, and at each step, update the current row based on the previous row.

**Code:**

```java
class Solution {
    int numberOfPaths(int m, int n) {
        // Initialize a 1D DP array
        int[] dp = new int[n];
        
        // Set base case: there is only one way to reach any cell in the first column
        dp[0] = 1;
        
        // Fill the DP array
        for (int i = 1; i < m; i++) {
            for (int j = 1; j < n; j++) {
                dp[j] += dp[j-1]; // Add the number of ways to reach the cell from the left
            }
        }
        
        // The result is the value at the last column
        return dp[n-1];
    }
}
```

**Explanation:**
- We use a 1D array `dp[]` to keep track of the number of ways to reach each column in the current row.
- `dp[j] += dp[j-1]` ensures that we add the number of ways to reach a cell from the left column.
- The final answer is found in `dp[n-1]`, which represents the number of ways to reach the bottom-right corner.

**Time Complexity:**
- **O(m * n)**: We still have to iterate over the matrix, but we're using only a 1D array for storage.

**Space Complexity:**
- **O(n)**: We only use a 1D array of size `n` for storage.

**3. Combinatorial Approach (Mathematical)**
The problem of counting unique paths from the top-left corner to the bottom-right corner in a grid can also be viewed as a combinatorial problem. To get from the top-left to the bottom-right, we need to make `m-1` down moves and `n-1` right moves. The total number of moves is `m + n - 2`, and we need to choose `m - 1` of those moves to be down (or equivalently, `n - 1` to be right). 

The number of ways to do this is given by the binomial coefficient:
```
C(m+n-2, m-1) = (m+n-2)! / ((m-1)! * (n-1)!)
```

**Code:**

```java
class Solution {
    public int numberOfPaths(int m, int n) {
        // Calculate the binomial coefficient C(m+n-2, m-1)
        return factorial(m+n-2) / (factorial(m-1) * factorial(n-1));
    }
    
    // Helper method to calculate factorial
    private int factorial(int num) {
        int result = 1;
        for (int i = 1; i <= num; i++) {
            result *= i;
        }
        return result;
    }
}
```

**Explanation:**
- The factorial method computes the factorial of a number, which is then used to compute the binomial coefficient.
- We compute the number of unique paths using the formula: `C(m+n-2, m-1)`.

**Time Complexity:**
- **O(m + n)**: The factorial computation takes linear time.

**Space Complexity:**
- **O(1)**: No additional space is required apart from the input parameters and a few variables.

**Conclusion:**
- **Dynamic Programming (Bottom-Up)**: This is the most intuitive approach, where we fill a DP table to compute the number of ways to reach each cell.
- **Space Optimization (1D DP)**: This reduces the space complexity to **O(n)** while still keeping the time complexity at **O(m * n)**.
- **Combinatorial Approach**: This approach uses mathematical formulas to directly compute the result, which is very efficient in terms of space complexity.

Each approach has its advantages and can be chosen based on the problem constraints and requirements.


## Problem Statement: Best Time to Buy and Sell Stock with Transaction Fee

You are given an array `prices` where `prices[i]` is the price of a given stock on the `i`-th day, and an integer `fee` representing a transaction fee. Your goal is to find the maximum profit you can achieve. You may complete as many transactions as you like, but you need to pay the transaction fee for each transaction.

**Note:**
- You may not engage in multiple transactions simultaneously (i.e., you must sell the stock before you buy again).
- The transaction fee is only charged once for each stock purchase and sale.

**Input:**
- An array of integers `prices[]` representing the prices of the stock on each day.
- An integer `fee` representing the transaction fee.

**Output:**
- The maximum profit you can achieve.

**Example:**
**Input:**
```java
prices = [1, 3, 2, 8, 4, 9], fee = 2
```

**Output:**
```java
8
```
**Explanation:**
- Buy at day 0 (price 1), sell at day 3 (price 8), profit = 8 - 1 - 2 = 5.
- Buy at day 4 (price 4), sell at day 5 (price 9), profit = 9 - 4 - 2 = 3.
- Total profit = 5 + 3 = 8.

**Approach 1: Dynamic Programming (Optimized with State Tracking)**

In this problem, we can use dynamic programming to keep track of two states:
1. **Hold state**: This represents the maximum profit when we are holding the stock (i.e., we have bought it).
2. **Cash state**: This represents the maximum profit when we are not holding the stock (i.e., we have sold it).

We can define two variables:
- **hold**: The maximum profit if we hold the stock at the end of a day.
- **cash**: The maximum profit if we do not hold the stock at the end of a day.

**Steps:**
1. **Initialization:**
   - On the first day, the maximum profit if we hold the stock is `-prices[0]` (because we buy the stock).
   - The maximum profit if we are not holding the stock is `0` (because no transaction has been made yet).

2. **State Transitions:**
   - On each subsequent day, we update our `hold` and `cash`:
     - To update `hold`, we can either:
       - Keep holding the stock from the previous day (`hold`).
       - Buy the stock on the current day, which means selling the stock before and subtracting the price (`cash - prices[i]`).
     - To update `cash`, we can either:
       - Keep the cash from the previous day (`cash`).
       - Sell the stock on the current day, which means buying it earlier and subtracting the transaction fee (`hold + prices[i] - fee`).

3. **Final Result:**
   - The maximum profit will be in the `cash` variable at the end of the last day because `cash` represents the maximum profit without holding any stock.

**Code:**
```java
class Solution {
    public int maxProfit(int[] prices, int fee) {
        int n = prices.length;
        if (n == 0) return 0;

        // Initialize variables for holding and cash states
        int hold = -prices[0];  // The maximum profit if we hold a stock
        int cash = 0;           // The maximum profit if we do not hold a stock

        // Loop through each day to update the states
        for (int i = 1; i < n; i++) {
            // Update the hold state (either continue holding or buy on day i)
            hold = Math.max(hold, cash - prices[i]);

            // Update the cash state (either continue with cash or sell on day i)
            cash = Math.max(cash, hold + prices[i] - fee);
        }

        // The result will be the maximum profit in the cash state
        return cash;
    }
}
```

**Explanation:**
1. **Initialization:**
   - `hold = -prices[0]`: Initially, if we buy the stock on day 0, the profit is negative, which means we have spent money on the stock.
   - `cash = 0`: Initially, if we don't hold any stock, our profit is zero.

2. **Iterating Over Each Day:**
   - For each day, update the `hold` and `cash` states:
     - `hold = Math.max(hold, cash - prices[i])`: We can either keep the stock from the previous day or buy the stock today (subtract the price from cash).
     - `cash = Math.max(cash, hold + prices[i] - fee)`: We can either keep the cash from the previous day or sell the stock today (add the stock price minus the transaction fee to the profit).

3. **Final Answer:**
   - After iterating through all the days, the result is the value of `cash`, as it represents the maximum profit when no stock is held.

**Approach 2: Brute Force (Naive Approach)**

In this approach, you would have to simulate each possible buy-sell transaction and calculate the profit, considering the transaction fee for each buy and sell action. This method is highly inefficient for large input sizes, as it involves nested loops to calculate the profit for each pair of buy and sell days.

**Steps:**
1. **For each possible pair of days (i, j)**, where `i` < `j`, simulate the process of buying on day `i` and selling on day `j`.
2. **Subtract the fee** from the profit each time a sell occurs.
3. Track the maximum profit obtained across all pairs.

**Time Complexity:**
- **O(n^2)**: Since there are two loops iterating through all possible pairs of days, the time complexity is quadratic, making it inefficient for large inputs.

**Code Example:**
```java
class Solution {
    public int maxProfit(int[] prices, int fee) {
        int maxProfit = 0;
        int n = prices.length;

        // Brute force: check all possible pairs of days (i, j)
        for (int i = 0; i < n; i++) {
            for (int j = i + 1; j < n; j++) {
                // Calculate the profit for buying on day i and selling on day j
                int profit = prices[j] - prices[i] - fee;
                if (profit > maxProfit) {
                    maxProfit = profit;
                }
            }
        }

        return maxProfit;
    }
}
```

**Explanation:**
1. We iterate through all possible days to buy and sell.
2. For each pair, we calculate the profit, subtracting the fee, and update the `maxProfit` if the current transaction yields a higher profit.
3. This brute-force approach is simple but inefficient for large arrays of prices.

**Approach 3: Greedy Approach (Less Common, but Applicable)**

In some cases, a greedy approach can be applied to determine whether we should buy or sell based on the previous day’s price and the fee. However, for this specific problem, the dynamic programming approach is far more efficient and accurate. A greedy approach might miss opportunities where multiple transactions are involved, making it a less preferred choice.

**Time Complexity:**
- **O(n)** for the Dynamic Programming approach.
- **O(n^2)** for the Brute Force approach.

**Space Complexity:**
- **O(1)** for the Dynamic Programming approach (optimized with two variables).
- **O(1)** for the Brute Force approach.

**Conclusion:**
The **Dynamic Programming approach** is the most efficient and scalable solution to the problem. The greedy approach might not work for all cases, and the brute force solution is not feasible for large inputs due to its quadratic time complexity. The DP approach leverages optimal substructure and avoids unnecessary recalculations, making it both time-efficient and space-efficient for this problem.

## Problem Statement: Longest Bitonic Subsequence

Given an array of positive integers, find the maximum length of a **Bitonic subsequence**. A subsequence of an array is called **Bitonic** if it is first strictly increasing, then strictly decreasing.

**Input:**
- An array `nums[]` of length `n` representing the integers.

**Output:**
- The maximum length of the Bitonic subsequence.

**Constraints:**
- 1 ≤ n ≤ 1000
- 1 ≤ nums[i] ≤ 1000

**Example:**
**Input:**
```java
nums = [1, 11, 2, 10, 4, 5, 2, 1]
```

**Output:**
```java
6
```

**Explanation:**
- The Bitonic subsequence with maximum length is [1, 2, 4, 5, 2, 1], which has length 6.

**Approach 1: Dynamic Programming (LIS and LDS)**

To solve the problem efficiently, we can use Dynamic Programming to find two subsequences:
1. **Longest Increasing Subsequence (LIS)**: Find the longest increasing subsequence that ends at each element.
2. **Longest Decreasing Subsequence (LDS)**: Find the longest decreasing subsequence that starts at each element.

The maximum length of the bitonic subsequence will be the sum of the LIS and LDS lengths minus 1 (since the peak element is counted twice, once in LIS and once in LDS).

**Steps:**
1. **Find LIS for each element**: Iterate through the array and for each element, find the longest subsequence ending at that element where the elements are strictly increasing.
2. **Find LDS for each element**: Iterate through the array in reverse order and for each element, find the longest subsequence starting at that element where the elements are strictly decreasing.
3. **Combine LIS and LDS**: For each element, the length of the bitonic subsequence with that element as the peak will be the sum of the LIS and LDS at that index minus 1 (to avoid counting the peak twice).

**Code Implementation:**

```java
class Solution {
    public static int LongestBitonicSequence(int n, int[] nums) {
        // Initialize arrays to store the LIS and LDS lengths for each element
        int dp1[] = new int[n]; // LIS array
        int dp2[] = new int[n]; // LDS array
        Arrays.fill(dp1, 1); // Initialize LIS to 1 for all elements
        Arrays.fill(dp2, 1); // Initialize LDS to 1 for all elements

        // Compute the Longest Increasing Subsequence (LIS)
        for (int i = 0; i < n; i++) {
            for (int j = 0; j < i; j++) {
                if (nums[i] > nums[j] && 1 + dp1[j] > dp1[i]) {
                    dp1[i] = 1 + dp1[j];
                }
            }
        }

        // Compute the Longest Decreasing Subsequence (LDS) from the right
        for (int i = n - 1; i >= 0; i--) {
            for (int j = n - 1; j > i; j--) {
                if (nums[i] > nums[j] && 1 + dp2[j] > dp2[i]) {
                    dp2[i] = 1 + dp2[j];
                }
            }
        }

        // Compute the maximum length of the Bitonic subsequence
        int maxi = 1;
        for (int i = 0; i < n; i++) {
            // Ensure that LIS and LDS are greater than 1 before combining
            if (dp1[i] != 1 && dp2[i] != 1) {
                maxi = Math.max(maxi, dp1[i] + dp2[i] - 1);
            }
        }
        
        return maxi;
    }
}
```

**Explanation:**
1. **dp1 array**: `dp1[i]` holds the length of the longest increasing subsequence that ends at index `i`. We compute this by iterating over all previous elements and updating `dp1[i]` if a longer subsequence can be formed by appending `nums[i]`.
2. **dp2 array**: `dp2[i]` holds the length of the longest decreasing subsequence that starts at index `i`. We compute this by iterating over all subsequent elements and updating `dp2[i]` if a longer subsequence can be formed by appending `nums[i]`.
3. **Bitonic subsequence**: The maximum length of the bitonic subsequence that has `i` as the peak is `dp1[i] + dp2[i] - 1`. We subtract 1 because `nums[i]` is counted both in the LIS and LDS.

**Time Complexity:**
- The time complexity of this approach is **O(n^2)**, as we use two nested loops to compute both LIS and LDS.

**Space Complexity:**
- The space complexity is **O(n)**, as we use two arrays of size `n` to store LIS and LDS values.

**Approach 2: Optimized Dynamic Programming (Binary Search for LIS)**

A more optimized solution for computing the LIS can be implemented using binary search, reducing the time complexity of LIS computation from **O(n^2)** to **O(n log n)**.

**Steps:**
1. **Compute LIS using Binary Search**: Use a dynamic array `lis` to store the elements of the longest increasing subsequence as we iterate through the array. For each element in the array, we use binary search to find its position in the `lis` array and update it.
2. **Compute LDS**: Compute the LDS in a similar manner, iterating from the right and applying binary search.
3. **Combine LIS and LDS**: After calculating LIS and LDS for each index, the length of the bitonic subsequence is obtained by adding the two, subtracting 1 to avoid counting the peak twice.

**Code Implementation:**

```java
import java.util.*;

class Solution {
    public static int LongestBitonicSequence(int n, int[] nums) {
        int[] dp1 = new int[n]; // LIS
        int[] dp2 = new int[n]; // LDS
        
        // Compute LIS using Binary Search
        for (int i = 0; i < n; i++) {
            dp1[i] = 1;
            for (int j = 0; j < i; j++) {
                if (nums[i] > nums[j] && dp1[j] + 1 > dp1[i]) {
                    dp1[i] = dp1[j] + 1;
                }
            }
        }

        // Compute LDS using Binary Search
        for (int i = n - 1; i >= 0; i--) {
            dp2[i] = 1;
            for (int j = n - 1; j > i; j--) {
                if (nums[i] > nums[j] && dp2[j] + 1 > dp2[i]) {
                    dp2[i] = dp2[j] + 1;
                }
            }
        }

        int maxLen = 1;
        for (int i = 0; i < n; i++) {
            if (dp1[i] > 1 && dp2[i] > 1) {
                maxLen = Math.max(maxLen, dp1[i] + dp2[i] - 1);
            }
        }

        return maxLen;
    }
}
```

**Time Complexity:**
- **O(n^2)**: Even though the optimized approach uses binary search, it still uses dynamic programming which takes **O(n^2)** due to the two nested loops for LIS and LDS.

**Space Complexity:**
- **O(n)**: We only need two arrays for LIS and LDS computation.

**Conclusion:**
The problem of finding the longest bitonic subsequence can be efficiently solved using dynamic programming. The solution with a time complexity of **O(n^2)** is the most straightforward, while the optimized solution using binary search can reduce the time complexity for the LIS computation but still relies on dynamic programming. Both approaches are suitable depending on the input size and performance requirements.

## Problem Statement: Print All LCS Sequences

Given two strings `s1` and `s2`, your task is to print all distinct longest common subsequences (LCS) in lexicographical order.

**Input:**
- Two strings `s1` and `s2` of lengths `n` and `m` respectively.

**Output:**
- A list of all distinct longest common subsequences, sorted in lexicographical order.

**Constraints:**
- 1 ≤ n, m ≤ 1000

**Explanation:**
The problem requires finding **all distinct longest common subsequences (LCS)** between two given strings. A subsequence is derived from another string by deleting some or none of the characters without changing the order of the remaining characters. 

**Approach:**
1. **Dynamic Programming (DP) Approach**: 
   - We can solve this problem using dynamic programming by first calculating the length of the longest common subsequence (LCS) using a 2D table `dp[][]`.
   - Once the LCS length is computed, we need to **backtrack** from the last cell of the `dp` table to generate all distinct LCS sequences.
   
2. **Memoization**:
   - To avoid recomputation and redundant calculations, memoization is used to store previously computed results for substrings.
   
3. **Recursive Backtracking**:
   - We use a recursive approach to backtrack from the DP table to generate all LCS sequences. This involves considering different paths in the DP table based on character matches or moves.
   
4. **Sorting**:
   - Finally, the sequences are stored in a set and then converted to a sorted list to ensure lexicographical order.

**Steps:**
1. **Compute the Length of LCS**: Using a DP table, calculate the length of the LCS between the two strings.
2. **Backtrack to Find All LCS**: Starting from the last cell of the DP table, recursively find all LCS sequences by considering both possible moves (up or left in the DP table).
3. **Memoization**: Store the results of recursive calls in a memoization map to avoid redundant calculations and optimize performance.
4. **Sort Results**: Once all LCS sequences are found, sort them lexicographically.

**Code Implementation:**

```java
import java.util.*;

class Solution {
    public List<String> allLCS(String s, String t) {
        int n = s.length();
        int m = t.length();
        int[][] dp = new int[n+1][m+1];
        
        // Step 1: Calculate the length of the LCS using DP
        for (int i = 0; i <= n; i++) {
            for (int j = 0; j <= m; j++) {
                if (i == 0 || j == 0) {
                    dp[i][j] = 0;
                } else if (s.charAt(i-1) == t.charAt(j-1)) {
                    dp[i][j] = 1 + dp[i-1][j-1];
                } else {
                    dp[i][j] = Math.max(dp[i-1][j], dp[i][j-1]);
                }
            }
        }
        
        int lcsLength = dp[n][m];
        Map<String, Set<String>> memo = new HashMap<>();
        
        // Step 2: Recursively backtrack to find all LCS sequences
        Set<String> lcsSet = getAllLCS(s, t, n, m, dp, memo, lcsLength);
        
        // Step 3: Sort the results lexicographically
        List<String> result = new ArrayList<>(lcsSet);
        Collections.sort(result);
        
        return result;
    }
    
    // Helper function to recursively find all LCS subsequences
    private Set<String> getAllLCS(String s, String t, int i, int j, int[][] dp, 
                                  Map<String, Set<String>> memo, int lcsLength) {
        String key = i + "," + j;
        
        if (memo.containsKey(key)) {
            return memo.get(key);
        }
        
        Set<String> result = new HashSet<>();
        
        // Base Case: If we reach the top or left boundary, the LCS is an empty string
        if (i == 0 || j == 0) {
            result.add("");
            memo.put(key, result);
            return result;
        }
        
        // Case 1: Characters match, include this character in the LCS
        if (s.charAt(i-1) == t.charAt(j-1)) {
            Set<String> subLCS = getAllLCS(s, t, i-1, j-1, dp, memo, lcsLength);
            for (String subseq : subLCS) {
                result.add(subseq + s.charAt(i-1));
            }
        } 
        // Case 2: No match, move up or left in the DP table
        else {
            if (dp[i-1][j] >= dp[i][j-1]) {
                result.addAll(getAllLCS(s, t, i-1, j, dp, memo, lcsLength));
            }
            if (dp[i][j-1] >= dp[i-1][j]) {
                result.addAll(getAllLCS(s, t, i, j-1, dp, memo, lcsLength));
            }
        }
        
        // Filter out invalid subsequences that do not have enough remaining characters
        int remainingChars = lcsLength - dp[i][j];
        Set<String> filtered = new HashSet<>();
        for (String seq : result) {
            if (seq.length() + remainingChars >= lcsLength) {
                filtered.add(seq);
            }
        }
        
        memo.put(key, filtered);
        return filtered;
    }
}
```

**Explanation:**

1. **DP Table Calculation**:
   - The `dp[i][j]` stores the length of the LCS for the substrings `s[0...i-1]` and `t[0...j-1]`.
   - We fill the DP table using the following rules:
     - If `s[i-1] == t[j-1]`, then `dp[i][j] = dp[i-1][j-1] + 1` (since we include the matching character).
     - Otherwise, `dp[i][j] = max(dp[i-1][j], dp[i][j-1])` (we skip one character from either string).

2. **Recursive Backtracking**:
   - The `getAllLCS` function recursively finds all subsequences by considering two cases:
     1. **If characters match**, we include the current character and look at the previous substrings.
     2. **If characters do not match**, we look at two possible paths: one moving up in the DP table and one moving left.

3. **Memoization**:
   - To avoid redundant calculations, the results of recursive calls are stored in a memoization map `memo`.

4. **Sorting**:
   - Once all LCS sequences are found, they are stored in a `Set` to ensure uniqueness. We then sort the set to return the subsequences in lexicographical order.

**Time Complexity:**
- **O(n * m)** for filling the DP table where `n` and `m` are the lengths of the two strings.
- The recursive backtracking calls can result in **O(2^(n + m))** in the worst case. However, memoization reduces redundant calculations significantly.

**Space Complexity:**
- **O(n * m)** for storing the DP table.
- **O(n * m)** for the memoization map.


## Problem Statement: Minimum Cost to Fill Given Weight in a Bag

You are given an array `cost[]` of positive integers of size `n` and an integer `w`, where `cost[i]` represents the cost of an `(i+1)` kg packet of oranges. The goal is to find the minimum cost to buy exactly `w` kg of oranges. If buying exactly `w` kg is impossible, return `-1`.

**Constraints:**
* `cost[i] = -1` indicates that the `i+1` kg packet is unavailable.
* You can use an infinite number of available packets.

**Approach 1: Bottom-Up Dynamic Programming (2D Table)**
We use a 2D DP table where `dp[i][j]` represents the minimum cost to achieve `j` kg using the first `i` packet sizes.

**Java Code:**

```java
class Solution {
    public static int minimumCost(int n, int w, int[] cost) {
        int[][] dp = new int[n + 1][w + 1];

        for (int i = 0; i < n; i++) {
            if (cost[i] == -1) {
                cost[i] = Integer.MAX_VALUE;
            }
        }

        for (int i = 0; i <= n; i++) {
            for (int j = 0; j <= w; j++) {
                dp[i][j] = Integer.MAX_VALUE;
            }
        }

        for (int i = 0; i <= n; i++) {
            dp[i][0] = 0; // 0 cost to fill 0 kg
        }

        for (int i = 1; i <= n; i++) {
            for (int j = 1; j <= w; j++) {
                if (cost[i - 1] != Integer.MAX_VALUE && j >= i && dp[i][j - i] != Integer.MAX_VALUE) {
                    dp[i][j] = Math.min(dp[i - 1][j], cost[i - 1] + dp[i][j - i]);
                } else {
                    dp[i][j] = dp[i - 1][j];
                }
            }
        }

        return dp[n][w] == Integer.MAX_VALUE ? -1 : dp[n][w];
    }
}
```

**Time Complexity:**
* `O(n * w)`

**Space Complexity:**
* `O(n * w)`

**Approach 2: Space Optimized DP (1D Array)**
We optimize space by using a 1D array where `dp[j]` stores the minimum cost to fill `j` kg.

**Java Code:**

```java
class Solution {
    public static int minimumCost(int n, int w, int[] cost) {
        int[] dp = new int[w + 1];
        Arrays.fill(dp, Integer.MAX_VALUE);
        dp[0] = 0;

        for (int i = 0; i < n; i++) {
            if (cost[i] == -1) continue;
            for (int j = i + 1; j <= w; j++) {
                if (dp[j - (i + 1)] != Integer.MAX_VALUE) {
                    dp[j] = Math.min(dp[j], dp[j - (i + 1)] + cost[i]);
                }
            }
        }

        return dp[w] == Integer.MAX_VALUE ? -1 : dp[w];
    }
}
```

**Time Complexity:**
* `O(n * w)`

**Space Complexity:**
* `O(w)`

**Summary:**
| Approach    | Time Complexity | Space Complexity | Notes                                 |
| ----------- | --------------- | ---------------- | ------------------------------------- |
| 2D DP Table | `O(n * w)`      | `O(n * w)`       | Clear tabulation of all possibilities |
| 1D DP Array | `O(n * w)`      | `O(w)`           | Efficient space usage                 |

Both approaches provide correct answers. Use the 1D space-optimized version for better memory performance in large input cases.


## Problem Statement: Minimum Sum Partition

Given an array `arr[]` containing non-negative integers, the task is to divide it into two sets `set1` and `set2` such that the absolute difference between their sums is minimum and return this minimum difference.

**Constraints**
- `1 <= arr.length <= 100`
- `0 <= arr[i] <= 100`

**Objective**
Minimize:
\[ \left| \text{sum}(\text{set1}) - \text{sum}(\text{set2}) \right| \]

**Approach 1: Recursive + Memoization**
This is a variation of the subset sum problem. We try all subsets and compute the sum closest to totalSum/2.

**Code:**
```java
class Solution {
    public int minDifference(int[] arr) {
        int n = arr.length;
        int sum = 0;
        for (int num : arr) sum += num;
        Boolean[][] memo = new Boolean[n + 1][sum + 1];
        int closest = subsetSum(n, sum / 2, arr, memo);
        return sum - 2 * closest;
    }

    private int subsetSum(int n, int sum, int[] arr, Boolean[][] memo) {
        if (n == 0 || sum == 0) return 0;
        if (memo[n][sum] != null) return memo[n][sum] ? 1 : 0;

        int exclude = subsetSum(n - 1, sum, arr, memo);
        int include = 0;
        if (arr[n - 1] <= sum)
            include = arr[n - 1] + subsetSum(n - 1, sum - arr[n - 1], arr, memo);

        memo[n][sum] = include > exclude;
        return Math.max(include, exclude);
    }
}
```

**Approach 2: Dynamic Programming (Tabulation)**

**Idea:**

Let `dp[i][j]` be true if a subset of the first `i` elements has a sum equal to `j`.

**Code:**

```java
class Solution {
    public int minDifference(int[] arr) {
        int n = arr.length;
        int totalSum = 0;
        for (int num : arr) totalSum += num;

        boolean[][] dp = new boolean[n + 1][totalSum + 1];
        for (int i = 0; i <= n; i++) dp[i][0] = true;

        for (int i = 1; i <= n; i++) {
            for (int j = 1; j <= totalSum; j++) {
                if (arr[i - 1] <= j)
                    dp[i][j] = dp[i - 1][j] || dp[i - 1][j - arr[i - 1]];
                else
                    dp[i][j] = dp[i - 1][j];
            }
        }

        int minDiff = Integer.MAX_VALUE;
        for (int j = 0; j <= totalSum / 2; j++) {
            if (dp[n][j])
                minDiff = Math.min(minDiff, totalSum - 2 * j);
        }
        return minDiff;
    }
}
```

**Approach 3: Recursive DFS + Boolean DP (Given)**

This approach explores all combinations using DFS and marks reachable sums using a boolean matrix.

**Code:**

```java
class Solution {
    private void sumPossible(int i, int sum, int[] arr, boolean[][] dp){
        if(i >= arr.length) return;

        if(!dp[i][sum + arr[i]]) {
            sumPossible(i + 1, sum + arr[i], arr, dp);
            dp[i][sum + arr[i]] = true;
        }

        if(!dp[i][sum]) {
            sumPossible(i + 1, sum, arr, dp);
            dp[i][sum] = true;
        }
    }

    public int minDifference(int arr[]) {
        int n = arr.length;
        int totalSum = 0;
        for(int ele : arr) totalSum += ele;

        boolean[][] dp = new boolean[n][totalSum + 1];
        sumPossible(0, 0, arr, dp);

        for(int i = totalSum / 2; i >= 0; i--) {
            if(dp[n - 1][i]) {
                return totalSum - 2 * i;
            }
        }
        return 1;
    }
}
```

**Time and Space Complexity**
| Approach                | Time Complexity   | Space Complexity |
| ----------------------- | ----------------- | ---------------- |
| Recursive + Memoization | O(N\*Sum)         | O(N\*Sum)        |
| DP Tabulation           | O(N\*Sum)         | O(N\*Sum)        |
| DFS + Boolean DP        | O(2^N) worst case | O(N\*Sum)        |

**Conclusion**

This problem is a classic variation of the Subset Sum problem. The most optimal and practical approach is to use dynamic programming to avoid redundant calculations and minimize the subset sum difference efficiently.

## Problem Statement: Number of Paths in a Matrix with K Coins

Given an `n x n` matrix such that each of its cells contains some coins, count the number of ways to collect exactly `k` coins while moving from the top-left corner of the matrix to the bottom-right corner.
From a cell `(i, j)`, you can only move to `(i+1, j)` or `(i, j+1)`.

**Constraints**
* You must collect exactly `k` coins.
* Movement is only allowed to the right or downward.

**Approach 1: Recursive Backtracking (Brute Force)**
**Code**
```java
class Solution {
    long ans;
    void helper(int[][] arr, int k, int i, int j) {
        if (k < 0) return;
        if (i >= arr.length || j >= arr.length) return;
        if (i == arr.length - 1 && j == arr.length - 1 && k == arr[i][j]) ans++;
        helper(arr, k - arr[i][j], i + 1, j);
        helper(arr, k - arr[i][j], i, j + 1);
    }

    long numberOfPath(int n, int k, int [][]arr) {
        ans = 0;
        helper(arr, k, 0, 0);
        return ans;
    }
}
```

**Time Complexity**

* Exponential: O(2^(2n))

**Space Complexity**

* O(n) for recursion stack.

**Approach 2: Memoization (Top-Down DP)**

**Idea**

Store the result of subproblems using a 3D DP array: `dp[i][j][k]` - number of ways to reach bottom-right from `(i, j)` with exactly `k` coins left.

**Code**

```java
class Solution {
    Long[][][] dp;
    
    long dfs(int[][] arr, int i, int j, int k) {
        if (i >= arr.length || j >= arr.length || k < 0) return 0;
        if (i == arr.length - 1 && j == arr.length - 1)
            return k == arr[i][j] ? 1 : 0;

        if (dp[i][j][k] != null) return dp[i][j][k];

        long right = dfs(arr, i, j + 1, k - arr[i][j]);
        long down = dfs(arr, i + 1, j, k - arr[i][j]);

        return dp[i][j][k] = right + down;
    }

    long numberOfPath(int n, int k, int[][] arr) {
        dp = new Long[n][n][k + 1];
        return dfs(arr, 0, 0, k);
    }
}
```

**Time Complexity**

* O(n^2 \* k)

**Space Complexity**

* O(n^2 \* k) for memoization.

**Approach 3: Bottom-Up Dynamic Programming (Tabulation)**

**Idea**

Build a 3D DP table iteratively.

**Code**

```java
class Solution {
    long numberOfPath(int n, int k, int[][] arr) {
        long[][][] dp = new long[n][n][k + 1];

        if (arr[0][0] <= k) dp[0][0][arr[0][0]] = 1;

        for (int i = 0; i < n; i++) {
            for (int j = 0; j < n; j++) {
                for (int coins = 0; coins <= k; coins++) {
                    if (i > 0 && coins >= arr[i][j])
                        dp[i][j][coins] += dp[i - 1][j][coins - arr[i][j]];
                    if (j > 0 && coins >= arr[i][j])
                        dp[i][j][coins] += dp[i][j - 1][coins - arr[i][j]];
                }
            }
        }
        return dp[n - 1][n - 1][k];
    }
}
```

**Time Complexity**

* O(n^2 \* k)

**Space Complexity**

* O(n^2 \* k)

**Summary**
| Approach    | Time Complexity | Space Complexity | Notes                       |
| ----------- | --------------- | ---------------- | --------------------------- |
| Brute Force | Exponential     | O(n)             | Not efficient for large `n` |
| Memoization | O(n^2 \* k)     | O(n^2 \* k)      | Efficient and recommended   |
| Tabulation  | O(n^2 \* k)     | O(n^2 \* k)      | Iterative and clear         |


## Problem: Rearrange a String so that All Same Characters Become `d` Distance Away

Given a string `s` and a positive integer `d`, rearrange the characters of the string so that any two identical characters are at least `d` positions apart. If no such arrangement is possible, print:

```
Cannot be rearranged
```

**Example**

```java
Input: s = "aabbcc", d = 3  
Output: abcabc
```

**🧠 Approach**

**✅Greedy + Priority Queue (Max-Heap)**

**Key Idea:**
We place characters with the **highest frequency first** and then ensure that the same character is placed at least `d` positions apart by skipping `d` positions in each placement.

**🔧 Steps:**

1. **Frequency Count:**
   Count the frequency of each character in the input string.

2. **Max-Heap:**
   Use a max-heap (priority queue) to always get the character with the highest frequency to place next.

3. **Placement with Distance Constraint:**
   Use an array of size `n` to place characters. Start placing each character every `d` positions.
   If it's not possible to place a character (i.e., index exceeds the array or there's a conflict), return `"Cannot be rearranged"`.

4. **Waiting Queue (optional enhancement):**
   Use a wait queue to hold characters temporarily so the same character isn't reused before `d` other characters are placed (handled implicitly via index skipping in this solution).

**Java Code**

```java
import java.util.*;

public class Rearrange {
    public static void rearrange(String s, int d) {
        int n = s.length();

        // Frequency map for characters
        Map<Character, Integer> freq = new HashMap<>();
        for (char ch : s.toCharArray()) {
            freq.put(ch, freq.getOrDefault(ch, 0) + 1);
        }

        // Max heap (sorting by frequency in descending order)
        PriorityQueue<Map.Entry<Character, Integer>> pq =
            new PriorityQueue<>((a, b) -> b.getValue() - a.getValue());

        pq.addAll(freq.entrySet());

        // Result array
        char[] result = new char[n];
        Arrays.fill(result, '\u0000');

        // Queue to store delayed elements
        Queue<Map.Entry<Character, Integer>> waitQueue = new LinkedList<>();

        int index = 0;

        while (!pq.isEmpty()) {
            Map.Entry<Character, Integer> current = pq.poll();
            char ch = current.getKey();
            int count = current.getValue();

            for (int i = 0; i < count; i++) {
                // Find the next available position
                while (index < n && result[index] != '\u0000') {
                    index++;
                }

                if (index >= n) {
                    System.out.println("Cannot be rearranged");
                    return;
                }

                result[index] = ch;
                index += d;  // Place next occurrence `d` positions apart

                if (index >= n) {
                    index = 0;
                    while (index < n && result[index] != '\u0000') {
                        index++;
                    }
                }
            }

            if (count > 1) {
                waitQueue.add(new AbstractMap.SimpleEntry<>(ch, count - 1));
            }

            // After `d` iterations, reinsert from the waitQueue
            if (waitQueue.size() >= d) {
                pq.offer(waitQueue.poll());
            }
        }

        System.out.println(new String(result));
    }

    public static void main(String[] args) {
        String s = "aabbcc";
        rearrange(s, 3);
    }
}
```

**Time and Space Complexity**

* **Time Complexity:** O(n log k), where `k` is the number of unique characters.
* **Space Complexity:** O(n + k), for the result array and priority queue.

**Notes**

* Greedy approach using a max-heap and queue to maintain valid distances.
* It handles the case when a rearrangement is not possible.



## Problem: Maximum Meetings in One Room

There is one meeting room in a firm. There are `N` meetings in the form of `(S[i], F[i])` where `S[i]` is the **start time** and `F[i]` is the **finish time** of meeting `i`. The task is to find the **maximum number of meetings** that can be accommodated in the meeting room.
You can accommodate a meeting **if the start time is strictly greater than the finish time of the previous meeting**.
**Note:** If two meetings can be chosen for the same slot, choose the one that finishes earlier.
You must **print all meeting numbers** (1-based index).

**🔢 Sample Input / Output**
**Input:**

```
N = 6
S = [1, 3, 0, 5, 8, 5]
F = [2, 4, 6, 7, 9, 9]
```

**Output:**

```
1 2 4 5
```

**💡 Approach: Greedy Algorithm**
**Key Idea:** Schedule meetings in such a way that we always choose the one which finishes earliest. This leaves the room free for as many subsequent meetings as possible.

**🔹 Steps:**
1. **Store** each meeting's start time, end time, and original index.
2. **Sort** the meetings by their finish times.
3. **Iterate** through the sorted list, and select a meeting if its start time is greater than the finish time of the last selected meeting.
4. **Collect and return** the indices of selected meetings in sorted order.

**📅 Java Code**

```java
import java.util.*;

class Solution {
    public static ArrayList<Integer> maxMeetings(int N, int[] S, int[] F) {
        int[][] meet = new int[N][3];
        for (int i = 0; i < N; i++) {
            meet[i][0] = S[i];  // Start time
            meet[i][1] = F[i];  // Finish time
            meet[i][2] = i + 1; // Meeting number (1-based index)
        }

        Arrays.sort(meet, (a, b) -> a[1] - b[1]); // Sort by finish time

        ArrayList<Integer> arr = new ArrayList<>();
        int endTime = meet[0][1];
        arr.add(meet[0][2]);

        for (int i = 1; i < N; i++) {
            if (meet[i][0] > endTime) {
                arr.add(meet[i][2]);
                endTime = meet[i][1];
            }
        }

        Collections.sort(arr); // Sort meeting numbers in ascending order
        return arr;
    }
}
```

**⏱️ Time and Space Complexity**

* **Time Complexity:** `O(N log N)` — for sorting the meetings
* **Space Complexity:** `O(N)` — for storing meeting data and results


## Problem Statement: Sliding Window Median

The median is the middle value in an ordered integer list. If the size of the list is even, there is no middle value. So the median is the mean of the two middle values.
- If `arr = [2,3,4]`, the median is `3`.
- If `arr = [1,2,3,4]`, the median is `(2 + 3) / 2 = 2.5`.

You are given an integer array `nums` and an integer `k`. There is a sliding window of size `k` which is moving from the very left of the array to the very right. You can only see the `k` numbers in the window. Each time the sliding window moves right by one position.
Return the median array for each window in the original array.
> **Note:** Answers within `10^-5` of the actual value will be accepted.

**Approach**
**1. Using Two Heaps (Max-Heap + Min-Heap)**

**Idea:**
- Use two heaps:
  - A max-heap to keep the smaller half of the numbers.
  - A min-heap to keep the larger half of the numbers.
- Maintain balance so the max-heap has at most 1 more element than min-heap.
- The median is:
  - Top of max-heap if `k` is odd.
  - Average of tops of both heaps if `k` is even.

**Edge Case Handling:**
- When removing the outgoing number from the heap, instead of removing directly (which is expensive), use a map `outgoingNum` to delay removal and clean up the heap top when needed.

**Java Code**

```java
import java.util.*;

class Solution {
    public double[] medianSlidingWindow(int[] nums, int k) {
        double[] medians = new double[nums.length - k + 1];
        Map<Integer, Integer> outgoingNum = new HashMap<>();
        PriorityQueue<Integer> smallList = new PriorityQueue<>(Collections.reverseOrder());
        PriorityQueue<Integer> largeList = new PriorityQueue<>();

        for (int i = 0; i < k; i++) {
            smallList.offer(nums[i]);
        }

        for (int i = 0; i < k / 2; i++) {
            largeList.offer(smallList.poll());
        }

        int balance = 0;
        int index = 0;

        for (int i = k;; i++) {
            if (k % 2 == 1) {
                medians[index] = smallList.peek();
            } else {
                medians[index] = ((double) smallList.peek() + largeList.peek()) / 2.0;
            }

            if (i >= nums.length)
                break;

            int outNum = nums[i - k];
            int inNum = nums[i];
            index++;

            balance += (outNum <= smallList.peek()) ? -1 : 1;

            outgoingNum.put(outNum, outgoingNum.getOrDefault(outNum, 0) + 1);

            if (inNum <= smallList.peek()) {
                balance++;
                smallList.offer(inNum);
            } else {
                balance--;
                largeList.offer(inNum);
            }

            if (balance < 0) {
                smallList.offer(largeList.poll());
            } else if (balance > 0) {
                largeList.offer(smallList.poll());
            }

            balance = 0;

            while (outgoingNum.getOrDefault(smallList.peek(), 0) > 0) {
                outgoingNum.put(smallList.peek(), outgoingNum.get(smallList.peek()) - 1);
                smallList.poll();
            }

            while (!largeList.isEmpty() && outgoingNum.getOrDefault(largeList.peek(), 0) > 0) {
                outgoingNum.put(largeList.peek(), outgoingNum.get(largeList.peek()) - 1);
                largeList.poll();
            }
        }

        return medians;
    }
}
```

* **Time Complexity:** `O(n log k)` where `n` is the number of elements and `k` is the window size.
* **Space Complexity:** `O(k)` for heaps and hashmap.

## Problem Statement: Minimum Number of Refueling Stops

A car travels from a starting position to a destination which is `target` miles east of the starting position.
There are gas stations along the way. The gas stations are represented as an array `stations` where `stations[i] = [positioni, fueli]` indicates that the ith gas station is `positioni` miles east of the starting position and has `fueli` liters of gas.
The car starts with an infinite tank of gas, which initially has `startFuel` liters of fuel in it. It uses one liter of gas per one mile that it drives. When the car reaches a gas station, it may stop and refuel, transferring all the gas from the station into the car.

Return the **minimum number of refueling stops** the car must make in order to reach its destination. If it **cannot** reach the destination, return **-1**.

> Note: If the car reaches a gas station or the destination with 0 fuel left, it is still considered to have arrived.

**Approach**
**Greedy + Max Heap**

We use a **greedy approach** with a **max heap** (priority queue) to keep track of the best fuel options passed so far.

**Steps:**
1. Traverse the list of stations and keep checking whether the car can reach the next station or the target with the current fuel.
2. If it **cannot**, pick the gas station with the most fuel that has been passed (stored in max heap) and refuel from there.
3. Every time fuel is added from the heap, increase the number of refueling stops.
4. If no station is available to refuel from when needed, return -1.

**Java Code**

```java
import java.util.*;

class Solution {
    public int minRefuelStops(int target, int startFuel, int[][] stations) {
        int n = stations.length, refuels = 0;
        PriorityQueue<Integer> pq = new PriorityQueue<>((a, b) -> b - a); // max heap

        for (int i = 0; i <= n; i++) {
            int dist = i == n ? target : stations[i][0];

            while (startFuel < dist) {
                if (pq.isEmpty()) return -1;
                startFuel += pq.poll();
                refuels++;
            }

            if (i < n) {
                pq.add(stations[i][1]); // add current station's fuel
            }
        }

        return refuels;
    }
}
```

* **Time Complexity:** O(n log n)
  * Each fuel addition/removal to/from heap takes log n.

* **Space Complexity:** O(n)
  * Max heap can store up to `n` fuel values.


## Problem Statement: Convert Min Heap to Max Heap

You are given an array `arr` of `N` integers representing a **Min Heap**. The task is to convert it into a **Max Heap**.

A **Max-Heap** is a complete binary tree in which the value at each internal node is **greater than or equal** to the values in its children.


**Example**
**Input:**

```text
arr = [1, 3, 5, 7, 9, 2]
```
**Output (after conversion):**

```text
arr = [9, 7, 5, 1, 3, 2]
```

**Approach**
**Idea**

To convert a **Min Heap** to a **Max Heap**, we need to rearrange the elements such that they follow the **Max Heap property**. This can be efficiently done using a bottom-up **heapify** process starting from the last internal node all the way up to the root.

**Steps**
1. The last non-leaf node is at index `(N - 2) / 2`.
2. Traverse all internal nodes in reverse level order.
3. For each node, apply `MaxHeapify()` to ensure the Max Heap property is maintained.

**Java Code**

```java
class Solution {

    static void MaxHeapify(int arr[], int i, int N) {
        int l = 2 * i + 1;   // left child
        int r = 2 * i + 2;   // right child
        int largest = i;

        if (l < N && arr[l] > arr[largest])
            largest = l;
        if (r < N && arr[r] > arr[largest])
            largest = r;

        if (largest != i) {
            int temp = arr[i];
            arr[i] = arr[largest];
            arr[largest] = temp;

            // Recursively heapify the affected subtree
            MaxHeapify(arr, largest, N);
        }
    }

    static void convertMinToMaxHeap(int N, int arr[]) {
        for (int i = (N - 2) / 2; i >= 0; --i)
            MaxHeapify(arr, i, N);
    }
}
```

**Time & Space Complexity**

* **Time Complexity:** O(N)
  Heap construction via bottom-up heapify takes linear time.

* **Space Complexity:** O(log N)
  Due to recursive stack space in `MaxHeapify`.


## Problem Statement: k Largest Elements

Given an array `arr[]` of positive integers and an integer `k`, your task is to return the **k largest elements** in **decreasing order**.

**Constraints**
- 1 ≤ K ≤ N ≤ 10⁵  
- 1 ≤ Arr[i] ≤ 10⁶


**Example**
**Input:**

```text
arr = [1, 23, 12, 9, 30, 2, 50]
k = 3
```

**Output:

```text
[50, 30, 23]
```


**Approach**
**Idea**
To find the `k` largest elements in an array efficiently:

1. Use a **Min Heap** (PriorityQueue in Java) of size `k`.
2. Insert the first `k` elements into the heap.
3. Traverse the remaining elements:
   - If the current element is greater than the root of the heap (smallest among the largest seen so far), replace it.
4. Finally, extract all elements from the heap into a list and sort it in **descending order**.

**Why Min Heap?**
- A **Min Heap** helps maintain the smallest of the `k` largest elements at the top.
- This ensures we can discard smaller elements efficiently as we encounter bigger ones.

**Java Code**
```java
import java.util.*;

class Solution {
    public ArrayList<Integer> kLargest(int[] arr, int k) {
        // Min Heap to store the top k elements
        PriorityQueue<Integer> minHeap = new PriorityQueue<>(k);

        // Build the initial heap with first k elements
        for (int i = 0; i < k; i++) {
            minHeap.add(arr[i]);
        }

        // Process remaining elements
        for (int i = k; i < arr.length; i++) {
            if (arr[i] > minHeap.peek()) {
                minHeap.poll();
                minHeap.add(arr[i]);
            }
        }

        // Extract and reverse the result for descending order
        ArrayList<Integer> res = new ArrayList<>();
        while (!minHeap.isEmpty()) {
            res.add(minHeap.poll());
        }
        Collections.reverse(res);

        return res;
    }
}
```

**Time and Space Complexity**

* **Time Complexity:** O(N log K)
  * Each insertion/removal in the heap takes log K time.

* **Space Complexity:** O(K)
  * To store the top `k` elements.


## Problem Statement: Top K Frequent Elements

Given an integer array `nums` and an integer `k`, return the **k most frequent elements**.  
You may return the answer in **any order**.

**Example**

**Input:**
```text
nums = [1,1,1,2,2,3], k = 2
````

**Output:**

```text
[1,2]
```

**Constraints**
* 1 ≤ nums.length ≤ 10⁵
* -10⁴ ≤ nums\[i] ≤ 10⁴
* k is in the range \[1, the number of unique elements in the array]
* It is guaranteed that the answer is unique

**Approach**
**Idea**

Use a **HashMap** to count the frequency of each number and then use a **Min Heap** to track the top `k` most frequent elements.

**Steps:**

1. **Count Frequencies:**
   Use a HashMap to store frequency of each element.

2. **Build a Min Heap of Size k:**

   * Insert elements with their frequency into the heap.
   * If heap size exceeds `k`, remove the least frequent element.

3. **Extract Elements from the Heap:**
   The heap will contain the top `k` frequent elements.

**Java Code**

```java
import java.util.*;

class Solution {
    public int[] topKFrequent(int[] nums, int k) {
        // Count frequency of each element
        Map<Integer, Integer> freqMap = new HashMap<>();
        for (int num : nums)
            freqMap.put(num, freqMap.getOrDefault(num, 0) + 1);

        // Min Heap to keep top k frequent elements
        PriorityQueue<Map.Entry<Integer, Integer>> minHeap =
            new PriorityQueue<>((a, b) -> a.getValue() - b.getValue());

        // Maintain the heap of size k
        for (Map.Entry<Integer, Integer> entry : freqMap.entrySet()) {
            minHeap.offer(entry);
            if (minHeap.size() > k)
                minHeap.poll();
        }

        // Extract the elements from the heap
        int[] result = new int[k];
        int i = 0;
        while (!minHeap.isEmpty())
            result[i++] = minHeap.poll().getKey();

        return result;
    }
}
```

**Time and Space Complexity**

* **Time Complexity:** O(N log K)
  * N for frequency map, log K for heap operations.

* **Space Complexity:** O(N)
  * For storing frequencies and the heap.

**Alternate Approaches**
**1. Bucket Sort**

* Group elements by frequency.
* Start from highest frequency and collect top `k` elements.
* Time: O(N), Space: O(N)


## Problem Statement: Top K Frequent Words

Given an array of strings `words` and an integer `k`, return the `k` most frequent strings.
The result should be:
- Sorted by **frequency** (from highest to lowest).
- For words with the **same frequency**, sort them **lexicographically** (i.e., alphabetically).

**Example**
**Input:**
```text
words = ["i","love","leetcode","i","love","coding"], k = 2
````

**Output:**
```text
["i","love"]
```

**Explanation:**
* "i" appears 2 times.
* "love" appears 2 times.
* "leetcode" and "coding" appear 1 time each.
* "i" and "love" are returned because of highest frequency.
* Lexicographically, "i" < "love".

**Approach**

**1. Count Frequencies**

* Use a HashMap to store the frequency of each word.

**2. Use a Min-Heap (PriorityQueue)**

* Custom comparator:

  * Sort by frequency ascending.
  * If frequencies are equal, sort lexicographically **descending** to maintain correct pop order.
* Keep only the top `k` elements in the heap.

**3. Collect Results**

* Pop all elements from the heap and reverse them to get the correct order.

**Java Code**

```java
import java.util.*;

class Solution {
    public List<String> topKFrequent(String[] words, int k) {
        // Step 1: Count frequencies
        Map<String, Integer> freqMap = new HashMap<>();
        for (String word : words)
            freqMap.put(word, freqMap.getOrDefault(word, 0) + 1);

        // Step 2: Create min heap with custom comparator
        PriorityQueue<String> heap = new PriorityQueue<>(
            (a, b) -> {
                int freqCompare = freqMap.get(a) - freqMap.get(b);
                if (freqCompare == 0) {
                    return b.compareTo(a); // reverse lexicographical
                }
                return freqCompare;
            }
        );

        // Step 3: Maintain heap of size k
        for (String word : freqMap.keySet()) {
            heap.offer(word);
            if (heap.size() > k) {
                heap.poll();
            }
        }

        // Step 4: Extract from heap to list
        List<String> result = new ArrayList<>();
        while (!heap.isEmpty()) {
            result.add(heap.poll());
        }

        Collections.reverse(result); // reverse to get correct order

        return result;
    }
}
```

**Complexity**

* **Time Complexity:** O(N log K), where N is the number of unique words.
* **Space Complexity:** O(N) for the map and heap.
