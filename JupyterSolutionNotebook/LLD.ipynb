
## 📞🚕 Call Taxi Booking Application - Problem Statement

This problem is designed to test your skills in object-oriented programming, resource allocation, and algorithmic reasoning. It was asked in Zoho interviews and involves simulating a real-time call taxi booking system with well-defined constraints.

**📌 Problem Statement**

You are required to build a **Call Taxi Booking Application** that simulates the booking process for taxis operating between a set of linear points. Your solution should allocate taxis optimally and maintain detailed records of all bookings.

**🔧 Constraints:**

- There are **6 locations**: `A, B, C, D, E, F`, placed linearly.
- The distance between two consecutive locations is **15 km**.
- Travel time between two consecutive points is **1 hour**.
- All taxis are initially stationed at **point A**.
- The fare rules are:
  - **Minimum Fare**: Rs. 100 for the first 5 km.
  - **Rs. 10** is charged for every additional kilometer.
- A **free taxi closest** to the customer’s pickup point is allocated.
  - If two taxis are equidistant, choose the one with **less total earnings**.
- If no taxi is available at the requested pickup time, **reject the booking**.
- After each booking, **display taxi earnings and all booking details**.

**🧠 Solution Design**

The solution follows an object-oriented approach with three main components:

**1. `Taxi` Class**

Encapsulates the properties and behavior of a taxi.

```java
class Taxi {
    int id;
    char currentPoint = 'A';
    int totalEarnings = 0;
    List<Booking> bookings = new ArrayList<>();

    public Taxi(int id) {
        this.id = id;
    }

    public boolean isAvailable(int requestTime) {
        if (bookings.isEmpty()) return true;
        Booking lastBooking = bookings.get(bookings.size() - 1);
        return lastBooking.dropTime <= requestTime;
    }

    public int calculateEarnings(char from, char to) {
        int distance = Math.abs(to - from) * 15;
        return 100 + Math.max(0, (distance - 5) * 10);
    }

    public void addBooking(Booking booking) {
        bookings.add(booking);
        totalEarnings += booking.amount;
        currentPoint = booking.to;
    }
}
```

Stores details about a customer booking.

```java
public class TaxiBookingSystem {
    static List<Taxi> taxis = new ArrayList<>();
    static Scanner sc = new Scanner(System.in);
    static int customerCounter = 1;

    public static void main(String[] args) {
        System.out.print("Enter number of taxis: ");
        int numTaxis = sc.nextInt();

        while (true) {
            System.out.println("\n1. Book Taxi\n2. Display Taxi Details\n3. Exit");
            System.out.print("Enter your choice: ");
            int choice = sc.nextInt();

            switch (choice) {
                case 1:
                    bookTaxi();
                    break;
                case 2:
                    displayTaxiDetails();
                    break;
                case 3:
                    System.out.println("Exiting...");
                    return;
                default:
                    System.out.println("Invalid choice. Try again.");
            }
        }
    }

    public static void initializeTaxis(int n) {
        for (int i = 1; i <= n; i++) {
            taxis.add(new Taxi(i));
        }
    }

    public static void bookTaxi() {
        int customerId = customerCounter++;
        System.out.print("Enter Pickup Point (A-F): ");
        char pickup = sc.next().toUpperCase().charAt(0);
        System.out.print("Enter Drop Point (A-F): ");
        char drop = sc.next().toUpperCase().charAt(0);
        System.out.print("Enter Pickup Time (in hours): ");
        int pickupTime = sc.nextInt();

        Taxi selectedTaxi = null;
        int minDistance = Integer.MAX_VALUE;

        for (Taxi taxi : taxis) {
            if (taxi.isAvailable(pickupTime)) {
                int distance = Math.abs(taxi.currentPoint - pickup);
                if (distance < minDistance ||
                    (distance == minDistance && taxi.totalEarnings < selectedTaxi.totalEarnings)) {
                    selectedTaxi = taxi;
                    minDistance = distance;
                }
            }
        }

        if (selectedTaxi == null) {
            System.out.println("Booking rejected. No taxis available.");
            return;
        }

        int dropTime = pickupTime + Math.abs(drop - pickup);
        int amount = selectedTaxi.calculateEarnings(pickup, drop);
        int bookingId = selectedTaxi.bookings.size() + 1;

        Booking booking = new Booking(bookingId, customerId, pickup, drop, pickupTime, dropTime, amount);
        selectedTaxi.addBooking(booking);

        System.out.println("Taxi-" + selectedTaxi.id + " is allocated.");
    }

    public static void displayTaxiDetails() {
        for (Taxi taxi : taxis) {
            System.out.println("Taxi-" + taxi.id + " Total Earnings: Rs." + taxi.totalEarnings);
            System.out.printf("%-10s %-12s %-5s %-5s %-12s %-9s %-6s%n",
                    "BookingID", "CustomerID", "From", "To", "PickupTime", "DropTime", "Amount");

            for (Booking booking : taxi.bookings) {
                System.out.printf("%-10d %-12d %-5c %-5c %-12d %-9d %-6d%n",
                        booking.bookingId, booking.customerId, booking.from, booking.to,
                        booking.pickupTime, booking.dropTime, booking.amount);
            }
        }
    }
}
```

## ✈️ Automated Air Traffic Control System (ATCS)

Design a program to automate the **Air Traffic Control** system that manages flight operations such as **landing**, **takeoff**, and **emergency landing** based on flight type, weight, and runway availability.

**📌 Problem Statement**

The system must simulate the behavior of an **Air Traffic Controller (ATC)**. Given flight requests with their details, the system must:

- Accept flight operation requests:
  - `Landing`
  - `Takeoff`
  - `Emergency Landing`
- Compute required time to halt or take off based on **flight type** and **weight**
- Allocate the most suitable **runway**
- Handle **multiple requests** and queue management
- Free the runway after the flight completes **taxiing**

**✈️ Terminologies**

| Term             | Meaning                                                                 |
|------------------|-------------------------------------------------------------------------|
| **Touchdown**     | Flight touches the ground                                               |
| **Time to halt**  | Time from touchdown until the flight stops completely                   |
| **Takeoff**       | Flight leaves the ground                                                |
| **Taxiing**       | Movement of the plane to/from the runway to its stop after landing/takeoff |

**🛩️ Flight Details**

All planes move at **200 km/h**.

**Flight Types**

| Type    | Max Weight (tons) | Full Time to Halt / Takeoff (seconds) |
|---------|-------------------|----------------------------------------|
| ATR     | 12                | 30 sec                                 |
| AirBus  | 20                | 40 sec                                 |
| Boeing  | 40                | 50 sec                                 |
| Cargo   | 100               | 60 sec                                 |

**⚙️ Time Calculation Logic**

Time to halt or takeoff is directly proportional to weight.

**Weight Impact:**

- **>75% of max weight**: Full time
- **50%–75% of max weight**: `Full time - 10%`
- **<50% of max weight**: `Full time - 20%`

**Example (ATR):**
- Max Weight: 12 tons, Full Time: 30 sec
- If Weight = 9+ tons (≥75%) → 30 sec
- If Weight = 6–9 tons (50%–75%) → 27 sec
- If Weight < 6 tons (<50%) → 24 sec

**🛬 Runway Details**

Each runway has a maximum operation time (both landing & takeoff):

| Runway | Max Duration | Notes                             |
|--------|--------------|-----------------------------------|
| R1     | 40 sec       | Fastest, preferred for small jets |
| R2     | 60 sec       |                                   |
| R3     | 80 sec       |                                   |
| R4     | 90 sec       | Longest, reserved for emergencies |

**🧠 Runway Assignment Rule:**
> Always assign a runway that offers **at least +10 sec more** than required time.

**🧠 Controller Logic**

**🗂️ Task Types:**
- **Landing**: Highest priority
- **Emergency Landing**: Overrides everything, uses **longest runway**
- **Takeoff**: Normal priority

**🎯 Task Handling:**
- Input: Flight type & weight
- Determine weight percentage → calculate halt/takeoff time
- Assign the most optimal runway
- Mark the runway as engaged
- Free the runway after completion + taxiing

**🧵 Request Management**

- Handle **multiple requests** concurrently
- Use **queues** when more than 4 requests are pending
- A request can **wait up to 3 times** (FIFO-based)
- Each operation has a **15-second pre-processing** time

- Flights: 2 Cargo, 2 ATR
- Vary ATR weights
- Emergency added mid-way → gets R4

**✅ Scenario 2:**
- 5 requests → 4 handled, 5th added to queue
- Queue respects FIFO and wait count

**1. `Flight.java`**

```java
public class Flight {
    String type;
    double weight;
    double maxWeight;
    int fullTime;

    public Flight(String type, double weight) {
        this.type = type.toLowerCase();
        this.weight = weight;
        switch (this.type) {
            case "atr":
                maxWeight = 12;
                fullTime = 30;
                break;
            case "airbus":
                maxWeight = 20;
                fullTime = 40;
                break;
            case "boeing":
                maxWeight = 40;
                fullTime = 50;
                break;
            case "cargo":
                maxWeight = 100;
                fullTime = 60;
                break;
            default:
                throw new IllegalArgumentException("Invalid flight type");
        }
    }

    public int getAdjustedTime() {
        double percent = (weight / maxWeight) * 100;
        if (percent > 75)
            return fullTime;
        else if (percent > 50)
            return fullTime - (int)(fullTime * 0.10);
        else
            return fullTime - (int)(fullTime * 0.20);
    }
}
```

**2. `Runway.java`**

```java
public class Runway {
    String id;
    int maxTime;
    boolean engaged;
    Flight currentFlight;

    public Runway(String id, int maxTime) {
        this.id = id;
        this.maxTime = maxTime;
        this.engaged = false;
    }

    public boolean canAssign(Flight flight) {
        return !engaged && (maxTime >= flight.getAdjustedTime() + 10);
    }

    public void assignFlight(Flight flight) {
        this.currentFlight = flight;
        this.engaged = true;
    }

    public void freeRunway() {
        this.engaged = false;
        this.currentFlight = null;
    }

    public String status() {
        return "Runway " + id + " : " + (engaged ? "Engaged (" + currentFlight.type + ")" : "Free");
    }
}
```

```java
import java.util.*;

public class ATCController {
    static List<Runway> runways = new ArrayList<>();
    static Queue<Request> requestQueue = new LinkedList<>();

    public static void main(String[] args) {
        initRunways();
        Scanner sc = new Scanner(System.in);

        while (true) {
            System.out.println("\nRequest Type:\n1. Takeoff\n2. Landing\n3. Emergency Landing\n4. Exit");
            int choice = sc.nextInt();
            if (choice == 4) break;

            sc.nextLine(); // flush
            System.out.println("Enter Plane Type (ATR, Airbus, Boeing, Cargo): ");
            String type = sc.nextLine();

            System.out.println("Enter Plane Weight (in tons): ");
            double weight = sc.nextDouble();

            Flight flight = new Flight(type, weight);
            Request req = new Request(flight, choice);
            handleRequest(req);
        }

        sc.close();
        System.out.println("Final Runway Status:");
        runways.forEach(r -> System.out.println(r.status()));
    }

    static void initRunways() {
        runways.add(new Runway("R1", 40));
        runways.add(new Runway("R2", 60));
        runways.add(new Runway("R3", 80));
        runways.add(new Runway("R4", 90));
    }

    static void handleRequest(Request req) {
        System.out.println("Processing request in 15 sec...");
        try { Thread.sleep(15000); } catch (InterruptedException e) {}

        if (req.type == 3) { // Emergency → R4
            Runway r4 = runways.get(3);
            if (!r4.engaged) {
                r4.assignFlight(req.flight);
                System.out.println("Emergency landing on R4");
            } else {
                System.out.println("R4 busy! Emergency request queued");
                requestQueue.add(req);
            }
            return;
        }

        // Landing has higher priority
        runways.sort(Comparator.comparingInt(r -> r.maxTime));
        for (Runway r : runways) {
            if (r.canAssign(req.flight)) {
                r.assignFlight(req.flight);
                printAssignment(req, r);
                return;
            }
        }

        if (requestQueue.size() < 4 && req.waitCount < 3) {
            System.out.println("All runways busy. Request queued.");
            req.waitCount++;
            requestQueue.add(req);
        } else {
            System.out.println("Request failed. Waited too long.");
        }
    }

    static void printAssignment(Request req, Runway r) {
        int t = req.flight.getAdjustedTime();
        System.out.println((req.type == 1 ? "Takeoff" : "Landing") + " approved for " + req.flight.type.toUpperCase() +
            " (" + req.flight.weight + " tons) on Runway " + r.id);
        System.out.println("Touchdown/Takeoff in 15 sec");
        System.out.println("Complete stop in " + t + " sec");
        System.out.println("Runway " + r.id + " will be free in ~1 minute");
    }
}

class Request {
    Flight flight;
    int type; // 1 = Takeoff, 2 = Landing, 3 = Emergency
    int waitCount = 0;

    Request(Flight flight, int type) {
        this.flight = flight;
        this.type = type;
    }
}
```

```java
import java.util.*;

class Highway {
    List<Toll> tollPoints;
    Map<String, Vehicle> vehicleRecords;

    public Highway(List<Toll> tollPoints) {
        this.tollPoints = tollPoints;
        this.vehicleRecords = new HashMap<>();
    }

    public void processJourney(String vehicleNumber, String vehicleType, boolean isVIP, String start, String end, List<Integer> tollRoute) {
        Vehicle vehicle = vehicleRecords.computeIfAbsent(vehicleNumber, vn -> new Vehicle(vn, vehicleType, isVIP));
        int totalAmount = 0;
        for (int tollId : tollRoute) {
            Toll toll = tollPoints.get(tollId);
            int charge = toll.calculateToll(vehicleType, isVIP);
            toll.recordVehicle(vehicle, charge);
            totalAmount += charge;
        }
        Journey journey = new Journey(start, end, tollRoute, totalAmount);
        vehicle.addJourney(journey);
        System.out.println("Journey completed! Total Toll Paid: ₹" + totalAmount);
    }

    public void displayTollDetails() {
        for (Toll toll : tollPoints) {
            toll.displayDetails();
        }
    }

    public void displayVehicleDetails() {
        for (Vehicle vehicle : vehicleRecords.values()) {
            vehicle.displayDetails();
        }
    }

    public List<Integer> findCircularRoute(int start, int end) {
        List<Integer> forward = new ArrayList<>();
        List<Integer> backward = new ArrayList<>();
        int total = tollPoints.size();

        for (int i = start; i != end; i = (i + 1) % total) forward.add(i);
        forward.add(end);

        for (int i = start; i != end; i = (i - 1 + total) % total) backward.add(i);
        backward.add(end);

        return forward.size() <= backward.size() ? forward : backward;
    }

    public int calculateRegularTollForRoute(int start, int end, String vehicleType, boolean isVIP) {
        int totalCost = 0;
        if (start <= end) {
            for (int i = start; i <= end; i++)
                totalCost += tollPoints.get(i).calculateToll(vehicleType, isVIP);
        } else {
            for (int i = start; i < tollPoints.size(); i++)
                totalCost += tollPoints.get(i).calculateToll(vehicleType, isVIP);
            for (int i = 0; i <= end; i++)
                totalCost += tollPoints.get(i).calculateToll(vehicleType, isVIP);
        }
        return totalCost;
    }

    public int calculateTollForRoute(List<Integer> tollRoute, String vehicleType, boolean isVIP) {
        return tollRoute.stream().mapToInt(id -> tollPoints.get(id).calculateToll(vehicleType, isVIP)).sum();
    }
}
```


**2. Flight Class**

This class handles operations related to a specific flight, such as booking, cancellation, and managing passenger data.

```java
import java.util.*;

class Flight {
    private String flightName;
    private int availableSeats = 50;
    private int ticketPrice = 5000;
    private Map<String, Passenger> bookings = new HashMap<>();
    private int bookingCounter = 0;

    public Flight(String flightName) {
        this.flightName = flightName;
    }

    public String bookTickets(String passengerName, int age, int seats) {
        if (seats <= availableSeats) {
            bookingCounter++;
            String bookingId = "T" + bookingCounter;
            Passenger passenger = new Passenger(bookingId, passengerName, age, seats);
            bookings.put(bookingId, passenger);
            availableSeats -= seats;
            ticketPrice += 200 * seats;
            return bookingId;
        } else {
            System.out.println("Booking failed: Not enough seats available.");
            return null;
        }
    }

    public boolean cancelBooking(String bookingId) {
        Passenger passenger = bookings.get(bookingId);
        if (passenger != null) {
            int seats = passenger.getSeatsBooked();
            availableSeats += seats;
            ticketPrice -= 200 * seats;
            bookings.remove(bookingId);
            System.out.println("Booking canceled successfully. Refund issued for " + seats + " seats.");
            return true;
        } else {
            System.out.println("Cancellation failed: Booking ID not found.");
            return false;
        }
    }

    public void displayDetails() {
        System.out.println("Flight: " + flightName);
        System.out.println("Available Seats: " + availableSeats);
        System.out.println("Current Ticket Price: ₹" + ticketPrice);
    }

    public void printDetails() {
        System.out.println("Flight: " + flightName);
        System.out.println("Available Seats: " + availableSeats);
        System.out.println("Current Ticket Price: ₹" + ticketPrice);
        System.out.println("Passengers:");
        for (Passenger passenger : bookings.values()) {
            System.out.println(passenger);
        }
    }
}
```

**3. FlightReservationSystem Class**

This class manages multiple flights and delegates operations to the relevant `Flight` objects.

```java
import java.util.*;

class FlightReservationSystem {
    Map<String, Flight> flights = new HashMap<>();

    public FlightReservationSystem() {
        flights.put("Indigo", new Flight("Indigo"));
        flights.put("SpiceJet", new Flight("SpiceJet"));
    }

    public String bookTicket(String flightName, String passengerName, int age, int seats) {
        Flight flight = flights.get(flightName);
        if (flight != null) {
            return flight.bookTickets(passengerName, age, seats);
        } else {
            System.out.println("Booking failed: Flight not found.");
            return null;
        }
    }

    public boolean cancelTicket(String flightName, String bookingId) {
        Flight flight = flights.get(flightName);
        if (flight != null) {
            return flight.cancelBooking(bookingId);
        } else {
            System.out.println("Cancellation failed: Flight not found.");
            return false;
        }
    }

    public void displayFlightDetails(String flightName) {
        Flight flight = flights.get(flightName);
        if (flight != null) {
            flight.displayDetails();
        } else {
            System.out.println("Flight not found.");
        }
    }

    public void printFlightDetails(String flightName) {
        Flight flight = flights.get(flightName);
        if (flight != null) {
            flight.printDetails();
        } else {
            System.out.println("Flight not found.");
        }
    }
}
```

**4. Main Class**

This is the entry point for the application, where users interact with the system.

```java
import java.util.*;

public class Main {
    public static void main(String[] args) {
        FlightReservationSystem system = new FlightReservationSystem();
        Scanner scanner = new Scanner(System.in);

        while (true) {
            System.out.println("\n--- Flight Reservation System ---");
            System.out.println("1. Book Ticket");
            System.out.println("2. Cancel Ticket");
            System.out.println("3. Print Flight Details");
            System.out.println("4. Exit");
            System.out.print("Enter your choice: ");
            int choice = scanner.nextInt();
            scanner.nextLine();

            switch (choice) {
                case 1:
                    System.out.print("Enter flight name (Indigo/SpiceJet): ");
                    String flightName = scanner.nextLine();
                    System.out.print("Enter passenger name: ");
                    String passengerName = scanner.nextLine();
                    System.out.print("Enter passenger age: ");
                    int age = scanner.nextInt();
                    System.out.print("Enter number of seats to book: ");
                    int seats = scanner.nextInt();
                    String bookingId = system.bookTicket(flightName, passengerName, age, seats);
                    if (bookingId != null) {
                        System.out.println("Booking successful! Your booking ID is: " + bookingId);
                    }
                    break;

                case 2:
                    System.out.print("Enter flight name (Indigo/SpiceJet): ");
                    flightName = scanner.nextLine();
                    System.out.print("Enter booking ID: ");
                    bookingId = scanner.nextLine();
                    system.cancelTicket(flightName, bookingId);
                    break;

                case 3:
                    System.out.print("Enter flight name (Indigo/SpiceJet): ");
                    flightName = scanner.nextLine();
                    system.printFlightDetails(flightName);
                    break;

                case 4:
                    System.out.println("Exiting system. Thank you!");
                    scanner.close();
                    return;

                default:
                    System.out.println("Invalid choice. Please try again.");
            }
        }
    }
}
```

**Conclusion**

This solution captures the essence of a real-world flight booking system with essential functionalities. By modularizing the application into distinct classes, we ensure a clean, maintainable, and extensible design.

## Railway Ticket Booking Application – LLD

**Understanding the Problem**

Design a Railway Ticket Booking Application with four main functionalities:

1. **Book Ticket**
2. **Cancel Ticket**
3. **Print Booked Tickets** (with details and summary)
4. **Print Available Tickets** (with details and summary)

**Ticket Categories:**
- **Confirmed Tickets**: 63 berths (Lower, Upper, Middle)
- **RAC Tickets**: 9 berths (18 passengers)
- **Waiting List Tickets**: Max 10

**Booking Conditions:**
- Senior citizens (> 60) or females should get a **lower berth** if available.
- If all confirmed berths are full → assign **RAC**.
- If RAC is full → assign **Waiting List**.
- If all are full → display **“No tickets available”**.

**Cancellation Conditions:**
- On cancellation:
  - One RAC ticket should move to Confirmed.
  - One Waiting List ticket should move to RAC.

**Solution Structure**

We break the system into three core classes:

- **Passenger**: Represents the passenger's ticket data.
- **TicketSystem**: Handles business logic (book, cancel, print).
- **TicketBooking**: Main menu interface for user interaction.

**Code Walkthrough**

**Passenger Class**

```java
public class Passenger {
    String name;
    int age;
    String gender;
    String berthPreference;
    String allottedBerth;
    String ticketId;

    public Passenger(String name, int age, String gender, String berthPreference, String allottedBerth, String ticketId) {
        this.name = name;
        this.age = age;
        this.gender = gender;
        this.berthPreference = berthPreference;
        this.allottedBerth = allottedBerth;
        this.ticketId = ticketId;
    }

    @Override
    public String toString() {
        return "Ticket ID: " + ticketId + ", Name: " + name + ", Age: " + age +
                ", Gender: " + gender + ", Berth: " + allottedBerth;
    }
}
````

```java
import java.util.*;

public class TicketSystem {
    private final List<String> availableBerths = new ArrayList<>(Arrays.asList("L", "U", "M"));
    private final Queue<Passenger> racQueue = new LinkedList<>();
    private final Queue<Passenger> waitingListQueue = new LinkedList<>();
    private final List<Passenger> confirmedPassengers = new ArrayList<>();
    private int ticketCounter = 1;

    public void bookTicket(String name, int age, String gender, String berthPreference) {
        String ticketId = "T" + ticketCounter++;
        Passenger passenger;

        if (!availableBerths.isEmpty()) {
            String allocatedBerth = allocateBerth(age, gender, berthPreference);
            passenger = new Passenger(name, age, gender, berthPreference, allocatedBerth, ticketId);
            confirmedPassengers.add(passenger);
            availableBerths.remove(allocatedBerth);
            System.out.println("Ticket confirmed: " + passenger);
        } else if (racQueue.size() < 1) {
            passenger = new Passenger(name, age, gender, berthPreference, "RAC", ticketId);
            racQueue.offer(passenger);
            System.out.println("Ticket in RAC: " + passenger);
        } else if (waitingListQueue.size() < 1) {
            passenger = new Passenger(name, age, gender, berthPreference, "Waiting List", ticketId);
            waitingListQueue.offer(passenger);
            System.out.println("Ticket in Waiting List: " + passenger);
        } else {
            System.out.println("No tickets available");
        }
    }

    private String allocateBerth(int age, String gender, String preference) {
        if (age > 60 || gender.equalsIgnoreCase("female")) {
            if (availableBerths.contains("L")) return "L";
        }
        if (availableBerths.contains(preference)) {
            return preference;
        }
        return availableBerths.get(0);
    }

    public void cancelTicket(String ticketId) {
        Optional<Passenger> passengerOpt = confirmedPassengers.stream()
                .filter(p -> p.ticketId.equals(ticketId))
                .findFirst();

        if (passengerOpt.isPresent()) {
            Passenger passenger = passengerOpt.get();
            confirmedPassengers.remove(passenger);
            availableBerths.add(passenger.allottedBerth);

            if (!racQueue.isEmpty()) {
                Passenger racPassenger = racQueue.poll();
                String allocatedBerth = allocateBerth(racPassenger.age, racPassenger.gender, racPassenger.berthPreference);
                racPassenger.allottedBerth = allocatedBerth;
                confirmedPassengers.add(racPassenger);
                availableBerths.remove(allocatedBerth);
                System.out.println("RAC ticket moved to confirmed: " + racPassenger);
            }

            if (!waitingListQueue.isEmpty()) {
                Passenger waitingPassenger = waitingListQueue.poll();
                racQueue.offer(waitingPassenger);
                waitingPassenger.allottedBerth = "RAC";
                System.out.println("Waiting list ticket moved to RAC: " + waitingPassenger);
            }

            System.out.println("Ticket cancelled successfully for ID: " + ticketId);
        } else {
            System.out.println("No ticket found with ID: " + ticketId);
        }
    }

    public void printBookedTickets() {
        if (confirmedPassengers.isEmpty()) {
            System.out.println("No confirmed tickets.");
        } else {
            System.out.println("Confirmed Tickets:");
            for (Passenger passenger : confirmedPassengers) {
                System.out.println(passenger);
            }
        }
    }

    public void printAvailableTickets() {
        System.out.println("Available Berths: " + availableBerths.size());
        System.out.println("Available RAC Tickets: " + (1 - racQueue.size()));
        System.out.println("Available Waiting List Tickets: " + (1 - waitingListQueue.size()));
    }
}
```


```java
class Member {
    String username;
    List<Book> borrowedBooks;

    Member(String username) {
        this.username = username;
        this.borrowedBooks = new ArrayList<>();
    }

    boolean borrowBook(Book book) {
        if (borrowedBooks.size() < 5 && book.isAvailable) {
            borrowedBooks.add(book);
            book.isAvailable = false;
            return true;
        }
        return false;
    }

    void returnBook(Book book) {
        borrowedBooks.remove(book);
        book.isAvailable = true;
    }
}
```

**Core Library Management System**

The `LibraryManagementSystem` class ties everything together.

```java
public class LibraryManagementSystem {
    static Scanner scanner = new Scanner(System.in);
    static List<Book> books = new ArrayList<>();
    static List<Member> members = new ArrayList<>();
    static Member loggedInMember = null;
    static final String adminUsername = "admin";
    static final String adminPassword = "password";

    public static void main(String[] args) {
        books.add(new Book("Science", "John Doe", "Educational"));
        books.add(new Book("Social Studies", "Jane Smith", "Educational"));

        while (true) {
            System.out.print("Enter role (admin/user): ");
            String role = scanner.nextLine();

            if (role.equals("admin")) loginAsAdmin();
            else if (role.equals("user")) loginAsUser();
            else System.out.println("Invalid role. Please try again.");
        }
    }

    // Admin Login
    private static void loginAsAdmin() {
        System.out.print("Enter username: ");
        String username = scanner.nextLine();
        System.out.print("Enter password: ");
        String password = scanner.nextLine();

        if (username.equals(adminUsername) && password.equals(adminPassword)) {
            System.out.println("Logged in as admin.");
            adminMenu();
        } else {
            System.out.println("Invalid admin credentials.");
        }
    }

    // User Login
    private static void loginAsUser() {
        System.out.print("Enter username: ");
        String username = scanner.nextLine();
        loggedInMember = findOrCreateMember(username);
        System.out.println("Logged in as user.");
        userMenu();
    }

    private static Member findOrCreateMember(String username) {
        for (Member member : members) {
            if (member.username.equals(username)) return member;
        }
        Member newMember = new Member(username);
        members.add(newMember);
        return newMember;
    }

    // Admin Menu
    private static void adminMenu() {
        while (true) {
            System.out.println("\nAdmin Menu:");
            System.out.println("1. Add Book");
            System.out.println("2. Update Book");
            System.out.println("3. Remove Book");
            System.out.println("4. Add Member");
            System.out.println("5. Display All Books");
            System.out.println("6. Display All Members");
            System.out.println("7. Exit");

            int choice = scanner.nextInt();
            scanner.nextLine();

            switch (choice) {
                case 1 -> addBook();
                case 2 -> updateBook();
                case 3 -> removeBook();
                case 4 -> addMember();
                case 5 -> displayAllBooks();
                case 6 -> displayAllMembers();
                case 7 -> { return; }
                default -> System.out.println("Invalid choice.");
            }
        }
    }

    // User Menu
    private static void userMenu() {
        while (true) {
            System.out.println("\nUser Menu:");
            System.out.println("1. Borrow Book");
            System.out.println("2. Return Book");
            System.out.println("3. Display All Books");
            System.out.println("4. Display All Members");
            System.out.println("5. Exit");

            int choice = scanner.nextInt();
            scanner.nextLine();

            switch (choice) {
                case 1 -> borrowBook();
                case 2 -> returnBook();
                case 3 -> displayAllBooks();
                case 4 -> displayAllMembers();
                case 5 -> { return; }
                default -> System.out.println("Invalid choice.");
            }
        }
    }

    private static void addBook() {
        System.out.print("Enter title: ");
        String title = scanner.nextLine();
        System.out.print("Enter author: ");
        String author = scanner.nextLine();
        System.out.print("Enter genre: ");
        String genre = scanner.nextLine();
        books.add(new Book(title, author, genre));
        System.out.println("Book added.");
    }

    private static void updateBook() {
        System.out.print("Enter title to update: ");
        String title = scanner.nextLine();
        Book book = findBookByTitle(title);
        if (book != null) {
            System.out.print("New author: ");
            book.author = scanner.nextLine();
            System.out.print("New genre: ");
            book.genre = scanner.nextLine();
            System.out.println("Book updated.");
        } else {
            System.out.println("Book not found.");
        }
    }

    private static void removeBook() {
        System.out.print("Enter title to remove: ");
        String title = scanner.nextLine();
        Book book = findBookByTitle(title);
        if (book != null) {
            books.remove(book);
            System.out.println("Book removed.");
        } else {
            System.out.println("Book not found.");
        }
    }

    private static void addMember() {
        System.out.print("Enter username: ");
        String username = scanner.nextLine();
        members.add(new Member(username));
        System.out.println("Member added.");
    }

    private static void borrowBook() {
        System.out.print("Enter book title: ");
        String title = scanner.nextLine();
        Book book = findBookByTitle(title);
        if (book != null && loggedInMember.borrowBook(book)) {
            System.out.println("Book borrowed.");
        } else {
            System.out.println("Unavailable or limit reached.");
        }
    }

    private static void returnBook() {
        System.out.print("Enter book title: ");
        String title = scanner.nextLine();
        Book book = findBookByTitle(title);
        if (book != null && !book.isAvailable) {
            loggedInMember.returnBook(book);
            System.out.println("Book returned.");
        } else {
            System.out.println("Book not found or not borrowed.");
        }
    }

    private static Book findBookByTitle(String title) {
        for (Book book : books) {
            if (book.title.equalsIgnoreCase(title)) return book;
        }
        return null;
    }

    private static void displayAllBooks() {
        if (books.isEmpty()) {
            System.out.println("No books available.");
        } else {
            for (Book book : books) {
                System.out.printf("Title: %s, Author: %s, Genre: %s, Available: %b%n",
                    book.title, book.author, book.genre, book.isAvailable);
            }
        }
    }

    private static void displayAllMembers() {
        if (members.isEmpty()) {
            System.out.println("No members.");
        } else {
            for (Member member : members) {
                System.out.printf("Username: %s, Borrowed Books: %d%n",
                    member.username, member.borrowedBooks.size());
            }
        }
    }
}
```

**Conclusion**

This Library Management System is a practical example of object-oriented programming, user roles, and enforcing business rules (like borrowing limits). The system can be expanded with features like overdue fines, book reservations, or even a GUI for better usability.


## Tic Tac Toe Game - Low-Level Design (LLD)

A robust command-line implementation of an **n x n Tic Tac Toe** game, inspired by a real technical interview round at **Zoho (Round 3)**.

**The Question: Build an n x n Tic Tac Toe Game**

**Requirements:**
1. **Dynamic grid size**: The game should support any grid size `n`.
2. **Two-player functionality**: Players X and O alternate turns.
3. **Win detection**: Check for matches in rows, columns, and diagonals.
4. **Edge cases**: Handle invalid input, repeated moves, and draw conditions.
5. **Clean code**: Maintain readability and avoid resource leaks.

**Approach and Design**

- **Board Initialization**: A 2D char array initialized with `-`.
- **Turn Management**: Players take turns, validated per move.
- **Win Checking**: Efficient logic for rows, columns, and diagonals.
- **Game State**: Track total moves and determine win or draw.
- **Validation**: Prevent out-of-bounds and repeated moves.

**Java Code: Step-by-Step Implementation**

```java
import java.util.Scanner;

public class TicTacToe {
    private static char[][] board;
    private static int n;
    private static char currentPlayer;

    public static void main(String[] args) {
        Scanner scanner = new Scanner(System.in);
        System.out.print("Enter the size of the grid (n x n): ");
        n = scanner.nextInt();
        board = new char[n][n];
        initializeBoard();

        currentPlayer = 'X';
        int moves = 0;
        boolean gameWon = false;

        while (moves < n * n && !gameWon) {
            printBoard();
            System.out.printf("Player %c, enter your move (row and column): ", currentPlayer);
            int row = scanner.nextInt();
            int col = scanner.nextInt();

            if (isValidMove(row, col)) {
                board[row][col] = currentPlayer;
                moves++;

                if (checkWin(row, col)) {
                    gameWon = true;
                    printBoard();
                    System.out.printf("Player %c wins!%n", currentPlayer);
                } else {
                    currentPlayer = (currentPlayer == 'X') ? 'O' : 'X';
                }
            } else {
                System.out.println("Invalid move. Try again.");
            }
        }

        if (!gameWon) {
            printBoard();
            System.out.println("The game is a draw!");
        }

        scanner.close();
    }

    private static void initializeBoard() {
        for (int i = 0; i < n; i++)
            for (int j = 0; j < n; j++)
                board[i][j] = '-';
    }

    private static void printBoard() {
        for (int i = 0; i < n; i++) {
            for (int j = 0; j < n; j++)
                System.out.print(board[i][j] + " ");
            System.out.println();
        }
    }

    private static boolean isValidMove(int row, int col) {
        return row >= 0 && row < n && col >= 0 && col < n && board[row][col] == '-';
    }

    private static boolean checkWin(int row, int col) {
        return checkRow(row) || checkColumn(col) || checkDiagonals();
    }

    private static boolean checkRow(int row) {
        for (int col = 0; col < n; col++)
            if (board[row][col] != currentPlayer)
                return false;
        return true;
    }

    private static boolean checkColumn(int col) {
        for (int row = 0; row < n; row++)
            if (board[row][col] != currentPlayer)
                return false;
        return true;
    }

    private static boolean checkDiagonals() {
        boolean diag1 = true, diag2 = true;
        for (int i = 0; i < n; i++) {
            if (board[i][i] != currentPlayer)
                diag1 = false;
            if (board[i][n - i - 1] != currentPlayer)
                diag2 = false;
        }
        return diag1 || diag2;
    }
}
```

**Key Features**

**1. Dynamic Grid Size**

* Supports any board size (`n x n`) provided at runtime.

**2. Efficient Win Checks**

* Optimized checks for rows, columns, and both diagonals.

**3. Edge Case Handling**

* Prevents invalid and repeated moves.
* Detects draw when all cells are filled without a winner.

**4. Optimized Logic**

* Clean separation of concerns.
* No unnecessary loops or redundant checks.

**Conclusion**

This implementation showcases how to build a fully functioning, scalable Tic Tac Toe game using core Java. The approach prioritizes:

* Code clarity
* Efficient logic
* Robust edge-case handling

## Snake Game - Low Level Design (LLD) in Java

This document walks through building a **Snake Game** in Java that runs directly in the command line.

We'll explore key concepts like grid initialization, snake movement, food placement, and collision detection. By the end, you’ll have a working game that you can tweak and improve.

**1. Game Requirements**

Before diving into the code, let’s outline the core requirements:

- **Grid Initialization**: Set up a fixed-dimension grid for gameplay.
- **Snake Movement**: Use keyboard inputs to control the snake.
- **Food Placement**: Randomly place food on the grid. Eating food extends the snake’s length.
- **Collision Detection**: End the game when the snake collides with itself or the grid boundaries.
- **Advanced Features** *(optional)*: Score tracking, dynamic grid resizing, or obstacles.

**2. Solution Overview**

The game uses the following Java components:

- A 2D array to represent the game grid.
- A queue to track the snake’s body positions (FIFO).
- Input handling to read movement directions.
- Basic game logic to handle food placement, movement, and collisions.

**3. Implementation**

```java
import java.util.*;

public class Main {
    public static void main(String[] args) {  
        SnakeGame snakeGame = new SnakeGame(6, 6); 
        snakeGame.start(); 
    }
}

class SnakeGame {
    private final char[][] board;
    private final Queue<Node> snakeBody = new LinkedList<>();
    private final Queue<Node> foodQueue = new LinkedList<>();
    private final Random random = new Random();
    private int currentRow = 0;
    private int currentCol = 0;
    private char currentHead = '>';

    public SnakeGame(int rows, int cols) {
        board = new char[rows][cols];
        snakeBody.add(new Node(0, 0));
        board[0][0] = currentHead;
        generateFoodPositions();
        placeFood();
    }

    public void start() {
        Scanner scanner = new Scanner(System.in);
        printBoard();
        while (true) {
            System.out.print("Enter move (U, D, L, R): ");
            char move = scanner.next().toUpperCase().charAt(0);
            if (makeMove(move)) {
                printBoard();
            } else {
                System.out.println("Game Over!");
                break;
            }
        }
        scanner.close();
    }

    private boolean makeMove(char direction) {
        int newRow = currentRow, newCol = currentCol;
        char newHead;

        switch (direction) {
            case 'U': newRow--; newHead = '^'; break;
            case 'D': newRow++; newHead = 'v'; break;
            case 'L': newCol--; newHead = '<'; break;
            case 'R': newCol++; newHead = '>'; break;
            default:
                System.out.println("Invalid move! Use U, D, L, R.");
                return true;
        }

        if (!isValidMove(newRow, newCol)) return false;

        if (board[newRow][newCol] == 'X') {
            placeFood();
        } else {
            Node tail = snakeBody.poll();
            board[tail.row][tail.col] = '\0';
        }

        board[currentRow][currentCol] = '.';
        snakeBody.add(new Node(newRow, newCol));
        currentRow = newRow;
        currentCol = newCol;
        currentHead = newHead;
        board[newRow][newCol] = currentHead;

        return true;
    }

    private boolean isValidMove(int row, int col) {
        return row >= 0 && row < board.length && col >= 0 && col < board[0].length && board[row][col] != '.';
    }

    private void generateFoodPositions() {
        foodQueue.add(new Node(1, 0));
        foodQueue.add(new Node(2, 2));
        foodQueue.add(new Node(3, 4));
        foodQueue.add(new Node(5, 2));
        foodQueue.add(new Node(4, 5));
    }

    private void placeFood() {
        Node food;
        if (!foodQueue.isEmpty()) {
            food = foodQueue.poll();
        } else {
            do {
                food = new Node(random.nextInt(board.length), random.nextInt(board[0].length));
            } while (board[food.row][food.col] != '\0');
        }
        board[food.row][food.col] = 'X';
    }

    private void printBoard() {
        for (char[] row : board) {
            for (char cell : row) {
                System.out.print((cell == '\0' ? '■' : cell) + " ");
            }
            System.out.println();
        }
        System.out.println();
    }
}

class Node {
    final int row, col;

    Node(int row, int col) {
        this.row = row;
        this.col = col;
    }
}
````

**4. Key Concepts Explained**

**Grid Representation**

* A 2D char array represents the game board.
* Symbols:

  * `'■'` – Empty cell
  * `'.'` – Snake body
  * `'X'` – Food
  * `'>'`, `'<', '^', 'v'` – Snake head in respective directions

**Snake Movement**

* Controlled by input commands:

  * `U` = Up
  * `D` = Down
  * `L` = Left
  * `R` = Right

**Food Placement**

* Initially placed from a queue of predefined locations.
* When exhausted, new food appears randomly in empty cells.

**Collision Detection**

* The game ends if the snake:

  * Runs into its own body
  * Crosses the grid boundary

```java
class Encryption {
    public String encrypt(String input) {
        StringBuilder encrypted = new StringBuilder();
        for (char ch : input.toCharArray()) {
            if (Character.isUpperCase(ch)) encrypted.append((char) ((ch - 'A' + 1) % 26 + 'A'));
            else if (Character.isLowerCase(ch)) encrypted.append((char) ((ch - 'a' + 1) % 26 + 'a'));
            else if (Character.isDigit(ch)) encrypted.append((char) ((ch - '0' + 1) % 10 + '0'));
            else encrypted.append(ch);
        }
        return encrypted.toString();
    }
```

## Brick Breaker - Low Level Design (LLD)

A grid-based simulation game inspired by classic brick-breaking mechanics. This LLD includes data structures, commands, gameplay logic, and example implementations.


**1. System Overview**

A 7x7 grid-based brick breaker game. The player controls a ball that can move in three directions to break bricks. Bricks have strength values that decrease with each hit. The game simulates ball movement, collision with walls/bricks, and tracking ball lives.

**2. Game Grid Representation**

- `w`: Wall  
- `g`: Ground  
- `o`: Ball  
- `1`, `2`, `3`, etc.: Bricks with strength values

**3. Initial Setup**

- **Grid size**: 7x7
- **Number of bricks**: 6
- **Brick positions**:
  - (2, 2), (2, 3), (2, 4)
  - (3, 2), (3, 3), (3, 4)
- **Ball initial position**: (6, 3)
- **Ball lives**: 5

**4. Game Mechanics**

- Ball can move in 3 directions:
  - `St`: Straight Up
  - `Lt`: Left-Up Diagonal
  - `Rt`: Right-Up Diagonal
- Ball behavior:
  - Hits reduce brick strength.
  - If a brick disappears, ball advances to that position.
  - If not, ball returns to its origin.
  - Hitting a wall ends the turn and deducts a life.

**5. Gameplay Commands**

| Command | Description |
|---------|-------------|
| `St`    | Straight up movement |
| `Lt`    | Left diagonal movement |
| `Rt`    | Right diagonal movement |

**6. Java Implementation**

```java
import java.util.*;

public class BrickBreaker {
    static char[][] grid = new char[7][7];
    static int[][] bricks = new int[7][7];
    static int ballRow = 6, ballCol = 3;
    static int lives = 5;

    public static void main(String[] args) {
        setupGrid();
        Scanner sc = new Scanner(System.in);

        while (lives > 0) {
            printGrid();
            System.out.print("Enter command (St, Lt, Rt): ");
            String cmd = sc.nextLine().trim();

            switch (cmd.toLowerCase()) {
                case "st" -> moveStraight();
                case "lt" -> moveDiagonal(-1);
                case "rt" -> moveDiagonal(1);
                default -> System.out.println("Invalid command");
            }
        }

        System.out.println("Game Over!");
    }

    static void setupGrid() {
        for (int i = 0; i < 7; i++)
            Arrays.fill(grid[i], ' ');

        for (int i = 0; i < 7; i++) {
            grid[0][i] = grid[6][i] = 'w';
            grid[i][0] = grid[i][6] = 'w';
        }

        for (int i = 1; i < 6; i++)
            grid[6][i] = 'g';

        grid[6][3] = 'o';

        int[][] brickPos = {{2, 2}, {2, 3}, {2, 4}, {3, 2}, {3, 3}, {3, 4}};
        for (int[] pos : brickPos) {
            grid[pos[0]][pos[1]] = '1';
            bricks[pos[0]][pos[1]] = 1;
        }
    }

    static void moveStraight() {
        int r = ballRow - 1;
        while (r >= 0 && grid[r][ballCol] == ' ') r--;

        if (r <= 0 || grid[r][ballCol] == 'w') {
            System.out.println("Hit wall. Life lost.");
            lives--;
            return;
        }

        if (Character.isDigit(grid[r][ballCol])) {
            bricks[r][ballCol]--;
            if (bricks[r][ballCol] == 0) {
                grid[r][ballCol] = ' ';
                grid[ballRow][ballCol] = 'g';
                ballRow = r;
            }
            lives--;
        }
    }

    static void moveDiagonal(int direction) {
        int r = ballRow - 1;
        int c = ballCol + direction;

        while (r > 0 && c > 0 && c < 6 && grid[r][c] == ' ') {
            r--;
            c += direction;
        }

        if (r <= 0 || c <= 0 || c >= 6 || grid[r][c] == 'w') {
            System.out.println("Hit wall. Life lost.");
            lives--;
            return;
        }

        if (Character.isDigit(grid[r][c])) {
            bricks[r][c]--;
            if (bricks[r][c] == 0) {
                grid[r][c] = ' ';
                grid[ballRow][ballCol] = 'g';
                ballRow = r;
                ballCol = c;

                // Move downward
                while (ballRow + 1 < 6 && grid[ballRow + 1][ballCol] == 'g') {
                    ballRow++;
                }
            }
            lives--;
        }

        updateBall();
    }

    static void updateBall() {
        for (int i = 1; i < 6; i++) {
            for (int j = 1; j < 6; j++) {
                if (grid[i][j] == 'o') grid[i][j] = 'g';
            }
        }
        grid[ballRow][ballCol] = 'o';
    }

    static void printGrid() {
        for (char[] row : grid) {
            for (char ch : row) System.out.print(ch + " ");
            System.out.println();
        }
        System.out.println("Ball count: " + lives);
    }
}
```

## Lift System - Low Level Design (LLD)

This is a simulation of a lift (elevator) allocation system. It consists of multiple modules to manage lift position, assignment, direction logic, capacity, and maintenance status.

**1. System Overview**

The system controls **5 lifts** (`L1` to `L5`) and assigns them to user requests based on proximity, direction, floor restrictions, capacity, and maintenance status.

**2. Functional Modules**

1. **Display lift positions**
2. **Assign lift to users**
3. **Nearest lift selection**
4. **Tie-breaker by movement direction**
5. **Lift floor range constraints**
6. **Least number of stops**
7. **Lift capacity constraint**
8. **Maintenance mode**

**3. Lift Assignment Rules**

| Rule # | Condition/Logic                                                                 |
|--------|----------------------------------------------------------------------------------|
| 1      | Show current positions of lifts                                                 |
| 2      | Assign first available lift for simple request                                  |
| 3      | Assign nearest lift based on absolute floor difference                          |
| 4      | Direction-aware tie-breaking if distances match                                 |
| 5      | L1 & L2 → 0-5, L3 & L4 → 6-10, L5 → 0-10                                         |
| 6      | Choose lift with least number of intermediate stops                             |
| 7      | Respect lift capacity                                                           |
| 8      | Skip lifts with current position marked as `-1` (maintenance)                   |

**4. Java Implementation**

```java
import java.util.*;

class Lift {
    String id;
    int floor;
    int minFloor, maxFloor;
    int stops;
    int capacity;
    boolean underMaintenance;

    Lift(String id, int min, int max, int cap) {
        this.id = id;
        this.floor = 0;
        this.minFloor = min;
        this.maxFloor = max;
        this.stops = 0;
        this.capacity = cap;
        this.underMaintenance = false;
    }

    boolean canServe(int source, int dest, int people) {
        return !underMaintenance &&
               source >= minFloor && source <= maxFloor &&
               dest >= minFloor && dest <= maxFloor &&
               people <= capacity;
    }

    int stopsToReach(int source, int dest) {
        return Math.abs(floor - source) + Math.abs(source - dest);
    }
}

public class LiftSystem {
    static List<Lift> lifts = new ArrayList<>();

    public static void main(String[] args) {
        setupLifts();
        Scanner sc = new Scanner(System.in);

        while (true) {
            System.out.print("\nEnter source, dest, people (or -1 to exit): ");
            int src = sc.nextInt();
            if (src == -1) break;
            int dst = sc.nextInt();
            int people = sc.nextInt();

            assignLift(src, dst, people);
            displayLiftPositions();
        }
    }

    static void setupLifts() {
        lifts.add(new Lift("L1", 0, 5, 4));
        lifts.add(new Lift("L2", 0, 5, 4));
        lifts.add(new Lift("L3", 6, 10, 6));
        lifts.add(new Lift("L4", 6, 10, 6));
        lifts.add(new Lift("L5", 0, 10, 10));
    }

    static void assignLift(int source, int dest, int people) {
        Lift bestLift = null;
        int minStops = Integer.MAX_VALUE;

        for (Lift lift : lifts) {
            if (!lift.canServe(source, dest, people)) continue;

            int stops = lift.stopsToReach(source, dest);
            if (stops < minStops) {
                minStops = stops;
                bestLift = lift;
            } else if (stops == minStops) {
                boolean userGoingUp = dest > source;
                boolean liftGoingUp = lift.floor < source;
                if (userGoingUp == liftGoingUp) {
                    bestLift = lift;
                }
            }
        }

        if (bestLift != null) {
            System.out.println(bestLift.id + " is assigned.");
            bestLift.floor = dest;
            bestLift.stops++;
        } else {
            System.out.println("No lift available for this request.");
        }
    }

    static void displayLiftPositions() {
        System.out.print("Lift   : ");
        for (Lift l : lifts) System.out.print(l.id + " ");
        System.out.print("\nFloor  : ");
        for (Lift l : lifts) System.out.print((l.underMaintenance ? "-1" : l.floor) + " ");
        System.out.println();
    }
}
```
