## Problem Statement:108  
Given a **singly linked list** of `n` nodes and a **key**, the task is to check whether the key is present in the linked list or not.  

**Input**  
- A **singly linked list** with `n` nodes.  
- An **integer key** to search in the linked list.  

**Output**  
- `true` if the key is found in the linked list.  
- `false` otherwise.  

**Constraints**  
- `1 ≤ n ≤ 10^4`  
- `1 ≤ Node data, Key ≤ 10^6`  

**Example**  
**Input:**  
- Linked List: 10 → 20 → 30 → 40  
- Key: 30  

**Output:**  
true  

**Goal of the Code**  
To efficiently search for a **key** in a **singly linked list** by traversing the list iteratively.  

**Approach**  
1. Start from the **head** of the linked list.  
2. Traverse the list **node by node**.  
3. If the **current node's data** matches the key, return `true`.  
4. If the traversal reaches the end without finding the key, return `false`.  

**🧮 Flowchart/ Diagram:**  
1. Start  
2. Set `current` to `head`  
3. While `current` is not null:  
   - If `current.data == key`: return true  
   - Else: move to `current.next`  
4. If end reached without match: return false  
5. End 

**Pseudocode/Algorithm**  
```java
function searchKey(head, key):  
  current = head  
  while current is not null:  
    if current.data == key:  
      return true  
    current = current.next  
  return false  
  ```

  **Code:**
  ```java

/* Node of a linked list
  class Node {
   int data;
    Node next;
    Node(int d)  { data = d;  next = null; }
}
*/
class Solution {
    static boolean searchKey(int n, Node head, int key) {
        // Code here
        Node curr=head;
        while(curr!=null){
            if(curr.data ==key){
                return true;
            }
            curr=curr.next;
        }
        return false;
    }
}
  ```
**Time Complexity:** O(n) → Each node is visited once in the worst case.
**Space Complexity:** O(1) → No extra space is used.

**Additional Notes/Hints**
This approach works for unsorted linked lists as well.
In a sorted linked list, we can stop searching early if the current node's data exceeds the key.
If frequent searches are needed, hashing or a balanced BST might be better alternatives.

**Notes:**  
- The linear search works on both sorted and unsorted singly linked lists.  
- No additional data structures are needed, making it memory efficient.  
- For large-scale or frequent searches, alternate data structures may improve performance.  



## Problem Statement: 109
**Reverse a Linked List**  
Given the **head** of a **singly linked list**, the task is to **reverse** the linked list and return the **new head** of the reversed list.  

**Input**  
- A **singly linked list** with `n` nodes.  

**Output**  
- The **head** of the reversed linked list.  

**Constraints**  
- `1 ≤ n ≤ 10^5`  
- `1 ≤ Node data ≤ 10^6`  

**Example**  
**Input:**  
Linked List: 1 → 2 → 3 → 4 → 5  
**Output:**  
Reversed List: 5 → 4 → 3 → 2 → 1  

**Goal of the Code**  
To reverse a **singly linked list** in-place without using extra space.  

**Approach**  
1. Initialize three pointers:  
   - `prev` → initially `null` (tracks the previous node).  
   - `curr` → initially `head` (tracks the current node).  
   - `next` → stores the next node temporarily.  
2. Traverse the list while `curr` is not null:  
   - Store `next = curr.next` (save the next node).  
   - Reverse the link: `curr.next = prev`.  
   - Move `prev` and `curr` one step forward.  
3. Return `prev` as the new head (since `curr` becomes `null` at the end).  

**🧮 Flowchart/ Diagram:**  
1. Start  
2. Initialize `prev = null`, `curr = head`  
3. While `curr != null`:  
   - Store `next = curr.next`  
   - Set `curr.next = prev`  
   - Update `prev = curr`, `curr = next`  
4. Return `prev` as new head  
5. End  

**Pseudocode/Algorithm**  
```java
function reverseList(head):  
  prev = null  
  curr = head  
  while curr is not null:  
    next = curr.next  
    curr.next = prev  
    prev = curr  
    curr = next  
  return prev  
```
**Code:**
```java

    int data;
    Node next;
    Node(int value) {
        this.value = value;
    }
}

*/

    Node reverseList(Node head) {
        // code here
        Node curr=head, prev=null, next;
     
        while(curr!=null){
            next=curr.next;
            curr.next=prev;
            
            prev=curr;
            curr=next;
        }
        return prev;
    }
}
```
**Time Complexity:** O(n) → Each node is visited once.
**Space Complexity:** O(1) → Only a few extra pointers are used.

**Additional Notes/Hints**
This approach reverses the linked list in-place without extra memory.
For recursive reversal, we can use a stack or recursion, but it increases space complexity to O(n).
This method maintains constant space O(1), making it optimal for large lists.

**Notes:**  
- This in-place iterative reversal ensures optimal performance for large lists.  
- Useful in algorithms involving stack reversal, backtracking, or undo mechanisms.  

## Problem Statement: 110
**Palindrome Linked List** 
Given the **head** of a **singly linked list** of **positive integers**, the task is to check if the linked list is a **palindrome** or not.  

**Input**  
- A **singly linked list** with `n` nodes.  

**Output**  
- `true` if the linked list is a **palindrome**.  
- `false` otherwise.  

**Constraints**  
- `1 ≤ n ≤ 10^5`  
- `1 ≤ Node data ≤ 10^6`  

**Example**  
**Input:**  
Linked List: 1 → 2 → 3 → 2 → 1  
**Output:**  
true  

**Goal of the Code**  
To determine if a **singly linked list** reads the same **forward and backward**.  

**Approach**  
1. Use **two pointers** (`slow` and `fast`) to find the **middle** of the linked list.  
2. Reverse the **second half** of the linked list.  
3. Compare the **first half** with the **reversed second half**.  
4. Restore the original list structure by reversing the second half again.  
5. Return `true` if both halves match; otherwise, return `false`.  

**🧮 Flowchart/ Diagram:**  
1. Start  
2. Find the middle of the list using slow and fast pointers  
3. Reverse the second half of the list  
4. Compare both halves  
5. Restore the list  
6. Return true if matched, else false  
7. End  

**Pseudocode/Algorithm**  
```java
function isPalindrome(head):  
  if head is null or head.next is null:  
    return true  

  slow = head  
  fast = head  

  while fast.next is not null and fast.next.next is not null:  
    slow = slow.next  
    fast = fast.next.next  

  head2 = reverseList(slow.next)  
  slow.next = null  

  result = isMatch(head, head2)  

  head2 = reverseList(head2)  
  slow.next = head2

  curr = head

    next = curr.next  
    curr.next = prev  
    prev = curr  
    curr = next  
  return prev  

function isMatch(n1, n2):  
  while n1 is not null and n2 is not null:  
    if n1.data ≠ n2.data:  
      return false  
    n1 = n1.next  
    n2 = n2.next  
  return true  
```

## Problem Statement: 111
**Insert in Middle of Linked List**  
Given the **head** of a **Singly Linked List** and a **value `x`**, the task is to **insert** the key `x` in the **middle** of the linked list.  

**Input**  
- A **singly linked list** with `n` nodes.  
- An **integer `x`**, representing the value to be inserted.  

**Output**  
- The **updated linked list** after inserting `x` in the middle.  

**Constraints**  
- `1 ≤ n ≤ 10^5`  
- `1 ≤ Node data, x ≤ 10^6`  

**Example**  
**Input:**  
Linked List: 1 → 2 → 4 → 5  
x = 3  
**Output:**  
1 → 2 → 3 → 4 → 5  

**Goal of the Code**  
To efficiently insert an **element in the middle** of a singly linked list while maintaining the list structure.  

**Approach**  
1. **Find the length** of the linked list.  
2. Calculate the **middle position** using `(length / 2)` (considering 0-based indexing).  
3. Traverse to the **middle node**.  
4. Insert the **new node** after the middle node.  
5. Return the **updated head** of the linked list.  

**🧮 Flowchart/ Diagram:**  
1. Start  
2. Traverse to calculate the length  
3. Compute middle index  
4. Traverse to node before middle  
5. Create and link new node  
6. Return head  
7. End  

**Pseudocode/Algorithm**  
```java
function insertInMiddle(head, x):  
  if head is null:  
    return new Node(x)  

  newNode = new Node(x)  
  currNode = head  
  length = 0  

  while currNode is not null:  
    length++  
    currNode = currNode.next  

  count = (length / 2) if length is even else (length + 1) / 2  
  currNode = head  

  while count-- > 1:  
    currNode = currNode.next  

  newNode.next = currNode.next  
  currNode.next = newNode  

  return head  
```

**Additional Notes/Hints**
If the list is empty, the new node becomes the head.
If the list has only one element, the new node is inserted after it.
This method modifies the list in-place without using extra memory.

**Notes:**  
- For even-length lists, the element is inserted after `n/2` elements.  
- This logic ensures the inserted node lies exactly in the middle region.  
- The code maintains a stable list structure and is suitable for large datasets due to O(1) space.


## Problem Statement: 112
**Intersection in Y Shaped Lists**  
Given the **head of two singly linked lists**, return the **point where these two linked lists intersect**.  

**Note:** It is guaranteed that the **intersected node always exists**.  

**Input**  
- `head1`: The **head of the first linked list**.  
- `head2`: The **head of the second linked list**.  

**Output**  
- The **intersection node** where both linked lists merge.  

**Constraints**  
- `1 ≤ n1, n2 ≤ 10^5`  
- `1 ≤ Node data ≤ 10^6`  
- The two linked lists **must intersect** at some point.  

**Example**  
**Input:**  
List1: 10 → 15 → 30  
List2: 3 → 6 → 9 → 15 → 30  
**Output:**  
15  

**Goal of the Code**  
To efficiently **find the intersection node** of two linked lists while ensuring **optimal time complexity**.  

**Approach**  
1. **Find the length** of both linked lists (`len1` and `len2`).  
2. Compute the **difference `diff`** between the lengths.  
3. Move the **longer list ahead by `diff` nodes**, so both lists become of equal length.  
4. Traverse both lists **simultaneously** until the intersection node is found.  
5. Return the **first common node**.  

**🧮 Flowchart/ Diagram:**  
1. Start  
2. Get lengths of both lists  
3. Calculate length difference  
4. Advance the longer list by the difference  
5. Traverse both lists simultaneously  
6. Return the first common node  
7. End  

**Pseudocode/Algorithm**  
```java
function getCount(head):
    count = 0
    currNode = head
    while currNode is not null:
        count++
        currNode = currNode.next
    return count

function getIntersection(diff, head1, head2):
    curr1 = head1
    curr2 = head2

    for i from 0 to diff:
        if curr1 is null:
            return null
        curr1 = curr1.next

    while curr1 is not null and curr2 is not null:
        if curr1 == curr2:
            return curr1
        curr1 = curr1.next
        curr2 = curr2.next

    return null

function intersectPoint(head1, head2):
    len1 = getCount(head1)
    len2 = getCount(head2)
    diff = abs(len1 - len2)

    if len1 > len2:
        return getIntersection(diff, head1, head2)
    else:
        return getIntersection(diff, head2, head1)
```


Using hashing can solve it in O(n1 + n2) time but uses extra space.
The two-pointer technique used in this approach is optimal, working in O(n1 + n2) time and O(1) space.

**Notes:**  
- Intersection is by **reference**, not value. Two nodes with same value but different memory addresses are not considered intersection.  
- This approach does not require modifying the input lists.  
- The first intersection node returned is guaranteed due to input constraints.

## Problem Statement: 115
**Union of Two Linked Lists**  
Given two linked lists (**L1** & **L2**), return the **union** list of both lists, containing only **distinct** elements in **sorted order**.  

**Input**  
- Two linked lists: `head1` and `head2`.  

**Output**  
- A **sorted linked list** containing the **union** of elements from both lists.  

**Constraints**  
- `1 ≤ Length of L1, L2 ≤ 10^4`  
- `0 ≤ Node Value ≤ 10^5`  

**Example**  
**Input:**  
L1: 9 → 6 → 4 → 2  
L2: 5 → 2 → 8 → 6  
**Output:**  
2 → 4 → 5 → 6 → 8 → 9  

**Goal of the Code**  
To efficiently find the **union** of two linked lists while ensuring all elements are **distinct** and sorted in **ascending order**.  

**Approach**  
1. Use a **TreeSet** (Self-balancing BST) to store **unique elements** from both linked lists.  
2. Traverse both linked lists and **insert** their elements into the **TreeSet**.  
3. Create a **new linked list** from the sorted elements of the **TreeSet**.  
4. Return the **head** of the newly constructed linked list.  

**🧮 Flowchart/ Diagram:**  
1. Start  
2. Initialize a `TreeSet`  
3. Traverse both linked lists  
4. Add all elements to `TreeSet`  
5. Create new linked list using sorted elements in set  
6. Return head of new list  
7. End  

**Pseudocode/Algorithm**  
```java
function findUnion(head1, head2):  
    Initialize a TreeSet `set` to store unique elements in sorted order.  

    Traverse the first linked list:
        Add each element to `set`.  

    Traverse the second linked list:
        Add each element to `set`.  

    Create a new dummy node `dummyHead`.  
    Initialize `current` pointer to `dummyHead`.  

    For each element `num` in `set`:
        Create a new node with value `num` and link it to `current`.  
        Move `current` pointer forward.  

    Return `dummyHead.next` (head of the new linked list).  

```

**Code:**
```java

class Solution {
    public static Node findUnion(Node head1, Node head2) {
        // Add your code here.
        TreeSet<Integer> set=new TreeSet<>();
        Node temp=head1;
        while(temp!=null){
            set.add(temp.data);
            temp=temp.next;
        }
        temp=head2;
        while(temp!=null){
            set.add(temp.data);
            temp=temp.next;
        }
        
        Node dummy=new Node(-1);
        Node curr=dummy;
        
        for(int num: set){
            curr.next=new Node(num);
            curr=curr.next;
        }
        
        return dummy.next;
        
    }
}
```

**Time Complexity:** O(N log N) (due to TreeSet insertions, where N is the total number of nodes).
**Space Complexity:** O(N) (for storing unique elements in TreeSet).

**Additional Notes/Hints**
Alternative Approach: Using HashSet + Sorting can also be used but would require additional sorting overhead.
Edge Case Handling: If either linked list is empty, return the non-empty list as the union.

**Notes:**  
- TreeSet ensures **sorting and uniqueness** automatically.  
- Dummy node simplifies linked list construction.  
- The output list does not modify the original lists.


**Pseudocode/Algorithm:**

```java
function getIntersection(head1, head2):
    set = empty hash set
    result = null

    for each node in head1:
        set.add(node.data)

    for each node in head2:
        if set.contains(node.data):
            create new node with node.data
            insert at beginning of result list

    return result

function getUnion(head1, head2):
    unionSet = empty hash set

    for each node in head1:
        unionSet.add(node.data)

    for each node in head2:
        unionSet.add(node.data)

    result = null
    tail = null
    for each value in unionSet:
        create new node with value
        if result is null:
            result = node
            tail = node
        else:
            tail.next = node
            tail = node

    return result
```

**Code:**

```java
import java.util.HashSet;
import java.util.Set;

class Node {
    int data;
    Node next;

    Node(int x) {
        data = x;
        next = null;
    }
}

class GfG {

    static void printList(Node head) {
        Node curr = head;
        while (curr != null) {
            System.out.print(curr.data + " ");
            curr = curr.next;
        }
        System.out.println();
    }

    // Function to get the intersection of two lists
    static Node getIntersection(Node head1, Node head2) {
        Set<Integer> set = new HashSet<>();
        Node result = null;

        // Traverse the first list and store elements in a hash set
        Node p = head1;
        while (p != null) {
            set.add(p.data);
            p = p.next;
        }

        // Traverse the second list and check for common elements
        p = head2;
        while (p != null) {
            if (set.contains(p.data)) {
                // Insert at the beginning of the result list
                Node newNode = new Node(p.data);
                newNode.next = result;
                result = newNode;
            }
            p = p.next;
        }

        return result;
    }

    // Function to get the union of two lists
    static Node getUnion(Node head1, Node head2) {
        Set<Integer> unionSet = new HashSet<>();
        Node p = head1;
        while (p != null) {
            unionSet.add(p.data);
            p = p.next;
        }

        p = head2;
        while (p != null) {
            unionSet.add(p.data);
            p = p.next;
        }

        Node result = null;
        Node tail = null;

        // Insert all unique values into the result list
        for (int i : unionSet) {
            Node newNode = new Node(i);
            if (result == null) {
                result = newNode;
                tail = newNode;
            } else {
                tail.next = newNode;
                tail = tail.next;
            }
        }

        return result;
    }

    // Function to print both union and intersection of two lists
    static void printUnionIntersection(Node head1, Node head2) {
        Node intersectionList = getIntersection(head1, head2);
        Node unionList = getUnion(head1, head2);

        System.out.println("Intersection list is:");
        printList(intersectionList);

        System.out.println("Union list is:");
        printList(unionList);
    }

    public static void main(String[] args) {

        // List 1: 1 -> 2 -> 3 -> 3 -> 4 -> 5
        Node head1 = new Node(1);
        head1.next = new Node(2);
        head1.next.next = new Node(3);
        head1.next.next.next = new Node(3);
        head1.next.next.next.next = new Node(4);
        head1.next.next.next.next.next = new Node(5);

        // List 2: 1 -> 5 -> 6
        Node head2 = new Node(1);
        head2.next = new Node(5);
        head2.next.next = new Node(6);

        printUnionIntersection(head1, head2);
    }
}
```

**Time Complexity:** O(m + n) where m and n are the lengths of the two lists.
**Space Complexity:** O(m + n) for storing elements in hash sets.

**Notes:**

* Insertion in the intersection list is done at the head for simplicity; order is not important.
* The union list preserves no specific order due to the nature of hash sets.
* Handles duplicates internally in lists by virtue of hashing only unique values.

## Problem Statement: 113
**Delete without Head Pointer**  
You are given a **node (del_node) of a Singly Linked List** and must delete it from the linked list.  

**Constraints:**  
- You **do not have access to the head** of the linked list.  
- The given `del_node` **will not be the last node** in the linked list.  
- The **number of nodes should decrease by one**, but memory deallocation is **not required**.  
- The **order of remaining nodes must be maintained**.  

**Input**  
- `del_node`: A reference to the node that needs to be deleted.  

**Output**  
- The linked list with `del_node` removed.  

**Example**  
**Input:**  
Linked List: 1 → 2 → 3 → 4  
del_node: Node with value 3  

**Output:**  
1 → 2 → 4  

**Approach**  
Since we **do not have access to the head**, we cannot traverse the list normally. Instead, we **modify the given node itself**:  
1. **Copy the data** of the **next node** into `del_node`.  
2. **Change the pointer** of `del_node` to **skip** the next node.  
3. This effectively **removes the next node** and replaces `del_node` with its successor.  

**Algorithm / Pseudocode**  
```java
function deleteNode(node):
    if node is null:
        return
    if node.next is null: 
        return  // Edge case (but problem guarantees it's not the last node)
    
    node.data = node.next.data  // Copy next node's data to current node
    node.next = node.next.next  // Skip the next node (effectively deleting it)
```

**Code:**
```java


/*
class Node
{
    int data ;
    Node next;
    Node(int d)
    {
        data = d;
        next = null;
    }
}
*/

// Function to delete a node without any reference to head pointer.
class Solution {
    void deleteNode(Node node) {
        // Your code here
        if(node==null){
            return;
        }
        else if(node.next==null){
            return;
        }
        
        node.data=node.next.data;
        node.next=node.next.next;
    }
}
```

**Time Complexity:** O(1) (Constant time deletion)
**Space Complexity:** O(1) (No extra space used)

**Note:**  
- Memory deallocation is not required since Java uses garbage collection.
- Works only if `del_node` is **not the last node**, since we rely on accessing `del_node.next`.


## Problem Statement: 116
**Count Pairs Whose Sum is Equal to X**
Given two linked lists **head1** and **head2** with **distinct elements**, find the count of all **distinct pairs** whose sum equals a given value **X**.  

**Note:**  
A valid pair is in the form `(a, b)`, where:  
- `a` is from **head1**  
- `b` is from **head2**  
- **Order matters**: `(1, 3)` and `(3, 1)` are different pairs.  

**Input**  
- Two **linked lists** (`head1` and `head2`).  
- An **integer X** representing the required sum.  

**Output**  
- An **integer** representing the count of valid pairs.  

**Constraints**  
- `1 ≤ Length of L1, L2 ≤ 10^4`  
- `1 ≤ Node Value, X ≤ 10^5`  

**Example**  
**Input:**  
head1: 1 → 2 → 3  
head2: 3 → 4 → 5  
X = 6  

**Output:**  
2  

**Explanation:**  
The valid pairs are (1, 5) and (2, 4) whose sums equal 6.  

**Goal of the Code**  
To efficiently count **distinct pairs** whose sum equals **X** using **Hashing** for optimal performance.  

**Approach**  
1. **Use a HashSet** to store elements of `head1`.  
2. Traverse `head2`, and for each element `b`, check if `X - b` exists in the **HashSet**.  
3. If found, **increase the count**.  
4. Return the **total count** of such pairs.  

**Pseudocode/Algorithm**  
```java
function countPairs(head1, head2, X):  
    Initialize HashSet `set`  
    Initialize `count = 0`  

    Traverse `head1`:  
        Insert each element into `set`  

    Traverse `head2`:  
        If `(X - current node value)` exists in `set`:  
            Increment `count`  

    Return `count`  
```

**Code:**
```java

/*class Node {
    int data;
    Node next;

    Node(int x) {
        data = x;
        next = null;
    }
}*/

class Solution {
    // Function to count pairs in two linked lists whose sum is equal to x
    public int countPairs(Node head1, Node head2, int x) {
        // add your code 
        HashSet<Integer> set=new HashSet<>();
        int count=0;
        Node temp=head1;
        while(temp!=null){
            set.add(temp.data);
            temp=temp.next;
        }
        
        temp=head2;
        while(temp!=null){
            if(set.contains(x-temp.data)){
                count++;
            }
            temp=temp.next;
        }
        return count;
    }
}

```

**Time Complexity:** O(N + M) (where N and M are the lengths of head1 and head2).
**Space Complexity:** O(N) (for storing elements in the HashSet).

**Additional Notes/Hints**  
- Each element in the linked lists is **distinct**, which avoids counting duplicates.  
- The problem does **not require returning the actual pairs**, only their **count**.  
- Using a HashSet ensures **constant time lookups**, making the approach efficient.  
- Be careful not to reverse the roles of `head1` and `head2`, as `(a, b)` and `(b, a)` are treated as **different** and only `a ∈ head1`, `b ∈ head2` is valid.  
- Edge Case: If one of the lists is empty, the result is `0` as no pairs can be formed.


## Problem Statement: Count pairs from two linked lists whose sum is equal to a given value

Given two linked lists (can be sorted or unsorted) of sizes n1 and n2 containing distinct elements, and a value x, count all pairs formed by taking one element from each linked list such that their sum equals x.

**Example:**  
List 1: 3 → 1 → 5 → 7  
List 2: 8 → 2 → 5 → 3  
Target sum (x): 10

**Output:**  
Count = 2  
Explanation:  
Pairs are (3,7) and (5,5) since 3+7=10 and 5+5=10.

**Input:**  
- Two linked lists of distinct integers.  
- An integer value x representing the target sum.

**Output:**  
- An integer representing the count of pairs from both lists whose sum is equal to x.

**Constraints:**  
- Lists can be sorted or unsorted.  
- Elements in each list are distinct.

**Goal:**  
Count pairs efficiently without modifying original lists beyond sorting.

**Approach:**  
1. Sort the first linked list in ascending order.  
2. Sort the second linked list in descending order.  
3. Use two iterators/pointers starting at the beginning of each list:  
   - If the sum of elements pointed by the iterators equals x, increment count and move both iterators forward.  
   - If the sum is greater than x, move the iterator of the descending list forward.  
   - If the sum is less than x, move the iterator of the ascending list forward.  
4. Continue until one of the iterators reaches the end.

**🧮 Flowchart/ Diagram:**  
```plaintext
Start
  |
  V
Sort list1 ascending, list2 descending
  |
  V
Initialize iterators itr1 at head1, itr2 at head2
  |
  V
While itr1 and itr2 not at end:
  - sum = itr1.data + itr2.data
  - if sum == x: count++; move both itr1, itr2
  - else if sum > x: move itr2
  - else move itr1
  |
  V
Return count
````

```java
class Solution {
    // Function to pairwise swap elements of a linked list.
    public Node pairwiseSwap(Node head) {
        // Create a dummy node to simplify pointer manipulation
        Node dummy = new Node(-1);
        dummy.next = head;
        Node prev = dummy;

        // Loop through the list in pairs
        while (prev.next != null && prev.next.next != null) {
            // Identify nodes to swap
            Node first = prev.next;

            prev = first;
        }

        return dummy.next;
    }
}
```

1️⃣ **Traverse the list** while swapping the `next` and `prev` pointers for each node.  
2️⃣ **Update the head** to point to the last node (which becomes the new head).  

**🧮 Flowchart/ Diagram:**

```mermaid
flowchart TD
    Start([Start]) --> CheckHead{head == null or head.next == null?}
    CheckHead -- Yes --> ReturnHead[Return head]
    CheckHead -- No --> Init[Set currNode = head, prevNode = null]
    Init --> LoopStart{currNode != null?}
    
    LoopStart -- Yes --> SwapPointers[Swap currNode.prev and currNode.next]
    SwapPointers --> MoveCurr[Set currNode = currNode.prev (next node)]
    MoveCurr --> LoopStart
    
    LoopStart -- No --> UpdateHead[Set head = prevNode.prev]
    UpdateHead --> ReturnNewHead[Return head]
```

**📜 Algorithm (Pseudocode)**
```java
function reverseDLL(head):
  if head is null or head.next is null:
    return head  // No change needed for empty or single-node list

  currNode = head
  prevNode = null

  while currNode is not null:
    prevNode = currNode.prev  // Store previous node
    currNode.prev = currNode.next  // Swap next and prev pointers
    currNode.next = prevNode  

    currNode = currNode.prev  // Move to next node (which is now prev)

  head = prevNode.prev  // Update head to new first node
  return head
```

```java
class Solution {
    public DLLNode reverseDLL(DLLNode head) {
        // Your code here
        if(head==null ||  head.next==null){
            return head;
        }
        
        DLLNode currNode=head, prevNode=null;
        while(currNode!=null){
            prevNode=currNode.prev;
            currNode.prev=currNode.next;
            currNode.next=prevNode;
            
            currNode=currNode.prev;
        }
        
        head=prevNode.prev;
        
        return head;
    }
}
```
**Time Complexity:** O(n) → We iterate through the list once.
**Space Complexity:** O(1) → In-place reversal, no extra space used.

**📌 Additional Notes/Hints**
✅ Edge Cases

Single node list → Remains unchanged.
Already reversed list → No unnecessary operations are performed.
Large input size (10^5 nodes) → Algorithm efficiently handles it in O(n).

**Notes:**

* Swapping `prev` and `next` pointers reverses the traversal direction.
* The head is updated to the last node processed (`prevNode.prev`).
* Works efficiently in-place for large lists.


## Problem Statement: 135   
**Find Pairs with Given Sum in Doubly Linked List**  
You are given a **sorted doubly linked list** of **positive distinct elements** and a **target sum**.  
Find all **pairs of nodes** whose sum is equal to the given **target value**.

**Example:**  

Given doubly linked list (sorted):  
`1 <-> 2 <-> 3 <-> 4 <-> 5 <-> 6`  
Target sum: `7`  

**Output:**  
Pairs with sum 7 are:  
- (1, 6)  
- (2, 5)  
- (3, 4)  

Explanation:  
- 1 + 6 = 7  
- 2 + 5 = 7  
- 3 + 4 = 7

**Explanation:**  
Pairs `(1,6)` and `(2,5)` add up to `7`.

**🔹 Constraints:**  
- `1 ≤ n ≤ 10^5` (size of the doubly linked list)  
- `1 ≤ Node.data ≤ 10^6`  
- Elements in the list are **distinct** and **sorted**.

**🚀 Efficient Two-Pointer Approach (O(n) Time, O(1) Space)**  
✅ **Logic:**  
1️⃣ Use **two pointers**:  
   - `first` → Points to the **head**.  
   - `second` → Moves to the **last node**.  

2️⃣ While `first < second`:  
   - If `first.data + second.data == target` → Store the pair & move both pointers.  
   - If `sum < target` → Move `first` forward.  
   - If `sum > target` → Move `second` backward.  

**🧮 Flowchart/ Diagram:**

```mermaid
flowchart TD
    Start([Start]) --> InitPointers[Initialize first = head, second = last node]
    InitPointers --> LoopCheck{first != second AND second.next != first?}
    
    LoopCheck -- No --> End[Return result list]
    LoopCheck -- Yes --> CalcSum[sum = first.data + second.data]
    CalcSum --> CheckSum{sum == target?}
    
    CheckSum -- Yes --> AddPair[Add (first.data, second.data) to result]
    AddPair --> MovePointers[Move first = first.next, second = second.prev]
    MovePointers --> LoopCheck
    
    CheckSum -- No --> CheckLess{sum < target?}
    
    CheckLess -- Yes --> MoveFirst[Move first = first.next]
    MoveFirst --> LoopCheck
    
    CheckLess -- No --> MoveSecond[Move second = second.prev]
    MoveSecond --> LoopCheck
```

**📜 Algorithm (Pseudocode)**
```java
function findPairsWithGivenSum(target, head):
    result = []
    
    // Step 1: Set up two pointers
    first = head
    second = head
    while second.next is not NULL:
        second = second.next // Move to the last node
    
    // Step 2: Find pairs using two-pointer technique
    while (first != second && second.next != first):
        sum = first.data + second.data
        
        if (sum == target):
            result.add([first.data, second.data])
            first = first.next
            second = second.prev
        else if (sum < target):
            first = first.next
        else:
            second = second.prev
    
    return result
```
