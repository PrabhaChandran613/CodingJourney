### Problem Statement No.: 1

**Problem Description:** Check whether a given number is even or odd.  

**Input:** An integer number.  
**Output:** A string indicating either "even" or "odd".  

**Examples:**

**Input:** n = 15

**Output:** false

**Explanation:** The number is not divisible by 2

**Constraints:**  
1. The input number can be positive, negative, or zero.  
2. The input number will always be an integer.  

**Goal:** Determine if the input number is divisible by 2 (even) or not (odd).  

**Approach:**  
- Use the modulus operator `%` to check the remainder when the number is divided by 2.  
- If the remainder is `0`, the number is **even**; otherwise, it is **odd**. 

**🧮 Flowchart: Check if Number is Even or Odd**

```text
     +---------------------+
     |   Input: num        |
     +---------------------+
                |
                v
     +---------------------+
     | Is num % 2 == 0 ?   |
     +---------------------+
         |          |
       Yes         No
        |           |
        v           v
+---------------+ +---------------+
|  Output: even | |  Output: odd  |
+---------------+ +---------------+
```

**Pseudocode/Algorithm:**  
1. Read the integer `num`.  
2. If `num % 2 == 0`, output "even".  
3. Otherwise, output "odd".  

**Code:**
```java
public class Solution {

    public static boolean isEven(int n)  {
        return (n % 2 == 0);
    }

    public static void main(String[] args) {
        int n = 101;
        if (isEven(n) == true)
            System.out.print("true");
        else
            System.out.print("false");
    }
}
```

**Time Complexity:** \(O(1)\)  
**Space Complexity:** \(O(1)\)  

**Notes:**  
- Negative numbers follow the same rule as positives (e.g., -2 is even, -5 is odd).  
- The modulus operation is efficient even for large numbers.


### Problem Statement No.: 2
**Problem Description:** Find the first and last digits of a given number.  
**Input:** An integer number.  
**Output:** Two integers representing the first digit and the last digit of the number.  
**Constraints:**  
1. The input number can be positive, negative, or zero.  
2. The input number will always be an integer.  
**Goal:** Extract the first and last digits of the input number.  
**Example**

**Input:** n = 12345  
**Output:** First digit: 1, Last digit: 5  

**🧮 Flowchart/ Diagram:**

```text
     +------------------------+
     |      Input: n          |
     +------------------------+
                |
                v
     +------------------------+
     | n = abs(n)             |
     +------------------------+
                |
                v
     +------------------------+
     | last_digit = n % 10    |
     +------------------------+
                |
                v
     +------------------------+
     | first_digit = n        |
     | while first_digit >=10 |
     |     first_digit /= 10  |
     +------------------------+
                |
                v
     +------------------------------+
     | Output: first_digit, last_digit |
     +------------------------------+
```
**Approach:**  
- **Last Digit:** Use the modulo operator `%` with 10 (`n % 10`) to extract the last digit.  
- **First Digit:** Repeatedly divide the number by 10 until it is less than 10, which gives the first digit.  
**Pseudocode/Algorithm:**  
1. Read the integer `n`.  
2. Calculate `last_digit = abs(n) % 10` to handle negatives.  
3. Set `first_digit = abs(n)` and repeatedly divide by 10 until `first_digit < 10`.  
4. Output `first_digit` and `last_digit`.  

**Code:**
```java
public static void findFirstAndLastDigits(int number) {
    int lastDigit = Math.abs(number) % 10; // Find the last digit
    int firstDigit = Math.abs(number); // Use absolute value to handle negatives
    // Divide the number repeatedly until only one digit is left
    while (firstDigit >= 10) {
        firstDigit /= 10;
    }
    // Output the first and last digits
    System.out.println("First digit: " + firstDigit);
    System.out.println("Last digit: " + lastDigit);
}
int number = 12345; 
findFirstAndLastDigits(number);
```
**Time Complexity:** \(O(log{10}(n))\) due to repeated division for the first digit.  
**Space Complexity:** \(O(1)\) since no additional storage is required.  
**Notes:**  
- Ensure negative numbers are handled by taking their absolute value.  
- For single-digit numbers, the first and last digits are the same.

### Problem Statement No.: 3
**Find Last Digit of a^b for Large Numbers**

You are given two integer numbers as **strings**, `a` and `b`. Your task is to compute the **last digit** of `a^b` (i.e., `a` raised to the power `b`), even when the numbers are large.


**Constraints**

* 1 ≤ |a|, |b| ≤ 1000
* |a|: Length of string `a`
* |b|: Length of string `b`
* No test cases will involve undefined results (e.g., 0^0)

**Examples**
```
Input: a = "3", b = "10"
Output: 9
Explanation: 3^10 = 59049 → Last digit is 9
```

**Input:**  
Two strings `a` and `b`, representing integers.

**Output:**  
An integer representing the last digit of `a^b`.

**Goal:**  
Efficiently compute the last digit of `a^b` without calculating the entire power for large values of `a` and `b`.

**🧮 Flowchart/ Diagram:**

```text
     +--------------------------+
     |     Input: a, b (as str) |
     +--------------------------+
                |
                v
     +--------------------------+
     |  If b == "0" → return 1  |
     +--------------------------+
                |
                v
     +--------------------------+
     | Get lastDigit = a[-1]    |
     +--------------------------+
                |
                v
     +--------------------------+
     | Use predefined cycles[]  |
     | for lastDigit            |
     +--------------------------+
                |
                v
     +--------------------------+
     | Compute b mod cycle.length |
     +--------------------------+
                |
                v
     +------------------------------+
     | Return cycle[(bMod - 1)]     |
     +------------------------------+
```
**Pseudocode/Algorithm:**

1. If `b == "0"`, return 1.
2. Set `lastDigit = last character of a`.
3. Retrieve the digit cycle for `lastDigit`.
4. Compute `b mod cycle.length` using string arithmetic.
5. If result is 0, use the last element of the cycle.
6. Return the appropriate element from the digit cycle.

**Your Task**
Implement the method `getLastDigit(String a, String b)` that:

- Returns the last digit of `a^b`
- Handles `a` and `b` as strings with up to 1000 digits

**Expected Complexity**

* **Time Complexity**: O(|b|)
* **Auxiliary Space**: O(1)
**Space Complexity:** $O(1)$ — Uses fixed-size structures and constant space for computation.


**Approach**

**Key Concepts:**

* The last digit of `a^b` cycles in a predictable pattern for each digit from 0–9.
* We only need to know the **last digit** of the base `a`, and compute `b mod cycleLength` to find the result.

**Example:**

For base digit `2`:

* Pattern: \[2, 4, 8, 6]
* Repeat every 4 steps: 2^1=2, 2^2=4, 2^3=8, 2^4=16, 2^5=32...

**Java Implementation**

```java
// User function Template for Java
class Solution {
    static int getLastDigit(String a, String b) {
        if (b.equals("0")) return 1;  // any number to power 0 is 1

        int lastDigit = a.charAt(a.length() - 1) - '0';

        // Predefined cycles for digits 0-9
        int[][] cycles = {
            {0},             // 0^1 = 0
            {1},             // 1^n = 1
            {2, 4, 8, 6},    // cycle length = 4
            {3, 9, 7, 1},
            {4, 6},
            {5},             // 5^n = 5
            {6},             // 6^n = 6
            {7, 9, 3, 1},
            {8, 4, 2, 6},
            {9, 1}
        };

        int[] cycle = cycles[lastDigit];

        // Calculate b mod cycle length
        int bMod = 0;
        for (int i = 0; i < b.length(); i++) {
            bMod = (bMod * 10 + (b.charAt(i) - '0')) % cycle.length;
        }

        if (bMod == 0) bMod = cycle.length;

        return cycle[bMod - 1];
    }
}
```

**Notes**

* Handles inputs of size up to 1000 digits.
* Avoids full computation of `a^b` which would be infeasible for large `b`.
* Efficient modulo calculation using string parsing.

### Problem Statement No.: 4
**Problem Description:** Count the number of digits in a given integer.  
**Input:** An integer `n`.  
**Output:** An integer representing the total number of digits in `n`.  
**Constraints:**  
1. The input number can be positive, negative, or zero.  
2. The input number will always be an integer.  

**Goal:** Calculate the number of digits in the given number.  

**Approach:**  
- Use a loop to repeatedly divide the number by 10, which removes the rightmost digit.  
- Increment a counter each time to keep track of the number of divisions.  
- Handle the base case where the input is `0` (it has 1 digit).  

**Example**

**Input:** n = 58964  
**Output:** 5  

**🧮 Flowchart/ Diagram:**

```text
     +---------------------+
     |   Input: n          |
     +---------------------+
                |
                v
     +---------------------+
     | Is n == 0 ?         |
     +---------------------+
         |          |
       Yes         No
        |           |
        v           v
+----------------+  +-------------------------+
|  Output: 1     |  | Initialize count = 0    |
+----------------+  +-------------------------+
                              |
                              v
              +------------------------------+
              | While n != 0:                |
              |   n = n / 10                 |
              |   count = count + 1          |
              +------------------------------+
                              |
                              v
               +-----------------------------+
               |       Output: count         |
               +-----------------------------+
```
**Pseudocode/Algorithm:**  
1. If `n == 0`, return `1` (base case).  
2. Initialize `count = 0`.  
3. While `n != 0`:  
   - Divide `n` by 10.  
   - Increment `count`.  
4. Return `count`.  

 **Code:**
 ```java
 import java.io.*;

class GfG {

    // function to count digits
    static int countDigit(int n) {
        // Base case
        if (n == 0)
            return 1;

        int count = 0;
      
        // Iterate till n has digits remaining
        while (n != 0) {
          
            // Remove rightmost digit
            n = n / 10;
          
            // Increment digit count by 1
            ++count;
        }
        return count;
    }

    public static void main(String[] args) {
        int n = 58964;
        System.out.println( countDigit(n));
    }
}
```
**Time Complexity:** \(O(\log_{10}(n))\) due to repeated division by 10.  
**Space Complexity:** \(O(1)\) since no additional storage is required.  
**Notes:**  
- Negative numbers are handled naturally since the sign doesn't affect the digit count.  
- For `0`, ensure the result is `1`. 

### Problem Statement No.: 5
**Problem Description:** Count the number of digits in an integer \( n \) that divide \( n \) evenly.  
**Input:** A positive integer `n`.  
**Output:** An integer representing the count of digits in `n` that evenly divide `n`.  
**Constraints:** \( n > 0 \), ignore `0` since division by zero is undefined.  
**Goal:** Identify and count the digits of \( n \) that divide \( n \) without a remainder.  
**Example**

**Input:** n = 122  
**Output:** 2  
**Explanation:** Digits 1 and 2 divide 122 evenly; the second '2' is counted again.

**🧮 Flowchart/ Diagram:**

```text
     +----------------------------+
     |        Input: n           |
     +----------------------------+
                  |
                  v
     +----------------------------+
     |   Initialize count = 0     |
     |   Set temp = n             |
     +----------------------------+
                  |
                  v
     +----------------------------+
     | While temp > 0             |
     |   digit = temp % 10        |
     |   if digit != 0 and        |
     |      n % digit == 0        |
     |       → count++            |
     |   temp = temp / 10         |
     +----------------------------+
                  |
                  v
     +----------------------------+
     |      Return count          |
     +----------------------------+
```
**Approach:** Extract digits individually, check divisibility, ignore `0`, maintain a counter.  
**Pseudocode/Algorithm:**  
1. Initialize `count = 0`, store original `n` in temp.  
2. While `temp > 0`: extract last digit (`digit = temp % 10`), check divisibility (`n % digit == 0`), increment `count`, remove last digit (`temp = temp / 10`).  
3. Return `count`.  
**Code**:
```java
// User function Template for Java

class Solution {
    static int evenlyDivides(int n) {
        // code here
     int count=0, temp=n;
     while(temp>0){
         int digit=temp%10;
         if(digit!=0 && n%digit==0){
             count++;
         }
         temp/=10;
     }
     return count;
          
        }
    }
```
**Time Complexity:** \( O(d) \), where \( d \) is the number of digits in \( n \).  
**Space Complexity:** \( O(1) \), only integer variables used.  
**Notes:** Efficient for large values, ignores zero to prevent errors.  


### Problem Statement No.: 6
**Problem Description:** Reverse the given array so that the first element becomes the last, the second element becomes the second last, and so on.  
**Input:** An integer array `arr[]`.  
**Output:** The same array with elements in reversed order.  
**Constraints:**  
1. The array can contain positive or negative integers.  
2. The size of the array is \(1 \leq n \leq 10^5\). 

**Example**

**Input:** arr = [1, 4, 3, 2, 6, 5]  
**Output:** [5, 6, 2, 3, 4, 1]  

**🧮 Flowchart/ Diagram:**

```text
     +-----------------------------+
     |      Input: arr[]          |
     +-----------------------------+
                 |
                 v
     +-----------------------------+
     | Initialize: left = 0        |
     |              right = n - 1 |
     +-----------------------------+
                 |
                 v
     +----------------------------------+
     | While left < right:              |
     |   Swap arr[left] and arr[right]  |
     |   left++, right--                |
     +----------------------------------+
                 |
                 v
     +-----------------------------+
     |       Output arr[]         |
     +-----------------------------+
```
**Goal:** Rearrange the elements of the array in reverse order.  
**Approach:**  
- Create a temporary array to store the elements of the original array in reverse order.  
- Use a loop to copy the reversed elements back to the original array.  
**Pseudocode/Algorithm:**  
1. Read the array `arr[]` of size `n`.  
2. Create a temporary array `temp[]` of size `n`.  
3. For `i` from `0` to `n-1`, assign `temp[i] = arr[n - i - 1]`.  
4. Copy the elements from `temp[]` back to `arr[]`.  
5. Output the reversed `arr[]`. 
 
**Code:**
```java
// Java Program to reverse an array using Two Pointers

import java.util.Arrays;

class GfG {
    
    // function to reverse an array
    static void reverseArray(int[] arr) {
        
        // Initialize left to the beginning and right to the end
        int left = 0, right = arr.length - 1;

        // Iterate till left is less than right
        while (left < right) {
            
            // Swap the elements at left and right position
            int temp = arr[left];
            arr[left] = arr[right];
            arr[right] = temp;

            // Increment the left pointer
            left++;

            // Decrement the right pointer
            right--;
        }
    }

    public static void main(String[] args) {
        int[] arr = { 1, 4, 3, 2, 6, 5 };

        reverseArray(arr);

        for (int i = 0; i < arr.length; i++) 
            System.out.print(arr[i] + " ");
    }
}
```
**Time Complexity:** \(O(n)\) for both copying and reversing the elements.  
**Space Complexity:** \(O(n)\) due to the temporary array.  
**Notes:**  
- This approach uses extra space. An in-place reversal (swapping elements) can achieve \(O(1)\) space.

**Goal:** Calculate the result of \(x^n\) efficiently using iteration.  
**Approach:**  
- Initialize the result to 1.  
- Multiply `x` with the result `n` times.  
**Pseudocode/Algorithm:**  
1. Read integers `x` and `n`.  
2. Initialize `pow = 1`.  
3. For `i` from `0` to `n-1`:  
   - Multiply `pow` by `x`.

```java
class Gfg {
    // Naive iterative solution to calculate pow(x, n)
    public static long power(int x, int n){       
         // Initialize result by 1
        long pow = 1L;
        // Multiply x for n times
        for (int i = 0; i < n; i++) {
            pow = pow * x;
        }
        return pow;
    }}
int x = 2;
int n = 3;
System.out.println(Gfg.power(x, n));
``` 
**Time Complexity:** \(O(n)\), since the loop runs \(n\) times.  
**Space Complexity:** \(O(1)\), as no additional space is required.  
**Notes:**  
- This approach is simple but inefficient for large values of `n`. An optimized approach is exponentiation by squaring.  


### Problem Statement No.: 9
**Problem Description:** Given a number `n`, find the value of `n` raised to the power of its own reverse.  
**Input:** A positive integer `n`.  
**Output:** The result of `n` raised to the power of its own reverse.  
**Constraints:**  
1. The result will always fit into a 32-bit signed integer.  
2. \(n > 0\).  

**Example**

**Input:** n = 12  
**Output:** 8916100448256  
**Explanation:** Reverse of 12 is 21, so 12^21 = 8916100448256

**🧮 Flowchart/ Diagram:**

```text
     +----------------------------+
     |        Input: n           |
     +----------------------------+
                 |
                 v
     +----------------------------+
     |   Reverse digits of n      |
     +----------------------------+
                 |
                 v
     +----------------------------+
     |   result = 1               |
     |   Repeat reversed times:   |
     |   result *= original n     |
     +----------------------------+
                 |
                 v
     +----------------------------+
     |     Return result          |
     +----------------------------+
```

### Problem Statement No.: 10
**Problem Description:** Find the GCD (Greatest Common Divisor) or HCF (Highest Common Factor) of two integers `a` and `b`.  
**Input:** Two integers `a` and `b`.  
**Output:** An integer representing the GCD of `a` and `b`.  
**Constraints:**  
1. \(a > 0\) and \(b > 0\).  
2. Both numbers are integers.  

**Example**

**Input:** a = 98, b = 56  
**Output:** 14  
**Explanation:** The GCD of 98 and 56 is 14 because 14 is the largest number that divides both 98 and 56.

**🧮 Flowchart/ Diagram:**

```text
     +----------------------------+
     |        Input: a, b         |
     +----------------------------+
                 |
                 v
     +----------------------------+
     |   Initialize result = min(a, b) |
     +----------------------------+
                 |
                 v
     +----------------------------+
     |   While result > 0:        |
     |   Check divisibility of a  |
     |   and b by result          |
     +----------------------------+
                 |
                 v
     +----------------------------+
     |     Return result as GCD   |
     +----------------------------+
```
**Goal:** Find the largest number that divides both `a` and `b`.  
**Approach:**  
- Use a brute-force method to find the smallest of `a` and `b`.  
- Decrement this value until a common divisor for both numbers is found.  
**Pseudocode/Algorithm:**  
1. Read integers `a` and `b`.  
2. Initialize `result = min(a, b)`.  
3. While `result > 0`:  
   - If `a % result == 0` and `b % result == 0`, break the loop.  
   - Otherwise, decrement `result`.  
4. Return `result` as the GCD.  

**Code:**
```java
public class GFG {

    // Function to return gcd of a and b
    static int gcd(int a, int b)
    {
        // Find Minimum of a and b
        int result = Math.min(a, b);
        while (result > 0) {
            if (a % result == 0 && b % result == 0) {
                break;
            }
            result--;
        }

        // Return gcd of a and b
        return result;
    }}

int a = 98, b = 56;
System.out.print("GCD of " + a + " and " + b
                 + " is " + GFG.gcd(a, b));
```
**Time Complexity:** \(O(\min(a, b))\), where \(a\) and \(b\) are the two numbers.  
**Space Complexity:** \(O(1)\), as no additional storage is required.  
**Notes:**  
- This is a naive approach. A more efficient solution uses the **Euclidean Algorithm**, which reduces the complexity to \(O(\log(\min(a, b)))\).

- It works because the GCD of two numbers also divides their difference.
- Avoid using recursion if you're working with very large inputs to prevent stack overflow in some languages.

```java
// User function Template for Java
class Solution {
    static boolean isPerfectNumber(int n) {
        // Initial sum is set to 1, since 1 is a divisor of all numbers greater than 1
        int sum = 1;

        // Loop from 2 to sqrt(n) to find divisors
        for (int i = 2; i * i <= n; i++) {
            if (n % i == 0) {
                sum += i;  // Add divisor i
                if (i != n / i) {
                    sum += n / i;  // Add the corresponding pair divisor n / i
                }
            }
        }

        // Return true if the sum of divisors is equal to the number itself, false otherwise
        return sum == n;
    }
}
```
**✅ Time Complexity Clarification**

* **Divisor Calculation:** We iterate through values from 2 to $\sqrt{n}$ to find divisors. This results in a time complexity of $O(\sqrt{n})$.
* **Space Complexity:** The space complexity is $O(1)$ because only a constant amount of extra space is used (the variables `sum` and `i`).

**✅ Additional Notes**

* **Perfect Numbers:** A perfect number is a positive integer that is equal to the sum of its proper divisors (excluding itself). The first few perfect numbers are 6, 28, 496, and 8128.
* **Optimization:** The algorithm efficiently checks divisors up to $\sqrt{n}$, which reduces unnecessary checks. By adding both $i$ and $n / i$ when $i$ is a divisor, we minimize the number of iterations needed.
* **For Large Numbers:** For extremely large values of $n$, there are more efficient algorithms for checking perfect numbers (e.g., leveraging the Euclidean formula for even perfect numbers). However, this approach is sufficient for most use cases within the given constraints.
