**Constraints:**
- The array `arr[]` is sorted in ascending order.
- The array may contain zero or more elements.
- The integer `k` can be any valid integer value.

**Goal of the Code:**
Efficiently determine the presence of the integer `k` in the sorted array `arr[]`.

**Approach:**
Utilize the binary search algorithm to leverage the sorted nature of the array, achieving efficient search performance.

**Pseudocode/Algorithm:**
1. Initialize two pointers: `left` at the start of the array and `right` at the end.
2. While `left` is less than or equal to `right`:
   - Calculate the middle index: `mid = left + (right - left) / 2`.
   - If `arr[mid]` equals `k`, return `true`.
   - If `arr[mid]` is less than `k`, move the `left` pointer to `mid + 1`.
   - If `arr[mid]` is greater than `k`, move the `right` pointer to `mid - 1`.
3. If the loop ends without finding `k`, return `false`.

**Time and Space Complexity:**
- **Time Complexity:** O(log n), where `n` is the number of elements in the array. This is due to the binary search algorithm, which halves the search space with each iteration.
- **Space Complexity:** O(1), as the algorithm uses a constant amount of extra space.

**Additional Notes/Hints:**
- Binary search is optimal for searching in sorted arrays due to its logarithmic time complexity.
- Ensure that edge cases, such as empty arrays, are handled appropriately.

**Java Implementation:**
```java
class Solution {
    static boolean searchInSorted(int arr[], int k) {
        int left = 0;
        int right = arr.length - 1;
        
        while (left <= right) {
            int mid = left + (right - left) / 2;
            
            if (arr[mid] == k) {
                return true;
            } else if (arr[mid] < k) {
                left = mid + 1;
            } else {
                right = mid - 1;
            }
        }
        
        return false;
    }
}


### Problem Statement: 73
Given a sorted array `arr[]` with unique elements and an integer `k`, find the index (0-based) of the largest element in `arr[]` that is less than or equal to `k`. This element is called the "floor" of `k`. If such an element does not exist, return `-1`.

**Input:**
- A sorted array `arr[]` of unique integers.
- An integer `k`.

**Output:**
- An integer representing the index of the floor of `k` in `arr[]`. If no such element exists, return `-1`.

**Constraints:**
- The array `arr[]` is sorted in ascending order.
- All elements in `arr[]` are unique.
- The array may contain zero or more elements.
- The integer `k` can be any valid integer value.

**Goal of the Code:**
Efficiently determine the index of the largest element in the sorted array `arr[]` that is less than or equal to the integer `k`.

**Approach:**
Utilize the binary search algorithm to efficiently find the floor of `k` in the sorted array `arr[]`.

**Pseudocode/Algorithm:**
1. Initialize two pointers: `lo` at the start of the array and `hi` at the end.
2. Initialize a variable `floorIndex` to `-1` to keep track of the potential floor index.
3. While `lo` is less than or equal to `hi`:
   - Calculate the middle index: `mid = lo + (hi - lo) / 2`.
   - If `arr[mid]` equals `k`, return `mid` as the floor index.
   - If `arr[mid]` is less than `k`:
     - Update `floorIndex` to `mid`.
     - Move the `lo` pointer to `mid + 1` to search in the right subarray.
   - If `arr[mid]` is greater than `k`, move the `hi` pointer to `mid - 1` to search in the left subarray.
4. After the loop, return `floorIndex`. If no floor exists, `floorIndex` will remain `-1`.

**Time and Space Complexity:**
- **Time Complexity:** O(log n), where `n` is the number of elements in the array. This efficiency is achieved through the binary search algorithm.
- **Space Complexity:** O(1), as the algorithm uses a constant amount of extra space.

**Additional Notes/Hints:**
- Binary search is optimal for searching in sorted arrays due to its logarithmic time complexity.
- Ensure that edge cases, such as empty arrays or `k` being smaller than the smallest element in the array, are handled appropriately.

**Java Implementation:**
```java
class Solution {
    static int findFloor(int[] arr, int k) {
        int lo = 0, hi = arr.length - 1;
        int floorIndex = -1;

        while (lo <= hi) {
            int mid = lo + (hi - lo) / 2;

            if (arr[mid] == k) {
                return mid;
            } else if (arr[mid] < k) {
                floorIndex = mid;
                lo = mid + 1;
            } else {
                hi = mid - 1;
            }
        }

        return floorIndex;
    }
}


### Problem Statement: 74
Given a sorted array `arr[]` with possibly some duplicates, find the first and last occurrences of an element `x` in the array. If `x` is not found, return both indices as `-1`.

**Input:**
- A sorted array `arr[]` that may contain duplicates.
- An integer `x` whose first and last occurrences need to be found.

**Output:**
- An array with two integers:
  - The first integer is the index of the first occurrence of `x`.
  - The second integer is the index of the last occurrence of `x`.

  If `x` is not present in the array, return both indices as `-1`.

**Constraints:**
- The array `arr[]` is sorted in ascending order and may contain duplicates.
- The array can have `0 ≤ n ≤ 10^5` elements.
- The value `x` can be any integer.

**Goal of the Code:**
Efficiently find the first and last occurrences of `x` in the sorted array `arr[]`.

**Approach:**
1. Utilize binary search to efficiently find both the first and last occurrences of `x` in the sorted array.
2. Perform the first binary search to locate the first occurrence by adjusting the search boundaries if `x` is found.
3. Perform a second binary search to locate the last occurrence similarly.

**Pseudocode/Algorithm:**
1. **First Occurrence Search:**
   - Perform binary search to locate the first occurrence of `x`.
   - If `arr[mid] == x`, update the index and search the left half of the array by setting `high = mid - 1`.
   - Continue adjusting the boundaries until the first occurrence is found or the boundaries converge.

2. **Last Occurrence Search:**
   - Perform binary search to locate the last occurrence of `x`.
   - If `arr[mid] == x`, update the index and search the right half of the array by setting `low = mid + 1`.
   - Continue adjusting the boundaries until the last occurrence is found or the boundaries converge.

3. Return the result with both first and last indices.

**Time and Space Complexity:**
- **Time Complexity:** O(log n) for both the first and last occurrence searches, where `n` is the size of the array.
- **Space Complexity:** O(1), as only a constant amount of extra space is used for variables.

**Additional Notes/Hints:**
- The problem leverages the fact that the array is sorted, making binary search a suitable solution.
- Special edge cases like empty arrays or arrays where `x` is not present need to be handled by returning `-1` for both indices.

**Java Implementation:**
```java
class GFG {
    ArrayList<Integer> find(int arr[], int x) {
        // code here
        ArrayList<Integer> result = new ArrayList<>();
        int first = firstOccurence(arr, x);
        int last = lastOccurence(arr, x);
        result.add(first);
        result.add(last);
        return result;
    }

    private int firstOccurence(int arr[], int x) {
        int low = 0, high = arr.length - 1, firstOccurrence = -1;
        while (low <= high) {
            int mid = low + (high - low) / 2;
            if (arr[mid] == x) {
                firstOccurrence = mid;
                high = mid - 1;
            } else if (arr[mid] < x) {
                low = mid + 1;
            } else {
                high = mid - 1;
            }
        }
        return firstOccurrence;
    }

    private int lastOccurence(int arr[], int x) {
        int low = 0, high = arr.length - 1, lastOccurrence = -1;
        while (low <= high) {
            int mid = low + (high - low) / 2;
            if (arr[mid] == x) {
                lastOccurrence = mid;
                low = mid + 1;
            } else if (arr[mid] < x) {
                low = mid + 1;
            } else {
                high = mid - 1;
            }
        }
        return lastOccurrence;
    }
}


### Problem Statement: 75
Given an array `nums` containing `n` distinct numbers in the range `[0, n]`, return the only number in the range that is missing from the array.

**Input:**
- An integer array `nums` of length `n` containing distinct numbers from the range `[0, n]`.

**Output:**
- The single integer from the range `[0, n]` that is missing in `nums`.

**Constraints:**
- The array `nums` contains `n` unique numbers.
- The length of `nums` is `1 <= n <= 10^4`.
- All elements in `nums` are within the range `[0, n]`.

**Goal of the Code:**
Efficiently identify and return the missing number from the array `nums`.

**Approach:**
1. **Mathematical Summation:**
   - Calculate the expected sum of the first `n` natural numbers using the formula: `expectedSum = n * (n + 1) / 2`.
   - Compute the actual sum of the elements present in `nums`.
   - The difference between the `expectedSum` and the `actualSum` will yield the missing number.

**Pseudocode/Algorithm:**
1. Initialize `sum` to `0` and determine `n` as the length of `nums`.
2. Iterate through each number in `nums` and accumulate the sum.
3. Compute the `expectedSum` using the formula for the sum of the first `n` natural numbers.
4. The missing number is `expectedSum - sum`.
5. Return the missing number.

**Time and Space Complexity:**
- **Time Complexity:** O(n), where `n` is the length of the array `nums`. This is due to the single pass required to compute the sum of elements.
- **Space Complexity:** O(1), as the solution utilizes a constant amount of extra space.

**Java Implementation:**
```java
class Solution {
    public int missingNumber(int[] nums) {
        int sum = 0, n = nums.length;
        for (int num : nums) {
            sum += num;
        }
        int expectedSum = (n * (n + 1)) / 2;
        return expectedSum - sum;
    }
}


### Problem Statement: 76
**Square Root:**
Given a positive integer `n`, determine its square root. If `n` is not a perfect square, return the floor value of its square root. The floor value of a number is the greatest integer less than or equal to that number.

**Input:**
- A single positive integer `n`.

**Output:**
- An integer representing the floor value of the square root of `n`.

**Constraints:**
- \( 1 \leq n \leq 10^9 \)

**Goal of the Code:**
Efficiently compute and return the floor value of the square root of the given integer `n`.

**Approach:**
1. **Binary Search Method:**
   - Utilize binary search to find the floor value of the square root:
     - Initialize `low` to 1 and `high` to `n`.
     - While `low` is less than or equal to `high`:
       - Compute `mid` as the average of `low` and `high`.
       - If `mid * mid` equals `n`, `mid` is the exact square root; return `mid`.
       - If `mid * mid` is less than `n`, move the `low` pointer to `mid + 1` and update the result to `mid`.
       - If `mid * mid` is greater than `n`, move the `high` pointer to `mid - 1`.
     - Return the result after the loop concludes.

**Pseudocode/Algorithm:**
1. **Function:** `floorSqrt(n)`
2. **Input:** Integer `n`
3. **Output:** Integer representing the floor value of the square root of `n`
4. **Steps:**
   - **Edge Case:** If `n` is less than 2, return `n`.
   - Initialize `low` to 1 and `high` to `n`.
   - Initialize `result` to 1.
   - **While** `low` ≤ `high`:
     - Calculate `mid` as `low + (high - low) // 2`.
     - **If** `mid * mid` equals `n`:
       - Return `mid`.
     - **Else If** `mid * mid` is less than `n`:
       - Set `result` to `mid`.
       - Set `low` to `mid + 1`.
     - **Else:**
       - Set `high` to `mid - 1`.
   - Return `result`.

**Time and Space Complexity:**
- **Time Complexity:** \( O(\log n) \) — The binary search approach reduces the search space by half in each iteration, leading to a logarithmic time complexity.
- **Space Complexity:** \( O(1) \) — The algorithm uses a constant amount of extra space.

**Java Implementation:**
```java
class Solution {
    int floorSqrt(int n) {
        if (n < 2) {
            return n;
        }
        int low = 1, high = n, result = 1;
        while (low <= high) {
            int mid = low + (high - low) / 2;
            if (mid * mid == n) {
                return mid;
            } else if (mid * mid < n) {
                result = mid;
                low = mid + 1;
            } else {
                high = mid - 1;
            }
        }
        return result;
    }
}


### Problem Statement: 77
Given a sorted and rotated array `arr` that may contain duplicates, and a target key `k`, determine whether the key is present in the array.

**Input:**
- An integer `k` representing the target key.
- An array `arr[]` of integers which is sorted, rotated at some pivot, and may contain duplicates.

**Output:**
- Return `true` if the key `k` is present in the array; otherwise, return `false`.

**Constraints:**
- The array may contain duplicate elements.
- The array is sorted and then rotated at an unknown pivot.
- The array can be of any size within reasonable limits.

**Goal of the code:**
To efficiently determine if the target key `k` exists in the rotated sorted array `arr`.

**Approach:**
1. **Binary Search with Adjustments for Duplicates:**
   - Initialize two pointers, `low` and `high`, to the start and end of the array, respectively.
   - While `low` is less than or equal to `high`:
     - Calculate the middle index `mid`.
     - If `arr[mid]` equals `k`, return `true`.
     - If `arr[mid]`, `arr[low]`, and `arr[high]` are equal, increment `low` and decrement `high` to skip duplicates.
     - Determine which part of the array is sorted:
       - If the left part is sorted:
         - Check if `k` lies within this range; adjust `low` and `high` accordingly.
       - Otherwise, the right part must be sorted:
         - Check if `k` lies within this range; adjust `low` and `high` accordingly.
   - If the loop ends without finding `k`, return `false`.

```java
boolean searchInRotatedArray(int[] arr, int k) {
    int low = 0, high = arr.length - 1;
    while (low <= high) {
        int mid = low + (high - low) / 2;
        if (arr[mid] == k) {
            return true;
        }
        if (arr[mid] == arr[low] && arr[mid] == arr[high]) {
            ++low;
            --high;
            continue;
        }
        if (arr[low] <= arr[mid]) {
            if (k >= arr[low] && k < arr[mid]) {
                high = mid - 1;
            } else {
                low = mid + 1;
            }
        } else {
            if (k > arr[mid] && k <= arr[high]) {
                low = mid + 1;
            } else {
                high = mid - 1;
            }
        }
    }
    return false;
}


## Problem Statement:78
**Search Element with distinct in sorted Rotated Array:**
Given a sorted and rotated array `arr[]` of distinct elements, find the index of a target key. Return -1 if the key is not found.

**Input**:  
- A rotated sorted array `arr[]` of distinct integers.
- An integer `key` representing the target value to search for.

**Output**:  
- An integer representing the index of the target `key` in the array. If the `key` is not found, return -1.

**Constraints**:  
- The array contains distinct integers.
- The array was originally sorted in ascending order.
- The array has been rotated at an unknown pivot.
- The search algorithm should have a time complexity of O(log n).

**Goal of the code**:  
Efficiently locate the index of the target `key` in a rotated sorted array using a modified binary search approach.

**Approach**:  
1. **Initialize Pointers**:
   - Set `low` to 0 and `high` to the last index of the array.
2. **Binary Search with Rotation Consideration**:
   - While `low` is less than or equal to `high`:
     - Calculate `mid` as the average of `low` and `high`.
     - If `arr[mid]` equals the `key`, return `mid`.
     - Determine which part of the array is sorted:
       - If `arr[low]` is less than or equal to `arr[mid]`, the left part is sorted:
         - If `key` lies between `arr[low]` and `arr[mid]`, adjust `high` to `mid - 1`.
         - Otherwise, adjust `low` to `mid + 1`.
       - Else, the right part is sorted:
         - If `key` lies between `arr[mid]` and `arr[high]`, adjust `low` to `mid + 1`.
         - Otherwise, adjust `high` to `mid - 1`.
3. **Return Result**:
   - If the `key` is not found, return -1.


**Time and Space Complexity**:
- **Time Complexity**: O(log n), where n is the number of elements in the array. The algorithm performs a binary search, effectively halving the search space with each iteration.
- **Space Complexity**: O(1), as the algorithm uses a constant amount of extra space.

**Additional Notes**:
- This approach leverages the properties of the rotated sorted array to maintain the efficiency of binary search.
- By identifying which part of the array is sorted, the algorithm decides the direction to continue the search, ensuring that the target `key` is found in logarithmic time.


**Pseudocode/Algorithm**:
function searchInRotatedArray(arr, key): low = 0 high = len(arr) - 1
```java
while low <= high:
    mid = (low + high) // 2

    if arr[mid] == key:
        return mid

    if arr[low] <= arr[mid]:  // Left part is sorted
        if arr[low] <= key < arr[mid]:
            high = mid - 1
        else:
            low = mid + 1
    else:  // Right part is sorted
        if arr[mid] < key <= arr[high]:
            low = mid + 1
        else:
            high = mid - 1

return -1

## Problem Statement: 79
**Minimum Number in a Sorted Rotated Array:**
Given an array of distinct elements that was initially sorted in ascending order and then rotated at some unknown point, find the minimum element in the array.

**Input:**
- An array `arr[]` of size `N` containing distinct integers, which is sorted and then rotated.

**Output:**
- An integer representing the minimum element in the array.

**Constraints:**
- The array does not contain duplicate elements.
- The array is non-empty.

**Goal of the Code:**
To efficiently find and return the minimum element in a sorted and rotated array.

**Approach:**
Utilize a modified binary search algorithm to achieve an efficient solution with a time complexity of O(log N).

**Pseudocode/Algorithm:**
1. Initialize two pointers:
   - `low` to the start of the array (`0`).
   - `high` to the end of the array (`N - 1`).

2. While `low` is less than `high`:
   - If the subarray from `low` to `high` is already sorted (i.e., `arr[low] < arr[high]`), then the minimum element is `arr[low]`. Return `arr[low]`.
   - Calculate the middle index: `mid = low + (high - low) / 2`.
   - If `arr[mid]` is greater than `arr[high]`, it means the minimum element is in the right half of the array. Update `low` to `mid + 1`.
   - Otherwise, the minimum element is in the left half or could be the `mid` element itself. Update `high` to `mid`.

3. After exiting the loop, `low` will point to the minimum element. Return `arr[low]`.

**Time and Space Complexity:**
- **Time Complexity:** O(log N) – The binary search approach ensures that the search space is halved in each iteration.
- **Space Complexity:** O(1) – Only a constant amount of extra space is used.

**Additional Notes/Hints:**
- This problem leverages the properties of a rotated sorted array. By comparing the middle element with the high (or low) element, we can determine which part of the array contains the minimum element.
- It's crucial to handle edge cases, such as when the array is not rotated at all or when the rotation point is at the end of the array.

**Java Implementation:**
```java
class Solution {
    static int minNumber(int arr[], int low, int high) {
        while (low < high) {
            if (arr[low] < arr[high]) {
                return arr[low];
            }
            int mid = low + (high - low) / 2;
            if (arr[mid] > arr[high]) {
                low = mid + 1;
            } else {
                high = mid;
            }
        }
        return arr[low];
    }
}


## Problem Statement: 80
**Maximum Element in a Sorted and Rotated Array:**
Given a sorted array `arr[]` of size `n` that has been rotated at some unknown point and may contain duplicates, the task is to find the maximum element in the array.

**Input:**
- An integer array `arr[]` of size `n`, which is sorted in ascending order and then rotated at an unknown point. The array may contain duplicate elements.

**Output:**
- An integer representing the maximum element in the array.

**Constraints:**
- The array may contain duplicate elements.
- The array is non-empty.

**Goal of the Code:**
To efficiently find and return the maximum element in a sorted and rotated array.

**Approach:**
Utilize a modified binary search algorithm to achieve an efficient solution with a time complexity of O(log n) in the average case. However, due to the presence of duplicates, the worst-case time complexity can degrade to O(n).

**Pseudocode/Algorithm:**
1. Initialize two pointers:
   - `lo` to the start of the array (`0`).
   - `hi` to the end of the array (`n - 1`).

2. While `lo` is less than `hi`:
   - If the subarray from `lo` to `hi` is already sorted (i.e., `arr[lo] <= arr[hi]`), then the maximum element is `arr[hi]`. Return `arr[hi]`.
   - Calculate the middle index: `mid = lo + (hi - lo) / 2`.
   - If `arr[mid] > arr[lo]`, it indicates that the left half is sorted, and the maximum must be either `arr[mid]` or in the right half. Update `lo` to `mid`.
   - Otherwise, the maximum element is in the left half. Update `hi` to `mid - 1`.

3. After exiting the loop, `lo` will point to the maximum element. Return `arr[lo]`.

**Time and Space Complexity:**
- **Time Complexity:** O(log n) in the average case. Due to the presence of duplicates, the worst-case time complexity can degrade to O(n).
- **Space Complexity:** O(1) – Only a constant amount of extra space is used.

**Additional Notes/Hints:**
- This problem leverages the properties of a rotated sorted array. By comparing the middle element with the low element, we can determine which part of the array contains the maximum element.
- It's crucial to handle edge cases, such as when the array is not rotated at all or when the rotation point is at the end of the array.

**Java Implementation:**
```java
public class Main {
    public static int findMax(int[] arr) {
        int lo = 0, hi = arr.length - 1;

        while (lo < hi) {
            // If the current subarray is already sorted,
            // the maximum is at the hi index
            if (arr[lo] <= arr[hi])
                return arr[hi];

            int mid = (lo + hi) / 2;

            // The left half is sorted, the maximum must
            // be either arr[mid] or in the right half.
            if (arr[mid] > arr[lo])
                lo = mid;
            else
                hi = mid - 1;
        }

        return arr[lo];
    }

    public static void main(String[] args) {
        int[] arr = {7, 8, 9, 10, 1, 2, 3, 4, 5};
        System.out.println(findMax(arr)); // Output: 10
    }
}


## Problem Statement: 81
**Find the Index of the First Occurrence of 1 in an Infinite Sorted Binary Array:**
You are given an infinite array consisting of only 0s and 1s, sorted in non-decreasing order. Your task is to find the index of the first occurrence of 1 in this array.

**Input:**
- Since the array is infinite, you cannot access its length. Instead, you have access to a function `get(index)` that returns the value at the specified index.

**Output:**
- Return the index of the first occurrence of 1. If 1 is not present in the array, return `-1`.

**Constraints:**
- The array is infinite and sorted.
- The array contains only 0s and 1s.

**Approach:**
To find the index of the first occurrence of 1 in an infinite sorted binary array, we can use a combination of exponential search and binary search:

1. **Exponential Search to Find Boundaries:**
   - Start with an initial range where `low = 0` and `high = 1`.
   - Double the `high` index until `get(high)` returns 1. This helps in identifying a range `[low, high]` where the first occurrence of 1 lies.

2. **Binary Search Within the Identified Range:**
   - Once the range is identified, perform a binary search within `[low, high]` to find the first occurrence of 1.
   - In each iteration, calculate the middle index `mid`.
   - If `get(mid)` is 1 and either `mid` is 0 or `get(mid - 1)` is 0, then `mid` is the index of the first occurrence of 1.
   - If `get(mid)` is 1, move the `high` pointer to `mid - 1` to search in the left half.
   - If `get(mid)` is 0, move the `low` pointer to `mid + 1` to search in the right half.

**Algorithm:**
1. Initialize `low = 0` and `high = 1`.
2. While `get(high)` is 0:
   - Set `low = high`.
   - Double the `high` index (`high = 2 * high`).
3. Perform binary search within the range `[low, high]`:
   - While `low` is less than or equal to `high`:
     - Calculate `mid = low + (high - low) / 2`.
     - If `get(mid)` is 1 and (`mid` is 0 or `get(mid - 1)` is 0), return `mid`.
     - If `get(mid)` is 1, set `high = mid - 1`.
     - If `get(mid)` is 0, set `low = mid + 1`.
4. If no 1 is found, return `-1`.

**Time Complexity:**
- The exponential search phase takes O(log p) time, where `p` is the position of the first occurrence of 1.
- The binary search phase takes O(log p) time.
- Thus, the overall time complexity is O(log p).

**Space Complexity:**
- The algorithm uses a constant amount of extra space, i.e., O(1).

**Java Implementation:**
```java
public class Solution {

    // Mock function to simulate access to an infinite array.
    // In a real scenario, this function would access the actual data source.
    public static int get(long index) {
        // Example infinite array: [0, 0, 0, 0, 1, 1, 1, ...]
        // The first occurrence of 1 is at index 4.
        if (index < 4) {
            return 0;
        } else {
            return 1;
        }
    }

    public static long firstOne() {
        long low = 0;
        long high = 1;

        // Exponential search to find the range where the first 1 is located
        while (get(high) == 0) {
            low = high;
            high *= 2;
        }

        // Binary search within the identified range
        while (low <= high) {
            long mid = low + (high - low) / 2;
            if (get(mid) == 1 && (mid == 0 || get(mid - 1) == 0)) {
                return mid;
            } else if (get(mid) == 1) {
                high = mid - 1;
            } else {
                low = mid + 1;
            }
        }

        // If 1 is not found, return -1
        return -1;
    }

    public static void main(String[] args) {
        System.out.println("The index of the first occurrence of 1 is: " + firstOne());
    }
}


## Problem Statement:  82
**Search in Infinite sorted Array:**
Given an infinite sorted array consisting of only 0s and 1s, find the index of the first occurrence of 1.

**Input**:  
- An infinite sorted array of 0s and 1s.

**Output**:  
- An integer representing the index of the first occurrence of 1.

**Constraints**:  
- The array is infinite and sorted in non-decreasing order.
- The array contains only 0s and 1s.

**Goal of the code**:  
Efficiently determine the index of the first occurrence of 1 in an infinite sorted binary array.

**Approach**:  
1. **Initialize Pointers**:
   - Start with `low = 0` and `high = 1`.
2. **Expand the Search Range**:
   - While `get(high) == 0`, set `low = high` and double the `high` index (`high = 2 * high`).
3. **Binary Search Within the Range**:
   - Once `get(high) == 1`, perform a binary search between `low` and `high` to find the first occurrence of 1.

**Time and Space Complexity**:
- **Time Complexity**: O(log n), where n is the index of the first occurrence of 1. The range expansion takes O(log n) time, and the subsequent binary search within that range also takes O(log n) time.
- **Space Complexity**: O(1), as the algorithm uses a constant amount of extra space.

**Additional Notes**:
- This approach efficiently narrows down the search space by first expanding the range exponentially until a '1' is found, and then applying binary search within that range to find the exact index of the first occurrence of 1.
- The `get(index)` function is assumed to be a predefined method that returns the value at the specified index in the infinite array.

**Pseudocode/Algorithm**:
```java
// Expand the range to find an upper bound where '1' is present
while get(high) == 0:
    low = high
    high = 2 * high

// Perform binary search within the identified range
while low <= high:
    mid = low + (high - low) / 2
    if get(mid) == 1 and (mid == 0 or get(mid - 1) == 0):
        return mid
    elif get(mid) == 1:
        high = mid - 1
    else:
        low = mid + 1

return -1  // In case '1' is not found


## Problem Statement: 83

**Problem Description**:  
Given an increasing sorted rotated array `arr` of distinct integers, determine the number of times the array has been right-rotated, denoted as `k`.

**Input**:  
- A list of distinct integers `arr` representing a sorted array that has been right-rotated `k` times.

**Output**:  
- An integer `k` representing the number of times the array has been right-rotated.

**Constraints**:  
- The array contains distinct integers.
- The array was originally sorted in ascending order.
- The array has been rotated at an unknown pivot.
- The solution should have a time complexity of O(log n).

**Goal of the code**:  
Efficiently determine the number of right rotations `k` applied to a sorted array using a modified binary search approach.

**Approach**:  
1. **Identify the Pivot Point**:
   - The pivot point is the index where the smallest element in the array resides. This index corresponds to the number of rotations `k`.
2. **Binary Search to Find the Pivot**:
   - Initialize two pointers: `low` at the start of the array and `high` at the end.
   - While `low` is less than or equal to `high`:
     - Calculate `mid` as the average of `low` and `high`.
     - Check if `mid` is the pivot:
       - If `arr[mid]` is less than or equal to `arr[high]`, the smallest element is in the left half or at `mid`. Adjust `high` to `mid`.
       - Otherwise, the smallest element is in the right half. Adjust `low` to `mid + 1`.
   - The index `low` will point to the smallest element, which is the pivot. This index is the value of `k`.


**Time and Space Complexity**:
- **Time Complexity**: O(log n), where n is the number of elements in the array. The algorithm performs a binary search to find the pivot point.
- **Space Complexity**: O(1), as the algorithm uses a constant amount of extra space.

**Additional Notes**:
- This approach leverages the properties of the rotated sorted array to maintain the efficiency of binary search.
- By identifying the pivot point, the algorithm determines the number of rotations `k` applied to the original sorted array.

**Pseudocode/Algorithm**:
```java
function findKRotation(arr): low = 0 high = len(arr) - 1

while low < high:
    mid = low + (high - low) // 2

    if arr[mid] <= arr[high]:
        high = mid
    else:
        low = mid + 1

return low


## Problem Statement: 84
Given an array `arr[]` where no two adjacent elements are the same, find the index of a peak element. An element is considered to be a peak if it is greater than its adjacent elements (if they exist). If there are multiple peak elements, return the index of any one of them. The output will be "true" if the index returned by your function is correct; otherwise, it will be "false".

*Note:* Consider the element before the first element and the element after the last element to be negative infinity.

**Input:**
- An integer array `arr[]` of length `n` where no two adjacent elements are the same.

**Output:**
- An integer representing the index of any one peak element in the array.

**Constraints:**
- The array contains distinct elements.
- No two adjacent elements are the same.

**Goal of the code:**
To efficiently find and return the index of any peak element in the given array.

**Approach:**
Utilize a binary search algorithm to find a peak element in logarithmic time. The idea is to leverage the properties of the array and the definition of a peak to narrow down the search space efficiently.

**Pseudocode/Algorithm:**
1. **Initial Checks:**
   - If the array has only one element, return index `0` since it's trivially a peak.
   - If the first element is greater than the second, return index `0`.
   - If the last element is greater than the second last, return index `n - 1`.

2. **Binary Search:**
   - Initialize `low` to `1` and `high` to `n - 2` (since the edge cases are already handled).
   - While `low` is less than or equal to `high`:
     - Calculate `mid` as `low + (high - low) / 2`.
     - If `arr[mid]` is greater than both `arr[mid - 1]` and `arr[mid + 1]`, return `mid` as it is a peak.
     - If `arr[mid]` is less than `arr[mid + 1]`, move `low` to `mid + 1` to search in the right half.
     - Otherwise, move `high` to `mid - 1` to search in the left half.

3. **Return Default:**
   - If no peak is found in the binary search loop (which is unlikely due to problem constraints), return index `0`.

**Time and Space Complexity:**
- **Time Complexity:** O(log n) due to the binary search approach.
- **Space Complexity:** O(1) as no additional space is used beyond a few variables.

**Additional Notes:**
- This approach ensures that we find a peak element efficiently without needing to examine every element in the array.
- The problem guarantees the existence of at least one peak due to the constraints provided.


## Problem Statement: 85
Given a binary matrix of order NxN, where each element is either 0 or 1 and all rows are sorted in non-decreasing order, find the row number with the maximum number of 1s. Additionally, determine the count of 1s in that row.

**Note:**
1. In case of a tie, return the smallest row number.
2. Row numbering starts from 0.

**Input:**
- An integer `N` representing the order of the matrix.
- A binary matrix `mat` of size `N x N` where each row is sorted in non-decreasing order.

**Output:**
- An array of two integers:
  - The index of the row with the maximum number of 1s.
  - The count of 1s in that row.

**Constraints:**
- The matrix is square with dimensions `N x N`.
- Each row is sorted in non-decreasing order.
- The matrix contains only binary values (0s and 1s).

**Goal of the code:**
Efficiently identify the row with the maximum number of 1s in a sorted binary matrix and return both the row index and the count of 1s.

**Approach:**
Leverage the sorted property of each row to efficiently determine the count of 1s using binary search. By finding the first occurrence of 1 in each row, we can calculate the number of 1s and track the row with the maximum count.

**Pseudocode/Algorithm:**
1. **Initialize Variables:**
   - `maxOnesCount` to 0: Tracks the maximum number of 1s found.
   - `maxRowIndex` to 0: Tracks the index of the row with the maximum number of 1s.

2. **Define Helper Function `firstOneIndex(row)`:**
   - Perform a binary search on the input `row` to find the index of the first occurrence of 1.
   - Return the index if found; otherwise, return -1.

3. **Iterate Through Each Row:**
   - For each row `i` in the matrix:
     - Use `firstOneIndex` to find the first occurrence of 1.
     - If a 1 is found (`index != -1`):
       - Calculate the number of 1s in the row as `N - index`.
       - If this count is greater than `maxOnesCount`:
         - Update `maxOnesCount` with the new count.
         - Update `maxRowIndex` with the current row index `i`.

4. **Return Result:**
   - Return an array containing `maxRowIndex` and `maxOnesCount`.

**Time and Space Complexity:**
- **Time Complexity:** O(N log N)
  - The binary search within each row operates in O(log N) time.
  - Iterating through all N rows results in a total time complexity of O(N log N).

- **Space Complexity:** O(1)
  - The algorithm uses a constant amount of extra space for variables.

**Additional Notes:**
- This approach efficiently utilizes the sorted property of each row to minimize the search time for the first occurrence of 1.
- In the event of multiple rows having the same number of 1s, the algorithm naturally returns the smallest row index due to the sequential iteration from the top of the matrix.


## Problem Statement: 86

Given a 2D integer matrix `mat[][]` of size `n x m`, where every row and column is sorted in increasing order, and an integer `x`, determine whether `x` is present in the matrix.

**Input:**
- An integer `n` representing the number of rows in the matrix.
- An integer `m` representing the number of columns in the matrix.
- A 2D integer matrix `mat` of size `n x m`, where each row and column is sorted in increasing order.
- An integer `x` representing the target value to search for.

**Output:**
- Return `true` if `x` is found in the matrix; otherwise, return `false`.

**Constraints:**
- The matrix dimensions `n` and `m` are positive integers.
- Each row and column in the matrix is sorted in increasing order.
- The matrix may contain duplicate elements.

**Goal of the code:**
Efficiently determine whether the target value `x` exists within the given row-wise and column-wise sorted matrix.

**Approach:**
Utilize a search strategy that leverages the sorted properties of the matrix to efficiently locate the target value. Start the search from the top-right corner of the matrix and iteratively narrow down the search space based on comparisons with the target value.

**Pseudocode/Algorithm:**
1. Initialize `i` to 0 (first row) and `j` to `m - 1` (last column).
2. While `i` is within the number of rows and `j` is non-negative:
   - If `mat[i][j]` equals `x`, return `true`.
   - If `mat[i][j]` is greater than `x`, decrement `j` to move left.
   - If `mat[i][j]` is less than `x`, increment `i` to move down.
3. If the loop terminates without finding `x`, return `false`.

**Time and Space Complexity:**
- **Time Complexity:** O(n + m), where `n` is the number of rows and `m` is the number of columns. In the worst case, the algorithm may need to traverse the entire first row and last column.
- **Space Complexity:** O(1), as the algorithm uses a constant amount of extra space.

**Additional Notes/Hints:**
- Starting the search from the top-right corner is strategic because it allows elimination of either a row or a column in each step:
  - If the current element is greater than the target, all elements in the current column below are also greater, so move left.
  - If the current element is less than the target, all elements in the current row to the left are also less, so move down.
- This approach ensures that each step brings us closer to the target while efficiently narrowing down the search space.
