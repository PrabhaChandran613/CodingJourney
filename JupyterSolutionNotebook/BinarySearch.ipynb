Efficiently determine the presence of the integer `k` in the sorted array `arr[]`.

**Approach:**
Utilize the binary search algorithm to leverage the sorted nature of the array, achieving efficient search performance.

**Pseudocode/Algorithm:**
1. Initialize two pointers: `left` at the start of the array and `right` at the end.
2. While `left` is less than or equal to `right`:
   - Calculate the middle index: `mid = left + (right - left) / 2`.
   - If `arr[mid]` equals `k`, return `true`.
   - If `arr[mid]` is less than `k`, move the `left` pointer to `mid + 1`.
   - If `arr[mid]` is greater than `k`, move the `right` pointer to `mid - 1`.
3. If the loop ends without finding `k`, return `false`.

**Time and Space Complexity:**
- **Time Complexity:** O(log n), where `n` is the number of elements in the array. This is due to the binary search algorithm, which halves the search space with each iteration.

- Binary search is optimal for searching in sorted arrays due to its logarithmic time complexity.

**Java Implementation:**
```java
class Solution {
    static boolean searchInSorted(int arr[], int k) {
        int left = 0;
        int right = arr.length - 1;
        
        while (left <= right) {
            int mid = left + (right - left) / 2;
            
            if (arr[mid] == k) {
                return true;
            } else if (arr[mid] < k) {
                left = mid + 1;
            } else {
                right = mid - 1;
            }
        }
        
        return false;
    }
}


### Problem Statement: 73
Given a sorted array `arr[]` with unique elements and an integer `k`, find the index (0-based) of the largest element in `arr[]` that is less than or equal to `k`. This element is called the "floor" of `k`. If such an element does not exist, return `-1`.

**Input:**
- A sorted array `arr[]` of unique integers.
- An integer `k`.

**Output:**
- An integer representing the index of the floor of `k` in `arr[]`. If no such element exists, return `-1`.

**Constraints:**
- The array `arr[]` is sorted in ascending order.
- All elements in `arr[]` are unique.
- The array may contain zero or more elements.
- The integer `k` can be any valid integer value.

**Goal of the Code:**
Efficiently determine the index of the largest element in the sorted array `arr[]` that is less than or equal to the integer `k`.

**Approach:**
Utilize the binary search algorithm to efficiently find the floor of `k` in the sorted array `arr[]`.

**Pseudocode/Algorithm:**
1. Initialize two pointers: `lo` at the start of the array and `hi` at the end.
2. Initialize a variable `floorIndex` to `-1` to keep track of the potential floor index.
3. While `lo` is less than or equal to `hi`:
   - Calculate the middle index: `mid = lo + (hi - lo) / 2`.
   - If `arr[mid]` equals `k`, return `mid` as the floor index.
   - If `arr[mid]` is less than `k`:
     - Update `floorIndex` to `mid`.
     - Move the `lo` pointer to `mid + 1` to search in the right subarray.
   - If `arr[mid]` is greater than `k`, move the `hi` pointer to `mid - 1` to search in the left subarray.
4. After the loop, return `floorIndex`. If no floor exists, `floorIndex` will remain `-1`.

**Time and Space Complexity:**
- **Time Complexity:** O(log n), where `n` is the number of elements in the array. This efficiency is achieved through the binary search algorithm.
- **Space Complexity:** O(1), as the algorithm uses a constant amount of extra space.

**Additional Notes/Hints:**
- Binary search is optimal for searching in sorted arrays due to its logarithmic time complexity.
- Ensure that edge cases, such as empty arrays or `k` being smaller than the smallest element in the array, are handled appropriately.

**Java Implementation:**
```java
class Solution {
    static int findFloor(int[] arr, int k) {
        int lo = 0, hi = arr.length - 1;
        int floorIndex = -1;

        while (lo <= hi) {
            int mid = lo + (hi - lo) / 2;

            if (arr[mid] == k) {
                return mid;
            } else if (arr[mid] < k) {
                floorIndex = mid;
                lo = mid + 1;
            } else {
                hi = mid - 1;
            }
        }

        return floorIndex;
    }
}


### Problem Statement: 74
Given a sorted array `arr[]` with possibly some duplicates, find the first and last occurrences of an element `x` in the array. If `x` is not found, return both indices as `-1`.

**Input:**
- A sorted array `arr[]` that may contain duplicates.
- An integer `x` whose first and last occurrences need to be found.

**Output:**
- An array with two integers:
  - The first integer is the index of the first occurrence of `x`.
  - The second integer is the index of the last occurrence of `x`.

  If `x` is not present in the array, return both indices as `-1`.

**Constraints:**
- The array `arr[]` is sorted in ascending order and may contain duplicates.
- The array can have `0 ≤ n ≤ 10^5` elements.
- The value `x` can be any integer.

**Goal of the Code:**
Efficiently find the first and last occurrences of `x` in the sorted array `arr[]`.

**Approach:**
1. Utilize binary search to efficiently find both the first and last occurrences of `x` in the sorted array.
2. Perform the first binary search to locate the first occurrence by adjusting the search boundaries if `x` is found.
3. Perform a second binary search to locate the last occurrence similarly.

**Pseudocode/Algorithm:**
1. **First Occurrence Search:**
   - Perform binary search to locate the first occurrence of `x`.
   - If `arr[mid] == x`, update the index and search the left half of the array by setting `high = mid - 1`.
   - Continue adjusting the boundaries until the first occurrence is found or the boundaries converge.

2. **Last Occurrence Search:**
   - Perform binary search to locate the last occurrence of `x`.
   - If `arr[mid] == x`, update the index and search the right half of the array by setting `low = mid + 1`.
   - Continue adjusting the boundaries until the last occurrence is found or the boundaries converge.

3. Return the result with both first and last indices.

**Time and Space Complexity:**
- **Time Complexity:** O(log n) for both the first and last occurrence searches, where `n` is the size of the array.
- **Space Complexity:** O(1), as only a constant amount of extra space is used for variables.

**Additional Notes/Hints:**
- The problem leverages the fact that the array is sorted, making binary search a suitable solution.
- Special edge cases like empty arrays or arrays where `x` is not present need to be handled by returning `-1` for both indices.

**Java Implementation:**
```java
class GFG {
    ArrayList<Integer> find(int arr[], int x) {
        // code here
        ArrayList<Integer> result = new ArrayList<>();
        int first = firstOccurence(arr, x);
        int last = lastOccurence(arr, x);
        result.add(first);
        result.add(last);
        return result;
    }

    private int firstOccurence(int arr[], int x) {
        int low = 0, high = arr.length - 1, firstOccurrence = -1;
        while (low <= high) {
            int mid = low + (high - low) / 2;
            if (arr[mid] == x) {
                firstOccurrence = mid;
                high = mid - 1;
            } else if (arr[mid] < x) {
                low = mid + 1;
            } else {
                high = mid - 1;
            }
        }
        return firstOccurrence;
    }

    private int lastOccurence(int arr[], int x) {
        int low = 0, high = arr.length - 1, lastOccurrence = -1;
        while (low <= high) {
            int mid = low + (high - low) / 2;
            if (arr[mid] == x) {
                lastOccurrence = mid;
                low = mid + 1;
            } else if (arr[mid] < x) {
                low = mid + 1;
            } else {
                high = mid - 1;
            }
        }
        return lastOccurrence;
    }
}


### Problem Statement: 75
Given an array `nums` containing `n` distinct numbers in the range `[0, n]`, return the only number in the range that is missing from the array.

**Input:**
- An integer array `nums` of length `n` containing distinct numbers from the range `[0, n]`.

**Output:**
- The single integer from the range `[0, n]` that is missing in `nums`.

**Constraints:**
- The array `nums` contains `n` unique numbers.
- The length of `nums` is `1 <= n <= 10^4`.
- All elements in `nums` are within the range `[0, n]`.

**Goal of the Code:**
Efficiently identify and return the missing number from the array `nums`.

**Approach:**
1. **Mathematical Summation:**
   - Calculate the expected sum of the first `n` natural numbers using the formula: `expectedSum = n * (n + 1) / 2`.
   - Compute the actual sum of the elements present in `nums`.
   - The difference between the `expectedSum` and the `actualSum` will yield the missing number.

**Pseudocode/Algorithm:**
1. Initialize `sum` to `0` and determine `n` as the length of `nums`.
2. Iterate through each number in `nums` and accumulate the sum.
3. Compute the `expectedSum` using the formula for the sum of the first `n` natural numbers.
4. The missing number is `expectedSum - sum`.
5. Return the missing number.

**Time and Space Complexity:**
- **Time Complexity:** O(n), where `n` is the length of the array `nums`. This is due to the single pass required to compute the sum of elements.
- **Space Complexity:** O(1), as the solution utilizes a constant amount of extra space.

**Java Implementation:**
```java
class Solution {
    public int missingNumber(int[] nums) {
        int sum = 0, n = nums.length;
        for (int num : nums) {
            sum += num;
        }
        int expectedSum = (n * (n + 1)) / 2;
        return expectedSum - sum;
    }
}


### Problem Statement: 76
**Square Root:**
Given a positive integer `n`, determine its square root. If `n` is not a perfect square, return the floor value of its square root. The floor value of a number is the greatest integer less than or equal to that number.

**Input:**
- A single positive integer `n`.

**Output:**
- An integer representing the floor value of the square root of `n`.

**Constraints:**
- \( 1 \leq n \leq 10^9 \)

**Goal of the Code:**
Efficiently compute and return the floor value of the square root of the given integer `n`.

**Approach:**
1. **Binary Search Method:**
   - Utilize binary search to find the floor value of the square root:
     - Initialize `low` to 1 and `high` to `n`.
     - While `low` is less than or equal to `high`:
       - Compute `mid` as the average of `low` and `high`.
       - If `mid * mid` equals `n`, `mid` is the exact square root; return `mid`.
       - If `mid * mid` is less than `n`, move the `low` pointer to `mid + 1` and update the result to `mid`.
       - If `mid * mid` is greater than `n`, move the `high` pointer to `mid - 1`.
     - Return the result after the loop concludes.

**Pseudocode/Algorithm:**
1. **Function:** `floorSqrt(n)`
2. **Input:** Integer `n`
3. **Output:** Integer representing the floor value of the square root of `n`
4. **Steps:**
   - **Edge Case:** If `n` is less than 2, return `n`.
   - Initialize `low` to 1 and `high` to `n`.
   - Initialize `result` to 1.
   - **While** `low` ≤ `high`:
     - Calculate `mid` as `low + (high - low) // 2`.
     - **If** `mid * mid` equals `n`:
       - Return `mid`.
     - **Else If** `mid * mid` is less than `n`:
       - Set `result` to `mid`.
       - Set `low` to `mid + 1`.
     - **Else:**
       - Set `high` to `mid - 1`.
   - Return `result`.

**Time and Space Complexity:**
- **Time Complexity:** \( O(\log n) \) — The binary search approach reduces the search space by half in each iteration, leading to a logarithmic time complexity.
- **Space Complexity:** \( O(1) \) — The algorithm uses a constant amount of extra space.

**Java Implementation:**
```java
class Solution {
    int floorSqrt(int n) {
        if (n < 2) {
            return n;
        }
        int low = 1, high = n, result = 1;
        while (low <= high) {
            int mid = low + (high - low) / 2;
            if (mid * mid == n) {
                return mid;
            } else if (mid * mid < n) {
                result = mid;
                low = mid + 1;
            } else {
                high = mid - 1;
            }
        }
        return result;
    }
}


### Problem Statement: 77
Given a sorted and rotated array `arr` that may contain duplicates, and a target key `k`, determine whether the key is present in the array.

**Input:**
- An integer `k` representing the target key.
- An array `arr[]` of integers which is sorted, rotated at some pivot, and may contain duplicates.

**Output:**
- Return `true` if the key `k` is present in the array; otherwise, return `false`.

**Constraints:**
- The array may contain duplicate elements.
- The array is sorted and then rotated at an unknown pivot.
- The array can be of any size within reasonable limits.

**Goal of the code:**
To efficiently determine if the target key `k` exists in the rotated sorted array `arr`.

**Approach:**
1. **Binary Search with Adjustments for Duplicates:**
   - Initialize two pointers, `low` and `high`, to the start and end of the array, respectively.
   - While `low` is less than or equal to `high`:
     - Calculate the middle index `mid`.
     - If `arr[mid]` equals `k`, return `true`.
     - If `arr[mid]`, `arr[low]`, and `arr[high]` are equal, increment `low` and decrement `high` to skip duplicates.
     - Determine which part of the array is sorted:
       - If the left part is sorted:
         - Check if `k` lies within this range; adjust `low` and `high` accordingly.
       - Otherwise, the right part must be sorted:
         - Check if `k` lies within this range; adjust `low` and `high` accordingly.
   - If the loop ends without finding `k`, return `false`.

```java
boolean searchInRotatedArray(int[] arr, int k) {
    int low = 0, high = arr.length - 1;
    while (low <= high) {
        int mid = low + (high - low) / 2;
        if (arr[mid] == k) {
            return true;
        }
        if (arr[mid] == arr[low] && arr[mid] == arr[high]) {
            ++low;
            --high;
            continue;
        }
        if (arr[low] <= arr[mid]) {
            if (k >= arr[low] && k < arr[mid]) {
                high = mid - 1;
            } else {
                low = mid + 1;
            }
        } else {
            if (k > arr[mid] && k <= arr[high]) {
                low = mid + 1;
            } else {
                high = mid - 1;
            }
        }
    }
    return false;
}


## Problem Statement:78
**Search Element with distinct in sorted Rotated Array:**
Given a sorted and rotated array `arr[]` of distinct elements, find the index of a target key. Return -1 if the key is not found.

**Input**:  
- A rotated sorted array `arr[]` of distinct integers.
- An integer `key` representing the target value to search for.

**Output**:  
- An integer representing the index of the target `key` in the array. If the `key` is not found, return -1.

**Constraints**:  
- The array contains distinct integers.
- The array was originally sorted in ascending order.
- The array has been rotated at an unknown pivot.
- The search algorithm should have a time complexity of O(log n).

**Goal of the code**:  
Efficiently locate the index of the target `key` in a rotated sorted array using a modified binary search approach.

**Approach**:  
1. **Initialize Pointers**:
   - Set `low` to 0 and `high` to the last index of the array.
2. **Binary Search with Rotation Consideration**:
   - While `low` is less than or equal to `high`:
     - Calculate `mid` as the average of `low` and `high`.
     - If `arr[mid]` equals the `key`, return `mid`.
     - Determine which part of the array is sorted:
       - If `arr[low]` is less than or equal to `arr[mid]`, the left part is sorted:
         - If `key` lies between `arr[low]` and `arr[mid]`, adjust `high` to `mid - 1`.
         - Otherwise, adjust `low` to `mid + 1`.
       - Else, the right part is sorted:
         - If `key` lies between `arr[mid]` and `arr[high]`, adjust `low` to `mid + 1`.
         - Otherwise, adjust `high` to `mid - 1`.
3. **Return Result**:
   - If the `key` is not found, return -1.


**Time and Space Complexity**:
- **Time Complexity**: O(log n), where n is the number of elements in the array. The algorithm performs a binary search, effectively halving the search space with each iteration.
- **Space Complexity**: O(1), as the algorithm uses a constant amount of extra space.

**Additional Notes**:
- This approach leverages the properties of the rotated sorted array to maintain the efficiency of binary search.
- By identifying which part of the array is sorted, the algorithm decides the direction to continue the search, ensuring that the target `key` is found in logarithmic time.


**Pseudocode/Algorithm**:
function searchInRotatedArray(arr, key): low = 0 high = len(arr) - 1
```java
while low <= high:
    mid = (low + high) // 2

    if arr[mid] == key:
        return mid

    if arr[low] <= arr[mid]:  // Left part is sorted
        if arr[low] <= key < arr[mid]:
            high = mid - 1
        else:
            low = mid + 1
    else:  // Right part is sorted
        if arr[mid] < key <= arr[high]:
            low = mid + 1
        else:
            high = mid - 1

return -1

## Problem Statement: 79
**Minimum Number in a Sorted Rotated Array:**
Given an array of distinct elements that was initially sorted in ascending order and then rotated at some unknown point, find the minimum element in the array.

**Input:**
- An array `arr[]` of size `N` containing distinct integers, which is sorted and then rotated.

**Output:**
- An integer representing the minimum element in the array.

**Constraints:**
- The array does not contain duplicate elements.
- The array is non-empty.

**Goal of the Code:**
To efficiently find and return the minimum element in a sorted and rotated array.

**Approach:**
Utilize a modified binary search algorithm to achieve an efficient solution with a time complexity of O(log N).

**Pseudocode/Algorithm:**
1. Initialize two pointers:
   - `low` to the start of the array (`0`).
   - `high` to the end of the array (`N - 1`).

2. While `low` is less than `high`:
   - If the subarray from `low` to `high` is already sorted (i.e., `arr[low] < arr[high]`), then the minimum element is `arr[low]`. Return `arr[low]`.
   - Calculate the middle index: `mid = low + (high - low) / 2`.
   - If `arr[mid]` is greater than `arr[high]`, it means the minimum element is in the right half of the array. Update `low` to `mid + 1`.
   - Otherwise, the minimum element is in the left half or could be the `mid` element itself. Update `high` to `mid`.

3. After exiting the loop, `low` will point to the minimum element. Return `arr[low]`.

**Time and Space Complexity:**
- **Time Complexity:** O(log N) – The binary search approach ensures that the search space is halved in each iteration.
- **Space Complexity:** O(1) – Only a constant amount of extra space is used.

**Additional Notes/Hints:**
- This problem leverages the properties of a rotated sorted array. By comparing the middle element with the high (or low) element, we can determine which part of the array contains the minimum element.
- It's crucial to handle edge cases, such as when the array is not rotated at all or when the rotation point is at the end of the array.

**Java Implementation:**
```java
class Solution {
    static int minNumber(int arr[], int low, int high) {
        while (low < high) {
            if (arr[low] < arr[high]) {
                return arr[low];
            }
            int mid = low + (high - low) / 2;
            if (arr[mid] > arr[high]) {
                low = mid + 1;
            } else {
                high = mid;
            }
        }
        return arr[low];
    }
}


## Problem Statement: 80
**Maximum Element in a Sorted and Rotated Array:**
Given a sorted array `arr[]` of size `n` that has been rotated at some unknown point and may contain duplicates, the task is to find the maximum element in the array.

**Input:**
- An integer array `arr[]` of size `n`, which is sorted in ascending order and then rotated at an unknown point. The array may contain duplicate elements.

**Output:**
- An integer representing the maximum element in the array.

**Constraints:**
- The array may contain duplicate elements.
- The array is non-empty.

**Goal of the Code:**
To efficiently find and return the maximum element in a sorted and rotated array.

**Approach:**
Utilize a modified binary search algorithm to achieve an efficient solution with a time complexity of O(log n) in the average case. However, due to the presence of duplicates, the worst-case time complexity can degrade to O(n).

**Pseudocode/Algorithm:**
1. Initialize two pointers:
   - `lo` to the start of the array (`0`).
   - `hi` to the end of the array (`n - 1`).

2. While `lo` is less than `hi`:
   - If the subarray from `lo` to `hi` is already sorted (i.e., `arr[lo] <= arr[hi]`), then the maximum element is `arr[hi]`. Return `arr[hi]`.
   - Calculate the middle index: `mid = lo + (hi - lo) / 2`.
   - If `arr[mid] > arr[lo]`, it indicates that the left half is sorted, and the maximum must be either `arr[mid]` or in the right half. Update `lo` to `mid`.
   - Otherwise, the maximum element is in the left half. Update `hi` to `mid - 1`.

3. After exiting the loop, `lo` will point to the maximum element. Return `arr[lo]`.

**Time and Space Complexity:**
- **Time Complexity:** O(log n) in the average case. Due to the presence of duplicates, the worst-case time complexity can degrade to O(n).
- **Space Complexity:** O(1) – Only a constant amount of extra space is used.

**Additional Notes/Hints:**
- This problem leverages the properties of a rotated sorted array. By comparing the middle element with the low element, we can determine which part of the array contains the maximum element.
- It's crucial to handle edge cases, such as when the array is not rotated at all or when the rotation point is at the end of the array.

**Java Implementation:**
```java
public class Main {
    public static int findMax(int[] arr) {
        int lo = 0, hi = arr.length - 1;

        while (lo < hi) {
            // If the current subarray is already sorted,
            // the maximum is at the hi index
            if (arr[lo] <= arr[hi])
                return arr[hi];

            int mid = (lo + hi) / 2;

            // The left half is sorted, the maximum must
            // be either arr[mid] or in the right half.
            if (arr[mid] > arr[lo])
                lo = mid;
            else
                hi = mid - 1;
        }

        return arr[lo];
    }

    public static void main(String[] args) {
        int[] arr = {7, 8, 9, 10, 1, 2, 3, 4, 5};
        System.out.println(findMax(arr)); // Output: 10
    }
}


## Problem Statement: 81
**Find the Index of the First Occurrence of 1 in an Infinite Sorted Binary Array:**
You are given an infinite array consisting of only 0s and 1s, sorted in non-decreasing order. Your task is to find the index of the first occurrence of 1 in this array.

**Input:**
- Since the array is infinite, you cannot access its length. Instead, you have access to a function `get(index)` that returns the value at the specified index.

**Output:**
- Return the index of the first occurrence of 1. If 1 is not present in the array, return `-1`.

**Constraints:**
- The array is infinite and sorted.
- The array contains only 0s and 1s.

**Approach:**
To find the index of the first occurrence of 1 in an infinite sorted binary array, we can use a combination of exponential search and binary search:

1. **Exponential Search to Find Boundaries:**
   - Start with an initial range where `low = 0` and `high = 1`.
   - Double the `high` index until `get(high)` returns 1. This helps in identifying a range `[low, high]` where the first occurrence of 1 lies.

2. **Binary Search Within the Identified Range:**
   - Once the range is identified, perform a binary search within `[low, high]` to find the first occurrence of 1.
   - In each iteration, calculate the middle index `mid`.
   - If `get(mid)` is 1 and either `mid` is 0 or `get(mid - 1)` is 0, then `mid` is the index of the first occurrence of 1.
   - If `get(mid)` is 1, move the `high` pointer to `mid - 1` to search in the left half.
   - If `get(mid)` is 0, move the `low` pointer to `mid + 1` to search in the right half.

**Algorithm:**
1. Initialize `low = 0` and `high = 1`.
2. While `get(high)` is 0:
   - Set `low = high`.
   - Double the `high` index (`high = 2 * high`).
3. Perform binary search within the range `[low, high]`:
   - While `low` is less than or equal to `high`:
     - Calculate `mid = low + (high - low) / 2`.
     - If `get(mid)` is 1 and (`mid` is 0 or `get(mid - 1)` is 0), return `mid`.
     - If `get(mid)` is 1, set `high = mid - 1`.
     - If `get(mid)` is 0, set `low = mid + 1`.
4. If no 1 is found, return `-1`.

**Time Complexity:**
- The exponential search phase takes O(log p) time, where `p` is the position of the first occurrence of 1.
- The binary search phase takes O(log p) time.
- Thus, the overall time complexity is O(log p).

**Space Complexity:**
- The algorithm uses a constant amount of extra space, i.e., O(1).

**Java Implementation:**
```java
public class Solution {

    // Mock function to simulate access to an infinite array.
    // In a real scenario, this function would access the actual data source.
    public static int get(long index) {
        // Example infinite array: [0, 0, 0, 0, 1, 1, 1, ...]
        // The first occurrence of 1 is at index 4.
        if (index < 4) {
            return 0;
        } else {
            return 1;
        }
    }

    public static long firstOne() {
        long low = 0;
        long high = 1;

        // Exponential search to find the range where the first 1 is located
        while (get(high) == 0) {
            low = high;
            high *= 2;
        }

        // Binary search within the identified range
        while (low <= high) {
            long mid = low + (high - low) / 2;
            if (get(mid) == 1 && (mid == 0 || get(mid - 1) == 0)) {
                return mid;
            } else if (get(mid) == 1) {
                high = mid - 1;
            } else {
                low = mid + 1;
            }
        }

        // If 1 is not found, return -1
        return -1;
    }

    public static void main(String[] args) {
        System.out.println("The index of the first occurrence of 1 is: " + firstOne());
    }
}


## Problem Statement:  82
**Search in Infinite sorted Array:**
Given an infinite sorted array consisting of only 0s and 1s, find the index of the first occurrence of 1.

**Input**:  
- An infinite sorted array of 0s and 1s.

**Output**:  
- An integer representing the index of the first occurrence of 1.

**Constraints**:  
- The array is infinite and sorted in non-decreasing order.
- The array contains only 0s and 1s.

**Goal of the code**:  
Efficiently determine the index of the first occurrence of 1 in an infinite sorted binary array.

**Approach**:  
1. **Initialize Pointers**:
   - Start with `low = 0` and `high = 1`.
2. **Expand the Search Range**:
   - While `get(high) == 0`, set `low = high` and double the `high` index (`high = 2 * high`).
3. **Binary Search Within the Range**:
   - Once `get(high) == 1`, perform a binary search between `low` and `high` to find the first occurrence of 1.

**Time and Space Complexity**:
- **Time Complexity**: O(log n), where n is the index of the first occurrence of 1. The range expansion takes O(log n) time, and the subsequent binary search within that range also takes O(log n) time.
- **Space Complexity**: O(1), as the algorithm uses a constant amount of extra space.

**Additional Notes**:
- This approach efficiently narrows down the search space by first expanding the range exponentially until a '1' is found, and then applying binary search within that range to find the exact index of the first occurrence of 1.
- The `get(index)` function is assumed to be a predefined method that returns the value at the specified index in the infinite array.

**Pseudocode/Algorithm**:
```java
// Expand the range to find an upper bound where '1' is present
while get(high) == 0:
    low = high
    high = 2 * high

// Perform binary search within the identified range
while low <= high:
    mid = low + (high - low) / 2
    if get(mid) == 1 and (mid == 0 or get(mid - 1) == 0):
        return mid
    elif get(mid) == 1:
        high = mid - 1
    else:
        low = mid + 1

return -1  // In case '1' is not found


## Problem Statement: 83

**Problem Description**:  
Given an increasing sorted rotated array `arr` of distinct integers, determine the number of times the array has been right-rotated, denoted as `k`.

**Input**:  
- A list of distinct integers `arr` representing a sorted array that has been right-rotated `k` times.

**Output**:  
- An integer `k` representing the number of times the array has been right-rotated.

**Constraints**:  
- The array contains distinct integers.
- The array was originally sorted in ascending order.
- The array has been rotated at an unknown pivot.
- The solution should have a time complexity of O(log n).

**Goal of the code**:  
Efficiently determine the number of right rotations `k` applied to a sorted array using a modified binary search approach.

**Approach**:  
1. **Identify the Pivot Point**:
   - The pivot point is the index where the smallest element in the array resides. This index corresponds to the number of rotations `k`.
2. **Binary Search to Find the Pivot**:
   - Initialize two pointers: `low` at the start of the array and `high` at the end.
   - While `low` is less than or equal to `high`:
     - Calculate `mid` as the average of `low` and `high`.
     - Check if `mid` is the pivot:
       - If `arr[mid]` is less than or equal to `arr[high]`, the smallest element is in the left half or at `mid`. Adjust `high` to `mid`.
       - Otherwise, the smallest element is in the right half. Adjust `low` to `mid + 1`.
   - The index `low` will point to the smallest element, which is the pivot. This index is the value of `k`.


**Time and Space Complexity**:
- **Time Complexity**: O(log n), where n is the number of elements in the array. The algorithm performs a binary search to find the pivot point.
- **Space Complexity**: O(1), as the algorithm uses a constant amount of extra space.

**Additional Notes**:
- This approach leverages the properties of the rotated sorted array to maintain the efficiency of binary search.
- By identifying the pivot point, the algorithm determines the number of rotations `k` applied to the original sorted array.

**Pseudocode/Algorithm**:
```java
function findKRotation(arr): low = 0 high = len(arr) - 1

while low < high:
    mid = low + (high - low) // 2

    if arr[mid] <= arr[high]:
        high = mid
    else:
        low = mid + 1

return low


## Problem Statement: 84
Given an array `arr[]` where no two adjacent elements are the same, find the index of a peak element. An element is considered to be a peak if it is greater than its adjacent elements (if they exist). If there are multiple peak elements, return the index of any one of them. The output will be "true" if the index returned by your function is correct; otherwise, it will be "false".

*Note:* Consider the element before the first element and the element after the last element to be negative infinity.

**Input:**
- An integer array `arr[]` of length `n` where no two adjacent elements are the same.

**Output:**
- An integer representing the index of any one peak element in the array.

**Constraints:**
- The array contains distinct elements.
- No two adjacent elements are the same.

**Goal of the code:**
To efficiently find and return the index of any peak element in the given array.

**Approach:**
Utilize a binary search algorithm to find a peak element in logarithmic time. The idea is to leverage the properties of the array and the definition of a peak to narrow down the search space efficiently.

**Pseudocode/Algorithm:**
1. **Initial Checks:**
   - If the array has only one element, return index `0` since it's trivially a peak.
   - If the first element is greater than the second, return index `0`.
   - If the last element is greater than the second last, return index `n - 1`.

2. **Binary Search:**
   - Initialize `low` to `1` and `high` to `n - 2` (since the edge cases are already handled).
   - While `low` is less than or equal to `high`:
     - Calculate `mid` as `low + (high - low) / 2`.
     - If `arr[mid]` is greater than both `arr[mid - 1]` and `arr[mid + 1]`, return `mid` as it is a peak.
     - If `arr[mid]` is less than `arr[mid + 1]`, move `low` to `mid + 1` to search in the right half.
     - Otherwise, move `high` to `mid - 1` to search in the left half.

3. **Return Default:**
   - If no peak is found in the binary search loop (which is unlikely due to problem constraints), return index `0`.

**Time and Space Complexity:**
- **Time Complexity:** O(log n) due to the binary search approach.
- **Space Complexity:** O(1) as no additional space is used beyond a few variables.

**Additional Notes:**
- This approach ensures that we find a peak element efficiently without needing to examine every element in the array.
- The problem guarantees the existence of at least one peak due to the constraints provided.


## Problem Statement: 85
Given a binary matrix of order NxN, where each element is either 0 or 1 and all rows are sorted in non-decreasing order, find the row number with the maximum number of 1s. Additionally, determine the count of 1s in that row.

**Note:**
1. In case of a tie, return the smallest row number.
2. Row numbering starts from 0.

**Input:**
- An integer `N` representing the order of the matrix.
- A binary matrix `mat` of size `N x N` where each row is sorted in non-decreasing order.

**Output:**
- An array of two integers:
  - The index of the row with the maximum number of 1s.
  - The count of 1s in that row.

**Constraints:**
- The matrix is square with dimensions `N x N`.
- Each row is sorted in non-decreasing order.
- The matrix contains only binary values (0s and 1s).

**Goal of the code:**
Efficiently identify the row with the maximum number of 1s in a sorted binary matrix and return both the row index and the count of 1s.

**Approach:**
Leverage the sorted property of each row to efficiently determine the count of 1s using binary search. By finding the first occurrence of 1 in each row, we can calculate the number of 1s and track the row with the maximum count.

**Pseudocode/Algorithm:**
1. **Initialize Variables:**
   - `maxOnesCount` to 0: Tracks the maximum number of 1s found.
   - `maxRowIndex` to 0: Tracks the index of the row with the maximum number of 1s.

2. **Define Helper Function `firstOneIndex(row)`:**
   - Perform a binary search on the input `row` to find the index of the first occurrence of 1.
   - Return the index if found; otherwise, return -1.

3. **Iterate Through Each Row:**
   - For each row `i` in the matrix:
     - Use `firstOneIndex` to find the first occurrence of 1.
     - If a 1 is found (`index != -1`):
       - Calculate the number of 1s in the row as `N - index`.
       - If this count is greater than `maxOnesCount`:
         - Update `maxOnesCount` with the new count.
         - Update `maxRowIndex` with the current row index `i`.

4. **Return Result:**
   - Return an array containing `maxRowIndex` and `maxOnesCount`.

**Time and Space Complexity:**
- **Time Complexity:** O(N log N)
  - The binary search within each row operates in O(log N) time.
  - Iterating through all N rows results in a total time complexity of O(N log N).

- **Space Complexity:** O(1)
  - The algorithm uses a constant amount of extra space for variables.

**Additional Notes:**
- This approach efficiently utilizes the sorted property of each row to minimize the search time for the first occurrence of 1.
- In the event of multiple rows having the same number of 1s, the algorithm naturally returns the smallest row index due to the sequential iteration from the top of the matrix.


## Problem Statement: 86

Given a 2D integer matrix `mat[][]` of size `n x m`, where every row and column is sorted in increasing order, and an integer `x`, determine whether `x` is present in the matrix.

**Input:**
- An integer `n` representing the number of rows in the matrix.
- An integer `m` representing the number of columns in the matrix.
- A 2D integer matrix `mat` of size `n x m`, where each row and column is sorted in increasing order.
- An integer `x` representing the target value to search for.

**Output:**
- Return `true` if `x` is found in the matrix; otherwise, return `false`.

**Constraints:**
- The matrix dimensions `n` and `m` are positive integers.
- Each row and column in the matrix is sorted in increasing order.
- The matrix may contain duplicate elements.

**Goal of the code:**
Efficiently determine whether the target value `x` exists within the given row-wise and column-wise sorted matrix.

**Approach:**
Utilize a search strategy that leverages the sorted properties of the matrix to efficiently locate the target value. Start the search from the top-right corner of the matrix and iteratively narrow down the search space based on comparisons with the target value.

**Pseudocode/Algorithm:**
1. Initialize `i` to 0 (first row) and `j` to `m - 1` (last column).
2. While `i` is within the number of rows and `j` is non-negative:
   - If `mat[i][j]` equals `x`, return `true`.
   - If `mat[i][j]` is greater than `x`, decrement `j` to move left.
   - If `mat[i][j]` is less than `x`, increment `i` to move down.
3. If the loop terminates without finding `x`, return `false`.

**Time and Space Complexity:**
- **Time Complexity:** O(n + m), where `n` is the number of rows and `m` is the number of columns. In the worst case, the algorithm may need to traverse the entire first row and last column.
- **Space Complexity:** O(1), as the algorithm uses a constant amount of extra space.

**Additional Notes/Hints:**
- Starting the search from the top-right corner is strategic because it allows elimination of either a row or a column in each step:
  - If the current element is greater than the target, all elements in the current column below are also greater, so move left.
  - If the current element is less than the target, all elements in the current row to the left are also less, so move down.
- This approach ensures that each step brings us closer to the target while efficiently narrowing down the search space.


## Problem Statement: 87
Given a strictly sorted 2D matrix `mat[][]` of size `n x m` and an integer `x`, determine whether `x` is present in the matrix or not.

*Note:* In a strictly sorted matrix:
- Each row is sorted in strictly increasing order.
- The first element of the ith row (i ≠ 0) is greater than the last element of the (i-1)th row.

**Input:**
- An integer `n` representing the number of rows in the matrix.
- An integer `m` representing the number of columns in the matrix.
- A 2D integer matrix `mat` of size `n x m`, where each row is sorted in strictly increasing order, and the first element of each row is greater than the last element of the previous row.
- An integer `x` representing the target value to search for.

**Output:**
- Return `true` if `x` is found in the matrix; otherwise, return `false`.

**Constraints:**
- The matrix dimensions `n` and `m` are positive integers.
- Each row is sorted in strictly increasing order.
- The first element of each row is greater than the last element of the previous row.
- The matrix does not contain duplicate elements.

**Goal of the code:**
Efficiently determine whether the target value `x` exists within the given strictly sorted 2D matrix.

**Approach:**
Utilize a binary search strategy over the entire matrix by treating it as a flattened 1D array. Given the strict ordering properties, this approach allows for efficient searching.

**Pseudocode/Algorithm:**
1. Initialize `low` to 0 and `high` to `n * m - 1`.
2. While `low` is less than or equal to `high`:
   - Calculate `mid` as `(low + high) / 2`.
   - Determine the element at the `mid` position:
     - `row = mid / m`
     - `col = mid % m`
     - `mid_element = mat[row][col]`
   - If `mid_element` equals `x`, return `true`.
   - If `mid_element` is less than `x`, set `low = mid + 1`.
   - If `mid_element` is greater than `x`, set `high = mid - 1`.
3. If the loop terminates without finding `x`, return `false`.

**Time and Space Complexity:**
- **Time Complexity:** O(log(n * m)), where `n` is the number of rows and `m` is the number of columns. This is due to the binary search over the `n * m` elements.
- **Space Complexity:** O(1), as the algorithm uses a constant amount of extra space.

**Additional Notes/Hints:**
- Treating the 2D matrix as a flattened 1D array leverages the strict ordering properties, allowing for a straightforward binary search implementation.
- Ensure that the matrix adheres to the strict ordering properties to guarantee the correctness of this approach.

**Code Implementation:**
```java
class Solution {
    public boolean searchMatrix(int[][] mat, int x) {
        int n = mat.length;
        int m = mat[0].length;
        int low = 0, high = n * m - 1;
        
        while (low <= high) {
            int mid = (low + high) / 2;
            int row = mid / m;
            int col = mid % m;
            int mid_element = mat[row][col];
            
            if (mid_element == x) {
                return true;
            } else if (mid_element < x) {
                low = mid + 1;
            } else {
                high = mid - 1;
            }
        }
        return false;
    }
}


## Problem Statement: 88
You are given an array `stalls[]` containing unique integers that represent the positions of stalls along a straight line. Additionally, you are provided with an integer `k` denoting the number of aggressive cows. The task is to assign the cows to these stalls such that the minimum distance between any two cows is maximized.

**Input:**
- An integer `n` representing the number of stalls.
- An integer `k` representing the number of cows.
- An array `stalls[]` of size `n`, where each element denotes the position of a stall.

**Output:**
- A single integer representing the largest possible minimum distance between any two of the `k` cows.

**Constraints:**
- `2 ≤ n ≤ 100,000`
- `2 ≤ k ≤ n`
- `0 ≤ stalls[i] ≤ 1,000,000,000`
- All stall positions are unique.

**Goal of the code:**
Determine the maximum possible minimum distance between any two of the `k` aggressive cows when placed in the given stalls.

**Approach:**
To solve this problem efficiently, we can use a combination of sorting and binary search:

1. **Sorting:** First, sort the `stalls[]` array to arrange the stall positions in ascending order.
2. **Binary Search on Distance:** Use binary search to find the maximum possible minimum distance (`d`) between any two cows.
3. **Greedy Placement Check:** For each midpoint `d` during the binary search, use a greedy approach to check if it's feasible to place all `k` cows such that each cow is at least `d` distance apart.

**Pseudocode/Algorithm:**
1. Sort the `stalls[]` array.
2. Initialize `low` to 1 (minimum possible distance) and `high` to `stalls[n-1] - stalls[0]` (maximum possible distance).
3. While `low` is less than or equal to `high`:
   - Calculate `mid` as the average of `low` and `high`.
   - Check if it's possible to place `k` cows with at least `mid` distance apart:
     - Place the first cow in the first stall.
     - For each subsequent stall, if the distance from the last placed cow is greater than or equal to `mid`, place another cow.
     - If all `k` cows are placed successfully, update the result to `mid` and set `low` to `mid + 1` to try for a larger distance.
     - If not, set `high` to `mid - 1` to try for a smaller distance.
4. Return the result, which represents the largest minimum distance possible.

**Time and Space Complexity:**
- **Time Complexity:** O(n log d), where `n` is the number of stalls and `d` is the difference between the maximum and minimum stall positions. Sorting the stalls takes O(n log n), and the binary search combined with the feasibility check takes O(n log d).
- **Space Complexity:** O(1), as the algorithm uses a constant amount of extra space.

**Additional Notes/Hints:**
- The key insight is to use binary search on the minimum distance between cows and verify the feasibility of each distance using a greedy approach.
- This problem is a classic example of the "maximize the minimum distance" type, which is efficiently solved using binary search combined with a greedy placement strategy.


## Problem Statement: 89
Given an array `arr[]` of integers, where each element `arr[i]` represents the number of pages in the `i`-th book, and an integer `k` representing the number of students, allocate books to each student such that:

1. Each student receives at least one book.
2. Each student is assigned a contiguous sequence of books.
3. No book is assigned to more than one student.

The objective is to minimize the maximum number of pages assigned to any student. If a valid assignment is not possible, return `-1`.

**Input:**
- An integer `n` representing the number of books.
- An array `arr[]` of size `n`, where `arr[i]` denotes the number of pages in the `i`-th book.
- An integer `k` representing the number of students.

**Output:**
- A single integer representing the minimized value of the maximum number of pages assigned to any student. If a valid assignment is not possible, return `-1`.

**Constraints:**
- `1 ≤ n ≤ 10^5`
- `1 ≤ arr[i] ≤ 10^9`
- `1 ≤ k ≤ 10^5`

**Goal of the code:**
Determine the optimal way to allocate books to `k` students such that the maximum number of pages assigned to any student is minimized.

**Approach:**
To achieve the objective, we can utilize a binary search strategy combined with a greedy allocation method:

1. **Binary Search on Maximum Pages:**
   - Define the search range for the maximum number of pages (`low` to `high`):
     - `low`: The maximum number of pages in a single book (i.e., `max(arr)`), since no student can be assigned less than the largest book.
     - `high`: The total number of pages in all books (i.e., `sum(arr)`), representing the scenario where one student gets all the books.
   - Perform binary search within this range to find the minimal feasible maximum number of pages.

2. **Feasibility Check (Greedy Allocation):**
   - For a mid-value in the binary search, determine if it's possible to allocate books to `k` students such that no student receives more than `mid` pages:
     - Iterate through the books, cumulatively adding pages to the current student's allocation.
     - If adding another book exceeds `mid` pages, assign the current set of books to the current student and proceed to the next student.
     - If the number of students required exceeds `k`, the allocation isn't feasible for that `mid` value.

**Pseudocode/Algorithm:**
1. Initialize `low` to `max(arr)` and `high` to `sum(arr)`.
2. While `low` is less than or equal to `high`:
   - Calculate `mid` as the average of `low` and `high`.
   - Check if it's feasible to allocate books with `mid` as the maximum number of pages per student:
     - Initialize `required_students` to 1 and `current_pages` to 0.
     - For each book in `arr`:
       - If adding the book to `current_pages` exceeds `mid`:
         - Increment `required_students` by 1.
         - Set `current_pages` to the number of pages in the current book.
       - Otherwise, add the book's pages to `current_pages`.
     - If `required_students` is less than or equal to `k`, the allocation is feasible; otherwise, it's not.
   - If feasible, update the result to `mid` and set `high` to `mid - 1` to search for a potentially smaller maximum.
   - If not feasible, set `low` to `mid + 1` to search for a larger maximum.
3. If a valid allocation is found, return the result; otherwise, return `-1`.

**Time and Space Complexity:**
- **Time Complexity:** O(n log S), where `n` is the number of books and `S` is the sum of all pages. The binary search runs in O(log S) time, and for each mid-value, the feasibility check runs in O(n) time.
- **Space Complexity:** O(1), as the algorithm uses a constant amount of extra space.

**Additional Notes/Hints:**
- If `k` is greater than `n`, it's impossible to allocate at least one book to each student, and the function should return `-1`.
- This problem is a classic example of using binary search on the answer space combined with a greedy approach to verify feasibility.


## Problem Statement: 90
Dilpreet wants to paint his dog's home, which consists of `n` boards of varying lengths. The length of the `i`-th board is given by `arr[i]`, where `arr[]` is an array of `n` integers. He has hired `k` painters, and each painter takes 1 unit time to paint 1 unit length of the board. The goal is to determine the minimum time required to paint all the boards if all painters start simultaneously, with the constraint that each painter can only paint continuous sections of boards.

**Input:**
- An integer `n` representing the number of boards.
- An array `arr[]` of `n` integers, where `arr[i]` denotes the length of the `i`-th board.
- An integer `k` representing the number of painters.

**Output:**
- An integer representing the minimum time required to paint all boards.

**Constraints:**
- 1 ≤ `n` ≤ 10^5
- 1 ≤ `arr[i]` ≤ 10^9
- 1 ≤ `k` ≤ 10^5

**Goal of the code:**
To allocate the painting job among `k` painters such that the maximum time taken by any painter is minimized.

**Approach:**
The problem can be solved using a binary search combined with a feasibility check. The idea is to determine the minimum possible maximum time required by any painter to paint the boards.

**Pseudocode/Algorithm:**
1. **Initialize Variables:**
   - Set `low` to the maximum length in `arr[]` (since at least one painter has to paint the longest board).
   - Set `high` to the sum of all lengths in `arr[]` (if one painter were to paint all boards).

2. **Binary Search:**
   - While `low` is less than or equal to `high`:
     - Calculate `mid` as the average of `low` and `high`.
     - Check if it's feasible to paint all boards within `mid` time using at most `k` painters:
       - Initialize `requiredPainters` to 1 and `currentTime` to 0.
       - For each board length in `arr[]`:
         - If adding the current board's length to `currentTime` exceeds `mid`:
           - Increment `requiredPainters` by 1.
           - Set `currentTime` to the current board's length.
         - Otherwise, add the current board's length to `currentTime`.
       - If `requiredPainters` is less than or equal to `k`, set `high` to `mid - 1` (try for a smaller maximum time).
       - Otherwise, set `low` to `mid + 1` (increase the allowed time).

3. **Result:**
   - The minimum time required is stored in `low` after the loop terminates.

**Time and Space Complexity:**
- **Time Complexity:** O(n log S), where `n` is the number of boards and `S` is the sum of all board lengths. The binary search runs in O(log S) time, and for each mid value, we check feasibility in O(n) time.
- **Space Complexity:** O(1), as we are using a constant amount of extra space.

**Additional Notes/Hints:**
- This problem is analogous to the "Allocate Minimum Pages" problem, where the goal is to allocate books to students such that the maximum number of pages assigned to a student is minimized.
- Understanding the feasibility function is crucial. It checks whether it's possible to allocate boards to painters such that no painter paints more than `mid` units of board length.


## Problem Statement: 91
Given an array `arr` of `N` non-negative integers and an integer `K`, split the array into `K` contiguous subarrays such that the maximum sum among these subarrays is minimized. Return this minimized maximum subarray sum.

**Input:**
- An integer array `arr` of size `N`.
- An integer `K` denoting the number of subarrays.

**Output:**
- A single integer representing the minimized largest sum among the `K` subarrays.

**Constraints:**
- `1 ≤ N ≤ 1000`
- `1 ≤ K ≤ min(50, N)`
- `0 ≤ arr[i] ≤ 10^6`

**Goal of the code:**
To determine the minimum possible value of the maximum sum among `K` contiguous subarrays into which the given array can be split.

**Approach:**
Utilize a binary search algorithm to efficiently find the minimal possible maximum subarray sum:

1. **Initialize Binary Search Bounds:**
   - Set `low` to the maximum element in the array (smallest possible maximum subarray sum).
   - Set `high` to the sum of all elements in the array (largest possible maximum subarray sum).

2. **Binary Search Execution:**
   - While `low` is less than or equal to `high`:
     - Calculate `mid` as the average of `low` and `high`.
     - Check if it's feasible to split the array into `K` or fewer subarrays with each subarray sum not exceeding `mid`:
       - If feasible, update the result to `mid` and set `high` to `mid - 1` to search for a smaller maximum sum.
       - If not feasible, set `low` to `mid + 1` to allow larger subarray sums.

3. **Feasibility Check Function (`canSplit`):**
   - Iterate through the array, accumulating the sum of elements.
   - If adding the next element exceeds `mid`, increment the subarray count and start a new subarray.
   - If the subarray count exceeds `K`, return `false` (indicating the current `mid` is too small).
   - If the loop completes without exceeding `K` subarrays, return `true`.

**Pseudocode/Algorithm:**
```java
function splitArray(arr, N, K):
    define function canSplit(arr, K, mid):
        count = 1
        currentSum = 0
        for each number in arr:
            if currentSum + number > mid:
                count += 1
                currentSum = number
                if count > K:
                    return false
            else:
                currentSum += number
        return true

    maxElement = maximum value in arr
    totalSum = sum of all elements in arr
    low = maxElement
    high = totalSum
    result = high

    while low <= high:
        mid = (low + high) // 2
        if canSplit(arr, K, mid):
            result = mid
            high = mid - 1
        else:
            low = mid + 1

    return result


## Problem Statement: 92
Given three integers `M`, `N`, and `K`, consider a grid of size `M x N` where the element at position `(i, j)` is defined as `mat[i][j] = i * j` (1-based index). The task is to find the `K`-th smallest element in this multiplication table.

**Input:**
- Three integers:
  - `M`: Number of rows in the multiplication table.
  - `N`: Number of columns in the multiplication table.
  - `K`: The position of the desired smallest element when the table's elements are sorted in ascending order.

**Output:**
- A single integer representing the `K`-th smallest element in the `M x N` multiplication table.

**Constraints:**
- `1 ≤ M, N ≤ 30,000`
- `1 ≤ K ≤ M * N`

**Goal of the code:**
To efficiently determine the `K`-th smallest number in an `M x N` multiplication table without explicitly constructing the table.

**Approach:**
Utilize a binary search strategy to identify the `K`-th smallest element:

1. **Define Search Range:**
   - The smallest possible value in the table is `1` (i.e., `1 * 1`).
   - The largest possible value is `M * N` (i.e., `M * N`).

2. **Binary Search Execution:**
   - While the search range is valid (`low` ≤ `high`):
     - Compute the midpoint `mid` of the current range.
     - Determine the count of elements in the table that are less than or equal to `mid`.
     - If this count is less than `K`, adjust the search range to consider larger values (`low = mid + 1`).
     - If the count is greater than or equal to `K`, adjust the range to consider smaller values (`high = mid - 1`).

3. **Counting Function:**
   - For each row `i` from `1` to `M`, calculate the number of elements in that row that are less than or equal to `mid` using `min(mid // i, N)`.
   - Sum these counts to get the total number of elements in the table that are less than or equal to `mid`.

**Pseudocode/Algorithm:**
```java
function findKthNumber(M, N, K):
    low = 1
    high = M * N

    while low <= high:
        mid = (low + high) // 2
        count = 0

        for i from 1 to M:
            count += min(mid // i, N)

        if count < K:
            low = mid + 1
        else:
            high = mid - 1

    return low


## Problem Statement: 93
You are tasked with transporting packages from one port to another within a specified number of days, `D`. The weights of the packages are given in an array `weights`, where the `i-th` package has a weight of `weights[i]`. Each day, you load the ship with packages in the order they appear in the array, ensuring that the total weight does not exceed the ship's maximum weight capacity. Your goal is to determine the minimum weight capacity of the ship that allows all packages to be shipped within `D` days.

**Input**
- An array `weights` of length `n`, where `weights[i]` represents the weight of the `i-th` package.
- An integer `D` representing the number of days within which all packages must be shipped.

**Output**
- An integer representing the minimum weight capacity of the ship required to ship all packages within `D` days.

**Constraints**
- The number of packages, `n`, and the number of days, `D`, are positive integers.
- The weight of each package is a positive integer.
- It is guaranteed that it is possible to ship all packages within `D` days.

**Goal of the Code**
Determine the minimum ship capacity that allows all packages to be shipped within the given number of days, adhering to the order of packages as they appear in the input array.

**Approach**
To solve this problem, we can use a binary search approach to find the minimum ship capacity:

1. **Determine the Search Range:**
   - The minimum possible capacity (`left`) is the weight of the heaviest package, since the ship must be able to carry at least that package.
   - The maximum possible capacity (`right`) is the sum of all package weights, representing the scenario where the ship carries all packages in one day.

2. **Binary Search:**
   - While `left` is less than `right`:
     - Calculate the midpoint `mid` of the current range.
     - Use a helper function to determine if it's possible to ship all packages within `D` days with a ship capacity of `mid`.
     - If possible, adjust the `right` boundary to `mid` to explore smaller capacities.
     - If not possible, adjust the `left` boundary to `mid + 1` to increase the capacity.

3. **Helper Function (`canShip`):**
   - Simulate the shipping process:
     - Initialize `days` to 1 and `currentLoad` to 0.
     - Iterate through each weight in the `weights` array:
       - If adding the current weight `w` to `currentLoad` exceeds the `capacity`, increment the `days` count and reset `currentLoad` to 0.
       - Add the current weight `w` to `currentLoad`.
     - After processing all weights, check if the total `days` used is less than or equal to `D`.

**Time and Space Complexity**
- **Time Complexity:** The binary search runs in `O(log(sum(weights) - max(weights)))` iterations. Each iteration involves a linear scan of the `weights` array to determine if the current capacity can ship all packages within `D` days, which takes `O(n)` time. Thus, the overall time complexity is `O(n log(sum(weights) - max(weights)))`.

- **Space Complexity:** The space complexity is `O(1)` since we are using a constant amount of extra space.

**Additional Notes/Hints**
- This problem can be approached using binary search on the potential ship capacities because the feasibility of shipping within `D` days is a monotonic function with respect to the ship's capacity.
- Ensuring that the ship's capacity is at least the weight of the heaviest package is crucial, as any capacity less than this would make it impossible to ship that package. 

**Pseudocode/Algorithm**
```java
function shipWithinDays(weights, D):
 left = max(weights) 
 right = sum(weights)

while left < right:
    mid = (left + right) / 2
    if canShip(weights, D, mid):
        right = mid
    else:
        left = mid + 1

return left

function canShip(weights, D, capacity): 
days = 1 
currentLoad = 0 
for weight in weights:
 if currentLoad + weight > capacity: 
 days += 1 
 currentLoad = 0 
 
 currentLoad += weight 
 return days <= D




## Problem Statement: 94
Given two sorted arrays `a[]` and `b[]`, the task is to find and return the median of the combined array after merging them into a single sorted array.

**Input**
- Two sorted arrays `a[]` and `b[]` of sizes `n` and `m` respectively.

**Output**
- A single number representing the median of the merged sorted array.

**Constraints**
- The arrays `a[]` and `b[]` are non-empty and sorted in non-decreasing order.
- The sizes of the arrays, `n` and `m`, are positive integers.

**Goal of the Code**
To efficiently find the median of two sorted arrays without fully merging them, achieving a time complexity better than O(n + m).

**Approach**
To find the median of two sorted arrays without merging them, we can utilize a binary search approach. The idea is to partition the arrays such that the left halves contain the smaller elements and the right halves contain the larger elements. By ensuring the partitions are balanced, the median can be determined from the maximum of the left halves and the minimum of the right halves.

**Steps:**
1. **Ensure the First Array is Smaller:** If array `a` is larger than array `b`, swap them to make sure `a` is the smaller array. This simplifies the binary search on the smaller array.

2. **Binary Search on the Smaller Array:** Perform binary search on array `a` to find the correct partition:
   - Calculate `i` (the partition index for `a`) and `j` (the partition index for `b`) such that `i + j = (n + m + 1) / 2`.
   - Determine the elements around the partition:
     - `aLeft`: Element just before the partition in `a` (or negative infinity if partition is at the start).
     - `aRight`: Element just after the partition in `a` (or positive infinity if partition is at the end).
     - `bLeft`: Element just before the partition in `b`.
     - `bRight`: Element just after the partition in `b`.
   - Check if the partitions are correct:
     - If `aLeft <= bRight` and `bLeft <= aRight`, the correct partition is found.
     - If `aLeft > bRight`, move the partition in `a` to the left.
     - If `bLeft > aRight`, move the partition in `a` to the right.

3. **Calculate the Median:**
   - If the total number of elements is odd, the median is the maximum of the left partition elements.
   - If even, the median is the average of the maximum of the left partition elements and the minimum of the right partition elements.

**Time and Space Complexity:**

**Time Complexity:** O(log(min(n, m))), where n and m are the sizes of arrays a and b respectively. The binary search is performed on the smaller array, ensuring logarithmic time complexity.
**Space Complexity:** O(1), as no additional space proportional to the input size is used.

**Pseudocode/Algorithm**
```java
public double findMedianSortedArrays(int[] a, int[] b) {
    if (a.length > b.length) {
        return findMedianSortedArrays(b, a);
    }
    int n = a.length;
    int m = b.length;
    int low = 0, high = n;
    while (low <= high) {
        int i = (low + high) / 2;
        int j = (n + m + 1) / 2 - i;
        int aLeft = (i == 0) ? Integer.MIN_VALUE : a[i - 1];
        int aRight = (i == n) ? Integer.MAX_VALUE : a[i];
        int bLeft = (j == 0) ? Integer.MIN_VALUE : b[j - 1];
        int bRight = (j == m) ? Integer.MAX_VALUE : b[j];
        if (aLeft <= bRight && bLeft <= aRight) {
            if ((n + m) % 2 == 0) {
                return (Math.max(aLeft, bLeft) + Math.min(aRight, bRight)) / 2.0;
            } else {
                return Math.max(aLeft, bLeft);
            }
        } else if (aLeft > bRight) {
            high = i - 1;
        } else {
            low = i + 1;
        }
    }
    throw new IllegalArgumentException("Input arrays are not sorted.");
}


## Problem Statement: 95
Koko loves to eat bananas. Given an array `arr[]` where each element represents a pile of bananas, and an integer `k` representing the number of hours Koko has to finish all the piles, determine the minimum number of bananas `s` Koko must eat per hour to finish all the bananas within `k` hours. Each hour, Koko chooses a pile and eats `s` bananas from it. If the pile has fewer than `s` bananas, she consumes the entire pile for that hour and doesn't eat any more bananas during that hour.

**Input**
- An integer array `arr[]` where `arr[i]` represents the number of bananas in the `i`-th pile.
- An integer `k` representing the total hours available to finish all the piles.

**Output**
- An integer `s` representing the minimum number of bananas Koko must eat per hour to finish all the bananas within `k` hours.

**Constraints**
- The length of `arr[]` is `n` where `1 ≤ n ≤ 10^4`.
- Each element in `arr[]` is a positive integer representing the number of bananas in that pile.
- `1 ≤ k ≤ 10^9`.

**Goal of the Code**
To find the minimum eating speed `s` (bananas per hour) such that Koko can consume all the bananas in all piles within `k` hours.

**Approach**
To determine the minimum number of bananas Koko must eat per hour, we can utilize a binary search approach on the possible eating speeds.

**Steps:**
1. **Define the Search Range:**
   - The minimum possible eating speed is `1` banana per hour.
   - The maximum possible eating speed is the size of the largest pile, `max(arr)`, since eating faster than this would mean Koko finishes the largest pile in less than an hour.

2. **Binary Search on Eating Speed:**
   - Initialize `low` to `1` and `high` to `max(arr)`.
   - While `low` is less than `high`:
     - Calculate `mid` as the average of `low` and `high`.
     - Determine if Koko can finish all piles at an eating speed of `mid` bananas per hour by calculating the total hours required:
       - For each pile, the hours needed to finish that pile is `(pile + mid - 1) // mid`, which is equivalent to `ceil(pile / mid)`.
       - Sum these hours across all piles to get the total hours required.
     - If the total hours required is less than or equal to `k`, then `mid` is a feasible eating speed, and we search the lower half by setting `high` to `mid`.
     - If the total hours required is greater than `k`, then `mid` is too slow, and we search the upper half by setting `low` to `mid + 1`.

3. **Result:**
   - After the binary search completes, `low` will be the minimum eating speed that allows Koko to finish all the bananas within `k` hours.

**Time Complexity:** O(n log m), where n is the number of piles and m is the maximum number of bananas in a pile. The binary search runs in O(log m) time, and for each candidate eating speed, we compute the total hours required in O(n) time.

**Space Complexity:** O(1), as we use a constant amount of extra space.

**Pseudocode/Algorithm**
```java
public int minEatingSpeed(int[] piles, int h) {
    int low = 1;
    int high = Arrays.stream(piles).max().getAsInt();

    while (low < high) {
        int mid = (low + high) / 2;
        int totalHours = 0;

        for (int pile : piles) {
            totalHours += (pile + mid - 1) / mid; // Equivalent to Math.ceil(pile / mid)
        }

        if (totalHours <= h) {
            high = mid; // Try a slower speed
        } else {
            low = mid + 1; // Need a faster speed
        }
    }

    return low;
}


## Problem Statement: 96 
**Minimum Days to Make M Bouquets**
You have a row of flowers, where each flower blooms after a specific day. The array `arr` represents the blooming schedule: `arr[i]` is the day the flower at position `i` will bloom. To create a bouquet, you need to collect `k` adjacent bloomed flowers. Each flower can only be used in one bouquet.

Your goal is to find the minimum number of days required to make exactly `m` bouquets. If it is not possible to make `m` bouquets with the given arrangement, return `-1`.

**Input**
- An integer array `arr` where `arr[i]` represents the day the flower at position `i` will bloom.
- Two integers `m` and `k` representing the number of bouquets needed and the number of adjacent flowers required per bouquet, respectively.

**Output**
- An integer representing the minimum number of days required to make `m` bouquets. If it's not possible, return `-1`.

**Constraints**
- `1 <= arr.length <= 10^5`
- `1 <= arr[i] <= 10^9`
- `1 <= m <= 10^6`
- `1 <= k <= arr.length`

**Goal of the Code**
Determine the minimum number of days required to make exactly `m` bouquets, each consisting of `k` adjacent bloomed flowers. If it's not possible to form `m` bouquets, return `-1`.

**Approach**
1. **Binary Search on Days:** Utilize binary search to determine the minimum number of days required to form `m` bouquets.
2. **Feasibility Check:** For each mid-point in the binary search, check if it's possible to form `m` bouquets by that day using a helper function.
3. **Helper Function:** Traverse the `arr` array to count the number of bouquets that can be formed by the given day. Reset the count when a flower hasn't bloomed by that day.

**Pseudocode/Algorithm**
1. **Initialize Binary Search Parameters:**
   - `low` = minimum value in `arr`
   - `high` = maximum value in `arr`
   - `result` = -1

2. **Binary Search Loop:**
   - While `low` <= `high`:
     - Calculate `mid` = (low + high) / 2
     - If `canMakeBouquets(arr, m, k, mid)`:
       - Set `result` = `mid`
       - Update `high` = `mid` - 1
     - Else:
       - Update `low` = `mid` + 1

3. **Helper Function `canMakeBouquets(arr, m, k, days)`:**
   - Initialize `bouquets` = 0 and `flowers` = 0
   - For each day in `arr`:
     - If `day` <= `days`:
       - Increment `flowers`
     - Else:
       - Calculate `bouquets` += `flowers` // `k`
       - Reset `flowers` = 0
   - Add any remaining bouquets from the last segment
   - Return `bouquets` >= `m`

4. **Return Result:**
   - If `result` remains -1, return -1
   - Otherwise, return `result`

**Time and Space Complexity**

- **Time Complexity:** O(n log D), where `n` is the length of `arr` and `D` is the range of days between the minimum and maximum values in `arr`. The binary search runs in O(log D) and for each mid-point, we traverse the array in O(n) to count possible bouquets.
- **Space Complexity:** O(1), as we are using a constant amount of extra space.

**Additional Notes**
- If `m * k` is greater than the length of `arr`, it's impossible to form `m` bouquets, and the function should immediately return `-1`.
- This problem is a classic example of using binary search in combination with a feasibility function to optimize the search for a minimum or maximum value under given constraints.

## Problem Statement: 97
An **ugly number** is a positive integer whose prime factors are limited to 2, 3, and 5. Given an integer `n`, return the `n`th ugly number.

**Constraints:**
- `1 <= n <= 1690`

**Approach**
To find the `n`th ugly number efficiently, we can use a dynamic programming approach with three pointers. The idea is to generate ugly numbers in ascending order by multiplying previously found ugly numbers by 2, 3, and 5, and ensuring no duplicates.

**Algorithm:**
1. Initialize an array `uglyNumbers` of size `n` to store the first `n` ugly numbers.
2. Set the first ugly number: `uglyNumbers[0] = 1`.
3. Initialize three pointers `p2`, `p3`, and `p5` to track the position for multiples of 2, 3, and 5 respectively.
4. For each position from 1 to `n-1`:
   - Compute the next potential ugly numbers by multiplying the values at pointers `p2`, `p3`, and `p5` by 2, 3, and 5 respectively.
   - The next ugly number is the minimum of these values.
   - Increment the pointer(s) that match the chosen minimum value to avoid duplicates.
5. The last element in `uglyNumbers` will be the `n`th ugly number.

**Time Complexity:** O(n)  
**Space Complexity:** O(n)

**Code Implementation**
```java
public class Solution {
    public int nthUglyNumber(int n) {
        int[] uglyNumbers = new int[n];
        uglyNumbers[0] = 1;
        int p2 = 0, p3 = 0, p5 = 0;
        int nextMultipleOf2 = 2;
        int nextMultipleOf3 = 3;
        int nextMultipleOf5 = 5;

        for (int i = 1; i < n; i++) {
            int nextUgly = Math.min(nextMultipleOf2, Math.min(nextMultipleOf3, nextMultipleOf5));
            uglyNumbers[i] = nextUgly;

            if (nextUgly == nextMultipleOf2) {
                p2++;
                nextMultipleOf2 = uglyNumbers[p2] * 2;
            }
            if (nextUgly == nextMultipleOf3) {
                p3++;
                nextMultipleOf3 = uglyNumbers[p3] * 3;
            }
            if (nextUgly == nextMultipleOf5) {
                p5++;
                nextMultipleOf5 = uglyNumbers[p5] * 5;
            }
        }
        return uglyNumbers[n - 1];
    }
}


## Problem Statement: 98
Given an array of size `n` containing positive integers and a number `k`, the task is to find the k-th smallest absolute difference between any two distinct elements in the array. The absolute difference between values at indices `i` and `j` is defined as `|a[i] - a[j]|`.

**Input:**
- An array `arr` of size `n` containing positive integers.
- An integer `k`.

**Output:**
- An integer representing the k-th smallest absolute difference among all possible pairs in the array.

**Constraints:**
- `2 ≤ n ≤ 10^5`
- `1 ≤ arr[i] ≤ 10^9`
- `1 ≤ k ≤ n*(n-1)/2`

**Goal of the code:**
Determine the k-th smallest absolute difference between any two distinct elements in the given array.

**Approach:**
1. **Sorting:** Sort the array to facilitate efficient calculation of differences.
2. **Binary Search:** Use binary search to find the minimum possible absolute difference that could be the k-th smallest.
3. **Counting Pairs:** For a mid value in binary search, count how many pairs have an absolute difference less than or equal to mid.
4. **Adjust Search Range:** Based on the count, adjust the binary search range to converge on the k-th smallest difference.

**Pseudocode/Algorithm:**
1. Sort the array arr.
2. Initialize low to 0 and high to the difference between the maximum and minimum elements in arr.
3. While low is less than high: a. Set mid to the average of low and high. b. Initialize count to 0 and j to 0. c. For each i from 0 to n-1: i. Increment j while j < n and arr[j] - arr[i] <= mid. ii. Add j - i - 1 to count. d. If count is greater than or equal to k, set high to mid. e. Else, set low to mid + 1.
4. Return low as the k-th smallest absolute difference.

**Time and Space Complexity:**
- **Time Complexity:** O(n log n) for sorting the array, and O(n log D) for the binary search and pair counting, where D is the difference between the maximum and minimum elements. Overall: O(n log n + n log D).
- **Space Complexity:** O(1) additional space, not including input storage.

**Additional Notes/Hints:**
- Efficient counting of pairs with differences less than or equal to a given value is crucial for reducing the time complexity.
- Understanding the properties of sorted arrays can help in optimizing the pair counting process.

## Problem Statement: 99
Given an array of integers `nums` and an integer `threshold`, the task is to find the smallest positive integer divisor such that the sum of the division results of all elements in the array, rounded up to the nearest integer, is less than or equal to the `threshold`.

**Input:**
- An array `nums` of size `n` containing positive integers.
- An integer `threshold`.

**Output:**
- An integer representing the smallest divisor that ensures the sum of the division results is less than or equal to the `threshold`.

**Constraints:**
- `1 ≤ n ≤ 5 * 10^4`
- `1 ≤ nums[i] ≤ 10^6`
- `n ≤ threshold ≤ 10^6`

**Goal of the code:**
Determine the smallest positive integer divisor such that the sum of the division results of all elements in the array, rounded up to the nearest integer, does not exceed the given `threshold`.

**Approach:**
1. **Binary Search:** Utilize binary search to efficiently find the smallest divisor.
2. **Define Search Range:** Set the initial search range for the divisor between 1 and the maximum value in the array.
3. **Calculate Sum for Midpoint Divisor:** For each midpoint divisor in the current search range, compute the sum of the division results of all elements, rounding up each result to the nearest integer.
4. **Adjust Search Range:** If the computed sum is less than or equal to the `threshold`, adjust the search range to explore smaller divisors; otherwise, explore larger divisors.

**Pseudocode/Algorithm:**
1. Initialize left to 1 and right to the maximum value in nums.
2. While left is less than or equal to right: a. Calculate mid as the average of left and right. b. Initialize sum to 0. c. For each number num in nums: i. Add the result of (num + mid - 1) // mid to sum (this computes the ceiling of num / mid). d. If sum is less than or equal to threshold: i. Set right to mid - 1 (try smaller divisors). e. Else: i. Set left to mid + 1 (try larger divisors).
3. Return left as the smallest valid divisor.

**Time and Space Complexity:**
- **Time Complexity:** O(n log m), where `n` is the number of elements in `nums` and `m` is the maximum value in `nums`. The binary search runs in O(log m) time, and for each midpoint, calculating the sum takes O(n) time.
- **Space Complexity:** O(1), as the algorithm uses a constant amount of extra space.

**Additional Notes/Hints:**
- The division results are rounded up to the nearest integer, which can be efficiently computed using integer arithmetic: `(num + divisor - 1) // divisor`.
- Binary search is effective here due to the monotonic nature of the sum with respect to the divisor: as the divisor increases, the sum of the division results decreases.


## Problem Statement: 100
Given a 2D array (matrix) `mat[][]`, identify a peak element. An element is considered a peak if it is greater than or equal to its four immediate neighbors: left, right, top, and bottom. For elements on the boundaries or corners of the matrix, any missing neighbors are treated as having a value of negative infinity. Note that a peak element is not necessarily the global maximum; multiple peak elements can exist within the matrix.

**Input:**
- A 2D array `mat` of dimensions `n x m` containing positive integers.

**Output:**
- The coordinates `(i, j)` of any one peak element in the matrix.

**Constraints:**
- The matrix dimensions `n` and `m` are such that `1 ≤ n, m ≤ 1000`.
- Each element in the matrix is a positive integer.

**Goal of the Code:**
Implement an efficient algorithm to locate any peak element in the given 2D matrix.

**Approach:**
Utilize a **divide and conquer** strategy, similar to binary search, to efficiently find a peak element:

1. **Select a Middle Column:**
   - Choose the middle column of the current submatrix.

2. **Find the Global Maximum in the Middle Column:**
   - Scan the middle column to find the maximum element. Let its position be `(max_row, mid_col)`.

3. **Compare the Maximum Element with Its Neighbors:**
   - Check the elements to the left and right of `(max_row, mid_col)`.
     - If the middle column's maximum is greater than or equal to both neighbors (or if the neighbor doesn't exist), then it's a peak element.
     - If the left neighbor is greater, recursively apply the algorithm to the left submatrix (columns before `mid_col`).
     - If the right neighbor is greater, recursively apply the algorithm to the right submatrix (columns after `mid_col`).

4. **Repeat the Process:**
   - Continue the above steps on the selected submatrix until a peak element is found.


**Time and Space Complexity:**

- **Time Complexity:** O(n log m), where `n` is the number of rows and `m` is the number of columns. This efficiency is achieved by reducing the problem size by half in each step (log m) and scanning a column to find the maximum element (n).

- **Space Complexity:** O(1), as the algorithm uses a constant amount of extra space.

**Additional Notes/Hints:**

- This approach ensures that a peak element is found without examining every element in the matrix, making it more efficient than a brute-force search.
- Understanding the properties of peak elements and their neighbors is crucial for implementing the divide and conquer strategy effectively.

**Pseudocode/Algorithm:**
```java
public class PeakElement2D {

    // Function to find a peak element in the matrix
    public static int[] findPeakElement(int[][] mat) {
        return findPeakRec(mat, 0, mat[0].length - 1);
    }

    // Recursive function to locate a peak element
    private static int[] findPeakRec(int[][] mat, int startCol, int endCol) {
        int midCol = (startCol + endCol) / 2;
        int maxRow = findMaxRow(mat, midCol);

        // Check if the middle column's maximum is a peak
        if ((midCol == 0 || mat[maxRow][midCol] >= mat[maxRow][midCol - 1]) &&
            (midCol == mat[0].length - 1 || mat[maxRow][midCol] >= mat[maxRow][midCol + 1])) {
            return new int[]{maxRow, midCol};
        }

        // If the left neighbor is greater, search the left submatrix
        if (midCol > 0 && mat[maxRow][midCol - 1] > mat[maxRow][midCol]) {
            return findPeakRec(mat, startCol, midCol - 1);
        }

        // Otherwise, search the right submatrix
        return findPeakRec(mat, midCol + 1, endCol);
    }

    // Helper function to find the row index of the maximum element in a given column
    private static int findMaxRow(int[][] mat, int col) {
        int maxRow = 0;
        for (int i = 1; i < mat.length; i++) {
            if (mat[i][col] > mat[maxRow][col]) {
                maxRow = i;
            }
        }
        return maxRow;
    }

    public static void main(String[] args) {
        int[][] mat = {
            {10, 8, 10, 10},
            {14, 13, 12, 11},
            {15, 9, 11, 21},
            {16, 17, 19, 20}
        };

        int[] peak = findPeakElement(mat);
        System.out.println("Peak element found at: (" + peak[0] + ", " + peak[1] + ")");
        System.out.println("Peak element value: " + mat[peak[0]][peak[1]]);
    }
}



## Problem Statement: 101
Given a bitonic sequence `arr` of `n` distinct integers and an integer `x`, write a program to find the index of `x` in the sequence in `O(log n)` time.
A **bitonic sequence** is a sequence of numbers that is first strictly increasing and then, after a certain point, strictly decreasing.

**Input:**
- An integer array `arr` representing the bitonic sequence.
- An integer `x` representing the target value to search for.

**Output:**
- The index of `x` in the array `arr` if present; otherwise, `-1`.

**Constraints:**
- `3 <= n <= 10^5`
- `1 <= arr[i], x <= 10^8`
- The array `arr` always contains a bitonic point.
- All elements in `arr` are distinct.

**Goal of the code:**
Efficiently locate the index of the target value `x` in a bitonic array using a logarithmic time complexity approach.

**Approach:**
1. **Find the Peak Element:**
   - Utilize a modified binary search to identify the peak (maximum) element in the array, which divides the array into increasing and decreasing parts.

2. **Search in the Increasing Part:**
   - Perform a binary search on the subarray from the start up to the peak.

3. **Search in the Decreasing Part:**
   - If the element is not found in the increasing part, perform a binary search on the subarray from the peak to the end, adjusting for the descending order.

**Pseudocode/Algorithm:**
1. **Finding the Peak Element:**
   - Initialize `low` to 0 and `high` to `n - 1`.
   - While `low` <= `high`:
     - Calculate `mid` as `(low + high) / 2`.
     - If `arr[mid]` is greater than both `arr[mid - 1]` and `arr[mid + 1]`, return `mid` as the peak.
     - If `arr[mid]` is greater than `arr[mid - 1]`, set `low` to `mid + 1`.
     - Else, set `high` to `mid - 1`.

2. **Binary Search in Ascending Order:**
   - Perform standard binary search on the subarray from index 0 to the peak index.

3. **Binary Search in Descending Order:**
   - Perform binary search on the subarray from the peak index + 1 to `n - 1`, adjusting comparisons for descending order.

**Time and Space Complexity:**

- **Time Complexity:** `O(log n)` for finding the peak element, and `O(log n)` for each binary search in the subarrays, resulting in an overall time complexity of `O(log n)`.
- **Space Complexity:** `O(1)`, as the algorithm uses a constant amount of extra space.

**Additional Notes/Hints:**
- Ensure to handle edge cases where the array is strictly increasing or strictly decreasing.
- The peak element serves as a pivot, allowing the application of binary search on both subarrays efficiently.


## Problem Statement:102
Implement the Bubble Sort algorithm to sort a given array `arr[]` in ascending order.

**Input:**
- An integer array `arr[]` of size `n` containing unsorted elements.

**Output:**
- The array `arr[]` sorted in non-decreasing order.

**Constraints:**
- `1 <= n <= 10^3`
- `-10^3 <= arr[i] <= 10^3`

**Goal of the code:**
To sort the input array `arr[]` using the Bubble Sort algorithm.

**Approach:**
1. **Iterate through the array multiple times:**
   - For each element in the array, compare it with the next element.
   - If the current element is greater than the next element, swap them.

2. **Optimization with a flag:**
   - Use a boolean flag `swapped` to monitor if any elements were swapped during a pass.
   - If no elements were swapped, the array is already sorted, and the algorithm can terminate early.

**Pseudocode/Algorithm:**
bubbleSort(arr):
 n = length of arr
  for i from 0 to n-1: 
  swapped = false 
  for j from 0 to n-i-2:
   if arr[j] > arr[j+1]:
    swap arr[j] and arr[j+1]
     swapped = true      
     if not swapped: break

**Time and Space Complexity:**
- **Time Complexity:**
  - Worst-case: O(n^2)
  - Best-case: O(n) when the array is already sorted
  - Average-case: O(n^2)

- **Space Complexity:**
  - O(1), as the algorithm sorts the array in place without requiring additional storage.

**Additional Notes/Hints:**
- Bubble Sort is a simple comparison-based sorting algorithm.
- It's not suitable for large datasets due to its quadratic time complexity.
- The algorithm is stable, meaning it preserves the relative order of equal elements.
- An optimized version stops the algorithm if the inner loop didn't cause any swap, indicating that the array is already sorted.


## Problem Statement:103
Implement the **Selection Sort** algorithm to sort a given array `arr[]` in ascending order.

**Input:**
- An integer array `arr[]` of size `n` containing unsorted elements.

**Output:**
- The array `arr[]` sorted in non-decreasing order.

**Constraints:**
- `1 <= n <= 10^3`
- `-10^3 <= arr[i] <= 10^3`

**Goal of the Code:**
To sort the input array `arr[]` using the Selection Sort algorithm.

**Approach:**
1. **Iterate through the array:**
   - For each position `i` from `0` to `n-2`, assume the element at `i` is the minimum.
   - Search the remaining unsorted portion of the array to find the actual minimum element.
   - If a smaller element is found, swap it with the element at position `i`.

2. **Repeat the process:**
   - Move to the next position and repeat the process until the entire array is sorted.

**Pseudocode/Algorithm:**
selectionSort(arr):
 n = length of arr 
 for i from 0 to n-1: 
 min_idx = i 
 for j from i+1 to n: 
 if arr[j] < arr[min_idx]: 
 min_idx = j
  if min_idx != i:
   swap arr[i] and arr[min_idx]

**Time and Space Complexity:**
- **Time Complexity:**
  - Best-case: O(n²)
  - Average-case: O(n²)
  - Worst-case: O(n²)

- **Space Complexity:**
  - O(1), as the algorithm sorts the array in place without requiring additional storage.

**Additional Notes/Hints:**
- Selection Sort is a simple comparison-based sorting algorithm.
- It is not suitable for large datasets due to its quadratic time complexity.
- The algorithm is not stable, meaning it may not preserve the relative order of equal elements.
- Selection Sort performs well on small arrays or when memory write operations are costly, as it makes the minimum possible number of swaps.


## Problem Statement: 104
Implement the **Insertion Sort** algorithm to sort a given array `arr[]` in ascending order.

**Input:**
- An integer array `arr[]` of size `n` containing unsorted elements.

**Output:**
- The array `arr[]` sorted in non-decreasing order.

**Constraints:**
- `1 <= n <= 10^3`
- `-10^3 <= arr[i] <= 10^3`

**Goal of the Code:**
To sort the input array `arr[]` using the Insertion Sort algorithm.

**Approach:**
1. **Iterate through the array starting from the second element:**
   - Assume the first element is already sorted.
   - For each subsequent element (`key`), compare it with the elements in the sorted portion to its left.

2. **Shift elements in the sorted portion:**
   - While elements in the sorted portion are greater than the `key`, shift them one position to the right.

3. **Insert the `key` into its correct position:**
   - Place the `key` in the position where all elements to its left are smaller or equal, and all elements to its right are greater.

**Pseudocode/Algorithm:**
insertionSort(arr): n = length of arr for i from 1 to n-1: key = arr[i] j = i - 1 while j >= 0 and arr[j] > key: arr[j + 1] = arr[j] j = j - 1 arr[j + 1] = key

**Time and Space Complexity:**
- **Time Complexity:**
  - Best-case: O(n) when the array is already sorted.
  - Average-case: O(n²) due to the nested loops.
  - Worst-case: O(n²) when the array is sorted in reverse order.

- **Space Complexity:**
  - O(1), as the algorithm sorts the array in place without requiring additional storage.

**Additional Notes/Hints:**
- Insertion Sort is efficient for small datasets or nearly sorted arrays.
- It is a stable sorting algorithm, meaning it preserves the relative order of equal elements.
- The algorithm is adaptive, performing better on datasets that are already partially sorted.
- Insertion Sort is commonly used in practice for sorting small arrays or as a part of more complex algorithms.



## Problem Statement: 105
Implement the **Merge Sort** algorithm to sort a given array `arr[]` in ascending order.

**Input:**
- An integer array `arr[]` of size `n` containing unsorted elements.

**Output:**
- The array `arr[]` sorted in non-decreasing order.

**Constraints:**
- `1 <= n <= 10^3`
- `-10^3 <= arr[i] <= 10^3`

**Goal of the Code:**
To sort the input array `arr[]` using the Merge Sort algorithm.

**Approach:**
1. **Divide:** Recursively split the array into two halves until each subarray contains a single element.
2. **Conquer:** Merge the subarrays by comparing elements and sorting them during the merge process.
3. **Combine:** Repeat the merging process until all subarrays are merged into a single sorted array.

**Pseudocode/Algorithm:**
```java
mergeSort(arr, l, r): 
if l < r: m = l + (r - l) // 2
 mergeSort(arr, l, m) 
 mergeSort(arr, m + 1, r)
 merge(arr, l, m, r)

merge(arr, l, m, r): 
n1 = m - l + 1 
n2 = r - m 
L = new array of size n1 
R = new array of size n2 
for i from 0 to n1 - 1: 
L[i] = arr[l + i] 
for j from 0 to n2 - 1: 
R[j] = arr[m + 1 + j]
 i = 0 j = 0 k = l
  while i < n1 and j < n2: 
  if L[i] <= R[j]:
   arr[k] = L[i] 
   i = i + 1 
   else:
    arr[k] = R[j] 
    j = j + 1 
    k = k + 1 
    while i < n1: 
    arr[k] = L[i] 
    i = i + 1 
    k = k + 1 
    while j < n2: 
    arr[k] = R[j] 
    j = j + 1 
    k = k + 1
```
