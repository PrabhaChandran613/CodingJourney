**Constraints**

- `1 ≤ Node data, Key ≤ 10^6`  

**Goal of the Code**  
To efficiently search for a **key** in a **singly linked list** by traversing the list iteratively.  

**Approach**  
1. Start from the **head** of the linked list.  
2. Traverse the list **node by node**.  
3. If the **current node's data** matches the key, return `true`.  
4. If the traversal reaches the end without finding the key, return `false`.  

**Pseudocode/Algorithm**

**Time Complexity:** O(n) → Each node is visited once.
**Space Complexity:** O(1) → Only a few extra pointers are used.

**Additional Notes/Hints**
This approach reverses the linked list in-place without extra memory.
For recursive reversal, we can use a stack or recursion, but it increases space complexity to O(n).
This method maintains constant space O(1), making it optimal for large lists.

To determine if a **singly linked list** reads the same **forward and backward**.  

**Approach**  
1. Use **two pointers** (`slow` and `fast`) to find the **middle** of the linked list.  
2. Reverse the **second half** of the linked list.  
3. Compare the **first half** with the **reversed second half**.  
4. Restore the original list structure by reversing the second half again.  
5. Return `true` if both halves match; otherwise, return `false`.  

**Pseudocode/Algorithm**  
```java
function isPalindrome(head):  
  if head is null or head.next is null:  
    return true  

  slow = head  
  fast = head  

  while fast.next is not null and fast.next.next is not null:  
    slow = slow.next  
    fast = fast.next.next  

  head2 = reverseList(slow.next)  
  slow.next = null  

  result = isMatch(head, head2)  

  head2 = reverseList(head2)  
  slow.next = head2  

  return result  

function reverseList(head):  
  prev = null  
  curr = head  
  while curr is not null:  
    next = curr.next  
    curr.next = prev  
    prev = curr  
    curr = next  
  return prev  

function isMatch(n1, n2):  
  while n1 is not null and n2 is not null:  
    if n1.data ≠ n2.data:  
      return false  
    n1 = n1.next  
    n2 = n2.next  
  return true  
```

- `1 ≤ Node data ≤ 10^6`

**Goal of the Code**  
To efficiently **find the intersection node** of two linked lists while ensuring **optimal time complexity**.  

**Approach**  
1. **Find the length** of both linked lists (`len1` and `len2`).  
2. Compute the **difference `diff`** between the lengths.  
3. Move the **longer list ahead by `diff` nodes**, so both lists become of equal length.  
4. Traverse both lists **simultaneously** until the intersection node is found.  
5. Return the **first common node**.  

**Pseudocode/Algorithm**  
```java
function getCount(head):
    count = 0
    currNode = head
    while currNode is not null:
        count++
        currNode = currNode.next
    return count

function getIntersection(diff, head1, head2):
    curr1 = head1
    curr2 = head2

    for i from 0 to diff:
        if curr1 is null:
            return null
        curr1 = curr1.next

    while curr1 is not null and curr2 is not null:
        if curr1 == curr2:
            return curr1
        curr1 = curr1.next
        curr2 = curr2.next

    return null

function intersectPoint(head1, head2):
    len1 = getCount(head1)
    len2 = getCount(head2)
    diff = abs(len1 - len2)

    if len1 > len2:
        return getIntersection(diff, head1, head2)
    else:
        return getIntersection(diff, head2, head1)
```
**Time Complexity:** O(n1 + n2) → Since we traverse both lists only once.

**Additional Notes/Hints**
The naive approach using a nested loop has O(n1 × n2) complexity, which is inefficient.
Using hashing can solve it in O(n1 + n2) time but uses extra space.
The two-pointer technique used in this approach is optimal, working in O(n1 + n2) time and O(1) space.

**Input**  
- A **singly linked list** `head` of size `n`.  
- An **integer** `k`, representing the group size.  

**Output**  
- The **head** of the modified linked list after reversing nodes in groups of `k`.  

**Constraints**  
- `1 ≤ n ≤ 10^5`  
- `1 ≤ Node value ≤ 10^9`  
- `1 ≤ k ≤ n`  

**Goal of the Code**  
To efficiently **reverse the linked list in groups of `k`** while maintaining the order of remaining nodes.  

**Approach**  
1. **Initialize Pointers**  
   - `curr` → Tracks the current node.  
   - `prev` → Stores the previous node while reversing.  
   - `newHead` → Stores the head of the new reversed list.  
   - `tail` → Points to the last node of the previous reversed group.  

2. **Reverse K Nodes at a Time**  
   - Traverse and reverse `k` nodes at a time.  
   - Keep track of `groupHead` (the first node of the current group before reversal).  
   - Update `tail.next` to link the reversed group properly.  

3. **Handle Remaining Nodes**  
   - If nodes left are **less than k**, reverse them as well.  

**Pseudocode/Algorithm**  
```java
function reverseKGroup(head, k): 
if head is null:
    return head

curr = head
newHead = null
tail = null

while curr is not null:
    groupHead = curr
    prev = null
    nextNode = null
    count = 0

    # Reverse k nodes
    while curr is not null and count < k:
        nextNode = curr.next
        curr.next = prev
        prev = curr
        curr = nextNode
        count += 1

    # Update new head for the first reversed group
    if newHead is null:
        newHead = prev

    # Connect the previous groups tail to the current reversed group
    if tail is not null:
        tail.next = prev

    # Move the tail to the last node of the reversed group
    tail = groupHead

return newHead
```
**Best & Average Case (O(n)):** Since we traverse the list once and reverse nodes in constant time.
**Worst Case (O(n)):** Even if nodes are not a multiple of k, we still traverse the entire list.
**Space Complexity (O(1)):** No extra space is used apart from pointers.

**Additional Notes/Hints**
This is an in-place reversal method, meaning it does not use extra space.
The function ensures that the order of groups is maintained, while reversing k elements in each step.
If the last set of elements is less than k, they are still reversed.

## Problem Statement: 120 
**Remove Loop in Linked List**
Given the `head` of a **singly linked list**, detect and **remove** the loop (if present).  
A **loop** means that the **last node** in the list **points back** to an earlier node, forming a cycle.

**Input Format**
- `head`: Head node of a singly linked list.
- `pos`: **(1-based index)** Position of the node to which the last node **points**.
  - If `pos = 0`, there is **no loop**.
  - If `pos > 0`, a loop **exists**.

**Output Format**
- **Modified linked list** where the loop (if present) is removed.
- The program prints:
  - `"true"` if the loop is successfully removed.
  - `"false"` if the loop still exists or the list is modified incorrectly.

**Constraints**
- `1 ≤ Number of Nodes ≤ 10^4`
- `1 ≤ Node Value ≤ 10^5`

**Approach**
To **detect and remove** a loop, we use **Floyd’s Cycle Detection Algorithm**:
**Step 1: Detect the Loop**
- Use two pointers: `slow` and `fast`.
- Move `slow` **one step** and `fast` **two steps** at a time.
- If `slow == fast`, **a loop is detected**.
**Step 2: Find the Start of the Loop**
- Reset `slow` to `head`.
- Move `slow` and `fast` **one step at a time**.
- The **first meeting point** is the **start of the loop**.
**Step 3: Remove the Loop**
- Find the **last node** of the loop.
- Set its `next` pointer to `null`, **removing the loop**.

**Algorithm (Pseudocode)**
```java
function removeLoop(head):
    slow = head, fast = head
    
    // Step 1: Detect loop
    while (fast != null && fast.next != null):
        slow = slow.next
        fast = fast.next.next

        if (slow == fast): // Loop detected
            break

    if (slow == fast): // Step 2: Find start of loop
        slow = head
        if (slow != fast):  // Loop is not at head
            while (slow.next != fast.next):
                slow = slow.next
                fast = fast.next
        else:  // Loop is at head, find last node
            while (fast.next != slow):
                fast = fast.next
        
        fast.next = null  // Step 3: Remove loop
```
**Time Complexity:** O(N) (Detects and removes loop in linear time)
**Space Complexity:** O(1) (Uses constant extra space)

```java
class Solution {
    // Function to sort a linked list of 0s, 1s and 2s.
    static Node segregate(Node head) {
        // Step 1: Count occurrences of 0s, 1s, and 2s
        int[] count = {0, 0, 0};
        Node ptr = head;
        while (ptr != null) {
            count[ptr.data]++; // Increment count for respective data (0,1,2)
            ptr = ptr.next;
        }
        
        // Step 2: Modify the linked list in sorted order
        int idx = 0;
        ptr = head;
        while (ptr != null) {
            if (count[idx] == 0) { // Move to next index when count is zero
                idx++;
            } else {
                ptr.data = idx;  // Assign value 0,1, or 2
                count[idx]--;    // Decrease the count
                ptr = ptr.next;  // Move to next node
            }
        }
        return head;  // Return sorted list
    }
}
```
**Time and Space Complexity**

Modifying List              	O(N)
Total	                        O(N)
Space Complexity	            O(1) (In-place sorting)

## Problem Statement: 122
**Remove Duplicates from an Unsorted Linked List**
Given an **unsorted singly linked list**, the task is to **remove all duplicate elements** while keeping the first occurrence of each element.

**Input Format**
- `head`: Head node of a **singly linked list**.

**Output Format**
- The modified linked list with **only unique elements**, keeping their original order.

**Constraints**
- `1 ≤ Number of Nodes ≤ 10^4`
- `0 ≤ Node Data ≤ 10^3`

**Approach: Hashing (Optimal)**
**Steps**
1. **Use a HashSet** to keep track of already seen values.
2. **Iterate through the list**, checking if a node's value is in the set:
   - If **not in the set**, add it and move to the next node.
   - If **already in the set**, remove the node by adjusting pointers.
3. **Return the modified linked list**.

**Algorithm**
1. **Initialize** a `HashSet<Integer>` to store seen values.
2. **Use two pointers**:
   - `curr` for iteration.
   - `prev` to track the previous node.
3. **Traverse the list**:
   - If `curr.data` is in `hashSet`, skip the node (`prev.next = curr.next`).
   - Otherwise, add `curr.data` to `hashSet` and move `prev` forward.
4. **Return the modified list**.

**Implementation**
```java
import java.util.HashSet;

class Solution {
    // Function to remove duplicates from unsorted linked list.
    public Node removeDuplicates(Node head) {
        // Edge case: empty list or single node
        if (head == null || head.next == null) return head;

        // HashSet to store seen values
        HashSet<Integer> hashSet = new HashSet<>();
        Node curr = head;
        Node prev = null;

        while (curr != null) {
            if (hashSet.contains(curr.data)) {
                // Duplicate found, remove it
                prev.next = curr.next;
            } else {
                // Add to set and move prev
                hashSet.add(curr.data);
                prev = curr;
            }
            curr = curr.next;
        }
        return head;
    }
}
```
**Time and Space Complexity**

HashSet Operations (Insert, Lookup)	        O(1)

**Pairwise Swap Elements of a Linked List**
Given a **singly linked list**, swap **adjacent nodes** in pairs without swapping data. If the list has an odd number of nodes, the last node remains unchanged.

**Approach: Iterative (Using Pointers)**
**Steps**
1. **Use a dummy node** to simplify handling edge cases.
2. **Traverse the list in pairs**:
   - Swap **two adjacent nodes**.
   - Update pointers correctly to maintain the correct structure.
3. **Continue until the end of the list is reached**.

**Algorithm**
1. **Create a dummy node** pointing to `head` (for easier pointer manipulation).
2. **Use a `prev` pointer** (starting at the dummy node) to track the node before the pair.
3. **Loop while at least two nodes remain**:
   - Identify `first` and `second` nodes in the pair.
   - Swap them by adjusting `next` pointers.
   - Move `prev` to the next pair.
4. **Return `dummy.next`** (new head of the list).

**Implementation**
```java
class Solution {
    // Function to pairwise swap elements of a linked list.
    public Node pairwiseSwap(Node head) {
        // Create a dummy node to simplify pointer manipulation
        Node dummy = new Node(-1);
        dummy.next = head;
        Node prev = dummy;

        // Loop through the list in pairs
        while (prev.next != null && prev.next.next != null) {
            // Identify nodes to swap
            Node first = prev.next;
            Node second = first.next;

            // Swapping nodes
            first.next = second.next;
            second.next = first;
            prev.next = second;

            // Move prev to the next pair
            prev = first;
        }

        // Return the new head
        return dummy.next;
    }
}

```

**Approach**

2. **Maintain a pointer (`curr`)** to track the last node in the merged list.
3. **Compare nodes from both lists (`head1` and `head2`)**:
   - Append the smaller node to `curr.next`.
   - Move the corresponding pointer (`head1` or `head2`) forward.
4. **After one list is exhausted**, append the remaining nodes of the other list.
5. **Return `dummy.next`**, which is the head of the merged list.

**Algorithm**
1. Create a **dummy node (`dummy`)** and set `curr = dummy`.
2. While both lists are non-empty (`head1 != null` and `head2 != null`):
   - Compare `head1.data` and `head2.data`.
   - Append the smaller node to `curr.next` and move the pointer forward.
3. Append the remaining nodes from `head1` or `head2` (whichever is not null).
4. Return `dummy.next`.

**Implementation**
```java
class Solution {
    Node sortedMerge(Node head1, Node head2) {
        Node dummy = new Node(-1);
        Node curr = dummy;

        while (head1 != null && head2 != null) {
            if (head1.data <= head2.data) {
                curr.next = head1;
                head1 = head1.next;
            } else {
                curr.next = head2;
                head2 = head2.next;
            }
            curr = curr.next;
        }

        // Append the remaining nodes
        if (head1 != null) {
            curr.next = head1;
        } else {
            curr.next = head2;
        }

        return dummy.next;
    }
}

```

**Approach**

   - One for **even nodes** (`eStart`).
   - One for **odd nodes** (`oStart`).
2. **Traverse the list** and distribute nodes into these lists.
3. **Merge the even and odd lists** so that even nodes appear first.
4. **Return the head of the modified list**.

**Algorithm**
1. **Create two dummy nodes**:
   - `eStart` (for even numbers).
   - `oStart` (for odd numbers).
2. Maintain **two pointers**:
   - `eEnd` (tracks the last even node).
   - `oEnd` (tracks the last odd node).
3. **Traverse the given list**:
   - If the current node's value is even, append it to `eEnd.next`.
   - If odd, append it to `oEnd.next`.
   - Move `eEnd` and `oEnd` accordingly.
4. **Merge both lists**:
   - Set `eEnd.next = oStart.next` to attach odd nodes after even nodes.
   - Set `oEnd.next = null` to mark the end of the list.
5. **Return `eStart.next`** (head of the modified list).

**Implementation**
```java
class Solution {
    Node divide(Node head) {
        // Create two dummy nodes
        Node eStart = new Node(0);
        Node oStart = new Node(0);

        // End pointers for even and odd lists
        Node eEnd = eStart;
        Node oEnd = oStart;

        // Traverse the list
        Node curr = head;
        while (curr != null) {
            if (curr.data % 2 == 0) {
                eEnd.next = curr;
                eEnd = eEnd.next;
            } else {
                oEnd.next = curr;
                oEnd = oEnd.next;
            }
            curr = curr.next;
        }

        // Merge even and odd lists
        eEnd.next = oStart.next;
        oEnd.next = null; // Mark end of list

        // Return head of the modified list
        return eStart.next;
    }
}

```

**Add 1 to a Linked List Number**
Given a linked list where each node contains a single digit, treat the linked list as a number (in left-to-right order). Add 1 to this number and return the updated linked list.

**Approach**
1. **Reverse the linked list** to process addition from the least significant digit.
2. **Perform addition** while handling carry.
3. **Reverse the list back** to restore the original order.

**Algorithm**
1. **Reverse the linked list** so the least significant digit comes first.
2. **Add 1** to the first digit and handle the carry.
3. **Traverse the list**:
   - If `sum >= 10`, set carry to `1`, else `0`.
   - Update the node’s value with `sum % 10`.
   - Move to the next node.
4. **If carry remains**, create a new node with value `1` at the end.
5. **Reverse the linked list back** to restore the original order.

**Implementation**
```java
class Solution {
    // Function to reverse a linked list
    private Node reverse(Node head) {
        Node prev = null, curr = head, next;
        while (curr != null) {
            next = curr.next;
            curr.next = prev;
            prev = curr;
            curr = next;
        }
        return prev;
    }

    // Function to add one in the reversed list
    private Node addOneInReverse(Node head) {
        Node res = head, curr = head, last = null;
        int carry = 1, sum;

        while (curr != null) {
            sum = carry + curr.data;
            carry = (sum >= 10) ? 1 : 0;
            curr.data = sum % 10;
            last = curr;
            curr = curr.next;
        }

        // If carry is still remaining, add a new node
        if (carry > 0) {
            last.next = new Node(carry);
        }

        return res;
    }

    public Node addOne(Node head) {
        head = reverse(head); // Step 1: Reverse list
        head = addOneInReverse(head); // Step 2: Add 1
        return reverse(head); // Step 3: Reverse back
    }
}

```

**Delete Nodes Having Greater Value on Right**
Given a singly linked list, remove all nodes that have a node with a greater value anywhere to their right in the list. Return the head of the modified linked list.

**Approach**
1. **Reverse the linked list** so that we can process it from right to left.
2. **Traverse and delete nodes** that have a smaller value than the maximum encountered so far.
3. **Reverse the list back** to restore the original order.

**Algorithm**
1. **Reverse the linked list**.
2. **Traverse the reversed list**:
   - Maintain a `maxNode` that stores the largest value seen so far.
   - If `curr.next.data < maxNode.data`, delete `curr.next`.
   - Otherwise, update `maxNode` to `curr.next`.
3. **Reverse the list back** to get the final result.

**Implementation**
```java
class Solution {
    // Function to reverse a linked list
    private Node reverse(Node head) {
        Node prev = null, curr = head, next;
        while (curr != null) {
            next = curr.next;
            curr.next = prev;
            prev = curr;
            curr = next;
        }
        return prev;
    }

    // Function to delete nodes with greater values on the right
    private Node deleteNodes(Node head) {
        Node curr = head, maxNode = head, temp;

        while (curr != null && curr.next != null) {
            if (curr.next.data < maxNode.data) {
                temp = curr.next;
                curr.next = temp.next; // Delete the node
            } else {
                curr = curr.next;
                maxNode = curr; // Update maxNode
            }
        }
        return head;
    }

    public Node compute(Node head) {
        head = reverse(head);   // Step 1: Reverse the list
        head = deleteNodes(head); // Step 2: Delete nodes
        return reverse(head);   // Step 3: Reverse back to original order
    }
}

```
