## Problem Statement:108  
Given a **singly linked list** of `n` nodes and a **key**, the task is to check whether the key is present in the linked list or not.  

**Input**  
- A **singly linked list** with `n` nodes.  
- An **integer key** to search in the linked list.  

**Output**  
- `true` if the key is found in the linked list.  
- `false` otherwise.  

**Constraints**  
- `1 ≤ n ≤ 10^4`  
- `1 ≤ Node data, Key ≤ 10^6`  

**Goal of the Code**  
To efficiently search for a **key** in a **singly linked list** by traversing the list iteratively.  

**Approach**  
1. Start from the **head** of the linked list.  
2. Traverse the list **node by node**.  
3. If the **current node's data** matches the key, return `true`.  
4. If the traversal reaches the end without finding the key, return `false`.  

**Pseudocode/Algorithm**  
```java
function searchKey(head, key):  
  current = head  
  while current is not null:  
    if current.data == key:  
      return true  
    current = current.next  
  return false  
  ```
**Time Complexity:** O(n) → Each node is visited once in the worst case.
**Space Complexity:** O(1) → No extra space is used.

**Additional Notes/Hints**
This approach works for unsorted linked lists as well.
In a sorted linked list, we can stop searching early if the current node's data exceeds the key.
If frequent searches are needed, hashing or a balanced BST might be better alternatives.


## Problem Statement: 109
**Reverse a Linked List**  
Given the **head** of a **singly linked list**, the task is to **reverse** the linked list and return the **new head** of the reversed list.  

**Input**  
- A **singly linked list** with `n` nodes.  

**Output**  
- The **head** of the reversed linked list.  

**Constraints**  
- `1 ≤ n ≤ 10^5`  
- `1 ≤ Node data ≤ 10^6`  

**Goal of the Code**  
To reverse a **singly linked list** in-place without using extra space.  

**Approach**  
1. Initialize three pointers:  
   - `prev` → initially `null` (tracks the previous node).  
   - `curr` → initially `head` (tracks the current node).  
   - `next` → stores the next node temporarily.  
2. Traverse the list while `curr` is not null:  
   - Store `next = curr.next` (save the next node).  
   - Reverse the link: `curr.next = prev`.  
   - Move `prev` and `curr` one step forward.  
3. Return `prev` as the new head (since `curr` becomes `null` at the end).  

**Pseudocode/Algorithm**  
```java
function reverseList(head):  
  prev = null  
  curr = head  
  while curr is not null:  
    next = curr.next  
    curr.next = prev  
    prev = curr  
    curr = next  
  return prev  
```

**Time Complexity:** O(n) → Each node is visited once.
**Space Complexity:** O(1) → Only a few extra pointers are used.

**Additional Notes/Hints**
This approach reverses the linked list in-place without extra memory.
For recursive reversal, we can use a stack or recursion, but it increases space complexity to O(n).
This method maintains constant space O(1), making it optimal for large lists.

## Problem Statement: 110
**Palindrome Linked List** 
Given the **head** of a **singly linked list** of **positive integers**, the task is to check if the linked list is a **palindrome** or not.  

**Input**  
- A **singly linked list** with `n` nodes.  

**Output**  
- `true` if the linked list is a **palindrome**.  
- `false` otherwise.  

**Constraints**  
- `1 ≤ n ≤ 10^5`  
- `1 ≤ Node data ≤ 10^6`  

**Goal of the Code**  
To determine if a **singly linked list** reads the same **forward and backward**.  

**Approach**  
1. Use **two pointers** (`slow` and `fast`) to find the **middle** of the linked list.  
2. Reverse the **second half** of the linked list.  
3. Compare the **first half** with the **reversed second half**.  
4. Restore the original list structure by reversing the second half again.  
5. Return `true` if both halves match; otherwise, return `false`.  

**Pseudocode/Algorithm**  
```java
function isPalindrome(head):  
  if head is null or head.next is null:  
    return true  

  slow = head  
  fast = head  

  while fast.next is not null and fast.next.next is not null:  
    slow = slow.next  
    fast = fast.next.next  

  head2 = reverseList(slow.next)  
  slow.next = null  

  result = isMatch(head, head2)  

  head2 = reverseList(head2)  
  slow.next = head2  

  return result  

function reverseList(head):  
  prev = null  
  curr = head  
  while curr is not null:  
    next = curr.next  
    curr.next = prev  
    prev = curr  
    curr = next  
  return prev  

function isMatch(n1, n2):  
  while n1 is not null and n2 is not null:  
    if n1.data ≠ n2.data:  
      return false  
    n1 = n1.next  
    n2 = n2.next  
  return true  
```
**Time Complexity:** O(n) → The linked list is traversed a few times.
**Space Complexity:** O(1) → The reversal is done in-place without extra space.

**Additional Notes/Hints**
This approach modifies the linked list temporarily but restores it after checking.
If modifying the list is not allowed, use a stack to store the first half and compare it with the second half.
This method ensures an optimal O(n) time and O(1) space complexity.

## Problem Statement: 111
**Insert in Middle of Linked List**  
Given the **head** of a **Singly Linked List** and a **value `x`**, the task is to **insert** the key `x` in the **middle** of the linked list.  

**Input**  
- A **singly linked list** with `n` nodes.  
- An **integer `x`**, representing the value to be inserted.  

**Output**  
- The **updated linked list** after inserting `x` in the middle.  

**Constraints**  
- `1 ≤ n ≤ 10^5`  
- `1 ≤ Node data, x ≤ 10^6`  

**Goal of the Code**  
To efficiently insert an **element in the middle** of a singly linked list while maintaining the list structure.  

**Approach**  
1. **Find the length** of the linked list.  
2. Calculate the **middle position** using `(length / 2)` (considering 0-based indexing).  
3. Traverse to the **middle node**.  
4. Insert the **new node** after the middle node.  
5. Return the **updated head** of the linked list.  

**Pseudocode/Algorithm**  
```java
function insertInMiddle(head, x):  
  if head is null:  
    return new Node(x)  

  newNode = new Node(x)  
  currNode = head  
  length = 0  

  while currNode is not null:  
    length++  
    currNode = currNode.next  

  count = (length / 2) if length is even else (length + 1) / 2  
  currNode = head  

  while count-- > 1:  
    currNode = currNode.next  

  newNode.next = currNode.next  
  currNode.next = newNode  

  return head  
```
**Time Complexity:** O(n) → Since we traverse the list twice (once to find length, once to insert).
**Space Complexity:** O(1) → No extra space is used apart from the new node.

**Additional Notes/Hints**
If the list is empty, the new node becomes the head.
If the list has only one element, the new node is inserted after it.
This method modifies the list in-place without using extra memory.

## Problem Statement: 112
**Intersection in Y Shaped Lists**  
Given the **head of two singly linked lists**, return the **point where these two linked lists intersect**.  

**Note:** It is guaranteed that the **intersected node always exists**.  

**Input**  
- `head1`: The **head of the first linked list**.  
- `head2`: The **head of the second linked list**.  

**Output**  
- The **intersection node** where both linked lists merge.  

**Constraints**  
- `1 ≤ n1, n2 ≤ 10^5`  
- `1 ≤ Node data ≤ 10^6`  
- The two linked lists **must intersect** at some point.  

**Goal of the Code**  
To efficiently **find the intersection node** of two linked lists while ensuring **optimal time complexity**.  

**Approach**  
1. **Find the length** of both linked lists (`len1` and `len2`).  
2. Compute the **difference `diff`** between the lengths.  
3. Move the **longer list ahead by `diff` nodes**, so both lists become of equal length.  
4. Traverse both lists **simultaneously** until the intersection node is found.  
5. Return the **first common node**.  

**Pseudocode/Algorithm**  
```java
function getCount(head):
    count = 0
    currNode = head
    while currNode is not null:
        count++
        currNode = currNode.next
    return count

function getIntersection(diff, head1, head2):
    curr1 = head1
    curr2 = head2

    for i from 0 to diff:
        if curr1 is null:
            return null
        curr1 = curr1.next

    while curr1 is not null and curr2 is not null:
        if curr1 == curr2:
            return curr1
        curr1 = curr1.next
        curr2 = curr2.next

    return null

function intersectPoint(head1, head2):
    len1 = getCount(head1)
    len2 = getCount(head2)
    diff = abs(len1 - len2)

    if len1 > len2:
        return getIntersection(diff, head1, head2)
    else:
        return getIntersection(diff, head2, head1)
```
**Time Complexity:** O(n1 + n2) → Since we traverse both lists only once.
**Space Complexity:** O(1) → No extra space is used apart from variables.

**Additional Notes/Hints**
The naive approach using a nested loop has O(n1 × n2) complexity, which is inefficient.
Using hashing can solve it in O(n1 + n2) time but uses extra space.
The two-pointer technique used in this approach is optimal, working in O(n1 + n2) time and O(1) space.

## Problem Statement: 113
**Delete without Head Pointer**  
You are given a **node (del_node) of a Singly Linked List** and must delete it from the linked list.  

**Constraints:**  
- You **do not have access to the head** of the linked list.  
- The given `del_node` **will not be the last node** in the linked list.  
- The **number of nodes should decrease by one**, but memory deallocation is **not required**.  
- The **order of remaining nodes must be maintained**.  

**Input**  
- `del_node`: A reference to the node that needs to be deleted.  

**Output**  
- The linked list with `del_node` removed.  

**Approach**  
Since we **do not have access to the head**, we cannot traverse the list normally. Instead, we **modify the given node itself**:  
1. **Copy the data** of the **next node** into `del_node`.  
2. **Change the pointer** of `del_node` to **skip** the next node.  
3. This effectively **removes the next node** and replaces `del_node` with its successor.  

**Algorithm / Pseudocode**  
```java
function deleteNode(node):
    if node is null:
        return
    if node.next is null: 
        return  // Edge case (but problem guarantees it's not the last node)
    
    node.data = node.next.data  // Copy next node's data to current node
    node.next = node.next.next  // Skip the next node (effectively deleting it)
```

**Time Complexity:** O(1) (Constant time deletion)
**Space Complexity:** O(1) (No extra space used)

## Problem Statement: 114
**Reverse Linked List in Groups of Size K**  
Given the **head** of a singly linked list and an integer `k`, reverse every `k` nodes in the linked list.  
If the number of nodes is **not a multiple of k**, reverse the remaining nodes as a group.  

**Input**  
- A **singly linked list** `head` of size `n`.  
- An **integer** `k`, representing the group size.  

**Output**  
- The **head** of the modified linked list after reversing nodes in groups of `k`.  

**Constraints**  
- `1 ≤ n ≤ 10^5`  
- `1 ≤ Node value ≤ 10^9`  
- `1 ≤ k ≤ n`  

**Goal of the Code**  
To efficiently **reverse the linked list in groups of `k`** while maintaining the order of remaining nodes.  

**Approach**  
1. **Initialize Pointers**  
   - `curr` → Tracks the current node.  
   - `prev` → Stores the previous node while reversing.  
   - `newHead` → Stores the head of the new reversed list.  
   - `tail` → Points to the last node of the previous reversed group.  

2. **Reverse K Nodes at a Time**  
   - Traverse and reverse `k` nodes at a time.  
   - Keep track of `groupHead` (the first node of the current group before reversal).  
   - Update `tail.next` to link the reversed group properly.  

3. **Handle Remaining Nodes**  
   - If nodes left are **less than k**, reverse them as well.  

**Pseudocode/Algorithm**  
```java
function reverseKGroup(head, k): 
if head is null:
    return head

curr = head
newHead = null
tail = null

while curr is not null:
    groupHead = curr
    prev = null
    nextNode = null
    count = 0

    # Reverse k nodes
    while curr is not null and count < k:
        nextNode = curr.next
        curr.next = prev
        prev = curr
        curr = nextNode
        count += 1

    # Update new head for the first reversed group
    if newHead is null:
        newHead = prev

    # Connect the previous groups tail to the current reversed group
    if tail is not null:
        tail.next = prev

    # Move the tail to the last node of the reversed group
    tail = groupHead

return newHead
```
**Best & Average Case (O(n)):** Since we traverse the list once and reverse nodes in constant time.
**Worst Case (O(n)):** Even if nodes are not a multiple of k, we still traverse the entire list.
**Space Complexity (O(1)):** No extra space is used apart from pointers.

**Additional Notes/Hints**
This is an in-place reversal method, meaning it does not use extra space.
The function ensures that the order of groups is maintained, while reversing k elements in each step.
If the last set of elements is less than k, they are still reversed.
