### Problem Statement: 27
**Find Repeating Number in Array:**
Given an array `arr[]` of size `n`, where each element is between `1` and `n-1`, find the repeating number in the array.

**Note:** The array has exactly one element that repeats.
**Approach:**
We can solve this problem efficiently by using **Floyd's Tortoise and Hare Algorithm** (Cycle detection in a linked list). This approach helps in detecting cycles without extra space.

**Explanation of Approach:**
1. **Cycle Detection Approach (Floyd's Algorithm):**
   - We treat each number in the array as a "pointer" to another index in the array.
   - This transforms the problem into finding the cycle in the array, where the "tortoise" and "hare" pointers will eventually meet at the repeating element.

2. **Steps:**
   - First, we initialize two pointers, `slow` and `fast`. Both pointers will traverse the array, but the `fast` pointer moves two steps at a time while the `slow` pointer moves one step at a time.
   - Eventually, the two pointers will meet at some point within the cycle. Then, to find the repeating number, we reset one pointer to the start and move both pointers one step at a time until they meet again at the entry point of the cycle, which is the repeating number.

**Code:**

```java
public class Solution {

    // Function to find the repeating number
    static int findRepeatingNumber(int arr[]) {
        // Phase 1: Finding intersection point in the cycle
        int slow = arr[0];
        int fast = arr[0];

        // Loop to detect the cycle
        do {
            slow = arr[slow];
            fast = arr[arr[fast]];
        } while (slow != fast);

        // Phase 2: Find the entry point of the cycle (repeating number)
        slow = arr[0];
        while (slow != fast) {
            slow = arr[slow];
            fast = arr[fast];
        }

        return slow;  // This is the repeating number
    }

    // Main method to test the function
    public static void main(String[] args) {
        // Example array with a repeating number
        int arr[] = { 1, 3, 4, 2, 2 };

        // Call the function and print the result
        System.out.println("The repeating number is: " + findRepeatingNumber(arr));
    }
}


### Problem Statement:28

**Sort 0s, 1s, and 2s**
Given an array `arr[]` containing only 0s, 1s, and 2s, sort the array in ascending order.
You need to solve this problem **without utilizing the built-in sort function**.

 **Approach**

The problem asks us to sort the array containing only 0s, 1s, and 2s. The ideal solution for this is the **Dutch National Flag Problem**, which can be solved efficiently in **one pass** using the **three-pointer technique**.

**Algorithm:**
1. **Initial Pointers:**
   - We will use three pointers: `low`, `mid`, and `high`.
   - `low`: This pointer keeps track of the boundary for the 0s.
   - `mid`: This pointer is used to scan the array.
   - `high`: This pointer keeps track of the boundary for the 2s.
   
2. **Three-Way Partitioning:**
   - The `mid` pointer starts from the beginning of the array and moves towards the end.
   - If the current element (`arr[mid]`) is 0, we swap it with the element at the `low` pointer, then increment both `low` and `mid`.
   - If the current element (`arr[mid]`) is 1, we simply increment the `mid` pointer.
   - If the current element (`arr[mid]`) is 2, we swap it with the element at the `high` pointer, and decrement the `high` pointer.

3. **Termination Condition:**
   - The algorithm terminates when the `mid` pointer surpasses the `high` pointer.

This approach guarantees that the array is sorted in **one pass** and uses only **constant extra space** (i.e., no additional arrays or data structures).

**Time Complexity:**
- **O(n):** The array is traversed once, and each element is processed at most once.

**Space Complexity:**
- **O(1):** No extra space is used for sorting (constant space).


### Problem Statement:29
Given two arrays `a[]` and `b[]` of equal size, the task is to check if they contain the same elements with the same frequencies. The order of elements may be different, but both arrays should have identical elements in equal counts.

**Input:**
- Two arrays `a[]` and `b[]`, both of equal size.

**Output:**
- A boolean value `true` if the arrays are equal (contain the same set of elements with the same frequencies).
- A boolean value `false` otherwise.

**Constraints:**
- The arrays `a[]` and `b[]` have equal size.
- The size of the arrays can range from 1 to 10^5 elements.
- Array elements can be integers, and the values may be positive, negative, or zero.

**Goal of the code:**
The main objective is to determine if two arrays are equal by checking if they contain the same elements with identical counts, irrespective of the order.

**Approach:**

1. First, check if the lengths of both arrays are equal. If not, return `false` immediately.
2. Sort both arrays. Sorting helps in comparing the elements in order, which makes it easier to check if both arrays are identical.
3. Compare each corresponding element in the sorted arrays. If any element is different, return `false`.
4. If all elements match, return `true`.

**Pseudocode/Algorithm:**
function checkEqual(a[], b[]): if length of a[] != length of b[]: return false
 sort a[] 
 sort b[] 
 for i = 0 to length of a[] - 1: 
 if a[i] != b[i]:
  return false
  
   return true

- **Time Complexity**: `O(n log n)` where `n` is the length of the arrays. Sorting both arrays takes `O(n log n)` time.
- **Space Complexity**: `O(1)` if sorting is done in-place, or `O(n)` if additional space is required for sorting.

**Additional Notes/Hints:**
- Sorting the arrays ensures that you can directly compare corresponding elements.
- For optimal performance, sorting is chosen over other methods like using hashmaps or counting frequencies manually.

### Problem Statement:30
Given an array `arr`, rotate the array by one position in a clockwise direction, such that the last element becomes the first element, and all other elements are shifted one position to the right.

**Input:**
- A single array `arr[]` of integers.

**Output:**
- The array after rotating by one position in the clockwise direction.

**Constraints:**
- The size of the array can range from 1 to 10^5 elements.
- The array elements can be integers, and the values may be positive, negative, or zero.

**Goal of the code:**
The main objective is to rotate the array by one position in a clockwise direction.

**Approach:**
1. First, store the last element of the array because it needs to be placed at the start after rotation.
2. Shift each element of the array one position to the right, starting from the last element (except the first element).
3. Assign the stored last element to the first position of the array.

**Pseudocode/Algorithm:**
function rotate(arr[]):
 lastElement = arr[arr.length - 1] 
 for i = arr.length - 1 to 1:
  arr[i] = arr[i - 1]
   arr[0] = lastElement

- **Time Complexity**: `O(n)`, where `n` is the length of the array. The loop goes through the entire array once to shift the elements.
- **Space Complexity**: `O(1)`, as we only need a constant amount of extra space (for storing the last element).

**Additional Notes/Hints:**
- This solution works in-place, meaning it doesn't require additional arrays, making it space efficient.
- The last element is temporarily stored in a variable, and then shifted elements are replaced, ensuring the array is rotated by one position efficiently.





### Problem Statement: 31
Given an array of size `N` and a value `K`, rotate the array `K` times in a clockwise direction and print the resulting array.

**Input:**
- An integer array `arr[]` of size `N`.
- An integer `K`, representing the number of right rotations.

**Output:**
- The array after being right rotated `K` times.

**Constraints:**
- `1 ≤ N ≤ 10^5`
- `1 ≤ K ≤ 10^9` (K can be greater than N, so we take `K % N`).
- The array elements can be positive, negative, or zero.

**Goal of the code:**
The main objective is to efficiently rotate the array `K` times to the right and print the updated array.

**Approach:**
1. Since rotating the array `N` times results in the same array, take `K % N` to optimize rotations.
2. Reverse the entire array.
3. Reverse the first `K` elements.
4. Reverse the remaining `N - K` elements.
5. Print the modified array.

**Algorithm:**
function rightRotate(arr[], N, K):
 K = K % N # Reduce K if it's greater than N
  reverse(arr, 0, N-1) # Reverse the entire array
   reverse(arr, 0, K-1) # Reverse the first K elements
    reverse(arr, K, N-1) # Reverse the remaining elements 
    print(arr)
**Time and Space Complexity:**
- **Time Complexity**: `O(N)`, as reversing the array takes linear time.
- **Space Complexity**: `O(1)`, since no extra space is used.

**Additional Notes/Hints:**
- Instead of performing `K` rotations one by one, use the reversal technique to achieve optimal performance.
- If `K` is greater than `N`, taking `K % N` ensures unnecessary rotations are avoided.
**Code:**

```java
// Java Implementation of Right Rotation 
// of an Array K number of times
import java.util.*;
import java.lang.*;
import java.io.*;

class Array_Rotation
{

// Function to rightRotate array
static void RightRotate(int a[], 
                        int n, int k)
{
    
    // If rotation is greater 
    // than size of array
    k=k%n;

    for(int i = 0; i < n; i++)
    {
        if(i<k)
        {
            // Printing rightmost 
            // kth elements
            System.out.print(a[n + i - k] 
                             + " ");
        }
        else
        {
            // Prints array after
            // 'k' elements
            System.out.print(a[i - k] 
                             + " ");
        }
    }
    System.out.println();
}
    
// Driver program
public static void main(String args[])
{
    int Array[] = {1, 2, 3, 4, 5};
    int N = Array.length;

    int K = 2;
    RightRotate(Array, N, K);

}
}
// This code is contributed by M Vamshi Krishna


### Problem Statement: 32
Given an array of size `N` and a value `K`, rotate the array `K` times in a counterclockwise (left) direction and print the resulting array.

**Input:**
- An integer array `arr[]` of size `N`.
- An integer `K`, representing the number of left rotations.

**Output:**
- The array after being left rotated `K` times.

**Constraints:**
- `1 ≤ N ≤ 10^5`
- `1 ≤ K ≤ 10^9` (K can be greater than N, so we take `K % N`).
- The array elements can be positive, negative, or zero.

**Goal of the code:**
The main objective is to efficiently rotate the array `K` times to the left and print the updated array.

**Approach:**
1. Since rotating the array `N` times results in the same array, take `K % N` to optimize rotations.
2. Reverse the first `K` elements.
3. Reverse the remaining `N - K` elements.
4. Reverse the entire array.
5. Print the modified array.

**Algorithm:**
Function leftRotate(arr[], N, K):
K = K % N  // Reduce K if it's greater than N
Reverse the first K elements from index 0 to K-1
Reverse the remaining elements from index K to N-1
Reverse the entire array from index 0 to N-1
Return the modified array


**Time and Space Complexity:**
- **Time Complexity**: `O(N)`, as reversing the array takes linear time.
- **Space Complexity**: `O(1)`, since no extra space is used.

**Additional Notes/Hints:**
- Instead of performing `K` rotations one by one, use the reversal technique to achieve optimal performance.
- If `K` is greater than `N`, taking `K % N` ensures unnecessary rotations are avoided.


### Problem Statement: 33

**Description**
Given two arrays `a[]` and `b[]` of sizes `m` and `n` respectively, the task is to determine whether `b[]` is a subset of `a[]`.  
Both arrays are **not sorted**, and elements are **distinct**.

**Input**
- Two integer arrays `a[]` and `b[]`.
- `m` → Size of array `a[]`.
- `n` → Size of array `b[]`.

**Output**
- Return `true` if `b[]` is a subset of `a[]`, otherwise return `false`.

**Constraints**
- `1 ≤ n ≤ m ≤ 10^5`
- `1 ≤ a[i], b[i] ≤ 10^9`

**Goal of the Code**
To efficiently determine whether all elements of array `b[]` exist in array `a[]`.

**Approach**
1. **Brute Force Approach (O(n * m))**  
   - Iterate through each element of `b[]` and check if it exists in `a[]`.
   - If any element is missing, return `false`; otherwise, return `true`.
  
2. **Optimized Approach using HashSet (O(m + n))**  
   - Store all elements of `a[]` in a **HashSet** for `O(1)` lookup time.
   - Check each element of `b[]` in the HashSet.
   - If any element is missing, return `false`, otherwise return `true`.

**Pseudocode:**
function isSubset(a, b): 
Create a HashSet from elements of array
 a For each element in b:
  If element is not in HashSet:
   Return False
    Return True
**Time and Space Complexity**
Approach	   Time Complexity	  Space Complexity
Brute Force	     O(n * m)	          O(1)
HashSet Method	 O(m + n)	          O(m)

**Additional Notes**
The Brute Force approach is inefficient for large m, n values.
The HashSet approach is optimal and ensures O(1) lookup time.

**Code:**

```java

import java.util.HashSet;

class GfG {
    static boolean isSubset(int[] a, int[] b) {
        HashSet<Integer> set = new HashSet<>();
        
        // Add all elements of a[] to the set
        for (int num : a) {
            set.add(num);
        }

        // Check if all elements of b[] exist in the set
        for (int num : b) {
            if (!set.contains(num)) {
                return false;
            }
        }
        return true;
    }

    public static void main(String[] args) {
        int[] a = {11, 1, 13, 21, 3, 7};
        int[] b = {11, 3, 7, 1};

        System.out.println(isSubset(a, b)); // Output: true
    }
}


### Problem Statement: 34
**Frequencies in a Limited Array**
You are given an array `arr[]` containing positive integers. The elements in the array `arr[]` range from `1` to `n` (where `n` is the size of the array), and some numbers may be repeated or absent. Your task is to count the frequency of all numbers in the range `1` to `n` and return an array of size `n` such that `result[i]` represents the frequency of the number `i` (1-based indexing).

**Input**
- An integer array `arr[]` of size `n`.
- The array contains positive integers between `1` and `n` (inclusive).

**Output**
- Return an array `result[]` of size `n` where `result[i]` represents the frequency of the integer `i` in `arr[]`.

**Constraints**
- `1 ≤ n ≤ 10^5`
- `1 ≤ arr[i] ≤ n`

**Goal of the Code**
To return an array containing the frequency of each number in the range `1` to `n` from the given array `arr[]`.

**Approach**
1. **Create a Result Array**  
   Initialize an array `result[]` of size `n` with all values set to `0`.
   
2. **Count Frequencies**  
   Loop through each element in `arr[]` and increment the corresponding index in the `result[]` array. The value at index `i-1` in `result[]` will represent the frequency of `i` in `arr[]`.

3. **Return the Result**  
   After counting all the frequencies, return the `result[]` array.

**Pseudocode**
function countFrequencies(arr): 
Initialize an array result of size n with all elements set to 0 
For each element in arr:
 Increment result[element - 1] by 1 
 Return result
 **Complexity:**
 Time Complexity: The solution iterates over the array once, making it O(n) in time.
Space Complexity: We need an additional array of size n to store the frequency counts.


### Problem Statement: 35
**Counting Frequencies of Array Elements**
Given an array which may contain duplicates, the task is to print all unique elements in the array along with their frequencies.

**Input**
- An integer array `arr[]` of size `n` (the length of the array).
- The array may contain duplicate values.

**Output**
- Print each unique element and its frequency in the array, one per line. The elements and their frequencies should be printed in the order of their first occurrence.

**Constraints**
- `1 ≤ n ≤ 10^5`
- Array elements are integers, and they may range from negative to positive values.

**Goal of the Code**
To print the frequency of each unique element in the array `arr[]`.

**Approach**
1. **Use a Map (HashMap)**: 
   - A HashMap is used to store each unique element as a key and its frequency as the value.
   
2. **Traverse the Array**:
   - For each element in the array, check if it exists in the map.
   - If it exists, increment its frequency; otherwise, add it to the map with a frequency of 1.
   
3. **Print the Frequencies**:
   - After processing the entire array, print each element and its frequency by iterating through the entries in the map.

**Pseudocode**
function countFreq(arr, n): Initialize an empty map mp For each element in arr: If element exists in mp: Increment its count in mp Else: Add it to mp with count 1 For each entry in map: Print key and value (element and frequency)


**Implementation (Java)**

```java
// Java program to count frequencies of array items
import java.util.*;

class GFG {

    static void countFreq(int arr[], int n) {
        Map<Integer, Integer> mp = new HashMap<>();

        // Traverse through array elements and count frequencies
        for (int i = 0; i < n; i++) {
            if (mp.containsKey(arr[i])) {
                mp.put(arr[i], mp.get(arr[i]) + 1);
            } else {
                mp.put(arr[i], 1);
            }
        }
        
        // Traverse through map and print frequencies
        for (Map.Entry<Integer, Integer> entry : mp.entrySet()) {
            System.out.println(entry.getKey() + " " + entry.getValue());
        }
    }

    // Driver code
    public static void main(String args[]) {
        int arr[] = {10, 20, 20, 10, 10, 20, 5, 20};
        int n = arr.length;
        countFreq(arr, n);
    }
}



### Problem Statement: 36

**Two Sum - Pair with Given Sum**
Given an array `arr[]` of positive integers and another integer `target`, determine if there exist two distinct indices such that the sum of their corresponding elements equals the target.

**Input**
- An integer array `arr[]` of size `n` (where `n` is the number of elements in the array).
- An integer `target`, which is the sum we want to find by adding two distinct elements from the array.

**Output**
- Return `true` if there exist two distinct indices such that the sum of the elements at those indices equals the `target`.
- Return `false` if no such pair exists.

**Constraints**
- `2 ≤ n ≤ 10^5`
- `1 ≤ arr[i] ≤ 10^9`
- `1 ≤ target ≤ 10^9`

**Goal of the Code**
To determine if there exists a pair of distinct elements in the array whose sum equals the given `target`.

**Approach**
1. **Use a HashSet**:
   - As we iterate through the array, we keep track of the elements we have seen so far in a HashSet.
   
2. **Find the complement**:
   - For each element in the array, calculate its complement as `complement = target - arr[i]`.
   - If the complement already exists in the HashSet, that means we have previously encountered an element that, when added to the current element, results in the `target` sum.
   
3. **Return the result**:
   - If such a pair is found, return `true`. Otherwise, after the loop, return `false`.

**Pseudocode**
function twoSum(arr, target):
 Initialize an empty set set For each element arr[i]:
  Calculate complement = target - arr[i] 
  If complement exists in set: 
  return true 
  Add arr[i] to set 
  return false

**Time Complexity:** The loop iterates through the array once, performing constant time operations (checking for the complement and adding to the set). Thus, the time complexity is O(n), where n is the size of the array.

**Space Complexity:** The HashSet stores elements from the array, resulting in a space complexity of O(n) in the worst case, where all elements are distinct.

### Problem Statement: 37  
**Triplet Sum in Array**
Given an array `arr[]` and an integer `target`, determine if there exists a triplet in the array whose sum equals the given `target`.  
Return `true` if such a triplet exists, otherwise, return `false`.  

**Input**  
- An integer array `arr[]` of size `n`.  
- An integer `target` representing the required sum.  

**Output**  
- Return `true` if a triplet with the sum equal to `target` exists.  
- Otherwise, return `false`.

**Constraints**  
- \( 1 \leq n \leq 10^3 \)  
- \( -10^5 \leq arr[i] \leq 10^5 \)  
- \( -10^5 \leq target \leq 10^5 \) 

**Goal of the Code**  
Find whether there exist three numbers in the array whose sum is equal to `target` using an efficient approach.  

**Approach**  
1. **Sorting & Two-Pointer Technique:**  
   - Sort the array to enable a structured search.  
   - Fix one element and use the two-pointer technique to find the remaining two numbers that sum to the required value.  
2. **Steps:**  
   - Sort the array.  
   - Iterate through each element as the first element of the triplet.  
   - Use two pointers (`left` and `right`) to find the remaining two elements.  
   - If their sum matches the required sum, return `true`.  
   - If the sum is smaller, move the left pointer right (`l++`).  
   - If the sum is larger, move the right pointer left (`r--`).  

**Algorithm (Pseudocode)**  
1. **Sort** the given array.  
2. Iterate over the array, choosing each element as a potential first element of the triplet:  
   - Set `l = i + 1` (left pointer).  
   - Set `r = n - 1` (right pointer).  
   - Compute `requiredSum = target - arr[i]`.  
   - While `l < r`:  
     - If `arr[l] + arr[r] == requiredSum`, return `true`.  
     - Else if `arr[l] + arr[r] < requiredSum`, move `l++`.  
     - Else move `r--`.  
3. If no such triplet is found, return `false`. 

**Time Complexity:** O(n^2)
**Space Complexity:** O(1) (if the sort is done in-place) or O(n) (if a separate array is used for sorting)

**Code Implementation (Java)**  
```java
import java.util.Arrays;

class Solution {
    // Function to check if there exists a triplet with the given sum
    public static boolean hasTripletSum(int arr[], int target) {
        int n = arr.length;
        Arrays.sort(arr);  // Step 1: Sort the array

        // Step 2: Iterate through each element
        for (int i = 0; i < n - 2; i++) {
            int l = i + 1, r = n - 1;
            int requiredSum = target - arr[i];

            // Step 3: Two-pointer approach
            while (l < r) {
                if (arr[l] + arr[r] == requiredSum) {
                    return true;
                }
                if (arr[l] + arr[r] < requiredSum) {
                    l++; // Move left pointer to increase sum
                } else {
                    r--; // Move right pointer to decrease sum
                }
            }
        }
        return false; // No valid triplet found
    }

    public static void main(String[] args) {
        int[] arr = {1, 4, 45, 6, 10, 8};
        int target = 13;
        System.out.println(hasTripletSum(arr, target)); // Output: true
    }
}


### Problem Statement: 38 
**Four Sum - Find all unique quadruplets that sum to a target**  
Given an integer array `nums` and an integer `target`, return all **unique** quadruplets `[nums[a], nums[b], nums[c], nums[d]]` such that:  
- `a`, `b`, `c`, and `d` are distinct indices.
- `nums[a] + nums[b] + nums[c] + nums[d] == target`.

**Input:**  
- An integer array `nums` (can contain both positive and negative numbers).
- An integer `target`.

 **Output:**  
- A list of lists containing all unique quadruplets that sum up to `target`.

**Constraints:**  
- `1 <= nums.length <= 200`
- `-10^9 <= nums[i] <= 10^9`
- `-10^9 <= target <= 10^9`

**Goal of the Code:**  
To find all **unique** sets of four numbers in `nums` that add up to `target` using a two-pointer approach.

**Approach:**  
1. **Sorting the Array**:  
   - Sorting helps efficiently find quadruplets and skip duplicate elements.  
2. **Two Nested Loops for Fixing Two Numbers (`i` and `j`)**:  
   - Iterate over `nums` with `i`, then pick a second number `j` such that `j > i`.  
   - Skip duplicate elements for `i` and `j` to avoid duplicate quadruplets.  
3. **Two Pointer Technique (`k` and `l`)**:  
   - After fixing `i` and `j`, use two pointers (`k` from `j+1`, `l` from `n-1`).  
   - Calculate the sum of `nums[i] + nums[j] + nums[k] + nums[l]`.  
   - If `sum == target`, add the quadruplet to the result and move both pointers.  
   - If `sum < target`, increment `k` to increase sum.  
   - If `sum > target`, decrement `l` to decrease sum.  
   - Skip duplicate values for `k` and `l` to avoid duplicate quadruplets.

**Time and Space Complexity:**
Sorting the array:  O(N log N)
Nested loops:  O(N^2)
Two-pointer search inside the loops:  O(N)
Overall Time Complexity:  O(N^3), which is optimal for a four-sum problem.
Space Complexity:  O(1) (excluding output storage).

**Pseudocode/Algorithm:**  
```plaintext
1. Sort the array.
2. Iterate `i` from `0` to `n-3`:
   - If `i > 0` and `nums[i] == nums[i-1]`, continue (skip duplicate).
3. Iterate `j` from `i+1` to `n-2`:
   - If `j > i+1` and `nums[j] == nums[j-1]`, continue (skip duplicate).
4. Initialize `k = j+1` and `l = n-1` (two-pointer approach).
5. While `k < l`:
   - Compute `sum = nums[i] + nums[j] + nums[k] + nums[l]`.
   - If `sum == target`, add quadruplet to the result.
     - Move `k++`, `l--` and skip duplicate values.
   - If `sum < target`, move `k++`.
   - If `sum > target`, move `l--`.
6. Return result list.
