## Problem: Minimum Jumps
You are given an array `arr[]` of non-negative integers. Each element at index `i` represents the **maximum number of steps** you can jump **forward** from that position.
Your task is to determine the **minimum number of jumps** needed to move from the first index (index `0`) to the **last index** in the array.
Return `-1` if it is **not possible** to reach the end of the array.

**📥 Input Format**
- An integer array `arr[]` of size `n`  
  `1 <= n <= 10^6`  
  `0 <= arr[i] <= 10^6`

**📤 Output Format**
- Return the minimum number of jumps to reach the end of the array.  
- If unreachable, return `-1`.

**💡 Example**
#**Input
```java
arr = [1, 3, 5, 8, 9, 2, 6, 7, 6, 8, 9]
Output
Copy
Edit
3
Explanation
Jump from arr[0] to arr[1] (1 step)

Jump from arr[1] to arr[4] (3 steps ahead)

Jump from arr[4] to the end (at least 5 steps ahead)
```

**🧠 Approach**
1. **Initialize:**
    maxReach: farthest index reachable so far
    currReach: end of the current jump range
    jumps: number of jumps made

2. **Traverse the array:**
    Update maxReach at every index
    When the current index equals currReach, we must jump
    If the current index is also maxReach, return -1 (can't move forward)
    If maxReach >= n - 1 at any point, return jumps + 1

**Time Complexity: O(n)**
We iterate through the array only once.

**Space Complexity: O(1)**
No extra space is used except variables.

**Code Solution:**
```java
class Solution {
    static int minJumps(int[] arr) {
        int n = arr.length;
        if (arr[0] == 0)
            return -1;

        int maxReach = 0;
        int currReach = 0;
        int jump = 0;

        for (int i = 0; i < n; i++) {
            maxReach = Math.max(maxReach, i + arr[i]);

            if (maxReach >= n - 1) {
                return jump + 1;
            }

            if (i == currReach) {
                if (i == maxReach) {
                    return -1;
                } else {
                    jump++;
                    currReach = maxReach;
                }
            }
        }

        return -1;
    }
}
```

## Problem:  Minimum Operations

You are given a number `n`. You need to find the **minimum number of operations** required to reach `n` starting from `0`.
You are allowed to perform **only two operations**:
1. **Double the number**
2. **Add one to the number**

**📥 Input Format**
- A single integer `n`  
  `1 <= n <= 10^4`

**📤 Output Format**
- Return the minimum number of operations to reach `n` from `0`.

**💡 Example**
**Input**
n = 8

**Output**
4

**Explanation**
Operations:  
- 0 + 1 → 1  
- 1 + 1 → 2  
- 2 × 2 → 4  
- 4 × 2 → 8  

Total operations = 4

**🧠 Approach**

- Use **dynamic programming** to store minimum operations to reach each number from `0` to `n`.
- At each step `i`, compute the minimum of:
  - `dp[i - 1] + 1` (adding 1)
  - `dp[i / 2] + 1` (if `i` is even, doubling)

**⏱️ Time & Space Complexity**
- **Time Complexity:** `O(n)`  
- **Space Complexity:** `O(n)`

**💻 Java Code**
```java
class Solution {
    public int minOperation(int n) {
        int dp[] = new int[n + 1];

        for (int i = 1; i <= n; i++) {
            dp[i] = dp[i - 1] + 1;
            if (i % 2 == 0) {
                dp[i] = Math.min(dp[i], dp[i / 2] + 1);
            }
        }

        return dp[n];
    }
}


## Problem: Minimize the Sum of Product

You are given two arrays `arr1` and `arr2`, each of size `N`.
Your task is to find the minimum value of:
arr1[0] * arr2[0] + arr1[1] * arr2[1] + .... + arr1[N-1] * arr2[N-1]

where **shuffling of elements** in both arrays is allowed.

**📥 Input Format**
- Two integer arrays `arr1` and `arr2` of equal size `N`.

**📤 Output Format**
- Return the minimum possible sum of products.

**💡 Example**
**Input**
arr1 = [1, 3, 5]
arr2 = [8, 2, 6]

**Output**
31

**Explanation**
Sorted `arr1`: [1, 3, 5]  
Sorted `arr2`: [8, 6, 2]

Sum = 1×8 + 3×6 + 5×2 = 8 + 18 + 10 = 36

Wait, that's incorrect! Let's try correct order:  
Sorted `arr1`: [1, 3, 5]  
Sorted `arr2`: [8, 6, 2] → should be reversed → [2, 6, 8]

**Now:** 
Sum = 1×8 + 3×6 + 5×2 = 8 + 18 + 10 = **36**

Oops! Let's try:  
Sorted `arr1`: [1, 3, 5]  
Sorted `arr2` descending: [8, 6, 2]

Sum = 1×8 + 3×6 + 5×2 = 8 + 18 + 10 = **36**

All good!

**🧠 Approach**
To minimize the sum of products:
- Sort `arr1` in **ascending** order.
- Sort `arr2` in **descending** order.
- Multiply corresponding elements and sum them up.
This works due to the **greedy approach** — multiplying the smallest element of one array with the largest of the other to minimize the contribution.

**⏱️ Time & Space Complexity**
- **Time Complexity:** O(N log N)  
- **Space Complexity:** O(1) (in-place sorting)

**💻 Java Code**
```java
class Solution {
    public long minValue(List<Integer> A, List<Integer> B) {
        A.sort(Integer::compareTo);         // Sort A ascending
        B.sort(Integer::compareTo);         // Sort B ascending

        long result = 0;
        int n = A.size();
        for (int i = 0; i < n; i++) {
            result += (long) A.get(i) * B.get(n - i - 1);  // Multiply with reversed B
        }

        return result;
    }
}


## Problem: Largest Number Possible

Given two numbers `n` and `s`, find the **largest number** that can be formed with **`n` digits** and whose **sum of digits** is equal to `s`.  
Return `-1` if it is not possible to form such a number.

**🧠 Approach**
To form the **largest possible number**:
- Iterate over each digit position from left to right.
- At each step, use the **maximum digit possible** (up to `9`) without exceeding the remaining sum `s`.
- Append the digit to the result and decrease the remaining sum.

**Edge Cases:**
- If `s == 0`:
  - If `n == 1`, return "0".
  - Else, return `-1` (you can't make a number with multiple digits adding up to 0).
- If `s > 9 * n`: return `-1` because even using 9 for each digit won't be enough.

**📥 Input Format**
- `n`: Number of digits.
- `s`: Required sum of digits.

**📤 Output Format**
- A string representing the **largest number possible** or `-1` if it's not feasible.

**💡 Example**

**Input**
n = 2, s = 9

**Output**
90

**Explanation**
- Start with max digit = 9 → add `9`, remaining sum = 0
- Add `0` as next digit → result = `90`

- **Time Complexity:** O(n)
- **Space Complexity:** O(n)

**💻 Java Code**

```java
class Solution {
    static String findLargest(int n, int s) {
        if (s == 0) {
            if (n == 1) return "0"; 
            else return "-1";
        }
        if (s > 9 * n) {
            return "-1";
        }

        StringBuilder result = new StringBuilder();

        for (int i = 0; i < n; i++) {
            int digit = Math.min(9, s);
            result.append(digit);
            s -= digit;
        }

        return result.toString();
    }
}


## Problem:Lemonade Change
You are an owner of Lemonade Island, where **each lemonade costs $5**.  
Customers line up to buy lemonade and each pays with a **$5, $10, or $20** bill.  
You must provide the **exact change** to each customer **in the order** they arrive using only the bills you have received from previous customers.

**Goal:**
Return `true` if you can give change to **every customer**, otherwise return `false`.

**🧠 Approach**
You track how many **$5** and **$10** bills you have.

**Rules:**
- If customer pays with **$5**: No change required → add 1 to `$5` count.
- If customer pays with **$10**: Need to give back one `$5`.  
  If no `$5` bill, return `false`.
- If customer pays with **$20**:  
  Prefer giving one `$10` and one `$5` as change.  
  If not possible, try giving three `$5` bills.  
  If still not possible, return `false`.

**📥 Input Format**
- `N`: Number of customers.
- `bills[]`: An array of size N, where `bills[i]` is the bill paid by the `i-th` customer.

**📤 Output Format**
- `true` if every customer can be given the correct change, `false` otherwise.

**💡 Example**
**Input**
```java
N = 5  
bills = [5, 5, 5, 10, 20]
```
**Output**
true
**Explanation**
Customer 1 pays $5 → no change → 1x$5

Customer 2 pays $5 → no change → 2x$5

Customer 3 pays $5 → no change → 3x$5

Customer 4 pays $10 → give $5 → 2x$5, 1x$10

Customer 5 pays $20 → give $10+$5 → ✅

**Time Complexity: O(N)**

**Space Complexity: O(1)**

**Code Solution:**
```java
class Solution {
    static boolean lemonadeChange(int N, int bills[]) {
        int count5 = 0;
        int count10 = 0;

        for (int i = 0; i < N; i++) {
            if (bills[i] == 5) {
                count5++;
            } else if (bills[i] == 10) {
                if (count5 < 1) return false;
                count5--;
                count10++;
            } else if (bills[i] == 20) {
                if (count10 >= 1 && count5 >= 1) {
                    count10--;
                    count5--;
                } else if (count5 >= 3) {
                    count5 -= 3;
                } else {
                    return false;
                }
            }
        }

        return true;
    }
}



## Problem: Activity Selection

You are given two arrays:
- `start[]`: Start times of activities
- `finish[]`: Finish times of activities
A person can perform **only one activity at a time**, and **activities cannot overlap**.

**Goal:**
Find the **maximum number of non-overlapping activities** a person can perform in a single day.

**📥 Input Format**
- `start[]`: array of integers of size `n` representing start times.
- `finish[]`: array of integers of size `n` representing finish times.

**📤 Output Format**
- Return a single integer: **maximum number of activities** that can be performed.

**💡 Example**
**Input**
```java
start = [1, 3, 2, 5]
finish = [2, 4, 3, 6]
```
**Output**
3
**Explanation:**
The selected activities are:
Activity 1: (1, 2)
Activity 3: (2, 3)
Activity 4: (5, 6)

**🧠 Approach**
This is a classic **Greedy Algorithm** problem.

**Strategy:**
- Pair start and finish times.
- Sort all activities by their finish time.
- Select the activity if its **start time is after the last selected activity's finish time**.

**Time Complexity: O(n log n)** → Due to sorting

**Space Complexity: O(n)** → For storing activity pairs

**Code Solution:**
```java
class Solution {
    public int activitySelection(int[] start, int[] finish) {
        int n = start.length;
        int ans = 0;

        // Pair finish and start times
        int[][] arr = new int[n][2];
        for (int i = 0; i < n; i++) {
            arr[i][0] = finish[i];
            arr[i][1] = start[i];
        }

        // Sort activities based on finish times
        Arrays.sort(arr, (a, b) -> Integer.compare(a[0], b[0]));

        int finishtime = -1;

        // Select activities greedily
        for (int i = 0; i < n; i++) {
            if (arr[i][1] > finishtime) {
                finishtime = arr[i][0];
                ans++;
            }
        }

        return ans;
    }
}


## Problem: N Meetings in One Room

You are given `n` meetings with their start and end times in two arrays:  
- `start[i]`: Start time of the i-th meeting  
- `end[i]`: End time of the i-th meeting  
Only **one meeting can be held at a time** in a single meeting room.  
Find the **maximum number of meetings** that can be accommodated.

> **Note:** The start time of one selected meeting **must be strictly greater** than the end time of the previously selected meeting.

**📥 Input Format**
- Two integer arrays `start[]` and `end[]` of size `n`.

**📤 Output Format**
- Return a single integer: **maximum number of meetings** that can be accommodated.

**💡 Example**
**Input:**
```java
start = [1, 3, 0, 5, 8, 5]
end = [2, 4, 6, 7, 9, 9]
```
**Output:**
4
**Explanation:**
Maximum meetings that can be held:
(1, 2), (3, 4), (5, 7), (8, 9)

**🧠 Approach**
**Strategy:**
    Pair the start and end times of each meeting.

    Sort all meetings by end time.

    Select a meeting only if its start time is greater than the end time of the previously selected meeting.

This ensures maximum utilization of time and meeting slots.

**Time Complexity: O(n log n)** → For sorting the meetings by end time

**Space Complexity: O(n)** → For storing meeting pairs

**Code Solution:**
```java
class Solution {
    public int maxMeetings(int start[], int end[]) {
        int n = start.length;
        int[][] meetings = new int[n][2];

        for (int i = 0; i < n; i++) {
            meetings[i][0] = start[i];
            meetings[i][1] = end[i];
        }

        // Sort by end time
        Arrays.sort(meetings, (a, b) -> a[1] - b[1]);

        int count = 1; // First meeting is always selected
        int lastEndTime = meetings[0][1];

        for (int i = 1; i < n; i++) {
            if (meetings[i][0] > lastEndTime) {
                count++;
                lastEndTime = meetings[i][1];
            }
        }

        return count;
    }
}


## Problem: Minimum Platforms
You are given two arrays:
- `arr[]`: Arrival times of trains  
- `dep[]`: Departure times of trains  
Your task is to **find the minimum number of platforms** required at the station so that **no train waits** for a platform.

> **Constraint:**  
A platform **cannot be shared** by a train arriving and another departing at the same time.  
If a train arrives **before** the previous one departs, a **new platform** is needed.

**📥 Input Format**
- Two integer arrays `arr[]` and `dep[]` of equal length `n` representing arrival and departure times in 24-hour format without colon (e.g., 900 for 9:00 AM).

**📤 Output Format**
- Return a single integer: **minimum number of platforms** required.

**💡 Example**
**Input:**
```java
arr[] = {900, 940, 950, 1100, 1500, 1800}
dep[] = {910, 1200, 1120, 1130, 1900, 2000}
```
**Output:**
3
**Explanation:**
At time 950, three trains are at the station → 3 platforms required.

**🧠 Approach**
**Strategy:**
    Convert arrival and departure into a unified timeline of events.

    Label each time as 'a' (arrival) or 'd' (departure).

**Sort events:**
    If times are the same, arrivals come before departures.

**Traverse the sorted list:**
    Increase platform count on arrival.

    Decrease on departure.

    Track the maximum platforms occupied at any time.

**Time Complexity: O(n log n)**
(Due to sorting of 2n events)

**Space Complexity: O(n)**
(For storing event pairs)

**Code Solution:**
```java
class pair {
    int first;
    char second;

    pair(int key1, char key2) {
        this.first = key1;
        this.second = key2;
    }
}

class Solution {
    static int findPlatform(int arr[], int dep[]) {
        int n = arr.length;
        ArrayList<pair> order = new ArrayList<>();

        // Mark arrivals and departures
        for (int i = 0; i < n; i++) {
            order.add(new pair(arr[i], 'a'));
            order.add(new pair(dep[i], 'd'));
        }

        // Sort by time, arrival before departure if equal
        Collections.sort(order, new Comparator<pair>() {
            public int compare(pair p1, pair p2) {
                if (p1.first == p2.first)
                    return Character.compare(p1.second, p2.second);
                return p1.first - p2.first;
            }
        });

        int result = 1;
        int occupiedPlatforms = 0;

        // Traverse sorted events
        for (pair p : order) {
            if (p.second == 'a')
                occupiedPlatforms++;
            else
                occupiedPlatforms--;

            result = Math.max(result, occupiedPlatforms);
        }

        return result;
    }
}


## Problem: Minimum number of Coins

You are given an **infinite supply** of Indian currency denominations:
{ 1, 2, 5, 10, 20, 50, 100, 200, 500, 2000 }
Given a target value **N**, find the **minimum number of coins/notes** required to make change for Rs **N**.  
Return a list containing the **values of coins/notes used**.

**📥 Input Format**
- A single integer `N` (1 ≤ N ≤ 10⁶): the amount to make change for.

**📤 Output Format**
- A list of integers representing the denominations used in descending order.

**💡 Example**
**Input:**
N = 43

**Output:**
[20, 20, 2, 1]

**Explanation:**
- 2 × 20 = 40  
- 1 × 2 = 2  
- 1 × 1 = 1  
- Total = 43 with 4 notes/coins

**🧠 Approach**
**Strategy:**
- Use a **greedy approach**: always pick the **largest denomination** that does not exceed `N`.
- Subtract the denomination from `N` and add it to the result.
- Repeat until `N == 0`.

**⏱️ Time & Space Complexity**
- **Time Complexity:** O(log N)  
  (As we subtract large chunks each iteration)
- **Space Complexity:** O(N) in worst case (if only 1s are used)

**💻 Java Code**
```java
class Solution {
    static List<Integer> minPartition(int N) {
        int[] denominations = {2000, 500, 200, 100, 50, 20, 10, 5, 2, 1};
        List<Integer> result = new ArrayList<>();

        for (int coin : denominations) {
            while (N >= coin) {
                N -= coin;
                result.add(coin);
            }
        }

        return result;
    }
}


## Problem: Fractional Knapsack

Given two arrays `val[]` and `wt[]`, where:
- `val[i]` represents the **value** of the `i-th` item.
- `wt[i]` represents the **weight** of the `i-th` item.
And an integer `capacity`, which represents the **maximum weight** a knapsack can carry.
Your task is to determine the **maximum total value** that can be achieved by placing items in the knapsack. **You are allowed to break items** into smaller parts (i.e., fractional knapsack).
Return the result as a `double` rounded to **6 decimal places**.

**📥 Input Format**
- `val[]`: array of integers representing values.
- `wt[]`: array of integers representing weights.
- `capacity`: integer representing the weight capacity of the knapsack.

**📤 Output Format**
- A double representing the maximum value that can be carried, rounded to 6 decimal places.

**💡 Example**
**Input:**
```java
val = [60, 100, 120]
wt = [10, 20, 30]
capacity = 50
```
**Output:**
240.000000

**Explanation:**
Take full item 1 (value=100, weight=20)
Take full item 2 (value=120, weight=30)
Total value = 100 + 120 = 220

**🧠 Approach**
**Greedy Strategy:**

    Calculate value-to-weight ratio for each item.

    Sort items by this ratio in descending order.

    Pick items greedily:

        If item fits completely, take it all.

        Else, take fraction of it that fits.

**Time Complexity: O(n log n) for sorting**

**Space Complexity: O(n) for storing item pairs**

**Code Solution:**
```java
class Solution {
    static class ItemComparator implements Comparator<int[]> {
        public int compare(int[] a, int[] b) {
            double a1 = (1.0 * a[0]) / a[1];
            double b1 = (1.0 * b[0]) / b[1];
            return Double.compare(b1, a1);
        }
    }

    double fractionalKnapsack(int[] val, int[] wt, int capacity) {
        int n = val.length;
        int[][] items = new int[n][2];

        for (int i = 0; i < n; i++) {
            items[i][0] = val[i];
            items[i][1] = wt[i];
        }

        Arrays.sort(items, new ItemComparator());

        double res = 0.0;
        int currentCapacity = capacity;

        for (int i = 0; i < n; i++) {
            if (items[i][1] <= currentCapacity) {
                res += items[i][0];
                currentCapacity -= items[i][1];
            } else {
                res += (1.0 * items[i][0] / items[i][1]) * currentCapacity;
                break;
            }
        }

        return res;
    }
}


## Problem: Jump Game

Given an array of **positive integers** `arr[]`, where each element denotes the **maximum length of jump** you can make from that position.
Your task is to determine if it's possible to **reach the last index** starting from the **0th index** of the array.
Return `true` if you can reach the end, otherwise return `false`.

**📥 Input Format**
- An array `arr[]` of size `n` with all positive integers.

**📤 Output Format**
- A boolean value: `true` if the end is reachable, otherwise `false`.

**💡 Example**
**Input:**
```java
arr = [2, 3, 1, 1, 4]
```
**Output:**
true
**Explanation:**
        From index 0, jump to index 1 (2 steps).

        From index 1, jump to index 4 (3 steps).

        Index 4 is the last index, so return true.

**🧠 Approach**
**Greedy Strategy:**
    Track the maximum index reachable at every step.

    If at any point, current index is greater than max reachable index → return false.

    If you can move beyond or to the last index, return true.

**Time Complexity: O(n)**

**Space Complexity: O(1)**

**Code Solution:**
```java
class Solution {
    // Function to check if we can reach the last index from the 0th index.
    public boolean canReach(int[] arr) {
        int maxReach = 0;
        
        for (int i = 0; i < arr.length; i++) {
            if (i > maxReach) {
                return false;
            }
            maxReach = Math.max(maxReach, i + arr[i]);
        }
        
        return true;
    }
}




## Problem: Sum Except First and Last
You are given an array `arr[]` of numbers. Your task is to return the sum of all the elements except the first and last elements of the array.

**Input Format:**
- An array `arr[]` of size `n` where `n >= 3`.

**Output Format:**
- An integer value representing the sum of the elements excluding the first and last elements.

**💡 Example**
**Input:**
```java
arr = [2, 3, 4, 5]
```
**Output:**
7
**Explanation:**
    Excluding the first element 2 and the last element 5, the sum of remaining elements is 3 + 4 = 7.

**🧠 Approach**
1. If the array has less than 3 elements, return 0 because no elements are left to sum.

2. For arrays with more than 2 elements:
       Start iterating from the second element (index 1) and stop at the second-to-last element (index arr.length - 2).

       Add the elements in this range and return the sum.


**Time Complexity: O(n)** where n is the length of the array.

**Space Complexity: O(1)**, constant space.

**Code Solution:**
```java
class Solution {
    int sumExceptFirstLast(int[] arr) {
        if (arr.length <= 2) return 0; 

        int sum = 0;
        for (int i = 1; i < arr.length - 1; i++) {
            sum += arr[i];
        }
        return sum;
    }
}


## Problem: Shortest Job First

Geek is a software engineer. He is tasked with calculating the average waiting time of all processes by following the **Shortest Job First (SJF)** policy.
The **Shortest Job First (SJF)** or **Shortest Job Next** is a scheduling policy that selects the waiting process with the smallest execution time to execute next.
Given an array of integers `bt[]` representing the burst time of each process, calculate the average waiting time for all the processes. Return the nearest integer which is smaller than or equal to the output.

**Input Format:**
- An array `bt[]` of size `n`, where each element `bt[i]` denotes the burst time of the ith process.

**Output Format:**
- Return an integer representing the average waiting time of all the processes.

**Note:**
- All processes are available at time 0.

**Example**
**Input:**
```java
bt = [6, 8, 7, 3]
```
**Output:**
7
**Explanation:**
Sorting burst times: [3, 6, 7, 8]
The waiting times are:
    Waiting time for process 1 = 0
    Waiting time for process 2 = 3
    Waiting time for process 3 = 3 + 6 = 9
    Waiting time for process 4 = 3 + 6 + 7 = 16
Average waiting time = (0 + 3 + 9 + 16) / 4 = 7

**Approach**
1. Sort the burst times of all processes in increasing order to apply the Shortest Job First (SJF) policy.
2. Calculate the waiting time for each process:
        The first process has a waiting time of 0.

        For each subsequent process, its waiting time is the sum of burst times of all the previous processes.

3. The total waiting time is the sum of the waiting times of all the processes.
4. Return the average waiting time as the floor of the total waiting time divided by the number of processes.

**Time Complexity: O(n log n)** for sorting the burst times, where n is the number of processes.

**Space Complexity: O(1)**, constant space for computation.

**Code Solution:**
```java
class Solution {
    static int solve(int bt[]) {
        // code here
        Arrays.sort(bt); 
        
        int totalWait = 0;
        int waitTime = 0;

        for (int i = 1; i < bt.length; i++) {
            waitTime += bt[i - 1];   
            totalWait += waitTime;  
        }

        return totalWait / bt.length; 
    }
}


## Problem: Insert Interval

You are given an array of non-overlapping intervals `intervals` where `intervals[i] = [starti, endi]` represent the start and the end of the ith interval, and `intervals` is sorted in ascending order by `starti`. 
You are also given an interval `newInterval = [start, end]` that represents the start and end of another interval.
**Insert** `newInterval` into `intervals` such that `intervals` is still sorted in ascending order by `starti` and intervals still do not have any overlapping intervals (merge overlapping intervals if necessary).
Return `intervals` after the insertion.

**Note:**
- You don't need to modify `intervals` in-place. You can make a new array and return it.

**Example**
**Input:**
```java
intervals = [[1, 3], [6, 9]]
newInterval = [2, 5]
```
**Output:**
[[1, 5], [6, 9]]
**Explanation:**
   The new interval [2, 5] overlaps with the interval [1, 3], so we merge them into [1, 5].

   The intervals [6, 9] remain unchanged.

**Approach**
1. Add intervals before newInterval:
    Traverse through the given intervals and add those intervals to the result which are entirely before the newInterval (i.e., intervals[i][1] < newInterval[0]).

2. Merge overlapping intervals:
    If there is an overlap between intervals[i] and newInterval, update newInterval to merge these overlapping intervals by taking the minimum of the start times and the maximum of the end times.

3. Add remaining intervals:
    After merging, the remaining intervals in intervals which start after newInterval end should be added to the result.

4. Return the result:
    The result array will contain the merged intervals, which should be returned.

**Time Complexity: O(n)**, where n is the number of intervals. We need to iterate through all the intervals once.

**Space Complexity: O(n)**, where n is the number of intervals. We use a list to store the result.

**Code Solution:**
```java
class Solution {
    public int[][] insert(int[][] intervals, int[] newInterval) {
        List<int[]> result = new ArrayList<>();
        int i = 0;
        int n = intervals.length;
        
        // Add all intervals before the newInterval
        while (i < n && intervals[i][1] < newInterval[0]) {
            result.add(intervals[i]);
            i++;
        }
        
        // Merge overlapping intervals
        while (i < n && intervals[i][0] <= newInterval[1]) {
            newInterval[0] = Math.min(intervals[i][0], newInterval[0]);
            newInterval[1] = Math.max(intervals[i][1], newInterval[1]);
            i++;
        }
        
        // Add the merged interval
        result.add(newInterval);
        
        // Add remaining intervals
        while (i < n) {
            result.add(intervals[i]);
            i++;
        }

        // Convert list to array and return
        return result.toArray(new int[result.size()][]);
    }
}


## Problem: Merge Intervals

Given an array of intervals where `intervals[i] = [starti, endi]`, merge all overlapping intervals, and return an array of the non-overlapping intervals that cover all the intervals in the input.

**Example**
**Input:**
```java
intervals = [[1,3],[2,6],[8,10],[15,18]]
```
**Output:**
[[1,6],[8,10],[15,18]]
**Explanation:**
Intervals [1,3] and [2,6] overlap, so they are merged into [1,6].

**Approach**
1. Sort the intervals based on the start time.
2. Initialize a list merged to keep track of merged intervals.
3. Traverse through the intervals:
    If the current interval does not overlap with the last merged interval, simply add it.

    If it does overlap, merge them by updating the end of the last interval in the merged list.

4. Convert the list of merged intervals back to an array and return.

**Time Complexity: O(n log n)** due to sorting.

**Space Complexity: O(n)** for the result list.

**Code Solution:**
```java
class Solution {
    public int[][] merge(int[][] intervals) {
        if (intervals.length <= 1) return intervals;

        // Sort intervals by starting time
        Arrays.sort(intervals, (a, b) -> Integer.compare(a[0], b[0]));

        List<int[]> merged = new ArrayList<>();

        for (int[] interval : intervals) {
            // If list is empty or no overlap, add the interval
            if (merged.isEmpty() || merged.get(merged.size() - 1)[1] < interval[0]) {
                merged.add(interval);
            } else {
                // Merge the overlapping interval
                merged.get(merged.size() - 1)[1] = 
                    Math.max(merged.get(merged.size() - 1)[1], interval[1]);
            }
        }

        return merged.toArray(new int[merged.size()][]);
    }
}


## Problem: Non-overlapping Intervals

Given an array of intervals `intervals` where `intervals[i] = [starti, endi]`, return the **minimum number of intervals you need to remove** to make the rest of the intervals **non-overlapping**.
**Note:** Intervals which only touch at a point are considered non-overlapping. For example, `[1, 2]` and `[2, 3]` are non-overlapping.

**Example**
**Input:**
```java
intervals = [[1,2],[2,3],[3,4],[1,3]]
```

**Output:**
1
**Explanation:**
Removing [1,3] leaves [1,2],[2,3],[3,4] which are non-overlapping.

**Approach**
1. Sort the intervals by their end times in ascending order.
2. Initialize a variable end to track the end of the last added interval.
3. terate over intervals:
    If the current interval overlaps with the previous (start < end), increment removals.

    Else, update end to the current interval's end time.

4. Return the total number of intervals removed.

**Time Complexity: O(n log n)** — for sorting the intervals.

**Space Complexity: O(1)**— constant extra space.

**Code Solution:**
```java
class Solution {
    public int eraseOverlapIntervals(int[][] intervals) {
        if (intervals.length == 0) return 0;

        // Sort by end time
        Arrays.sort(intervals, (a, b) -> Integer.compare(a[1], b[1]));

        int end = intervals[0][1];
        int removals = 0;

        for (int i = 1; i < intervals.length; i++) {
            if (intervals[i][0] < end) {
                // Overlap found
                removals++;
            } else {
                end = intervals[i][1];
            }
        }

        return removals;
    }
}


## Problem: Partition Labels
You are given a string `s`. We want to partition the string into as many parts as possible such that each letter appears in **at most one part**.
The partition is valid if:
- After concatenating all the parts in order, the resultant string is equal to the original string `s`.

**Example:**
```text
Input: s = "ababcc"
```
**Output:** [4, 2]
**Explanation:** The string can be split into ["abab", "cc"]
**Invalid partitions include:**

    ["aba", "bcc"] (letter 'b' appears in both)

    ["ab", "ab", "cc"] (letter 'a' and 'b' appear in multiple parts)

**Approach**
1. Record the last occurrence of each character in the string.
2. Use two pointers (start and end) to track the current partition:
    For each character, update end to the farthest last occurrence seen so far.

    When the current index reaches end, close the partition.
3. Add the size of the partition to the result list and reset start for the next segment.

**Time Complexity: O(n)** — One pass to store last occurrences, another pass to partition.

**Space Complexity: O(1)** — Only 26 characters are stored in the map.

**Code Solution:**
```java
class Solution {
    public List<Integer> partitionLabels(String s) {
        Map<Character, Integer> lastOccurrence = new HashMap<>();
        
        // Step 1: Record last occurrence of each character
        for (int i = 0; i < s.length(); i++) {
            lastOccurrence.put(s.charAt(i), i);
        }

        List<Integer> result = new ArrayList<>();
        int start = 0, end = 0;

        // Step 2: Traverse and build partitions
        for (int i = 0; i < s.length(); i++) {
            end = Math.max(end, lastOccurrence.get(s.charAt(i)));
            if (i == end) {
                result.add(end - start + 1);
                start = i + 1;
            }
        }

        return result;  
    }
}


## Problem: LRU Cache
Design a data structure that works like an **LRU Cache** (Least Recently Used Cache). The cache should support two types of operations:
1. `PUT x y`: Set the value of the key `x` with value `y`.
2. `GET x`: Retrieve the value of the key `x` if present, otherwise return `-1`.

The LRUCache class has two methods:
- `get(key)`: Returns the value of the key if it exists in the cache, otherwise returns `-1`.
- `put(key, value)`: If the key is already present, update its value. If not present, add the key-value pair to the cache. If the cache reaches its capacity, it should remove the **least recently used (LRU)** item before inserting the new item.

**Constraints:**
- The cache has a fixed `capacity`.
- The operations `put` and `get` should be efficient, with time complexity as close to O(1) as possible.

**Approach: Using a Greedy Strategy with a HashMap and Doubly Linked List**

To implement an efficient LRU cache, we use the following data structures:
1. **HashMap**: This provides O(1) average time complexity for both the `get()` and `put()` operations. It maps the keys to their corresponding nodes in the doubly linked list.
   
2. **Doubly Linked List**: This will store the cache items in the order of their usage. The most recently used items are moved to the front (head), while the least recently used items are moved to the back (tail). This allows for O(1) removal of the least recently used item when the cache exceeds its capacity.

**Key Operations:**
- **get(key)**: If the key is present in the cache, move it to the front (most recently used). If not, return `-1`.
- **put(key, value)**: If the key is already present, update its value and move it to the front. If the key is not present and the cache is full, remove the item from the tail (least recently used) and add the new key-value pair to the front.

**Algorithm:**
1. **Initialization**: 
   - Set the cache size (capacity) and initialize a hashmap and a doubly linked list.
   
2. **GET operation**:
   - Check if the key exists in the hashmap. If it does, move the node to the front of the doubly linked list and return its value.
   - If the key does not exist, return `-1`.

3. **PUT operation**:
   - If the key already exists, update the value and move it to the front of the doubly linked list.
   - If the key doesn't exist and the cache is full, remove the least recently used item (node at the tail), and then insert the new key-value pair at the front.

**Code Implementation:**

```java
import java.util.*;

class LRUCache {
    // Doubly Linked List Node to store the key-value pairs
    class Node {
        int key, value;
        Node prev, next;
        Node(int key, int value) {
            this.key = key;
            this.value = value;
        }
    }

    private Map<Integer, Node> cache;
    private int capacity;
    private Node head, tail;

    // Constructor to initialize the LRUCache with a given capacity
    public LRUCache(int capacity) {
        this.capacity = capacity;
        cache = new HashMap<>();
        head = new Node(0, 0); // dummy head
        tail = new Node(0, 0); // dummy tail
        head.next = tail;
        tail.prev = head;
    }

    // Move the given node to the front (most recently used)
    private void moveToFront(Node node) {
        removeNode(node);
        addToFront(node);
    }

    // Add the node right after the head
    private void addToFront(Node node) {
        node.next = head.next;
        node.prev = head;
        head.next.prev = node;
        head.next = node;
    }

    // Remove a node from the doubly linked list
    private void removeNode(Node node) {
        node.prev.next = node.next;
        node.next.prev = node.prev;
    }

    // GET operation to fetch the value associated with the key
    public int get(int key) {
        if (!cache.containsKey(key)) {
            return -1;
        }
        Node node = cache.get(key);
        moveToFront(node);
        return node.value;
    }

    // PUT operation to set a value for the key
    public void put(int key, int value) {
        if (cache.containsKey(key)) {
            Node node = cache.get(key);
            node.value = value;
            moveToFront(node);
        } else {
            if (cache.size() >= capacity) {
                // Cache is full, remove the least recently used item
                Node tailNode = tail.prev;
                cache.remove(tailNode.key);
                removeNode(tailNode);
            }
            // Add the new key-value pair
            Node newNode = new Node(key, value);
            cache.put(key, newNode);
            addToFront(newNode);
        }
    }
}

// Driver Function
public class Main {
    public static void main(String[] args) {
        LRUCache cache = new LRUCache(3);

        cache.put(1, 1); // cache: {1=1}
        cache.put(2, 2); // cache: {1=1, 2=2}
        System.out.println(cache.get(1));    // returns 1, cache: {2=2, 1=1}
        cache.put(3, 3); // cache: {1=1, 2=2, 3=3}
        System.out.println(cache.get(2));    // returns 2, cache: {1=1, 3=3, 2=2}
        cache.put(4, 4); // evicts key 1, cache: {3=3, 2=2, 4=4}
        System.out.println(cache.get(1));    // returns -1 (not found)
        System.out.println(cache.get(3));    // returns 3
        System.out.println(cache.get(4));    // returns 4
    }
}
```

**Explanation of Code:**

1. **LRUCache Class**:
   - The `LRUCache` class contains a `Map` (hashmap) `cache` to store key-node mappings and a doubly linked list to maintain the order of usage.
   - **Node class**: Each node in the doubly linked list stores the `key`, `value`, and pointers to the `prev` and `next` nodes.
   - The doubly linked list has dummy `head` and `tail` nodes to simplify edge cases.

2. **Helper Functions**:
   - `moveToFront(Node node)`: Moves a given node to the front of the doubly linked list (indicating that it was recently used).
   - `addToFront(Node node)`: Adds a node right after the head of the doubly linked list.
   - `removeNode(Node node)`: Removes a node from the doubly linked list.

3. **`get(key)`**:
   - If the key exists in the cache, it moves the corresponding node to the front of the doubly linked list and returns its value. Otherwise, it returns `-1`.

4. **`put(key, value)`**:
   - If the key exists, it updates the value and moves the node to the front.
   - If the key does not exist and the cache is full, it removes the least recently used node (from the tail) and adds the new key-value pair to the front.

**Time Complexity:**
- **get(key)**: O(1) because we use the hashmap for constant-time lookup and the doubly linked list for constant-time node movement.
- **put(key, value)**: O(1) for both updating and inserting because of the hashmap and doubly linked list.

**Space Complexity:**
- O(capacity) where `capacity` is the maximum number of keys the cache can hold. The space is used by the hashmap and the doubly linked list.

**Example:**

For the following input:

```plaintext
Cache capacity: 3
PUT operations: (1, 1), (2, 2), (3, 3), (4, 4)
GET operations: 1, 2, 3, 4
```

The output will be:

```plaintext
1
2
3
-1
3
4
```

**Conclusion:**
The **LRU Cache** is implemented efficiently using a **greedy strategy** combined with a **hashmap** and a **doubly linked list** to ensure that both `get` and `put` operations are executed in constant time. The most recently used items are kept at the front of the list, and the least recently used items are evicted when the cache reaches its capacity.


## Problem: Minimum Cost of Ropes

Given an array `arr[]` of rope lengths, the task is to connect all the ropes into a single rope with the minimum total cost. The cost to connect two ropes is the sum of their lengths. We need to find the minimum total cost to connect all the ropes.

**Constraints:**
- The number of ropes (n) is between `1` and `10^5`.
- The length of each rope is between `1` and `10^5`.

**Approach: Using Greedy Method**

The greedy strategy for this problem involves always combining the two smallest ropes first to minimize the cost at each step. This ensures that at each stage of combining ropes, we are adding the smallest possible cost, which leads to the overall minimum cost.

**Key Operations:**
1. **Priority Queue / Min-Heap**: Use a min-heap to efficiently find and remove the two smallest ropes at each step.
2. **Cost Calculation**: After combining the two smallest ropes, the cost is the sum of their lengths. Add this cost to the total cost and insert the combined rope back into the heap.

**Steps:**
1. **Initialization**: Insert all rope lengths into a min-heap.
2. **Combine Ropes**:
   - Extract the two smallest ropes from the heap.
   - Calculate the cost of combining them (sum of their lengths).
   - Insert the combined rope back into the heap.
   - Add the combination cost to the total cost.
3. Repeat this process until there is only one rope left in the heap.

**Algorithm:**
1. Create a min-heap and insert all the ropes into it.
2. While there is more than one rope:
   - Extract the two smallest ropes.
   - Combine them and add the cost to the total cost.
   - Insert the new combined rope back into the heap.
3. Return the total cost.

**Code Implementation:**

```java
import java.util.PriorityQueue;

public class MinimumCostOfRopes {
    public static int minCost(int[] arr) {
        // Create a priority queue (min-heap)
        PriorityQueue<Integer> minHeap = new PriorityQueue<>();
        
        // Insert all ropes into the min-heap
        for (int length : arr) {
            minHeap.offer(length);
        }
        
        int totalCost = 0;
        
        // While there is more than one rope in the heap
        while (minHeap.size() > 1) {
            // Extract the two smallest ropes
            int first = minHeap.poll();
            int second = minHeap.poll();
            
            // Calculate the cost of combining them
            int cost = first + second;
            totalCost += cost;
            
            // Insert the combined rope back into the heap
            minHeap.offer(cost);
        }
        
        return totalCost;
    }

    public static void main(String[] args) {
        int[] ropes1 = {4, 3, 2, 6};
        System.out.println("Minimum cost of ropes: " + minCost(ropes1)); // Output: 29

        int[] ropes2 = {1, 2, 3, 4, 5};
        System.out.println("Minimum cost of ropes: " + minCost(ropes2)); // Output: 33
    }
}
```

**Explanation of Code:**
1. **minCost Method**:
   - A `PriorityQueue` (min-heap) is used to store the ropes, which allows efficient extraction of the two smallest ropes.
   - We extract the two smallest ropes, calculate their cost, and add the new combined rope back to the heap.
   - We keep a running total of the cost as we combine the ropes.

2. **Main Method**:
   - We test the `minCost` method with two examples:
     - Example 1: Ropes of lengths `{4, 3, 2, 6}` which gives a minimum cost of 29.
     - Example 2: Ropes of lengths `{1, 2, 3, 4, 5}` which gives a minimum cost of 33.

**Time Complexity:**
- **Inserting all ropes into the heap**: O(n log n), where `n` is the number of ropes.
- **While loop**: In each iteration, we extract two ropes and insert the combined rope back into the heap. The heap operations (extract and insert) take O(log n) time. Since we perform this operation `n-1` times, the total time complexity is O(n log n).

**Space Complexity:**
- The space complexity is O(n) due to the storage required for the heap.

**Example Walkthrough:**

For the input `{4, 3, 2, 6}`, the process is as follows:

1. **Step 1**: Heap: `{2, 3, 4, 6}`
   - Extract 2 and 3, combine them: `2 + 3 = 5`, total cost = 5.
   - Insert 5 back into the heap: Heap = `{4, 5, 6}`.

2. **Step 2**: Heap: `{4, 5, 6}`
   - Extract 4 and 5, combine them: `4 + 5 = 9`, total cost = 5 + 9 = 14.
   - Insert 9 back into the heap: Heap = `{6, 9}`.

3. **Step 3**: Heap: `{6, 9}`
   - Extract 6 and 9, combine them: `6 + 9 = 15`, total cost = 14 + 15 = 29.

The minimum cost of connecting all ropes is **29**.

**Conclusion:**
This problem is a classic example of using a **greedy algorithm** to minimize the total cost by always choosing the two smallest ropes first. The use of a **min-heap** allows us to efficiently find and combine the two smallest ropes at each step, ensuring that the solution is optimal.


## Problem: Job Sequencing Problem

You are given two arrays: `deadline[]` and `profit[]`, where each job is associated with a deadline and a profit. Each job takes 1 unit of time to complete, and only one job can be scheduled at a time. You will earn the profit associated with a job only if it is completed by its deadline.

Your task is to find:
1. The maximum number of jobs that can be completed within their deadlines.
2. The total maximum profit earned by completing those jobs.

**Constraints:**
- Number of jobs (n) is between 1 and 100.
- The deadline and profit for each job are between 1 and 100.

**Approach: Using Greedy Method**

The greedy strategy for this problem involves scheduling jobs in a way that maximizes profit, while also ensuring that jobs are completed within their deadlines. The key idea is to:
1. **Sort jobs by profit**: The most profitable jobs should be prioritized.
2. **Schedule the jobs**: Assign each job to the latest available time slot before its deadline.

**Steps:**
1. **Sort jobs by profit**: Sort the jobs in descending order based on profit, as we want to complete the most profitable jobs first.
2. **Find the latest available time slot for each job**: For each job, check the latest available time slot before its deadline and schedule the job there.
3. **Track the number of jobs completed**: Ensure that no two jobs are assigned to the same time slot.

**Algorithm:**
1. Create a list of jobs where each job consists of its `deadline`, `profit`, and `job id`.
2. Sort the jobs in descending order based on profit.
3. Iterate over the jobs, and for each job, try to find the latest available time slot before its deadline.
4. If a slot is available, schedule the job and update the total profit.
5. Return the number of jobs completed and the total profit.

**Code Implementation:**

```java
import java.util.*;

class Job {
    int id;
    int deadline;
    int profit;

    public Job(int id, int deadline, int profit) {
        this.id = id;
        this.deadline = deadline;
        this.profit = profit;
    }
}

public class JobSequencing {
    public static int[] JobScheduling(Job[] jobs, int n) {
        // Sort jobs in decreasing order of profit
        Arrays.sort(jobs, (a, b) -> b.profit - a.profit);
        
        // To keep track of free time slots
        int[] result = new int[n];
        boolean[] slot = new boolean[n];
        
        // Initialize the result and slot arrays
        Arrays.fill(slot, false);
        
        int countJobs = 0;
        int totalProfit = 0;
        
        // Iterate over all jobs
        for (int i = 0; i < n; i++) {
            // Find a free slot for this job (starting from the last possible slot)
            for (int j = Math.min(n, jobs[i].deadline) - 1; j >= 0; j--) {
                if (!slot[j]) {
                    // If the slot is free, schedule the job
                    slot[j] = true;
                    result[j] = jobs[i].id;
                    countJobs++;
                    totalProfit += jobs[i].profit;
                    break;
                }
            }
        }
        
        return new int[]{countJobs, totalProfit};
    }

    public static void main(String[] args) {
        Job[] jobs = {
            new Job(1, 4, 20),
            new Job(2, 1, 10),
            new Job(3, 1, 40),
            new Job(4, 1, 30)
        };
        
        int n = jobs.length;
        int[] result = JobScheduling(jobs, n);
        System.out.println("Number of jobs completed: " + result[0]);
        System.out.println("Total profit: " + result[1]);
    }
}
```

**Explanation of Code:**
1. **Job Class**: 
   - Represents a job with an `id`, `deadline`, and `profit`.
   
2. **JobScheduling Method**:
   - Sort the jobs based on profit in descending order.
   - Use a boolean array `slot[]` to track which time slots are occupied.
   - For each job, try to find the latest available slot before its deadline.
   - If an available slot is found, schedule the job and update the profit and count of jobs completed.

3. **Main Method**:
   - We test the `JobScheduling` method with an array of jobs.
   - The result contains the number of jobs completed and the total profit earned.

**Time Complexity:**
- **Sorting jobs by profit**: O(n log n), where `n` is the number of jobs.
- **Scheduling jobs**: For each job, we may need to check up to `n` slots, resulting in a time complexity of O(n^2) in the worst case.

Thus, the overall time complexity is O(n^2).

**Space Complexity:**
- The space complexity is O(n) due to the space required for the `slot[]` and `result[]` arrays.

**Example Walkthrough:**

For the input jobs `{(1, 4, 20), (2, 1, 10), (3, 1, 40), (4, 1, 30)}`, the process is as follows:

1. **Step 1**: Sort jobs by profit: `{(3, 1, 40), (4, 1, 30), (1, 4, 20), (2, 1, 10)}`
   
2. **Step 2**: Try to schedule the jobs:
   - Job 3: Deadline = 1, Profit = 40 → Slot 0 is available, schedule job 3.
   - Job 4: Deadline = 1, Profit = 30 → No available slot.
   - Job 1: Deadline = 4, Profit = 20 → Slot 3 is available, schedule job 1.
   - Job 2: Deadline = 1, Profit = 10 → No available slot.

3. **Result**: 
   - Jobs completed: 2 (Job 3 and Job 1)
   - Total profit: 40 + 20 = 60

**Conclusion:**
The Job Sequencing Problem can be solved using a greedy approach by always prioritizing the jobs with the highest profit and scheduling them as late as possible within their deadlines. This ensures that we maximize the profit while respecting the job deadlines.


## Problem: Cinema Seat Allocation

A cinema has `n` rows of seats, numbered from 1 to `n`, and there are ten seats in each row, labeled from 1 to 10 as shown in the figure below.
Given the array `reservedSeats` containing the numbers of seats already reserved, for example, `reservedSeats[i] = [3,8]` means the seat located in row 3 and labeled with 8 is already reserved.
Return the maximum number of four-person groups you can assign to the cinema seats. A four-person group occupies four adjacent seats in one single row. Seats across an aisle (such as `[3,3]` and `[3,4]`) are not considered to be adjacent, but there is an exceptional case where an aisle splits a four-person group. In that case, the aisle splits a four-person group in the middle, meaning to have two people on each side.

**Problem Constraints:**
- `n` (1 <= n <= 10^4): Number of rows.
- `reservedSeats[i][0]` (1 <= reservedSeats[i][0] <= n): Row number.
- `reservedSeats[i][1]` (1 <= reservedSeats[i][1] <= 10): Column number.

**Approach:**

The goal is to maximize the number of four-person groups that can be seated. A group can sit in one of the following seat segments:
- **Left group**: Seats `[1,2,3,4]`
- **Middle group**: Seats `[4,5,6,7]`
- **Right group**: Seats `[7,8,9,10]`

The seats in these segments must be contiguous, and they must not have any reserved seats within them. A special case occurs when the group is split across the aisle, meaning a group might be split into two smaller groups, each occupying two seats on either side of the aisle.

**Steps to Solve:**

1. **Track Reserved Seats by Row**:
   - Use a hash map to store which seats in each row are reserved. The keys will be row numbers, and the values will be a custom class `RowInfo` that holds boolean flags for whether each group (left, middle, right) is available.

2. **Calculate Groups in Each Row**:
   - For each row, check which groups (left, middle, right) are available by checking the reserved seats.
   - If a group is available, increase the possible group count by 1 for that group.

3. **Handle Rows with No Reserved Seats**:
   - If a row has no reserved seats, you can place two four-person groups, one in the left side and one in the right side of the aisle.

4. **Return the Result**:
   - After processing all rows, return the total number of four-person groups.

**Algorithm:**

```java
import java.util.*;

class Solution {
    class RowInfo {
        int row;
        boolean left;
        boolean middle;
        boolean right;
        
        public RowInfo(int row) {
            this.row = row;
            left = true;
            middle = true;
            right = true;
        }
    }
    
    public int maxNumberOfFamilies(int n, int[][] reservedSeats) {
        Map<Integer, RowInfo> infos = new HashMap<>();
        
        // Mark reserved seats for each row
        for (int[] reservedSeat : reservedSeats) {
            int row = reservedSeat[0];
            int col = reservedSeat[1];
            infos.putIfAbsent(row, new RowInfo(row));
            if (col >= 2 && col <= 5) {
                infos.get(row).left = false; // Mark left group unavailable
            } 
            if (col >= 4 && col <= 7) {
                infos.get(row).middle = false; // Mark middle group unavailable
            } 
            if (col >= 6 && col <= 9) {
                infos.get(row).right = false; // Mark right group unavailable
            }
        }
        
        // Calculate the maximum number of groups
        int result = 0;
        for (Integer row : infos.keySet()) {
            RowInfo info = infos.get(row);
            result += Math.max((info.left ? 1 : 0) + (info.right ? 1 : 0), info.middle ? 1 : 0);
        }
        
        // Handle rows with no reserved seats (n - infos.size() rows)
        result += (n - infos.size()) * 2;
        
        return result;
    }
}
```

**Explanation of the Code:**

1. **RowInfo Class**:
   - The `RowInfo` class holds information about each row. Specifically, it tracks whether a group (left, middle, or right) is available (`true`) or unavailable (`false`).

2. **maxNumberOfFamilies Method**:

## Problem: Minimum Add to Make Parentheses Valid

A parentheses string is valid if and only if:
1. It is the empty string.
2. It can be written as `AB` (A concatenated with B), where A and B are valid strings.
3. It can be written as `(A)`, where A is a valid string.

You are given a parentheses string `s`. In one move, you can insert a parenthesis at any position of the string. The task is to return the minimum number of moves required to make the string valid.

**Example:**
**Input:** `"()))"`

**Output:** `4`

**Explanation:**
- To make the string valid, we can insert an opening parenthesis at the beginning, resulting in `(()))`. Then, insert a closing parenthesis at the end to make it `()())`. Finally, insert one more closing parenthesis to get `()()()`.

**Approach:**

We can use a simple greedy approach to solve this problem. The key observation is:
- We need to track how many opening and closing parentheses are unmatched as we iterate through the string.
- If we encounter a closing parenthesis without a corresponding opening parenthesis, it indicates that we need an additional opening parenthesis.
- If we encounter an opening parenthesis, we expect to find a matching closing parenthesis later. If no matching closing parenthesis is found, it means we need one more closing parenthesis.

**Steps:**
1. Traverse through the string character by character.
2. Maintain two counters:
   - `open`: Tracks unmatched opening parentheses.
   - `close`: Tracks unmatched closing parentheses.
3. If an opening parenthesis `(` is encountered, increase the `open` counter.
4. If a closing parenthesis `)` is encountered:
   - If there's an unmatched opening parenthesis (`open > 0`), decrease the `open` counter.
   - Otherwise, increase the `close` counter as we need an additional opening parenthesis for the unmatched closing parenthesis.
5. The result will be the sum of the `open` and `close` counters, representing the number of moves required to make the string valid.

**Code Implementation:**

```java
class Solution {
    public int minAddToMakeValid(String s) {
        int open = 0; // Tracks unmatched opening parentheses
        int close = 0; // Tracks unmatched closing parentheses
        
        for (char c : s.toCharArray()) {
            if (c == '(') {
                open++; // Increase open counter for '('
            } else if (c == ')') {
                if (open > 0) {
                    open--; // Decrease open counter if we find a matching ')'
                } else {
                    close++; // Increase close counter if no matching '(' is found
                }
            }
        }
        
        // The result is the sum of unmatched open and close parentheses
        return open + close;
    }
}
```

**Explanation of the Code:**
1. **open** and **close** are initialized to 0. These variables track the number of unmatched opening and closing parentheses respectively.
2. The `for` loop iterates through each character in the input string `s`.
3. If the character is an opening parenthesis `(`, we increment the `open` counter.
4. If the character is a closing parenthesis `)`, we check if there is an unmatched opening parenthesis (`open > 0`). If yes, we decrement the `open` counter, indicating we have matched an opening parenthesis with this closing one. If `open == 0`, it means there is no matching opening parenthesis, so we increment the `close` counter.
5. After iterating through the entire string, the result is the sum of the `open` and `close` counters, which represents the minimum number of moves required to balance the string.

**Time Complexity:**
- **O(n)** where `n` is the length of the string. We only need to make a single pass through the string to calculate the result.

**Space Complexity:**
- **O(1)**. We are using only a constant amount of space for the `open` and `close` counters.

**Example Walkthrough:**

For the input:

```java
String s = "()))";
```

- Initially, `open = 0` and `close = 0`.
- Iterating through the string:
  - First character is `)`, since `open == 0`, we increment `close`, so `close = 1`.
  - Second character is `)`, again no matching `(`, so we increment `close`, so `close = 2`.
  - Third character is `)`, still no matching `(`, so we increment `close`, so `close = 3`.
  - Fourth character is `)`, no matching `(`, so we increment `close`, so `close = 4`.
- Finally, the result is `open + close = 0 + 4 = 4`.

**Output:** `4`

**Conclusion:**
This solution efficiently calculates the minimum number of moves required to make a given parentheses string valid by using a greedy approach to track unmatched opening and closing parentheses.


## Problem: Minimize the Heights I

Given an array `arr[]` representing the heights of towers and a positive integer `k`, you need to modify the height of each tower either by increasing or decreasing it by `k` only once.

The task is to find the possible minimum difference between the height of the shortest and longest towers after modifying each tower.

**Constraints:**
- You can modify each tower's height by either adding `k` or subtracting `k`.
- The goal is to minimize the difference between the maximum and minimum heights of the towers after modification.

**Example:**
**Input:**
```java
arr[] = [1, 5, 8, 10]
k = 2
```

**Output:**
```java
The minimum difference between the tallest and shortest towers is 5.
```

Explanation: After modifying the towers:
- Increase the height of tower at index 0 (1 + 2 = 3).
- Decrease the height of tower at index 3 (10 - 2 = 8).
- The heights of the towers become: `[3, 5, 8, 8]`.
- The minimum difference between the tallest and shortest towers is `8 - 3 = 5`.

**Approach:**
1. **Sorting the Array**: The first step is to sort the array of tower heights in ascending order. This allows us to easily access the smallest and largest heights and efficiently calculate the difference between them after modification.
   
2. **Initial Answer**: The difference between the tallest and shortest towers without any modification is the initial value of the answer. This is simply `arr[n - 1] - arr[0]`, where `arr[n - 1]` is the tallest tower and `arr[0]` is the shortest tower.

3. **Iterative Calculation**: For each tower, consider the two possible modifications:
   - Decrease the height of the tallest tower (or increase the smallest tower).
   - Increase the height of the shortest tower (or decrease the tallest tower).
   
   Then, calculate the new minimum and maximum heights after applying these modifications and update the answer accordingly.

4. **Return the Result**: The answer will be the smallest difference between the tallest and shortest towers found during the iteration.

**Code Implementation:**

```java
class Solution {
    public int getMinDiff(int[] arr, int k) {
        // Step 1: Sort the array to process the tower heights in ascending order
        Arrays.sort(arr);

        int n = arr.length;

        // Initialize the answer with the difference between the tallest and shortest
        // towers without any modifications
        int ans = arr[n - 1] - arr[0];

        // Step 2: Iterate through the array and try to modify each tower's height
        for (int i = 1; i < n; i++) {

            // Calculate the minimum possible height after modification:
            // Either increase the smallest tower by k or decrease the current tower by
            // k
            int mn = Math.min(arr[0] + k, arr[i] - k);

            // Calculate the maximum possible height after modification:
            // Either increase the previous tower by k or decrease the tallest tower by
            // k
            int mx = Math.max(arr[i - 1] + k, arr[n - 1] - k);

            // Update the answer with the smallest difference found so far
            ans = Math.min(ans, mx - mn);
        }

        // Return the minimum possible difference between the tallest and shortest
        // towers after modifications
        return ans;
    }
}
```

**Explanation of the Code:**

1. **Sorting**: We start by sorting the array to get the towers in ascending order, which helps in efficiently calculating the difference between the tallest and shortest towers after modifications.

2. **Initial Answer Calculation**: We first calculate the difference between the tallest and shortest tower without any modifications. This gives us an initial answer.

3. **Iterating through the Array**:
   - For each tower from index 1 to `n-1`, we calculate the possible minimum (`mn`) and maximum (`mx`) tower heights after modifying the current tower.
   - We check if increasing the smallest tower by `k` or decreasing the current tower by `k` results in the minimum possible height.
   - Similarly, we check if increasing the previous tower by `k` or decreasing the largest tower by `k` results in the maximum possible height.
   
4. **Final Answer**: We keep track of the smallest difference between the maximum and minimum tower heights during the iteration and return the result.

**Time Complexity:**
- Sorting the array takes **O(n log n)** time.
- The iteration through the array to compute the minimum difference takes **O(n)** time.
Thus, the overall time complexity is **O(n log n)**.

**Space Complexity:**
- The space complexity is **O(1)** as we are using a constant amount of extra space apart from the input array.

**Example Walkthrough:**
For the input:

```java
arr[] = [1, 5, 8, 10]
k = 2
```

1. After sorting: `arr = [1, 5, 8, 10]`.
2. The initial difference between the tallest and shortest towers: `10 - 1 = 9`.
3. Iterating through the towers:
   - For `i = 1`: 
     - Minimum possible height: `min(1 + 2, 5 - 2) = 3`.
     - Maximum possible height: `max(1 + 2, 10 - 2) = 8`.
     - The new difference is `8 - 3 = 5`.
   - For `i = 2`:
     - Minimum possible height: `min(1 + 2, 8 - 2) = 3`.
     - Maximum possible height: `max(5 + 2, 10 - 2) = 8`.
     - The new difference remains `5`.
4. The minimum possible difference is `5`.

**Output:** `5`

**Conclusion:**
This solution efficiently minimizes the difference between the tallest and shortest towers after modifying the tower heights using the given `k` value. The greedy approach ensures that we get the optimal solution with a time complexity of **O(n log n)**.
