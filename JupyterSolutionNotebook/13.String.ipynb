## Problem Statement: Count and Say

The count-and-say sequence is a sequence of digit strings defined by the recursive formula:

- countAndSay(1) = "1"
- countAndSay(n) is the run-length encoding of countAndSay(n - 1).

Run-length encoding (RLE) is a string compression method that works by replacing consecutive identical characters (repeated 2 or more times) with the concatenation of the character and the number marking the count of the characters (length of the run). 

For example, to compress the string "3322251", we replace "33" with "23", replace "222" with "32", replace "5" with "15", and replace "1" with "11". Thus the compressed string becomes "23321511".

Given a positive integer `n`, return the nth element of the count-and-say sequence.

**Example**

```java
Input: n = 4
Output: "1211"
Explanation: 
    countAndSay(1) = "1"
    countAndSay(2) = "11" (one 1)
    countAndSay(3) = "21" (two 1's)
    countAndSay(4) = "1211" (one 2, one 1)
```


**Input:**  
```plaintext
n = 4
```

**Output:**

```plaintext
"1211"
```

**Constraints:**

* $1 \leq n \leq 30$

**Goal:**
To generate the nth term of the count-and-say sequence using iterative construction based on the description of digits in the previous term.

**🧮 Flowchart/ Diagram:**
*Not provided in the original, optional to include diagram illustrating the iterative generation process from "1" → "11" → "21" → "1211", etc.*

**Approach**

We will use an iterative approach to generate the sequence. Start with `countAndSay(1)` as `"1"`. For each subsequent value, we build the next string by processing the previous string using run-length encoding.

**Steps:**
1. Start with `"1"`.
2. For each next number in the sequence (from 2 to n):
    - Traverse through the current string.
    - Count consecutive identical characters.
    - Append the count and the character to the next string.
3. Repeat this process until we have generated the nth term.

**Pseudocode/Algorithm:**

1. Initialize `result = "1"`
2. Loop from 2 to `n`:
   a. Initialize empty string `nextResult`
   b. Initialize `count = 1`
   c. Loop through characters of `result`:

   * If current char equals previous, increment `count`
   * Else, append `count` and previous char to `nextResult`, reset `count = 1`
     d. After loop, append the last `count` and character
     e. Set `result = nextResult`
3. Return `result`
**Code (Java)**

```java
class Solution {
    public String countAndSay(int n) {
        String result = "1"; // base case
        for (int i = 2; i <= n; i++) {
            StringBuilder nextResult = new StringBuilder();
            int count = 1;
            for (int j = 1; j < result.length(); j++) {
                if (result.charAt(j) == result.charAt(j - 1)) {
                    count++; // Increment the count for the same character
                } else {
                    nextResult.append(count).append(result.charAt(j - 1));
                    count = 1; // reset the count for the next group
                }
            }
            nextResult.append(count).append(result.charAt(result.length() - 1)); // Append the last group
            result = nextResult.toString(); // Update the result for the next iteration
        }

        return result;
    }
}
```

**Time and Space Complexity**

* **Time Complexity**: O(n \* m), where `n` is the number of terms (up to n) and `m` is the average length of the string at each term. This is because for each term, we process the entire string from the previous term.
* **Space Complexity**: O(m), where `m` is the length of the string for the nth term.

**Notes:**

* This is a classic recursive string problem, but efficiently implemented using iteration to avoid stack overhead.
* The string grows approximately exponentially; however, the value of `n` is capped to 30, so this remains efficient in practice.

## Problem Statement: Count the Reversals

Given a string `s` consisting of only opening and closing curly brackets `{` and `}`, find out the minimum number of reversals required to convert the string into a balanced expression. A reversal means changing `{` to `}` or vice-versa.

**A string is considered balanced if:**
- Every opening curly bracket `{` has a corresponding closing curly bracket `}`.
- The curly brackets are correctly nested.

**Example**

```java
Input: s = "}{{}}{{{"
Output: 3
Explanation: 
   - Reverse the first closing bracket to opening: `{{}}{{{`
   - Reverse the last three openings to closing: `{{}}}}`
   - Now the string is balanced, requiring 3 reversals in total.
```


**Input:**  
```plaintext
s = "}{{}}{{{"
```

**Output:**

```plaintext
3
```

**Constraints:**

* $1 \leq |s| \leq 10^5$
* `s` contains only `{` and `}` characters

**Goal:**
To compute the minimum number of bracket reversals required to balance the input string using linear time and constant space.

**🧮 Flowchart/ Diagram:**
*Not provided — could visualize as a stack-based simulation where matching `{` and `}` cancel, and unbalanced portions are counted.*

**Approach**

**1. Initial Check for Even Length**:
   - If the length of the string `s` is odd, it cannot be balanced. This is because a balanced string must contain an even number of brackets (each opening bracket needs a closing one). If the length is odd, return `-1` as it's impossible to balance it.

**2. Track Unmatched Brackets**:
   - Traverse through the string and maintain a count of unmatched opening and closing brackets.
   - For each opening bracket `{`, increment the unmatched opening count.
   - For each closing bracket `}`, if there is an unmatched opening bracket (i.e., a `{`), match it and decrement the unmatched opening count. If there is no unmatched opening bracket, increment the unmatched closing count.

**3. Reversals Calculation**:
   - Once the string is traversed, the unmatched opening and closing brackets will give the minimum number of reversals needed.
   - The number of reversals will be:
     - `(unmatched opening count + 1) / 2` to balance unmatched opening brackets.
     - `(unmatched closing count + 1) / 2` to balance unmatched closing brackets.

**Code (Java)**

```java
class Solution {
    public int countReversals(String s) {
        // If the length of the string is odd, it's impossible to balance
        if (s.length() % 2 != 0) {
            return -1;
        }

        int unmatchedOpening = 0;
        int unmatchedClosing = 0;

        // Traverse the string to count unmatched brackets
        for (int i = 0; i < s.length(); i++) {
            char c = s.charAt(i);
            if (c == '{') {
                unmatchedOpening++;
            } else {
                if (unmatchedOpening > 0) {
                    unmatchedOpening--; // Match the current closing with an opening
                } else {
                    unmatchedClosing++; // This closing bracket is unmatched
                }
            }
        }

        // The number of reversals needed is the half of unmatched openings and closings
        return (unmatchedOpening + 1) / 2 + (unmatchedClosing + 1) / 2;
    }
}
```

**Time and Space Complexity**

* **Time Complexity**: O(n), where `n` is the length of the string `s`, because we traverse the string once.
* **Space Complexity**: O(1), as we only use a few integer variables to keep track of unmatched brackets.

**Code-2**
```java 
// Returns count of minimum reversals for making
// expr balanced. Returns -1 if expr cannot be
// balanced.
class Solution {
    public int countMinReversals(String expr) {
        int len = expr.length();

        // Expressions of odd lengths cannot be balanced
        if (len % 2 != 0) {
            return -1;
        }
        int left_brace = 0, right_brace = 0;
        int ans;
        for (int i = 0; i < len; i++) {

            // If we find a left bracket then we
            // simply increment the left bracket
            if (expr.charAt(i) == '{') {
                left_brace++;
            }

            // Else if left bracket is 0 then we find
            // unbalanced right bracket and increment
            // right bracket or if the expression
            // is balanced then we decrement left
            else {
                if (left_brace == 0) {
                    right_brace++;
                }
                else {
                    left_brace--;
                }
            }
        }
        ans = (int)Math.ceil(left_brace / 2.0) + (int)Math.ceil(right_brace / 2.0);
        return ans;
    }

    public static void main(String[] args) {
        String s = "}{{}}{{{";
        Solution sol = new Solution();
        System.out.println(sol.countMinReversals(s));
    }
}
```



**Pseudocode/Algorithm:**

1. If length of string is odd → return -1 (cannot be balanced).
2. Initialize `left_brace = 0`, `right_brace = 0`.
3. For each character `c` in string:

   * If `c == '{'`, increment `left_brace`.
   * Else if `left_brace == 0`, increment `right_brace`.
   * Else, decrement `left_brace`.
4. Return `ceil(left_brace / 2) + ceil(right_brace / 2)`.

**Notes:**

* If you use integer division, `(x + 1) / 2` achieves `ceil(x / 2)` in integer arithmetic.
* Two variants of the code demonstrate the same logic — one uses integer math tricks, another uses `Math.ceil()` with casting.


## Problem Statement: Search Pattern (Rabin-Karp Algorithm)

Given two strings, one is a text string and the other is a pattern string. The task is to print the starting indexes of all the occurrences of the pattern string in the text string. For printing, the starting index of a string should be taken as `1`. The strings will only contain lowercase English alphabets ('a' to 'z').

**Input:**  
```plaintext
pattern = "abc"
text = "abcabcabc"
```

**Output:**

```plaintext
[1, 4, 7]
```

**Constraints:**

* $1 \leq |text|, |pattern| \leq 10^5$
* Strings consist of only lowercase letters `'a'` to `'z'`

**Example**
```java
Input: pattern = "abc", text = "abcabcabc"
Output: [1, 4, 7]
Explanation: 
The pattern "abc" occurs at index 1, 4, and 7 in the text string "abcabcabc".
```

**Goal:**
Efficiently locate all positions where the pattern appears in the text using Rabin-Karp algorithm for better average-case performance.

**🧮 Flowchart/ Diagram:**
*Not included — could depict a sliding window over the text comparing hash values of substrings with the pattern.*

**Approach**

The problem can be approached using the **Rabin-Karp Algorithm** or a simple brute-force approach.

**Steps for Brute-Force Approach**:
1. Iterate through the `text` string.
2. For each position in the `text`, check if the substring starting from that position matches the `pattern` string.
3. If a match is found, record the index (1-based).
4. Continue this until all possible positions are checked.

**Time Complexity**:
- **Brute-force**: O(n * m), where `n` is the length of the `text` string and `m` is the length of the `pattern` string.
- **Rabin-Karp**: O(n + m) on average for string matching using hashing.

Here, we'll solve it using the brute-force method for simplicity.

**Code (Java)**

```java
class Solution {
    ArrayList<Integer> search(String pattern, String text) {
        // Result list to store the starting indices of the pattern in the text
        ArrayList<Integer> ls = new ArrayList<Integer>();
        
        int textLength = text.length();
        int patternLength = pattern.length();
        
        // Iterate through the text and check for pattern matches
        for(int i = 0; i <= textLength - patternLength; i++) {
            // Compare substring of text with pattern
            if(text.substring(i, i + patternLength).equals(pattern)) {
                // If match is found, add the starting index (1-based)
                ls.add(i + 1);
            }
        }
        // Return the list of starting indices
        return ls;
    }
}
```

**Time and Space Complexity**
* **Time Complexity**: O(n \* m), where `n` is the length of the `text` string and `m` is the length of the `pattern` string, because for each position in the text, we compare a substring of length `m`.
* **Space Complexity**: O(1) for the algorithm itself, excluding the space used to store the result list.

**Code-2**
```java
// Following program is a Java implementation
// of Rabin Karp Algorithm given in the CLRS book

public class Main {
    // d is the number of characters in the input alphabet
    public final static int d = 256;

    // Search the pat string in the txt string
    static void search(String pat, String txt, int q)
    {
        int M = pat.length();
        int N = txt.length();
        int i, j;
        int p = 0; // hash value for pattern
        int t = 0; // hash value for txt
        int h = 1;

        // The value of h would be "pow(d, M-1)%q"
        for (i = 0; i < M - 1; i++)
            h = (h * d) % q;

        // Calculate the hash value of pattern and first
        // window of text
        for (i = 0; i < M; i++) {
            p = (d * p + pat.charAt(i)) % q;
            t = (d * t + txt.charAt(i)) % q;
        }

        // Slide the pattern over text one by one
        for (i = 0; i <= N - M; i++) {

            // Check the hash values of current window of
            // text and pattern. If the hash values match
            // then only check for characters one by one
            if (p == t) {
                /* Check for characters one by one */
                for (j = 0; j < M; j++) {
                    if (txt.charAt(i + j) != pat.charAt(j))
                        break;
                }

                // if p == t and pat[0...M-1] = txt[i, i+1,
                // ...i+M-1]
                if (j == M)
                    System.out.println(
                        "Pattern found at index " + i);
            }

            // Calculate hash value for next window of text:
            // Remove leading digit, add trailing digit
            if (i < N - M) {
                t = (d * (t - txt.charAt(i) * h)
                     + txt.charAt(i + M))
                    % q;

                // We might get negative value of t,
                // converting it to positive
                if (t < 0)
                    t = (t + q);
            }
        }
    }

    /* Driver Code */
    public static void main(String[] args)
    {
        String txt = "GEEKS FOR GEEKS";
        String pat = "GEEK";

        // A prime number
        int q = 101;

        // Function Call
        search(pat, txt, q);
    }
}

// This code is contributed by nuclode
```

**Pseudocode/Algorithm:**

1. Compute hash `p` for the pattern and `t` for the first substring of the text of equal length.
2. Slide a window over the text of size `m` (length of pattern):

   * If hashes match, check characters one by one to confirm.
   * Recompute hash for next window in O(1) time.
3. Record 1-based indices of all matches found.

**Notes:**

* Rabin-Karp is efficient on average due to constant-time rolling hash updates.
* Choose a prime modulus `q` to reduce hash collisions and avoid overflow.


## Problem Statement: Boyer Moore Algorithm For Pattern Searching

You are given a string `text` and a string `pattern`. Your task is to find all occurrences of the pattern in the string `text` and return an array of indexes of all those occurrences. You may assume that the length of `text` is always greater than the length of `pattern`.

**Example:**
Let `text = "this is a good place to have a good start"`, and `pattern = "good"`.  
So, you have to return `{10, 31}` because the pattern "good" is present at indices 10 and 31 in the text.

**Note:**
- If there are no such indexes in the text, return an array containing `-1`.

**Input:**  
```plaintext
text = "this is a good place to have a good start"
pattern = "good"
```

**Output:**

```plaintext
[10, 31]
```

**Constraints:**

* $1 \leq |text|, |pattern| \leq 10^5$
* All characters are printable ASCII (up to 256 characters)

**Example**
```java
Input:
text = "this is a good place to have a good start"
pattern = "good"

Output:
[10, 31]

Explanation:
The pattern "good" occurs at index 10 and index 31 in the text string "this is a good place to have a good start".
```

**Goal:**
Efficient pattern searching using the Boyer-Moore algorithm with bad character heuristic, which skips unnecessary comparisons for faster search.

**🧮 Flowchart/ Diagram:**
*Not included — would typically illustrate shifting the pattern along the text using bad character mismatches.*

**Approach**

We can solve this problem using the **Boyer-Moore Algorithm** for pattern searching. This algorithm uses two primary components for efficient string searching:
1. **Bad character heuristic**: It shifts the pattern to the right in case of a mismatch using the last occurrence of the mismatched character in the pattern.
2. **Good suffix heuristic**: Although not used in this implementation, it can further optimize the search.

We only implement the **Bad Character Heuristic** here for simplicity.

**Steps:**
1. Preprocess the pattern and build a bad character shift array. This array stores the last occurrence of each character in the pattern.
2. Start scanning the text from left to right and compare the pattern with the substring of text.
3. If a mismatch occurs, use the bad character heuristic to shift the pattern appropriately.
4. If a match is found, add the starting index to the result list.
5. If no match is found, return `-1`.

**Pseudocode/Algorithm:**

1. Create `badCharShift[]` to store the last index of every character in the pattern.
2. Align the pattern with the text and compare characters from right to left.
3. On mismatch, shift the pattern based on how far the mismatched character is from the last seen index in the pattern.
4. On match, store the current index and shift by full pattern length (or according to next char).
5. Repeat until the pattern can no longer fit in the remaining text.

**Code (Java)**

```java
import java.util.* ;
import java.io.*; 

public class Solution {

    // Boyer-Moore Pattern Search Algorithm
    public static ArrayList<Integer> patternSearch(String text, String pattern) {
        ArrayList<Integer> result = new ArrayList<>();
        int[] badCharShift = new int[256]; 
        Arrays.fill(badCharShift, -1); // Initialize shift array
        
        // Preprocessing the pattern to fill the bad character shift array
        for (int i = 0; i < pattern.length(); i++) {
            badCharShift[pattern.charAt(i)] = i;
        }
        
        int m = pattern.length();
        int n = text.length();
        int i = 0;
        
        // Start searching the pattern in the text
        while (i <= n - m) {
            int j = m - 1;
            
            // Check for matching pattern from the end
            while (j >= 0 && pattern.charAt(j) == text.charAt(i + j)) {
                j--;
            }
            
            // If a match is found, add the index to result list
            if (j < 0) {
                result.add(i);
                // Shift the pattern using the bad character heuristic
                i += (i + m < n) ? m - badCharShift[text.charAt(i + m)] : 1;
            } else {
                // Shift the pattern based on the last occurrence of the mismatch
                i += Math.max(1, j - badCharShift[text.charAt(i + j)]);
            }
        }
        
        // If no match is found, return [-1]
        if (result.isEmpty()) {
            result.add(-1);
        }
        
        return result;
    }
    
    public static void main(String[] args) {
        String text = "this is a good place to have a good start";
        String pattern = "good";
        
        // Search for the pattern in the text
        ArrayList<Integer> result = patternSearch(text, pattern);
        
        // Print the result
        System.out.println(result);
    }
}
```

**Time and Space Complexity**

* **Time Complexity**:

  * Preprocessing the pattern: O(m), where `m` is the length of the pattern.
  * Searching the text: O(n), where `n` is the length of the text.
  * Overall, the time complexity is O(n + m).

* **Space Complexity**:

  * The space complexity is O(1) for the bad character shift array since it is of constant size (256 for ASCII characters).

**Notes:**

* Boyer-Moore performs better than naive and Rabin-Karp for large texts and small patterns due to efficient character skipping.


## Problem Statement: Longest Prefix Suffix

Given a string of characters `s`, find the length of the **longest prefix** which is also a **suffix**.

**Note:**
- Prefix and suffix can be **overlapping**, but they should **not** be equal to the entire string.

**Example**

**Input:**
```
s = "abab"
```

**Output:**
```
2
```

**Explanation:**
- The longest prefix which is also a suffix is "ab".  
- It is of length 2.

**Constraints:**

* $1 \leq |s| \leq 10^5$
* `s` contains only lowercase English alphabets ('a' to 'z')

**Goal:**
Efficiently compute the longest prefix which is also a suffix using the LPS (Longest Prefix Suffix) logic from the KMP pattern matching algorithm.

**🧮 Diagram/Visual Explanation:**
*Not included — usually a table showing character matches and the LPS array being filled.*

**Example Walkthrough**

For `s = "abcdabc"`:

* LPS array: `[0, 0, 0, 0, 1, 2, 3]`
* Result: `3` (prefix = "abc", suffix = "abc")

For `s = "aaaa"`:

* LPS array: `[0, 1, 2, 3]`
* Result: `3` (prefix = "aaa", suffix = "aaa")


**Approach**

This problem can be efficiently solved using the **KMP (Knuth-Morris-Pratt) algorithm's LPS (Longest Prefix Suffix) array**.

**Key Concepts:**
- The LPS array stores at each index `i` the length of the longest proper prefix which is also a suffix in the substring `s[0..i]`.
- Proper prefix: A prefix that is not equal to the string itself.
- The value at the last index of the LPS array gives the length of the longest prefix which is also a suffix for the entire string.

**Steps:**

1. Initialize an LPS array of size equal to the length of the string.
2. Use two pointers:
   - `i` for iterating through the string.
   - `len` for tracking the current matching prefix and suffix length.
3. Update the LPS array using comparison logic.
4. Return the last value of the LPS array.

**Pseudocode/Algorithm:**

1. Initialize `lps[0] = 0`
2. For each character from index 1 to `n-1`:

   * If `s[i] == s[len]`, increment `len` and assign `lps[i] = len`
   * Else if `len != 0`, update `len = lps[len - 1]` and continue without incrementing `i`
   * Else assign `lps[i] = 0` and move to next character
3. Return `lps[n-1]`

**Code (Java)**

```java
class Solution {
    int longestPrefixSuffix(String s) {
        int n = s.length();
        int[] lps = new int[n];
        int len = 0; // length of the previous longest prefix suffix
        lps[0] = 0; // lps[0] is always 0

        int i = 1;
        while (i < n) {
            if (s.charAt(i) == s.charAt(len)) {
                len++;
                lps[i] = len;
                i++;
            } else {
                if (len != 0) {
                    // Try to find a shorter previous match
                    len = lps[len - 1];
                } else {
                    lps[i] = 0;
                    i++;
                }
            }
        }

        // lps[n - 1] contains the length of the longest prefix which is also a suffix
        return lps[n - 1];
    }
}
```

**Time Complexity:**
- **O(n)** where `n` is the length of the string.

**Space Complexity:**
- **O(n)** for the LPS array.

**Note:**
This LPS technique is foundational in string pattern matching and is particularly useful in problems involving repeated patterns and efficient substring search.


## Problem Statement: Remove All Adjacent Duplicates In String

You are given a string `s` consisting of **lowercase English letters**. A *duplicate removal* consists of choosing two **adjacent and equal letters** and removing them.

You must **repeatedly** make duplicate removals on `s` until no more can be made.

Return the **final string** after all such duplicate removals have been made.

> It can be proven that the answer is **unique**.

**Input Format:**
```plaintext
A single string `s`.
```

**Output Format:**

```plaintext
The resulting string after all adjacent duplicate removals.
```

**Constraints:**
- \(1 \leq |s| \leq 10^5\)
- `s` consists only of lowercase English letters (`'a'` to `'z'`)

**Example-1**

**Input:**
```text
s = "abbaca"
```

**Output:**

```text
"ca"
```

**Explanation:**
* "abbaca" → Remove "bb" → "aaca"
* "aaca" → Remove "aa" → "ca"
* Final result: "ca"


**Example 2**

**Input:**

```text
s = "azxxzy"
```

**Output:**

```text
"ay"
```

**Explanation:**

* Remove "xx" → "azzy"
* Remove "zz" → "ay"
* Final result: "ay"

**Edge Cases:**

* If the string becomes empty after all removals, return an empty string `""`.

**Approach**
To solve this problem efficiently, we can use a **stack** data structure.

**Steps:**
1. Traverse the string character by character.
2. For each character:

   * If the stack is not empty and the top character is equal to the current character, pop the top (i.e., remove the duplicate pair).
   * Otherwise, push the current character onto the stack.
3. After processing all characters, the stack will contain the final characters of the result (in reverse order).
4. Reverse the stack to get the final string.

**Code (Java)**

```java
class Solution {
    public String removeDuplicates(String s) {
        Stack<Character> stack = new Stack<>();
        
        for (char c : s.toCharArray()) {
            if (!stack.isEmpty() && stack.peek() == c) {
                stack.pop(); // Remove adjacent duplicate
            } else {
                stack.push(c);
            }
        }
        
        // Build final string from stack
        StringBuilder result = new StringBuilder();
        while (!stack.isEmpty()) {
            result.append(stack.pop());
        }

        return result.reverse().toString();
    }
}
```


**Time Complexity:**

* **O(n)**, where `n` is the length of the input string.

**Space Complexity:**

* **O(n)** in the worst case (if no duplicates are removed).


**Alternate Implementations:**

* This can also be solved using a character array and pointer to simulate stack behavior to improve performance and avoid stack overhead.


## Problem Statement: Group Anagrams

Given an array of strings `strs`, **group the anagrams** together.

You can return the answer in **any order**.

**Constraints:**
- \(1 \leq \text{strs.length} \leq 10^4\)
- \(0 \leq \text{strs[i].length} \leq 100\)
- `strs[i]` consists of lowercase English letters.

**Input Format:**
```plaintext
A list of lowercase strings, e.g., ["eat", "tea", "tan"]
```

**Output Format:**

```plaintext
A list of groups of anagrams, each group as a list of strings.
```

**Example 1**

**Input:**

```text
strs = [""]
```

**Output:**

```text
[[""]]
```

**Example**

**Input:**
```text
strs = ["eat", "tea", "tan", "ate", "nat", "bat"]
```

**Output:**

```text
[
  ["eat", "tea", "ate"],
  ["tan", "nat"],
  ["bat"]
]
```

**Approach**

We use a **HashMap** to group strings that are anagrams of each other.

**Key Insight:**

Two strings are anagrams if their **sorted versions are equal**.
So, we can use the **sorted string as a key** and group all original strings with the same sorted form.

**Steps:**

1. Create a HashMap where the key is the sorted string, and the value is a list of strings.
2. Loop through each string:

   * Convert the string to a character array.
   * Sort the character array.
   * Convert the sorted array back to a string.
   * Use this string as a key and store the original string in the map.
3. Finally, return the values of the map as the list of grouped anagrams.

**Pseudocode**

```plaintext
function groupAnagrams(strs):
    create empty hashmap map

    for each string s in strs:
        key = sort_characters(s)
        if key not in map:
            map[key] = new empty list
        map[key].append(s)

    return all values in map as a list
```

**Java Code**

```java
class Solution {
    public List<List<String>> groupAnagrams(String[] strs) {
        Map<String, List<String>> map = new HashMap<>();

        for (String s : strs) {
            // Convert string to char array, sort it, and convert back to string
            char[] chars = s.toCharArray();
            Arrays.sort(chars);
            String key = new String(chars); // this becomes the group ID for anagrams

            // Add the original string to the corresponding group
            map.computeIfAbsent(key, k -> new ArrayList<>()).add(s);
        }

        // Return all grouped anagrams
        return new ArrayList<>(map.values());
    }
}
```

**Time and Space Complexity**

* **Time Complexity**: O(n \* k log k), where `n` is the number of strings, and `k` is the maximum length of a string.
* **Space Complexity**: O(n \* k), for storing the output and the hash map.

**Notes:**

* Sorting each string ensures all anagrams become the same key (e.g., "eat", "tea", "ate" → "aet").
* You could optimize by using a frequency count of characters instead of sorting for very long strings.
* This approach guarantees that each group contains only strings that are anagrams of each other.
* Useful in natural language processing, spell-check systems, and cryptography.

## Problem Statement: Search Pattern (Z-algorithm)

Given two strings, one is a **text** string and the other is a **pattern** string.  
The task is to **print the indexes** of all the occurrences of the pattern string in the text string.

📌 **Note**:  
- Starting index of the string should be taken as **1-based**.
- If the pattern is not found, return a list containing `-1`.

**Example**

**Input:**
```
Text = "abcabcabc"
Pattern = "abc"
```

**Output:
```
\[1, 4, 7]
```

**Explanation:**
Pattern `"abc"` occurs at positions 1, 4, and 7 (1-based indexing).

**Approach**

Although the problem mentions Z-algorithm, the provided solution uses a **naive sliding window** method.

**Steps:**
1. Iterate over each substring of the text of length equal to the pattern.
2. Compare each substring to the pattern.
3. If they match, store the **1-based** index.
4. If no matches are found, return a list with `-1`.

**PseudoCode:**

function search(pat, S):

    Initialize an empty list al
    m = length of pat
    n = length of S

    for i = 0 to n - m:
        if substring of S starting from i and having length m equals pat:
            add (i + 1) to al

    if no match found:
        add -1 to al
    
    return al

**Java Code**

```java
class Solution {
    ArrayList<Integer> search(String pat, String S) {
        int m = pat.length();
        int n = S.length();
        ArrayList<Integer> al = new ArrayList<>();
        
        for (int i = 0; i <= n - m; i++) {
            String str = S.substring(i, i + m);
            if (str.equals(pat)) {
                al.add(i + 1); // 1-based indexing
            }
        }
        
        if (n < m || al.size() == 0) {
            al.add(-1);
        }
        
        return al;
    }
}
```

**Time & Space Complexity**

- **Time Complexity**: O((n - m + 1) * m), where `n` is the length of the text, and `m` is the length of the pattern.
- **Space Complexity**: O(1) (excluding the output list).

**Note**
- To solve this efficiently in practice (especially for large inputs), consider using the **Z-algorithm** or **KMP algorithm** for pattern searching.
Notes:

- The solution provided above is a simple brute-force approach, which compares each substring of length m with the pattern. This method is useful for small inputs but becomes inefficient for large text and pattern sizes.

**Z-algorithm:**

     The Z-algorithm is a linear time complexity algorithm used for pattern matching. It computes an array called Z-array that represents the length of the longest substring starting from each position in the string that matches the prefix of the string. You can use the Z-algorithm to efficiently find pattern matches in the text in O(n + m) time complexity.

- For large inputs or performance-critical applications, it's recommended to use Z-algorithm or KMP algorithm (Knuth-Morris-Pratt) for better time complexity.


## Problem Statement: Shortest Palindrome

You are given a string `s`. You can convert `s` to a palindrome by adding characters in **front** of it.

Return the **shortest palindrome** you can find by performing this transformation.

**Example 1:**
```

Input: s = "aacecaaa"
Output: "aaacecaaa"

```

**Example 2:**
```

Input: s = "abcd"
Output: "dcbabcd"

```

**Constraints**

- `0 <= s.length <= 5 * 10^4`
- `s` consists of lowercase English letters only.

**Goal**

To find the shortest palindrome by **only** adding characters in front of the original string.

**Approach**

We use the **KMP (Knuth-Morris-Pratt)** algorithm to find the longest palindromic prefix in the string.

**Steps:**
1. Reverse the original string and call it `rev`.
2. Concatenate the original string with a delimiter (`#`) and the reversed string:  
   `combined = s + "#" + rev`
3. Compute the **Longest Prefix Suffix (LPS)** array for the combined string.
4. The value at the end of the LPS array tells us the longest palindromic prefix in the original string.
5. Add the remaining characters (from `rev`) in front of `s` to form the shortest palindrome.

**pseudocode:**
```java
function shortestPalindrome(s):
    rev = reverse(s)
    combined = s + "#" + rev
    lps = computeLPS(combined)
    palindromeLength = lps[length of combined - 1]
    suffixToAdd = rev[0 to rev.length() - palindromeLength]
    return suffixToAdd + s

function computeLPS(s):
    Initialize an array lps of length len(s)
    len = 0
    i = 1
    while i < len(s):
        if s[i] == s[len]:
            len++
            lps[i] = len
            i++
        else:
            if len != 0:
                len = lps[len - 1]
            else:
                lps[i] = 0
                i++
    return lps
```
**Code (Java)**

```java
class Solution {
    public String shortestPalindrome(String s) {
        String rev = new StringBuilder(s).reverse().toString();
        String combined = s + "#" + rev;

        int[] lps = computeLPS(combined);

        int palindromeLength = lps[combined.length() - 1];
        String suffixToAdd = rev.substring(0, rev.length() - palindromeLength);

        return suffixToAdd + s;
    }

    private int[] computeLPS(String s) {
        int[] lps = new int[s.length()];
        int len = 0;  // length of the previous longest prefix suffix
        int i = 1;

        while (i < s.length()) {
            if (s.charAt(i) == s.charAt(len)) {
                len++;
                lps[i] = len;
                i++;
            } else {
                if (len != 0) {
                    len = lps[len - 1];
                } else {
                    lps[i] = 0;
                    i++;
                }
            }
        }
        return lps;
    }
}
```

**Time and Space Complexity**

* **Time Complexity:** `O(n)`

  * Reversing the string: `O(n)`
  * Constructing `combined`: `O(n)`
  * Computing LPS array using KMP logic: `O(n)`

* **Space Complexity:** `O(n)`

  * Due to LPS array and the `rev`/`combined` strings.

**Notes**

* The key insight is to find the longest palindromic prefix, so we only need to add the **minimum** characters required at the front.
* Using KMP's prefix table ensures we do this efficiently even for large strings.


## Problem Statement: Circle of Strings

Given an array `arr` of lowercase strings, determine if the strings can be chained together to form a circle.  

A string `X` can be chained together with another string `Y` if the **last character of `X`** is the same as the **first character of `Y`**.  

If **every string** of the array can be chained with **exactly two other strings** (one with the first character and one with the last), it will form a circle.

**Example**

**Input:**
```java
arr[] = {"for", "geek", "rig", "kaf"}
```

**Output:**

```
1
```

**Explanation:**
The strings can be ordered as `"for"` → `"rig"` → `"geek"` → `"kaf"` → `"for"` to form a circle.

**Approach**

We treat the problem as a graph where each **character** (`a` to `z`) is a **node**, and each string represents a **directed edge** from its **starting character** to its **ending character**.

**Key Observations:**

1. **Eulerian Circuit Conditions**:

   * The in-degree and out-degree of every node (character) must be the same.
   * All non-zero degree nodes must belong to the same connected component.

2. We build an **adjacency list** and a **degree array** for each character.

3. **DFS** is used to check if all connected characters are part of a single component.

**Psuedocode:**

function isCircle(arr):
    Initialize an adjacency list for 26 characters
    Initialize degree array dp to 0

    for each string in arr:
        a = first character of the string - 'a'
        b = last character of the string - 'a'
        Increment dp[a] and dp[b]
        Add directed edge a -> b in adjacency list

    for each degree in dp:
        if degree is odd:
            return 0  // No Eulerian circuit possible

    Initialize visited array for 26 characters
    Call DFS from the first character of the first string

    for each character in dp:
        if degree > 0 and character is not visited:
            return 0  // Not all connected nodes are visited

    return 1  // Valid Eulerian circuit

**Java Code**

```java
class Solution {
    public int isCircle(String arr[]) {
        List<List<Integer>> list = new ArrayList<>();
        for (int i = 0; i < 26; i++) list.add(new ArrayList<>());

        int[] dp = new int[26];
        for (String s : arr) {
            int a = s.charAt(0) - 'a';
            int b = s.charAt(s.length() - 1) - 'a';
            dp[a]++;
            dp[b]++;
            list.get(a).add(b);
        }

        // Check if all characters have even degree
        for (int i : dp) {
            if (i % 2 == 1) return 0;
        }

        boolean[] vis = new boolean[26];
        solve(arr[0].charAt(0) - 'a', list, vis);

        // Check all nodes with edges are visited
        for (int i = 0; i < 26; i++) {
            if (dp[i] > 0 && !vis[i]) return 0;
        }

        return 1;
    }

    void solve(int i, List<List<Integer>> list, boolean[] vis) {
        if (vis[i]) return;
        vis[i] = true;
        for (int nbr : list.get(i)) {
            solve(nbr, list, vis);
        }
    }
}
```

**Time & Space Complexity**

* **Time Complexity**: O(N + E), where N is number of strings and E is number of edges (characters).
* **Space Complexity**: O(1), since the graph has a max of 26 characters.

**Note**

- This solution ensures that the set of strings can be rearranged to form a circular chain by verifying the necessary conditions from graph theory related to **Eulerian circuits**.

- The solution is based on Eulerian circuit theory in graph theory, which requires that every node (character) with edges has equal in-degree and out-degree.

- DFS is used to check if all nodes with edges are part of the same connected component. This ensures that the entire graph is reachable, forming a circle.

- Time Complexity is linear in terms of the number of strings and the number of characters (edges) due to the construction of the graph and the DFS traversal.

- The problem is efficiently solved in O(N + E) time, where N is the number of strings and E is the total number of edges (characters).



## Problem Statement: Roman Number to Integer

Given a string in **Roman numeral format**, your task is to convert it to an **integer**.

**Roman Symbols and Values**

| Symbol | Value |
|--------|-------|
| I      | 1     |
| V      | 5     |
| X      | 10    |
| L      | 50    |
| C      | 100   |
| D      | 500   |
| M      | 1000  |

**Rules**

- Roman numerals are usually written **largest to smallest** from left to right.
- But if a smaller value symbol appears **before** a larger one, it is **subtracted**.  
  For example:
  - `IV = 4`
  - `IX = 9`
  - `XL = 40`
  - `XC = 90`
  - `CD = 400`
  - `CM = 900`


**Example**
**Input:**
```text
s = "MCMIV"
```

**Output:**
```text
1904
```

**Explanation:**
* M = 1000
* CM = 900
* IV = 4
  \=> 1000 + 900 + 4 = **1904**

**Approach**

- Traverse the Roman string from **left to right**.
- For each character, compare its value to the **next character's** value:
  - If it's **greater than or equal**, add it to the result.
  - If it's **less**, subtract it from the result.

**PseudoCode:**

function romanToInt(s):
    Initialize sum as 0
    Initialize n as the length of the string

    for i from 0 to n-1:
        value1 = getValue(s[i])

        if i + 1 < n:
            value2 = getValue(s[i+1])

            if value1 >= value2:
                sum += value1
            else:
                sum += (value2 - value1)
                i += 1  // Skip the next character
        else:
            sum += value1

    return sum

function getValue(c):
    switch c:
        case 'I': return 1
        case 'V': return 5
        case 'X': return 10
        case 'L': return 50
        case 'C': return 100
        case 'D': return 500
        case 'M': return 1000
        default: return 0

**Java Code (Using `String` methods)**

```java
class Solution {
    public int romanToInt(String s) {
        int sum = 0;
        int n = s.length();
        
        for (int i = 0; i < n; i++) {
            int value1 = getValue(s.charAt(i));

            if (i + 1 < n) {
                int value2 = getValue(s.charAt(i + 1));

                if (value1 >= value2) {
                    sum += value1;
                } else {
                    sum += value2 - value1;
                    i++; // skip next character
                }
            } else {
                sum += value1;
            }
        }

        return sum;
    }

    private int getValue(char c) {
        switch (c) {
            case 'I': return 1;
            case 'V': return 5;
            case 'X': return 10;
            case 'L': return 50;
            case 'C': return 100;
            case 'D': return 500;
            case 'M': return 1000;
            default: return 0;
        }
    }
}
```

**Time and Space Complexity**

* **Time Complexity**: O(n), where n is the length of the input string.
* **Space Complexity**: O(1), constant space used for mapping symbols.

**Notes:**

- The Roman numeral system requires subtracting smaller values from larger ones when a smaller numeral precedes a larger one (e.g., IV for 4).

- The approach efficiently checks each character in the string once and handles the subtraction rule based on the following character.

- The getValue function maps each Roman numeral character to its respective integer value.

- The algorithm runs in linear time, O(n), where n is the length of the string, and uses constant space, O(1), apart from the input string.
